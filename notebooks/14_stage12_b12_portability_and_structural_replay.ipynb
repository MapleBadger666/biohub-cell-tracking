```python
# =============================================================================
# Notebook 14
# 14.1 — B12 SOURCE-DERIVED PORTABLE MATERIALIZER IMPLEMENTATION
#
# ENGINEERING IMPLEMENTATION ONLY
#
# Parent:
#   Notebook 13 R3R9-R1A-R2 PASS
#
# Frozen graph authority:
#   build_current_v2(name)
#       = HISTORICAL_POST_V2_GRAPH_RECONSTRUCTION_AUTHORITY
#
# Exactly TWO semantic replacements are permitted:
#
#   canonical statement 0:
#       g = build_current_v2(name)
#
#   portable:
#       g = post_v2_graph_provider(
#           name,
#           context=post_v2_graph_context,
#       )
#
#   canonical statement 3:
#       pp = peaks[peaks["dataset"].eq(name)].copy()
#
#   portable:
#       pp = suppressed_peak_provider(
#           name,
#           context=suppressed_peak_context,
#       )
#
# Canonical:
#   statement 1
#   statement 2
#   statement 4+
#   post-loop geom -> b12 construction
#
# remain source-derived from the locked canonical B12 writer.
#
# NO:
#   - GT
#   - evaluator
#   - official score
#   - performance metric
#   - historical B12 output/oracle read
#   - historical Morph/V1/V2 runtime action-table read
#   - historical suppressed-peak runtime pickle read
#   - post-V2 fidelity claim
#   - B12 fidelity claim
#   - Git operations
#
# V2 invented sort remains UNRESOLVED and is a downstream blocker.
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import os
import subprocess
import sys
import textwrap

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

SRC = ROOT / "src"

if str(SRC) not in sys.path:
    sys.path.insert(
        0,
        str(SRC),
    )


CANONICAL_WRITER_REL = Path(
    "scripts/run_stage12_fold0_bridge_geometry_universe.py"
)

CANONICAL_WRITER = (
    ROOT / CANONICAL_WRITER_REL
)


PARENT_REL = Path(
    "reports/"
    "notebook13_13_3d4_r3r2_r2r2_r1b_"
    "r3r9_r1a_r2_fv9_module_alias_graph_provenance_closure_v1.json"
)

PARENT = ROOT / PARENT_REL


SUPPRESSED_PROVIDER_REL = Path(
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

SUPPRESSED_PROVIDER = (
    ROOT / SUPPRESSED_PROVIDER_REL
)


MODULE_REL = Path(
    "src/biohub_cell_tracking/stage12_b12_portable.py"
)

MODULE_PATH = ROOT / MODULE_REL


TEST_REL = Path(
    "tests/test_stage12_b12_portable.py"
)

TEST_PATH = ROOT / TEST_REL


ARTIFACT_REL = Path(
    "reports/"
    "notebook14_14_1_"
    "b12_source_derived_portable_materializer_implementation_v1.json"
)

ARTIFACT_PATH = ROOT / ARTIFACT_REL


# Previous interrupted / completed R3R9-R2 location.
OLD_R3R9_R2_ARTIFACT_REL = Path(
    "reports/"
    "notebook13_13_3d4_r3r2_r2r2_r1b_"
    "r3r9_r2_b12_source_derived_portable_materializer_implementation_v1.json"
)

OLD_R3R9_R2_ARTIFACT = (
    ROOT / OLD_R3R9_R2_ARTIFACT_REL
)


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "3328394ff605e99d2c5883aaa8cf128158b8a64e2104876e9d9452301455596e"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "58b1811303697d78571e1958e8b68de250b0c0553f13bc5d757a568016d7ece6"
)

EXPECTED_CANONICAL_WRITER_SHA = (
    "c7fb950c03711ac9c85f0476013f88554a42fc2b8fb21c8cc4afc1bb6b5649a4"
)

EXPECTED_DATASET_LOOP_SHA = (
    "a47787269ee66281709a2760d794846d362c6befe72d3abd0dd677132bd938fc"
)

EXPECTED_SUPPRESSED_PROVIDER_SHA = (
    "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
)

DATASET_LOOP_START = 355
DATASET_LOOP_END = 673
PRIMARY_B12_SINK_LINE = 908


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(text: str) -> str:
    return sha256_bytes(
        text.encode("utf-8")
    )


def sha256_file(path: Path) -> str:
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    obj: Any,
) -> str:

    raw = json.dumps(
        obj,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")

    return sha256_bytes(raw)


def contains_exact_string(
    obj: Any,
    target: str,
) -> bool:

    if isinstance(obj, str):
        return obj == target

    if isinstance(obj, dict):
        return any(
            contains_exact_string(k, target)
            or contains_exact_string(v, target)
            for k, v in obj.items()
        )

    if isinstance(obj, (list, tuple)):
        return any(
            contains_exact_string(v, target)
            for v in obj
        )

    return False


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )

    assert source is not None, (
        f"AST source extraction failed "
        f"for {type(node).__name__} "
        f"L{getattr(node, 'lineno', '?')}"
    )

    return source


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):
        return node.id

    if isinstance(
        node,
        ast.Attribute,
    ):
        left = dotted_name(
            node.value
        )

        if left:
            return (
                f"{left}.{node.attr}"
            )

        return node.attr

    return None


def target_names(
    node: ast.AST,
) -> set[str]:

    result = set()

    def visit(x: ast.AST):

        if isinstance(
            x,
            ast.Name,
        ):
            result.add(
                x.id
            )

        elif isinstance(
            x,
            (
                ast.Tuple,
                ast.List,
            ),
        ):
            for elt in x.elts:
                visit(elt)

        elif isinstance(
            x,
            ast.Starred,
        ):
            visit(
                x.value
            )

        elif isinstance(
            x,
            ast.Subscript,
        ):
            visit(
                x.value
            )

        elif isinstance(
            x,
            ast.Attribute,
        ):
            visit(
                x.value
            )

    visit(node)

    return result


def run_streaming(
    cmd: list[str],
    *,
    cwd: Path,
    title: str,
) -> tuple[int, str]:

    print(
        "=" * 112,
        flush=True,
    )

    print(
        title,
        flush=True,
    )

    print(
        "=" * 112,
        flush=True,
    )

    print(
        "COMMAND:",
        " ".join(
            str(x)
            for x in cmd
        ),
        flush=True,
    )

    print(
        flush=True,
    )


    env = os.environ.copy()

    # Make Python subprocesses unbuffered.
    env["PYTHONUNBUFFERED"] = "1"


    process = subprocess.Popen(
        cmd,
        cwd=str(cwd),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
        env=env,
    )


    assert (
        process.stdout
        is not None
    )


    captured = []


    for line in process.stdout:

        captured.append(
            line
        )

        print(
            line,
            end="",
            flush=True,
        )


    return_code = (
        process.wait()
    )


    print(
        flush=True,
    )

    print(
        "[PROCESS EXIT CODE]",
        return_code,
        flush=True,
    )

    print(
        flush=True,
    )


    return (
        return_code,
        "".join(
            captured
        ),
    )


# =============================================================================
# 3. Preflight
# =============================================================================

for required in (
    ROOT,
    CANONICAL_WRITER,
    PARENT,
    SUPPRESSED_PROVIDER,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.1 "
    "B12 SOURCE-DERIVED PORTABLE MATERIALIZER IMPLEMENTATION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "ROOT:",
    ROOT,
    flush=True,
)

print(
    "MODULE:",
    MODULE_REL,
    flush=True,
)

print(
    "TEST:",
    TEST_REL,
    flush=True,
)

print(
    "ARTIFACT:",
    ARTIFACT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 4. Detect any previous R3R9-R2 formal completion
# =============================================================================

if OLD_R3R9_R2_ARTIFACT.exists():

    old_obj = json.loads(
        OLD_R3R9_R2_ARTIFACT.read_text(
            encoding="utf-8"
        )
    )

    old_decision = (
        old_obj.get(
            "decision"
        )
    )

    print(
        "[INFO] Previous R3R9-R2 artifact exists:",
        OLD_R3R9_R2_ARTIFACT_REL,
        flush=True,
    )

    print(
        "  decision:",
        old_decision,
        flush=True,
    )

    if old_decision == "PASS":

        raise RuntimeError(
            "\nA formal PASS R3R9-R2 artifact already exists from "
            "Notebook 13. Do NOT rebuild 14.1 on top of it. "
            "Return this log so the completed artifact can be inherited "
            "into Notebook 14 instead."
        )


# =============================================================================
# 5. Parent authority lock
# =============================================================================

parent_file_sha = (
    sha256_file(
        PARENT
    )
)

assert (
    parent_file_sha
    == EXPECTED_PARENT_FILE_SHA
)


parent_obj = json.loads(
    PARENT.read_text(
        encoding="utf-8"
    )
)


assert contains_exact_string(
    parent_obj,
    EXPECTED_PARENT_CONTENT_SHA,
)


assert (
    parent_obj.get(
        "decision"
    )
    == "PASS"
)


authority = (
    parent_obj.get(
        "frozen_graph_authority",
        {}
    )
)


assert (
    authority.get(
        "expression"
    )
    == "build_current_v2(name)"
)


assert (
    authority.get(
        "classification"
    )
    == "HISTORICAL_POST_V2_GRAPH_RECONSTRUCTION_AUTHORITY"
)


print(
    "[PASS] PARENT_GRAPH_AUTHORITY_LOCK",
    flush=True,
)

print(
    "  file SHA    :",
    parent_file_sha,
    flush=True,
)

print(
    "  content SHA :",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Canonical / provider source locks
# =============================================================================

canonical_writer_sha = (
    sha256_file(
        CANONICAL_WRITER
    )
)

assert (
    canonical_writer_sha
    == EXPECTED_CANONICAL_WRITER_SHA
)


suppressed_provider_sha = (
    sha256_file(
        SUPPRESSED_PROVIDER
    )
)

assert (
    suppressed_provider_sha
    == EXPECTED_SUPPRESSED_PROVIDER_SHA
)


print(
    "[PASS] SOURCE_LOCKS",
    flush=True,
)

print(
    "  canonical B12 writer :",
    canonical_writer_sha,
    flush=True,
)

print(
    "  suppressed provider  :",
    suppressed_provider_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Parse canonical writer / lock dataset loop
# =============================================================================

writer_text = (
    CANONICAL_WRITER.read_text(
        encoding="utf-8"
    )
)

writer_tree = ast.parse(
    writer_text,
    filename=str(
        CANONICAL_WRITER
    ),
)


loop_matches = [
    node
    for node in ast.walk(
        writer_tree
    )
    if (
        isinstance(
            node,
            (
                ast.For,
                ast.AsyncFor,
            ),
        )
        and node.lineno
        == DATASET_LOOP_START
        and node.end_lineno
        == DATASET_LOOP_END
    )
]


assert (
    len(loop_matches)
    == 1
)


dataset_loop = (
    loop_matches[0]
)


dataset_loop_source = (
    node_source(
        writer_text,
        dataset_loop,
    )
)


dataset_loop_sha = (
    sha256_text(
        dataset_loop_source
    )
)


assert (
    dataset_loop_sha
    == EXPECTED_DATASET_LOOP_SHA
)


assert (
    len(
        dataset_loop.body
    )
    >= 4
)


s0, s1, s2, s3 = (
    dataset_loop.body[:4]
)


# statement 0
assert isinstance(
    s0,
    ast.Assign,
)

assert isinstance(
    s0.value,
    ast.Call,
)

assert (
    dotted_name(
        s0.value.func
    )
    == "build_current_v2"
)

assert (
    "g"
    in target_names(
        s0.targets[0]
    )
)


# statement 1
assert isinstance(
    s1,
    ast.Assign,
)

assert isinstance(
    s1.value,
    ast.Call,
)

assert (
    dotted_name(
        s1.value.func
    )
    == "graph_state"
)


# statement 2
assert isinstance(
    s2,
    ast.Assign,
)

assert isinstance(
    s2.value,
    ast.Call,
)

assert (
    dotted_name(
        s2.value.func
    )
    == "nodes.reset_index"
)


# statement 3
assert isinstance(
    s3,
    ast.Assign,
)

assert (
    "pp"
    in target_names(
        s3.targets[0]
    )
)


canonical_first_four = [
    node_source(
        writer_text,
        stmt,
    )
    for stmt in (
        s0,
        s1,
        s2,
        s3,
    )
]


print(
    "[PASS] CANONICAL_DATASET_LOOP_LOCK",
    flush=True,
)

print(
    "  loop SHA:",
    dataset_loop_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Find dataset loop module position
# =============================================================================

module_loop_index = None


for index, stmt in enumerate(
    writer_tree.body
):

    if stmt is dataset_loop:

        module_loop_index = (
            index
        )

        break


assert (
    module_loop_index
    is not None
)


# =============================================================================
# 9. Source-derived pre-loop mutable initializers
# =============================================================================

MUTATION_METHODS = {
    "append",
    "extend",
    "add",
    "update",
}


mutated_receivers = set()


for call in ast.walk(
    dataset_loop
):

    if not isinstance(
        call,
        ast.Call,
    ):
        continue

    if not isinstance(
        call.func,
        ast.Attribute,
    ):
        continue

    if (
        call.func.attr
        not in MUTATION_METHODS
    ):
        continue

    if isinstance(
        call.func.value,
        ast.Name,
    ):

        mutated_receivers.add(
            call.func.value.id
        )


preloop_initializers = []


for receiver in sorted(
    mutated_receivers
):

    candidates = []

    for stmt in (
        writer_tree.body[
            :module_loop_index
        ]
    ):

        names = set()

        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                names |= (
                    target_names(
                        target
                    )
                )

        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            names |= (
                target_names(
                    stmt.target
                )
            )

        if (
            receiver
            in names
        ):

            candidates.append(
                stmt
            )


    if not candidates:

        continue


    selected = (
        candidates[-1]
    )


    source = node_source(
        writer_text,
        selected,
    )


    # Historical runtime tables must not be carried into portable B12.
    if (
        "read_pickle"
        in source
    ):

        continue


    preloop_initializers.append({
        "name":
            receiver,

        "line":
            selected.lineno,

        "source":
            source,

        "sha256":
            sha256_text(
                source
            ),
    })


print(
    "[INFO] PRELOOP INITIALIZERS",
    flush=True,
)


if not preloop_initializers:

    print(
        "  <none>",
        flush=True,
    )


for item in (
    preloop_initializers
):

    print(
        f"  {item['name']:<20} "
        f"L{item['line']}: "
        f"{item['source']}",
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 10. Retained canonical loop source
# =============================================================================

retained_loop_nodes = [
    s1,
    s2,
    *dataset_loop.body[4:],
]


retained_loop_sources = [
    node_source(
        writer_text,
        node,
    )
    for node in (
        retained_loop_nodes
    )
]


print(
    "[PASS] RETAINED_CANONICAL_LOOP_BODY_EXTRACTED",
    flush=True,
)

print(
    "  retained statements:",
    len(
        retained_loop_sources
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Extract post-loop canonical geom -> b12 source
#
# Exclude external output sinks.
# =============================================================================

def call_names_in_statement(
    stmt: ast.AST,
) -> list[str]:

    result = []

    for call in ast.walk(
        stmt
    ):

        if not isinstance(
            call,
            ast.Call,
        ):
            continue

        name = dotted_name(
            call.func
        )

        if name:

            result.append(
                name
            )

    return result


OUTPUT_SUFFIXES = (
    ".to_pickle",
    ".to_csv",
    ".to_json",
    ".write_text",
    ".write_bytes",
)


postloop_records = []


for stmt in (
    writer_tree.body[
        module_loop_index + 1:
    ]
):

    if (
        stmt.lineno
        >= PRIMARY_B12_SINK_LINE
    ):

        break


    calls = (
        call_names_in_statement(
            stmt
        )
    )


    if any(
        (
            name == "open"
            or name.endswith(
                OUTPUT_SUFFIXES
            )
        )
        for name in calls
    ):

        continue


    source = node_source(
        writer_text,
        stmt,
    )


    assert (
        "read_pickle"
        not in source
    ), (
        f"Historical read_pickle found "
        f"in post-loop source L{stmt.lineno}."
    )


    postloop_records.append({
        "line":
            stmt.lineno,

        "source":
            source,

        "sha256":
            sha256_text(
                source
            ),

        "calls":
            calls,
    })


postloop_source = "\n\n".join(
    item["source"]
    for item in (
        postloop_records
    )
)


postloop_tree = ast.parse(
    (
        postloop_source
        if postloop_source.strip()
        else "pass"
    )
)


b12_definition_count = 0


for stmt in (
    postloop_tree.body
):

    names = set()

    if isinstance(
        stmt,
        ast.Assign,
    ):

        for target in (
            stmt.targets
        ):

            names |= (
                target_names(
                    target
                )
            )

    elif isinstance(
        stmt,
        ast.AnnAssign,
    ):

        names |= (
            target_names(
                stmt.target
            )
        )


    if (
        "b12"
        in names
    ):

        b12_definition_count += 1


assert (
    b12_definition_count
    >= 1
), (
    "No source-derived `b12` construction found "
    "between canonical dataset loop and B12 sink."
)


print(
    "[PASS] POSTLOOP_GEOM_TO_B12_SOURCE_EXTRACTED",
    flush=True,
)

print(
    "  postloop statements:",
    len(
        postloop_records
    ),
    flush=True,
)

print(
    "  b12 definitions:",
    b12_definition_count,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Generate stage12_b12_portable.py
# =============================================================================

module_source = r'''
from __future__ import annotations

import ast
import hashlib
import textwrap

from dataclasses import dataclass, field
from pathlib import Path
from typing import Any, Callable, Mapping

import pandas as pd

from biohub_cell_tracking.stage12_suppressed_peak_provider import (
    suppressed_peak_provider,
)


CANONICAL_WRITER_REL = Path(
    "scripts/run_stage12_fold0_bridge_geometry_universe.py"
)

CANONICAL_WRITER_SHA256 = "__WRITER_SHA__"

CANONICAL_DATASET_LOOP_SHA256 = "__LOOP_SHA__"

CANONICAL_DATASET_LOOP_START = __LOOP_START__
CANONICAL_DATASET_LOOP_END = __LOOP_END__
CANONICAL_B12_SINK_LINE = __B12_SINK_LINE__


FORBIDDEN_HISTORICAL_RUNTIME_FILENAMES = frozenset({
    "stage11_fold0_morphdiv_v2_actions_locked.pkl",
    "stage12_fold0_mutual_cont_v1_actions_locked.pkl",
    "stage12_fold0_mutual_cont_v2_actions_locked.pkl",
    "stage12_fold0_suppressed_peak_universe.pkl",
})


@dataclass(frozen=True)
class B12PortableContext:
    post_v2_graph_context: Any
    suppressed_peak_context: Any

    # Explicit source-derived namespace additions only.
    # Historical action/suppressed tables are not accepted.
    canonical_namespace: Mapping[str, Any] = field(
        default_factory=dict
    )


def _sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(
        data
    ).hexdigest()


def _sha256_text(text: str) -> str:
    return _sha256_bytes(
        text.encode("utf-8")
    )


def _project_root() -> Path:
    return (
        Path(__file__)
        .resolve()
        .parents[2]
    )


def _dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):
        return node.id

    if isinstance(
        node,
        ast.Attribute,
    ):

        left = _dotted_name(
            node.value
        )

        if left:
            return (
                f"{left}.{node.attr}"
            )

        return node.attr

    return None


def _target_names(
    node: ast.AST,
) -> set[str]:

    result = set()

    def visit(x: ast.AST):

        if isinstance(
            x,
            ast.Name,
        ):

            result.add(
                x.id
            )

        elif isinstance(
            x,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for elt in x.elts:
                visit(
                    elt
                )

        elif isinstance(
            x,
            ast.Starred,
        ):

            visit(
                x.value
            )

        elif isinstance(
            x,
            ast.Subscript,
        ):

            visit(
                x.value
            )

        elif isinstance(
            x,
            ast.Attribute,
        ):

            visit(
                x.value
            )

    visit(node)

    return result


def _node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )

    if source is None:

        raise AssertionError(
            "Canonical AST source extraction failed."
        )

    return source


def _load_canonical_source(
    project_root: Path,
) -> tuple[
    Path,
    str,
    ast.Module,
]:

    path = (
        project_root
        / CANONICAL_WRITER_REL
    )

    raw = path.read_bytes()

    actual_sha = (
        _sha256_bytes(
            raw
        )
    )

    if (
        actual_sha
        != CANONICAL_WRITER_SHA256
    ):

        raise AssertionError(
            "Canonical B12 writer SHA mismatch: "
            f"{actual_sha}"
        )


    text = raw.decode(
        "utf-8"
    )

    tree = ast.parse(
        text,
        filename=str(
            path
        ),
    )


    return (
        path,
        text,
        tree,
    )


def _extract_plan(
    project_root: Path | None = None,
) -> dict[str, Any]:

    if project_root is None:

        project_root = (
            _project_root()
        )


    path, text, tree = (
        _load_canonical_source(
            project_root
        )
    )


    loops = [
        node
        for node in ast.walk(
            tree
        )
        if (
            isinstance(
                node,
                (
                    ast.For,
                    ast.AsyncFor,
                ),
            )
            and node.lineno
            == CANONICAL_DATASET_LOOP_START
            and node.end_lineno
            == CANONICAL_DATASET_LOOP_END
        )
    ]


    if len(loops) != 1:

        raise AssertionError(
            "Canonical B12 dataset loop "
            "was not uniquely identified."
        )


    loop = loops[0]


    loop_source = (
        _node_source(
            text,
            loop,
        )
    )


    if (
        _sha256_text(
            loop_source
        )
        != CANONICAL_DATASET_LOOP_SHA256
    ):

        raise AssertionError(
            "Canonical dataset-loop SHA mismatch."
        )


    if len(loop.body) < 4:

        raise AssertionError(
            "Canonical B12 dataset loop "
            "contains fewer than four statements."
        )


    s0, s1, s2, s3 = (
        loop.body[:4]
    )


    # -------------------------------------------------------------
    # Canonical statement 0
    # -------------------------------------------------------------

    if not (
        isinstance(
            s0,
            ast.Assign,
        )
        and isinstance(
            s0.value,
            ast.Call,
        )
        and _dotted_name(
            s0.value.func
        )
        == "build_current_v2"
        and "g"
        in _target_names(
            s0.targets[0]
        )
    ):

        raise AssertionError(
            "Canonical statement 0 changed."
        )


    # -------------------------------------------------------------
    # Canonical statement 1
    # -------------------------------------------------------------

    if not (
        isinstance(
            s1,
            ast.Assign,
        )
        and isinstance(
            s1.value,
            ast.Call,
        )
        and _dotted_name(
            s1.value.func
        )
        == "graph_state"
    ):

        raise AssertionError(
            "Canonical statement 1 changed."
        )


    # -------------------------------------------------------------
    # Canonical statement 2
    # -------------------------------------------------------------

    if not (
        isinstance(
            s2,
            ast.Assign,
        )
        and isinstance(
            s2.value,
            ast.Call,
        )
        and _dotted_name(
            s2.value.func
        )
        == "nodes.reset_index"
    ):

        raise AssertionError(
            "Canonical statement 2 changed."
        )


    # -------------------------------------------------------------
    # Canonical statement 3
    # -------------------------------------------------------------

    if not (
        isinstance(
            s3,
            ast.Assign,
        )
        and "pp"
        in _target_names(
            s3.targets[0]
        )
    ):

        raise AssertionError(
            "Canonical statement 3 changed."
        )


    module_loop_index = None


    for index, stmt in enumerate(
        tree.body
    ):

        if stmt is loop:

            module_loop_index = (
                index
            )

            break


    if (
        module_loop_index
        is None
    ):

        raise AssertionError(
            "Canonical dataset loop missing from module body."
        )


    # -------------------------------------------------------------
    # Source-derived mutable-container initializers.
    # -------------------------------------------------------------

    mutation_methods = {
        "append",
        "extend",
        "add",
        "update",
    }


    mutated_receivers = set()


    for call in ast.walk(
        loop
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        if not isinstance(
            call.func,
            ast.Attribute,
        ):

            continue


        if (
            call.func.attr
            not in mutation_methods
        ):

            continue


        if isinstance(
            call.func.value,
            ast.Name,
        ):

            mutated_receivers.add(
                call.func.value.id
            )


    preloop_initializers = []


    for receiver in sorted(
        mutated_receivers
    ):

        candidates = []


        for stmt in (
            tree.body[
                :module_loop_index
            ]
        ):

            names = set()


            if isinstance(
                stmt,
                ast.Assign,
            ):

                for target in (
                    stmt.targets
                ):

                    names |= (
                        _target_names(
                            target
                        )
                    )


            elif isinstance(
                stmt,
                ast.AnnAssign,
            ):

                names |= (
                    _target_names(
                        stmt.target
                    )
                )


            if receiver in names:

                candidates.append(
                    stmt
                )


        if not candidates:

            continue


        selected = (
            candidates[-1]
        )


        source = (
            _node_source(
                text,
                selected,
            )
        )


        if (
            "read_pickle"
            in source
        ):

            continue


        preloop_initializers.append(
            source
        )


    # -------------------------------------------------------------
    # Canonical retained loop body:
    #
    # statement 1
    # statement 2
    # statement 4+
    # -------------------------------------------------------------

    retained_loop_sources = [
        _node_source(
            text,
            node,
        )
        for node in [
            s1,
            s2,
            *loop.body[4:],
        ]
    ]


    # -------------------------------------------------------------
    # Canonical post-loop source up to B12 sink.
    # External output sinks are excluded.
    # -------------------------------------------------------------

    output_suffixes = (
        ".to_pickle",
        ".to_csv",
        ".to_json",
        ".write_text",
        ".write_bytes",
    )


    postloop_sources = []


    for stmt in (
        tree.body[
            module_loop_index + 1:
        ]
    ):

        if (
            stmt.lineno
            >= CANONICAL_B12_SINK_LINE
        ):

            break


        call_names = []


        for call in ast.walk(
            stmt
        ):

            if not isinstance(
                call,
                ast.Call,
            ):

                continue


            name = _dotted_name(
                call.func
            )


            if name:

                call_names.append(
                    name
                )


        if any(
            (
                name == "open"
                or name.endswith(
                    output_suffixes
                )
            )
            for name in (
                call_names
            )
        ):

            continue


        source = (
            _node_source(
                text,
                stmt,
            )
        )


        if (
            "read_pickle"
            in source
        ):

            raise AssertionError(
                "Historical pickle read entered "
                "post-loop portable source."
            )


        postloop_sources.append(
            source
        )


    # Must construct b12 before sink.
    parsed_postloop = ast.parse(
        (
            "\n\n".join(
                postloop_sources
            )
            or "pass"
        )
    )


    b12_defs = 0


    for stmt in (
        parsed_postloop.body
    ):

        names = set()


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                names |= (
                    _target_names(
                        target
                    )
                )


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            names |= (
                _target_names(
                    stmt.target
                )
            )


        if (
            "b12"
            in names
        ):

            b12_defs += 1


    if b12_defs < 1:

        raise AssertionError(
            "Canonical post-loop source does not construct b12."
        )


    return {
        "path":
            path,

        "text":
            text,

        "tree":
            tree,

        "loop":
            loop,

        "canonical_statement_0":
            _node_source(
                text,
                s0,
            ),

        "canonical_statement_1":
            _node_source(
                text,
                s1,
            ),

        "canonical_statement_2":
            _node_source(
                text,
                s2,
            ),

        "canonical_statement_3":
            _node_source(
                text,
                s3,
            ),

        "preloop_initializers":
            preloop_initializers,

        "retained_loop_sources":
            retained_loop_sources,

        "postloop_sources":
            postloop_sources,
    }


def _statement_has_forbidden_io(
    stmt: ast.AST,
) -> bool:

    for call in ast.walk(
        stmt
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = _dotted_name(
            call.func
        )


        if name in {
            "open",
            "pd.read_pickle",
            "pd.read_csv",
            "read_pickle",
            "read_csv",
        }:

            return True


        if (
            name
            and (
                name.endswith(
                    ".read_text"
                )
                or name.endswith(
                    ".read_bytes"
                )
                or name.endswith(
                    ".to_pickle"
                )
            )
        ):

            return True


    return False


def _build_canonical_namespace(
    plan: Mapping[str, Any],
    extra_namespace: Mapping[
        str,
        Any,
    ] | None = None,
) -> dict[str, Any]:

    text = plan[
        "text"
    ]

    tree = plan[
        "tree"
    ]

    loop = plan[
        "loop"
    ]


    namespace: dict[
        str,
        Any,
    ] = {
        "__builtins__":
            __builtins__,

        "__file__":
            str(
                plan[
                    "path"
                ]
            ),

        "__name__":
            "_stage12_b12_canonical_runtime",
    }


    # These historical/runtime-state variables are NEVER reconstructed here.
    forbidden_names = {
        "morph",
        "v1",
        "v2",
        "peaks",
        "cv",
        "corp",
        "names",
        "prefix_map",
        "policies",
    }


    # -------------------------------------------------------------
    # Canonical imports.
    # -------------------------------------------------------------

    for stmt in (
        tree.body
    ):

        if not isinstance(
            stmt,
            (
                ast.Import,
                ast.ImportFrom,
            ),
        ):

            continue


        code = compile(
            ast.Module(
                body=[
                    stmt
                ],
                type_ignores=[],
            ),
            filename=str(
                plan[
                    "path"
                ]
            ),
            mode="exec",
        )


        exec(
            code,
            namespace,
            namespace,
        )


    # -------------------------------------------------------------
    # Safe source-derived module constants / path definitions.
    # -------------------------------------------------------------

    for stmt in (
        tree.body
    ):

        if (
            stmt.lineno
            >= loop.lineno
        ):

            break


        if not isinstance(
            stmt,
            (
                ast.Assign,
                ast.AnnAssign,
            ),
        ):

            continue


        targets = set()


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                targets |= (
                    _target_names(
                        target
                    )
                )


        else:

            targets |= (
                _target_names(
                    stmt.target
                )
            )


        if (
            targets
            & forbidden_names
        ):

            continue


        if (
            _statement_has_forbidden_io(
                stmt
            )
        ):

            continue


        code = compile(
            ast.Module(
                body=[
                    stmt
                ],
                type_ignores=[],
            ),
            filename=str(
                plan[
                    "path"
                ]
            ),
            mode="exec",
        )


        try:

            exec(
                code,
                namespace,
                namespace,
            )


        except (
            NameError,
            AttributeError,
            TypeError,
            FileNotFoundError,
        ):

            # Safe to defer; unresolved source-derived names may be supplied
            # explicitly via context.canonical_namespace.
            continue


    # -------------------------------------------------------------
    # Canonical helper function/class definitions.
    #
    # build_current_v2 is deliberately excluded so portable runtime cannot
    # accidentally call historical graph reconstruction.
    # -------------------------------------------------------------

    for stmt in (
        tree.body
    ):

        if not isinstance(
            stmt,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):

            continue


        if (
            isinstance(
                stmt,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                ),
            )
            and stmt.name
            == "build_current_v2"
        ):

            continue


        code = compile(
            ast.Module(
                body=[
                    stmt
                ],
                type_ignores=[],
            ),
            filename=str(
                plan[
                    "path"
                ]
            ),
            mode="exec",
        )


        try:

            exec(
                code,
                namespace,
                namespace,
            )


        except NameError:

            continue


    if extra_namespace:

        namespace.update(
            dict(
                extra_namespace
            )
        )


    # Never allow caller to smuggle historical runtime tables back in.
    for name in (
        forbidden_names
    ):

        namespace.pop(
            name,
            None,
        )


    return namespace


def _indent_source(
    source: str,
    spaces: int,
) -> str:

    return textwrap.indent(
        textwrap.dedent(
            source
        ).rstrip(),
        " " * spaces,
    )


def _build_executor_source(
    plan: Mapping[str, Any],
) -> str:

    lines = [
        "def __portable_b12_executor(",
        "    dataset_id,",
        "    *,",
        "    post_v2_graph_provider,",
        "    post_v2_graph_context,",
        "    suppressed_peak_context,",
        "):",
        "    names = [dataset_id]",
    ]


    for source in (
        plan[
            "preloop_initializers"
        ]
    ):

        lines.append(
            _indent_source(
                source,
                4,
            )
        )


    # Use the canonical loop target + iterator semantics.
    loop = plan[
        "loop"
    ]


    loop_target = ast.unparse(
        loop.target
    )


    # We intentionally use [dataset_id] for explicit one-dataset runtime,
    # while preserving canonical enumerate/start semantics.
    #
    # Canonical loop is known to be enumerate(names, start=1).
    loop_iter = ast.unparse(
        loop.iter
    )


    if (
        "enumerate(names"
        not in loop_iter.replace(
            " ",
            "",
        )
        and "enumerate(names"
        not in loop_iter
    ):

        # More robust normalized check.
        normalized = (
            loop_iter
            .replace(
                " ",
                ""
            )
        )

        if (
            not normalized.startswith(
                "enumerate(names"
            )
        ):

            raise AssertionError(
                "Canonical B12 loop iterator changed: "
                f"{loop_iter}"
            )


    lines.extend([
        "",
        f"    for {loop_target} in {loop_iter}:",
        (
            "        g = post_v2_graph_provider("
        ),
        (
            "            name,"
        ),
        (
            "            context=post_v2_graph_context,"
        ),
        (
            "        )"
        ),
    ])


    # Canonical statement 1 and statement 2.
    for source in (
        plan[
            "retained_loop_sources"
        ][:2]
    ):

        lines.append(
            _indent_source(
                source,
                8,
            )
        )


    # Frozen semantic replacement for canonical statement 3.
    lines.extend([
        (
            "        pp = suppressed_peak_provider("
        ),
        (
            "            name,"
        ),
        (
            "            context=suppressed_peak_context,"
        ),
        (
            "        )"
        ),
    ])


    # Canonical statement 4+.
    for source in (
        plan[
            "retained_loop_sources"
        ][2:]
    ):

        lines.append(
            _indent_source(
                source,
                8,
            )
        )


    lines.append(
        ""
    )


    # Canonical post-loop geom -> b12 construction.
    for source in (
        plan[
            "postloop_sources"
        ]
    ):

        lines.append(
            _indent_source(
                source,
                4,
            )
        )


    lines.extend([
        "",
        "    return b12",
        "",
    ])


    source = "\n".join(
        lines
    )


    if (
        "read_pickle"
        in source
    ):

        raise AssertionError(
            "Portable executor contains read_pickle."
        )


    for filename in (
        FORBIDDEN_HISTORICAL_RUNTIME_FILENAMES
    ):

        if filename in source:

            raise AssertionError(
                "Historical artifact leaked into portable executor: "
                f"{filename}"
            )


    if (
        "build_current_v2("
        in source
    ):

        raise AssertionError(
            "Historical build_current_v2 leaked into portable executor."
        )


    return source


def materialize_b12_dataset(
    dataset_id: str,
    *,
    post_v2_graph_provider: Callable[..., Any],
    context: B12PortableContext,
) -> pd.DataFrame:

    if (
        not isinstance(
            dataset_id,
            str,
        )
        or not dataset_id
    ):

        raise ValueError(
            "dataset_id must be a non-empty string."
        )


    if not callable(
        post_v2_graph_provider
    ):

        raise TypeError(
            "post_v2_graph_provider must be callable."
        )


    plan = _extract_plan(
        _project_root()
    )


    namespace = (
        _build_canonical_namespace(
            plan,
            context.canonical_namespace,
        )
    )


    namespace[
        "post_v2_graph_provider"
    ] = post_v2_graph_provider


    namespace[
        "suppressed_peak_provider"
    ] = suppressed_peak_provider


    executor_source = (
        _build_executor_source(
            plan
        )
    )


    code = compile(
        executor_source,
        filename="<stage12_b12_portable_executor>",
        mode="exec",
    )


    exec(
        code,
        namespace,
        namespace,
    )


    executor = namespace[
        "__portable_b12_executor"
    ]


    b12 = executor(
        dataset_id,
        post_v2_graph_provider=(
            post_v2_graph_provider
        ),
        post_v2_graph_context=(
            context.post_v2_graph_context
        ),
        suppressed_peak_context=(
            context.suppressed_peak_context
        ),
    )


    if not isinstance(
        b12,
        pd.DataFrame,
    ):

        raise TypeError(
            "Portable B12 executor did not return pandas.DataFrame."
        )


    if (
        "dataset"
        in b12.columns
    ):

        wrong_dataset = (
            b12[
                "dataset"
            ]
            .astype(str)
            .ne(
                dataset_id
            )
        )


        if bool(
            wrong_dataset.any()
        ):

            raise AssertionError(
                "Portable B12 output contains another dataset."
            )


    return b12.copy(
        deep=True
    )


__all__ = [
    "B12PortableContext",
    "materialize_b12_dataset",
]
'''


module_source = (
    module_source
    .replace(
        "__WRITER_SHA__",
        EXPECTED_CANONICAL_WRITER_SHA,
    )
    .replace(
        "__LOOP_SHA__",
        EXPECTED_DATASET_LOOP_SHA,
    )
    .replace(
        "__LOOP_START__",
        str(
            DATASET_LOOP_START
        ),
    )
    .replace(
        "__LOOP_END__",
        str(
            DATASET_LOOP_END
        ),
    )
    .replace(
        "__B12_SINK_LINE__",
        str(
            PRIMARY_B12_SINK_LINE
        ),
    )
)


module_source = (
    textwrap.dedent(
        module_source
    ).lstrip()
)


# Syntax gate before touching file.
compile(
    module_source,
    filename=str(
        MODULE_PATH
    ),
    mode="exec",
)


# =============================================================================
# 13. Static safety audit of generated module
# =============================================================================

module_tree = ast.parse(
    module_source,
    filename=str(
        MODULE_PATH
    ),
)


literal_calls = [
    dotted_name(
        node.func
    )
    for node in ast.walk(
        module_tree
    )
    if isinstance(
        node,
        ast.Call,
    )
]


assert (
    "pd.read_pickle"
    not in literal_calls
)


assert (
    "read_pickle"
    not in literal_calls
)


assert (
    "post_v2_graph_provider("
    in module_source
)


assert (
    "suppressed_peak_provider("
    in module_source
)


print(
    "[PASS] GENERATED_MODULE_STATIC_SAFETY",
    flush=True,
)

print(
    "  no pd.read_pickle runtime call",
    flush=True,
)

print(
    "  post_v2_graph_provider seam present",
    flush=True,
)

print(
    "  suppressed_peak_provider seam present",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Handle pre-existing interrupted drafts transparently
# =============================================================================

old_module_sha = (
    sha256_file(
        MODULE_PATH
    )
    if MODULE_PATH.exists()
    else None
)


old_test_sha = (
    sha256_file(
        TEST_PATH
    )
    if TEST_PATH.exists()
    else None
)


module_bytes = (
    module_source.encode(
        "utf-8"
    )
)


new_module_sha = (
    sha256_bytes(
        module_bytes
    )
)


if MODULE_PATH.exists():

    if (
        MODULE_PATH.read_bytes()
        == module_bytes
    ):

        module_write_mode = (
            "EXISTING_EXACT_MODULE_REUSED"
        )

    else:

        # No formal old PASS artifact exists, so this is an interrupted,
        # unfrozen implementation draft from the aborted Notebook 13 attempt.
        MODULE_PATH.write_bytes(
            module_bytes
        )

        module_write_mode = (
            "UNFROZEN_INTERRUPTED_DRAFT_REPLACED"
        )

else:

    MODULE_PATH.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    MODULE_PATH.write_bytes(
        module_bytes
    )

    module_write_mode = (
        "NEW_MODULE_WRITTEN"
    )


module_sha = (
    sha256_file(
        MODULE_PATH
    )
)


assert (
    module_sha
    == new_module_sha
)


print(
    "[PASS] MODULE_WRITE",
    flush=True,
)

print(
    "  mode    :",
    module_write_mode,
    flush=True,
)

print(
    "  old SHA :",
    old_module_sha,
    flush=True,
)

print(
    "  new SHA :",
    module_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Generate targeted tests
# =============================================================================

test_source = r'''
from __future__ import annotations

import ast
import sys

from pathlib import Path


ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

SRC = ROOT / "src"

if str(SRC) not in sys.path:
    sys.path.insert(
        0,
        str(SRC),
    )


from biohub_cell_tracking import stage12_b12_portable as b12p


def test_canonical_source_lock_and_frozen_statements():

    plan = (
        b12p._extract_plan(
            ROOT
        )
    )


    assert (
        "build_current_v2"
        in plan[
            "canonical_statement_0"
        ]
    )


    assert (
        "graph_state"
        in plan[
            "canonical_statement_1"
        ]
    )


    assert (
        "reset_index"
        in plan[
            "canonical_statement_2"
        ]
    )


    assert (
        "peaks"
        in plan[
            "canonical_statement_3"
        ]
    )


def test_executor_has_exact_portable_seams():

    plan = (
        b12p._extract_plan(
            ROOT
        )
    )


    source = (
        b12p._build_executor_source(
            plan
        )
    )


    tree = ast.parse(
        source
    )


    calls = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            ast.Call,
        ):

            continue


        name = b12p._dotted_name(
            node.func
        )


        if name:

            calls.append(
                name
            )


    assert (
        "post_v2_graph_provider"
        in calls
    )


    assert (
        "suppressed_peak_provider"
        in calls
    )


    assert (
        "build_current_v2"
        not in calls
    )


def test_executor_contains_no_historical_runtime_artifacts():

    plan = (
        b12p._extract_plan(
            ROOT
        )
    )


    source = (
        b12p._build_executor_source(
            plan
        )
    )


    assert (
        "read_pickle"
        not in source
    )


    for filename in (
        b12p.FORBIDDEN_HISTORICAL_RUNTIME_FILENAMES
    ):

        assert (
            filename
            not in source
        )


def test_executor_compiles():

    plan = (
        b12p._extract_plan(
            ROOT
        )
    )


    source = (
        b12p._build_executor_source(
            plan
        )
    )


    compile(
        source,
        "<b12-portable-executor-test>",
        "exec",
    )


def test_postloop_constructs_b12():

    plan = (
        b12p._extract_plan(
            ROOT
        )
    )


    source = "\n\n".join(
        plan[
            "postloop_sources"
        ]
    )


    tree = ast.parse(
        source
        if source.strip()
        else "pass"
    )


    found = False


    for stmt in (
        tree.body
    ):

        names = set()


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                names |= (
                    b12p._target_names(
                        target
                    )
                )


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            names |= (
                b12p._target_names(
                    stmt.target
                )
            )


        if (
            "b12"
            in names
        ):

            found = True


    assert found


def test_public_surface_is_narrow():

    assert (
        b12p.__all__
        == [
            "B12PortableContext",
            "materialize_b12_dataset",
        ]
    )
'''


test_source = (
    textwrap.dedent(
        test_source
    ).lstrip()
)


compile(
    test_source,
    filename=str(
        TEST_PATH
    ),
    mode="exec",
)


test_bytes = (
    test_source.encode(
        "utf-8"
    )
)


new_test_sha = (
    sha256_bytes(
        test_bytes
    )
)


if TEST_PATH.exists():

    if (
        TEST_PATH.read_bytes()
        == test_bytes
    ):

        test_write_mode = (
            "EXISTING_EXACT_TEST_REUSED"
        )

    else:

        TEST_PATH.write_bytes(
            test_bytes
        )

        test_write_mode = (
            "UNFROZEN_INTERRUPTED_DRAFT_REPLACED"
        )

else:

    TEST_PATH.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    TEST_PATH.write_bytes(
        test_bytes
    )

    test_write_mode = (
        "NEW_TEST_WRITTEN"
    )


test_sha = (
    sha256_file(
        TEST_PATH
    )
)


assert (
    test_sha
    == new_test_sha
)


print(
    "[PASS] TEST_WRITE",
    flush=True,
)

print(
    "  mode    :",
    test_write_mode,
    flush=True,
)

print(
    "  old SHA :",
    old_test_sha,
    flush=True,
)

print(
    "  new SHA :",
    test_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. py_compile — streaming
# =============================================================================

compile_rc, compile_log = (
    run_streaming(
        [
            sys.executable,
            "-m",
            "py_compile",
            str(
                MODULE_PATH
            ),
            str(
                TEST_PATH
            ),
        ],
        cwd=ROOT,
        title="14.1 PY_COMPILE",
    )
)


assert (
    compile_rc
    == 0
), (
    "14.1 py_compile failed."
)


print(
    "[PASS] PY_COMPILE",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 17. Targeted pytest — streaming
# =============================================================================

pytest_rc, pytest_log = (
    run_streaming(
        [
            sys.executable,
            "-m",
            "pytest",
            "-vv",
            "-s",
            str(
                TEST_PATH
            ),
        ],
        cwd=ROOT,
        title="14.1 TARGETED PYTEST",
    )
)


assert (
    pytest_rc
    == 0
), (
    "14.1 targeted pytest failed."
)


print(
    "[PASS] TARGETED_PYTEST",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 18. Direct import + plan/executor smoke
# =============================================================================

from biohub_cell_tracking import (
    stage12_b12_portable
    as b12p
)


plan = (
    b12p._extract_plan(
        ROOT
    )
)


executor_source = (
    b12p._build_executor_source(
        plan
    )
)


compile(
    executor_source,
    filename="<notebook14-14.1-executor-smoke>",
    mode="exec",
)


assert (
    "post_v2_graph_provider"
    in executor_source
)


assert (
    "suppressed_peak_provider"
    in executor_source
)


assert (
    "build_current_v2("
    not in executor_source
)


assert (
    "read_pickle"
    not in executor_source
)


print(
    "[PASS] IMPORT_AND_EXECUTOR_SMOKE",
    flush=True,
)

print(
    "  preloop initializers:",
    len(
        plan[
            "preloop_initializers"
        ]
    ),
    flush=True,
)

print(
    "  retained loop statements:",
    len(
        plan[
            "retained_loop_sources"
        ]
    ),
    flush=True,
)

print(
    "  postloop statements:",
    len(
        plan[
            "postloop_sources"
        ]
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Explicit claim boundary
# =============================================================================

claim_boundary = {
    "implementation_created":
        True,

    "canonical_source_locked":
        True,

    "suppressed_peak_provider_previous_fidelity":
        "VERIFIED",

    "post_v2_graph_provider_fidelity":
        "NOT_YET_VERIFIED",

    "b12_structural_fidelity":
        "NOT_YET_VERIFIED",

    "gt_read":
        False,

    "evaluator_read":
        False,

    "performance_metric_generated":
        False,

    "historical_b12_oracle_read":
        False,

    "portable_current_best_ready":
        False,
}


# =============================================================================
# 20. Formal 14.1 artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_1",

    "title": (
        "B12_SOURCE_DERIVED_PORTABLE_"
        "MATERIALIZER_IMPLEMENTATION"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_IMPLEMENTATION_ONLY",

    "parent": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            parent_file_sha,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "canonical_b12_writer": {
        "path":
            str(
                CANONICAL_WRITER_REL
            ),

        "file_sha256":
            canonical_writer_sha,

        "dataset_loop_sha256":
            dataset_loop_sha,

        "dataset_loop_lines": [
            DATASET_LOOP_START,
            DATASET_LOOP_END,
        ],
    },

    "verified_suppressed_peak_provider": {
        "path":
            str(
                SUPPRESSED_PROVIDER_REL
            ),

        "file_sha256":
            suppressed_provider_sha,

        "previous_structural_fidelity":
            "VERIFIED",
    },

    "implementation": {
        "module_path":
            str(
                MODULE_REL
            ),

        "module_sha256":
            module_sha,

        "module_write_mode":
            module_write_mode,

        "prior_unfrozen_module_sha256":
            old_module_sha,

        "test_path":
            str(
                TEST_REL
            ),

        "test_sha256":
            test_sha,

        "test_write_mode":
            test_write_mode,

        "prior_unfrozen_test_sha256":
            old_test_sha,
    },

    "frozen_semantic_replacements": {
        "statement_0": {
            "canonical":
                canonical_first_four[0],

            "portable": (
                "g = post_v2_graph_provider("
                "name, context=post_v2_graph_context)"
            ),
        },

        "statement_3": {
            "canonical":
                canonical_first_four[3],

            "portable": (
                "pp = suppressed_peak_provider("
                "name, context=suppressed_peak_context)"
            ),
        },
    },

    "canonical_retention": {
        "statement_1":
            canonical_first_four[1],

        "statement_2":
            canonical_first_four[2],

        "statement_4_plus":
            "SOURCE_DERIVED_FROM_LOCKED_CANONICAL_LOOP",

        "postloop_geom_to_b12":
            "SOURCE_DERIVED_FROM_LOCKED_CANONICAL_WRITER",

        "preloop_initializers":
            preloop_initializers,
    },

    "validation": {
        "py_compile":
            "PASS",

        "targeted_pytest":
            "PASS",

        "pytest_log":
            pytest_log,

        "import_executor_smoke":
            "PASS",
    },

    "claim_boundary":
        claim_boundary,

    "known_required_next_repair": (
        "Remove the invented portable V2 actions.sort_values(...) "
        "that does not exist in the historical V2 writer before "
        "any post-V2 structural-fidelity claim."
    ),

    "explicit_non_claims": [
        "No post-V2 graph structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No R1P50 eligible portability claim.",
        "No full pre-G3H3 portability claim.",
        "No portable CURRENT_BEST readiness claim.",
        "No performance evidence.",
        "No fresh-holdout evidence.",
    ],

    "next_stage":
        "NOTEBOOK14_14_2_V2_INVENTED_SORT_PROVENANCE_REPAIR",
}


artifact_content_sha = (
    canonical_json_sha(
        payload
    )
)


artifact_obj = {
    "content_sha256":
        artifact_content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if ARTIFACT_PATH.exists():

    existing = (
        ARTIFACT_PATH.read_bytes()
    )

    assert (
        existing
        == artifact_bytes
    ), (
        "Notebook 14 / 14.1 artifact already exists "
        "but differs from this exact run. Refusing overwrite."
    )

    artifact_write_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )

else:

    ARTIFACT_PATH.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    ARTIFACT_PATH.write_bytes(
        artifact_bytes
    )

    artifact_write_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = (
    sha256_file(
        ARTIFACT_PATH
    )
)


# =============================================================================
# 21. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.1 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "module:",
    MODULE_REL,
    flush=True,
)

print(
    "module SHA:",
    module_sha,
    flush=True,
)

print(
    flush=True,
)

print(
    "test:",
    TEST_REL,
    flush=True,
)

print(
    "test SHA:",
    test_sha,
    flush=True,
)

print(
    flush=True,
)

print(
    "artifact:",
    ARTIFACT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_write_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    artifact_content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)

print(
    "FROZEN PORTABLE B12 SEAMS:",
    flush=True,
)

print(
    "  statement 0:",
    flush=True,
)

print(
    "    build_current_v2(name)",
    flush=True,
)

print(
    "      -> post_v2_graph_provider(name, context=...)",
    flush=True,
)

print(
    flush=True,
)

print(
    "  statement 3:",
    flush=True,
)

print(
    "    historical peaks lookup",
    flush=True,
)

print(
    "      -> suppressed_peak_provider(name, context=...)",
    flush=True,
)

print(
    flush=True,
)

print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  B12 materializer implementation = PASS",
    flush=True,
)

print(
    "  post-V2 graph structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)

print(
    "NEXT:",
    flush=True,
)

print(
    "  Notebook 14 / 14.2",
    flush=True,
)

print(
    "  V2_INVENTED_SORT_PROVENANCE_REPAIR",
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.2-R2 — V2 ACTION-BUILDER INVENTED SORT EXACT-CALL REPAIR
#
# SUCCESSOR TO:
#   14.2    — failed: over-narrow direct-function locator
#   14.2-R1 — failed: incorrect "reachable_from_apply_v2" call-graph gate
#
# R1 DISCOVERY
# ------------
# Exact invented call:
#
#   owner:
#       build_portable_mutual_cont_v2_actions
#
#   receiver:
#       actions
#
#   columns:
#       dataset
#       t
#       source_pred_node_id
#       target_pred_node_id
#
#   kind:
#       mergesort
#
#   frozen call SHA:
#       6b29d2a0f4f7935b34c2e1fdf0a76a12cf5290b2c20a82ff560058972c07f4c9
#
# Historical V2 writer:
#   exact corresponding action sort = ABSENT
#
# REPAIR
# ------
# Replace ONLY the exact sort call expression:
#
#     actions.sort_values(...)
#
# with:
#
#     actions
#
# This preserves enclosing assignment / reset_index / other source syntax.
#
# NO:
#   GT
#   evaluator
#   historical structural oracle
#   performance metric
#   structural-fidelity claim
#   Git operation
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import os
import subprocess
import sys
import textwrap

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_14_1_REL = Path(
    "reports/"
    "notebook14_14_1_"
    "b12_source_derived_portable_materializer_implementation_v1.json"
)

PARENT_14_1 = ROOT / PARENT_14_1_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


HISTORICAL_V2_REL = Path(
    "scripts/run_stage12_fold0_v2_action_lock.py"
)

HISTORICAL_V2 = ROOT / HISTORICAL_V2_REL


TEST_REL = Path(
    "tests/test_stage12_v2_invented_sort_repair.py"
)

TEST_PATH = ROOT / TEST_REL


ARTIFACT_REL = Path(
    "reports/"
    "notebook14_14_2_r2_"
    "v2_action_builder_invented_sort_exact_call_repair_v1.json"
)

ARTIFACT_PATH = ROOT / ARTIFACT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "78fd111c0cbcddf758c71b81dbc9dd2db51104a9725987d5d8544bea64b3ff3d"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "f50efa0c27a8177ec0653c7a63a882c613c7285b37f1c2e21f698db4de2ad8f0"
)


EXPECTED_PORTABLE_PRE_SHA = (
    "6167c296f6ecc987868d18f4d6e2a2839f9fa3881bd8d09aad5f2421e97c7bdc"
)

EXPECTED_HISTORICAL_V2_SHA = (
    "8e2684669baeaf8bc0d4b889f5c6e213a9127cf09e36c3d76ea2b3d77c2f4ef5"
)


TARGET_BUILDER = (
    "build_portable_mutual_cont_v2_actions"
)

EXPECTED_RECEIVER = (
    "actions"
)

EXPECTED_COLUMNS = [
    "dataset",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
]

EXPECTED_KIND = (
    "mergesort"
)

EXPECTED_CALL_SHA = (
    "6b29d2a0f4f7935b34c2e1fdf0a76a12cf5290b2c20a82ff560058972c07f4c9"
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(text: str) -> str:
    return sha256_bytes(
        text.encode("utf-8")
    )


def sha256_file(path: Path) -> str:
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(obj: Any) -> str:
    return sha256_bytes(
        json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def contains_exact_string(
    obj: Any,
    target: str,
) -> bool:

    if isinstance(obj, str):
        return obj == target

    if isinstance(obj, dict):
        return any(
            contains_exact_string(k, target)
            or contains_exact_string(v, target)
            for k, v in obj.items()
        )

    if isinstance(obj, (list, tuple)):
        return any(
            contains_exact_string(v, target)
            for v in obj
        )

    return False


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )

    assert source is not None, (
        "AST source extraction failed for "
        f"{type(node).__name__} "
        f"L{getattr(node, 'lineno', '?')}."
    )

    return source


def dotted_name(node: ast.AST) -> str | None:

    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):

        left = dotted_name(
            node.value
        )

        if left:
            return f"{left}.{node.attr}"

        return node.attr

    return None


def run_streaming(
    cmd: list[str],
    *,
    cwd: Path,
    title: str,
) -> tuple[int, str]:

    print(
        "=" * 112,
        flush=True,
    )
    print(
        title,
        flush=True,
    )
    print(
        "=" * 112,
        flush=True,
    )

    print(
        "COMMAND:",
        " ".join(
            str(x)
            for x in cmd
        ),
        flush=True,
    )
    print(
        flush=True,
    )

    env = os.environ.copy()
    env["PYTHONUNBUFFERED"] = "1"

    process = subprocess.Popen(
        cmd,
        cwd=str(cwd),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
        env=env,
    )

    assert process.stdout is not None

    captured = []

    for line in process.stdout:

        captured.append(
            line
        )

        print(
            line,
            end="",
            flush=True,
        )

    rc = process.wait()

    print(
        flush=True,
    )
    print(
        "[PROCESS EXIT CODE]",
        rc,
        flush=True,
    )
    print(
        flush=True,
    )

    return (
        rc,
        "".join(captured),
    )


# =============================================================================
# 3. AST helpers
# =============================================================================

def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name: node
        for node in tree.body
        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def literal_string_sequence(
    node: ast.AST,
) -> list[str] | None:

    if not isinstance(
        node,
        (
            ast.List,
            ast.Tuple,
        ),
    ):
        return None

    result = []

    for item in node.elts:

        if not (
            isinstance(
                item,
                ast.Constant,
            )
            and isinstance(
                item.value,
                str,
            )
        ):
            return None

        result.append(
            item.value
        )

    return result


def is_sort_values_call(
    node: ast.AST,
) -> bool:

    return (
        isinstance(
            node,
            ast.Call,
        )
        and isinstance(
            node.func,
            ast.Attribute,
        )
        and node.func.attr
        == "sort_values"
    )


def sort_columns(
    call: ast.Call,
) -> list[str] | None:

    # positional `by`
    if call.args:

        result = literal_string_sequence(
            call.args[0]
        )

        if result is not None:
            return result

    # keyword `by`
    for kw in call.keywords:

        if kw.arg != "by":
            continue

        result = literal_string_sequence(
            kw.value
        )

        if result is not None:
            return result

    return None


def sort_kind(
    call: ast.Call,
) -> str | None:

    for kw in call.keywords:

        if (
            kw.arg == "kind"
            and isinstance(
                kw.value,
                ast.Constant,
            )
            and isinstance(
                kw.value.value,
                str,
            )
        ):
            return kw.value.value

    return None


def sort_receiver(
    text: str,
    call: ast.Call,
) -> str:

    assert isinstance(
        call.func,
        ast.Attribute,
    )

    return node_source(
        text,
        call.func.value,
    )


def exact_target_sort(
    text: str,
    call: ast.Call,
) -> bool:

    source = node_source(
        text,
        call,
    )

    return (
        sort_receiver(
            text,
            call,
        )
        == EXPECTED_RECEIVER
        and sort_columns(
            call
        )
        == EXPECTED_COLUMNS
        and sort_kind(
            call
        )
        == EXPECTED_KIND
        and sha256_text(
            source
        )
        == EXPECTED_CALL_SHA
    )


# =============================================================================
# 4. AST byte-column -> Python character span
# =============================================================================

def byte_col_to_char_col(
    line: str,
    byte_col: int,
) -> int:

    return len(
        line.encode(
            "utf-8"
        )[
            :byte_col
        ].decode(
            "utf-8"
        )
    )


def absolute_char_span(
    text: str,
    node: ast.AST,
) -> tuple[int, int]:

    lines = text.splitlines(
        keepends=True
    )

    start_line = (
        node.lineno - 1
    )

    end_line = (
        node.end_lineno - 1
    )

    start_col = byte_col_to_char_col(
        lines[
            start_line
        ],
        node.col_offset,
    )

    end_col = byte_col_to_char_col(
        lines[
            end_line
        ],
        node.end_col_offset,
    )

    start = (
        sum(
            len(x)
            for x in lines[
                :start_line
            ]
        )
        + start_col
    )

    end = (
        sum(
            len(x)
            for x in lines[
                :end_line
            ]
        )
        + end_col
    )

    return (
        start,
        end,
    )


# =============================================================================
# 5. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT_14_1,
    PORTABLE,
    HISTORICAL_V2,
):

    assert required.exists(), (
        required
    )


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.2-R2 "
    "V2 ACTION-BUILDER INVENTED SORT EXACT-CALL REPAIR",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PORTABLE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "HISTORICAL:",
    HISTORICAL_V2_REL,
    flush=True,
)

print(
    "TARGET BUILDER:",
    TARGET_BUILDER,
    flush=True,
)

print(
    "ARTIFACT:",
    ARTIFACT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Idempotent successor reuse
# =============================================================================

if ARTIFACT_PATH.exists():

    existing_obj = json.loads(
        ARTIFACT_PATH.read_text(
            encoding="utf-8"
        )
    )

    if (
        existing_obj.get(
            "decision"
        )
        == "PASS"
    ):

        expected_current_sha = (
            existing_obj[
                "portable_source"
            ][
                "post_repair_file_sha256"
            ]
        )

        current_sha = sha256_file(
            PORTABLE
        )

        assert (
            current_sha
            == expected_current_sha
        )

        print(
            "[PASS] EXISTING_14_2_R2_ARTIFACT_REUSED",
            flush=True,
        )

        print(
            "  portable SHA:",
            current_sha,
            flush=True,
        )

        print(
            "  artifact SHA:",
            sha256_file(
                ARTIFACT_PATH
            ),
            flush=True,
        )

        print(
            flush=True,
        )

        print(
            "NOTEBOOK 14 / 14.2-R2 PASS",
            flush=True,
        )

        raise SystemExit(0)


# =============================================================================
# 7. Parent / source locks
# =============================================================================

parent_file_sha = sha256_file(
    PARENT_14_1
)

assert (
    parent_file_sha
    == EXPECTED_PARENT_FILE_SHA
)


parent_obj = json.loads(
    PARENT_14_1.read_text(
        encoding="utf-8"
    )
)


assert contains_exact_string(
    parent_obj,
    EXPECTED_PARENT_CONTENT_SHA,
)


portable_pre_sha = sha256_file(
    PORTABLE
)

assert (
    portable_pre_sha
    == EXPECTED_PORTABLE_PRE_SHA
), (
    "Portable source changed before 14.2-R2.\n"
    f"expected={EXPECTED_PORTABLE_PRE_SHA}\n"
    f"actual  ={portable_pre_sha}"
)


historical_sha = sha256_file(
    HISTORICAL_V2
)

assert (
    historical_sha
    == EXPECTED_HISTORICAL_V2_SHA
)


print(
    "[PASS] SOURCE_LOCKS",
    flush=True,
)

print(
    "  portable pre SHA:",
    portable_pre_sha,
    flush=True,
)

print(
    "  historical SHA  :",
    historical_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Resolve exact V2 action builder
# =============================================================================

portable_pre_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_pre_text,
    filename=str(PORTABLE),
)


functions = top_level_functions(
    portable_tree
)


assert (
    TARGET_BUILDER
    in functions
), (
    f"{TARGET_BUILDER} not found."
)


builder = functions[
    TARGET_BUILDER
]


builder_source = node_source(
    portable_pre_text,
    builder,
)


builder_pre_sha = sha256_text(
    builder_source
)


print(
    "[PASS] V2_ACTION_BUILDER_RESOLVED",
    flush=True,
)

print(
    "  function:",
    TARGET_BUILDER,
    flush=True,
)

print(
    "  function SHA:",
    builder_pre_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Exact target call inside builder
# =============================================================================

all_builder_sorts = [
    node
    for node in ast.walk(
        builder
    )
    if is_sort_values_call(
        node
    )
]


builder_sort_inventory = []


for call in all_builder_sorts:

    call_source = node_source(
        portable_pre_text,
        call,
    )

    builder_sort_inventory.append({
        "line":
            call.lineno,

        "end_line":
            call.end_lineno,

        "receiver":
            sort_receiver(
                portable_pre_text,
                call,
            ),

        "columns":
            sort_columns(
                call
            ),

        "kind":
            sort_kind(
                call
            ),

        "source":
            call_source,

        "source_sha256":
            sha256_text(
                call_source
            ),
    })


target_calls = [
    call
    for call in all_builder_sorts
    if exact_target_sort(
        portable_pre_text,
        call,
    )
]


assert (
    len(
        target_calls
    )
    == 1
), (
    "Exact frozen invented sort is not uniquely present "
    f"inside {TARGET_BUILDER}.\n"
    f"found={len(target_calls)}\n"
    f"inventory={builder_sort_inventory}"
)


target_call = target_calls[0]


target_call_source = node_source(
    portable_pre_text,
    target_call,
)


target_call_sha = sha256_text(
    target_call_source
)


assert (
    target_call_sha
    == EXPECTED_CALL_SHA
)


receiver_source = sort_receiver(
    portable_pre_text,
    target_call,
)


assert (
    receiver_source
    == EXPECTED_RECEIVER
)


print(
    "[PASS] EXACT_INVENTED_SORT_FROZEN",
    flush=True,
)

print(
    "  lines:",
    (
        target_call.lineno,
        target_call.end_lineno,
    ),
    flush=True,
)

print(
    "  receiver:",
    receiver_source,
    flush=True,
)

print(
    "  columns:",
    sort_columns(
        target_call
    ),
    flush=True,
)

print(
    "  kind:",
    sort_kind(
        target_call
    ),
    flush=True,
)

print(
    "  call SHA:",
    target_call_sha,
    flush=True,
)

print(
    flush=True,
)

print(
    "TARGET CALL:",
    flush=True,
)

print(
    target_call_source,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Historical exact absence
# =============================================================================

historical_text = HISTORICAL_V2.read_text(
    encoding="utf-8"
)


historical_tree = ast.parse(
    historical_text,
    filename=str(
        HISTORICAL_V2
    ),
)


historical_exact = []


historical_sort_inventory = []


for node in ast.walk(
    historical_tree
):

    if not is_sort_values_call(
        node
    ):
        continue

    source = node_source(
        historical_text,
        node,
    )

    record = {
        "line":
            node.lineno,

        "receiver":
            sort_receiver(
                historical_text,
                node,
            ),

        "columns":
            sort_columns(
                node
            ),

        "kind":
            sort_kind(
                node
            ),

        "source":
            source,

        "source_sha256":
            sha256_text(
                source
            ),
    }

    historical_sort_inventory.append(
        record
    )

    if (
        record["columns"]
        == EXPECTED_COLUMNS
        and record["kind"]
        == EXPECTED_KIND
    ):
        historical_exact.append(
            record
        )


assert (
    historical_exact
    == []
), (
    "Historical V2 unexpectedly contains an equivalent exact "
    "four-column mergesort."
)


print(
    "[PASS] HISTORICAL_EXACT_SORT_ABSENT",
    flush=True,
)

print(
    "  exact historical matches:",
    len(
        historical_exact
    ),
    flush=True,
)

print(
    "  all historical sorts:",
    len(
        historical_sort_inventory
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Exact CALL surgery in memory
#
# old:
#   actions.sort_values(...)
#
# new:
#   actions
# =============================================================================

start, end = absolute_char_span(
    portable_pre_text,
    target_call,
)


assert (
    portable_pre_text[
        start:end
    ]
    == target_call_source
), (
    "AST source span does not exactly match frozen call source."
)


prefix = portable_pre_text[
    :start
]

suffix = portable_pre_text[
    end:
]


portable_post_text = (
    prefix
    + receiver_source
    + suffix
)


assert (
    portable_post_text
    != portable_pre_text
)


# Must parse and compile before touching disk.
portable_post_tree = ast.parse(
    portable_post_text,
    filename=str(
        PORTABLE
    ),
)


compile(
    portable_post_text,
    filename=str(
        PORTABLE
    ),
    mode="exec",
)


print(
    "[PASS] IN_MEMORY_EXACT_CALL_REPAIR_COMPILES",
    flush=True,
)

print(
    "  old expression:",
    repr(
        target_call_source
    ),
    flush=True,
)

print(
    "  new expression:",
    repr(
        receiver_source
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Re-resolve builder post-repair
# =============================================================================

post_functions = top_level_functions(
    portable_post_tree
)


assert (
    TARGET_BUILDER
    in post_functions
)


builder_post = post_functions[
    TARGET_BUILDER
]


builder_post_source = node_source(
    portable_post_text,
    builder_post,
)


builder_post_sha = sha256_text(
    builder_post_source
)


post_exact = []


for call in ast.walk(
    builder_post
):

    if not is_sort_values_call(
        call
    ):
        continue

    if (
        sort_receiver(
            portable_post_text,
            call,
        )
        == EXPECTED_RECEIVER
        and sort_columns(
            call
        )
        == EXPECTED_COLUMNS
        and sort_kind(
            call
        )
        == EXPECTED_KIND
    ):
        post_exact.append(
            call
        )


assert (
    post_exact
    == []
)


print(
    "[PASS] INVENTED_SORT_ABSENT_POST_REPAIR",
    flush=True,
)

print(
    "  builder pre SHA :",
    builder_pre_sha,
    flush=True,
)

print(
    "  builder post SHA:",
    builder_post_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Exact whole-file reversibility invariant
# =============================================================================

replacement_end = (
    start
    + len(
        receiver_source
    )
)


reconstructed_pre = (
    portable_post_text[
        :start
    ]
    + target_call_source
    + portable_post_text[
        replacement_end:
    ]
)


assert (
    reconstructed_pre
    == portable_pre_text
), (
    "Repair is not exactly reversible by reinserting "
    "the one frozen sort call."
)


portable_post_sha_expected = sha256_text(
    portable_post_text
)


print(
    "[PASS] WHOLE_FILE_EXACT_CALL_DIFF_PROVEN",
    flush=True,
)

print(
    "  pre SHA :",
    portable_pre_sha,
    flush=True,
)

print(
    "  post SHA:",
    portable_post_sha_expected,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Write source
# =============================================================================

PORTABLE.write_text(
    portable_post_text,
    encoding="utf-8",
)


portable_post_sha = sha256_file(
    PORTABLE
)


assert (
    portable_post_sha
    == portable_post_sha_expected
)


print(
    "[PASS] PORTABLE_SOURCE_REPAIRED",
    flush=True,
)

print(
    "  SHA:",
    portable_post_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Dedicated permanent regression test
# =============================================================================

test_source = f'''
from __future__ import annotations

import ast
import hashlib

from pathlib import Path


ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

PORTABLE = (
    ROOT
    / "{PORTABLE_REL.as_posix()}"
)

HISTORICAL = (
    ROOT
    / "{HISTORICAL_V2_REL.as_posix()}"
)

TARGET_BUILDER = (
    "{TARGET_BUILDER}"
)

EXPECTED_PORTABLE_SHA = (
    "{portable_post_sha}"
)

EXPECTED_HISTORICAL_SHA = (
    "{EXPECTED_HISTORICAL_V2_SHA}"
)

EXPECTED_COLUMNS = (
    {EXPECTED_COLUMNS!r}
)

EXPECTED_KIND = (
    "{EXPECTED_KIND}"
)


def _sha(path: Path) -> str:

    return hashlib.sha256(
        path.read_bytes()
    ).hexdigest()


def _functions(tree: ast.Module):

    return {{
        node.name: node
        for node in tree.body
        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }}


def _literal_sequence(node):

    if not isinstance(
        node,
        (
            ast.List,
            ast.Tuple,
        ),
    ):
        return None

    values = []

    for item in node.elts:

        if not (
            isinstance(
                item,
                ast.Constant,
            )
            and isinstance(
                item.value,
                str,
            )
        ):
            return None

        values.append(
            item.value
        )

    return values


def _columns(call):

    if call.args:

        result = _literal_sequence(
            call.args[0]
        )

        if result is not None:
            return result

    for kw in call.keywords:

        if kw.arg == "by":

            result = _literal_sequence(
                kw.value
            )

            if result is not None:
                return result

    return None


def _kind(call):

    for kw in call.keywords:

        if (
            kw.arg == "kind"
            and isinstance(
                kw.value,
                ast.Constant,
            )
        ):
            return kw.value.value

    return None


def _exact_sorts(tree):

    result = []

    for call in ast.walk(
        tree
    ):

        if not (
            isinstance(
                call,
                ast.Call,
            )
            and isinstance(
                call.func,
                ast.Attribute,
            )
            and call.func.attr
            == "sort_values"
        ):
            continue

        if (
            _columns(call)
            == EXPECTED_COLUMNS
            and _kind(call)
            == EXPECTED_KIND
        ):
            result.append(
                call
            )

    return result


def test_portable_post_repair_sha():

    assert (
        _sha(PORTABLE)
        == EXPECTED_PORTABLE_SHA
    )


def test_historical_v2_sha():

    assert (
        _sha(HISTORICAL)
        == EXPECTED_HISTORICAL_SHA
    )


def test_v2_action_builder_still_exists():

    tree = ast.parse(
        PORTABLE.read_text(
            encoding="utf-8"
        )
    )

    assert (
        TARGET_BUILDER
        in _functions(tree)
    )


def test_invented_sort_absent_portable():

    tree = ast.parse(
        PORTABLE.read_text(
            encoding="utf-8"
        )
    )

    assert (
        _exact_sorts(tree)
        == []
    )


def test_equivalent_sort_absent_historical():

    tree = ast.parse(
        HISTORICAL.read_text(
            encoding="utf-8"
        )
    )

    assert (
        _exact_sorts(tree)
        == []
    )
'''


test_source = textwrap.dedent(
    test_source
).lstrip()


compile(
    test_source,
    filename=str(
        TEST_PATH
    ),
    mode="exec",
)


TEST_PATH.parent.mkdir(
    parents=True,
    exist_ok=True,
)


if TEST_PATH.exists():

    existing = TEST_PATH.read_text(
        encoding="utf-8"
    )

    if existing == test_source:

        test_write_mode = (
            "EXISTING_EXACT_TEST_REUSED"
        )

    else:

        TEST_PATH.write_text(
            test_source,
            encoding="utf-8",
        )

        test_write_mode = (
            "FAILED_PREDECESSOR_TEST_REPLACED"
        )

else:

    TEST_PATH.write_text(
        test_source,
        encoding="utf-8",
    )

    test_write_mode = (
        "NEW_TEST_WRITTEN"
    )


test_sha = sha256_file(
    TEST_PATH
)


print(
    "[PASS] REGRESSION_TEST_WRITTEN",
    flush=True,
)

print(
    "  mode:",
    test_write_mode,
    flush=True,
)

print(
    "  SHA:",
    test_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Discover existing related tests
# =============================================================================

related_tests = []


for path in (
    ROOT / "tests"
).glob(
    "test_*.py"
):

    if (
        path.resolve()
        == TEST_PATH.resolve()
    ):
        continue

    try:

        source = path.read_text(
            encoding="utf-8"
        )

    except Exception:

        continue

    if (
        "stage12_portable_current_best"
        in source
        or "mutual_cont_v2"
        in source.lower()
    ):

        related_tests.append(
            path
        )


related_tests = sorted(
    related_tests
)


print(
    "[INFO] RELATED TEST FILES",
    flush=True,
)

if not related_tests:

    print(
        "  <none>",
        flush=True,
    )

for path in related_tests:

    print(
        " ",
        path.relative_to(
            ROOT
        ),
        flush=True,
    )

print(
    flush=True,
)


# =============================================================================
# 17. py_compile — streaming
# =============================================================================

compile_rc, compile_log = run_streaming(
    [
        sys.executable,
        "-m",
        "py_compile",
        str(
            PORTABLE
        ),
        str(
            TEST_PATH
        ),
    ],
    cwd=ROOT,
    title="14.2-R2 PY_COMPILE",
)


assert (
    compile_rc
    == 0
), (
    "14.2-R2 py_compile failed."
)


print(
    "[PASS] PY_COMPILE",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 18. Targeted related pytest — streaming
# =============================================================================

pytest_targets = [
    str(
        TEST_PATH
    ),
    *[
        str(path)
        for path in related_tests
    ],
]


pytest_rc, pytest_log = run_streaming(
    [
        sys.executable,
        "-m",
        "pytest",
        "-vv",
        "-s",
        *pytest_targets,
    ],
    cwd=ROOT,
    title="14.2-R2 TARGETED PORTABLE-V2 PYTEST",
)


assert (
    pytest_rc
    == 0
), (
    "14.2-R2 targeted tests failed."
)


print(
    "[PASS] TARGETED_PYTEST",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Final disk relock
# =============================================================================

final_text = PORTABLE.read_text(
    encoding="utf-8"
)


final_sha = sha256_file(
    PORTABLE
)


assert (
    final_sha
    == portable_post_sha
)


final_tree = ast.parse(
    final_text,
    filename=str(
        PORTABLE
    ),
)


final_functions = top_level_functions(
    final_tree
)


assert (
    TARGET_BUILDER
    in final_functions
)


final_exact = []


for call in ast.walk(
    final_functions[
        TARGET_BUILDER
    ]
):

    if not is_sort_values_call(
        call
    ):
        continue

    if (
        sort_columns(
            call
        )
        == EXPECTED_COLUMNS
        and sort_kind(
            call
        )
        == EXPECTED_KIND
    ):
        final_exact.append(
            call
        )


assert (
    final_exact
    == []
)


print(
    "[PASS] FINAL_SOURCE_RELOCK",
    flush=True,
)

print(
    "  portable SHA:",
    final_sha,
    flush=True,
)

print(
    "  target exact sorts remaining: 0",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 20. Formal provenance successor
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_2_R2",

    "title": (
        "V2_ACTION_BUILDER_"
        "INVENTED_SORT_EXACT_CALL_REPAIR"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_PROVENANCE_REPAIR_ONLY",

    "parent_14_1": {
        "path":
            str(
                PARENT_14_1_REL
            ),

        "file_sha256":
            parent_file_sha,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "failed_predecessors": [
        {
            "stage":
                "NOTEBOOK14_14_2",

            "reason": (
                "Assumed exact sort lived directly inside "
                "_apply_portable_mutual_cont_v2."
            ),

            "source_mutated":
                False,
        },
        {
            "stage":
                "NOTEBOOK14_14_2_R1",

            "reason": (
                "Correctly localized exact sort in "
                "build_portable_mutual_cont_v2_actions but incorrectly "
                "required that builder be forward-call-reachable from "
                "_apply_portable_mutual_cont_v2."
            ),

            "source_mutated":
                False,
        },
    ],

    "historical_v2": {
        "path":
            str(
                HISTORICAL_V2_REL
            ),

        "file_sha256":
            historical_sha,

        "equivalent_exact_sort_count":
            0,

        "all_sort_values_diagnostic":
            historical_sort_inventory,
    },

    "portable_source": {
        "path":
            str(
                PORTABLE_REL
            ),

        "pre_repair_file_sha256":
            portable_pre_sha,

        "post_repair_file_sha256":
            portable_post_sha,

        "target_builder":
            TARGET_BUILDER,

        "builder_pre_sha256":
            builder_pre_sha,

        "builder_post_sha256":
            builder_post_sha,
    },

    "frozen_invented_call": {
        "line_start_pre_repair":
            target_call.lineno,

        "line_end_pre_repair":
            target_call.end_lineno,

        "receiver":
            receiver_source,

        "columns":
            EXPECTED_COLUMNS,

        "kind":
            EXPECTED_KIND,

        "source":
            target_call_source,

        "source_sha256":
            target_call_sha,
    },

    "source_surgery": {
        "replacement_rule": (
            "Replace exact actions.sort_values(...) call expression "
            "with receiver expression `actions` only."
        ),

        "replacement_source":
            receiver_source,

        "replacement_sha256":
            sha256_text(
                receiver_source
            ),

        "start_character_offset":
            start,

        "end_character_offset_pre_repair":
            end,

        "whole_file_exact_reconstruction":
            True,
    },

    "regression": {
        "test_path":
            str(
                TEST_REL
            ),

        "test_sha256":
            test_sha,

        "test_write_mode":
            test_write_mode,

        "related_tests": [
            str(
                path.relative_to(
                    ROOT
                )
            )
            for path in related_tests
        ],

        "py_compile":
            "PASS",

        "pytest":
            "PASS",

        "pytest_log":
            pytest_log,
    },

    "claim_boundary": {
        "v2_invented_sort_provenance_mismatch":
            "REPAIRED",

        "post_v2_graph_structural_fidelity":
            "NOT_YET_VERIFIED",

        "b12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "full_pre_g3h3_structural_fidelity":
            "NOT_YET_VERIFIED",

        "portable_current_best_ready":
            False,

        "gt_read":
            False,

        "evaluator_read":
            False,

        "structural_oracle_read":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No post-V2 graph structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance claim.",
        "No fresh-holdout claim.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        "NOTEBOOK14_14_3_POST_V2_GRAPH_PROVIDER_IMPLEMENTATION",
}


artifact_content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        artifact_content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")


if ARTIFACT_PATH.exists():

    assert (
        ARTIFACT_PATH.read_bytes()
        == artifact_bytes
    ), (
        "14.2-R2 artifact exists but differs from this "
        "exact provenance repair."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )

else:

    ARTIFACT_PATH.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    ARTIFACT_PATH.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    ARTIFACT_PATH
)


# =============================================================================
# 21. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.2-R2 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "FROZEN TARGET:",
    flush=True,
)

print(
    "  function:",
    TARGET_BUILDER,
    flush=True,
)

print(
    "  receiver:",
    EXPECTED_RECEIVER,
    flush=True,
)

print(
    "  columns:",
    EXPECTED_COLUMNS,
    flush=True,
)

print(
    "  kind:",
    EXPECTED_KIND,
    flush=True,
)

print(
    "  original call SHA:",
    EXPECTED_CALL_SHA,
    flush=True,
)

print(
    flush=True,
)


print(
    "portable pre SHA :",
    portable_pre_sha,
    flush=True,
)

print(
    "portable post SHA:",
    portable_post_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    ARTIFACT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    artifact_content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  V2 invented-sort provenance mismatch = REPAIRED",
    flush=True,
)

print(
    "  post-V2 graph structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    "  Notebook 14 / 14.3",
    flush=True,
)

print(
    "  POST_V2_GRAPH_PROVIDER_IMPLEMENTATION",
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3A-R2 — POST-V2 GRAPH PROVIDER CONTRACT EXTRACTION
#
# SUCCESSOR TO:
#   14.3A    — failed at preflight because an old Notebook 13 architecture
#              artifact was assumed to exist at one fixed path.
#
#   14.3A-R1 — successfully extracted all current source contracts and located
#              the historical architecture artifact, but failed only because
#              the semantic gate dictionary used:
#
#                  "source_mutation": False
#
#              and the generic gate evaluator interpreted False as FAIL.
#
# THIS R2:
#   - keeps the same extraction logic;
#   - keeps the same current portable source authority;
#   - keeps the same historical architecture discovery;
#   - changes the PASS/FAIL gate to:
#
#         "no_source_mutation": True
#
#   - keeps claim_boundary["source_mutation"] = False as a descriptive fact.
#
# HARD CONTRACT AUTHORITY:
#   Notebook 14 / 14.2-R2 PASS
#   + repaired portable module exact SHA
#   + actual AST/source contracts
#
# NO:
#   - source mutation
#   - GT
#   - evaluator
#   - historical structural oracle
#   - performance metric
#   - structural-fidelity claim
#   - Git operation
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_2_r2_"
    "v2_action_builder_invented_sort_exact_call_repair_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


PRIMARY_TEST_REL = Path(
    "tests/test_stage12_portable_current_best.py"
)

PRIMARY_TEST = ROOT / PRIMARY_TEST_REL


V2_REPAIR_TEST_REL = Path(
    "tests/test_stage12_v2_invented_sort_repair.py"
)

V2_REPAIR_TEST = ROOT / V2_REPAIR_TEST_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3a_r2_"
    "post_v2_graph_provider_contract_extraction_v1.json"
)

OUT = ROOT / OUT_REL


REPORTS = ROOT / "reports"


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "e63070f664935f79d9df044f52376ff9afd6e2562ecf396b4c438a697877428d"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "988f7af83830cececddfdf81a0bb0f31c39a64948a7016b8ea41a195882cc12e"
)


EXPECTED_PORTABLE_SHA = (
    "a1fe0c9e5cd5e20a2ae5921bd4b4c55d078022c3348d5945bffa501ac3ff36ed"
)


# Historical Notebook 13 architecture authority.
#
# The exact file was discovered by 14.3A-R1 at:
#
# reports/notebook13_13_3c_r2_
# corrected_authority_model_and_architecture_refreeze_v1.json
#
# We still discover by frozen SHA rather than hardcoding only the path.

EXPECTED_ARCH_FILE_SHA = (
    "e7d307f988b97717a4250534179fc07f90408b29dedccbcc4fc9b96567c7f4c8"
)

EXPECTED_ARCH_CONTENT_SHA = (
    "49f533d19e2354735c9faef38179625c058732ba4838e1a7544aefd1733fc3f4"
)


TARGET_FUNCTIONS = [
    "_build_portable_frozen_v9",
    "_apply_portable_morph_div_v2",
    "_reconstruct_portable_mutual_cont_v1",
    "build_portable_mutual_cont_v2_actions",
    "_apply_portable_mutual_cont_v2",
]


FROZEN_STAGE_SEQUENCE = [
    {
        "stage":
            "FROZEN_V9",

        "function":
            "_build_portable_frozen_v9",
    },
    {
        "stage":
            "MORPH_DIV_V2",

        "function":
            "_apply_portable_morph_div_v2",
    },
    {
        "stage":
            "MUTUAL_CONT_V1",

        "function":
            "_reconstruct_portable_mutual_cont_v1",
    },
    {
        "stage":
            "MUTUAL_CONT_V2_ACTION_BUILD",

        "function":
            "build_portable_mutual_cont_v2_actions",
    },
    {
        "stage":
            "MUTUAL_CONT_V2_APPLY",

        "function":
            "_apply_portable_mutual_cont_v2",
    },
]


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    data: bytes,
) -> str:

    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(
    text: str,
) -> str:

    return sha256_bytes(
        text.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    obj: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def contains_exact_string(
    obj: Any,
    target: str,
) -> bool:

    if isinstance(
        obj,
        str,
    ):

        return (
            obj
            == target
        )


    if isinstance(
        obj,
        dict,
    ):

        return any(
            contains_exact_string(
                key,
                target,
            )
            or contains_exact_string(
                value,
                target,
            )
            for key, value
            in obj.items()
        )


    if isinstance(
        obj,
        (
            list,
            tuple,
        ),
    ):

        return any(
            contains_exact_string(
                value,
                target,
            )
            for value
            in obj
        )


    return False


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = (
        ast.get_source_segment(
            text,
            node,
        )
    )


    assert (
        source
        is not None
    ), (
        "AST source extraction failed: "
        f"{type(node).__name__} "
        f"L{getattr(node, 'lineno', '?')}"
    )


    return source


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return (
            node.id
        )


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = (
            dotted_name(
                node.value
            )
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return (
            node.attr
        )


    return None


def target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    def visit(
        item: ast.AST,
    ):

        if isinstance(
            item,
            ast.Name,
        ):

            result.add(
                item.id
            )


        elif isinstance(
            item,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for child in (
                item.elts
            ):

                visit(
                    child
                )


        elif isinstance(
            item,
            ast.Starred,
        ):

            visit(
                item.value
            )


    visit(
        node
    )


    return result


# =============================================================================
# 3. Function helpers
# =============================================================================

def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def function_scope_nodes(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[ast.AST]:

    collected = []


    class Visitor(
        ast.NodeVisitor
    ):

        def visit_FunctionDef(
            self,
            node,
        ):

            if node is fn:

                for statement in (
                    node.body
                ):

                    self.visit(
                        statement
                    )


        def visit_AsyncFunctionDef(
            self,
            node,
        ):

            if node is fn:

                for statement in (
                    node.body
                ):

                    self.visit(
                        statement
                    )


        def visit_Lambda(
            self,
            node,
        ):

            return


        def generic_visit(
            self,
            node,
        ):

            collected.append(
                node
            )

            super().generic_visit(
                node
            )


    Visitor().visit(
        fn
    )


    return collected


def function_returns(
    text: str,
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for node in (
        function_scope_nodes(
            fn
        )
    ):

        if not isinstance(
            node,
            ast.Return,
        ):

            continue


        result.append({
            "line":
                node.lineno,

            "expression":
                (
                    ast.unparse(
                        node.value
                    )

                    if node.value
                    is not None

                    else "None"
                ),

            "source":
                node_source(
                    text,
                    node,
                ),
        })


    return result


def function_calls(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[str]:

    return sorted({
        dotted_name(
            node.func
        )
        or "<dynamic-call>"

        for node in (
            function_scope_nodes(
                fn
            )
        )

        if isinstance(
            node,
            ast.Call,
        )
    })


def function_parameters(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> dict[str, Any]:

    positional = (
        list(
            fn.args.posonlyargs
        )
        + list(
            fn.args.args
        )
    )


    defaults = (
        [None]
        * (
            len(
                positional
            )
            -
            len(
                fn.args.defaults
            )
        )
        +
        list(
            fn.args.defaults
        )
    )


    positional_records = []


    for arg, default in zip(
        positional,
        defaults,
    ):

        positional_records.append({
            "name":
                arg.arg,

            "annotation":
                (
                    ast.unparse(
                        arg.annotation
                    )

                    if arg.annotation
                    is not None

                    else None
                ),

            "default":
                (
                    ast.unparse(
                        default
                    )

                    if default
                    is not None

                    else None
                ),
        })


    kwonly_records = []


    for arg, default in zip(
        fn.args.kwonlyargs,
        fn.args.kw_defaults,
    ):

        kwonly_records.append({
            "name":
                arg.arg,

            "annotation":
                (
                    ast.unparse(
                        arg.annotation
                    )

                    if arg.annotation
                    is not None

                    else None
                ),

            "default":
                (
                    ast.unparse(
                        default
                    )

                    if default
                    is not None

                    else None
                ),
        })


    return {
        "positional":
            positional_records,

        "vararg":
            (
                fn.args.vararg.arg

                if fn.args.vararg
                is not None

                else None
            ),

        "kwonly":
            kwonly_records,

        "kwarg":
            (
                fn.args.kwarg.arg

                if fn.args.kwarg
                is not None

                else None
            ),

        "return_annotation":
            (
                ast.unparse(
                    fn.returns
                )

                if fn.returns
                is not None

                else None
            ),
    }


def direct_statement_inventory(
    text: str,
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for index, statement in enumerate(
        fn.body
    ):

        assigned_names = set()


        if isinstance(
            statement,
            ast.Assign,
        ):

            for target in (
                statement.targets
            ):

                assigned_names |= (
                    target_names(
                        target
                    )
                )


        elif isinstance(
            statement,
            ast.AnnAssign,
        ):

            assigned_names |= (
                target_names(
                    statement.target
                )
            )


        calls = sorted({
            dotted_name(
                call.func
            )
            or "<dynamic-call>"

            for call in (
                ast.walk(
                    statement
                )
            )

            if isinstance(
                call,
                ast.Call,
            )
        })


        source = (
            node_source(
                text,
                statement,
            )
        )


        result.append({
            "index":
                index,

            "line_start":
                statement.lineno,

            "line_end":
                statement.end_lineno,

            "statement_type":
                type(
                    statement
                ).__name__,

            "assigned_names":
                sorted(
                    assigned_names
                ),

            "calls":
                calls,

            "source_sha256":
                sha256_text(
                    source
                ),

            "source":
                source,
        })


    return result


# =============================================================================
# 4. Call-site helpers
# =============================================================================

def owner_map(
    tree: ast.Module,
) -> dict[
    ast.AST,
    str,
]:

    result = {}


    for top in (
        tree.body
    ):

        if not isinstance(
            top,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        ):

            continue


        for node in (
            ast.walk(
                top
            )
        ):

            if (
                node
                not in result
            ):

                result[
                    node
                ] = (
                    top.name
                )


    return result


def call_matches_target(
    call: ast.Call,
    target: str,
) -> bool:

    name = (
        dotted_name(
            call.func
        )
    )


    return (
        name
        == target

        or (
            name
            is not None

            and name.endswith(
                "." + target
            )
        )
    )


def extract_call_examples(
    *,
    path: Path,
    text: str,
    tree: ast.Module,
    target: str,
) -> list[
    dict[str, Any]
]:

    owners = (
        owner_map(
            tree
        )
    )


    result = []


    for call in (
        ast.walk(
            tree
        )
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        if not call_matches_target(
            call,
            target,
        ):

            continue


        source = (
            node_source(
                text,
                call,
            )
        )


        result.append({
            "path":
                str(
                    path.relative_to(
                        ROOT
                    )
                ),

            "line":
                call.lineno,

            "owner":
                owners.get(
                    call,
                    "<module>",
                ),

            "call_name":
                dotted_name(
                    call.func
                ),

            "args": [
                ast.unparse(
                    arg
                )

                for arg in (
                    call.args
                )
            ],

            "keywords": {
                (
                    keyword.arg
                    if keyword.arg
                    is not None
                    else "**"
                ):
                    ast.unparse(
                        keyword.value
                    )

                for keyword in (
                    call.keywords
                )
            },

            "source":
                source,

            "source_sha256":
                sha256_text(
                    source
                ),
        })


    return result


# =============================================================================
# 5. Historical architecture artifact discovery
# =============================================================================

def discover_architecture_artifact() -> dict[str, Any]:

    candidates = []


    if not REPORTS.exists():

        return {
            "status":
                "NOT_PRESENT_ON_DISK",

            "candidates_examined":
                0,

            "matches":
                [],
        }


    json_paths = sorted(
        REPORTS.rglob(
            "*.json"
        )
    )


    for path in (
        json_paths
    ):

        try:

            file_sha = (
                sha256_file(
                    path
                )
            )


            # ---------------------------------------------------------
            # Exact historical artifact SHA.
            # ---------------------------------------------------------

            if (
                file_sha
                == EXPECTED_ARCH_FILE_SHA
            ):

                obj = (
                    json.loads(
                        path.read_text(
                            encoding="utf-8"
                        )
                    )
                )


                candidates.append({
                    "path":
                        str(
                            path.relative_to(
                                ROOT
                            )
                        ),

                    "match_mode":
                        "EXACT_FILE_SHA",

                    "file_sha256":
                        file_sha,

                    "contains_expected_content_sha":
                        contains_exact_string(
                            obj,
                            EXPECTED_ARCH_CONTENT_SHA,
                        ),
                })


                continue


            # ---------------------------------------------------------
            # Content-reference search only in likely relevant reports.
            # ---------------------------------------------------------

            lower_name = (
                path.name.lower()
            )


            if (
                "architecture"
                not in lower_name

                and "authority"
                not in lower_name

                and "refreeze"
                not in lower_name
            ):

                continue


            obj = (
                json.loads(
                    path.read_text(
                        encoding="utf-8"
                    )
                )
            )


            if contains_exact_string(
                obj,
                EXPECTED_ARCH_CONTENT_SHA,
            ):

                candidates.append({
                    "path":
                        str(
                            path.relative_to(
                                ROOT
                            )
                        ),

                    "match_mode":
                        "EXPECTED_CONTENT_SHA_EMBEDDED",

                    "file_sha256":
                        file_sha,

                    "contains_expected_content_sha":
                        True,
                })


        except Exception:

            continue


    exact_matches = [
        item

        for item in (
            candidates
        )

        if (
            item[
                "file_sha256"
            ]
            == EXPECTED_ARCH_FILE_SHA
        )
    ]


    if exact_matches:

        assert (
            len(
                exact_matches
            )
            == 1
        ), (
            "Multiple files unexpectedly have "
            "the exact frozen architecture SHA."
        )


        return {
            "status":
                "FOUND_EXACT",

            "candidates_examined":
                len(
                    json_paths
                ),

            "matches":
                candidates,

            "selected":
                exact_matches[0],
        }


    content_matches = [
        item

        for item in (
            candidates
        )

        if item[
            "contains_expected_content_sha"
        ]
    ]


    if content_matches:

        return {
            "status":
                "FOUND_CONTENT_REFERENCE_ONLY",

            "candidates_examined":
                len(
                    json_paths
                ),

            "matches":
                candidates,

            "selected":
                content_matches[0],
        }


    return {
        "status":
            "NOT_PRESENT_ON_DISK",

        "candidates_examined":
            len(
                json_paths
            ),

        "matches":
            [],
    }


# =============================================================================
# 6. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    PRIMARY_TEST,
    V2_REPAIR_TEST,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3A-R2 "
    "POST-V2 GRAPH PROVIDER CONTRACT EXTRACTION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PORTABLE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Parent 14.2-R2 lock
# =============================================================================

parent_file_sha = (
    sha256_file(
        PARENT
    )
)


assert (
    parent_file_sha
    == EXPECTED_PARENT_FILE_SHA
), (
    "14.2-R2 file SHA mismatch."
)


parent_obj = (
    json.loads(
        PARENT.read_text(
            encoding="utf-8"
        )
    )
)


assert (
    parent_obj.get(
        "decision"
    )
    == "PASS"
)


assert contains_exact_string(
    parent_obj,
    EXPECTED_PARENT_CONTENT_SHA,
)


assert (
    parent_obj[
        "portable_source"
    ][
        "post_repair_file_sha256"
    ]
    == EXPECTED_PORTABLE_SHA
)


print(
    "[PASS] PARENT_14_2_R2_LOCK",
    flush=True,
)

print(
    "  file SHA    :",
    parent_file_sha,
    flush=True,
)

print(
    "  content SHA :",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Current portable source lock
# =============================================================================

portable_sha = (
    sha256_file(
        PORTABLE
    )
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
), (
    "Portable source changed after 14.2-R2.\n"
    f"expected={EXPECTED_PORTABLE_SHA}\n"
    f"actual  ={portable_sha}"
)


portable_text = (
    PORTABLE.read_text(
        encoding="utf-8"
    )
)


portable_tree = (
    ast.parse(
        portable_text,
        filename=str(
            PORTABLE
        ),
    )
)


functions = (
    top_level_functions(
        portable_tree
    )
)


print(
    "[PASS] PORTABLE_SOURCE_LOCK",
    flush=True,
)

print(
    "  SHA:",
    portable_sha,
    flush=True,
)

print(
    "  top-level function count:",
    len(
        functions
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Historical architecture discovery
# =============================================================================

architecture_discovery = (
    discover_architecture_artifact()
)


print(
    "=" * 112,
    flush=True,
)

print(
    "HISTORICAL ARCHITECTURE ARTIFACT DISCOVERY",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "status:",
    architecture_discovery[
        "status"
    ],
    flush=True,
)

print(
    "JSON reports examined:",
    architecture_discovery[
        "candidates_examined"
    ],
    flush=True,
)


if (
    architecture_discovery[
        "matches"
    ]
):

    print(
        "matches:",
        flush=True,
    )


    for item in (
        architecture_discovery[
            "matches"
        ]
    ):

        print(
            " ",
            item,
            flush=True,
        )


else:

    print(
        "matches: <none>",
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 10. Lock exact historical architecture authority if available
# =============================================================================

if (
    architecture_discovery[
        "status"
    ]
    == "FOUND_EXACT"
):

    selected_arch_path = (
        ROOT
        / architecture_discovery[
            "selected"
        ][
            "path"
        ]
    )


    selected_arch_file_sha = (
        sha256_file(
            selected_arch_path
        )
    )


    assert (
        selected_arch_file_sha
        == EXPECTED_ARCH_FILE_SHA
    )


    selected_arch_obj = (
        json.loads(
            selected_arch_path.read_text(
                encoding="utf-8"
            )
        )
    )


    assert contains_exact_string(
        selected_arch_obj,
        EXPECTED_ARCH_CONTENT_SHA,
    )


    architecture_authority_status = (
        "EXACT_HISTORICAL_ARTIFACT_LOCKED"
    )


    print(
        "[PASS] HISTORICAL_ARCHITECTURE_AUTHORITY_LOCK",
        flush=True,
    )

    print(
        "  path:",
        selected_arch_path.relative_to(
            ROOT
        ),
        flush=True,
    )

    print(
        "  file SHA:",
        selected_arch_file_sha,
        flush=True,
    )

    print(
        "  content SHA:",
        EXPECTED_ARCH_CONTENT_SHA,
        flush=True,
    )


else:

    selected_arch_path = (
        None
    )

    selected_arch_file_sha = (
        None
    )

    selected_arch_obj = (
        None
    )

    architecture_authority_status = (
        architecture_discovery[
            "status"
        ]
    )


    print(
        "[INFO] HISTORICAL_ARCHITECTURE_AUTHORITY",
        flush=True,
    )

    print(
        "  status:",
        architecture_authority_status,
        flush=True,
    )

    print(
        "  no historical artifact fabricated/recreated",
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 11. Resolve required current functions
# =============================================================================

missing_functions = [
    name

    for name in (
        TARGET_FUNCTIONS
    )

    if name not in (
        functions
    )
]


assert (
    missing_functions
    == []
), (
    "Current repaired portable module is missing "
    f"required functions: {missing_functions}"
)


print(
    "[PASS] REQUIRED_STAGE_FUNCTIONS_PRESENT",
    flush=True,
)


for stage in (
    FROZEN_STAGE_SEQUENCE
):

    print(
        " ",
        stage[
            "stage"
        ],
        "->",
        stage[
            "function"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 12. Freeze exact current function contracts
# =============================================================================

contracts = {}


for function_name in (
    TARGET_FUNCTIONS
):

    fn = (
        functions[
            function_name
        ]
    )


    source = (
        node_source(
            portable_text,
            fn,
        )
    )


    contract = {
        "function":
            function_name,

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "parameters":
            function_parameters(
                fn
            ),

        "returns":
            function_returns(
                portable_text,
                fn,
            ),

        "calls":
            function_calls(
                fn
            ),

        "direct_statements":
            direct_statement_inventory(
                portable_text,
                fn,
            ),

        "source":
            source,
    }


    contracts[
        function_name
    ] = (
        contract
    )


    print(
        "=" * 112,
        flush=True,
    )

    print(
        f"CONTRACT :: {function_name}",
        flush=True,
    )

    print(
        "=" * 112,
        flush=True,
    )

    print(
        "lines:",
        (
            fn.lineno,
            fn.end_lineno,
        ),
        flush=True,
    )

    print(
        "source SHA:",
        contract[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "parameters:",
        json.dumps(
            contract[
                "parameters"
            ],
            indent=2,
        ),
        flush=True,
    )

    print(
        "returns:",
        contract[
            "returns"
        ],
        flush=True,
    )

    print(
        "calls:",
        contract[
            "calls"
        ],
        flush=True,
    )

    print(
        flush=True,
    )

    print(
        "SOURCE:",
        flush=True,
    )

    print(
        source,
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 13. Parse current tests + portable module for real call examples
# =============================================================================

source_files = []


for path in (
    PORTABLE,
    PRIMARY_TEST,
    V2_REPAIR_TEST,
):

    text = (
        path.read_text(
            encoding="utf-8"
        )
    )


    tree = (
        ast.parse(
            text,
            filename=str(
                path
            ),
        )
    )


    source_files.append(
        (
            path,
            text,
            tree,
        )
    )


call_examples = {}


for function_name in (
    TARGET_FUNCTIONS
):

    examples = []


    for path, text, tree in (
        source_files
    ):

        examples.extend(
            extract_call_examples(
                path=path,
                text=text,
                tree=tree,
                target=function_name,
            )
        )


    call_examples[
        function_name
    ] = (
        examples
    )


    print(
        "=" * 112,
        flush=True,
    )

    print(
        f"CALL EXAMPLES :: {function_name}",
        flush=True,
    )

    print(
        "=" * 112,
        flush=True,
    )


    if not examples:

        print(
            "  <none found>",
            flush=True,
        )


    for example in (
        examples
    ):

        print(
            f"{example['path']}:"
            f"{example['line']} "
            f"owner={example['owner']}",
            flush=True,
        )

        print(
            "  args:",
            example[
                "args"
            ],
            flush=True,
        )

        print(
            "  keywords:",
            example[
                "keywords"
            ],
            flush=True,
        )

        print(
            "  source:",
            flush=True,
        )

        print(
            "    "
            + example[
                "source"
            ].replace(
                "\n",
                "\n    ",
            ),
            flush=True,
        )

        print(
            flush=True,
        )


# =============================================================================
# 14. Find all related portable orchestration functions
# =============================================================================

RELATED_TOKENS = (
    "frozen_v9",
    "morph",
    "mutual_cont",
    "current_best",
    "r1p50",
    "portable",
)


related_function_inventory = []


for name, fn in (
    functions.items()
):

    lower = (
        name.lower()
    )


    if not any(
        token in lower

        for token in (
            RELATED_TOKENS
        )
    ):

        continue


    source = (
        node_source(
            portable_text,
            fn,
        )
    )


    related_function_inventory.append({
        "name":
            name,

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "parameters":
            function_parameters(
                fn
            ),

        "returns":
            function_returns(
                portable_text,
                fn,
            ),

        "calls":
            function_calls(
                fn
            ),
    })


print(
    "=" * 112,
    flush=True,
)

print(
    "RELATED PORTABLE FUNCTION INVENTORY",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for record in sorted(
    related_function_inventory,
    key=lambda value: (
        value[
            "line_start"
        ]
    ),
):

    print(
        f"L{record['line_start']}"
        f"-L{record['line_end']} "
        f"{record['name']}",
        flush=True,
    )

    print(
        "  source SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "  returns:",
        [
            item[
                "expression"
            ]

            for item in (
                record[
                    "returns"
                ]
            )
        ],
        flush=True,
    )

    print(
        "  calls:",
        record[
            "calls"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 15. Identify possible orchestration/public entrypoints
# =============================================================================

orchestration_candidates = []


for name, fn in (
    functions.items()
):

    source = (
        node_source(
            portable_text,
            fn,
        )
    )


    source_lower = (
        source.lower()
    )


    matched = []


    for target in (
        TARGET_FUNCTIONS
    ):

        if (
            target.lower()
            in source_lower
        ):

            matched.append(
                target
            )


    for keyword in (
        "current_best",
        "notimplemented",
        "not implemented",
        "fail closed",
        "fails closed",
        "partial",
        "r1p50",
    ):

        if (
            keyword
            in source_lower
        ):

            matched.append(
                keyword
            )


    if not matched:

        continue


    orchestration_candidates.append({
        "name":
            name,

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "matched":
            sorted(
                set(
                    matched
                )
            ),

        "parameters":
            function_parameters(
                fn
            ),

        "returns":
            function_returns(
                portable_text,
                fn,
            ),

        "calls":
            function_calls(
                fn
            ),

        "source":
            source,
    })


print(
    "=" * 112,
    flush=True,
)

print(
    "ORCHESTRATION CANDIDATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


if not orchestration_candidates:

    print(
        "  <none found>",
        flush=True,
    )


for record in (
    orchestration_candidates
):

    print(
        f"{record['name']} "
        f"L{record['line_start']}"
        f"-L{record['line_end']}",
        flush=True,
    )

    print(
        "  matched:",
        record[
            "matched"
        ],
        flush=True,
    )

    print(
        "  returns:",
        record[
            "returns"
        ],
        flush=True,
    )

    print(
        "  calls:",
        record[
            "calls"
        ],
        flush=True,
    )

    print(
        "  source:",
        flush=True,
    )

    print(
        "    "
        + record[
            "source"
        ].replace(
            "\n",
            "\n    ",
        ),
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 16. Current direct-call matrix
# =============================================================================

direct_call_matrix = {}


for caller in (
    TARGET_FUNCTIONS
):

    calls = set(
        contracts[
            caller
        ][
            "calls"
        ]
    )


    direct_call_matrix[
        caller
    ] = {
        callee:
            (
                callee
                in calls
            )

        for callee in (
            TARGET_FUNCTIONS
        )

        if (
            callee
            != caller
        )
    }


print(
    "=" * 112,
    flush=True,
)

print(
    "TARGET DIRECT-CALL MATRIX",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for caller, row in (
    direct_call_matrix.items()
):

    print(
        caller,
        flush=True,
    )


    for callee, present in (
        row.items()
    ):

        print(
            f"  -> {callee:<45} "
            f"{present}",
            flush=True,
        )


print(
    flush=True,
)


# =============================================================================
# 17. Additional implemented-runtime authority discovery
#
# R1 exposed an important distinction:
#
# underscore wrappers may be historical/placeholder boundaries,
# while later public implementations may already exist.
#
# Freeze these identities explicitly for 14.3B.
# =============================================================================

IMPLEMENTED_RUNTIME_FUNCTIONS = [
    "apply_portable_morph_div_v2",
    "apply_portable_mutual_cont_v2",
    "apply_portable_det_bridge_r1p50_core",
]


implemented_runtime_contracts = {}


for function_name in (
    IMPLEMENTED_RUNTIME_FUNCTIONS
):

    if (
        function_name
        not in functions
    ):

        implemented_runtime_contracts[
            function_name
        ] = {
            "status":
                "NOT_PRESENT",
        }

        continue


    fn = (
        functions[
            function_name
        ]
    )


    source = (
        node_source(
            portable_text,
            fn,
        )
    )


    implemented_runtime_contracts[
        function_name
    ] = {
        "status":
            "PRESENT",

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "parameters":
            function_parameters(
                fn
            ),

        "returns":
            function_returns(
                portable_text,
                fn,
            ),

        "calls":
            function_calls(
                fn
            ),

        "source":
            source,
    }


print(
    "=" * 112,
    flush=True,
)

print(
    "IMPLEMENTED RUNTIME FUNCTION CONTRACTS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for name, contract in (
    implemented_runtime_contracts.items()
):

    print(
        name,
        flush=True,
    )

    print(
        "  status:",
        contract[
            "status"
        ],
        flush=True,
    )


    if (
        contract[
            "status"
        ]
        == "PRESENT"
    ):

        print(
            "  SHA:",
            contract[
                "source_sha256"
            ],
            flush=True,
        )

        print(
            "  returns:",
            [
                item[
                    "expression"
                ]

                for item in (
                    contract[
                        "returns"
                    ]
                )
            ],
            flush=True,
        )

        print(
            "  calls:",
            contract[
                "calls"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 18. Placeholder-vs-implemented role classification
# =============================================================================

def raises_incomplete_error(
    contract: dict[str, Any],
) -> bool:

    return (
        "PortableImplementationIncompleteError"
        in contract[
            "calls"
        ]
        and (
            contract[
                "returns"
            ]
            == []
        )
    )


authority_role_classification = {
    "frozen_v9": {
        "boundary_function":
            "_build_portable_frozen_v9",

        "boundary_status":
            (
                "IMPLEMENTED"
                if not raises_incomplete_error(
                    contracts[
                        "_build_portable_frozen_v9"
                    ]
                )
                else "PLACEHOLDER"
            ),
    },

    "morph_div_v2": {
        "legacy_boundary_function":
            "_apply_portable_morph_div_v2",

        "legacy_boundary_status":
            (
                "PLACEHOLDER"
                if raises_incomplete_error(
                    contracts[
                        "_apply_portable_morph_div_v2"
                    ]
                )
                else "IMPLEMENTED"
            ),

        "implemented_runtime_function":
            "apply_portable_morph_div_v2",

        "implemented_runtime_status":
            implemented_runtime_contracts[
                "apply_portable_morph_div_v2"
            ][
                "status"
            ],
    },

    "mutual_cont_v1": {
        "legacy_boundary_function":
            "_reconstruct_portable_mutual_cont_v1",

        "legacy_boundary_status":
            (
                "PLACEHOLDER"
                if raises_incomplete_error(
                    contracts[
                        "_reconstruct_portable_mutual_cont_v1"
                    ]
                )
                else "IMPLEMENTED"
            ),

        "implemented_matrix_core":
            "reconstruct_mutual_cont_v1_actions",

        "graph_level_runtime_status":
            "UNRESOLVED_IN_14_3A_R2",
    },

    "mutual_cont_v2": {
        "legacy_boundary_function":
            "_apply_portable_mutual_cont_v2",

        "legacy_boundary_status":
            (
                "PLACEHOLDER"
                if raises_incomplete_error(
                    contracts[
                        "_apply_portable_mutual_cont_v2"
                    ]
                )
                else "IMPLEMENTED"
            ),

        "action_builder":
            "build_portable_mutual_cont_v2_actions",

        "implemented_runtime_function":
            "apply_portable_mutual_cont_v2",

        "implemented_runtime_status":
            implemented_runtime_contracts[
                "apply_portable_mutual_cont_v2"
            ][
                "status"
            ],
    },
}


print(
    "=" * 112,
    flush=True,
)

print(
    "AUTHORITY ROLE CLASSIFICATION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        authority_role_classification,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Formal semantic gates
#
# IMPORTANT:
# Gates must be TRUE when the invariant is satisfied.
#
# Descriptive facts such as source_mutation=False belong in claim_boundary,
# NOT as a False-valued PASS/FAIL gate.
# =============================================================================

semantic_gates = {
    "parent_14_2_r2_lock":
        (
            parent_file_sha
            == EXPECTED_PARENT_FILE_SHA
        ),

    "repaired_portable_source_lock":
        (
            portable_sha
            == EXPECTED_PORTABLE_SHA
        ),

    "required_function_set_complete":
        (
            missing_functions
            == []
        ),

    "all_contract_sources_nonempty":
        all(
            bool(
                contracts[
                    name
                ][
                    "source"
                ].strip()
            )

            for name in (
                TARGET_FUNCTIONS
            )
        ),

    "all_contract_source_shas_frozen":
        all(
            len(
                contracts[
                    name
                ][
                    "source_sha256"
                ]
            )
            == 64

            for name in (
                TARGET_FUNCTIONS
            )
        ),

    "v2_action_builder_survives_repair":
        (
            "build_portable_mutual_cont_v2_actions"
            in functions
        ),

    "historical_architecture_not_fabricated":
        True,

    # -------------------------------------------------------------
    # FIX FROM 14.3A-R1:
    #
    # OLD:
    #     "source_mutation": False
    #
    # NEW:
    #     gate is phrased positively and therefore PASS == True.
    # -------------------------------------------------------------
    "no_source_mutation":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3A-R2 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed = [
    gate

    for gate, ok in (
        semantic_gates.items()
    )

    if not ok
]


assert (
    failed
    == []
), (
    "14.3A-R2 failed gates: "
    f"{failed}"
)


# =============================================================================
# 20. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3A_R2",

    "title":
        "POST_V2_GRAPH_PROVIDER_CONTRACT_EXTRACTION",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_SOURCE_CONTRACT_ONLY",

    "parent_14_2_r2": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            parent_file_sha,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "failed_predecessors": [
        {
            "stage":
                "NOTEBOOK14_14_3A",

            "failure": (
                "Preflight incorrectly required one old Notebook 13 "
                "architecture-refreeze artifact to exist at one fixed path."
            ),

            "source_mutation":
                False,

            "artifact_written":
                False,
        },
        {
            "stage":
                "NOTEBOOK14_14_3A_R1",

            "failure": (
                "Semantic-gate polarity bug: descriptive "
                "`source_mutation=False` was placed in the generic "
                "PASS/FAIL semantic-gate dictionary, causing False "
                "to be interpreted as a failed gate even though "
                "no source mutation occurred."
            ),

            "source_mutation":
                False,

            "artifact_written":
                False,
        },
    ],

    "portable_source": {
        "path":
            str(
                PORTABLE_REL
            ),

        "file_sha256":
            portable_sha,
    },

    "historical_architecture_artifact": {
        "expected_file_sha256":
            EXPECTED_ARCH_FILE_SHA,

        "expected_content_sha256":
            EXPECTED_ARCH_CONTENT_SHA,

        "discovery":
            architecture_discovery,

        "authority_status":
            architecture_authority_status,

        "selected_path":
            (
                str(
                    selected_arch_path.relative_to(
                        ROOT
                    )
                )
                if selected_arch_path
                is not None
                else None
            ),

        "selected_file_sha256":
            selected_arch_file_sha,

        "fabricated_or_recreated":
            False,
    },

    "stage_sequence_reference": {
        "status":
            "CURRENT_SOURCE_CONTRACT_SEQUENCE_FOR_PROVIDER_IMPLEMENTATION",

        "sequence":
            FROZEN_STAGE_SEQUENCE,

        "note": (
            "Current byte-locked source contracts are frozen here. "
            "Historical architecture authority is separately recorded "
            "when physically available."
        ),
    },

    "function_contracts":
        contracts,

    "implemented_runtime_contracts":
        implemented_runtime_contracts,

    "authority_role_classification":
        authority_role_classification,

    "call_examples":
        call_examples,

    "related_function_inventory":
        related_function_inventory,

    "orchestration_candidates":
        orchestration_candidates,

    "direct_call_matrix":
        direct_call_matrix,

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "provider_source_contracts":
            "FROZEN",

        "historical_architecture_artifact":
            architecture_authority_status,

        "post_v2_graph_provider_implemented":
            False,

        "post_v2_graph_structural_fidelity":
            "NOT_YET_VERIFIED",

        "b12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "gt_read":
            False,

        "evaluator_read":
            False,

        "historical_structural_oracle_read":
            False,

        "performance_metric_generated":
            False,

        # Descriptive fact. False is correct HERE.
        "source_mutation":
            False,
    },

    "explicit_non_claims": [
        "No post-V2 graph provider implementation claim.",
        "No post-V2 graph structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance evidence.",
        "No fresh-holdout evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage": (
        "NOTEBOOK14_14_3B_"
        "POST_V2_GRAPH_PROVIDER_AUTHORITY_ROLE_RECONCILIATION"
    ),
}


content_sha = (
    canonical_json_sha(
        payload
    )
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3A-R2 artifact already exists but differs "
        "from current exact source contracts."
    )


    write_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    write_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = (
    sha256_file(
        OUT
    )
)


# =============================================================================
# 21. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3A-R2 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "historical architecture artifact:",
    architecture_authority_status,
    flush=True,
)


if (
    selected_arch_path
    is not None
):

    print(
        "  path:",
        selected_arch_path.relative_to(
            ROOT
        ),
        flush=True,
    )

    print(
        "  file SHA:",
        selected_arch_file_sha,
        flush=True,
    )


print(
    "  fabricated/recreated: NO",
    flush=True,
)

print(
    flush=True,
)


print(
    "FROZEN CURRENT SOURCE CONTRACTS:",
    flush=True,
)


for function_name in (
    TARGET_FUNCTIONS
):

    contract = (
        contracts[
            function_name
        ]
    )


    print(
        " ",
        function_name,
        flush=True,
    )

    print(
        "    SHA:",
        contract[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "    returns:",
        [
            record[
                "expression"
            ]

            for record in (
                contract[
                    "returns"
                ]
            )
        ],
        flush=True,
    )


print(
    flush=True,
)


print(
    "IMPLEMENTED RUNTIME ROLES:",
    flush=True,
)

print(
    "  MorphDiv V2:",
    flush=True,
)

print(
    "    _apply_portable_morph_div_v2 =",
    authority_role_classification[
        "morph_div_v2"
    ][
        "legacy_boundary_status"
    ],
    flush=True,
)

print(
    "    apply_portable_morph_div_v2 =",
    authority_role_classification[
        "morph_div_v2"
    ][
        "implemented_runtime_status"
    ],
    flush=True,
)


print(
    "  Mutual-Cont V1:",
    flush=True,
)

print(
    "    _reconstruct_portable_mutual_cont_v1 =",
    authority_role_classification[
        "mutual_cont_v1"
    ][
        "legacy_boundary_status"
    ],
    flush=True,
)

print(
    "    graph-level runtime =",
    authority_role_classification[
        "mutual_cont_v1"
    ][
        "graph_level_runtime_status"
    ],
    flush=True,
)


print(
    "  Mutual-Cont V2:",
    flush=True,
)

print(
    "    _apply_portable_mutual_cont_v2 =",
    authority_role_classification[
        "mutual_cont_v2"
    ][
        "legacy_boundary_status"
    ],
    flush=True,
)

print(
    "    apply_portable_mutual_cont_v2 =",
    authority_role_classification[
        "mutual_cont_v2"
    ][
        "implemented_runtime_status"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    write_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  V2 invented-sort provenance mismatch = CLOSED",
    flush=True,
)

print(
    "  current post-V2 source contracts = FROZEN",
    flush=True,
)

print(
    "  source mutation in this stage = NONE",
    flush=True,
)

print(
    "  post-V2 provider = NOT YET IMPLEMENTED",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    "  Notebook 14 / 14.3B",
    flush=True,
)

print(
    "  POST_V2_GRAPH_PROVIDER_AUTHORITY_ROLE_RECONCILIATION",
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3B-R1 — POST-V2 AUTHORITY-ROLE RECONCILIATION
#            + V1 GRAPH-WIRING AUTHORITY DISCOVERY
#
# SUCCESSOR TO
# ------------
# 14.3B failed after successfully locking:
#
#   - 14.3A-R2
#   - architecture authority
#   - portable source
#   - placeholder/runtime role classification
#   - historical B12 writer
#   - build_current_v2
#   - add_actions
#
# Failure reason:
#
#     historical_build_source.find("v2")
#
# matched the substring "v2" in the FUNCTION NAME:
#
#     build_current_v2
#
# instead of the V2 add_actions stage.
#
# R1 REPAIR
# ---------
# Prove historical stage order structurally from AST:
#
#     add_actions(g, nodes, morph, name)
#     add_actions(g, nodes, v1,    name)
#     add_actions(g, nodes, v2,    name)
#
# Sort those calls by source position and inspect argument #3.
#
# NO:
#   - source mutation
#   - GT
#   - evaluator
#   - historical structural replay
#   - performance metric
#   - structural-fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3a_r2_"
    "post_v2_graph_provider_contract_extraction_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


B12_WRITER_REL = Path(
    "scripts/run_stage12_fold0_bridge_geometry_universe.py"
)

B12_WRITER = ROOT / B12_WRITER_REL


ARCH_REL = Path(
    "reports/"
    "notebook13_13_3c_r2_"
    "corrected_authority_model_and_architecture_refreeze_v1.json"
)

ARCH = ROOT / ARCH_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3b_r1_"
    "post_v2_authority_role_reconciliation_"
    "and_v1_graph_wiring_discovery_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "8de81c98306b623f21cbe5087549e362d8b88bb7a5c3fb35ae771da227ad9a07"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "0cca47421c42135c01dfe4e49901eebd73fbf3e357e31540414cf6c27efc6e6b"
)


EXPECTED_PORTABLE_SHA = (
    "a1fe0c9e5cd5e20a2ae5921bd4b4c55d078022c3348d5945bffa501ac3ff36ed"
)


EXPECTED_ARCH_FILE_SHA = (
    "e7d307f988b97717a4250534179fc07f90408b29dedccbcc4fc9b96567c7f4c8"
)

EXPECTED_ARCH_CONTENT_SHA = (
    "49f533d19e2354735c9faef38179625c058732ba4838e1a7544aefd1733fc3f4"
)


EXPECTED_B12_WRITER_SHA = (
    "c7fb950c03711ac9c85f0476013f88554a42fc2b8fb21c8cc4afc1bb6b5649a4"
)


EXPECTED_BUILD_CURRENT_V2_SHA = (
    "5e21fbd7d7ff443c87587398dc7ebd4247f5bc2eb87ac98986db89c91ea8182c"
)

EXPECTED_ADD_ACTIONS_SHA = (
    "36a6e189bbaea0fe2d9d2c4f114142087e5d99a42bd0c415085cec579bb07751"
)


EXPECTED_FUNCTION_SHAS = {
    "_build_portable_frozen_v9":
        "d4f91cd4d7c1cd97538f6dc01f0087f9c25af07bf3c7dd4d858891d1b3031505",

    "_apply_portable_morph_div_v2":
        "980f5dbbc18cffcec0872688bf1b6964f0087c4d4c28a6dbae7e9724c98f16d4",

    "_reconstruct_portable_mutual_cont_v1":
        "ea9cba47ba6992c0dab67d747ef119061eb92b3cd52f0762274d6300a076e44e",

    "_apply_portable_mutual_cont_v2":
        "377aebea342624d6d51be5231f606c614ca62a899bc217f2f8fb990106d741f7",

    "reconstruct_mutual_cont_v1_actions":
        "1a9e988329cf91c9a8a706987fede592e0d223e69ec1113e02e42726f6d09449",

    "apply_portable_morph_div_v2":
        "db173c58a83eddd282712915514c31739815195ccb99edaa84a3ff5e16f1ffcf",

    "build_portable_mutual_cont_v2_actions":
        "099dcec88cd225d75736c1f075d2c214e0e5c598db4b1a3850b1011b270f5f46",

    "apply_portable_mutual_cont_v2":
        "1901c1977a3f0f0cb6960aee426b8b726259b54e504646eeb9353febaca0d1dc",
}


EXPECTED_HISTORICAL_STAGE_SEQUENCE = [
    "morph",
    "v1",
    "v2",
]


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def contains_exact_string(
    value: Any,
    target: str,
) -> bool:

    if isinstance(
        value,
        str,
    ):

        return (
            value
            == target
        )


    if isinstance(
        value,
        dict,
    ):

        return any(
            contains_exact_string(
                key,
                target,
            )
            or
            contains_exact_string(
                item,
                target,
            )

            for key, item
            in value.items()
        )


    if isinstance(
        value,
        (
            list,
            tuple,
        ),
    ):

        return any(
            contains_exact_string(
                item,
                target,
            )

            for item
            in value
        )


    return False


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = (
        ast.get_source_segment(
            text,
            node,
        )
    )


    assert (
        source
        is not None
    ), (
        "Unable to extract exact AST source for "
        f"{type(node).__name__} "
        f"L{getattr(node, 'lineno', '?')}"
    )


    return source


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return (
            node.id
        )


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = (
            dotted_name(
                node.value
            )
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return (
            node.attr
        )


    return None


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def function_calls(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[str]:

    return sorted({
        dotted_name(
            call.func
        )
        or "<dynamic-call>"

        for call in (
            ast.walk(
                fn
            )
        )

        if isinstance(
            call,
            ast.Call,
        )
    })


def function_returns(
    text: str,
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for node in (
        ast.walk(
            fn
        )
    ):

        if not isinstance(
            node,
            ast.Return,
        ):

            continue


        result.append({
            "line":
                node.lineno,

            "expression":
                (
                    ast.unparse(
                        node.value
                    )

                    if node.value
                    is not None

                    else "None"
                ),

            "source":
                node_source(
                    text,
                    node,
                ),
        })


    return result


def function_record(
    *,
    text: str,
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> dict[str, Any]:

    source = (
        node_source(
            text,
            fn,
        )
    )


    return {
        "name":
            fn.name,

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "calls":
            function_calls(
                fn
            ),

        "returns":
            function_returns(
                text,
                fn,
            ),

        "source":
            source,
    }


# =============================================================================
# 3. V1 source-discovery helpers
# =============================================================================

V1_SEARCH_TERMS = [
    "mutual_cont_v1",
    "MUTUAL_CONT_V1",
    "MUTUAL-CONT-V1",
    "stage12_fold0_mutual_cont_v1",
    "source_pred_node_id",
    "target_pred_node_id",
    "round1",
    "association",
]


def matching_lines(
    text: str,
    *,
    terms: list[str],
    context: int = 2,
    max_blocks: int = 12,
) -> list[
    dict[str, Any]
]:

    lines = (
        text.splitlines()
    )


    lower_lines = [
        line.lower()
        for line in (
            lines
        )
    ]


    lower_terms = [
        term.lower()
        for term in (
            terms
        )
    ]


    hit_indices = []


    for index, line in enumerate(
        lower_lines
    ):

        if any(
            term in line
            for term in (
                lower_terms
            )
        ):

            hit_indices.append(
                index
            )


    blocks = []


    for index in (
        hit_indices[
            :max_blocks
        ]
    ):

        start = max(
            0,
            index - context,
        )


        end = min(
            len(
                lines
            ),
            index + context + 1,
        )


        blocks.append({
            "hit_line":
                index + 1,

            "start_line":
                start + 1,

            "end_line":
                end,

            "text":
                "\n".join(
                    lines[
                        start:end
                    ]
                ),
        })


    return blocks


def candidate_score(
    text: str,
) -> tuple[
    int,
    list[str],
]:

    lower = (
        text.lower()
    )


    score = 0
    reasons = []


    weighted_terms = [
        (
            "mutual_cont_v1",
            10,
        ),
        (
            "mutual-cont-v1",
            10,
        ),
        (
            "stage12_fold0_mutual_cont_v1",
            12,
        ),
        (
            "source_pred_node_id",
            3,
        ),
        (
            "target_pred_node_id",
            3,
        ),
        (
            "candidate_prob",
            2,
        ),
        (
            "association",
            2,
        ),
        (
            "add_edge",
            4,
        ),
        (
            "recovered_edge_dist",
            5,
        ),
        (
            "graph_state",
            3,
        ),
        (
            "to_pickle",
            2,
        ),
    ]


    for term, weight in (
        weighted_terms
    ):

        if (
            term
            in lower
        ):

            score += (
                weight
            )


            reasons.append(
                f"{term}:{weight}"
            )


    return (
        score,
        reasons,
    )


def scan_v1_source_candidates() -> list[
    dict[str, Any]
]:

    roots = [
        ROOT / "scripts",
        ROOT / "src",
        ROOT / "tests",
    ]


    candidates = []


    for base in (
        roots
    ):

        if not base.exists():

            continue


        for path in sorted(
            base.rglob(
                "*.py"
            )
        ):

            try:

                text = (
                    path.read_text(
                        encoding="utf-8"
                    )
                )

            except Exception:

                continue


            score, reasons = (
                candidate_score(
                    text
                )
            )


            if (
                score
                <= 0
            ):

                continue


            try:

                tree = ast.parse(
                    text,
                    filename=str(
                        path
                    ),
                )


                functions = (
                    top_level_functions(
                        tree
                    )
                )


                function_summaries = []


                for fn_name, fn in (
                    functions.items()
                ):

                    fn_source = (
                        node_source(
                            text,
                            fn,
                        )
                    )


                    fn_score, fn_reasons = (
                        candidate_score(
                            fn_source
                        )
                    )


                    if (
                        fn_score
                        <= 0
                    ):

                        continue


                    function_summaries.append({
                        "name":
                            fn_name,

                        "line_start":
                            fn.lineno,

                        "line_end":
                            fn.end_lineno,

                        "score":
                            fn_score,

                        "score_reasons":
                            fn_reasons,

                        "source_sha256":
                            sha256_text(
                                fn_source
                            ),

                        "calls":
                            function_calls(
                                fn
                            ),

                        "returns":
                            function_returns(
                                text,
                                fn,
                            ),
                    })


            except SyntaxError:

                function_summaries = []


            candidates.append({
                "path":
                    str(
                        path.relative_to(
                            ROOT
                        )
                    ),

                "file_sha256":
                    sha256_file(
                        path
                    ),

                "score":
                    score,

                "score_reasons":
                    reasons,

                "matching_blocks":
                    matching_lines(
                        text,
                        terms=V1_SEARCH_TERMS,
                    ),

                "function_candidates":
                    sorted(
                        function_summaries,
                        key=lambda item: (
                            -item[
                                "score"
                            ],
                            item[
                                "line_start"
                            ],
                        ),
                    ),
            })


    return sorted(
        candidates,
        key=lambda item: (
            -item[
                "score"
            ],
            item[
                "path"
            ],
        ),
    )


# =============================================================================
# 4. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    B12_WRITER,
    ARCH,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3B-R1 "
    "POST-V2 AUTHORITY-ROLE RECONCILIATION "
    "+ V1 GRAPH-WIRING AUTHORITY DISCOVERY",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "PORTABLE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "B12 HISTORICAL WRITER:",
    B12_WRITER_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Parent 14.3A-R2 lock
# =============================================================================

parent_file_sha = (
    sha256_file(
        PARENT
    )
)


assert (
    parent_file_sha
    == EXPECTED_PARENT_FILE_SHA
), (
    "14.3A-R2 file SHA mismatch."
)


parent_obj = (
    json.loads(
        PARENT.read_text(
            encoding="utf-8"
        )
    )
)


assert (
    parent_obj.get(
        "decision"
    )
    == "PASS"
)


assert contains_exact_string(
    parent_obj,
    EXPECTED_PARENT_CONTENT_SHA,
)


print(
    "[PASS] PARENT_14_3A_R2_LOCK",
    flush=True,
)

print(
    "  file SHA    :",
    parent_file_sha,
    flush=True,
)

print(
    "  content SHA :",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Historical architecture lock
# =============================================================================

arch_file_sha = (
    sha256_file(
        ARCH
    )
)


assert (
    arch_file_sha
    == EXPECTED_ARCH_FILE_SHA
)


arch_obj = (
    json.loads(
        ARCH.read_text(
            encoding="utf-8"
        )
    )
)


assert contains_exact_string(
    arch_obj,
    EXPECTED_ARCH_CONTENT_SHA,
)


print(
    "[PASS] HISTORICAL_ARCHITECTURE_LOCK",
    flush=True,
)

print(
    "  path:",
    ARCH_REL,
    flush=True,
)

print(
    "  file SHA:",
    arch_file_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Portable source lock
# =============================================================================

portable_sha = (
    sha256_file(
        PORTABLE
    )
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
), (
    "Portable CURRENT_BEST source changed "
    "after 14.3A-R2."
)


portable_text = (
    PORTABLE.read_text(
        encoding="utf-8"
    )
)


portable_tree = (
    ast.parse(
        portable_text,
        filename=str(
            PORTABLE
        ),
    )
)


portable_functions = (
    top_level_functions(
        portable_tree
    )
)


print(
    "[PASS] PORTABLE_SOURCE_LOCK",
    flush=True,
)

print(
    "  SHA:",
    portable_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Lock authority-role functions
# =============================================================================

missing_expected_functions = [
    name

    for name in (
        EXPECTED_FUNCTION_SHAS
    )

    if name not in (
        portable_functions
    )
]


assert (
    missing_expected_functions
    == []
)


current_function_records = {}


for name, expected_sha in (
    EXPECTED_FUNCTION_SHAS.items()
):

    fn = (
        portable_functions[
            name
        ]
    )


    record = (
        function_record(
            text=portable_text,
            fn=fn,
        )
    )


    assert (
        record[
            "source_sha256"
        ]
        == expected_sha
    ), (
        f"Source drift for {name}."
    )


    current_function_records[
        name
    ] = (
        record
    )


print(
    "[PASS] AUTHORITY_ROLE_FUNCTION_LOCKS",
    flush=True,
)


for name, record in (
    current_function_records.items()
):

    print(
        " ",
        name,
        flush=True,
    )

    print(
        "    SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 9. Placeholder / implementation classification
# =============================================================================

def is_incomplete_placeholder(
    record: dict[str, Any],
) -> bool:

    return (
        "PortableImplementationIncompleteError"
        in record[
            "calls"
        ]

        and (
            record[
                "returns"
            ]
            == []
        )
    )


assert (
    not is_incomplete_placeholder(
        current_function_records[
            "_build_portable_frozen_v9"
        ]
    )
)


assert is_incomplete_placeholder(
    current_function_records[
        "_apply_portable_morph_div_v2"
    ]
)


assert is_incomplete_placeholder(
    current_function_records[
        "_reconstruct_portable_mutual_cont_v1"
    ]
)


assert is_incomplete_placeholder(
    current_function_records[
        "_apply_portable_mutual_cont_v2"
    ]
)


assert (
    not is_incomplete_placeholder(
        current_function_records[
            "apply_portable_morph_div_v2"
        ]
    )
)


assert (
    not is_incomplete_placeholder(
        current_function_records[
            "reconstruct_mutual_cont_v1_actions"
        ]
    )
)


assert (
    not is_incomplete_placeholder(
        current_function_records[
            "apply_portable_mutual_cont_v2"
        ]
    )
)


print(
    "[PASS] PLACEHOLDER_VS_IMPLEMENTED_RUNTIME_CLASSIFICATION",
    flush=True,
)

print(
    "  FrozenV9                  = IMPLEMENTED",
    flush=True,
)

print(
    "  Morph underscore boundary = PLACEHOLDER",
    flush=True,
)

print(
    "  Morph public runtime      = IMPLEMENTED",
    flush=True,
)

print(
    "  V1 underscore boundary    = PLACEHOLDER",
    flush=True,
)

print(
    "  V1 matrix/action core     = IMPLEMENTED",
    flush=True,
)

print(
    "  V2 underscore boundary    = PLACEHOLDER",
    flush=True,
)

print(
    "  V2 public runtime         = IMPLEMENTED",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Authority role reconciliation
# =============================================================================

authority_roles = {
    "FROZEN_V9": {
        "historical_role":
            "BASE_GRAPH",

        "current_authority":
            "_build_portable_frozen_v9",

        "status":
            "IMPLEMENTED",
    },

    "MORPH_DIV_V2": {
        "historical_role":
            "POST_FROZEN_V9_GRAPH_MUTATION",

        "legacy_placeholder":
            "_apply_portable_morph_div_v2",

        "current_authority":
            "apply_portable_morph_div_v2",

        "status":
            "IMPLEMENTED_RUNTIME_PRESENT",
    },

    "MUTUAL_CONT_V1": {
        "historical_role":
            "POST_MORPH_GRAPH_MUTATION",

        "legacy_placeholder":
            "_reconstruct_portable_mutual_cont_v1",

        "implemented_matrix_action_core":
            "reconstruct_mutual_cont_v1_actions",

        "graph_level_runtime_authority":
            None,

        "status":
            "GRAPH_LEVEL_RUNTIME_UNRESOLVED",
    },

    "MUTUAL_CONT_V2": {
        "historical_role":
            "POST_V1_GRAPH_MUTATION",

        "legacy_placeholder":
            "_apply_portable_mutual_cont_v2",

        "action_builder":
            "build_portable_mutual_cont_v2_actions",

        "current_authority":
            "apply_portable_mutual_cont_v2",

        "status":
            "IMPLEMENTED_RUNTIME_PRESENT",
    },
}


print(
    "=" * 112,
    flush=True,
)

print(
    "CURRENT AUTHORITY ROLE RECONCILIATION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        authority_roles,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Historical B12 authority locks
# =============================================================================

b12_sha = (
    sha256_file(
        B12_WRITER
    )
)


assert (
    b12_sha
    == EXPECTED_B12_WRITER_SHA
)


b12_text = (
    B12_WRITER.read_text(
        encoding="utf-8"
    )
)


b12_tree = (
    ast.parse(
        b12_text,
        filename=str(
            B12_WRITER
        ),
    )
)


b12_functions = (
    top_level_functions(
        b12_tree
    )
)


assert (
    "build_current_v2"
    in b12_functions
)

assert (
    "add_actions"
    in b12_functions
)


build_current_v2_fn = (
    b12_functions[
        "build_current_v2"
    ]
)

add_actions_fn = (
    b12_functions[
        "add_actions"
    ]
)


build_current_v2_record = (
    function_record(
        text=b12_text,
        fn=build_current_v2_fn,
    )
)


add_actions_record = (
    function_record(
        text=b12_text,
        fn=add_actions_fn,
    )
)


assert (
    build_current_v2_record[
        "source_sha256"
    ]
    == EXPECTED_BUILD_CURRENT_V2_SHA
)


assert (
    add_actions_record[
        "source_sha256"
    ]
    == EXPECTED_ADD_ACTIONS_SHA
)


print(
    "[PASS] HISTORICAL_POST_V2_GRAPH_MUTATION_AUTHORITY_LOCK",
    flush=True,
)

print(
    "  B12 writer SHA:",
    b12_sha,
    flush=True,
)

print(
    "  build_current_v2 SHA:",
    build_current_v2_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  add_actions SHA:",
    add_actions_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. R1 FIX:
#     AST-level historical stage-order proof
# =============================================================================

add_action_call_records = []


for call in (
    ast.walk(
        build_current_v2_fn
    )
):

    if not isinstance(
        call,
        ast.Call,
    ):

        continue


    if (
        dotted_name(
            call.func
        )
        != "add_actions"
    ):

        continue


    assert (
        len(
            call.args
        )
        >= 4
    ), (
        "Historical add_actions call signature drift."
    )


    action_table_expression = (
        ast.unparse(
            call.args[
                2
            ]
        )
    )


    dataset_expression = (
        ast.unparse(
            call.args[
                3
            ]
        )
    )


    add_action_call_records.append({
        "line":
            call.lineno,

        "column":
            call.col_offset,

        "action_table_expression":
            action_table_expression,

        "dataset_expression":
            dataset_expression,

        "source":
            node_source(
                b12_text,
                call,
            ),

        "source_sha256":
            sha256_text(
                node_source(
                    b12_text,
                    call,
                )
            ),
    })


add_action_call_records = (
    sorted(
        add_action_call_records,
        key=lambda item: (
            item[
                "line"
            ],
            item[
                "column"
            ],
        ),
    )
)


assert (
    len(
        add_action_call_records
    )
    == 3
), (
    "Expected exactly three historical add_actions calls "
    f"inside build_current_v2; found "
    f"{len(add_action_call_records)}."
)


historical_stage_sequence = [
    record[
        "action_table_expression"
    ]

    for record in (
        add_action_call_records
    )
]


assert (
    historical_stage_sequence
    == EXPECTED_HISTORICAL_STAGE_SEQUENCE
), (
    "Historical action-stage order drift.\n"
    f"expected={EXPECTED_HISTORICAL_STAGE_SEQUENCE}\n"
    f"actual  ={historical_stage_sequence}"
)


assert all(
    record[
        "dataset_expression"
    ]
    == "name"

    for record in (
        add_action_call_records
    )
)


print(
    "[PASS] HISTORICAL_STAGE_ORDER_AST_LOCK",
    flush=True,
)

print(
    "  exact ordered action tables:",
    historical_stage_sequence,
    flush=True,
)


for index, record in enumerate(
    add_action_call_records,
    start=1,
):

    print(
        f"  [{index}] "
        f"L{record['line']} "
        f"table={record['action_table_expression']} "
        f"dataset={record['dataset_expression']}",
        flush=True,
    )

    print(
        "      SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "      source:",
        flush=True,
    )

    print(
        "        "
        + record[
            "source"
        ].replace(
            "\n",
            "\n        ",
        ),
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 13. Historical generic graph-mutation semantics
# =============================================================================

add_actions_source = (
    add_actions_record[
        "source"
    ]
)


assert (
    "source_pred_node_id"
    in add_actions_source
)

assert (
    "target_pred_node_id"
    in add_actions_source
)

assert (
    "candidate_prob"
    in add_actions_source
)

assert (
    "recovered_edge_dist"
    in add_actions_source
)

assert (
    ".add_edge"
    in add_actions_source
)


historical_generic_action_mutation = {
    "function":
        "add_actions",

    "function_sha256":
        add_actions_record[
            "source_sha256"
        ],

    "table_filter":
        'table[table["dataset"].eq(name)]',

    "source_column":
        "source_pred_node_id",

    "target_column":
        "target_pred_node_id",

    "probability_column":
        "candidate_prob",

    "distance_authority":
        "fv9.recovered_edge_dist",

    "graph_mutation":
        "g.add_edge",

    "validate_keys":
        True,

    "used_for_stages":
        historical_stage_sequence,
}


print(
    "[PASS] HISTORICAL_GENERIC_ACTION_MUTATION_SEMANTICS_LOCK",
    flush=True,
)

print(
    json.dumps(
        historical_generic_action_mutation,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Search repo for V1 historical source evidence
# =============================================================================

v1_candidates = (
    scan_v1_source_candidates()
)


print(
    "=" * 112,
    flush=True,
)

print(
    "V1 GRAPH-WIRING SOURCE CANDIDATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "candidate files:",
    len(
        v1_candidates
    ),
    flush=True,
)

print(
    flush=True,
)


for rank, candidate in enumerate(
    v1_candidates[
        :20
    ],
    start=1,
):

    print(
        f"[{rank}] "
        f"score={candidate['score']} "
        f"{candidate['path']}",
        flush=True,
    )

    print(
        "    SHA:",
        candidate[
            "file_sha256"
        ],
        flush=True,
    )

    print(
        "    reasons:",
        candidate[
            "score_reasons"
        ],
        flush=True,
    )


    if (
        candidate[
            "function_candidates"
        ]
    ):

        print(
            "    matching functions:",
            flush=True,
        )


        for function in (
            candidate[
                "function_candidates"
            ][
                :10
            ]
        ):

            print(
                "      "
                f"{function['name']} "
                f"L{function['line_start']}"
                f"-L{function['line_end']} "
                f"score={function['score']} "
                f"SHA={function['source_sha256']}",
                flush=True,
            )

            print(
                "        calls:",
                function[
                    "calls"
                ],
                flush=True,
            )


    if (
        candidate[
            "matching_blocks"
        ]
    ):

        print(
            "    source evidence:",
            flush=True,
        )


        for block in (
            candidate[
                "matching_blocks"
            ][
                :5
            ]
        ):

            print(
                f"      L{block['start_line']}"
                f"-L{block['end_line']}:",
                flush=True,
            )

            print(
                "        "
                + block[
                    "text"
                ].replace(
                    "\n",
                    "\n        ",
                ),
                flush=True,
            )


    print(
        flush=True,
    )


# =============================================================================
# 15. High-confidence V1-specific files
# =============================================================================

high_confidence_v1_files = []


for candidate in (
    v1_candidates
):

    path_lower = (
        candidate[
            "path"
        ].lower()
    )


    if (
        "mutual_cont_v1"
        in path_lower
        or
        "mutual-cont-v1"
        in path_lower
        or
        "stage12_fold0_mutual_cont_v1"
        in path_lower
    ):

        high_confidence_v1_files.append(
            candidate
        )


print(
    "=" * 112,
    flush=True,
)

print(
    "HIGH-CONFIDENCE V1-NAMED FILES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


if not high_confidence_v1_files:

    print(
        "  <none>",
        flush=True,
    )


for candidate in (
    high_confidence_v1_files
):

    print(
        " ",
        candidate[
            "path"
        ],
        flush=True,
    )

    print(
        "    score:",
        candidate[
            "score"
        ],
        flush=True,
    )

    print(
        "    SHA:",
        candidate[
            "file_sha256"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 16. V1 graph-mutation / action-generation function candidates
# =============================================================================

v1_function_candidates = []


for candidate in (
    v1_candidates
):

    for function in (
        candidate[
            "function_candidates"
        ]
    ):

        calls = set(
            function[
                "calls"
            ]
        )


        joined_calls = (
            " ".join(
                calls
            ).lower()
        )


        name_lower = (
            function[
                "name"
            ].lower()
        )


        has_graph_mutation = any(
            call.endswith(
                ".add_edge"
            )
            or call == "add_edge"

            for call in calls
        )


        looks_v1_specific = (
            "v1"
            in name_lower
            or
            "mutual"
            in name_lower
            or
            "association"
            in joined_calls
        )


        if (
            has_graph_mutation
            or
            looks_v1_specific
        ):

            v1_function_candidates.append({
                "path":
                    candidate[
                        "path"
                    ],

                **function,
            })


v1_function_candidates = sorted(
    v1_function_candidates,
    key=lambda item: (
        -item[
            "score"
        ],
        item[
            "path"
        ],
        item[
            "line_start"
        ],
    ),
)


print(
    "=" * 112,
    flush=True,
)

print(
    "V1 FUNCTION-LEVEL AUTHORITY CANDIDATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


if not v1_function_candidates:

    print(
        "  <none>",
        flush=True,
    )


for candidate in (
    v1_function_candidates[
        :30
    ]
):

    print(
        f"  {candidate['path']} :: "
        f"{candidate['name']} "
        f"L{candidate['line_start']}"
        f"-L{candidate['line_end']} "
        f"score={candidate['score']}",
        flush=True,
    )

    print(
        "    SHA:",
        candidate[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "    calls:",
        candidate[
            "calls"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 17. Strong source candidates
# =============================================================================

strong_v1_candidates = [
    candidate

    for candidate in (
        v1_candidates
    )

    if (
        candidate[
            "score"
        ]
        >= 18
    )
]


very_strong_v1_candidates = [
    candidate

    for candidate in (
        v1_candidates
    )

    if (
        candidate[
            "score"
        ]
        >= 26
    )
]


print(
    "=" * 112,
    flush=True,
)

print(
    "V1 AUTHORITY DISCOVERY SUMMARY",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "strong files (score >= 18):",
    len(
        strong_v1_candidates
    ),
    flush=True,
)

print(
    "very strong files (score >= 26):",
    len(
        very_strong_v1_candidates
    ),
    flush=True,
)

print(
    "high-confidence V1-named files:",
    len(
        high_confidence_v1_files
    ),
    flush=True,
)

print(
    "function-level candidates:",
    len(
        v1_function_candidates
    ),
    flush=True,
)

print(
    flush=True,
)


for candidate in (
    very_strong_v1_candidates[
        :10
    ]
):

    print(
        " ",
        candidate[
            "path"
        ],
        "score=",
        candidate[
            "score"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 18. Post-V2 provider readiness
# =============================================================================

frozen_v9_ready = (
    authority_roles[
        "FROZEN_V9"
    ][
        "status"
    ]
    == "IMPLEMENTED"
)


morph_ready = (
    authority_roles[
        "MORPH_DIV_V2"
    ][
        "status"
    ]
    == "IMPLEMENTED_RUNTIME_PRESENT"
)


v2_ready = (
    authority_roles[
        "MUTUAL_CONT_V2"
    ][
        "status"
    ]
    == "IMPLEMENTED_RUNTIME_PRESENT"
)


v1_graph_ready = (
    authority_roles[
        "MUTUAL_CONT_V1"
    ][
        "status"
    ]
    == "GRAPH_LEVEL_RUNTIME_IMPLEMENTED"
)


post_v2_provider_ready = (
    frozen_v9_ready
    and
    morph_ready
    and
    v1_graph_ready
    and
    v2_ready
)


assert (
    post_v2_provider_ready
    is False
)


provider_readiness = {
    "FrozenV9":
        frozen_v9_ready,

    "MorphDivV2":
        morph_ready,

    "MutualContV1ActionCore":
        True,

    "MutualContV1GraphLevel":
        v1_graph_ready,

    "MutualContV2":
        v2_ready,

    "post_v2_graph_provider_ready":
        post_v2_provider_ready,

    "blocking_stage":
        "MUTUAL_CONT_V1_GRAPH_LEVEL_RUNTIME",
}


print(
    "=" * 112,
    flush=True,
)

print(
    "POST-V2 PROVIDER READINESS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        provider_readiness,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Semantic gates
#
# PASS here means reconciliation + discovery completed.
# Provider readiness remains intentionally False.
# =============================================================================

semantic_gates = {
    "parent_14_3a_r2_locked":
        (
            parent_file_sha
            == EXPECTED_PARENT_FILE_SHA
        ),

    "portable_source_locked":
        (
            portable_sha
            == EXPECTED_PORTABLE_SHA
        ),

    "historical_architecture_locked":
        (
            arch_file_sha
            == EXPECTED_ARCH_FILE_SHA
        ),

    "historical_b12_writer_locked":
        (
            b12_sha
            == EXPECTED_B12_WRITER_SHA
        ),

    "build_current_v2_locked":
        (
            build_current_v2_record[
                "source_sha256"
            ]
            == EXPECTED_BUILD_CURRENT_V2_SHA
        ),

    "historical_add_actions_locked":
        (
            add_actions_record[
                "source_sha256"
            ]
            == EXPECTED_ADD_ACTIONS_SHA
        ),

    # R1 corrected structural proof.
    "historical_stage_order_ast_locked":
        (
            historical_stage_sequence
            == EXPECTED_HISTORICAL_STAGE_SEQUENCE
        ),

    "historical_generic_action_mutation_locked":
        (
            historical_generic_action_mutation[
                "used_for_stages"
            ]
            == EXPECTED_HISTORICAL_STAGE_SEQUENCE
        ),

    "frozen_v9_authority_reconciled":
        frozen_v9_ready,

    "morph_authority_reconciled":
        morph_ready,

    "v1_placeholder_identified":
        is_incomplete_placeholder(
            current_function_records[
                "_reconstruct_portable_mutual_cont_v1"
            ]
        ),

    "v1_action_core_identified":
        (
            not is_incomplete_placeholder(
                current_function_records[
                    "reconstruct_mutual_cont_v1_actions"
                ]
            )
        ),

    "v2_authority_reconciled":
        v2_ready,

    "v1_source_inventory_completed":
        True,

    "provider_remains_fail_closed_until_v1_resolved":
        (
            post_v2_provider_ready
            is False
        ),

    "no_source_mutation":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3B-R1 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok in (
        semantic_gates.items()
    )

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3B-R1 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 20. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3B_R1",

    "title": (
        "POST_V2_AUTHORITY_ROLE_RECONCILIATION_"
        "AND_V1_GRAPH_WIRING_AUTHORITY_DISCOVERY"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_AUTHORITY_RECONCILIATION_ONLY",

    "parent_14_3a_r2": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            parent_file_sha,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "failed_predecessor": {
        "stage":
            "NOTEBOOK14_14_3B",

        "failure": (
            "Historical stage-order verifier used substring "
            "positions from build_current_v2 source. "
            "Searching for `v2` matched the function name "
            "`build_current_v2` before the V2 add_actions call."
        ),

        "source_mutation":
            False,

        "artifact_written":
            False,
    },

    "portable_source": {
        "path":
            str(
                PORTABLE_REL
            ),

        "file_sha256":
            portable_sha,
    },

    "historical_architecture": {
        "path":
            str(
                ARCH_REL
            ),

        "file_sha256":
            arch_file_sha,

        "content_sha256":
            EXPECTED_ARCH_CONTENT_SHA,
    },

    "historical_post_v2_graph_authority": {
        "b12_writer_path":
            str(
                B12_WRITER_REL
            ),

        "b12_writer_file_sha256":
            b12_sha,

        "build_current_v2":
            build_current_v2_record,

        "add_actions":
            add_actions_record,

        "ordered_add_actions_calls":
            add_action_call_records,

        "stage_order":
            historical_stage_sequence,

        "stage_order_verification":
            "AST_CALL_ARGUMENT_ORDER",

        "generic_action_mutation":
            historical_generic_action_mutation,
    },

    "current_function_records":
        current_function_records,

    "authority_roles":
        authority_roles,

    "v1_authority_discovery": {
        "search_terms":
            V1_SEARCH_TERMS,

        "candidate_count":
            len(
                v1_candidates
            ),

        "strong_candidate_count":
            len(
                strong_v1_candidates
            ),

        "very_strong_candidate_count":
            len(
                very_strong_v1_candidates
            ),

        "high_confidence_named_file_count":
            len(
                high_confidence_v1_files
            ),

        "ranked_candidates":
            v1_candidates,

        "high_confidence_named_files":
            high_confidence_v1_files,

        "function_candidates":
            v1_function_candidates,
    },

    "provider_readiness":
        provider_readiness,

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "authority_role_reconciliation":
            "COMPLETE",

        "historical_stage_order":
            "MORPH_THEN_V1_THEN_V2_AST_VERIFIED",

        "historical_generic_action_mutation":
            "LOCKED",

        "frozen_v9_runtime":
            "IMPLEMENTED",

        "morph_runtime":
            "IMPLEMENTED",

        "mutual_cont_v1_action_core":
            "IMPLEMENTED",

        "mutual_cont_v1_graph_runtime":
            "UNRESOLVED",

        "mutual_cont_v2_runtime":
            "IMPLEMENTED",

        "post_v2_graph_provider":
            "BLOCKED_ON_V1_GRAPH_RUNTIME",

        "post_v2_graph_structural_fidelity":
            "NOT_YET_VERIFIED",

        "b12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "source_mutation":
            False,

        "gt_read":
            False,

        "evaluator_read":
            False,

        "historical_structural_oracle_read":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No post-V2 graph provider implementation claim.",
        "No V1 graph-level runtime implementation claim.",
        "No post-V2 structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance evidence.",
        "No fresh-holdout evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage": (
        "NOTEBOOK14_14_3C_"
        "MUTUAL_CONT_V1_GRAPH_WIRING_AUTHORITY_EXTRACTION"
    ),
}


content_sha = (
    canonical_json_sha(
        payload
    )
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3B-R1 artifact already exists but differs "
        "from current exact reconciliation."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = (
    sha256_file(
        OUT
    )
)


# =============================================================================
# 21. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3B-R1 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "HISTORICAL STAGE ORDER:",
    flush=True,
)

print(
    "  AST verified:",
    historical_stage_sequence,
    flush=True,
)

print(
    flush=True,
)


print(
    "HISTORICAL GENERIC ACTION MUTATION:",
    flush=True,
)

print(
    "  source column      = source_pred_node_id",
    flush=True,
)

print(
    "  target column      = target_pred_node_id",
    flush=True,
)

print(
    "  probability column = candidate_prob",
    flush=True,
)

print(
    "  distance authority = fv9.recovered_edge_dist",
    flush=True,
)

print(
    "  graph mutation     = g.add_edge",
    flush=True,
)

print(
    flush=True,
)


print(
    "AUTHORITY ROLES:",
    flush=True,
)

print(
    "  FrozenV9 = IMPLEMENTED",
    flush=True,
)

print(
    "  Morph V2 = IMPLEMENTED RUNTIME",
    flush=True,
)

print(
    "  V1 action core = IMPLEMENTED",
    flush=True,
)

print(
    "  V1 graph runtime = UNRESOLVED",
    flush=True,
)

print(
    "  V2 runtime = IMPLEMENTED",
    flush=True,
)

print(
    flush=True,
)


print(
    "V1 DISCOVERY:",
    flush=True,
)

print(
    "  candidate files:",
    len(
        v1_candidates
    ),
    flush=True,
)

print(
    "  strong candidates:",
    len(
        strong_v1_candidates
    ),
    flush=True,
)

print(
    "  very strong candidates:",
    len(
        very_strong_v1_candidates
    ),
    flush=True,
)

print(
    "  V1-named files:",
    len(
        high_confidence_v1_files
    ),
    flush=True,
)

print(
    "  function candidates:",
    len(
        v1_function_candidates
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "POST-V2 PROVIDER:",
    flush=True,
)

print(
    "  ready =",
    post_v2_provider_ready,
    flush=True,
)

print(
    "  blocker = MUTUAL_CONT_V1_GRAPH_LEVEL_RUNTIME",
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  authority-role reconciliation = COMPLETE",
    flush=True,
)

print(
    "  historical Morph -> V1 -> V2 order = AST VERIFIED",
    flush=True,
)

print(
    "  historical generic action mutation = LOCKED",
    flush=True,
)

print(
    "  V1 graph-level runtime = UNRESOLVED",
    flush=True,
)

print(
    "  post-V2 provider = BLOCKED ON V1",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    "  Notebook 14 / 14.3C",
    flush=True,
)

print(
    "  MUTUAL_CONT_V1_GRAPH_WIRING_AUTHORITY_EXTRACTION",
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3C-R1 — MUTUAL-CONT-V1 PRODUCER DATAFLOW CLOSURE
#
# SUCCESSOR TO
# ------------
# 14.3C successfully established:
#
#   [PASS] portable V1 action core lock
#   [PASS] association-provider source identity lock
#   [PASS] historical generic V1 mutation lock
#   [PASS] V2-confirmatory V1 baseline lock
#   [PASS] Round2 V1 mutation lock
#   [PASS] post-V1 topology lock
#   [PASS] historical edge-mutation convergence
#
# It then found:
#
#   historical V1 action artifact references in scripts/src/tests = 8
#   READ consumers                                      = present
#   WRITE producers                                     = 0
#
# and stopped because it incorrectly required the producer to exist in the
# already-scanned *.py universe.
#
#
# PURPOSE OF R1
# -------------
# Close producer provenance WITHOUT guessing.
#
# Search:
#
#   1. exact V1 lock JSON on disk;
#   2. provenance metadata inside that lock JSON;
#   3. notebooks/*.ipynb code cells, with cross-cell path-alias tracking;
#   4. reports/**/*.json / md / txt metadata references;
#   5. all repo text source for explicit action-artifact write sinks.
#
# IMPORTANT:
#   - the historical V1 *.pkl is NEVER opened;
#   - no pd.read_pickle on the historical action artifact;
#   - no structural oracle;
#   - no GT/evaluator;
#   - no source mutation.
#
#
# THIS STAGE MAY PASS WITH EITHER:
#
#   A. PRODUCER_SOURCE_FOUND
#      -> next = 14.3D V1 graph runtime implementation
#
#   B. PRODUCER_SOURCE_NOT_PRESENT_IN_CURRENT_REPO
#      -> next = 14.3C-R2 source-contract reconstruction from lock/provenance
#
# The discovery stage itself should not fail merely because old producer
# source is absent.
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3b_r1_"
    "post_v2_authority_role_reconciliation_"
    "and_v1_graph_wiring_discovery_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


B12_WRITER_REL = Path(
    "scripts/run_stage12_fold0_bridge_geometry_universe.py"
)

B12_WRITER = ROOT / B12_WRITER_REL


ITER_RECIP_REL = Path(
    "scripts/stage12_iterative_reciprocal.py"
)

ITER_RECIP = ROOT / ITER_RECIP_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3c_r1_"
    "mutual_cont_v1_producer_dataflow_closure_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "396d38246bd0a4302b1595c2e6261ae821f3e841b1beb725883710d38007ed74"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "4f9fc405aa9b0d7e6f51a450854058e9390c45988fa200738e00d7ec77d31112"
)


EXPECTED_PORTABLE_SHA = (
    "a1fe0c9e5cd5e20a2ae5921bd4b4c55d078022c3348d5945bffa501ac3ff36ed"
)


EXPECTED_V1_ACTION_CORE_SHA = (
    "1a9e988329cf91c9a8a706987fede592e0d223e69ec1113e02e42726f6d09449"
)


EXPECTED_ITER_RECIP_FILE_SHA = (
    "0f9a09e5a69b7ed55ac6613a95ef3eda7505918e258c4450b138dd49e255a550"
)

EXPECTED_ASSOCIATION_PROVIDER_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


EXPECTED_B12_WRITER_SHA = (
    "c7fb950c03711ac9c85f0476013f88554a42fc2b8fb21c8cc4afc1bb6b5649a4"
)

EXPECTED_B12_ADD_ACTIONS_SHA = (
    "36a6e189bbaea0fe2d9d2c4f114142087e5d99a42bd0c415085cec579bb07751"
)


TARGET_ACTION_FILENAME = (
    "stage12_fold0_mutual_cont_v1_actions_locked.pkl"
)

TARGET_LOCK_FILENAME = (
    "stage12_fold0_mutual_cont_v1_action_lock.json"
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    data: bytes,
) -> str:

    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(
    text: str,
) -> str:

    return sha256_bytes(
        text.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    obj: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def contains_exact_string(
    obj: Any,
    target: str,
) -> bool:

    if isinstance(
        obj,
        str,
    ):
        return (
            obj
            == target
        )

    if isinstance(
        obj,
        dict,
    ):
        return any(
            contains_exact_string(
                key,
                target,
            )
            or
            contains_exact_string(
                value,
                target,
            )
            for key, value in (
                obj.items()
            )
        )

    if isinstance(
        obj,
        (
            list,
            tuple,
        ),
    ):
        return any(
            contains_exact_string(
                item,
                target,
            )
            for item in obj
        )

    return False


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    result = ast.get_source_segment(
        text,
        node,
    )


    assert (
        result
        is not None
    ), (
        "AST source extraction failed: "
        f"{type(node).__name__} "
        f"L{getattr(node, 'lineno', '?')}"
    )


    return result


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):
        return (
            node.id
        )

    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return (
            node.attr
        )


    return None


def target_names(
    node: ast.AST,
) -> set[str]:

    names = set()


    def visit(
        item: ast.AST,
    ):

        if isinstance(
            item,
            ast.Name,
        ):
            names.add(
                item.id
            )

        elif isinstance(
            item,
            (
                ast.Tuple,
                ast.List,
            ),
        ):
            for child in (
                item.elts
            ):
                visit(
                    child
                )

        elif isinstance(
            item,
            ast.Starred,
        ):
            visit(
                item.value
            )


    visit(
        node
    )

    return names


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node
        for node in (
            tree.body
        )
        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def function_record(
    *,
    text: str,
    fn: ast.AST,
) -> dict[str, Any]:

    source = node_source(
        text,
        fn,
    )


    calls = sorted({
        dotted_name(
            call.func
        )
        or "<dynamic-call>"
        for call in ast.walk(
            fn
        )
        if isinstance(
            call,
            ast.Call,
        )
    })


    returns = []


    for node in ast.walk(
        fn
    ):

        if not isinstance(
            node,
            ast.Return,
        ):
            continue


        returns.append({
            "line":
                node.lineno,

            "expression":
                (
                    ast.unparse(
                        node.value
                    )
                    if node.value
                    is not None
                    else "None"
                ),
        })


    return {
        "name":
            getattr(
                fn,
                "name",
                None,
            ),

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "calls":
            calls,

        "returns":
            returns,

        "source":
            source,
    }


# =============================================================================
# 3. Recursive JSON provenance flattening
# =============================================================================

def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    records = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in (
            obj.items()
        ):

            child = (
                f"{prefix}.{key}"
            )


            records.extend(
                flatten_json(
                    value,
                    prefix=child,
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            child = (
                f"{prefix}[{index}]"
            )


            records.extend(
                flatten_json(
                    value,
                    prefix=child,
                )
            )


    else:

        records.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return records


PROVENANCE_KEYWORDS = (
    "producer",
    "source",
    "script",
    "notebook",
    "cell",
    "writer",
    "write",
    "function",
    "sha",
    "commit",
    "command",
    "pipeline",
    "generated",
    "creation",
    "created",
    "artifact",
)


def provenance_relevant_flat_records(
    obj: Any,
) -> list[
    dict[str, Any]
]:

    flat = flatten_json(
        obj
    )


    result = []


    for record in (
        flat
    ):

        path_lower = (
            record[
                "path"
            ].lower()
        )


        value = (
            record[
                "value"
            ]
        )


        value_text = (
            str(
                value
            )
            if value
            is not None
            else ""
        )


        value_lower = (
            value_text.lower()
        )


        path_hit = any(
            keyword
            in path_lower
            for keyword in (
                PROVENANCE_KEYWORDS
            )
        )


        value_hit = (
            ".py"
            in value_lower
            or ".ipynb"
            in value_lower
            or TARGET_ACTION_FILENAME.lower()
            in value_lower
            or TARGET_LOCK_FILENAME.lower()
            in value_lower
            or bool(
                re.fullmatch(
                    r"[0-9a-f]{64}",
                    value_lower,
                )
            )
        )


        if (
            path_hit
            or value_hit
        ):

            result.append(
                record
            )


    return result


# =============================================================================
# 4. Find exact files WITHOUT reading V1 pickle
# =============================================================================

def find_exact_named_paths(
    filename: str,
) -> list[Path]:

    return sorted(
        path
        for path in (
            ROOT.rglob(
                filename
            )
        )
        if (
            path.name
            == filename
        )
    )


# =============================================================================
# 5. Python alias propagation + read/write classification
# =============================================================================

def expr_references_artifact(
    expr: ast.AST,
    *,
    aliases: set[str],
    filename: str,
) -> bool:

    for node in ast.walk(
        expr
    ):

        if (
            isinstance(
                node,
                ast.Name,
            )
            and node.id
            in aliases
        ):

            return True


        if (
            isinstance(
                node,
                ast.Constant,
            )
            and isinstance(
                node.value,
                str,
            )
            and filename
            in node.value
        ):

            return True


    return False


def update_aliases_from_tree(
    tree: ast.AST,
    *,
    aliases: set[str],
    filename: str,
) -> set[str]:

    aliases = set(
        aliases
    )


    changed = True


    while changed:

        changed = False


        for stmt in ast.walk(
            tree
        ):

            value = None
            names = set()


            if isinstance(
                stmt,
                ast.Assign,
            ):

                for target in (
                    stmt.targets
                ):

                    names |= (
                        target_names(
                            target
                        )
                    )


                value = stmt.value


            elif isinstance(
                stmt,
                ast.AnnAssign,
            ):

                names |= (
                    target_names(
                        stmt.target
                    )
                )


                value = (
                    stmt.value
                )


            if (
                value
                is None
                or not names
            ):

                continue


            if expr_references_artifact(
                value,
                aliases=aliases,
                filename=filename,
            ):

                old_len = len(
                    aliases
                )


                aliases |= (
                    names
                )


                if (
                    len(
                        aliases
                    )
                    != old_len
                ):

                    changed = True


    return aliases


READ_CALL_NAMES = {
    "pd.read_pickle",
    "read_pickle",
}


def classify_tree_accesses(
    *,
    text: str,
    tree: ast.AST,
    aliases: set[str],
    filename: str,
    owner: str,
    source_ref: str,
) -> tuple[
    list[dict[str, Any]],
    list[dict[str, Any]],
]:

    reads = []
    writes = []


    for call in ast.walk(
        tree
    ):

        if not isinstance(
            call,
            ast.Call,
        ):
            continue


        name = dotted_name(
            call.func
        )


        # -------------------------------------------------------------
        # pd.read_pickle(PATH)
        # -------------------------------------------------------------

        if (
            name
            in READ_CALL_NAMES
            and call.args
        ):

            if expr_references_artifact(
                call.args[0],
                aliases=aliases,
                filename=filename,
            ):

                reads.append({
                    "source_ref":
                        source_ref,

                    "owner":
                        owner,

                    "line":
                        getattr(
                            call,
                            "lineno",
                            None,
                        ),

                    "call_name":
                        name,

                    "source":
                        node_source(
                            text,
                            call,
                        ),
                })


        # -------------------------------------------------------------
        # dataframe.to_pickle(PATH)
        # -------------------------------------------------------------

        if (
            name
            is not None
            and name.endswith(
                ".to_pickle"
            )
            and call.args
        ):

            if expr_references_artifact(
                call.args[0],
                aliases=aliases,
                filename=filename,
            ):

                writes.append({
                    "source_ref":
                        source_ref,

                    "owner":
                        owner,

                    "line":
                        getattr(
                            call,
                            "lineno",
                            None,
                        ),

                    "call_name":
                        name,

                    "receiver":
                        (
                            ast.unparse(
                                call.func.value
                            )
                            if isinstance(
                                call.func,
                                ast.Attribute,
                            )
                            else None
                        ),

                    "source":
                        node_source(
                            text,
                            call,
                        ),
                })


        # -------------------------------------------------------------
        # pickle.dump(obj, file_handle)
        #
        # We record it only if any argument directly references the path
        # alias. This catches some simple producer forms.
        # -------------------------------------------------------------

        if (
            name
            in {
                "pickle.dump",
                "dump",
            }
        ):

            if any(
                expr_references_artifact(
                    arg,
                    aliases=aliases,
                    filename=filename,
                )
                for arg in (
                    call.args
                )
            ):

                writes.append({
                    "source_ref":
                        source_ref,

                    "owner":
                        owner,

                    "line":
                        getattr(
                            call,
                            "lineno",
                            None,
                        ),

                    "call_name":
                        name,

                    "receiver":
                        None,

                    "source":
                        node_source(
                            text,
                            call,
                        ),
                })


    return (
        reads,
        writes,
    )


# =============================================================================
# 6. Notebook scanner with CROSS-CELL alias propagation
# =============================================================================

def normalize_cell_source(
    source_field: Any,
) -> str:

    if isinstance(
        source_field,
        list,
    ):

        return "".join(
            source_field
        )


    if isinstance(
        source_field,
        str,
    ):

        return source_field


    return ""


def parse_notebook_cell_safely(
    source: str,
) -> ast.Module | None:

    # Notebook magics make the cell invalid Python.
    #
    # For producer provenance, we do not silently execute/transform magics.
    # We only parse cells already valid as Python.
    try:

        return ast.parse(
            source
        )

    except SyntaxError:

        return None


def scan_notebook_for_artifact(
    path: Path,
    *,
    filename: str,
) -> dict[str, Any]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    aliases = set()


    matching_cells = []
    reads = []
    writes = []


    for cell_index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        if (
            cell.get(
                "cell_type"
            )
            != "code"
        ):

            continue


        source = normalize_cell_source(
            cell.get(
                "source",
                ""
            )
        )


        tree = parse_notebook_cell_safely(
            source
        )


        literal_hit = (
            filename
            in source
        )


        aliases_before = sorted(
            aliases
        )


        if (
            tree
            is not None
        ):

            aliases = update_aliases_from_tree(
                tree,
                aliases=aliases,
                filename=filename,
            )


            cell_reads, cell_writes = (
                classify_tree_accesses(
                    text=source,
                    tree=tree,
                    aliases=aliases,
                    filename=filename,
                    owner=(
                        f"cell[{cell_index}]"
                    ),
                    source_ref=(
                        f"{path.relative_to(ROOT)}"
                        f"#cell={cell_index}"
                    ),
                )
            )


            reads.extend(
                cell_reads
            )

            writes.extend(
                cell_writes
            )


        else:

            cell_reads = []
            cell_writes = []


        aliases_after = sorted(
            aliases
        )


        if (
            literal_hit
            or aliases_before
            != aliases_after
            or cell_reads
            or cell_writes
        ):

            matching_cells.append({
                "cell_index":
                    cell_index,

                "execution_count":
                    cell.get(
                        "execution_count"
                    ),

                "literal_filename_hit":
                    literal_hit,

                "python_ast_parseable":
                    (
                        tree
                        is not None
                    ),

                "aliases_before":
                    aliases_before,

                "aliases_after":
                    aliases_after,

                "source_sha256":
                    sha256_text(
                        source
                    ),

                "source":
                    source,
            })


    return {
        "path":
            str(
                path.relative_to(
                    ROOT
                )
            ),

        "file_sha256":
            sha256_file(
                path
            ),

        "final_aliases":
            sorted(
                aliases
            ),

        "matching_cells":
            matching_cells,

        "reads":
            reads,

        "writes":
            writes,
    }


# =============================================================================
# 7. Broad text-reference scanner
# =============================================================================

TEXT_SUFFIXES = {
    ".py",
    ".md",
    ".txt",
    ".json",
    ".yaml",
    ".yml",
    ".toml",
    ".log",
}


SKIP_PARTS = {
    ".git",
    ".venv",
    "venv",
    "__pycache__",
    ".pytest_cache",
    ".mypy_cache",
    ".ruff_cache",
    "node_modules",
}


def path_should_skip(
    path: Path,
) -> bool:

    return any(
        part
        in SKIP_PARTS
        for part in (
            path.parts
        )
    )


def text_context_blocks(
    text: str,
    *,
    needle: str,
    radius: int = 3,
    max_hits: int = 8,
) -> list[
    dict[str, Any]
]:

    lines = text.splitlines()


    hits = []


    for index, line in enumerate(
        lines
    ):

        if (
            needle
            not in line
        ):

            continue


        start = max(
            0,
            index - radius,
        )

        end = min(
            len(
                lines
            ),
            index + radius + 1,
        )


        hits.append({
            "hit_line":
                index + 1,

            "start_line":
                start + 1,

            "end_line":
                end,

            "text":
                "\n".join(
                    lines[
                        start:end
                    ]
                ),
        })


        if (
            len(
                hits
            )
            >= max_hits
        ):

            break


    return hits


def broad_text_reference_scan(
    filename: str,
) -> list[
    dict[str, Any]
]:

    records = []


    for path in (
        ROOT.rglob(
            "*"
        )
    ):

        if (
            not path.is_file()
            or path_should_skip(
                path
            )
        ):

            continue


        if (
            path.suffix.lower()
            not in TEXT_SUFFIXES
        ):

            continue


        # ipynb handled structurally below.
        if (
            path.suffix.lower()
            == ".ipynb"
        ):

            continue


        try:

            text = path.read_text(
                encoding="utf-8"
            )

        except Exception:

            continue


        if (
            filename
            not in text
        ):

            continue


        records.append({
            "path":
                str(
                    path.relative_to(
                        ROOT
                    )
                ),

            "file_sha256":
                sha256_file(
                    path
                ),

            "contexts":
                text_context_blocks(
                    text,
                    needle=filename,
                ),
        })


    return sorted(
        records,
        key=lambda item: (
            item[
                "path"
            ]
        ),
    )


# =============================================================================
# 8. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    B12_WRITER,
    ITER_RECIP,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3C-R1 "
    "MUTUAL-CONT-V1 PRODUCER DATAFLOW CLOSURE",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "TARGET ACTION:",
    TARGET_ACTION_FILENAME,
    flush=True,
)

print(
    "TARGET LOCK:",
    TARGET_LOCK_FILENAME,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Parent lock
# =============================================================================

parent_file_sha = sha256_file(
    PARENT
)


assert (
    parent_file_sha
    == EXPECTED_PARENT_FILE_SHA
)


parent_obj = json.loads(
    PARENT.read_text(
        encoding="utf-8"
    )
)


assert (
    parent_obj.get(
        "decision"
    )
    == "PASS"
)


assert contains_exact_string(
    parent_obj,
    EXPECTED_PARENT_CONTENT_SHA,
)


assert (
    parent_obj[
        "provider_readiness"
    ][
        "blocking_stage"
    ]
    == "MUTUAL_CONT_V1_GRAPH_LEVEL_RUNTIME"
)


print(
    "[PASS] PARENT_14_3B_R1_LOCK",
    flush=True,
)

print(
    "  file SHA    :",
    parent_file_sha,
    flush=True,
)

print(
    "  content SHA :",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Re-lock portable V1 action core
# =============================================================================

portable_sha = sha256_file(
    PORTABLE
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
)


portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_text,
    filename=str(
        PORTABLE
    ),
)


portable_functions = top_level_functions(
    portable_tree
)


assert (
    "reconstruct_mutual_cont_v1_actions"
    in portable_functions
)


v1_core_record = function_record(
    text=portable_text,
    fn=portable_functions[
        "reconstruct_mutual_cont_v1_actions"
    ],
)


assert (
    v1_core_record[
        "source_sha256"
    ]
    == EXPECTED_V1_ACTION_CORE_SHA
)


# Extract portable reference provider SHA.
reference_sha_matches = []


for stmt in (
    portable_tree.body
):

    targets = set()
    value = None


    if isinstance(
        stmt,
        ast.Assign,
    ):

        for target in (
            stmt.targets
        ):

            targets |= (
                target_names(
                    target
                )
            )


        value = stmt.value


    elif isinstance(
        stmt,
        ast.AnnAssign,
    ):

        targets |= (
            target_names(
                stmt.target
            )
        )

        value = stmt.value


    if (
        "REFERENCE_ASSOCIATION_PROVIDER_SHA256"
        not in targets
    ):

        continue


    assert (
        isinstance(
            value,
            ast.Constant,
        )
        and isinstance(
            value.value,
            str,
        )
    )


    reference_sha_matches.append(
        value.value
    )


assert (
    reference_sha_matches
    == [
        EXPECTED_ASSOCIATION_PROVIDER_SHA
    ]
)


print(
    "[PASS] PORTABLE_V1_CORE_RELOCK",
    flush=True,
)

print(
    "  module SHA:",
    portable_sha,
    flush=True,
)

print(
    "  action core SHA:",
    v1_core_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  provider reference SHA:",
    reference_sha_matches[
        0
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Re-lock association provider
# =============================================================================

iter_sha = sha256_file(
    ITER_RECIP
)


assert (
    iter_sha
    == EXPECTED_ITER_RECIP_FILE_SHA
)


iter_text = ITER_RECIP.read_text(
    encoding="utf-8"
)


iter_tree = ast.parse(
    iter_text,
    filename=str(
        ITER_RECIP
    ),
)


iter_functions = top_level_functions(
    iter_tree
)


assert (
    "fast_association_pair"
    in iter_functions
)


fast_assoc_record = function_record(
    text=iter_text,
    fn=iter_functions[
        "fast_association_pair"
    ],
)


assert (
    fast_assoc_record[
        "source_sha256"
    ]
    == EXPECTED_ASSOCIATION_PROVIDER_SHA
)


print(
    "[PASS] ASSOCIATION_PROVIDER_RELOCK",
    flush=True,
)

print(
    "  file SHA:",
    iter_sha,
    flush=True,
)

print(
    "  function SHA:",
    fast_assoc_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Re-lock historical generic graph mutation
# =============================================================================

b12_sha = sha256_file(
    B12_WRITER
)


assert (
    b12_sha
    == EXPECTED_B12_WRITER_SHA
)


b12_text = B12_WRITER.read_text(
    encoding="utf-8"
)


b12_tree = ast.parse(
    b12_text,
    filename=str(
        B12_WRITER
    ),
)


b12_functions = top_level_functions(
    b12_tree
)


assert (
    "add_actions"
    in b12_functions
)


add_actions_record = function_record(
    text=b12_text,
    fn=b12_functions[
        "add_actions"
    ],
)


assert (
    add_actions_record[
        "source_sha256"
    ]
    == EXPECTED_B12_ADD_ACTIONS_SHA
)


add_actions_source = (
    add_actions_record[
        "source"
    ]
)


for required_token in (
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_prob",
    "fv9.recovered_edge_dist",
    "g.add_edge",
    "validate_keys=True",
):

    assert (
        required_token
        in add_actions_source
    ), (
        f"Historical add_actions missing {required_token!r}."
    )


print(
    "[PASS] HISTORICAL_GENERIC_MUTATION_RELOCK",
    flush=True,
)

print(
    "  add_actions SHA:",
    add_actions_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Find exact lock JSON path(s)
# =============================================================================

lock_paths = find_exact_named_paths(
    TARGET_LOCK_FILENAME
)


print(
    "=" * 112,
    flush=True,
)

print(
    "V1 ACTION LOCK JSON DISCOVERY",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "matching lock files:",
    len(
        lock_paths
    ),
    flush=True,
)


for path in (
    lock_paths
):

    print(
        " ",
        path.relative_to(
            ROOT
        ),
        flush=True,
    )

    print(
        "    SHA:",
        sha256_file(
            path
        ),
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 14. Read provenance lock JSON only
#
# Historical *.pkl remains unopened.
# =============================================================================

lock_json_records = []


for path in (
    lock_paths
):

    try:

        obj = json.loads(
            path.read_text(
                encoding="utf-8"
            )
        )


        provenance_records = (
            provenance_relevant_flat_records(
                obj
            )
        )


        lock_json_records.append({
            "path":
                str(
                    path.relative_to(
                        ROOT
                    )
                ),

            "file_sha256":
                sha256_file(
                    path
                ),

            "json":
                obj,

            "provenance_records":
                provenance_records,
        })


    except Exception as exc:

        lock_json_records.append({
            "path":
                str(
                    path.relative_to(
                        ROOT
                    )
                ),

            "file_sha256":
                sha256_file(
                    path
                ),

            "json_parse_error":
                repr(
                    exc
                ),

            "provenance_records":
                [],
        })


print(
    "=" * 112,
    flush=True,
)

print(
    "V1 LOCK PROVENANCE METADATA",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


if not lock_json_records:

    print(
        "  <no lock JSON physically present>",
        flush=True,
    )


for record in (
    lock_json_records
):

    print(
        record[
            "path"
        ],
        flush=True,
    )

    print(
        "  file SHA:",
        record[
            "file_sha256"
        ],
        flush=True,
    )

    print(
        "  provenance-like fields:",
        len(
            record[
                "provenance_records"
            ]
        ),
        flush=True,
    )


    for item in (
        record[
            "provenance_records"
        ][
            :120
        ]
    ):

        print(
            "   ",
            item[
                "path"
            ],
            "=",
            repr(
                item[
                    "value"
                ]
            ),
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 15. Notebook producer search
# =============================================================================

notebook_records = []


notebook_root = (
    ROOT / "notebooks"
)


if notebook_root.exists():

    for path in sorted(
        notebook_root.rglob(
            "*.ipynb"
        )
    ):

        try:

            raw = path.read_text(
                encoding="utf-8"
            )

        except Exception:

            continue


        # Fast rejection.
        if (
            TARGET_ACTION_FILENAME
            not in raw
        ):

            continue


        record = scan_notebook_for_artifact(
            path,
            filename=TARGET_ACTION_FILENAME,
        )


        notebook_records.append(
            record
        )


notebook_write_records = []


for record in (
    notebook_records
):

    for write in (
        record[
            "writes"
        ]
    ):

        notebook_write_records.append({
            "notebook":
                record[
                    "path"
                ],

            "notebook_file_sha256":
                record[
                    "file_sha256"
                ],

            **write,
        })


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK V1 ARTIFACT PROVENANCE SEARCH",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "notebooks mentioning target:",
    len(
        notebook_records
    ),
    flush=True,
)

print(
    "detected notebook write sinks:",
    len(
        notebook_write_records
    ),
    flush=True,
)

print(
    flush=True,
)


for record in (
    notebook_records
):

    print(
        record[
            "path"
        ],
        flush=True,
    )

    print(
        "  notebook SHA:",
        record[
            "file_sha256"
        ],
        flush=True,
    )

    print(
        "  final aliases:",
        record[
            "final_aliases"
        ],
        flush=True,
    )

    print(
        "  reads:",
        len(
            record[
                "reads"
            ]
        ),
        flush=True,
    )

    print(
        "  writes:",
        len(
            record[
                "writes"
            ]
        ),
        flush=True,
    )


    for cell in (
        record[
            "matching_cells"
        ]
    ):

        print(
            f"    cell[{cell['cell_index']}] "
            f"exec={cell['execution_count']} "
            f"parseable={cell['python_ast_parseable']} "
            f"literal_hit={cell['literal_filename_hit']} "
            f"SHA={cell['source_sha256']}",
            flush=True,
        )


        # Full source is printed only for cells that actually mention the
        # filename or contain a detected write.
        cell_ref = (
            f"{record['path']}"
            f"#cell={cell['cell_index']}"
        )


        cell_has_write = any(
            write[
                "source_ref"
            ]
            == cell_ref
            for write in (
                record[
                    "writes"
                ]
            )
        )


        if (
            cell[
                "literal_filename_hit"
            ]
            or cell_has_write
        ):

            print(
                "      SOURCE:",
                flush=True,
            )

            print(
                "        "
                + cell[
                    "source"
                ].replace(
                    "\n",
                    "\n        ",
                ),
                flush=True,
            )


    for write in (
        record[
            "writes"
        ]
    ):

        print(
            "    WRITE:",
            write[
                "source_ref"
            ],
            flush=True,
        )

        print(
            "      receiver:",
            write[
                "receiver"
            ],
            flush=True,
        )

        print(
            "      source:",
            write[
                "source"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 16. Broad repo text reference search
# =============================================================================

action_text_refs = broad_text_reference_scan(
    TARGET_ACTION_FILENAME
)


lock_text_refs = broad_text_reference_scan(
    TARGET_LOCK_FILENAME
)


print(
    "=" * 112,
    flush=True,
)

print(
    "BROAD TEXT PROVENANCE REFERENCES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "action-artifact text refs:",
    len(
        action_text_refs
    ),
    flush=True,
)

print(
    "lock-artifact text refs:",
    len(
        lock_text_refs
    ),
    flush=True,
)

print(
    flush=True,
)


for record in (
    action_text_refs
):

    print(
        "ACTION REF:",
        record[
            "path"
        ],
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "file_sha256"
        ],
        flush=True,
    )


    for context in (
        record[
            "contexts"
        ][
            :4
        ]
    ):

        print(
            f"  L{context['start_line']}"
            f"-L{context['end_line']}:",
            flush=True,
        )

        print(
            "    "
            + context[
                "text"
            ].replace(
                "\n",
                "\n    ",
            ),
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 17. Extract explicit provenance path candidates from lock JSON
# =============================================================================

lock_path_candidates = []


for lock_record in (
    lock_json_records
):

    for item in (
        lock_record[
            "provenance_records"
        ]
    ):

        value = (
            item[
                "value"
            ]
        )


        if not isinstance(
            value,
            str,
        ):

            continue


        lower = (
            value.lower()
        )


        if (
            ".py"
            not in lower
            and ".ipynb"
            not in lower
        ):

            continue


        lock_path_candidates.append({
            "lock_path":
                lock_record[
                    "path"
                ],

            "metadata_path":
                item[
                    "path"
                ],

            "value":
                value,
        })


print(
    "=" * 112,
    flush=True,
)

print(
    "LOCK-METADATA SOURCE PATH CANDIDATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


if not lock_path_candidates:

    print(
        "  <none>",
        flush=True,
    )


for candidate in (
    lock_path_candidates
):

    print(
        " ",
        candidate[
            "metadata_path"
        ],
        "=",
        repr(
            candidate[
                "value"
            ]
        ),
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 18. Resolve metadata source paths against current repo
# =============================================================================

resolved_metadata_paths = []


for candidate in (
    lock_path_candidates
):

    raw = (
        candidate[
            "value"
        ]
    )


    possible_paths = []


    # Exact relative string.
    possible_paths.append(
        ROOT / raw
    )


    # Basename fallback.
    possible_paths.extend(
        ROOT.rglob(
            Path(
                raw
            ).name
        )
    )


    seen = set()


    for path in (
        possible_paths
    ):

        try:

            resolved = (
                path.resolve()
            )

        except Exception:

            continue


        if (
            resolved
            in seen
        ):

            continue


        seen.add(
            resolved
        )


        if not resolved.exists():

            continue


        try:

            rel = (
                resolved.relative_to(
                    ROOT
                )
            )

        except ValueError:

            continue


        resolved_metadata_paths.append({
            **candidate,

            "resolved_path":
                str(
                    rel
                ),

            "resolved_file_sha256":
                (
                    sha256_file(
                        resolved
                    )
                    if resolved.is_file()
                    else None
                ),
        })


print(
    "Resolved metadata paths:",
    len(
        resolved_metadata_paths
    ),
    flush=True,
)


for record in (
    resolved_metadata_paths
):

    print(
        " ",
        record[
            "metadata_path"
        ],
        "->",
        record[
            "resolved_path"
        ],
        flush=True,
    )

    print(
        "    SHA:",
        record[
            "resolved_file_sha256"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 19. Conservative producer-evidence classification
# =============================================================================

# Strongest evidence:
#   actual source write sink in notebook.
#
# The previous 14.3C already established no Python writer in scripts/src/tests.
# R1 does not assume a metadata mention alone proves a producer.

if (
    len(
        notebook_write_records
    )
    >= 1
):

    producer_status = (
        "PRODUCER_SOURCE_FOUND_IN_NOTEBOOK"
    )


elif (
    len(
        resolved_metadata_paths
    )
    >= 1
):

    producer_status = (
        "SOURCE_PATH_REFERENCED_BY_LOCK_METADATA_"
        "BUT_WRITE_SINK_NOT_YET_PROVEN"
    )


elif (
    lock_json_records
):

    producer_status = (
        "LOCK_METADATA_PRESENT_BUT_"
        "PRODUCER_SOURCE_NOT_IDENTIFIED"
    )


else:

    producer_status = (
        "PRODUCER_SOURCE_NOT_PRESENT_IN_CURRENT_REPO"
    )


print(
    "=" * 112,
    flush=True,
)

print(
    "V1 PRODUCER CLOSURE STATUS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    producer_status,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 20. Determine whether implementation authority is now complete
# =============================================================================

association_identity_ready = (
    fast_assoc_record[
        "source_sha256"
    ]
    == EXPECTED_ASSOCIATION_PROVIDER_SHA
)


action_core_ready = (
    v1_core_record[
        "source_sha256"
    ]
    == EXPECTED_V1_ACTION_CORE_SHA
)


mutation_contract_ready = (
    add_actions_record[
        "source_sha256"
    ]
    == EXPECTED_B12_ADD_ACTIONS_SHA
)


producer_source_proven = (
    producer_status
    == "PRODUCER_SOURCE_FOUND_IN_NOTEBOOK"
)


v1_runtime_authority_ready = (
    association_identity_ready
    and action_core_ready
    and mutation_contract_ready
    and producer_source_proven
)


readiness = {
    "association_provider_identity_ready":
        association_identity_ready,

    "portable_v1_action_core_ready":
        action_core_ready,

    "historical_graph_mutation_contract_ready":
        mutation_contract_ready,

    "producer_source_proven":
        producer_source_proven,

    "producer_status":
        producer_status,

    "v1_graph_runtime_source_authority_ready":
        v1_runtime_authority_ready,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "V1 GRAPH-RUNTIME IMPLEMENTATION READINESS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        readiness,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 21. Formal semantic gates
#
# These gates test whether the CLOSURE PROCEDURE completed correctly.
#
# producer_source_proven is intentionally NOT a mandatory PASS gate.
# =============================================================================

semantic_gates = {
    "parent_14_3b_r1_locked":
        (
            parent_file_sha
            == EXPECTED_PARENT_FILE_SHA
        ),

    "portable_v1_core_locked":
        action_core_ready,

    "association_provider_identity_locked":
        association_identity_ready,

    "historical_mutation_contract_locked":
        mutation_contract_ready,

    "lock_json_search_completed":
        True,

    "notebook_producer_search_completed":
        True,

    "broad_text_provenance_search_completed":
        True,

    "lock_metadata_path_resolution_completed":
        True,

    "producer_status_classified":
        (
            producer_status
            in {
                "PRODUCER_SOURCE_FOUND_IN_NOTEBOOK",
                "SOURCE_PATH_REFERENCED_BY_LOCK_METADATA_"
                "BUT_WRITE_SINK_NOT_YET_PROVEN",
                "LOCK_METADATA_PRESENT_BUT_"
                "PRODUCER_SOURCE_NOT_IDENTIFIED",
                "PRODUCER_SOURCE_NOT_PRESENT_IN_CURRENT_REPO",
            }
        ),

    "historical_action_pickle_not_opened":
        True,

    "no_source_mutation":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3C-R1 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate
    for gate, ok in (
        semantic_gates.items()
    )
    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3C-R1 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 22. Decide next stage WITHOUT guessing
# =============================================================================

if (
    v1_runtime_authority_ready
):

    next_stage = (
        "NOTEBOOK14_14_3D_"
        "MUTUAL_CONT_V1_GRAPH_RUNTIME_IMPLEMENTATION"
    )


elif (
    producer_status
    == (
        "SOURCE_PATH_REFERENCED_BY_LOCK_METADATA_"
        "BUT_WRITE_SINK_NOT_YET_PROVEN"
    )
):

    next_stage = (
        "NOTEBOOK14_14_3C_R2_"
        "LOCK_REFERENCED_SOURCE_PRODUCER_VERIFICATION"
    )


else:

    next_stage = (
        "NOTEBOOK14_14_3C_R2_"
        "V1_SOURCE_CONTRACT_RECONSTRUCTION_FROM_LOCK_AND_HISTORICAL_CONVERGENCE"
    )


# =============================================================================
# 23. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3C_R1",

    "title":
        "MUTUAL_CONT_V1_PRODUCER_DATAFLOW_CLOSURE",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_PROVENANCE_DISCOVERY_ONLY",

    "parent_14_3b_r1": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            parent_file_sha,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "failed_predecessor": {
        "stage":
            "NOTEBOOK14_14_3C",

        "failure": (
            "Producer search was limited to scripts/src/tests Python files "
            "and required at least one write sink. Eight consumers were "
            "found but zero writers, so the cell stopped before notebook "
            "and lock-metadata provenance could be searched."
        ),

        "successful_evidence_before_failure": [
            "portable V1 action core locked",
            "association-provider source identity locked",
            "historical generic V1 mutation locked",
            "historical V2-confirmatory V1 baseline locked",
            "historical Round2 V1 mutation locked",
            "historical post-V1 topology locked",
            "historical edge-mutation convergence established",
        ],

        "source_mutation":
            False,

        "formal_artifact_written":
            False,
    },

    "portable_authority": {
        "path":
            str(
                PORTABLE_REL
            ),

        "file_sha256":
            portable_sha,

        "v1_action_core":
            v1_core_record,

        "reference_association_provider_sha256":
            reference_sha_matches[
                0
            ],
    },

    "association_provider_authority": {
        "path":
            str(
                ITER_RECIP_REL
            ),

        "file_sha256":
            iter_sha,

        "fast_association_pair":
            fast_assoc_record,
    },

    "historical_graph_mutation_authority": {
        "path":
            str(
                B12_WRITER_REL
            ),

        "file_sha256":
            b12_sha,

        "add_actions":
            add_actions_record,
    },

    "historical_action_artifact": {
        "filename":
            TARGET_ACTION_FILENAME,

        "pickle_opened":
            False,
    },

    "lock_artifact_discovery": {
        "filename":
            TARGET_LOCK_FILENAME,

        "physical_path_count":
            len(
                lock_paths
            ),

        "records":
            lock_json_records,

        "source_path_candidates":
            lock_path_candidates,

        "resolved_source_paths":
            resolved_metadata_paths,
    },

    "notebook_provenance_search": {
        "notebooks_mentioning_target":
            len(
                notebook_records
            ),

        "records":
            notebook_records,

        "write_sink_count":
            len(
                notebook_write_records
            ),

        "write_sinks":
            notebook_write_records,
    },

    "broad_text_search": {
        "action_artifact_references":
            action_text_refs,

        "lock_artifact_references":
            lock_text_refs,
    },

    "producer_closure": {
        "status":
            producer_status,

        "producer_source_proven":
            producer_source_proven,
    },

    "readiness":
        readiness,

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "v1_action_core":
            "SOURCE_LOCKED",

        "v1_association_provider":
            "SOURCE_LOCKED",

        "historical_v1_graph_mutation":
            "SOURCE_LOCKED",

        "v1_producer_provenance":
            producer_status,

        "v1_graph_runtime_source_authority":
            (
                "READY_FOR_IMPLEMENTATION"
                if v1_runtime_authority_ready
                else "NOT_YET_CLOSED"
            ),

        "v1_graph_runtime_implemented":
            False,

        "post_v2_graph_provider_implemented":
            False,

        "post_v2_graph_structural_fidelity":
            "NOT_YET_VERIFIED",

        "b12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "historical_v1_action_pickle_opened":
            False,

        "source_mutation":
            False,

        "gt_read":
            False,

        "evaluator_read":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No historical V1 action pickle was opened.",
        "No V1 graph runtime implementation claim.",
        "No post-V2 graph provider implementation claim.",
        "No post-V2 structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance evidence.",
        "No fresh-holdout evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


artifact_content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        artifact_content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3C-R1 artifact exists but differs from "
        "the current exact provenance closure."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 24. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3C-R1 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "LOCK JSON:",
    flush=True,
)

print(
    "  physical copies:",
    len(
        lock_paths
    ),
    flush=True,
)


for path in (
    lock_paths
):

    print(
        " ",
        path.relative_to(
            ROOT
        ),
        flush=True,
    )


print(
    flush=True,
)


print(
    "NOTEBOOK PRODUCER SEARCH:",
    flush=True,
)

print(
    "  notebooks mentioning artifact:",
    len(
        notebook_records
    ),
    flush=True,
)

print(
    "  write sinks:",
    len(
        notebook_write_records
    ),
    flush=True,
)


for write in (
    notebook_write_records
):

    print(
        " ",
        write[
            "notebook"
        ],
        write[
            "source_ref"
        ],
        flush=True,
    )

    print(
        "    receiver:",
        write[
            "receiver"
        ],
        flush=True,
    )

    print(
        "    source:",
        write[
            "source"
        ],
        flush=True,
    )


print(
    flush=True,
)


print(
    "PRODUCER CLOSURE:",
    flush=True,
)

print(
    "  status:",
    producer_status,
    flush=True,
)

print(
    "  source proven:",
    producer_source_proven,
    flush=True,
)

print(
    flush=True,
)


print(
    "V1 GRAPH-RUNTIME AUTHORITY:",
    flush=True,
)

print(
    "  association provider:",
    association_identity_ready,
    flush=True,
)

print(
    "  action core:",
    action_core_ready,
    flush=True,
)

print(
    "  graph mutation:",
    mutation_contract_ready,
    flush=True,
)

print(
    "  producer source:",
    producer_source_proven,
    flush=True,
)

print(
    "  READY:",
    v1_runtime_authority_ready,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    artifact_content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  V1 producer provenance closure = COMPLETE SEARCH",
    flush=True,
)

print(
    "  historical V1 pickle opened = NO",
    flush=True,
)

print(
    "  V1 graph runtime implementation = NOT YET",
    flush=True,
)

print(
    "  post-V2 provider implementation = NOT YET",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3C-R2 — MUTUAL-CONT-V1 FALSE-WRITER CORRECTION
#             + CORRECTED AUTHORITY EXTRACTION
#
# SUCCESSOR TO
# ------------
# 14.3C:
#   searched scripts/src/tests and found historical consumers but no producer.
#
# 14.3C-R1:
#   broadened search to lock JSON + notebooks + repo text.
#
# Critical discovery:
#
#   historical Notebook13 / 13.3B3 authority:
#
#       Notebook12 cell 4
#       SHA b334a75e...
#
#   was NOT the V1 action writer.
#
# Notebook12 cell 4:
#
#   READS:
#       stage12_fold0_mutual_cont_v1_actions_locked.pkl
#
#   WRITES:
#       stage12_post_v1_linking_residuals.pkl
#       stage12_post_v1_linking_residual_summary.json
#
# Therefore:
#
#   old 13.3B3 "V1 exact writer cell" authority
#   = FALSE-WRITER PROVENANCE
#
#
# IMPORTANT
# ---------
# Notebook13 later contains explicit repair/audit machinery:
#
#   FAILED_FALSE_WRITER
#   VIABLE_WRITERS
#   CORRECTED_AUTHORITY
#   V1_AUTHORITY
#   V1_ARTIFACT_WRITER_CELLS
#
# and:
#
#   CELL70_ACTION_SCHEMA_COMPLETE
#   CELL70_CONFIG_FILTER_PRESENT
#   CELL70_CONFIG_UNIQUE
#   CELL70_CONFIG_VALUE
#   CELL70_HAS_LOCK_STRUCTURE
#   CELL70_HISTORICAL_SORT_ORDER
#   CELL70_RUNTIME_SORT_ORDER
#   CELL70_LOCK_COLS
#   CELL70_ROLE
#
# This stage recovers and freezes the LATER corrected authority.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical V1 pickle read
#   - graph loading
#   - GT
#   - evaluator
#   - performance metric
#   - structural fidelity claim
#   - Git operation
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3c_r1_"
    "mutual_cont_v1_producer_dataflow_closure_v1.json"
)

PARENT = ROOT / PARENT_REL


NOTEBOOK12_REL = Path(
    "notebooks/12_residual_edge_recovery.ipynb"
)

NOTEBOOK12 = ROOT / NOTEBOOK12_REL


NOTEBOOK13_REL = Path(
    "notebooks/13_stage12_handoff_and_fresh_holdout_readiness.ipynb"
)

NOTEBOOK13 = ROOT / NOTEBOOK13_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


CORRECTED_ARCH_REL = Path(
    "reports/"
    "notebook13_13_3c_r2_"
    "corrected_authority_model_and_architecture_refreeze_v1.json"
)

CORRECTED_ARCH = ROOT / CORRECTED_ARCH_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3c_r2_"
    "v1_false_writer_correction_and_"
    "corrected_authority_extraction_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PORTABLE_SHA = (
    "a1fe0c9e5cd5e20a2ae5921bd4b4c55d078022c3348d5945bffa501ac3ff36ed"
)


EXPECTED_NOTEBOOK12_SHA = (
    "510517ab4c3909d01b5c8d58dd90b23879d078e2009d2a12935fdcea75143a50"
)


EXPECTED_NOTEBOOK13_SHA = (
    "e7248f78388f8388e071bebd0c2ab5123f0d9703e55d4977b32dad4d961cfc55"
)


EXPECTED_FALSE_WRITER_CELL_INDEX = (
    4
)

EXPECTED_FALSE_WRITER_CELL_SHA = (
    "b334a75e1c47b7cdb69cd38c99b955815798161a25db6278c6ddd8662c249869"
)


EXPECTED_CORRECTED_ARCH_FILE_SHA = (
    "e7d307f988b97717a4250534179fc07f90408b29dedccbcc4fc9b96567c7f4c8"
)

EXPECTED_CORRECTED_ARCH_CONTENT_SHA = (
    "49f533d19e2354735c9faef38179625c058732ba4838e1a7544aefd1733fc3f4"
)


TARGET_ACTION_FILENAME = (
    "stage12_fold0_mutual_cont_v1_actions_locked.pkl"
)

FALSE_WRITER_REAL_OUTPUT = (
    "stage12_post_v1_linking_residuals.pkl"
)


# =============================================================================
# 2. Notebook13 correction markers
# =============================================================================

CORRECTION_MARKERS = [
    "FAILED_FALSE_WRITER",
    "VIABLE_WRITERS",
    "CORRECTED_AUTHORITY",
    "V1_AUTHORITY",
    "V1_ARTIFACT_WRITER_CELLS",
    "writer_resolved",
    "artifact_writer_verified",
]


CELL70_MARKERS = [
    "CELL70_ACTION_SCHEMA_COMPLETE",
    "CELL70_CONFIG_FILTER_PRESENT",
    "CELL70_CONFIG_FILTER_RECORDS",
    "CELL70_CONFIG_UNIQUE",
    "CELL70_CONFIG_VALUE",
    "CELL70_HAS_LOCK_STRUCTURE",
    "CELL70_HISTORICAL_SORT_ORDER",
    "CELL70_RUNTIME_SORT_ORDER",
    "CELL70_LOCK_COLS",
    "CELL70_ROLE",
]


ALL_AUDIT_MARKERS = (
    CORRECTION_MARKERS
    + CELL70_MARKERS
)


# =============================================================================
# 3. Helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    obj: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def notebook_cell_source(
    notebook: dict[str, Any],
    index: int,
) -> str:

    cells = notebook.get(
        "cells",
        [],
    )


    assert (
        0
        <= index
        < len(
            cells
        )
    ), (
        index,
        len(
            cells
        ),
    )


    cell = cells[
        index
    ]


    source = cell.get(
        "source",
        "",
    )


    if isinstance(
        source,
        list,
    ):

        return "".join(
            source
        )


    return str(
        source
    )


def parse_python(
    source: str,
) -> ast.Module | None:

    try:

        return ast.parse(
            source
        )

    except SyntaxError:

        return None


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def node_source(
    source: str,
    node: ast.AST,
) -> str:

    result = ast.get_source_segment(
        source,
        node,
    )


    assert (
        result
        is not None
    )


    return result


def target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    def visit(
        item: ast.AST,
    ):

        if isinstance(
            item,
            ast.Name,
        ):

            result.add(
                item.id
            )


        elif isinstance(
            item,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for child in (
                item.elts
            ):

                visit(
                    child
                )


        elif isinstance(
            item,
            ast.Starred,
        ):

            visit(
                item.value
            )


    visit(
        node
    )


    return result


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in (
            obj.items()
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


# =============================================================================
# 4. Formal-parent verifier
# =============================================================================

def verify_formal_artifact(
    path: Path,
) -> dict[str, Any]:

    assert path.exists(), (
        path
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    ), (
        "Parent artifact is not PASS."
    )


    stored_content_sha = (
        obj.get(
            "content_sha256"
        )
    )


    assert (
        isinstance(
            stored_content_sha,
            str,
        )
        and len(
            stored_content_sha
        )
        == 64
    )


    payload = dict(
        obj
    )


    payload.pop(
        "content_sha256"
    )


    recomputed = (
        canonical_json_sha(
            payload
        )
    )


    assert (
        recomputed
        == stored_content_sha
    ), (
        "Parent artifact content SHA does not verify.\n"
        f"stored    ={stored_content_sha}\n"
        f"recomputed={recomputed}"
    )


    return {
        "object":
            obj,

        "file_sha256":
            sha256_file(
                path
            ),

        "content_sha256":
            stored_content_sha,
    }


# =============================================================================
# 5. Exact per-cell path alias / I/O analysis
# =============================================================================

def literal_strings(
    node: ast.AST,
) -> set[str]:

    return {
        item.value

        for item in ast.walk(
            node
        )

        if (
            isinstance(
                item,
                ast.Constant,
            )
            and isinstance(
                item.value,
                str,
            )
        )
    }


def direct_artifact_path_aliases(
    tree: ast.Module,
    *,
    filename: str,
) -> set[str]:

    aliases = set()


    for stmt in (
        tree.body
    ):

        names = set()
        value = None


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                names |= (
                    target_names(
                        target
                    )
                )


            value = stmt.value


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            names |= (
                target_names(
                    stmt.target
                )
            )


            value = stmt.value


        if (
            value
            is None
        ):

            continue


        if (
            filename
            in literal_strings(
                value
            )
        ):

            aliases |= (
                names
            )


    return aliases


def expression_references_path(
    node: ast.AST,
    *,
    aliases: set[str],
    filename: str,
) -> bool:

    for item in ast.walk(
        node
    ):

        if (
            isinstance(
                item,
                ast.Name,
            )
            and item.id
            in aliases
        ):

            return True


        if (
            isinstance(
                item,
                ast.Constant,
            )
            and isinstance(
                item.value,
                str,
            )
            and item.value
            == filename
        ):

            return True


    return False


def exact_cell_artifact_io(
    source: str,
    *,
    filename: str,
) -> dict[str, Any]:

    tree = parse_python(
        source
    )


    assert (
        tree
        is not None
    ), (
        "Target authority cell is not valid Python."
    )


    aliases = direct_artifact_path_aliases(
        tree,
        filename=filename,
    )


    reads = []
    writes = []
    all_writes = []


    for call in ast.walk(
        tree
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        # -------------------------------------------------------------
        # read_pickle
        # -------------------------------------------------------------

        if (
            name
            in {
                "pd.read_pickle",
                "read_pickle",
            }
            and call.args
        ):

            if expression_references_path(
                call.args[
                    0
                ],
                aliases=aliases,
                filename=filename,
            ):

                reads.append({
                    "line":
                        call.lineno,

                    "call_name":
                        name,

                    "source":
                        node_source(
                            source,
                            call,
                        ),
                })


        # -------------------------------------------------------------
        # *.to_pickle(PATH)
        # -------------------------------------------------------------

        if (
            name
            is not None
            and name.endswith(
                ".to_pickle"
            )
            and call.args
        ):

            sink = {
                "line":
                    call.lineno,

                "call_name":
                    name,

                "receiver":
                    (
                        ast.unparse(
                            call.func.value
                        )
                        if isinstance(
                            call.func,
                            ast.Attribute,
                        )
                        else None
                    ),

                "destination":
                    ast.unparse(
                        call.args[
                            0
                        ]
                    ),

                "source":
                    node_source(
                        source,
                        call,
                    ),
            }


            all_writes.append(
                sink
            )


            if expression_references_path(
                call.args[
                    0
                ],
                aliases=aliases,
                filename=filename,
            ):

                writes.append(
                    sink
                )


    return {
        "aliases":
            sorted(
                aliases
            ),

        "reads":
            reads,

        "target_writes":
            writes,

        "all_pickle_writes":
            all_writes,
    }


# =============================================================================
# 6. Assignment extractor for Notebook13 audit cells
# =============================================================================

TRACKED_ASSIGNMENTS = set(
    ALL_AUDIT_MARKERS
    + [
        "COMPLETE_V1",
        "FULL_V1_CANDIDATES",
        "V1_RANK_PAIR_CANDIDATES",
        "WRITER_COUNT",
        "WRITER_CANDIDATES",
        "EXACT_WRITER_OCCURRENCES",
        "ALL_EXACT_WRITERS",
        "ARTIFACT_WRITER_VERIFIED",
        "CORRECTED_AUTHORITY_MODEL",
        "AUTHORITY_MODEL",
        "CANONICAL_RUNTIME_GENERATORS",
        "RECONSTRUCTION_AUTHORITY",
    ]
)


def tracked_assignments(
    source: str,
) -> list[
    dict[str, Any]
]:

    tree = parse_python(
        source
    )


    if (
        tree
        is None
    ):

        return []


    records = []


    for stmt in (
        tree.body
    ):

        names = set()
        value = None


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                names |= (
                    target_names(
                        target
                    )
                )


            value = (
                stmt.value
            )


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            names |= (
                target_names(
                    stmt.target
                )
            )


            value = (
                stmt.value
            )


        if (
            value
            is None
        ):

            continue


        for name in sorted(
            names
            & TRACKED_ASSIGNMENTS
        ):

            value_source = (
                ast.unparse(
                    value
                )
            )


            try:

                literal = (
                    ast.literal_eval(
                        value
                    )
                )


                literal_ok = True


            except Exception:

                literal = None
                literal_ok = False


            records.append({
                "name":
                    name,

                "line":
                    stmt.lineno,

                "value_source":
                    value_source,

                "literal_eval_ok":
                    literal_ok,

                "literal_value":
                    literal,
            })


    return records


# =============================================================================
# 7. Context extraction around marker hits
# =============================================================================

def source_contexts(
    source: str,
    *,
    markers: list[str],
    radius: int = 4,
) -> list[
    dict[str, Any]
]:

    lines = source.splitlines()


    result = []


    for index, line in enumerate(
        lines
    ):

        hits = [
            marker

            for marker in (
                markers
            )

            if marker in line
        ]


        if not hits:

            continue


        start = max(
            0,
            index - radius,
        )


        end = min(
            len(
                lines
            ),
            index + radius + 1,
        )


        block = "\n".join(
            lines[
                start:end
            ]
        )


        result.append({
            "hit_line":
                index + 1,

            "markers":
                hits,

            "start_line":
                start + 1,

            "end_line":
                end,

            "block_sha256":
                sha256_text(
                    block
                ),

            "text":
                block,
        })


    return result


# =============================================================================
# 8. Scan Notebook13 later correction/audit cells
# =============================================================================

def scan_notebook13_audit_cells(
    notebook: dict[str, Any],
) -> list[
    dict[str, Any]
]:

    records = []


    for index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        if (
            cell.get(
                "cell_type"
            )
            != "code"
        ):

            continue


        source = notebook_cell_source(
            notebook,
            index,
        )


        marker_hits = [
            marker

            for marker in (
                ALL_AUDIT_MARKERS
            )

            if marker in source
        ]


        if not marker_hits:

            continue


        records.append({
            "cell_index":
                index,

            "execution_count":
                cell.get(
                    "execution_count"
                ),

            "source_sha256":
                sha256_text(
                    source
                ),

            "markers":
                marker_hits,

            "assignments":
                tracked_assignments(
                    source
                ),

            "contexts":
                source_contexts(
                    source,
                    markers=marker_hits,
                ),
        })


    return records


# =============================================================================
# 9. Search formal Notebook13 reports for corrected-writer evidence
# =============================================================================

REPORT_KEYWORDS = [
    "FAILED_FALSE_WRITER",
    "VIABLE_WRITERS",
    "CORRECTED_AUTHORITY",
    "CELL70_ACTION_SCHEMA_COMPLETE",
    "CELL70_ROLE",
    TARGET_ACTION_FILENAME,
]


def scan_report_jsons() -> list[
    dict[str, Any]
]:

    reports_root = (
        ROOT / "reports"
    )


    records = []


    if not reports_root.exists():

        return records


    for path in sorted(
        reports_root.rglob(
            "*.json"
        )
    ):

        try:

            raw = path.read_text(
                encoding="utf-8"
            )


        except Exception:

            continue


        hits = [
            keyword

            for keyword in (
                REPORT_KEYWORDS
            )

            if keyword in raw
        ]


        if not hits:

            continue


        try:

            obj = json.loads(
                raw
            )


        except Exception:

            continue


        flat = flatten_json(
            obj
        )


        relevant = []


        for leaf in (
            flat
        ):

            path_lower = (
                leaf[
                    "path"
                ].lower()
            )


            value = (
                leaf[
                    "value"
                ]
            )


            value_text = (
                str(
                    value
                )
                if value
                is not None
                else ""
            )


            value_lower = (
                value_text.lower()
            )


            if (
                "writer"
                in path_lower
                or "authority"
                in path_lower
                or "v1"
                in path_lower
                or "cell70"
                in path_lower
                or TARGET_ACTION_FILENAME.lower()
                in value_lower
                or "mutual_cont_v1"
                in value_lower
            ):

                relevant.append(
                    leaf
                )


        records.append({
            "path":
                str(
                    path.relative_to(
                        ROOT
                    )
                ),

            "file_sha256":
                sha256_file(
                    path
                ),

            "keyword_hits":
                hits,

            "relevant_leaf_records":
                relevant[
                    :300
                ],
        })


    return records


# =============================================================================
# 10. Corrected architecture V1 extraction
# =============================================================================

def corrected_arch_v1_records(
    obj: Any,
) -> list[
    dict[str, Any]
]:

    flat = flatten_json(
        obj
    )


    records = []


    for leaf in (
        flat
    ):

        path_lower = (
            leaf[
                "path"
            ].lower()
        )


        value = (
            leaf[
                "value"
            ]
        )


        value_text = (
            str(
                value
            )
        )


        value_lower = (
            value_text.lower()
        )


        if (
            "mutual_cont_v1"
            in path_lower
            or "mutual_cont_v1"
            in value_lower
            or "mutual-cont-v1"
            in value_lower
            or "_reconstruct_portable_mutual_cont_v1"
            in value_lower
            or (
                "v1"
                in path_lower
                and (
                    "authority"
                    in path_lower
                    or "runtime"
                    in path_lower
                    or "source"
                    in path_lower
                )
            )
        ):

            records.append(
                leaf
            )


    return records


# =============================================================================
# 11. Independent Notebook12 cell70 inspection
# =============================================================================

def inspect_notebook12_cell70(
    notebook: dict[str, Any],
) -> dict[str, Any]:

    index = 70


    assert (
        index
        < len(
            notebook[
                "cells"
            ]
        )
    ), (
        "Notebook12 does not contain cell 70."
    )


    cell = notebook[
        "cells"
    ][
        index
    ]


    source = notebook_cell_source(
        notebook,
        index,
    )


    tree = parse_python(
        source
    )


    strings = (
        literal_strings(
            tree
        )
        if tree
        is not None
        else set()
    )


    calls = []


    if (
        tree
        is not None
    ):

        for call in ast.walk(
            tree
        ):

            if not isinstance(
                call,
                ast.Call,
            ):

                continue


            name = dotted_name(
                call.func
            )


            calls.append(
                name
                or "<dynamic-call>"
            )


    markers = {
        "mentions_target_action_filename":
            (
                TARGET_ACTION_FILENAME
                in source
            ),

        "mentions_source_pred_node_id":
            (
                "source_pred_node_id"
                in source
            ),

        "mentions_target_pred_node_id":
            (
                "target_pred_node_id"
                in source
            ),

        "mentions_candidate_prob":
            (
                "candidate_prob"
                in source
            ),

        "mentions_argmax":
            (
                "argmax"
                in source
            ),

        "mentions_sort_values":
            (
                "sort_values"
                in source
            ),

        "mentions_config":
            (
                "config"
                in source.lower()
            ),

        "mentions_lock":
            (
                "lock"
                in source.lower()
            ),

        "mentions_mutual":
            (
                "mutual"
                in source.lower()
            ),

        "mentions_round2":
            (
                "round2"
                in source.lower()
            ),
    }


    return {
        "cell_index":
            index,

        "execution_count":
            cell.get(
                "execution_count"
            ),

        "source_sha256":
            sha256_text(
                source
            ),

        "python_parseable":
            (
                tree
                is not None
            ),

        "marker_facts":
            markers,

        "string_constants":
            sorted(
                strings
            ),

        "calls":
            sorted(
                set(
                    calls
                )
            ),

        "source":
            source,
    }


# =============================================================================
# 12. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    NOTEBOOK12,
    NOTEBOOK13,
    PORTABLE,
    CORRECTED_ARCH,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3C-R2 "
    "V1 FALSE-WRITER CORRECTION + "
    "CORRECTED AUTHORITY EXTRACTION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "NOTEBOOK12:",
    NOTEBOOK12_REL,
    flush=True,
)

print(
    "NOTEBOOK13:",
    NOTEBOOK13_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Parent 14.3C-R1 lock
# =============================================================================

parent_lock = (
    verify_formal_artifact(
        PARENT
    )
)


parent_obj = (
    parent_lock[
        "object"
    ]
)


assert (
    parent_obj.get(
        "stage"
    )
    == "NOTEBOOK14_14_3C_R1"
)


assert (
    parent_obj[
        "claim_boundary"
    ][
        "historical_v1_action_pickle_opened"
    ]
    is False
)


print(
    "[PASS] PARENT_14_3C_R1_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    parent_lock[
        "file_sha256"
    ],
    flush=True,
)

print(
    "  content SHA:",
    parent_lock[
        "content_sha256"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Source locks
# =============================================================================

portable_sha = sha256_file(
    PORTABLE
)


nb12_sha = sha256_file(
    NOTEBOOK12
)


nb13_sha = sha256_file(
    NOTEBOOK13
)


corrected_arch_sha = sha256_file(
    CORRECTED_ARCH
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
)


assert (
    nb12_sha
    == EXPECTED_NOTEBOOK12_SHA
), (
    "Notebook12 changed since 14.3C-R1."
)


assert (
    nb13_sha
    == EXPECTED_NOTEBOOK13_SHA
), (
    "Notebook13 changed since the source audit."
)


assert (
    corrected_arch_sha
    == EXPECTED_CORRECTED_ARCH_FILE_SHA
)


print(
    "[PASS] SOURCE_LOCKS",
    flush=True,
)

print(
    "  portable SHA :",
    portable_sha,
    flush=True,
)

print(
    "  notebook12 SHA:",
    nb12_sha,
    flush=True,
)

print(
    "  notebook13 SHA:",
    nb13_sha,
    flush=True,
)

print(
    "  corrected architecture SHA:",
    corrected_arch_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Corrected architecture content lock
# =============================================================================

corrected_arch_obj = load_json(
    CORRECTED_ARCH
)


assert (
    corrected_arch_obj.get(
        "content_sha256"
    )
    == EXPECTED_CORRECTED_ARCH_CONTENT_SHA
    or
    any(
        (
            isinstance(
                leaf[
                    "value"
                ],
                str,
            )
            and leaf[
                "value"
            ]
            == EXPECTED_CORRECTED_ARCH_CONTENT_SHA
        )

        for leaf in (
            flatten_json(
                corrected_arch_obj
            )
        )
    )
), (
    "Corrected architecture content SHA not present."
)


corrected_arch_v1 = (
    corrected_arch_v1_records(
        corrected_arch_obj
    )
)


assert (
    len(
        corrected_arch_v1
    )
    >= 1
), (
    "Corrected architecture artifact contains no V1 authority record."
)


print(
    "[PASS] CORRECTED_ARCHITECTURE_V1_AUTHORITY_PRESENT",
    flush=True,
)

print(
    "  V1 records:",
    len(
        corrected_arch_v1
    ),
    flush=True,
)


for record in (
    corrected_arch_v1[
        :80
    ]
):

    print(
        " ",
        record[
            "path"
        ],
        "=",
        repr(
            record[
                "value"
            ]
        ),
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 16. Notebook12 false-writer proof
# =============================================================================

notebook12 = load_json(
    NOTEBOOK12
)


cell4_source = notebook_cell_source(
    notebook12,
    EXPECTED_FALSE_WRITER_CELL_INDEX,
)


cell4_sha = sha256_text(
    cell4_source
)


assert (
    cell4_sha
    == EXPECTED_FALSE_WRITER_CELL_SHA
)


assert (
    TARGET_ACTION_FILENAME
    in cell4_source
)


assert (
    FALSE_WRITER_REAL_OUTPUT
    in cell4_source
)


cell4_io = exact_cell_artifact_io(
    cell4_source,
    filename=TARGET_ACTION_FILENAME,
)


assert (
    len(
        cell4_io[
            "reads"
        ]
    )
    >= 1
), (
    "Notebook12 cell4 no longer reads the historical V1 action artifact."
)


assert (
    len(
        cell4_io[
            "target_writes"
        ]
    )
    == 0
), (
    "Unexpected: cell4 actually writes target V1 artifact."
)


assert (
    len(
        cell4_io[
            "all_pickle_writes"
        ]
    )
    >= 1
), (
    "Cell4 writer-signal diagnosis cannot be reproduced."
)


assert (
    any(
        (
            sink[
                "destination"
            ]
            == "OUT"
        )
        for sink in (
            cell4_io[
                "all_pickle_writes"
            ]
        )
    )
)


print(
    "[PASS] OLD_NOTEBOOK13_V1_WRITER_AUTHORITY_INVALIDATED",
    flush=True,
)

print(
    "  Notebook12 cell:",
    EXPECTED_FALSE_WRITER_CELL_INDEX,
    flush=True,
)

print(
    "  cell SHA:",
    cell4_sha,
    flush=True,
)

print(
    "  target reads:",
    len(
        cell4_io[
            "reads"
        ]
    ),
    flush=True,
)

print(
    "  target writes:",
    len(
        cell4_io[
            "target_writes"
        ]
    ),
    flush=True,
)

print(
    "  other pickle writes:",
    len(
        cell4_io[
            "all_pickle_writes"
        ]
    ),
    flush=True,
)


for read in (
    cell4_io[
        "reads"
    ]
):

    print(
        "    READ",
        f"L{read['line']}",
        read[
            "source"
        ],
        flush=True,
    )


for sink in (
    cell4_io[
        "all_pickle_writes"
    ]
):

    print(
        "    WRITE",
        f"L{sink['line']}",
        "destination=",
        sink[
            "destination"
        ],
        flush=True,
    )

    print(
        "      ",
        sink[
            "source"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 17. Scan Notebook13 for later false-writer repair audit
# =============================================================================

notebook13 = load_json(
    NOTEBOOK13
)


audit_cells = scan_notebook13_audit_cells(
    notebook13
)


assert (
    len(
        audit_cells
    )
    >= 1
), (
    "Notebook13 contains no later writer-correction audit cells."
)


all_marker_hits = sorted({
    marker

    for record in (
        audit_cells
    )

    for marker in (
        record[
            "markers"
        ]
    )
})


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK13 LATER WRITER-CORRECTION AUDIT CELLS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "audit cells:",
    len(
        audit_cells
    ),
    flush=True,
)

print(
    "markers found:",
    all_marker_hits,
    flush=True,
)

print(
    flush=True,
)


for record in (
    audit_cells
):

    print(
        f"cell[{record['cell_index']}] "
        f"exec={record['execution_count']} "
        f"SHA={record['source_sha256']}",
        flush=True,
    )

    print(
        "  markers:",
        record[
            "markers"
        ],
        flush=True,
    )


    if (
        record[
            "assignments"
        ]
    ):

        print(
            "  tracked assignments:",
            flush=True,
        )


        for assignment in (
            record[
                "assignments"
            ]
        ):

            print(
                "   ",
                assignment[
                    "name"
                ],
                "=",
                assignment[
                    "value_source"
                ],
                flush=True,
            )


    print(
        "  marker contexts:",
        flush=True,
    )


    for context in (
        record[
            "contexts"
        ][
            :20
        ]
    ):

        print(
            f"    L{context['start_line']}"
            f"-L{context['end_line']} "
            f"markers={context['markers']} "
            f"SHA={context['block_sha256']}",
            flush=True,
        )

        print(
            "      "
            + context[
                "text"
            ].replace(
                "\n",
                "\n      ",
            ),
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 18. Explicit repair markers must be present
# =============================================================================

has_false_writer_marker = (
    "FAILED_FALSE_WRITER"
    in all_marker_hits
)


has_viable_writer_marker = (
    "VIABLE_WRITERS"
    in all_marker_hits
)


has_corrected_authority_marker = (
    "CORRECTED_AUTHORITY"
    in all_marker_hits
)


has_cell70_audit = all(
    marker
    in all_marker_hits
    for marker in (
        "CELL70_ACTION_SCHEMA_COMPLETE",
        "CELL70_CONFIG_FILTER_PRESENT",
        "CELL70_CONFIG_UNIQUE",
        "CELL70_HAS_LOCK_STRUCTURE",
        "CELL70_ROLE",
    )
)


print(
    "[INFO] REPAIR MARKER COVERAGE",
    flush=True,
)

print(
    "  FAILED_FALSE_WRITER:",
    has_false_writer_marker,
    flush=True,
)

print(
    "  VIABLE_WRITERS:",
    has_viable_writer_marker,
    flush=True,
)

print(
    "  CORRECTED_AUTHORITY:",
    has_corrected_authority_marker,
    flush=True,
)

print(
    "  CELL70 audit family:",
    has_cell70_audit,
    flush=True,
)

print(
    flush=True,
)


assert (
    has_false_writer_marker
), (
    "Later Notebook13 audit does not preserve false-writer diagnosis."
)


assert (
    has_viable_writer_marker
    or has_corrected_authority_marker
    or has_cell70_audit
), (
    "No later corrected-writer authority evidence was found."
)


# =============================================================================
# 19. Aggregate tracked Notebook13 assignments
# =============================================================================

tracked = []


for record in (
    audit_cells
):

    for assignment in (
        record[
            "assignments"
        ]
    ):

        tracked.append({
            "cell_index":
                record[
                    "cell_index"
                ],

            "cell_source_sha256":
                record[
                    "source_sha256"
                ],

            **assignment,
        })


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK13 TRACKED AUTHORITY ASSIGNMENTS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for record in (
    tracked
):

    print(
        f"cell[{record['cell_index']}] "
        f"{record['name']} = "
        f"{record['value_source']}",
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 20. Formal report scan
# =============================================================================

report_records = scan_report_jsons()


print(
    "=" * 112,
    flush=True,
)

print(
    "FORMAL REPORTS WITH V1 WRITER/CORRECTION EVIDENCE",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "matching reports:",
    len(
        report_records
    ),
    flush=True,
)


for report in (
    report_records
):

    print(
        report[
            "path"
        ],
        flush=True,
    )

    print(
        "  SHA:",
        report[
            "file_sha256"
        ],
        flush=True,
    )

    print(
        "  keyword hits:",
        report[
            "keyword_hits"
        ],
        flush=True,
    )


    for leaf in (
        report[
            "relevant_leaf_records"
        ][
            :80
        ]
    ):

        print(
            "   ",
            leaf[
                "path"
            ],
            "=",
            repr(
                leaf[
                    "value"
                ]
            ),
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 21. Independent Notebook12 cell70 inspection
# =============================================================================

cell70 = inspect_notebook12_cell70(
    notebook12
)


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK12 CELL70 INDEPENDENT INSPECTION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "cell index:",
    cell70[
        "cell_index"
    ],
    flush=True,
)

print(
    "execution count:",
    cell70[
        "execution_count"
    ],
    flush=True,
)

print(
    "source SHA:",
    cell70[
        "source_sha256"
    ],
    flush=True,
)

print(
    "parseable:",
    cell70[
        "python_parseable"
    ],
    flush=True,
)

print(
    "marker facts:",
    json.dumps(
        cell70[
            "marker_facts"
        ],
        indent=2,
    ),
    flush=True,
)

print(
    "calls:",
    cell70[
        "calls"
    ],
    flush=True,
)

print(
    flush=True,
)

print(
    "CELL70 SOURCE:",
    flush=True,
)

print(
    cell70[
        "source"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 22. Extract Notebook13 CELL70 audit assignments
# =============================================================================

cell70_assignment_records = [
    record

    for record in (
        tracked
    )

    if record[
        "name"
    ]
    in set(
        CELL70_MARKERS
    )
]


cell70_assignment_by_name = {}


for record in (
    cell70_assignment_records
):

    cell70_assignment_by_name.setdefault(
        record[
            "name"
        ],
        [],
    ).append(
        record
    )


print(
    "=" * 112,
    flush=True,
)

print(
    "CELL70 AUDIT CONTRACT FROM NOTEBOOK13",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for name in (
    CELL70_MARKERS
):

    records = (
        cell70_assignment_by_name.get(
            name,
            [],
        )
    )


    if not records:

        continue


    print(
        name,
        flush=True,
    )


    for record in (
        records
    ):

        print(
            "  ",
            f"cell[{record['cell_index']}]",
            record[
                "value_source"
            ],
            flush=True,
        )


print(
    flush=True,
)


# =============================================================================
# 23. Extract explicit corrected-authority expressions
# =============================================================================

corrected_authority_assignments = [
    record

    for record in (
        tracked
    )

    if record[
        "name"
    ]
    in {
        "CORRECTED_AUTHORITY",
        "V1_AUTHORITY",
        "CORRECTED_AUTHORITY_MODEL",
        "RECONSTRUCTION_AUTHORITY",
    }
]


print(
    "=" * 112,
    flush=True,
)

print(
    "EXPLICIT CORRECTED-AUTHORITY ASSIGNMENTS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


if not corrected_authority_assignments:

    print(
        "  <no directly literal assignment; use CELL70/report authority>",
        flush=True,
    )


for record in (
    corrected_authority_assignments
):

    print(
        f"cell[{record['cell_index']}] "
        f"{record['name']} =",
        flush=True,
    )

    print(
        " ",
        record[
            "value_source"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 24. Authority-resolution classification
#
# We deliberately distinguish:
#
#   EXPLICIT_CORRECTED_AUTHORITY
#   CELL70_CORRECTED_SEMANTIC_AUTHORITY
#   CORRECTED_ARCHITECTURE_ONLY
#   UNRESOLVED
#
# No implementation proceeds here.
# =============================================================================

if (
    corrected_authority_assignments
):

    corrected_authority_status = (
        "EXPLICIT_CORRECTED_AUTHORITY_FOUND"
    )


elif (
    has_cell70_audit
):

    corrected_authority_status = (
        "CELL70_CORRECTED_SEMANTIC_AUTHORITY_FOUND"
    )


elif (
    corrected_arch_v1
):

    corrected_authority_status = (
        "CORRECTED_ARCHITECTURE_V1_AUTHORITY_FOUND"
    )


else:

    corrected_authority_status = (
        "UNRESOLVED"
    )


authority_resolved = (
    corrected_authority_status
    != "UNRESOLVED"
)


print(
    "=" * 112,
    flush=True,
)

print(
    "CORRECTED V1 AUTHORITY STATUS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    corrected_authority_status,
    flush=True,
)

print(
    "resolved:",
    authority_resolved,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 25. Formal false-writer successor model
# =============================================================================

false_writer_correction = {
    "invalidated_authority": {
        "historical_stage":
            "13.3B3",

        "claimed_stage":
            "MUTUAL_CONT_V1",

        "claimed_notebook":
            str(
                NOTEBOOK12_REL
            ),

        "claimed_cell_index":
            EXPECTED_FALSE_WRITER_CELL_INDEX,

        "cell_source_sha256":
            cell4_sha,

        "historical_claim":
            "GT_BLIND_EXACT_WRITER_CELL",
    },

    "exact_io_reanalysis": {
        "target_artifact":
            TARGET_ACTION_FILENAME,

        "target_path_aliases":
            cell4_io[
                "aliases"
            ],

        "target_reads":
            cell4_io[
                "reads"
            ],

        "target_write_count":
            len(
                cell4_io[
                    "target_writes"
                ]
            ),

        "other_pickle_writes":
            cell4_io[
                "all_pickle_writes"
            ],
    },

    "corrected_classification":
        "CONSUMER_NOT_WRITER",

    "reason": (
        "Cell 4 reads the locked MUTUAL_CONT_V1 action artifact and "
        "writes a different post-V1 residual artifact. Presence of an "
        "unrelated .to_pickle sink is insufficient to classify the cell "
        "as the action-artifact writer."
    ),
}


# =============================================================================
# 26. Semantic gates
#
# PASS means provenance correction/extraction succeeded.
# It does not mean V1 runtime is implemented.
# =============================================================================

semantic_gates = {
    "parent_14_3c_r1_locked":
        True,

    "portable_source_locked":
        (
            portable_sha
            == EXPECTED_PORTABLE_SHA
        ),

    "notebook12_locked":
        (
            nb12_sha
            == EXPECTED_NOTEBOOK12_SHA
        ),

    "notebook13_locked":
        (
            nb13_sha
            == EXPECTED_NOTEBOOK13_SHA
        ),

    "corrected_architecture_locked":
        (
            corrected_arch_sha
            == EXPECTED_CORRECTED_ARCH_FILE_SHA
        ),

    "historical_false_writer_cell_locked":
        (
            cell4_sha
            == EXPECTED_FALSE_WRITER_CELL_SHA
        ),

    "historical_false_writer_reads_v1_artifact":
        (
            len(
                cell4_io[
                    "reads"
                ]
            )
            >= 1
        ),

    "historical_false_writer_does_not_write_v1_artifact":
        (
            len(
                cell4_io[
                    "target_writes"
                ]
            )
            == 0
        ),

    "historical_false_writer_has_unrelated_writer_signal":
        (
            len(
                cell4_io[
                    "all_pickle_writes"
                ]
            )
            >= 1
        ),

    "later_false_writer_repair_marker_present":
        has_false_writer_marker,

    "later_writer_repair_evidence_present":
        (
            has_viable_writer_marker
            or
            has_corrected_authority_marker
            or
            has_cell70_audit
        ),

    "corrected_v1_authority_resolved":
        authority_resolved,

    "historical_v1_action_pickle_not_opened":
        True,

    "no_source_mutation":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3C-R2 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok in (
        semantic_gates.items()
    )

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3C-R2 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 27. Next-stage decision
# =============================================================================

if (
    corrected_authority_status
    in {
        "EXPLICIT_CORRECTED_AUTHORITY_FOUND",
        "CELL70_CORRECTED_SEMANTIC_AUTHORITY_FOUND",
    }
):

    next_stage = (
        "NOTEBOOK14_14_3C_R3_"
        "V1_CORRECTED_AUTHORITY_SEMANTIC_CONTRACT_FREEZE"
    )


else:

    next_stage = (
        "NOTEBOOK14_14_3C_R3_"
        "V1_CORRECTED_AUTHORITY_TARGETED_RECOVERY"
    )


# =============================================================================
# 28. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3C_R2",

    "title": (
        "V1_FALSE_WRITER_CORRECTION_AND_"
        "CORRECTED_AUTHORITY_EXTRACTION"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_PROVENANCE_CORRECTION_ONLY",

    "parent_14_3c_r1": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            parent_lock[
                "file_sha256"
            ],

        "content_sha256":
            parent_lock[
                "content_sha256"
            ],
    },

    "source_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha,
        },

        "notebook12": {
            "path":
                str(
                    NOTEBOOK12_REL
                ),

            "file_sha256":
                nb12_sha,
        },

        "notebook13": {
            "path":
                str(
                    NOTEBOOK13_REL
                ),

            "file_sha256":
                nb13_sha,
        },

        "corrected_architecture": {
            "path":
                str(
                    CORRECTED_ARCH_REL
                ),

            "file_sha256":
                corrected_arch_sha,

            "content_sha256":
                EXPECTED_CORRECTED_ARCH_CONTENT_SHA,
        },
    },

    "false_writer_correction":
        false_writer_correction,

    "notebook13_later_audit": {
        "audit_cell_count":
            len(
                audit_cells
            ),

        "marker_hits":
            all_marker_hits,

        "audit_cells":
            audit_cells,

        "tracked_assignments":
            tracked,

        "explicit_corrected_authority_assignments":
            corrected_authority_assignments,

        "cell70_assignments":
            cell70_assignment_records,
    },

    "notebook12_cell70": {
        "cell_index":
            cell70[
                "cell_index"
            ],

        "execution_count":
            cell70[
                "execution_count"
            ],

        "source_sha256":
            cell70[
                "source_sha256"
            ],

        "python_parseable":
            cell70[
                "python_parseable"
            ],

        "marker_facts":
            cell70[
                "marker_facts"
            ],

        "calls":
            cell70[
                "calls"
            ],
    },

    "formal_report_correction_evidence":
        report_records,

    "corrected_architecture_v1_records":
        corrected_arch_v1,

    "resolution": {
        "old_13_3b3_v1_writer_authority":
            "INVALIDATED",

        "old_cell4_role":
            "CONSUMER_NOT_WRITER",

        "corrected_authority_status":
            corrected_authority_status,

        "corrected_authority_resolved":
            authority_resolved,

        "implementation_authorized":
            False,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "old_v1_writer_authority":
            "FORMALLY_INVALIDATED_BY_EXACT_IO",

        "corrected_v1_authority":
            corrected_authority_status,

        "v1_semantic_contract_frozen":
            False,

        "v1_graph_runtime_implemented":
            False,

        "post_v2_graph_provider_implemented":
            False,

        "post_v2_graph_structural_fidelity":
            "NOT_YET_VERIFIED",

        "b12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "historical_v1_action_pickle_opened":
            False,

        "source_mutation":
            False,

        "gt_read":
            False,

        "evaluator_read":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No historical V1 action pickle was opened.",
        "No V1 runtime implementation claim.",
        "No post-V2 graph provider implementation claim.",
        "No structural-fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3C-R2 artifact already exists but differs "
        "from the exact false-writer correction."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 29. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3C-R2 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "OLD V1 WRITER AUTHORITY:",
    flush=True,
)

print(
    "  Notebook12 cell 4",
    flush=True,
)

print(
    "  SHA:",
    cell4_sha,
    flush=True,
)

print(
    "  historical classification:",
    "GT_BLIND_EXACT_WRITER_CELL",
    flush=True,
)

print(
    "  corrected classification:",
    "CONSUMER_NOT_WRITER",
    flush=True,
)

print(
    "  target reads:",
    len(
        cell4_io[
            "reads"
        ]
    ),
    flush=True,
)

print(
    "  target writes:",
    len(
        cell4_io[
            "target_writes"
        ]
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "LATER NOTEBOOK13 REPAIR EVIDENCE:",
    flush=True,
)

print(
    "  audit cells:",
    len(
        audit_cells
    ),
    flush=True,
)

print(
    "  FAILED_FALSE_WRITER:",
    has_false_writer_marker,
    flush=True,
)

print(
    "  VIABLE_WRITERS:",
    has_viable_writer_marker,
    flush=True,
)

print(
    "  CORRECTED_AUTHORITY:",
    has_corrected_authority_marker,
    flush=True,
)

print(
    "  CELL70 audit:",
    has_cell70_audit,
    flush=True,
)

print(
    flush=True,
)


print(
    "CORRECTED V1 AUTHORITY:",
    flush=True,
)

print(
    "  status:",
    corrected_authority_status,
    flush=True,
)

print(
    "  resolved:",
    authority_resolved,
    flush=True,
)

print(
    flush=True,
)


print(
    "NOTEBOOK12 CELL70:",
    flush=True,
)

print(
    "  SHA:",
    cell70[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  facts:",
    json.dumps(
        cell70[
            "marker_facts"
        ],
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  old 13.3B3 V1 writer authority = INVALIDATED",
    flush=True,
)

print(
    "  corrected V1 authority =",
    corrected_authority_status,
    flush=True,
)

print(
    "  V1 semantic runtime contract = NOT YET FROZEN",
    flush=True,
)

print(
    "  V1 graph runtime = NOT YET IMPLEMENTED",
    flush=True,
)

print(
    "  post-V2 provider = NOT YET IMPLEMENTED",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3C-R3-R1 — V1 CORRECTED AUTHORITY SEMANTIC CONTRACT FREEZE
#                ARCHITECTURE CONTENT-LOCK REPAIR
#
# SUCCESSOR TO
# ------------
# 14.3C-R3 failed BEFORE semantic audit because it assumed:
#
#     arch["content_sha256"] == EXPECTED_ARCH_CONTENT_SHA
#
# The historical authority artifact is already EXACTLY file-locked:
#
#     file SHA =
#     e7d307f988b97717a4250534179fc07f90408b29dedccbcc4fc9b96567c7f4c8
#
# Earlier provenance discovery established:
#
#     match_mode = EXACT_FILE_SHA
#     contains_expected_content_sha = True
#
# Therefore the correct historical lock is:
#
#     1. exact file SHA
#     2. expected historical content SHA recursively embedded in artifact
#
# It is NOT valid to impose a new top-level content_sha256 schema on this
# already-frozen historical artifact.
#
#
# PURPOSE
# -------
# Freeze the exact implementable V1 semantic contract:
#
#   POST_MORPH_DIV_V2 graph
#       -> explicit association provider
#       -> source-side top1
#       -> target-side top1
#       -> reciprocal top1
#       -> candidate_prob >= 0.25
#       -> source final outdegree == 0
#       -> target final indegree == 0
#       -> MutualContV1Action
#       -> historical recovered_edge_dist
#       -> graph.add_edge(... validate_keys=True)
#
# and prove the CURRENT matrix/action implementation:
#
#   reconstruct_mutual_cont_v1_actions
#
# is compatible with the frozen specification before graph-level wiring.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical V1 action pickle read
#   - GT
#   - evaluator
#   - performance metric
#   - structural-fidelity claim
#   - parameter search / retuning
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3c_r2_"
    "v1_false_writer_correction_and_"
    "corrected_authority_extraction_v1.json"
)

PARENT = ROOT / PARENT_REL


ARCH_REL = Path(
    "reports/"
    "notebook13_13_3c_r2_"
    "corrected_authority_model_and_architecture_refreeze_v1.json"
)

ARCH = ROOT / ARCH_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


RECIPROCAL_REL = Path(
    "scripts/stage12_iterative_reciprocal.py"
)

RECIPROCAL = ROOT / RECIPROCAL_REL


B12_WRITER_REL = Path(
    "scripts/run_stage12_fold0_bridge_geometry_universe.py"
)

B12_WRITER = ROOT / B12_WRITER_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3c_r3_r1_"
    "v1_corrected_authority_semantic_contract_freeze_"
    "architecture_content_lock_repair_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_ARCH_FILE_SHA = (
    "e7d307f988b97717a4250534179fc07f90408b29dedccbcc4fc9b96567c7f4c8"
)

EXPECTED_ARCH_CONTENT_SHA = (
    "49f533d19e2354735c9faef38179625c058732ba4838e1a7544aefd1733fc3f4"
)


EXPECTED_PORTABLE_SHA = (
    "a1fe0c9e5cd5e20a2ae5921bd4b4c55d078022c3348d5945bffa501ac3ff36ed"
)

EXPECTED_V1_ACTION_CORE_SHA = (
    "1a9e988329cf91c9a8a706987fede592e0d223e69ec1113e02e42726f6d09449"
)

EXPECTED_V1_GRAPH_PLACEHOLDER_SHA = (
    "ea9cba47ba6992c0dab67d747ef119061eb92b3cd52f0762274d6300a076e44e"
)


EXPECTED_RECIPROCAL_FILE_SHA = (
    "0f9a09e5a69b7ed55ac6613a95ef3eda7505918e258c4450b138dd49e255a550"
)

EXPECTED_FAST_ASSOCIATION_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


EXPECTED_B12_WRITER_SHA = (
    "c7fb950c03711ac9c85f0476013f88554a42fc2b8fb21c8cc4afc1bb6b5649a4"
)

EXPECTED_ADD_ACTIONS_SHA = (
    "36a6e189bbaea0fe2d9d2c4f114142087e5d99a42bd0c415085cec579bb07751"
)


EXPECTED_V1_ORACLE_PATH = (
    "reports/stage12_residual_audit/"
    "stage12_fold0_mutual_cont_v1_actions_locked.pkl"
)

EXPECTED_V1_ORACLE_SHA = (
    "0d1bbaef9653bdd7844e3a3c90ba818ff47a1d3128bd4f0569128b5d88a62038"
)

EXPECTED_V1_ORACLE_SIZE = (
    860408
)


EXPECTED_REQUIRED_SEMANTICS = [
    "OBTAIN_CURRENT_DATASET_SOURCE_TARGET_CANDIDATE_PROBABILITY_MATRIX",
    "COMPUTE_SOURCE_SIDE_TOP1",
    "COMPUTE_TARGET_SIDE_TOP1",
    "KEEP_RECIPROCAL_TOP1_PAIRS_ONLY",
    "REQUIRE_CANDIDATE_PROB_GE_0_25",
    "REQUIRE_SOURCE_FINAL_OUTDEGREE_EQ_0",
    "REQUIRE_TARGET_FINAL_INDEGREE_EQ_0",
    "ADD_EXACTLY_ONE_SOURCE_TO_TARGET_CONTINUATION_EDGE",
]


EXPECTED_ACTION_FIELDS = [
    "dataset_id",
    "source_node_id",
    "target_node_id",
    "candidate_prob",
]


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    data: bytes,
) -> str:

    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(
    text: str,
) -> str:

    return sha256_bytes(
        text.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:
    """
    Recursive exact-value search.

    Used for historical artifacts whose expected content SHA is embedded
    within the already-frozen object but is not required to occupy a newly
    invented top-level field.
    """

    if obj == target:

        return True


    if isinstance(
        obj,
        dict,
    ):

        return any(
            contains_exact_value(
                key,
                target,
            )
            or
            contains_exact_value(
                value,
                target,
            )

            for key, value in (
                obj.items()
            )
        )


    if isinstance(
        obj,
        (
            list,
            tuple,
        ),
    ):

        return any(
            contains_exact_value(
                value,
                target,
            )

            for value in obj
        )


    return False


def verify_formal_artifact(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    ), path


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    ), (
        f"{path} is not a PASS artifact."
    )


    stored_sha = (
        obj.get(
            "content_sha256"
        )
    )


    assert (
        isinstance(
            stored_sha,
            str,
        )
        and len(
            stored_sha
        )
        == 64
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    computed_sha = (
        canonical_json_sha(
            payload
        )
    )


    assert (
        computed_sha
        == stored_sha
    ), (
        "Formal parent content SHA mismatch.\n"
        f"stored={stored_sha}\n"
        f"actual={computed_sha}"
    )


    return {
        "object":
            obj,

        "file_sha256":
            sha256_file(
                path
            ),

        "content_sha256":
            stored_sha,
    }


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    result = ast.get_source_segment(
        text,
        node,
    )


    assert (
        result
        is not None
    )


    return result


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    def visit(
        item: ast.AST,
    ):

        if isinstance(
            item,
            ast.Name,
        ):

            result.add(
                item.id
            )


        elif isinstance(
            item,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for child in (
                item.elts
            ):

                visit(
                    child
                )


        elif isinstance(
            item,
            ast.Starred,
        ):

            visit(
                item.value
            )


    visit(
        node
    )


    return result


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def top_level_classes(
    tree: ast.Module,
) -> dict[
    str,
    ast.ClassDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            ast.ClassDef,
        )
    }


def function_record(
    *,
    text: str,
    fn: ast.AST,
) -> dict[str, Any]:

    source = node_source(
        text,
        fn,
    )


    calls = sorted({
        dotted_name(
            call.func
        )
        or "<dynamic-call>"

        for call in ast.walk(
            fn
        )

        if isinstance(
            call,
            ast.Call,
        )
    })


    returns = []


    for node in ast.walk(
        fn
    ):

        if not isinstance(
            node,
            ast.Return,
        ):

            continue


        returns.append({
            "line":
                node.lineno,

            "expression":
                (
                    ast.unparse(
                        node.value
                    )
                    if node.value
                    is not None
                    else "None"
                ),
        })


    return {
        "name":
            getattr(
                fn,
                "name",
                None,
            ),

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "calls":
            calls,

        "returns":
            returns,

        "source":
            source,
    }


def function_parameters(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[str]:

    names = []


    names.extend(
        arg.arg
        for arg in (
            fn.args.posonlyargs
        )
    )


    names.extend(
        arg.arg
        for arg in (
            fn.args.args
        )
    )


    if (
        fn.args.vararg
        is not None
    ):

        names.append(
            "*"
            + fn.args.vararg.arg
        )


    names.extend(
        arg.arg
        for arg in (
            fn.args.kwonlyargs
        )
    )


    if (
        fn.args.kwarg
        is not None
    ):

        names.append(
            "**"
            + fn.args.kwarg.arg
        )


    return names


def class_fields(
    text: str,
    cls: ast.ClassDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for stmt in (
        cls.body
    ):

        if not isinstance(
            stmt,
            ast.AnnAssign,
        ):

            continue


        if not isinstance(
            stmt.target,
            ast.Name,
        ):

            continue


        result.append({
            "name":
                stmt.target.id,

            "annotation":
                ast.unparse(
                    stmt.annotation
                ),

            "default":
                (
                    ast.unparse(
                        stmt.value
                    )
                    if stmt.value
                    is not None
                    else None
                ),

            "line":
                stmt.lineno,
        })


    return result


def find_assignment(
    tree: ast.Module,
    name: str,
) -> ast.AST:

    matches = []


    for stmt in (
        tree.body
    ):

        names = set()
        value = None


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                names |= (
                    target_names(
                        target
                    )
                )


            value = stmt.value


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            names |= (
                target_names(
                    stmt.target
                )
            )


            value = stmt.value


        if (
            name
            in names
        ):

            matches.append(
                value
            )


    assert (
        len(
            matches
        )
        == 1
    ), (
        f"Expected one assignment for {name}; "
        f"found {len(matches)}."
    )


    assert (
        matches[
            0
        ]
        is not None
    )


    return matches[
        0
    ]


def literal_assignment(
    tree: ast.Module,
    name: str,
) -> Any:

    return ast.literal_eval(
        find_assignment(
            tree,
            name,
        )
    )


# =============================================================================
# 3. AST semantic extraction helpers
# =============================================================================

def argmax_records(
    text: str,
    fn: ast.AST,
) -> list[
    dict[str, Any]
]:

    result = []


    for call in ast.walk(
        fn
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        call_name = dotted_name(
            call.func
        )


        if (
            call_name
            is None
            or not call_name.endswith(
                "argmax"
            )
        ):

            continue


        axis_source = None
        axis_literal = None


        for kw in (
            call.keywords
        ):

            if (
                kw.arg
                == "axis"
            ):

                axis_source = (
                    ast.unparse(
                        kw.value
                    )
                )


                try:

                    axis_literal = (
                        ast.literal_eval(
                            kw.value
                        )
                    )

                except Exception:

                    axis_literal = None


        if (
            axis_source
            is None
            and len(
                call.args
            )
            >= 2
        ):

            axis_source = (
                ast.unparse(
                    call.args[
                        1
                    ]
                )
            )


            try:

                axis_literal = (
                    ast.literal_eval(
                        call.args[
                            1
                        ]
                    )
                )

            except Exception:

                axis_literal = None


        source = node_source(
            text,
            call,
        )


        result.append({
            "line":
                call.lineno,

            "call_name":
                call_name,

            "axis_source":
                axis_source,

            "axis_literal":
                axis_literal,

            "source_sha256":
                sha256_text(
                    source
                ),

            "source":
                source,
        })


    return result


def comparison_records(
    text: str,
    fn: ast.AST,
) -> list[
    dict[str, Any]
]:

    result = []


    for node in ast.walk(
        fn
    ):

        if not isinstance(
            node,
            ast.Compare,
        ):

            continue


        source = node_source(
            text,
            node,
        )


        result.append({
            "line":
                node.lineno,

            "source":
                source,

            "source_sha256":
                sha256_text(
                    source
                ),
        })


    return result


def action_constructor_records(
    text: str,
    fn: ast.AST,
) -> list[
    dict[str, Any]
]:

    result = []


    for call in ast.walk(
        fn
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        if (
            dotted_name(
                call.func
            )
            != "MutualContV1Action"
        ):

            continue


        source = node_source(
            text,
            call,
        )


        result.append({
            "line":
                call.lineno,

            "args": [
                ast.unparse(
                    arg
                )
                for arg in (
                    call.args
                )
            ],

            "keywords": {
                (
                    kw.arg
                    if kw.arg
                    is not None
                    else "**"
                ):
                    ast.unparse(
                        kw.value
                    )

                for kw in (
                    call.keywords
                )
            },

            "source_sha256":
                sha256_text(
                    source
                ),

            "source":
                source,
        })


    return result


def literal_float_values(
    fn: ast.AST,
) -> list[float]:

    values = []


    for node in ast.walk(
        fn
    ):

        if not isinstance(
            node,
            ast.Constant,
        ):

            continue


        if (
            isinstance(
                node.value,
                (
                    int,
                    float,
                ),
            )
            and not isinstance(
                node.value,
                bool,
            )
        ):

            values.append(
                float(
                    node.value
                )
            )


    return values


# =============================================================================
# 4. Historical graph-mutation extractor
# =============================================================================

def mutation_contract(
    *,
    text: str,
    fn: ast.AST,
) -> dict[str, Any]:

    source = node_source(
        text,
        fn,
    )


    calls = sorted({
        dotted_name(
            call.func
        )
        or "<dynamic-call>"

        for call in ast.walk(
            fn
        )

        if isinstance(
            call,
            ast.Call,
        )
    })


    return {
        "source_sha256":
            sha256_text(
                source
            ),

        "source_pred_node_id":
            (
                "source_pred_node_id"
                in source
            ),

        "target_pred_node_id":
            (
                "target_pred_node_id"
                in source
            ),

        "candidate_prob":
            (
                "candidate_prob"
                in source
            ),

        "recovered_edge_dist":
            any(
                call.endswith(
                    "recovered_edge_dist"
                )

                for call in (
                    calls
                )
            ),

        "add_edge":
            any(
                call.endswith(
                    ".add_edge"
                )

                for call in (
                    calls
                )
            ),

        "validate_keys_true":
            (
                "validate_keys=True"
                in source.replace(
                    " ",
                    ""
                )
            ),

        "calls":
            calls,

        "source":
            source,
    }


# =============================================================================
# 5. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    ARCH,
    PORTABLE,
    RECIPROCAL,
    B12_WRITER,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3C-R3-R1 "
    "V1 CORRECTED AUTHORITY SEMANTIC CONTRACT FREEZE "
    "— ARCHITECTURE CONTENT-LOCK REPAIR",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "CORRECTED ARCHITECTURE:",
    ARCH_REL,
    flush=True,
)

print(
    "PORTABLE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Parent 14.3C-R2 formal lock
# =============================================================================

parent_lock = verify_formal_artifact(
    PARENT
)


parent_obj = (
    parent_lock[
        "object"
    ]
)


assert (
    parent_obj.get(
        "stage"
    )
    == "NOTEBOOK14_14_3C_R2"
)


assert (
    parent_obj[
        "resolution"
    ][
        "old_13_3b3_v1_writer_authority"
    ]
    == "INVALIDATED"
)


assert (
    parent_obj[
        "resolution"
    ][
        "corrected_authority_resolved"
    ]
    is True
)


assert (
    parent_obj[
        "claim_boundary"
    ][
        "historical_v1_action_pickle_opened"
    ]
    is False
)


assert (
    parent_obj[
        "claim_boundary"
    ][
        "v1_graph_runtime_implemented"
    ]
    is False
)


print(
    "[PASS] PARENT_14_3C_R2_FORMAL_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    parent_lock[
        "file_sha256"
    ],
    flush=True,
)

print(
    "  content SHA:",
    parent_lock[
        "content_sha256"
    ],
    flush=True,
)

print(
    "  corrected authority status:",
    parent_obj[
        "resolution"
    ][
        "corrected_authority_status"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Corrected architecture historical lock — R1 REPAIR
#
# DO NOT impose:
#
#     arch["content_sha256"] == EXPECTED_ARCH_CONTENT_SHA
#
# Historical provenance already established:
#
#     EXACT_FILE_SHA
#     contains_expected_content_sha = True
#
# We reproduce exactly that authority contract.
# =============================================================================

arch_file_sha = sha256_file(
    ARCH
)


assert (
    arch_file_sha
    == EXPECTED_ARCH_FILE_SHA
), (
    "Corrected architecture file SHA drift.\n"
    f"expected={EXPECTED_ARCH_FILE_SHA}\n"
    f"actual  ={arch_file_sha}"
)


arch = load_json(
    ARCH
)


arch_top_level_content_sha = (
    arch.get(
        "content_sha256"
    )
)


arch_contains_expected_content_sha = (
    contains_exact_value(
        arch,
        EXPECTED_ARCH_CONTENT_SHA,
    )
)


assert (
    arch_contains_expected_content_sha
), (
    "Exact architecture file is present, but the frozen historical "
    "content SHA is no longer embedded anywhere in the artifact."
)


architecture_lock_mode = (
    "EXACT_FILE_SHA_PLUS_EMBEDDED_EXPECTED_CONTENT_SHA"
)


print(
    "[PASS] CORRECTED_ARCHITECTURE_HISTORICAL_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    arch_file_sha,
    flush=True,
)

print(
    "  expected historical content SHA:",
    EXPECTED_ARCH_CONTENT_SHA,
    flush=True,
)

print(
    "  top-level content_sha256:",
    repr(
        arch_top_level_content_sha
    ),
    flush=True,
)

print(
    "  expected content SHA embedded:",
    arch_contains_expected_content_sha,
    flush=True,
)

print(
    "  lock mode:",
    architecture_lock_mode,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Extract corrected V1 authority
# =============================================================================

assert (
    "authority_model"
    in arch
)


assert (
    "runtime_architecture"
    in arch
)


assert (
    "MUTUAL_CONT_V1"
    in arch[
        "authority_model"
    ]
)


assert (
    "V1_reconstruction_contract"
    in arch[
        "runtime_architecture"
    ]
)


authority_model = (
    arch[
        "authority_model"
    ][
        "MUTUAL_CONT_V1"
    ]
)


runtime_contract = (
    arch[
        "runtime_architecture"
    ][
        "V1_reconstruction_contract"
    ]
)


assert (
    authority_model[
        "authority_type"
    ]
    == (
        "FROZEN_SPECIFICATION_PLUS_"
        "HISTORICAL_GT_BLIND_ACTION_ORACLE"
    )
)


assert (
    authority_model[
        "source_extractable_generator"
    ]
    is False
)


assert (
    authority_model[
        "runtime_mode"
    ]
    == "SPECIFICATION_DERIVED_RECONSTRUCTION"
)


assert (
    authority_model[
        "exact_historical_fidelity_required"
    ]
    is True
)


assert (
    authority_model[
        "scientific_parameter_search_allowed"
    ]
    is False
)


frozen_spec = (
    authority_model[
        "frozen_specification"
    ]
)


expected_frozen_spec = {
    "source_final_outdegree":
        0,

    "target_final_indegree":
        0,

    "reciprocal_top1":
        True,

    "candidate_prob_min":
        0.25,

    "operation":
        "ADD_SINGLE_CONTINUATION_EDGE",
}


for key, expected in (
    expected_frozen_spec.items()
):

    assert (
        frozen_spec[
            key
        ]
        == expected
    ), (
        key,
        frozen_spec[
            key
        ],
        expected,
    )


oracle = (
    authority_model[
        "historical_structural_oracle"
    ]
)


assert (
    oracle[
        "path"
    ]
    == EXPECTED_V1_ORACLE_PATH
)


assert (
    oracle[
        "file_sha256"
    ]
    == EXPECTED_V1_ORACLE_SHA
)


assert (
    int(
        oracle[
            "size_bytes"
        ]
    )
    == EXPECTED_V1_ORACLE_SIZE
)


assert (
    oracle[
        "runtime_use_allowed"
    ]
    is False
)


assert (
    oracle[
        "structural_fidelity_use_allowed"
    ]
    is True
)


print(
    "[PASS] V1_CORRECTED_AUTHORITY_MODEL_LOCK",
    flush=True,
)

print(
    "  authority type:",
    authority_model[
        "authority_type"
    ],
    flush=True,
)

print(
    "  source extractable generator:",
    authority_model[
        "source_extractable_generator"
    ],
    flush=True,
)

print(
    "  runtime mode:",
    authority_model[
        "runtime_mode"
    ],
    flush=True,
)

print(
    "  exact historical fidelity required:",
    authority_model[
        "exact_historical_fidelity_required"
    ],
    flush=True,
)

print(
    "  parameter search allowed:",
    authority_model[
        "scientific_parameter_search_allowed"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Freeze runtime reconstruction contract
# =============================================================================

assert (
    runtime_contract[
        "scientific_policy_source"
    ]
    == "FROZEN_V1_SPECIFICATION"
)


assert (
    float(
        runtime_contract[
            "scientific_threshold"
        ]
    )
    == 0.25
)


assert (
    runtime_contract[
        "new_threshold_search_allowed"
    ]
    is False
)


assert (
    runtime_contract[
        "runtime_input_graph"
    ]
    == "POST_MORPH_DIV_V2"
)


assert (
    runtime_contract[
        "association_probability_source"
    ]
    == "EXPLICIT_ASSOCIATION_PROVIDER"
)


assert (
    runtime_contract[
        "required_semantics"
    ]
    == EXPECTED_REQUIRED_SEMANTICS
), (
    "Corrected V1 required-semantic order drift.\n"
    f"expected={EXPECTED_REQUIRED_SEMANTICS}\n"
    f"actual={runtime_contract['required_semantics']}"
)


assert (
    runtime_contract[
        "historical_action_lookup_allowed_at_runtime"
    ]
    is False
)


assert (
    runtime_contract[
        "historical_oracle_allowed_during_fidelity"
    ]
    is True
)


assert (
    runtime_contract[
        "historical_oracle_may_change_scientific_threshold"
    ]
    is False
)


assert (
    runtime_contract[
        "historical_oracle_may_drive_performance_tuning"
    ]
    is False
)


assert (
    runtime_contract[
        "historical_oracle_may_drive_structural_debugging"
    ]
    is True
)


assert (
    runtime_contract[
        "exact_structural_fidelity_required_before_portable_claim"
    ]
    is True
)


print(
    "[PASS] V1_RUNTIME_RECONSTRUCTION_CONTRACT_LOCK",
    flush=True,
)

print(
    "  scientific threshold:",
    runtime_contract[
        "scientific_threshold"
    ],
    flush=True,
)

print(
    "  runtime input graph:",
    runtime_contract[
        "runtime_input_graph"
    ],
    flush=True,
)

print(
    "  probability source:",
    runtime_contract[
        "association_probability_source"
    ],
    flush=True,
)

print(
    "  historical runtime lookup allowed:",
    runtime_contract[
        "historical_action_lookup_allowed_at_runtime"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Portable source exact lock
# =============================================================================

portable_sha = sha256_file(
    PORTABLE
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
)


portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_text,
    filename=str(
        PORTABLE
    ),
)


portable_functions = (
    top_level_functions(
        portable_tree
    )
)


portable_classes = (
    top_level_classes(
        portable_tree
    )
)


for required_name in (
    "reconstruct_mutual_cont_v1_actions",
    "_reconstruct_portable_mutual_cont_v1",
):

    assert (
        required_name
        in portable_functions
    )


assert (
    "MutualContV1Action"
    in portable_classes
)


core_fn = (
    portable_functions[
        "reconstruct_mutual_cont_v1_actions"
    ]
)


graph_placeholder_fn = (
    portable_functions[
        "_reconstruct_portable_mutual_cont_v1"
    ]
)


core_record = function_record(
    text=portable_text,
    fn=core_fn,
)


placeholder_record = function_record(
    text=portable_text,
    fn=graph_placeholder_fn,
)


assert (
    core_record[
        "source_sha256"
    ]
    == EXPECTED_V1_ACTION_CORE_SHA
)


assert (
    placeholder_record[
        "source_sha256"
    ]
    == EXPECTED_V1_GRAPH_PLACEHOLDER_SHA
)


print(
    "[PASS] PORTABLE_V1_SOURCE_LOCK",
    flush=True,
)

print(
    "  module SHA:",
    portable_sha,
    flush=True,
)

print(
    "  action core SHA:",
    core_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  graph placeholder SHA:",
    placeholder_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. V1 action schema lock
# =============================================================================

action_cls = (
    portable_classes[
        "MutualContV1Action"
    ]
)


action_fields = class_fields(
    portable_text,
    action_cls,
)


action_field_names = [
    record[
        "name"
    ]

    for record in (
        action_fields
    )
]


assert (
    action_field_names
    == EXPECTED_ACTION_FIELDS
), (
    "MutualContV1Action schema drift.\n"
    f"expected={EXPECTED_ACTION_FIELDS}\n"
    f"actual={action_field_names}"
)


print(
    "[PASS] V1_ACTION_SCHEMA_LOCK",
    flush=True,
)

print(
    " ",
    action_field_names,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Reference association-provider identity
# =============================================================================

reference_provider_sha = (
    literal_assignment(
        portable_tree,
        "REFERENCE_ASSOCIATION_PROVIDER_SHA256",
    )
)


assert (
    reference_provider_sha
    == EXPECTED_FAST_ASSOCIATION_SHA
)


reciprocal_file_sha = sha256_file(
    RECIPROCAL
)


assert (
    reciprocal_file_sha
    == EXPECTED_RECIPROCAL_FILE_SHA
)


reciprocal_text = RECIPROCAL.read_text(
    encoding="utf-8"
)


reciprocal_tree = ast.parse(
    reciprocal_text,
    filename=str(
        RECIPROCAL
    ),
)


reciprocal_functions = (
    top_level_functions(
        reciprocal_tree
    )
)


assert (
    "fast_association_pair"
    in reciprocal_functions
)


fast_assoc_record = function_record(
    text=reciprocal_text,
    fn=reciprocal_functions[
        "fast_association_pair"
    ],
)


assert (
    fast_assoc_record[
        "source_sha256"
    ]
    == EXPECTED_FAST_ASSOCIATION_SHA
)


assert (
    fast_assoc_record[
        "source_sha256"
    ]
    == reference_provider_sha
)


print(
    "[PASS] V1_ASSOCIATION_REFERENCE_IDENTITY_LOCK",
    flush=True,
)

print(
    "  reference SHA:",
    reference_provider_sha,
    flush=True,
)

print(
    "  fast_association_pair SHA:",
    fast_assoc_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Reciprocal-top1 algorithm reference
#
# ALGORITHM AUTHORITY ONLY.
# NOT historical V1 generator authority.
# =============================================================================

assert (
    "residual_reciprocal_pairs"
    in reciprocal_functions
), (
    "Expected residual_reciprocal_pairs reference absent."
)


reciprocal_fn = (
    reciprocal_functions[
        "residual_reciprocal_pairs"
    ]
)


reciprocal_record = function_record(
    text=reciprocal_text,
    fn=reciprocal_fn,
)


reciprocal_argmax = (
    argmax_records(
        reciprocal_text,
        reciprocal_fn,
    )
)


reciprocal_axes = {
    record[
        "axis_literal"
    ]

    for record in (
        reciprocal_argmax
    )

    if record[
        "axis_literal"
    ]
    is not None
}


assert (
    0
    in reciprocal_axes
)


assert (
    1
    in reciprocal_axes
)


print(
    "[PASS] RECIPROCAL_TOP1_ALGORITHM_REFERENCE_LOCK",
    flush=True,
)

print(
    "  function SHA:",
    reciprocal_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  argmax axes:",
    sorted(
        reciprocal_axes
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Current V1 matrix/action-core semantic audit
# =============================================================================

core_parameters = (
    function_parameters(
        core_fn
    )
)


core_argmax = (
    argmax_records(
        portable_text,
        core_fn,
    )
)


core_axes = {
    record[
        "axis_literal"
    ]

    for record in (
        core_argmax
    )

    if record[
        "axis_literal"
    ]
    is not None
}


core_comparisons = (
    comparison_records(
        portable_text,
        core_fn,
    )
)


core_action_constructors = (
    action_constructor_records(
        portable_text,
        core_fn,
    )
)


core_numeric_literals = (
    literal_float_values(
        core_fn
    )
)


core_source = (
    core_record[
        "source"
    ]
)


print(
    "=" * 112,
    flush=True,
)

print(
    "CURRENT V1 MATRIX/ACTION CORE",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "parameters:",
    core_parameters,
    flush=True,
)

print(
    "argmax axes:",
    sorted(
        core_axes
    ),
    flush=True,
)

print(
    "action constructors:",
    len(
        core_action_constructors
    ),
    flush=True,
)

print(
    flush=True,
)


for record in (
    core_argmax
):

    print(
        f"  ARGMAX L{record['line']} "
        f"axis={record['axis_source']}: "
        f"{record['source']}",
        flush=True,
    )


print(
    flush=True,
)


for record in (
    core_comparisons
):

    print(
        f"  CMP L{record['line']}: "
        f"{record['source']}",
        flush=True,
    )


print(
    flush=True,
)


for record in (
    core_action_constructors
):

    print(
        f"  ACTION L{record['line']}:",
        flush=True,
    )

    print(
        "   ",
        record[
            "source"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 15. Prove matrix-core compatibility with frozen specification
# =============================================================================

# -----------------------------------------------------------------------------
# 15A. No runtime scientific-threshold parameter
# -----------------------------------------------------------------------------

forbidden_runtime_threshold_parameters = [
    name

    for name in (
        core_parameters
    )

    if (
        "threshold"
        in name.lower()
        or name.lower().endswith(
            "_min"
        )
        or "cutoff"
        in name.lower()
    )
]


assert (
    forbidden_runtime_threshold_parameters
    == []
), (
    "V1 action core exposes a runtime scientific-threshold parameter: "
    f"{forbidden_runtime_threshold_parameters}"
)


# -----------------------------------------------------------------------------
# 15B. Source-side and target-side top1
# -----------------------------------------------------------------------------

assert (
    1
    in core_axes
), (
    "V1 action core lacks source-side axis=1 top1."
)


assert (
    0
    in core_axes
), (
    "V1 action core lacks target-side axis=0 top1."
)


# -----------------------------------------------------------------------------
# 15C. Reciprocal selection
# -----------------------------------------------------------------------------

reciprocal_logic_present = (
    len(
        core_argmax
    )
    >= 2
    and (
        "=="
        in core_source
        or
        " in "
        in core_source
    )
)


assert (
    reciprocal_logic_present
), (
    "Unable to establish reciprocal-top1 logic in current action core."
)


# -----------------------------------------------------------------------------
# 15D. Frozen probability threshold = 0.25
# -----------------------------------------------------------------------------

threshold_constant_names = []


for stmt in (
    portable_tree.body
):

    names = set()
    value = None


    if isinstance(
        stmt,
        ast.Assign,
    ):

        for target in (
            stmt.targets
        ):

            names |= (
                target_names(
                    target
                )
            )


        value = (
            stmt.value
        )


    elif isinstance(
        stmt,
        ast.AnnAssign,
    ):

        names |= (
            target_names(
                stmt.target
            )
        )


        value = (
            stmt.value
        )


    if (
        value
        is None
    ):

        continue


    for name in (
        names
    ):

        if (
            "V1"
            not in name.upper()
            or
            "PROB"
            not in name.upper()
        ):

            continue


        try:

            literal = (
                ast.literal_eval(
                    value
                )
            )


        except Exception:

            continue


        if (
            isinstance(
                literal,
                (
                    int,
                    float,
                ),
            )
            and not isinstance(
                literal,
                bool,
            )
            and float(
                literal
            )
            == 0.25
        ):

            threshold_constant_names.append(
                name
            )


threshold_literal_present = (
    0.25
    in core_numeric_literals
)


threshold_constant_used = any(
    name
    in core_source

    for name in (
        threshold_constant_names
    )
)


assert (
    threshold_literal_present
    or
    threshold_constant_used
), (
    "Frozen V1 threshold 0.25 is not visible in current action core."
)


# -----------------------------------------------------------------------------
# 15E. Source/target topology eligibility
# -----------------------------------------------------------------------------

lower_core = (
    core_source.lower()
)


source_outdegree_semantics = (
    "source_outdegree"
    in lower_core
    or
    "outdegree"
    in lower_core
    or
    "out_degree"
    in lower_core
)


target_indegree_semantics = (
    "target_indegree"
    in lower_core
    or
    "indegree"
    in lower_core
    or
    "in_degree"
    in lower_core
)


zero_comparison_sources = [
    record[
        "source"
    ]

    for record in (
        core_comparisons
    )

    if (
        "0"
        in record[
            "source"
        ]
    )
]


assert (
    source_outdegree_semantics
), (
    "V1 action core lacks source-final-outdegree semantics."
)


assert (
    target_indegree_semantics
), (
    "V1 action core lacks target-final-indegree semantics."
)


assert (
    len(
        zero_comparison_sources
    )
    >= 1
), (
    "V1 action core lacks zero-degree eligibility comparison."
)


# -----------------------------------------------------------------------------
# 15F. Action construction
# -----------------------------------------------------------------------------

assert (
    len(
        core_action_constructors
    )
    >= 1
), (
    "V1 action core does not construct MutualContV1Action."
)


constructor_text = "\n".join(
    record[
        "source"
    ]

    for record in (
        core_action_constructors
    )
)


for field in (
    EXPECTED_ACTION_FIELDS
):

    assert (
        field
        in constructor_text
    ), (
        f"V1 action constructor does not map field {field!r}."
    )


# -----------------------------------------------------------------------------
# 15G. Historical action artifact runtime independence
# -----------------------------------------------------------------------------

assert (
    "read_pickle"
    not in core_source
)


assert (
    EXPECTED_V1_ORACLE_PATH
    not in core_source
)


assert (
    "stage12_fold0_mutual_cont_v1_actions_locked.pkl"
    not in core_source
)


print(
    "[PASS] CURRENT_V1_MATRIX_CORE_MATCHES_FROZEN_SPECIFICATION",
    flush=True,
)

print(
    "  runtime scientific threshold parameter:",
    "NONE",
    flush=True,
)

print(
    "  source-side top1 axis=1:",
    True,
    flush=True,
)

print(
    "  target-side top1 axis=0:",
    True,
    flush=True,
)

print(
    "  reciprocal top1 logic:",
    reciprocal_logic_present,
    flush=True,
)

print(
    "  frozen probability threshold:",
    0.25,
    flush=True,
)

print(
    "  source final outdegree=0 semantics:",
    source_outdegree_semantics,
    flush=True,
)

print(
    "  target final indegree=0 semantics:",
    target_indegree_semantics,
    flush=True,
)

print(
    "  action schema:",
    action_field_names,
    flush=True,
)

print(
    "  historical action runtime lookup:",
    False,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Graph-level historical mutation authority
# =============================================================================

b12_sha = sha256_file(
    B12_WRITER
)


assert (
    b12_sha
    == EXPECTED_B12_WRITER_SHA
)


b12_text = B12_WRITER.read_text(
    encoding="utf-8"
)


b12_tree = ast.parse(
    b12_text,
    filename=str(
        B12_WRITER
    ),
)


b12_functions = (
    top_level_functions(
        b12_tree
    )
)


assert (
    "add_actions"
    in b12_functions
)


add_actions_fn = (
    b12_functions[
        "add_actions"
    ]
)


add_actions_record = (
    function_record(
        text=b12_text,
        fn=add_actions_fn,
    )
)


assert (
    add_actions_record[
        "source_sha256"
    ]
    == EXPECTED_ADD_ACTIONS_SHA
)


graph_mutation = mutation_contract(
    text=b12_text,
    fn=add_actions_fn,
)


for required in (
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_prob",
    "recovered_edge_dist",
    "add_edge",
    "validate_keys_true",
):

    assert (
        graph_mutation[
            required
        ]
    ), (
        "Historical graph-mutation contract missing "
        f"{required!r}."
    )


print(
    "[PASS] V1_GRAPH_MUTATION_AUTHORITY_LOCK",
    flush=True,
)

print(
    "  add_actions SHA:",
    add_actions_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  recovered_edge_dist:",
    graph_mutation[
        "recovered_edge_dist"
    ],
    flush=True,
)

print(
    "  graph.add_edge:",
    graph_mutation[
        "add_edge"
    ],
    flush=True,
)

print(
    "  validate_keys=True:",
    graph_mutation[
        "validate_keys_true"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 17. Freeze action -> historical mutation field adapter
# =============================================================================

action_to_historical_edge_adapter = {
    "dataset_id":
        "dataset",

    "source_node_id":
        "source_pred_node_id",

    "target_node_id":
        "target_pred_node_id",

    "candidate_prob":
        "candidate_prob",
}


assert (
    list(
        action_to_historical_edge_adapter.keys()
    )
    == EXPECTED_ACTION_FIELDS
)


print(
    "[PASS] V1_ACTION_TO_GRAPH_FIELD_ADAPTER_FROZEN",
    flush=True,
)

print(
    json.dumps(
        action_to_historical_edge_adapter,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 18. Graph-level placeholder boundary
# =============================================================================

placeholder_parameters = (
    function_parameters(
        graph_placeholder_fn
    )
)


assert (
    placeholder_parameters
    == [
        "graph",
        "dataset_id",
        "context",
    ]
), (
    "V1 graph boundary signature drift.\n"
    f"actual={placeholder_parameters}"
)


assert (
    "PortableImplementationIncompleteError"
    in placeholder_record[
        "calls"
    ]
)


print(
    "[PASS] V1_GRAPH_BOUNDARY_SIGNATURE_FROZEN",
    flush=True,
)

print(
    "  function:",
    "_reconstruct_portable_mutual_cont_v1",
    flush=True,
)

print(
    "  parameters:",
    placeholder_parameters,
    flush=True,
)

print(
    "  current status:",
    "PLACEHOLDER",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Inventory implementation surface required by 14.3D
# =============================================================================

context_classes = [
    name

    for name in (
        portable_classes
    )

    if (
        "Context"
        in name
        or
        "Provider"
        in name
        or
        "Association"
        in name
    )
]


association_named_functions = [
    name

    for name in (
        portable_functions
    )

    if (
        "association"
        in name.lower()
        or
        "graph_state"
        in name.lower()
    )
]


context_records = {}


for name in (
    context_classes
):

    cls = (
        portable_classes[
            name
        ]
    )


    source = node_source(
        portable_text,
        cls,
    )


    context_records[
        name
    ] = {
        "source_sha256":
            sha256_text(
                source
            ),

        "fields":
            class_fields(
                portable_text,
                cls,
            ),
    }


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3D IMPLEMENTATION SURFACE INVENTORY",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "context/provider classes:",
    context_classes,
    flush=True,
)

print(
    "association/graph-state functions:",
    association_named_functions,
    flush=True,
)

print(
    flush=True,
)


for name, record in (
    context_records.items()
):

    print(
        name,
        "SHA=",
        record[
            "source_sha256"
        ],
        flush=True,
    )


    for field in (
        record[
            "fields"
        ]
    ):

        print(
            "  ",
            field[
                "name"
            ],
            ":",
            field[
                "annotation"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 20. Freeze exact graph-runtime implementation contract
# =============================================================================

graph_runtime_contract = {
    "boundary_function":
        "_reconstruct_portable_mutual_cont_v1",

    "input": {
        "graph":
            "POST_MORPH_DIV_V2_GRAPH",

        "dataset_id":
            "CURRENT_DATASET_ID",

        "context":
            "PORTABLE_CURRENT_BEST_CONTEXT",
    },

    "scientific_authority": {
        "type":
            authority_model[
                "authority_type"
            ],

        "runtime_mode":
            authority_model[
                "runtime_mode"
            ],

        "source_extractable_generator":
            False,

        "threshold":
            0.25,

        "threshold_search_allowed":
            False,
    },

    "association": {
        "source":
            "EXPLICIT_ASSOCIATION_PROVIDER",

        "reference_function":
            "fast_association_pair",

        "reference_function_sha256":
            EXPECTED_FAST_ASSOCIATION_SHA,

        "reference_is_historical_v1_generator_authority":
            False,
    },

    "selection": {
        "source_side_top1":
            True,

        "target_side_top1":
            True,

        "reciprocal_top1_only":
            True,

        "candidate_prob_min":
            0.25,

        "source_final_outdegree":
            0,

        "target_final_indegree":
            0,

        "action_core":
            "reconstruct_mutual_cont_v1_actions",

        "action_core_sha256":
            core_record[
                "source_sha256"
            ],
    },

    "action_schema":
        action_field_names,

    "graph_mutation": {
        "distance":
            "fv9.recovered_edge_dist",

        "operation":
            "graph.add_edge",

        "edge_attributes": [
            "edge_dist",
            "edge_prob",
        ],

        "edge_prob_source":
            "MutualContV1Action.candidate_prob",

        "validate_keys":
            True,

        "historical_authority_function":
            "add_actions",

        "historical_authority_sha256":
            add_actions_record[
                "source_sha256"
            ],
    },

    "historical_oracle": {
        "path":
            EXPECTED_V1_ORACLE_PATH,

        "file_sha256":
            EXPECTED_V1_ORACLE_SHA,

        "runtime_use_allowed":
            False,

        "structural_fidelity_use_allowed":
            True,

        "may_change_threshold":
            False,

        "may_drive_performance_tuning":
            False,

        "may_drive_structural_debugging":
            True,
    },

    "required_before_portable_claim": {
        "exact_historical_structural_fidelity":
            True,

        "GT_dependency":
            0,

        "fold_membership_dependency":
            0,

        "scientific_parameter_search":
            0,
    },
}


print(
    "=" * 112,
    flush=True,
)

print(
    "FROZEN V1 GRAPH-RUNTIME SEMANTIC CONTRACT",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        graph_runtime_contract,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 21. Semantic gates
# =============================================================================

semantic_gates = {
    "parent_14_3c_r2_formally_locked":
        True,

    "corrected_architecture_exact_file_locked":
        (
            arch_file_sha
            == EXPECTED_ARCH_FILE_SHA
        ),

    "corrected_architecture_expected_content_sha_embedded":
        arch_contains_expected_content_sha,

    "architecture_lock_does_not_invent_top_level_schema":
        True,

    "old_writer_authority_remains_invalidated":
        (
            parent_obj[
                "resolution"
            ][
                "old_13_3b3_v1_writer_authority"
            ]
            == "INVALIDATED"
        ),

    "V1_source_extractable_generator_false":
        (
            authority_model[
                "source_extractable_generator"
            ]
            is False
        ),

    "V1_runtime_mode_specification_derived":
        (
            authority_model[
                "runtime_mode"
            ]
            == "SPECIFICATION_DERIVED_RECONSTRUCTION"
        ),

    "V1_threshold_exact_0_25":
        (
            float(
                frozen_spec[
                    "candidate_prob_min"
                ]
            )
            == 0.25
        ),

    "V1_scientific_parameter_search_forbidden":
        (
            authority_model[
                "scientific_parameter_search_allowed"
            ]
            is False
        ),

    "historical_oracle_runtime_forbidden":
        (
            oracle[
                "runtime_use_allowed"
            ]
            is False
        ),

    "historical_oracle_fidelity_use_allowed":
        (
            oracle[
                "structural_fidelity_use_allowed"
            ]
            is True
        ),

    "association_reference_identity_locked":
        (
            fast_assoc_record[
                "source_sha256"
            ]
            == EXPECTED_FAST_ASSOCIATION_SHA
        ),

    "portable_matrix_core_locked":
        (
            core_record[
                "source_sha256"
            ]
            == EXPECTED_V1_ACTION_CORE_SHA
        ),

    "source_side_top1_present":
        (
            1
            in core_axes
        ),

    "target_side_top1_present":
        (
            0
            in core_axes
        ),

    "reciprocal_top1_logic_present":
        reciprocal_logic_present,

    "source_outdegree_zero_semantics_present":
        source_outdegree_semantics,

    "target_indegree_zero_semantics_present":
        target_indegree_semantics,

    "action_schema_exact":
        (
            action_field_names
            == EXPECTED_ACTION_FIELDS
        ),

    "historical_graph_mutation_locked":
        (
            graph_mutation[
                "source_sha256"
            ]
            == EXPECTED_ADD_ACTIONS_SHA
        ),

    "historical_recovered_edge_dist_locked":
        graph_mutation[
            "recovered_edge_dist"
        ],

    "historical_add_edge_locked":
        graph_mutation[
            "add_edge"
        ],

    "historical_validate_keys_true":
        graph_mutation[
            "validate_keys_true"
        ],

    "current_graph_boundary_still_placeholder":
        (
            "PortableImplementationIncompleteError"
            in placeholder_record[
                "calls"
            ]
        ),

    "historical_v1_pickle_not_opened":
        True,

    "no_source_mutation":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3C-R3-R1 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok in (
        semantic_gates.items()
    )

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3C-R3-R1 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 22. Implementation authorization
# =============================================================================

implementation_authorized = (
    all(
        semantic_gates.values()
    )
)


assert (
    implementation_authorized
    is True
)


next_stage = (
    "NOTEBOOK14_14_3D_"
    "MUTUAL_CONT_V1_GRAPH_RUNTIME_IMPLEMENTATION"
)


# =============================================================================
# 23. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3C_R3_R1",

    "title": (
        "V1_CORRECTED_AUTHORITY_SEMANTIC_CONTRACT_FREEZE_"
        "ARCHITECTURE_CONTENT_LOCK_REPAIR"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_SEMANTIC_CONTRACT_FREEZE_ONLY",

    "failed_predecessor": {
        "stage":
            "NOTEBOOK14_14_3C_R3",

        "failure": (
            "Historical corrected-architecture artifact was incorrectly "
            "required to expose EXPECTED_ARCH_CONTENT_SHA as a top-level "
            "`content_sha256` field. Earlier provenance had frozen this "
            "artifact by exact file SHA plus recursive presence of the "
            "expected historical content SHA."
        ),

        "failure_point":
            "CORRECTED_ARCHITECTURE_CONTENT_LOCK",

        "source_mutation":
            False,

        "formal_artifact_written":
            False,
    },

    "parent_14_3c_r2": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            parent_lock[
                "file_sha256"
            ],

        "content_sha256":
            parent_lock[
                "content_sha256"
            ],
    },

    "corrected_architecture": {
        "path":
            str(
                ARCH_REL
            ),

        "file_sha256":
            arch_file_sha,

        "expected_historical_content_sha256":
            EXPECTED_ARCH_CONTENT_SHA,

        "top_level_content_sha256":
            arch_top_level_content_sha,

        "contains_expected_historical_content_sha256":
            arch_contains_expected_content_sha,

        "lock_mode":
            architecture_lock_mode,
    },

    "corrected_V1_authority": {
        "authority_model":
            authority_model,

        "runtime_reconstruction_contract":
            runtime_contract,
    },

    "portable_source": {
        "path":
            str(
                PORTABLE_REL
            ),

        "file_sha256":
            portable_sha,

        "matrix_action_core":
            core_record,

        "graph_boundary_placeholder":
            placeholder_record,

        "action_schema":
            action_fields,

        "argmax_records":
            core_argmax,

        "comparison_records":
            core_comparisons,

        "action_constructor_records":
            core_action_constructors,

        "reference_association_provider_sha256":
            reference_provider_sha,
    },

    "association_reference": {
        "path":
            str(
                RECIPROCAL_REL
            ),

        "file_sha256":
            reciprocal_file_sha,

        "fast_association_pair":
            fast_assoc_record,

        "reciprocal_top1_algorithm":
            reciprocal_record,

        "reciprocal_top1_argmax_records":
            reciprocal_argmax,

        "authority_role": (
            "ENGINEERING_REFERENCE_PROVIDER_AND_"
            "ALGORITHM_REFERENCE_NOT_HISTORICAL_V1_GENERATOR"
        ),
    },

    "historical_graph_mutation": {
        "path":
            str(
                B12_WRITER_REL
            ),

        "file_sha256":
            b12_sha,

        "function":
            add_actions_record,

        "contract":
            graph_mutation,
    },

    "action_to_historical_edge_adapter":
        action_to_historical_edge_adapter,

    "graph_runtime_contract":
        graph_runtime_contract,

    "context_provider_surface_inventory":
        context_records,

    "semantic_gates":
        semantic_gates,

    "implementation_authorized":
        implementation_authorized,

    "claim_boundary": {
        "V1_semantic_contract":
            "FROZEN",

        "V1_matrix_action_core":
            "SOURCE_LOCKED_AND_CONTRACT_COMPATIBLE",

        "V1_graph_runtime_implementation":
            "AUTHORIZED_NOT_YET_IMPLEMENTED",

        "historical_V1_action_oracle_runtime_use":
            False,

        "historical_V1_action_oracle_fidelity_use":
            True,

        "V1_structural_fidelity":
            "NOT_YET_VERIFIED",

        "post_V2_graph_provider":
            "NOT_YET_IMPLEMENTED",

        "post_V2_structural_fidelity":
            "NOT_YET_VERIFIED",

        "B12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "scientific_parameter_search":
            False,

        "GT_read":
            False,

        "evaluator_read":
            False,

        "performance_metric_generated":
            False,

        "source_mutation":
            False,
    },

    "explicit_non_claims": [
        "No V1 graph-level runtime was implemented in this stage.",
        "No historical V1 action pickle was opened.",
        "No historical V1 action artifact is authorized as runtime input.",
        "No V1 structural-fidelity claim.",
        "No post-V2 graph structural-fidelity claim.",
        "No B12 structural-fidelity claim.",
        "No performance evidence.",
        "No scientific parameter search.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3C-R3-R1 artifact already exists but differs "
        "from the exact semantic-contract freeze."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = (
    sha256_file(
        OUT
    )
)


# =============================================================================
# 24. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3C-R3-R1 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "ARCHITECTURE LOCK REPAIR:",
    flush=True,
)

print(
    "  exact file SHA:",
    arch_file_sha,
    flush=True,
)

print(
    "  expected historical content SHA:",
    EXPECTED_ARCH_CONTENT_SHA,
    flush=True,
)

print(
    "  expected content SHA embedded:",
    arch_contains_expected_content_sha,
    flush=True,
)

print(
    "  top-level content_sha256:",
    repr(
        arch_top_level_content_sha
    ),
    flush=True,
)

print(
    "  lock mode:",
    architecture_lock_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "CORRECTED V1 AUTHORITY:",
    flush=True,
)

print(
    "  authority type:",
    authority_model[
        "authority_type"
    ],
    flush=True,
)

print(
    "  source-extractable generator:",
    authority_model[
        "source_extractable_generator"
    ],
    flush=True,
)

print(
    "  runtime mode:",
    authority_model[
        "runtime_mode"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "FROZEN SCIENTIFIC POLICY:",
    flush=True,
)

print(
    "  source final outdegree = 0",
    flush=True,
)

print(
    "  target final indegree  = 0",
    flush=True,
)

print(
    "  reciprocal top1        = True",
    flush=True,
)

print(
    "  candidate prob min     = 0.25",
    flush=True,
)

print(
    "  operation              = ADD_SINGLE_CONTINUATION_EDGE",
    flush=True,
)

print(
    flush=True,
)


print(
    "CURRENT MATRIX CORE:",
    flush=True,
)

print(
    "  SHA:",
    core_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  argmax axes:",
    sorted(
        core_axes
    ),
    flush=True,
)

print(
    "  reciprocal logic:",
    reciprocal_logic_present,
    flush=True,
)

print(
    "  source outdegree gate:",
    source_outdegree_semantics,
    flush=True,
)

print(
    "  target indegree gate:",
    target_indegree_semantics,
    flush=True,
)

print(
    "  action schema:",
    action_field_names,
    flush=True,
)

print(
    flush=True,
)


print(
    "HISTORICAL GRAPH MUTATION:",
    flush=True,
)

print(
    "  authority function:",
    "add_actions",
    flush=True,
)

print(
    "  authority SHA:",
    add_actions_record[
        "source_sha256"
    ],
    flush=True,
)

print(
    "  recovered_edge_dist:",
    graph_mutation[
        "recovered_edge_dist"
    ],
    flush=True,
)

print(
    "  graph.add_edge:",
    graph_mutation[
        "add_edge"
    ],
    flush=True,
)

print(
    "  validate_keys=True:",
    graph_mutation[
        "validate_keys_true"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "HISTORICAL V1 ORACLE:",
    flush=True,
)

print(
    "  path:",
    oracle[
        "path"
    ],
    flush=True,
)

print(
    "  SHA:",
    oracle[
        "file_sha256"
    ],
    flush=True,
)

print(
    "  runtime use:",
    oracle[
        "runtime_use_allowed"
    ],
    flush=True,
)

print(
    "  fidelity use:",
    oracle[
        "structural_fidelity_use_allowed"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "IMPLEMENTATION AUTHORIZATION:",
    flush=True,
)

print(
    "  V1 semantic contract frozen:",
    True,
    flush=True,
)

print(
    "  V1 graph runtime implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "  structural fidelity verified:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  V1 semantic contract = FROZEN",
    flush=True,
)

print(
    "  V1 graph runtime = NOT YET IMPLEMENTED",
    flush=True,
)

print(
    "  historical action oracle runtime use = FORBIDDEN",
    flush=True,
)

print(
    "  V1 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  post-V2 provider = NOT YET IMPLEMENTED",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3D — MUTUAL-CONT-V1 GRAPH RUNTIME IMPLEMENTATION
#
# PARENT
# ------
# 14.3C-R3-R1 PASS
#
# Frozen authority:
#
#   V1 runtime mode:
#       SPECIFICATION_DERIVED_RECONSTRUCTION
#
#   Scientific policy:
#       source final outdegree = 0
#       target final indegree  = 0
#       reciprocal top1        = True
#       candidate_prob >= 0.25
#       ADD_SINGLE_CONTINUATION_EDGE
#
#   Matrix/action core:
#       reconstruct_mutual_cont_v1_actions
#
#   Association:
#       explicit association provider
#       reference identity SHA =
#       7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1
#
#   Historical graph mutation:
#       fv9.recovered_edge_dist(...)
#       graph.add_edge(
#           source,
#           target,
#           {
#               "edge_dist": float(dist),
#               "edge_prob": float(candidate_prob),
#           },
#           validate_keys=True,
#       )
#
#   Historical V1 action oracle:
#       runtime use = FORBIDDEN
#       structural-fidelity use = ALLOWED LATER
#
#
# THIS STAGE
# ----------
# 1. Lock 14.3C-R3-R1.
# 2. Lock exact pre-mutation portable source.
# 3. Lock V1 placeholder and the engineering helper contracts used by runtime.
# 4. Replace ONLY _reconstruct_portable_mutual_cont_v1.
# 5. Add dedicated regression tests.
# 6. py_compile.
# 7. Stream pytest live.
# 8. Prove all other top-level portable functions are source-identical.
# 9. Freeze formal implementation artifact.
#
#
# IMPORTANT
# ---------
# This stage implements V1 graph wiring only.
#
# It does NOT claim:
#   - V1 historical structural fidelity
#   - post-V2 provider fidelity
#   - B12 fidelity
#   - performance
#
# NO:
#   - GT
#   - evaluator
#   - historical V1 action pickle read
#   - threshold search
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import subprocess
import sys

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3c_r3_r1_"
    "v1_corrected_authority_semantic_contract_freeze_"
    "architecture_content_lock_repair_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


PRIMARY_TEST_REL = Path(
    "tests/test_stage12_portable_current_best.py"
)

PRIMARY_TEST = ROOT / PRIMARY_TEST_REL


NEW_TEST_REL = Path(
    "tests/test_stage12_v1_graph_runtime.py"
)

NEW_TEST = ROOT / NEW_TEST_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3d_"
    "mutual_cont_v1_graph_runtime_implementation_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "c3d995a663119fa6570499d7bcb9b7080130e52ffa91e0fde8044831069794ee"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "013eefefae40b15db5a97fab0f2acf47576b2e0bf3523e8a392489428fb83752"
)


EXPECTED_PORTABLE_PRE_SHA = (
    "a1fe0c9e5cd5e20a2ae5921bd4b4c55d078022c3348d5945bffa501ac3ff36ed"
)


EXPECTED_V1_PLACEHOLDER_SHA = (
    "ea9cba47ba6992c0dab67d747ef119061eb92b3cd52f0762274d6300a076e44e"
)

EXPECTED_V1_ACTION_CORE_SHA = (
    "1a9e988329cf91c9a8a706987fede592e0d223e69ec1113e02e42726f6d09449"
)


EXPECTED_CLONE_GRAPH_SHA = (
    "1af5ac3f989d46a556c89e81b98fffcea724a4cdbe739afe2bb500591d9d2f1b"
)

EXPECTED_V2_RESOLVE_PROVIDER_SHA = (
    "82ee0876ffc3a11ed6811cf5ad7fa7dd5a1c92c067549a68d127f23418f4df96"
)

EXPECTED_V2_VALIDATE_OUTPUT_SHA = (
    "60283a40dbe245d57a9c13d809be9856f6a63cc19e33c0af3f34f65ace7b178a"
)

EXPECTED_R1P50_GRAPH_STATE_SHA = (
    "274917d87fc8c14fd9115a1a12e9172d226f089d56182a9b8f7a4a3c2a3118b0"
)


EXPECTED_ASSOCIATION_PROVIDER_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


HISTORICAL_ORACLE_FILENAME = (
    "stage12_fold0_mutual_cont_v1_actions_locked.pkl"
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    data: bytes,
) -> str:

    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(
    text: str,
) -> str:

    return sha256_bytes(
        text.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    result = ast.get_source_segment(
        text,
        node,
    )


    assert (
        result
        is not None
    )


    return result


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def function_source_sha(
    text: str,
    fn: ast.AST,
) -> str:

    return sha256_text(
        node_source(
            text,
            fn,
        )
    )


def function_calls(
    fn: ast.AST,
) -> list[str]:

    return sorted({
        dotted_name(
            call.func
        )
        or "<dynamic-call>"

        for call in ast.walk(
            fn
        )

        if isinstance(
            call,
            ast.Call,
        )
    })


def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert path.exists(), (
        path
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    ), (
        "14.3C-R3-R1 parent file SHA mismatch."
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "implementation_authorized"
        ]
        is True
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "V1_semantic_contract"
        ]
        == "FROZEN"
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "V1_graph_runtime_implementation"
        ]
        == "AUTHORIZED_NOT_YET_IMPLEMENTED"
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "historical_V1_action_oracle_runtime_use"
        ]
        is False
    )


    return obj


def stream_command(
    command: list[str],
    *,
    cwd: Path,
) -> int:

    print(
        "$",
        " ".join(
            command
        ),
        flush=True,
    )


    process = subprocess.Popen(
        command,
        cwd=str(
            cwd
        ),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )


    assert (
        process.stdout
        is not None
    )


    for line in (
        process.stdout
    ):

        print(
            line,
            end="",
            flush=True,
        )


    return int(
        process.wait()
    )


# =============================================================================
# 3. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    PRIMARY_TEST,
):

    assert (
        required.exists()
    ), required


assert (
    not NEW_TEST.exists()
), (
    f"Dedicated 14.3D test already exists unexpectedly: {NEW_TEST}"
)


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3D "
    "MUTUAL-CONT-V1 GRAPH RUNTIME IMPLEMENTATION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "SOURCE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "NEW TEST:",
    NEW_TEST_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 4. Parent authorization lock
# =============================================================================

parent_obj = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3C_R3_R1_AUTHORIZATION_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  implementation authorized:",
    parent_obj[
        "implementation_authorized"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Pre-mutation source lock
# =============================================================================

portable_pre_sha = sha256_file(
    PORTABLE
)


assert (
    portable_pre_sha
    == EXPECTED_PORTABLE_PRE_SHA
), (
    "Portable source drift before 14.3D.\n"
    f"expected={EXPECTED_PORTABLE_PRE_SHA}\n"
    f"actual  ={portable_pre_sha}"
)


pre_text = PORTABLE.read_text(
    encoding="utf-8"
)


pre_tree = ast.parse(
    pre_text,
    filename=str(
        PORTABLE
    ),
)


pre_functions = top_level_functions(
    pre_tree
)


required_functions = [
    "_reconstruct_portable_mutual_cont_v1",
    "reconstruct_mutual_cont_v1_actions",
    "_clone_portable_morph_graph",
    "_portable_v2_resolve_association_provider",
    "_portable_v2_validate_association_output",
    "_portable_r1p50_source_graph_state",
]


missing = [
    name

    for name in (
        required_functions
    )

    if name not in (
        pre_functions
    )
]


assert (
    missing
    == []
), (
    f"Missing required current helpers: {missing}"
)


expected_function_shas = {
    "_reconstruct_portable_mutual_cont_v1":
        EXPECTED_V1_PLACEHOLDER_SHA,

    "reconstruct_mutual_cont_v1_actions":
        EXPECTED_V1_ACTION_CORE_SHA,

    "_clone_portable_morph_graph":
        EXPECTED_CLONE_GRAPH_SHA,

    "_portable_v2_resolve_association_provider":
        EXPECTED_V2_RESOLVE_PROVIDER_SHA,

    "_portable_v2_validate_association_output":
        EXPECTED_V2_VALIDATE_OUTPUT_SHA,

    "_portable_r1p50_source_graph_state":
        EXPECTED_R1P50_GRAPH_STATE_SHA,
}


for name, expected_sha in (
    expected_function_shas.items()
):

    actual_sha = function_source_sha(
        pre_text,
        pre_functions[
            name
        ],
    )


    assert (
        actual_sha
        == expected_sha
    ), (
        f"Pre-mutation function drift: {name}\n"
        f"expected={expected_sha}\n"
        f"actual  ={actual_sha}"
    )


print(
    "[PASS] PRE_MUTATION_SOURCE_AND_HELPER_LOCKS",
    flush=True,
)

print(
    "  portable SHA:",
    portable_pre_sha,
    flush=True,
)


for name, expected_sha in (
    expected_function_shas.items()
):

    print(
        " ",
        name,
        expected_sha,
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 6. Freeze all pre-mutation top-level function SHAs
# =============================================================================

pre_top_level_shas = {
    name:
        function_source_sha(
            pre_text,
            fn,
        )

    for name, fn in (
        pre_functions.items()
    )
}


target_pre_fn = (
    pre_functions[
        "_reconstruct_portable_mutual_cont_v1"
    ]
)


target_pre_source = node_source(
    pre_text,
    target_pre_fn,
)


assert (
    sha256_text(
        target_pre_source
    )
    == EXPECTED_V1_PLACEHOLDER_SHA
)


assert (
    "PortableImplementationIncompleteError"
    in target_pre_source
)


print(
    "[PASS] EXACT_V1_PLACEHOLDER_FROZEN",
    flush=True,
)

print(
    "  SHA:",
    EXPECTED_V1_PLACEHOLDER_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Candidate implementation
#
# Notes:
#
# - uses existing explicit association-provider contract;
# - exact provider identity must equal frozen reference identity;
# - uses the already-frozen matrix/action core;
# - uses graph state extracted from the post-Morph graph;
# - clones input graph before mutation;
# - never reads historical V1 action artifact;
# - graph edge distance uses canonical frozen_v9 recovered_edge_dist;
# - candidate probability becomes historical edge_prob;
# - topology dictionaries are updated after each accepted action so later
#   frame-pair calls see the true current graph state.
# =============================================================================

NEW_FUNCTION_SOURCE = r'''def _reconstruct_portable_mutual_cont_v1(
    graph: Any,
    dataset_id: str,
    *,
    context: PortableCurrentBestContext,
) -> Any:
    """Reconstruct frozen MUTUAL-CONT-V1 from the corrected specification.

    Runtime authority is specification-derived reconstruction only.

    Historical locked V1 actions are forbidden as runtime input. They remain
    available only for a later structural-fidelity replay.

    Frozen scientific policy:
        - reciprocal source/target top-1
        - candidate probability >= 0.25
        - source final outdegree == 0
        - target final indegree == 0
        - add one continuation edge

    The association provider is explicit in ``context`` and is identity-locked
    to the frozen engineering reference provider.
    """

    dataset_id = _validate_dataset_id(
        dataset_id
    )

    if not isinstance(
        context,
        PortableCurrentBestContext,
    ):

        raise PortableRuntimeContractError(
            "context must be PortableCurrentBestContext."
        )

    (
        provider,
        provider_identity,
    ) = _portable_v2_resolve_association_provider(
        context
    )

    if (
        str(
            provider_identity
        )
        !=
        REFERENCE_ASSOCIATION_PROVIDER_SHA256
    ):

        raise PortableRuntimeContractError(
            "MUTUAL-CONT-V1 association-provider identity drift: "
            f"expected {REFERENCE_ASSOCIATION_PROVIDER_SHA256}, "
            f"got {provider_identity}."
        )

    output_graph = _clone_portable_morph_graph(
        graph
    )

    (
        nodes,
        _,
        source_outdegree,
        target_indegree,
    ) = _portable_r1p50_source_graph_state(
        output_graph
    )

    provider_nodes = (
        nodes
        .reset_index()
        .copy(
            deep=True
        )
    )

    node_ids = {
        int(
            value
        )
        for value in (
            nodes.index.tolist()
        )
    }

    times = sorted(
        {
            int(
                value
            )
            for value in (
                provider_nodes[
                    "t"
                ].tolist()
            )
        }
    )

    for t in times:

        result = provider(
            dataset_id,
            provider_nodes.copy(
                deep=True
            ),
            int(
                t
            ),
            context=context,
        )

        (
            source_node_ids,
            target_node_ids,
            candidate_prob_matrix,
        ) = _portable_v2_validate_association_output(
            result
        )

        if (
            candidate_prob_matrix
            is None
        ):

            continue

        missing_sources = sorted(
            {
                int(
                    value
                )
                for value in (
                    source_node_ids
                )
                if int(
                    value
                )
                not in node_ids
            }
        )

        missing_targets = sorted(
            {
                int(
                    value
                )
                for value in (
                    target_node_ids
                )
                if int(
                    value
                )
                not in node_ids
            }
        )

        if (
            missing_sources
            or missing_targets
        ):

            raise PortableRuntimeContractError(
                "MUTUAL-CONT-V1 association provider returned node IDs "
                "outside the current post-Morph graph: "
                f"missing_sources={missing_sources}, "
                f"missing_targets={missing_targets}."
            )

        actions = reconstruct_mutual_cont_v1_actions(
            dataset_id,
            source_node_ids,
            target_node_ids,
            candidate_prob_matrix,
            source_outdegree=source_outdegree,
            target_indegree=target_indegree,
        )

        for action in actions:

            source = int(
                action.source_node_id
            )

            target = int(
                action.target_node_id
            )

            probability = float(
                action.candidate_prob
            )

            if (
                source_outdegree.get(
                    source,
                    0,
                )
                !=
                0
            ):

                continue

            if (
                target_indegree.get(
                    target,
                    0,
                )
                !=
                0
            ):

                continue

            distance = _frozen_v9.recovered_edge_dist(
                nodes,
                source,
                target,
                _frozen_v9.FROZEN_V9_CONFIG,
            )

            output_graph.add_edge(
                source,
                target,
                {
                    "edge_dist":
                        float(
                            distance
                        ),

                    "edge_prob":
                        probability,
                },
                validate_keys=True,
            )

            source_outdegree[
                source
            ] = (
                source_outdegree.get(
                    source,
                    0,
                )
                +
                1
            )

            target_indegree[
                target
            ] = (
                target_indegree.get(
                    target,
                    0,
                )
                +
                1
            )

    return output_graph
'''


# =============================================================================
# 8. Candidate static semantic checks
# =============================================================================

candidate_fn_tree = ast.parse(
    NEW_FUNCTION_SOURCE
)


candidate_functions = top_level_functions(
    candidate_fn_tree
)


assert (
    list(
        candidate_functions
    )
    == [
        "_reconstruct_portable_mutual_cont_v1"
    ]
)


candidate_fn = candidate_functions[
    "_reconstruct_portable_mutual_cont_v1"
]


candidate_calls = set(
    function_calls(
        candidate_fn
    )
)


required_candidate_calls = {
    "_validate_dataset_id",
    "_portable_v2_resolve_association_provider",
    "_clone_portable_morph_graph",
    "_portable_r1p50_source_graph_state",
    "_portable_v2_validate_association_output",
    "reconstruct_mutual_cont_v1_actions",
    "_frozen_v9.recovered_edge_dist",
    "output_graph.add_edge",
}


missing_candidate_calls = sorted(
    required_candidate_calls
    -
    candidate_calls
)


assert (
    missing_candidate_calls
    == []
), (
    "Candidate implementation is missing required calls: "
    f"{missing_candidate_calls}"
)


for forbidden in (
    "read_pickle",
    HISTORICAL_ORACLE_FILENAME,
    "evaluate_graph",
    "ground_truth",
    "gt_graph",
):

    assert (
        forbidden
        not in NEW_FUNCTION_SOURCE
    ), (
        f"Forbidden V1 runtime dependency/signal present: {forbidden}"
    )


assert (
    "REFERENCE_ASSOCIATION_PROVIDER_SHA256"
    in NEW_FUNCTION_SOURCE
)


assert (
    "validate_keys=True"
    in NEW_FUNCTION_SOURCE.replace(
        " ",
        ""
    )
)


assert (
    '"edge_dist"'
    in NEW_FUNCTION_SOURCE
)


assert (
    '"edge_prob"'
    in NEW_FUNCTION_SOURCE
)


print(
    "[PASS] CANDIDATE_V1_RUNTIME_STATIC_CONTRACT",
    flush=True,
)

print(
    "  required calls:",
    sorted(
        required_candidate_calls
    ),
    flush=True,
)

print(
    "  historical action runtime lookup:",
    False,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Replace ONLY target function using AST source span
# =============================================================================

pre_lines = pre_text.splitlines(
    keepends=True
)


start_index = (
    target_pre_fn.lineno
    -
    1
)

end_index = (
    target_pre_fn.end_lineno
)


candidate_text = (
    "".join(
        pre_lines[
            :start_index
        ]
    )
    +
    NEW_FUNCTION_SOURCE
    +
    "\n\n"
    +
    "".join(
        pre_lines[
            end_index:
        ]
    )
)


# Candidate must compile before writing.
compile(
    candidate_text,
    str(
        PORTABLE
    ),
    "exec",
)


candidate_tree = ast.parse(
    candidate_text,
    filename=str(
        PORTABLE
    ),
)


candidate_top_functions = top_level_functions(
    candidate_tree
)


assert (
    set(
        candidate_top_functions
    )
    ==
    set(
        pre_functions
    )
), (
    "Top-level function set changed unexpectedly."
)


candidate_target_sha = function_source_sha(
    candidate_text,
    candidate_top_functions[
        "_reconstruct_portable_mutual_cont_v1"
    ],
)


assert (
    candidate_target_sha
    != EXPECTED_V1_PLACEHOLDER_SHA
)


assert (
    "PortableImplementationIncompleteError"
    not in node_source(
        candidate_text,
        candidate_top_functions[
            "_reconstruct_portable_mutual_cont_v1"
        ],
    )
)


# All other top-level functions must remain byte-source-identical.
other_function_drift = []


for name, pre_sha in (
    pre_top_level_shas.items()
):

    if (
        name
        == "_reconstruct_portable_mutual_cont_v1"
    ):

        continue


    candidate_sha = function_source_sha(
        candidate_text,
        candidate_top_functions[
            name
        ],
    )


    if (
        candidate_sha
        != pre_sha
    ):

        other_function_drift.append({
            "function":
                name,

            "before":
                pre_sha,

            "after":
                candidate_sha,
        })


assert (
    other_function_drift
    == []
), (
    "Unexpected non-V1 top-level source drift: "
    f"{other_function_drift}"
)


candidate_file_sha = sha256_text(
    candidate_text
)


print(
    "[PASS] SINGLE_FUNCTION_CANDIDATE_DIFF_PROVEN",
    flush=True,
)

print(
    "  pre file SHA:",
    portable_pre_sha,
    flush=True,
)

print(
    "  candidate file SHA:",
    candidate_file_sha,
    flush=True,
)

print(
    "  pre target SHA:",
    EXPECTED_V1_PLACEHOLDER_SHA,
    flush=True,
)

print(
    "  candidate target SHA:",
    candidate_target_sha,
    flush=True,
)

print(
    "  other top-level function drift:",
    other_function_drift,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Dedicated regression test
# =============================================================================

TEST_SOURCE = r'''from __future__ import annotations

import inspect

from pathlib import Path

import numpy as np
import pandas as pd
import pytest

import biohub_cell_tracking.stage12_portable_current_best as portable


class _Table:
    def __init__(self, frame):
        self._frame = frame.copy(deep=True)

    def to_pandas(self):
        return self._frame.copy(deep=True)


class _Graph:
    def __init__(self, nodes, edges=None):
        self._nodes = pd.DataFrame(nodes).copy(deep=True)

        if edges is None:
            edges = []

        self._edges = pd.DataFrame(
            edges,
            columns=[
                "edge_id",
                "source_id",
                "target_id",
                "edge_dist",
                "edge_prob",
            ],
        ).copy(deep=True)

    def node_attrs(self, attr_keys=None, unpack=True):
        frame = self._nodes.copy(deep=True)

        if attr_keys is not None:
            frame = frame[
                list(attr_keys)
            ].copy(deep=True)

        return _Table(frame)

    def edge_attrs(self, attr_keys=None, unpack=True):
        frame = self._edges.copy(deep=True)

        if attr_keys is not None:
            requested = list(attr_keys)

            base = [
                column
                for column in (
                    "edge_id",
                    "source_id",
                    "target_id",
                )
                if column in frame.columns
            ]

            requested = base + [
                column
                for column in requested
                if column not in base
            ]

            frame = frame[
                requested
            ].copy(deep=True)

        return _Table(frame)

    def add_edge(
        self,
        source,
        target,
        attrs,
        validate_keys=False,
    ):
        assert validate_keys is True

        record = {
            "edge_id":
                int(len(self._edges)),

            "source_id":
                int(source),

            "target_id":
                int(target),

            "edge_dist":
                float(attrs["edge_dist"]),

            "edge_prob":
                float(attrs["edge_prob"]),
        }

        self._edges = pd.concat(
            [
                self._edges,
                pd.DataFrame([record]),
            ],
            ignore_index=True,
        )

        return int(
            record[
                "edge_id"
            ]
        )


def _graph():
    return _Graph(
        [
            {
                "node_id": 1,
                "t": 0,
                "z": 0.0,
                "y": 0.0,
                "x": 0.0,
            },
            {
                "node_id": 2,
                "t": 1,
                "z": 0.0,
                "y": 3.0,
                "x": 4.0,
            },
        ]
    )


def _context(provider, identity=None):
    if identity is None:
        identity = (
            portable.
            REFERENCE_ASSOCIATION_PROVIDER_SHA256
        )

    return portable.PortableCurrentBestContext(
        b256_pred_995_root=Path("."),
        b256_pred_9975_root=Path("."),
        s9995_pred_995_root=Path("."),
        s9995_pred_9975_root=Path("."),
        dataset_root=Path("."),
        physical_scale_zyx_um=(1.0, 1.0, 1.0),
        frozen_v9_policy_bundle=object(),
        association_provider=provider,
        association_provider_identity_sha256=identity,
        strict=True,
    )


def _single_pair_provider(probability):
    calls = []

    def provider(
        dataset_id,
        nodes,
        t,
        *,
        context,
    ):
        calls.append(
            (
                dataset_id,
                int(t),
                tuple(nodes["node_id"].astype(int).tolist()),
            )
        )

        if int(t) == 0:
            return (
                np.asarray(
                    [1],
                    dtype=int,
                ),
                np.asarray(
                    [2],
                    dtype=int,
                ),
                np.asarray(
                    [[float(probability)]],
                    dtype=float,
                ),
            )

        return (
            np.asarray(
                [],
                dtype=int,
            ),
            np.asarray(
                [],
                dtype=int,
            ),
            np.empty(
                (0, 0),
                dtype=float,
            ),
        )

    provider.calls = calls

    return provider


def test_v1_graph_runtime_adds_frozen_continuation_edge(
    monkeypatch,
):
    graph = _graph()

    provider = _single_pair_provider(
        0.80
    )

    monkeypatch.setattr(
        portable._frozen_v9,
        "recovered_edge_dist",
        lambda nodes, source, target, config: 5.0,
    )

    result = (
        portable.
        _reconstruct_portable_mutual_cont_v1(
            graph,
            "ds",
            context=_context(
                provider
            ),
        )
    )

    # Runtime works on a clone.
    assert result is not graph

    assert len(
        graph._edges
    ) == 0

    assert len(
        result._edges
    ) == 1

    row = result._edges.iloc[0]

    assert int(
        row["source_id"]
    ) == 1

    assert int(
        row["target_id"]
    ) == 2

    assert float(
        row["edge_dist"]
    ) == pytest.approx(
        5.0
    )

    assert float(
        row["edge_prob"]
    ) == pytest.approx(
        0.80
    )

    assert provider.calls[0][0] == "ds"

    assert provider.calls[0][1] == 0


def test_v1_graph_runtime_keeps_exact_threshold_boundary(
    monkeypatch,
):
    monkeypatch.setattr(
        portable._frozen_v9,
        "recovered_edge_dist",
        lambda nodes, source, target, config: 1.0,
    )

    at_threshold = (
        portable.
        _reconstruct_portable_mutual_cont_v1(
            _graph(),
            "ds",
            context=_context(
                _single_pair_provider(
                    0.25
                )
            ),
        )
    )

    below_threshold = (
        portable.
        _reconstruct_portable_mutual_cont_v1(
            _graph(),
            "ds",
            context=_context(
                _single_pair_provider(
                    0.249999
                )
            ),
        )
    )

    assert len(
        at_threshold._edges
    ) == 1

    assert len(
        below_threshold._edges
    ) == 0


def test_v1_graph_runtime_respects_source_outdegree_zero_gate(
    monkeypatch,
):
    graph = _Graph(
        [
            {
                "node_id": 1,
                "t": 0,
                "z": 0.0,
                "y": 0.0,
                "x": 0.0,
            },
            {
                "node_id": 2,
                "t": 1,
                "z": 0.0,
                "y": 3.0,
                "x": 4.0,
            },
            {
                "node_id": 3,
                "t": 1,
                "z": 0.0,
                "y": 1.0,
                "x": 1.0,
            },
        ],
        [
            {
                "edge_id": 0,
                "source_id": 1,
                "target_id": 3,
                "edge_dist": 1.0,
                "edge_prob": 0.9,
            },
        ],
    )

    monkeypatch.setattr(
        portable._frozen_v9,
        "recovered_edge_dist",
        lambda nodes, source, target, config: 5.0,
    )

    result = (
        portable.
        _reconstruct_portable_mutual_cont_v1(
            graph,
            "ds",
            context=_context(
                _single_pair_provider(
                    0.95
                )
            ),
        )
    )

    # Existing edge only; source 1 is not a free endpoint.
    assert len(
        result._edges
    ) == 1

    row = result._edges.iloc[0]

    assert (
        int(
            row["source_id"]
        ),
        int(
            row["target_id"]
        ),
    ) == (
        1,
        3,
    )


def test_v1_graph_runtime_fails_closed_on_provider_identity_drift():
    provider = _single_pair_provider(
        0.90
    )

    with pytest.raises(
        portable.PortableRuntimeContractError,
        match="association-provider identity drift",
    ):
        (
            portable.
            _reconstruct_portable_mutual_cont_v1(
                _graph(),
                "ds",
                context=_context(
                    provider,
                    identity="f" * 64,
                ),
            )
        )


def test_v1_graph_runtime_fails_closed_on_unknown_provider_nodes():
    def provider(
        dataset_id,
        nodes,
        t,
        *,
        context,
    ):
        if int(t) == 0:
            return (
                np.asarray(
                    [999],
                    dtype=int,
                ),
                np.asarray(
                    [2],
                    dtype=int,
                ),
                np.asarray(
                    [[0.90]],
                    dtype=float,
                ),
            )

        return (
            np.asarray([], dtype=int),
            np.asarray([], dtype=int),
            np.empty((0, 0), dtype=float),
        )

    with pytest.raises(
        portable.PortableRuntimeContractError,
        match="outside the current post-Morph graph",
    ):
        (
            portable.
            _reconstruct_portable_mutual_cont_v1(
                _graph(),
                "ds",
                context=_context(
                    provider
                ),
            )
        )


def test_v1_graph_runtime_contains_no_historical_action_lookup():
    source = inspect.getsource(
        portable.
        _reconstruct_portable_mutual_cont_v1
    )

    assert "read_pickle" not in source

    assert (
        "stage12_fold0_mutual_cont_v1_actions_locked.pkl"
        not in source
    )

    assert "evaluate_graph" not in source
'''


# Compile test source before writing.
compile(
    TEST_SOURCE,
    str(
        NEW_TEST
    ),
    "exec",
)


test_sha = sha256_text(
    TEST_SOURCE
)


print(
    "[PASS] DEDICATED_TEST_SOURCE_COMPILES_IN_MEMORY",
    flush=True,
)

print(
    "  test SHA:",
    test_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Write source + dedicated test
# =============================================================================

PORTABLE.write_text(
    candidate_text,
    encoding="utf-8",
)


NEW_TEST.write_text(
    TEST_SOURCE,
    encoding="utf-8",
)


assert (
    PORTABLE.read_text(
        encoding="utf-8"
    )
    == candidate_text
)


assert (
    NEW_TEST.read_text(
        encoding="utf-8"
    )
    == TEST_SOURCE
)


portable_post_write_sha = (
    sha256_file(
        PORTABLE
    )
)


test_file_sha = (
    sha256_file(
        NEW_TEST
    )
)


assert (
    portable_post_write_sha
    == candidate_file_sha
)


assert (
    test_file_sha
    == test_sha
)


print(
    "[PASS] SOURCE_AND_TEST_WRITTEN",
    flush=True,
)

print(
    "  portable post-write SHA:",
    portable_post_write_sha,
    flush=True,
)

print(
    "  test file SHA:",
    test_file_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. py_compile — live
# =============================================================================

compile_exit = stream_command(
    [
        sys.executable,
        "-m",
        "py_compile",
        str(
            PORTABLE
        ),
        str(
            NEW_TEST
        ),
    ],
    cwd=ROOT,
)


assert (
    compile_exit
    == 0
), (
    f"py_compile failed with exit={compile_exit}"
)


print(
    "[PASS] PY_COMPILE",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Dedicated + portable regression pytest — LIVE
# =============================================================================

pytest_exit = stream_command(
    [
        sys.executable,
        "-m",
        "pytest",
        "-vv",
        "-s",
        str(
            NEW_TEST_REL
        ),
        str(
            PRIMARY_TEST_REL
        ),
    ],
    cwd=ROOT,
)


assert (
    pytest_exit
    == 0
), (
    f"Targeted pytest failed with exit={pytest_exit}"
)


print(
    "[PASS] TARGETED_PYTEST",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Post-test exact source audit
# =============================================================================

portable_post_sha = sha256_file(
    PORTABLE
)


assert (
    portable_post_sha
    == candidate_file_sha
), (
    "Portable source changed during tests."
)


post_text = PORTABLE.read_text(
    encoding="utf-8"
)


post_tree = ast.parse(
    post_text,
    filename=str(
        PORTABLE
    ),
)


post_functions = top_level_functions(
    post_tree
)


assert (
    set(
        post_functions
    )
    ==
    set(
        pre_functions
    )
)


target_post_fn = (
    post_functions[
        "_reconstruct_portable_mutual_cont_v1"
    ]
)


target_post_source = node_source(
    post_text,
    target_post_fn,
)


target_post_sha = sha256_text(
    target_post_source
)


assert (
    target_post_sha
    == candidate_target_sha
)


assert (
    "PortableImplementationIncompleteError"
    not in target_post_source
)


# Prove every other top-level function is unchanged.
post_other_function_drift = []


for name, pre_sha in (
    pre_top_level_shas.items()
):

    if (
        name
        == "_reconstruct_portable_mutual_cont_v1"
    ):

        continue


    actual_sha = function_source_sha(
        post_text,
        post_functions[
            name
        ],
    )


    if (
        actual_sha
        != pre_sha
    ):

        post_other_function_drift.append({
            "function":
                name,

            "before":
                pre_sha,

            "after":
                actual_sha,
        })


assert (
    post_other_function_drift
    == []
), (
    "Non-V1 function changed during 14.3D: "
    f"{post_other_function_drift}"
)


print(
    "[PASS] SINGLE_FUNCTION_POST_MUTATION_PROOF",
    flush=True,
)

print(
    "  portable pre SHA:",
    portable_pre_sha,
    flush=True,
)

print(
    "  portable post SHA:",
    portable_post_sha,
    flush=True,
)

print(
    "  V1 placeholder pre SHA:",
    EXPECTED_V1_PLACEHOLDER_SHA,
    flush=True,
)

print(
    "  V1 runtime post SHA:",
    target_post_sha,
    flush=True,
)

print(
    "  other-function drift:",
    post_other_function_drift,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Post-implementation semantic isolation
# =============================================================================

post_calls = set(
    function_calls(
        target_post_fn
    )
)


post_semantic_checks = {
    "uses_frozen_matrix_core":
        (
            "reconstruct_mutual_cont_v1_actions"
            in post_calls
        ),

    "uses_explicit_provider_resolver":
        (
            "_portable_v2_resolve_association_provider"
            in post_calls
        ),

    "validates_provider_output":
        (
            "_portable_v2_validate_association_output"
            in post_calls
        ),

    "clones_input_graph":
        (
            "_clone_portable_morph_graph"
            in post_calls
        ),

    "uses_current_graph_state":
        (
            "_portable_r1p50_source_graph_state"
            in post_calls
        ),

    "uses_historical_distance_semantics":
        (
            "_frozen_v9.recovered_edge_dist"
            in post_calls
        ),

    "mutates_graph_by_add_edge":
        (
            "output_graph.add_edge"
            in post_calls
        ),

    "contains_validate_keys_true":
        (
            "validate_keys=True"
            in target_post_source.replace(
                " ",
                ""
            )
        ),

    "historical_action_pickle_absent":
        (
            HISTORICAL_ORACLE_FILENAME
            not in target_post_source
        ),

    "read_pickle_absent":
        (
            "read_pickle"
            not in target_post_source
        ),

    "evaluator_absent":
        (
            "evaluate_graph"
            not in target_post_source
        ),

    "GT_absent":
        (
            "gt_graph"
            not in target_post_source
            and
            "ground_truth"
            not in target_post_source
        ),
}


for gate, ok in (
    post_semantic_checks.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_semantics = [
    name

    for name, ok in (
        post_semantic_checks.items()
    )

    if not ok
]


assert (
    failed_semantics
    == []
), (
    "14.3D semantic isolation failures: "
    f"{failed_semantics}"
)


print(
    "[PASS] POST_IMPLEMENTATION_SEMANTIC_ISOLATION",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Preserve repaired V2 invented-sort invariant
# =============================================================================

forbidden_v2_sort_fragment = (
    'actions.sort_values(["dataset", "t", '
    '"source_pred_node_id", "target_pred_node_id"], '
    'kind="mergesort")'
)


normalized_post = (
    post_text.replace(
        "\n",
        " "
    )
)


# AST-level exact reconstruction is safer than whitespace search.
exact_forbidden_sort_matches = []


for call in ast.walk(
    post_tree
):

    if not isinstance(
        call,
        ast.Call,
    ):

        continue


    func_name = dotted_name(
        call.func
    )


    if (
        func_name
        != "actions.sort_values"
    ):

        continue


    if not call.args:

        continue


    try:

        columns = ast.literal_eval(
            call.args[
                0
            ]
        )

    except Exception:

        continue


    kind = None


    for kw in (
        call.keywords
    ):

        if (
            kw.arg
            == "kind"
        ):

            try:

                kind = ast.literal_eval(
                    kw.value
                )

            except Exception:

                kind = None


    if (
        columns
        == [
            "dataset",
            "t",
            "source_pred_node_id",
            "target_pred_node_id",
        ]
        and kind
        == "mergesort"
    ):

        exact_forbidden_sort_matches.append(
            {
                "line":
                    call.lineno,

                "source":
                    node_source(
                        post_text,
                        call,
                    ),
            }
        )


assert (
    exact_forbidden_sort_matches
    == []
), (
    "Previously repaired V2 invented sort reappeared."
)


print(
    "[PASS] V2_INVENTED_SORT_REPAIR_PRESERVED",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 17. Formal gates
# =============================================================================

semantic_gates = {
    "parent_implementation_authorization_locked":
        (
            parent_obj[
                "implementation_authorized"
            ]
            is True
        ),

    "pre_source_exact_locked":
        (
            portable_pre_sha
            == EXPECTED_PORTABLE_PRE_SHA
        ),

    "pre_V1_placeholder_exact_locked":
        (
            sha256_text(
                target_pre_source
            )
            == EXPECTED_V1_PLACEHOLDER_SHA
        ),

    "V1_matrix_core_unchanged":
        (
            function_source_sha(
                post_text,
                post_functions[
                    "reconstruct_mutual_cont_v1_actions"
                ],
            )
            == EXPECTED_V1_ACTION_CORE_SHA
        ),

    "V1_placeholder_replaced":
        (
            "PortableImplementationIncompleteError"
            not in target_post_source
        ),

    "all_other_top_level_functions_unchanged":
        (
            post_other_function_drift
            == []
        ),

    "dedicated_test_written":
        NEW_TEST.exists(),

    "py_compile_passed":
        (
            compile_exit
            == 0
        ),

    "targeted_pytest_passed":
        (
            pytest_exit
            == 0
        ),

    "historical_V1_oracle_runtime_absent":
        (
            HISTORICAL_ORACLE_FILENAME
            not in target_post_source
        ),

    "V2_invented_sort_repair_preserved":
        (
            exact_forbidden_sort_matches
            == []
        ),

    "scientific_threshold_not_retuned":
        (
            function_source_sha(
                post_text,
                post_functions[
                    "reconstruct_mutual_cont_v1_actions"
                ],
            )
            == EXPECTED_V1_ACTION_CORE_SHA
        ),

    "GT_not_used":
        (
            "gt_graph"
            not in target_post_source
            and
            "ground_truth"
            not in target_post_source
        ),

    "evaluator_not_used":
        (
            "evaluate_graph"
            not in target_post_source
        ),
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3D SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok in (
        semantic_gates.items()
    )

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3D failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 18. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3D",

    "title":
        "MUTUAL_CONT_V1_GRAPH_RUNTIME_IMPLEMENTATION",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_IMPLEMENTATION_ONLY",

    "parent_14_3c_r3_r1": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "portable_source": {
        "path":
            str(
                PORTABLE_REL
            ),

        "pre_file_sha256":
            portable_pre_sha,

        "post_file_sha256":
            portable_post_sha,

        "target_function":
            "_reconstruct_portable_mutual_cont_v1",

        "target_pre_sha256":
            EXPECTED_V1_PLACEHOLDER_SHA,

        "target_post_sha256":
            target_post_sha,

        "matrix_core_sha256":
            EXPECTED_V1_ACTION_CORE_SHA,

        "other_top_level_function_drift":
            post_other_function_drift,
    },

    "implementation_contract": {
        "runtime_mode":
            "SPECIFICATION_DERIVED_RECONSTRUCTION",

        "association_provider_identity_sha256":
            EXPECTED_ASSOCIATION_PROVIDER_SHA,

        "candidate_prob_min":
            0.25,

        "source_final_outdegree":
            0,

        "target_final_indegree":
            0,

        "reciprocal_top1":
            True,

        "matrix_action_core":
            "reconstruct_mutual_cont_v1_actions",

        "distance_authority":
            "_frozen_v9.recovered_edge_dist",

        "graph_mutation":
            "graph.add_edge",

        "validate_keys":
            True,

        "historical_action_oracle_runtime_use":
            False,
    },

    "engineering_helper_locks": {
        "_clone_portable_morph_graph":
            EXPECTED_CLONE_GRAPH_SHA,

        "_portable_v2_resolve_association_provider":
            EXPECTED_V2_RESOLVE_PROVIDER_SHA,

        "_portable_v2_validate_association_output":
            EXPECTED_V2_VALIDATE_OUTPUT_SHA,

        "_portable_r1p50_source_graph_state":
            EXPECTED_R1P50_GRAPH_STATE_SHA,
    },

    "test": {
        "path":
            str(
                NEW_TEST_REL
            ),

        "file_sha256":
            test_file_sha,

        "pytest_exit_code":
            pytest_exit,

        "py_compile_exit_code":
            compile_exit,
    },

    "post_semantic_checks":
        post_semantic_checks,

    "V2_invented_sort_regression": {
        "exact_forbidden_sort_match_count":
            len(
                exact_forbidden_sort_matches
            ),

        "status":
            "PRESERVED",
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "V1_semantic_contract":
            "FROZEN",

        "V1_graph_runtime":
            "IMPLEMENTED",

        "V1_graph_runtime_tests":
            "PASS",

        "V1_structural_fidelity":
            "NOT_YET_VERIFIED",

        "historical_V1_action_oracle_runtime_use":
            False,

        "historical_V1_action_oracle_fidelity_use":
            True,

        "post_V2_graph_provider":
            "NOT_YET_IMPLEMENTED",

        "post_V2_structural_fidelity":
            "NOT_YET_VERIFIED",

        "B12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "scientific_parameter_search":
            False,

        "GT_read":
            False,

        "evaluator_read":
            False,

        "performance_metric_generated":
            False,

        "source_mutation":
            True,

        "source_mutation_scope":
            "ONLY__reconstruct_portable_mutual_cont_v1",
    },

    "explicit_non_claims": [
        "No V1 structural-fidelity claim.",
        "No historical V1 action pickle was opened.",
        "No historical V1 action artifact is used at runtime.",
        "No post-V2 graph provider implementation claim.",
        "No post-V2 structural-fidelity claim.",
        "No B12 structural-fidelity claim.",
        "No performance evidence.",
        "No scientific parameter search.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage": (
        "NOTEBOOK14_14_3E_"
        "POST_V2_GRAPH_PROVIDER_IMPLEMENTATION"
    ),
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3D artifact already exists but differs "
        "from current exact implementation."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 19. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3D PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "SOURCE MUTATION:",
    flush=True,
)

print(
    "  scope:",
    "ONLY _reconstruct_portable_mutual_cont_v1",
    flush=True,
)

print(
    "  portable pre SHA:",
    portable_pre_sha,
    flush=True,
)

print(
    "  portable post SHA:",
    portable_post_sha,
    flush=True,
)

print(
    "  V1 pre SHA:",
    EXPECTED_V1_PLACEHOLDER_SHA,
    flush=True,
)

print(
    "  V1 post SHA:",
    target_post_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "V1 RUNTIME:",
    flush=True,
)

print(
    "  specification-derived reconstruction:",
    True,
    flush=True,
)

print(
    "  explicit association provider:",
    True,
    flush=True,
)

print(
    "  provider identity:",
    EXPECTED_ASSOCIATION_PROVIDER_SHA,
    flush=True,
)

print(
    "  frozen threshold:",
    0.25,
    flush=True,
)

print(
    "  reciprocal top1:",
    True,
    flush=True,
)

print(
    "  source outdegree gate:",
    0,
    flush=True,
)

print(
    "  target indegree gate:",
    0,
    flush=True,
)

print(
    "  recovered_edge_dist:",
    True,
    flush=True,
)

print(
    "  graph.add_edge(validate_keys=True):",
    True,
    flush=True,
)

print(
    flush=True,
)


print(
    "HISTORICAL V1 ORACLE:",
    flush=True,
)

print(
    "  runtime use:",
    False,
    flush=True,
)

print(
    "  structural-fidelity use:",
    "LATER ONLY",
    flush=True,
)

print(
    flush=True,
)


print(
    "TEST:",
    flush=True,
)

print(
    "  file:",
    NEW_TEST_REL,
    flush=True,
)

print(
    "  SHA:",
    test_file_sha,
    flush=True,
)

print(
    "  py_compile:",
    "PASS",
    flush=True,
)

print(
    "  pytest:",
    "PASS",
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  V1 graph runtime = IMPLEMENTED",
    flush=True,
)

print(
    "  V1 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  post-V2 graph provider = NOT YET IMPLEMENTED",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    "  Notebook 14 / 14.3E",
    flush=True,
)

print(
    "  POST_V2_GRAPH_PROVIDER_IMPLEMENTATION",
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-A — POST-V2 ORCHESTRATION SEAM FREEZE
#
# PARENT
# ------
# 14.3D PASS
#
# Current state:
#
#   FrozenV9 runtime    = IMPLEMENTED
#   Morph public runtime= IMPLEMENTED
#   V1 graph runtime    = IMPLEMENTED
#   V2 public runtime   = IMPLEMENTED
#
# Historical post-V2 order:
#
#   FrozenV9 -> Morph -> V1 -> V2
#
# BUT:
#
# PortableCurrentBestContext currently exposes:
#
#   association_provider
#   association_provider_identity_sha256
#
# and DOES NOT visibly expose a MorphCandidateProvider field.
#
# Existing Morph runtime calls:
#
#   provider.build_candidates(...)
#
# Therefore, before implementing:
#
#   post_v2_graph_provider(dataset_id, *, context) -> graph
#
# we must source-lock the Morph provider seam. No guessing is allowed.
#
#
# PURPOSE
# -------
# 1. Lock 14.3D.
# 2. Lock current post-14.3D portable source.
# 3. Extract exact signatures and return behavior for:
#
#       _build_portable_frozen_v9
#       apply_portable_morph_div_v2
#       _reconstruct_portable_mutual_cont_v1
#       apply_portable_mutual_cont_v2
#
# 4. Freeze PortableCurrentBestContext fields.
# 5. Discover every MorphCandidateProvider construction/factory/source.
# 6. Discover every call site of apply_portable_morph_div_v2.
# 7. Determine whether Morph provider can be resolved WITHOUT inventing
#    a new runtime dependency.
# 8. Freeze implementation-readiness decision.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - GT
#   - evaluator
#   - historical action runtime lookup
#   - performance
#   - structural fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3d_"
    "mutual_cont_v1_graph_runtime_implementation_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


PRIMARY_TEST_REL = Path(
    "tests/test_stage12_portable_current_best.py"
)

PRIMARY_TEST = ROOT / PRIMARY_TEST_REL


V1_TEST_REL = Path(
    "tests/test_stage12_v1_graph_runtime.py"
)

V1_TEST = ROOT / V1_TEST_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_a_"
    "post_v2_orchestration_seam_freeze_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities from 14.3D
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "ece00f7edebe7937702645d019f3359d48ed1e67c39cdea0bf354457facfc7c8"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "5d1fefe6b803f00266dbb6c5326e360bc45b2cfbc8dab01be91d1f2cf17c2790"
)


EXPECTED_PORTABLE_SHA = (
    "5e1a2d64012b68e4e9b5cfb6909a4de849844ad5f5e10421a19c46b858ec9708"
)


EXPECTED_FROZEN_V9_SHA = (
    "d4f91cd4d7c1cd97538f6dc01f0087f9c25af07bf3c7dd4d858891d1b3031505"
)


EXPECTED_MORPH_PUBLIC_SHA = (
    "db173c58a83eddd282712915514c31739815195ccb99edaa84a3ff5e16f1ffcf"
)


EXPECTED_V1_RUNTIME_SHA = (
    "a159aeb6e4ebceede6194bf0663aa240e668b16f2b42cd79527030bee196ad3e"
)


EXPECTED_V2_PUBLIC_SHA = (
    "1901c1977a3f0f0cb6960aee426b8b726259b54e504646eeb9353febaca0d1dc"
)


EXPECTED_CONTEXT_SHA = (
    "40420de98e596e8f6b5eb67b7a2a5df9ab5a03b44fa1980e503fa9e4bd51c808"
)


EXPECTED_CONTEXT_FIELDS = [
    "b256_pred_995_root",
    "b256_pred_9975_root",
    "s9995_pred_995_root",
    "s9995_pred_9975_root",
    "dataset_root",
    "physical_scale_zyx_um",
    "frozen_v9_policy_bundle",
    "association_provider",
    "association_provider_identity_sha256",
    "strict",
]


EXPECTED_STAGE_ORDER = [
    "FROZEN_V9",
    "MORPH_DIV_V2",
    "MUTUAL_CONT_V1",
    "MUTUAL_CONT_V2",
]


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(
    data: bytes,
) -> str:

    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(
    text: str,
) -> str:

    return sha256_bytes(
        text.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )


    assert (
        source
        is not None
    )


    return source


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    def visit(
        value: ast.AST,
    ):

        if isinstance(
            value,
            ast.Name,
        ):

            result.add(
                value.id
            )


        elif isinstance(
            value,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for child in (
                value.elts
            ):

                visit(
                    child
                )


        elif isinstance(
            value,
            ast.Starred,
        ):

            visit(
                value.value
            )


    visit(
        node
    )


    return result


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def top_level_classes(
    tree: ast.Module,
) -> dict[
    str,
    ast.ClassDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            ast.ClassDef,
        )
    }


def source_sha(
    text: str,
    node: ast.AST,
) -> str:

    return sha256_text(
        node_source(
            text,
            node,
        )
    )


def function_calls(
    node: ast.AST,
) -> list[str]:

    return sorted({
        dotted_name(
            call.func
        )
        or "<dynamic-call>"

        for call in ast.walk(
            node
        )

        if isinstance(
            call,
            ast.Call,
        )
    })


def parameter_record(
    arg: ast.arg,
    default: ast.AST | None,
) -> dict[str, Any]:

    return {
        "name":
            arg.arg,

        "annotation":
            (
                ast.unparse(
                    arg.annotation
                )
                if arg.annotation
                is not None
                else None
            ),

        "default":
            (
                ast.unparse(
                    default
                )
                if default
                is not None
                else None
            ),
    }


def function_signature_record(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> dict[str, Any]:

    positional = (
        list(
            fn.args.posonlyargs
        )
        +
        list(
            fn.args.args
        )
    )


    positional_defaults = (
        [None]
        * (
            len(
                positional
            )
            -
            len(
                fn.args.defaults
            )
        )
        +
        list(
            fn.args.defaults
        )
    )


    kwonly = list(
        fn.args.kwonlyargs
    )


    kw_defaults = list(
        fn.args.kw_defaults
    )


    return {
        "positional": [
            parameter_record(
                arg,
                default,
            )

            for arg, default in zip(
                positional,
                positional_defaults,
            )
        ],

        "vararg":
            (
                fn.args.vararg.arg
                if fn.args.vararg
                is not None
                else None
            ),

        "kwonly": [
            parameter_record(
                arg,
                default,
            )

            for arg, default in zip(
                kwonly,
                kw_defaults,
            )
        ],

        "kwarg":
            (
                fn.args.kwarg.arg
                if fn.args.kwarg
                is not None
                else None
            ),

        "return_annotation":
            (
                ast.unparse(
                    fn.returns
                )
                if fn.returns
                is not None
                else None
            ),
    }


def return_records(
    text: str,
    fn: ast.AST,
) -> list[
    dict[str, Any]
]:

    result = []


    for node in ast.walk(
        fn
    ):

        if not isinstance(
            node,
            ast.Return,
        ):

            continue


        result.append({
            "line":
                node.lineno,

            "expression":
                (
                    ast.unparse(
                        node.value
                    )
                    if node.value
                    is not None
                    else "None"
                ),

            "source":
                node_source(
                    text,
                    node,
                ),
        })


    return sorted(
        result,
        key=lambda item: (
            item[
                "line"
            ]
        ),
    )


def function_contract(
    text: str,
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> dict[str, Any]:

    source = node_source(
        text,
        fn,
    )


    return {
        "name":
            fn.name,

        "line_start":
            fn.lineno,

        "line_end":
            fn.end_lineno,

        "source_sha256":
            sha256_text(
                source
            ),

        "signature":
            function_signature_record(
                fn
            ),

        "returns":
            return_records(
                text,
                fn,
            ),

        "calls":
            function_calls(
                fn
            ),

        "source":
            source,
    }


def class_fields(
    text: str,
    cls: ast.ClassDef,
) -> list[
    dict[str, Any]
]:

    fields = []


    for stmt in (
        cls.body
    ):

        if not isinstance(
            stmt,
            ast.AnnAssign,
        ):

            continue


        if not isinstance(
            stmt.target,
            ast.Name,
        ):

            continue


        fields.append({
            "name":
                stmt.target.id,

            "annotation":
                ast.unparse(
                    stmt.annotation
                ),

            "default":
                (
                    ast.unparse(
                        stmt.value
                    )
                    if stmt.value
                    is not None
                    else None
                ),

            "line":
                stmt.lineno,
        })


    return fields


def owner_map(
    tree: ast.Module,
) -> dict[
    ast.AST,
    str,
]:

    mapping = {}


    for top in (
        tree.body
    ):

        if not isinstance(
            top,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):

            continue


        for child in ast.walk(
            top
        ):

            if (
                child
                not in mapping
            ):

                mapping[
                    child
                ] = (
                    top.name
                )


    return mapping


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    ), path


    file_sha = sha256_file(
        path
    )


    assert (
        file_sha
        == EXPECTED_PARENT_FILE_SHA
    ), (
        "14.3D file SHA mismatch.\n"
        f"expected={EXPECTED_PARENT_FILE_SHA}\n"
        f"actual={file_sha}"
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )


    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "V1_graph_runtime"
        ]
        == "IMPLEMENTED"
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "post_V2_graph_provider"
        ]
        == "NOT_YET_IMPLEMENTED"
    )


    return {
        "object":
            obj,

        "file_sha256":
            file_sha,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    }


# =============================================================================
# 4. Call-site scanner
# =============================================================================

def scan_call_sites(
    path: Path,
    *,
    target_names_: set[str],
) -> list[
    dict[str, Any]
]:

    text = path.read_text(
        encoding="utf-8"
    )


    tree = ast.parse(
        text,
        filename=str(
            path
        ),
    )


    owners = owner_map(
        tree
    )


    records = []


    for call in ast.walk(
        tree
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if (
            name
            is None
        ):

            continue


        base = (
            name.split(
                "."
            )[
                -1
            ]
        )


        if (
            base
            not in target_names_
        ):

            continue


        records.append({
            "path":
                str(
                    path.relative_to(
                        ROOT
                    )
                ),

            "line":
                call.lineno,

            "owner":
                owners.get(
                    call,
                    "<module>",
                ),

            "call_name":
                name,

            "args": [
                ast.unparse(
                    arg
                )

                for arg in (
                    call.args
                )
            ],

            "keywords": {
                (
                    kw.arg
                    if kw.arg
                    is not None
                    else "**"
                ):
                    ast.unparse(
                        kw.value
                    )

                for kw in (
                    call.keywords
                )
            },

            "source":
                node_source(
                    text,
                    call,
                ),
        })


    return sorted(
        records,
        key=lambda item: (
            item[
                "path"
            ],
            item[
                "line"
            ],
        ),
    )


# =============================================================================
# 5. Morph-provider construction/factory discovery
# =============================================================================

MORPH_PROVIDER_SIGNALS = {
    "MorphCandidateProvider",
    "validate_morph_candidate_provider",
    "build_candidates",
    "provider_authority_status",
    "provider_source_sha256",
    "MORPH_DIV_V2_CONFIG_ID",
}


def morph_provider_source_inventory(
    text: str,
    tree: ast.Module,
) -> dict[str, Any]:

    functions = top_level_functions(
        tree
    )


    classes = top_level_classes(
        tree
    )


    provider_function_candidates = []


    for name, fn in (
        functions.items()
    ):

        source = node_source(
            text,
            fn,
        )


        hits = sorted(
            signal

            for signal in (
                MORPH_PROVIDER_SIGNALS
            )

            if signal in source
        )


        if (
            hits
        ):

            provider_function_candidates.append({
                "name":
                    name,

                "line_start":
                    fn.lineno,

                "line_end":
                    fn.end_lineno,

                "source_sha256":
                    sha256_text(
                        source
                    ),

                "signal_hits":
                    hits,

                "calls":
                    function_calls(
                        fn
                    ),
            })


    provider_class_candidates = []


    for name, cls in (
        classes.items()
    ):

        source = node_source(
            text,
            cls,
        )


        hits = sorted(
            signal

            for signal in (
                MORPH_PROVIDER_SIGNALS
            )

            if signal in source
        )


        if (
            hits
        ):

            provider_class_candidates.append({
                "name":
                    name,

                "line_start":
                    cls.lineno,

                "line_end":
                    cls.end_lineno,

                "source_sha256":
                    sha256_text(
                        source
                    ),

                "signal_hits":
                    hits,

                "fields":
                    class_fields(
                        text,
                        cls,
                    ),
            })


    constructor_calls = []


    owners = owner_map(
        tree
    )


    for call in ast.walk(
        tree
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if (
            name
            is None
            or not name.endswith(
                "MorphCandidateProvider"
            )
        ):

            continue


        constructor_calls.append({
            "line":
                call.lineno,

            "owner":
                owners.get(
                    call,
                    "<module>",
                ),

            "call_name":
                name,

            "args": [
                ast.unparse(
                    arg
                )

                for arg in (
                    call.args
                )
            ],

            "keywords": {
                (
                    kw.arg
                    if kw.arg
                    is not None
                    else "**"
                ):
                    ast.unparse(
                        kw.value
                    )

                for kw in (
                    call.keywords
                )
            },

            "source":
                node_source(
                    text,
                    call,
                ),
        })


    module_assignments = []


    for stmt in (
        tree.body
    ):

        names = set()
        value = None


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in (
                stmt.targets
            ):

                names |= (
                    target_names(
                        target
                    )
                )


            value = (
                stmt.value
            )


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            names |= (
                target_names(
                    stmt.target
                )
            )


            value = (
                stmt.value
            )


        if (
            value
            is None
        ):

            continue


        value_source = (
            ast.unparse(
                value
            )
        )


        if any(
            signal
            in value_source

            for signal in (
                MORPH_PROVIDER_SIGNALS
            )
        ):

            module_assignments.append({
                "names":
                    sorted(
                        names
                    ),

                "line":
                    stmt.lineno,

                "value":
                    value_source,

                "source":
                    node_source(
                        text,
                        stmt,
                    ),
            })


    return {
        "functions":
            provider_function_candidates,

        "classes":
            provider_class_candidates,

        "constructor_calls":
            constructor_calls,

        "module_assignments":
            module_assignments,
    }


# =============================================================================
# 6. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    PRIMARY_TEST,
    V1_TEST,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-A "
    "POST-V2 ORCHESTRATION SEAM FREEZE",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "PORTABLE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Parent lock
# =============================================================================

parent_lock = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3D_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    parent_lock[
        "file_sha256"
    ],
    flush=True,
)

print(
    "  content SHA:",
    parent_lock[
        "content_sha256"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Current portable source lock
# =============================================================================

portable_sha = sha256_file(
    PORTABLE
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
), (
    "Portable source drift after 14.3D.\n"
    f"expected={EXPECTED_PORTABLE_SHA}\n"
    f"actual  ={portable_sha}"
)


portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_text,
    filename=str(
        PORTABLE
    ),
)


portable_functions = top_level_functions(
    portable_tree
)


portable_classes = top_level_classes(
    portable_tree
)


print(
    "[PASS] POST_14_3D_PORTABLE_SOURCE_LOCK",
    flush=True,
)

print(
    "  SHA:",
    portable_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Required stage function locks
# =============================================================================

STAGE_FUNCTIONS = {
    "FROZEN_V9":
        "_build_portable_frozen_v9",

    "MORPH_DIV_V2":
        "apply_portable_morph_div_v2",

    "MUTUAL_CONT_V1":
        "_reconstruct_portable_mutual_cont_v1",

    "MUTUAL_CONT_V2":
        "apply_portable_mutual_cont_v2",
}


EXPECTED_STAGE_SHAS = {
    "FROZEN_V9":
        EXPECTED_FROZEN_V9_SHA,

    "MORPH_DIV_V2":
        EXPECTED_MORPH_PUBLIC_SHA,

    "MUTUAL_CONT_V1":
        EXPECTED_V1_RUNTIME_SHA,

    "MUTUAL_CONT_V2":
        EXPECTED_V2_PUBLIC_SHA,
}


contracts = {}


for stage in (
    EXPECTED_STAGE_ORDER
):

    name = (
        STAGE_FUNCTIONS[
            stage
        ]
    )


    assert (
        name
        in portable_functions
    ), (
        f"Missing stage function: {name}"
    )


    fn = (
        portable_functions[
            name
        ]
    )


    contract = function_contract(
        portable_text,
        fn,
    )


    assert (
        contract[
            "source_sha256"
        ]
        == EXPECTED_STAGE_SHAS[
            stage
        ]
    ), (
        f"{stage} source drift.\n"
        f"expected={EXPECTED_STAGE_SHAS[stage]}\n"
        f"actual={contract['source_sha256']}"
    )


    contracts[
        stage
    ] = contract


print(
    "[PASS] POST_V2_STAGE_SOURCE_LOCKS",
    flush=True,
)


for stage in (
    EXPECTED_STAGE_ORDER
):

    record = (
        contracts[
            stage
        ]
    )


    print(
        " ",
        stage,
        "->",
        record[
            "name"
        ],
        flush=True,
    )

    print(
        "    SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "    signature:",
        json.dumps(
            record[
                "signature"
            ],
            sort_keys=True,
        ),
        flush=True,
    )

    print(
        "    returns:",
        [
            item[
                "expression"
            ]
            for item in (
                record[
                    "returns"
                ]
            )
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 10. Context exact lock
# =============================================================================

assert (
    "PortableCurrentBestContext"
    in portable_classes
)


context_cls = (
    portable_classes[
        "PortableCurrentBestContext"
    ]
)


context_source = node_source(
    portable_text,
    context_cls,
)


context_sha = sha256_text(
    context_source
)


assert (
    context_sha
    == EXPECTED_CONTEXT_SHA
), (
    "PortableCurrentBestContext drift."
)


context_field_records = class_fields(
    portable_text,
    context_cls,
)


context_field_names = [
    record[
        "name"
    ]

    for record in (
        context_field_records
    )
]


assert (
    context_field_names
    == EXPECTED_CONTEXT_FIELDS
), (
    "PortableCurrentBestContext schema drift.\n"
    f"expected={EXPECTED_CONTEXT_FIELDS}\n"
    f"actual={context_field_names}"
)


morph_named_context_fields = [
    name

    for name in (
        context_field_names
    )

    if (
        "morph"
        in name.lower()
    )
]


print(
    "[PASS] CONTEXT_SCHEMA_LOCK",
    flush=True,
)

print(
    "  context SHA:",
    context_sha,
    flush=True,
)

print(
    "  fields:",
    context_field_names,
    flush=True,
)

print(
    "  Morph-named fields:",
    morph_named_context_fields,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Interpret public Morph signature
# =============================================================================

morph_signature = (
    contracts[
        "MORPH_DIV_V2"
    ][
        "signature"
    ]
)


morph_positional_names = [
    item[
        "name"
    ]

    for item in (
        morph_signature[
            "positional"
        ]
    )
]


morph_kwonly_names = [
    item[
        "name"
    ]

    for item in (
        morph_signature[
            "kwonly"
        ]
    )
]


morph_all_parameters = (
    morph_positional_names
    +
    morph_kwonly_names
)


morph_provider_parameters = [
    name

    for name in (
        morph_all_parameters
    )

    if (
        "provider"
        in name.lower()
    )
]


morph_context_parameter_present = (
    "context"
    in morph_all_parameters
)


morph_return_audit_present = (
    "return_audit"
    in morph_all_parameters
)


print(
    "=" * 112,
    flush=True,
)

print(
    "MORPH PUBLIC RUNTIME CONTRACT",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "parameters:",
    morph_all_parameters,
    flush=True,
)

print(
    "provider parameters:",
    morph_provider_parameters,
    flush=True,
)

print(
    "context parameter:",
    morph_context_parameter_present,
    flush=True,
)

print(
    "return_audit:",
    morph_return_audit_present,
    flush=True,
)

print(
    "calls provider.build_candidates:",
    (
        "provider.build_candidates"
        in contracts[
            "MORPH_DIV_V2"
        ][
            "calls"
        ]
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Interpret public V2 contract
# =============================================================================

v2_signature = (
    contracts[
        "MUTUAL_CONT_V2"
    ][
        "signature"
    ]
)


v2_positional_names = [
    item[
        "name"
    ]

    for item in (
        v2_signature[
            "positional"
        ]
    )
]


v2_kwonly_names = [
    item[
        "name"
    ]

    for item in (
        v2_signature[
            "kwonly"
        ]
    )
]


v2_all_parameters = (
    v2_positional_names
    +
    v2_kwonly_names
)


v2_expected_prefix_ok = (
    v2_positional_names[
        :2
    ]
    == [
        "dataset_id",
        "post_v1_graph",
    ]
)


v2_context_ok = (
    "context"
    in v2_kwonly_names
)


v2_return_audit_ok = (
    "return_audit"
    in v2_kwonly_names
)


assert (
    v2_expected_prefix_ok
), (
    "V2 public signature prefix is no longer "
    "(dataset_id, post_v1_graph)."
)


assert (
    v2_context_ok
)


assert (
    v2_return_audit_ok
)


print(
    "[PASS] V2_PUBLIC_ORCHESTRATION_CONTRACT",
    flush=True,
)

print(
    "  positional:",
    v2_positional_names,
    flush=True,
)

print(
    "  kwonly:",
    v2_kwonly_names,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. FrozenV9 return-shape contract
#
# _build_portable_frozen_v9 directly returns canonical apply_frozen_v9.
# Historical authority established canonical return:
#
#     (graph, scale, stats)
#
# We record that provider must select element 0 and must not silently treat
# the full tuple as a graph.
# =============================================================================

frozen_returns = (
    contracts[
        "FROZEN_V9"
    ][
        "returns"
    ]
)


assert (
    len(
        frozen_returns
    )
    == 1
)


assert (
    "_frozen_v9.apply_frozen_v9"
    in frozen_returns[
        0
    ][
        "expression"
    ]
)


frozen_v9_orchestration_return_contract = {
    "runtime_return":
        "CANONICAL_APPLY_FROZEN_V9_RETURN",

    "canonical_shape":
        "(graph, scale, stats)",

    "post_v2_provider_graph_element":
        0,
}


print(
    "[PASS] FROZEN_V9_ORCHESTRATION_RETURN_CONTRACT",
    flush=True,
)

print(
    json.dumps(
        frozen_v9_orchestration_return_contract,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Morph provider source inventory in portable module
# =============================================================================

morph_inventory = morph_provider_source_inventory(
    portable_text,
    portable_tree,
)


print(
    "=" * 112,
    flush=True,
)

print(
    "MORPH PROVIDER SOURCE INVENTORY :: PORTABLE MODULE",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "candidate functions:",
    len(
        morph_inventory[
            "functions"
        ]
    ),
    flush=True,
)


for record in (
    morph_inventory[
        "functions"
    ]
):

    print(
        " ",
        record[
            "name"
        ],
        f"L{record['line_start']}"
        f"-L{record['line_end']}",
        flush=True,
    )

    print(
        "    SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "    signals:",
        record[
            "signal_hits"
        ],
        flush=True,
    )


print(
    "candidate classes:",
    len(
        morph_inventory[
            "classes"
        ]
    ),
    flush=True,
)


for record in (
    morph_inventory[
        "classes"
    ]
):

    print(
        " ",
        record[
            "name"
        ],
        flush=True,
    )

    print(
        "    SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "    signals:",
        record[
            "signal_hits"
        ],
        flush=True,
    )


print(
    "MorphCandidateProvider constructor calls:",
    len(
        morph_inventory[
            "constructor_calls"
        ]
    ),
    flush=True,
)


for record in (
    morph_inventory[
        "constructor_calls"
    ]
):

    print(
        " ",
        f"L{record['line']}",
        f"owner={record['owner']}",
        flush=True,
    )

    print(
        "   ",
        record[
            "source"
        ],
        flush=True,
    )


print(
    "module assignments:",
    len(
        morph_inventory[
            "module_assignments"
        ]
    ),
    flush=True,
)


for record in (
    morph_inventory[
        "module_assignments"
    ]
):

    print(
        " ",
        record[
            "names"
        ],
        "=",
        record[
            "value"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 15. Call-site inventory
# =============================================================================

TARGET_CALL_NAMES = {
    "apply_portable_morph_div_v2",
    "apply_portable_mutual_cont_v2",
    "_reconstruct_portable_mutual_cont_v1",
    "_build_portable_frozen_v9",
}


call_sites = []


for path in (
    PORTABLE,
    PRIMARY_TEST,
    V1_TEST,
):

    call_sites.extend(
        scan_call_sites(
            path,
            target_names_=TARGET_CALL_NAMES,
        )
    )


print(
    "=" * 112,
    flush=True,
)

print(
    "ORCHESTRATION CALL-SITE INVENTORY",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for target in sorted(
    TARGET_CALL_NAMES
):

    target_records = [
        record

        for record in (
            call_sites
        )

        if (
            record[
                "call_name"
            ].split(
                "."
            )[
                -1
            ]
            == target
        )
    ]


    print(
        target,
        "calls:",
        len(
            target_records
        ),
        flush=True,
    )


    for record in (
        target_records
    ):

        print(
            " ",
            record[
                "path"
            ],
            f"L{record['line']}",
            f"owner={record['owner']}",
            flush=True,
        )

        print(
            "    args:",
            record[
                "args"
            ],
            flush=True,
        )

        print(
            "    keywords:",
            record[
                "keywords"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 16. Inspect test-side Morph providers
#
# Test providers do not become production authority.
# This section is only used to discover the actual callable contract.
# =============================================================================

primary_test_text = PRIMARY_TEST.read_text(
    encoding="utf-8"
)


primary_test_tree = ast.parse(
    primary_test_text,
    filename=str(
        PRIMARY_TEST
    ),
)


test_morph_inventory = morph_provider_source_inventory(
    primary_test_text,
    primary_test_tree,
)


print(
    "=" * 112,
    flush=True,
)

print(
    "TEST-SIDE MORPH PROVIDER CONTRACT EXAMPLES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "constructor calls:",
    len(
        test_morph_inventory[
            "constructor_calls"
        ]
    ),
    flush=True,
)


for record in (
    test_morph_inventory[
        "constructor_calls"
    ][
        :20
    ]
):

    print(
        " ",
        f"L{record['line']}",
        f"owner={record['owner']}",
        flush=True,
    )

    print(
        "   ",
        record[
            "source"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 17. Search for module-level production Morph provider resolution
# =============================================================================

production_morph_provider_instance_candidates = []


for record in (
    morph_inventory[
        "constructor_calls"
    ]
):

    # Module-level construction is the strongest existing fixed-runtime seam.
    if (
        record[
            "owner"
        ]
        == "<module>"
    ):

        production_morph_provider_instance_candidates.append(
            record
        )


production_morph_factory_candidates = []


for record in (
    morph_inventory[
        "functions"
    ]
):

    name_lower = (
        record[
            "name"
        ].lower()
    )


    calls = set(
        record[
            "calls"
        ]
    )


    looks_factory_like = (
        (
            "morph"
            in name_lower
        )
        and (
            "provider"
            in name_lower
            or
            "candidate"
            in name_lower
        )
    )


    constructs_provider = any(
        call.endswith(
            "MorphCandidateProvider"
        )
        for call in calls
    )


    if (
        looks_factory_like
        and constructs_provider
    ):

        production_morph_factory_candidates.append(
            record
        )


print(
    "=" * 112,
    flush=True,
)

print(
    "PRODUCTION MORPH PROVIDER RESOLUTION CANDIDATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "module-level instances:",
    len(
        production_morph_provider_instance_candidates
    ),
    flush=True,
)

print(
    "factory candidates:",
    len(
        production_morph_factory_candidates
    ),
    flush=True,
)


for record in (
    production_morph_factory_candidates
):

    print(
        " factory:",
        record[
            "name"
        ],
        record[
            "source_sha256"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 18. Determine what Morph runtime actually requires
# =============================================================================

morph_requires_explicit_provider_parameter = (
    len(
        morph_provider_parameters
    )
    >= 1
)


morph_provider_available_from_context = (
    len(
        morph_named_context_fields
    )
    >= 1
)


morph_provider_available_from_fixed_module_instance = (
    len(
        production_morph_provider_instance_candidates
    )
    == 1
)


morph_provider_available_from_source_factory = (
    len(
        production_morph_factory_candidates
    )
    == 1
)


if not (
    morph_requires_explicit_provider_parameter
):

    morph_provider_resolution_mode = (
        "MORPH_RUNTIME_RESOLVES_PROVIDER_INTERNALLY"
    )


elif (
    morph_provider_available_from_context
):

    morph_provider_resolution_mode = (
        "CONTEXT_FIELD"
    )


elif (
    morph_provider_available_from_fixed_module_instance
):

    morph_provider_resolution_mode = (
        "FIXED_MODULE_INSTANCE"
    )


elif (
    morph_provider_available_from_source_factory
):

    morph_provider_resolution_mode = (
        "SOURCE_FACTORY"
    )


else:

    morph_provider_resolution_mode = (
        "UNRESOLVED"
    )


print(
    "=" * 112,
    flush=True,
)

print(
    "MORPH PROVIDER SEAM RESOLUTION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "Morph requires explicit provider parameter:",
    morph_requires_explicit_provider_parameter,
    flush=True,
)

print(
    "provider available from context:",
    morph_provider_available_from_context,
    flush=True,
)

print(
    "provider available from fixed module instance:",
    morph_provider_available_from_fixed_module_instance,
    flush=True,
)

print(
    "provider available from source factory:",
    morph_provider_available_from_source_factory,
    flush=True,
)

print(
    "resolution mode:",
    morph_provider_resolution_mode,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Default return behavior
#
# Both Morph/V2 public runtimes expose graph on normal non-audit path.
# We prove source-visible graph return rather than guessing.
# =============================================================================

morph_return_expressions = [
    record[
        "expression"
    ]

    for record in (
        contracts[
            "MORPH_DIV_V2"
        ][
            "returns"
        ]
    )
]


v2_return_expressions = [
    record[
        "expression"
    ]

    for record in (
        contracts[
            "MUTUAL_CONT_V2"
        ][
            "returns"
        ]
    )
]


morph_has_plain_graph_return = (
    "graph"
    in morph_return_expressions
)


v2_has_plain_graph_return = (
    "graph"
    in v2_return_expressions
)


assert (
    morph_has_plain_graph_return
), (
    "Morph public runtime lacks a source-visible plain graph return."
)


assert (
    v2_has_plain_graph_return
), (
    "V2 public runtime lacks a source-visible plain graph return."
)


print(
    "[PASS] PUBLIC_RUNTIME_GRAPH_RETURN_PATHS",
    flush=True,
)

print(
    "  Morph plain graph return:",
    morph_has_plain_graph_return,
    flush=True,
)

print(
    "  V2 plain graph return:",
    v2_has_plain_graph_return,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 20. R1P50 exclusion proof
# =============================================================================

r1p50_stage_names = [
    name

    for name in (
        portable_functions
    )

    if (
        "r1p50"
        in name.lower()
    )
]


post_v2_required_functions = {
    contracts[
        stage
    ][
        "name"
    ]

    for stage in (
        EXPECTED_STAGE_ORDER
    )
}


post_v2_contains_r1p50 = any(
    "r1p50"
    in name.lower()

    for name in (
        post_v2_required_functions
    )
)


assert (
    post_v2_contains_r1p50
    is False
)


print(
    "[PASS] POST_V2_BOUNDARY_EXCLUDES_R1P50",
    flush=True,
)

print(
    "  stages:",
    EXPECTED_STAGE_ORDER,
    flush=True,
)

print(
    "  R1P50 included:",
    False,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 21. Implementation readiness
# =============================================================================

frozen_v9_ready = True

morph_runtime_ready = (
    contracts[
        "MORPH_DIV_V2"
    ][
        "source_sha256"
    ]
    == EXPECTED_MORPH_PUBLIC_SHA
)


v1_runtime_ready = (
    contracts[
        "MUTUAL_CONT_V1"
    ][
        "source_sha256"
    ]
    == EXPECTED_V1_RUNTIME_SHA
)


v2_runtime_ready = (
    contracts[
        "MUTUAL_CONT_V2"
    ][
        "source_sha256"
    ]
    == EXPECTED_V2_PUBLIC_SHA
)


morph_provider_seam_ready = (
    morph_provider_resolution_mode
    != "UNRESOLVED"
)


post_v2_provider_implementation_ready = (
    frozen_v9_ready
    and
    morph_runtime_ready
    and
    morph_provider_seam_ready
    and
    v1_runtime_ready
    and
    v2_runtime_ready
)


readiness = {
    "FrozenV9_runtime":
        frozen_v9_ready,

    "Morph_runtime":
        morph_runtime_ready,

    "Morph_provider_resolution_mode":
        morph_provider_resolution_mode,

    "Morph_provider_seam_ready":
        morph_provider_seam_ready,

    "V1_graph_runtime":
        v1_runtime_ready,

    "V2_runtime":
        v2_runtime_ready,

    "R1P50_excluded":
        True,

    "post_v2_provider_implementation_ready":
        post_v2_provider_implementation_ready,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "POST-V2 PROVIDER IMPLEMENTATION READINESS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        readiness,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 22. Semantic gates
#
# IMPORTANT:
#
# This discovery/freeze stage PASSES even if Morph provider seam is unresolved.
# In that case implementation remains blocked, rather than inventing a seam.
# =============================================================================

semantic_gates = {
    "parent_14_3d_locked":
        True,

    "portable_source_exact_locked":
        (
            portable_sha
            == EXPECTED_PORTABLE_SHA
        ),

    "FrozenV9_source_locked":
        frozen_v9_ready,

    "Morph_public_runtime_locked":
        morph_runtime_ready,

    "V1_graph_runtime_locked":
        v1_runtime_ready,

    "V2_public_runtime_locked":
        v2_runtime_ready,

    "context_schema_locked":
        (
            context_field_names
            == EXPECTED_CONTEXT_FIELDS
        ),

    "Morph_provider_inventory_complete":
        True,

    "orchestration_call_site_inventory_complete":
        True,

    "Morph_graph_return_path_proven":
        morph_has_plain_graph_return,

    "V2_graph_return_path_proven":
        v2_has_plain_graph_return,

    "R1P50_explicitly_excluded":
        (
            post_v2_contains_r1p50
            is False
        ),

    "no_source_mutation":
        True,

    "historical_action_runtime_lookup_not_used":
        True,

    "GT_not_used":
        True,

    "evaluator_not_used":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3E-A SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    name

    for name, ok in (
        semantic_gates.items()
    )

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-A failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 23. Next-stage decision
# =============================================================================

if (
    post_v2_provider_implementation_ready
):

    next_stage = (
        "NOTEBOOK14_14_3E_B_"
        "POST_V2_GRAPH_PROVIDER_IMPLEMENTATION"
    )


else:

    next_stage = (
        "NOTEBOOK14_14_3E_B_"
        "MORPH_PROVIDER_SEAM_CLOSURE"
    )


# =============================================================================
# 24. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_A",

    "title":
        "POST_V2_ORCHESTRATION_SEAM_FREEZE",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_ORCHESTRATION_CONTRACT_ONLY",

    "parent_14_3d": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "portable_source": {
        "path":
            str(
                PORTABLE_REL
            ),

        "file_sha256":
            portable_sha,
    },

    "stage_order":
        EXPECTED_STAGE_ORDER,

    "stage_contracts":
        contracts,

    "context_contract": {
        "class":
            "PortableCurrentBestContext",

        "source_sha256":
            context_sha,

        "fields":
            context_field_records,

        "morph_named_fields":
            morph_named_context_fields,
    },

    "FrozenV9_return_contract":
        frozen_v9_orchestration_return_contract,

    "Morph_provider_inventory":
        morph_inventory,

    "test_Morph_provider_examples":
        test_morph_inventory,

    "orchestration_call_sites":
        call_sites,

    "Morph_provider_resolution": {
        "runtime_parameters":
            morph_all_parameters,

        "explicit_provider_parameters":
            morph_provider_parameters,

        "context_parameter_present":
            morph_context_parameter_present,

        "return_audit_present":
            morph_return_audit_present,

        "context_provider_available":
            morph_provider_available_from_context,

        "fixed_module_instance_available":
            morph_provider_available_from_fixed_module_instance,

        "source_factory_available":
            morph_provider_available_from_source_factory,

        "resolution_mode":
            morph_provider_resolution_mode,
    },

    "V2_orchestration_contract": {
        "positional_parameters":
            v2_positional_names,

        "keyword_only_parameters":
            v2_kwonly_names,

        "normal_return":
            "graph",
    },

    "R1P50": {
        "module_functions":
            r1p50_stage_names,

        "included_in_post_v2_boundary":
            False,
    },

    "readiness":
        readiness,

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "post_V2_stage_order":
            "FROZEN",

        "post_V2_stage_source_contracts":
            "FROZEN",

        "Morph_provider_seam":
            morph_provider_resolution_mode,

        "post_V2_provider_implementation":
            (
                "AUTHORIZED"
                if post_v2_provider_implementation_ready
                else "BLOCKED_ON_MORPH_PROVIDER_SEAM"
            ),

        "post_V2_provider_implemented":
            False,

        "post_V2_structural_fidelity":
            "NOT_YET_VERIFIED",

        "V1_structural_fidelity":
            "NOT_YET_VERIFIED",

        "B12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "R1P50_included":
            False,

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_read":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No post-V2 provider was implemented in this stage.",
        "No Morph provider seam was invented.",
        "No R1P50 stage is part of the post-V2 boundary.",
        "No V1 structural-fidelity claim.",
        "No post-V2 structural-fidelity claim.",
        "No B12 structural-fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-A artifact exists but differs "
        "from current exact orchestration freeze."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 25. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-A PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "POST-V2 ORDER:",
    flush=True,
)

print(
    "  FrozenV9 -> Morph -> V1 -> V2",
    flush=True,
)

print(
    "  R1P50 included:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "CURRENT STAGE CONTRACTS:",
    flush=True,
)


for stage in (
    EXPECTED_STAGE_ORDER
):

    print(
        " ",
        stage,
        "->",
        contracts[
            stage
        ][
            "name"
        ],
        flush=True,
    )

    print(
        "    signature:",
        json.dumps(
            contracts[
                stage
            ][
                "signature"
            ],
            sort_keys=True,
        ),
        flush=True,
    )


print(
    flush=True,
)


print(
    "MORPH PROVIDER SEAM:",
    flush=True,
)

print(
    "  explicit provider params:",
    morph_provider_parameters,
    flush=True,
)

print(
    "  context Morph fields:",
    morph_named_context_fields,
    flush=True,
)

print(
    "  module-level provider instances:",
    len(
        production_morph_provider_instance_candidates
    ),
    flush=True,
)

print(
    "  source factories:",
    [
        record[
            "name"
        ]

        for record in (
            production_morph_factory_candidates
        )
    ],
    flush=True,
)

print(
    "  resolution mode:",
    morph_provider_resolution_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "IMPLEMENTATION READINESS:",
    flush=True,
)

print(
    "  FrozenV9:",
    frozen_v9_ready,
    flush=True,
)

print(
    "  Morph runtime:",
    morph_runtime_ready,
    flush=True,
)

print(
    "  Morph provider seam:",
    morph_provider_seam_ready,
    flush=True,
)

print(
    "  V1 runtime:",
    v1_runtime_ready,
    flush=True,
)

print(
    "  V2 runtime:",
    v2_runtime_ready,
    flush=True,
)

print(
    "  POST-V2 PROVIDER READY:",
    post_v2_provider_implementation_ready,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  post-V2 orchestration contracts = FROZEN",
    flush=True,
)

print(
    "  Morph provider seam =",
    morph_provider_resolution_mode,
    flush=True,
)

print(
    "  post-V2 provider = NOT YET IMPLEMENTED",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-B-R1 — MORPH PROVIDER SEAM CLOSURE
#               HISTORICAL AUTHORITY CONTENT-LOCK REPAIR
#
# SUCCESSOR TO
# ------------
# 14.3E-B failed after proving:
#
#   [PASS] parent 14.3E-A
#   [PASS] portable source
#   [PASS] current context schema
#   [PASS] MorphCandidateProvider Protocol
#   [PASS] Morph provider validator
#   [PASS] Morph runtime explicit-provider contract
#   [PASS] no hidden concrete Morph provider
#
# Failure occurred ONLY because 14.3E-B assumed:
#
#     authority["content_sha256"] ==
#         EXPECTED_MORPH_AUTHORITY_CONTENT_SHA
#
# That imposes a new top-level JSON schema on an already-frozen historical
# artifact.
#
#
# R1 REPAIR
# ---------
# Historical Morph authority is locked by:
#
#   1. exact file SHA — mandatory;
#   2. record top-level logical content SHA if present;
#   3. record whether expected historical content SHA is recursively embedded;
#   4. record canonical JSON hashes for diagnostic provenance;
#   5. DO NOT fail merely because an exact historical artifact does not expose
#      a newly invented top-level ``content_sha256`` field.
#
# Exact file SHA already cryptographically freezes the artifact bytes.
#
#
# PURPOSE
# -------
# Freeze the engineering dependency-injection seam:
#
#     PortableCurrentBestContext.morph_candidate_provider
#
# without implementing or inventing a concrete scientific Morph generator.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical Morph action pickle read
#   - GT
#   - evaluator
#   - performance
#   - scientific-policy changes
#   - structural-fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_a_"
    "post_v2_orchestration_seam_freeze_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


MORPH_AUTHORITY_REL = Path(
    "reports/"
    "notebook13_13_3d2r3_"
    "morph_authority_correction_and_"
    "semantic_reconstruction_freeze_v1.json"
)

MORPH_AUTHORITY = ROOT / MORPH_AUTHORITY_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_b_r1_"
    "morph_provider_seam_closure_"
    "historical_authority_content_lock_repair_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "e5b4791816b4a2b5dfd78a3318343401d08a59a422bbf3f133d082042f097e1e"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "435ea6322c079b297fcce5668af258d53a3ded2c7de748864fe79cfe0c0d9143"
)


EXPECTED_PORTABLE_SHA = (
    "5e1a2d64012b68e4e9b5cfb6909a4de849844ad5f5e10421a19c46b858ec9708"
)


EXPECTED_CONTEXT_SHA = (
    "40420de98e596e8f6b5eb67b7a2a5df9ab5a03b44fa1980e503fa9e4bd51c808"
)


EXPECTED_MORPH_PROVIDER_PROTOCOL_SHA = (
    "faeec0eef21035189ed1fba8f2a8c81df0446df2a79f244caa2545e341eecfc2"
)


EXPECTED_MORPH_PROVIDER_VALIDATOR_SHA = (
    "1d8b4dc334d9586d12c94298c68f6f4b5197299a353a9fe13185b2363e88c035"
)


EXPECTED_MORPH_RUNTIME_SHA = (
    "db173c58a83eddd282712915514c31739815195ccb99edaa84a3ff5e16f1ffcf"
)


EXPECTED_MORPH_AUTHORITY_FILE_SHA = (
    "ac0cb4959b8a9cb54784d0fd37d4da89a87b6e110298f67af00e712deebf8afb"
)

EXPECTED_MORPH_AUTHORITY_CONTENT_SHA = (
    "9a0ca7a1d2013a9cd0ddfe74ec00e9f787505c19ca69c6d0c4e297db03a701b0"
)


EXPECTED_MORPH_STRUCTURAL_ORACLE_SHA = (
    "9c08f88c9824912af72a5538e6f23cd6bc7e14ba4ffc30e26ad9ad6ca37dd9eb"
)


EXPECTED_MORPH_AUTHORITY_TYPE = (
    "SEMANTIC_SOURCE_LOCK_PLUS_"
    "HISTORICAL_GT_BLIND_STRUCTURAL_ORACLE"
)


EXPECTED_CONTEXT_FIELDS = [
    "b256_pred_995_root",
    "b256_pred_9975_root",
    "s9995_pred_995_root",
    "s9995_pred_9975_root",
    "dataset_root",
    "physical_scale_zyx_um",
    "frozen_v9_policy_bundle",
    "association_provider",
    "association_provider_identity_sha256",
    "strict",
]


EXPECTED_PROVIDER_IDENTITY_FIELDS = [
    "provider_id",
    "provider_version",
    "provider_source_sha256",
    "provider_authority_status",
]


AUTHORIZED_CONTEXT_FIELD = (
    "morph_candidate_provider"
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    data: bytes,
) -> str:

    return hashlib.sha256(
        data
    ).hexdigest()


def sha256_text(
    text: str,
) -> str:

    return sha256_bytes(
        text.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:

    if obj == target:

        return True


    if isinstance(
        obj,
        dict,
    ):

        return any(
            contains_exact_value(
                key,
                target,
            )
            or
            contains_exact_value(
                value,
                target,
            )

            for key, value
            in obj.items()
        )


    if isinstance(
        obj,
        (
            list,
            tuple,
        ),
    ):

        return any(
            contains_exact_value(
                item,
                target,
            )

            for item in obj
        )


    return False


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in (
            obj.items()
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )


    assert (
        source
        is not None
    )


    return source


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def top_level_classes(
    tree: ast.Module,
) -> dict[
    str,
    ast.ClassDef,
]:

    return {
        node.name:
            node

        for node in (
            tree.body
        )

        if isinstance(
            node,
            ast.ClassDef,
        )
    }


def source_sha(
    text: str,
    node: ast.AST,
) -> str:

    return sha256_text(
        node_source(
            text,
            node,
        )
    )


def class_fields(
    cls: ast.ClassDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for stmt in (
        cls.body
    ):

        if not isinstance(
            stmt,
            ast.AnnAssign,
        ):

            continue


        if not isinstance(
            stmt.target,
            ast.Name,
        ):

            continue


        result.append({
            "name":
                stmt.target.id,

            "annotation":
                ast.unparse(
                    stmt.annotation
                ),

            "default":
                (
                    ast.unparse(
                        stmt.value
                    )
                    if stmt.value
                    is not None
                    else None
                ),

            "line":
                stmt.lineno,
        })


    return result


def class_methods(
    cls: ast.ClassDef,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        stmt.name:
            stmt

        for stmt in (
            cls.body
        )

        if isinstance(
            stmt,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def function_signature(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> dict[str, Any]:

    positional = (
        list(
            fn.args.posonlyargs
        )
        +
        list(
            fn.args.args
        )
    )


    positional_defaults = (
        [None]
        * (
            len(
                positional
            )
            -
            len(
                fn.args.defaults
            )
        )
        +
        list(
            fn.args.defaults
        )
    )


    return {
        "positional": [
            {
                "name":
                    arg.arg,

                "annotation":
                    (
                        ast.unparse(
                            arg.annotation
                        )
                        if arg.annotation
                        is not None
                        else None
                    ),

                "default":
                    (
                        ast.unparse(
                            default
                        )
                        if default
                        is not None
                        else None
                    ),
            }

            for arg, default
            in zip(
                positional,
                positional_defaults,
            )
        ],

        "kwonly": [
            {
                "name":
                    arg.arg,

                "annotation":
                    (
                        ast.unparse(
                            arg.annotation
                        )
                        if arg.annotation
                        is not None
                        else None
                    ),

                "default":
                    (
                        ast.unparse(
                            default
                        )
                        if default
                        is not None
                        else None
                    ),
            }

            for arg, default
            in zip(
                fn.args.kwonlyargs,
                fn.args.kw_defaults,
            )
        ],

        "return_annotation":
            (
                ast.unparse(
                    fn.returns
                )
                if fn.returns
                is not None
                else None
            ),
    }


def function_calls(
    fn: ast.AST,
) -> set[str]:

    return {
        dotted_name(
            call.func
        )
        or "<dynamic-call>"

        for call in ast.walk(
            fn
        )

        if isinstance(
            call,
            ast.Call,
        )
    }


def target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    def visit(
        value: ast.AST,
    ):

        if isinstance(
            value,
            ast.Name,
        ):

            result.add(
                value.id
            )


        elif isinstance(
            value,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for child in (
                value.elts
            ):

                visit(
                    child
                )


        elif isinstance(
            value,
            ast.Starred,
        ):

            visit(
                value.value
            )


    visit(
        node
    )


    return result


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert path.exists(), (
        path
    )


    file_sha = sha256_file(
        path
    )


    assert (
        file_sha
        == EXPECTED_PARENT_FILE_SHA
    ), (
        "14.3E-A file SHA drift.\n"
        f"expected={EXPECTED_PARENT_FILE_SHA}\n"
        f"actual={file_sha}"
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )


    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "readiness"
        ][
            "Morph_provider_seam_ready"
        ]
        is False
    )


    assert (
        obj[
            "readiness"
        ][
            "post_v2_provider_implementation_ready"
        ]
        is False
    )


    assert (
        obj[
            "Morph_provider_resolution"
        ][
            "resolution_mode"
        ]
        == "UNRESOLVED"
    )


    return obj


# =============================================================================
# 4. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    MORPH_AUTHORITY,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-B-R1 "
    "MORPH PROVIDER SEAM CLOSURE "
    "— HISTORICAL AUTHORITY CONTENT-LOCK REPAIR",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "PORTABLE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "MORPH AUTHORITY:",
    MORPH_AUTHORITY_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Parent exact lock
# =============================================================================

parent = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_A_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  previous Morph seam:",
    parent[
        "Morph_provider_resolution"
    ][
        "resolution_mode"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Current portable source lock
# =============================================================================

portable_sha = sha256_file(
    PORTABLE
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
), (
    "Portable source changed after 14.3E-A."
)


portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_text,
    filename=str(
        PORTABLE
    ),
)


functions = top_level_functions(
    portable_tree
)


classes = top_level_classes(
    portable_tree
)


print(
    "[PASS] PORTABLE_SOURCE_LOCK",
    flush=True,
)

print(
    "  SHA:",
    portable_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. PortableCurrentBestContext lock
# =============================================================================

assert (
    "PortableCurrentBestContext"
    in classes
)


context_cls = (
    classes[
        "PortableCurrentBestContext"
    ]
)


context_sha = source_sha(
    portable_text,
    context_cls,
)


assert (
    context_sha
    == EXPECTED_CONTEXT_SHA
)


context_fields = class_fields(
    context_cls
)


context_field_names = [
    record[
        "name"
    ]

    for record in (
        context_fields
    )
]


assert (
    context_field_names
    == EXPECTED_CONTEXT_FIELDS
)


assert (
    AUTHORIZED_CONTEXT_FIELD
    not in context_field_names
)


print(
    "[PASS] CURRENT_CONTEXT_SCHEMA_LOCK",
    flush=True,
)

print(
    "  SHA:",
    context_sha,
    flush=True,
)

print(
    "  fields:",
    context_field_names,
    flush=True,
)

print(
    "  morph_candidate_provider currently present:",
    False,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. MorphCandidateProvider exact contract
# =============================================================================

assert (
    "MorphCandidateProvider"
    in classes
)


provider_cls = (
    classes[
        "MorphCandidateProvider"
    ]
)


provider_sha = source_sha(
    portable_text,
    provider_cls,
)


assert (
    provider_sha
    == EXPECTED_MORPH_PROVIDER_PROTOCOL_SHA
)


provider_fields = class_fields(
    provider_cls
)


provider_field_names = [
    record[
        "name"
    ]

    for record in (
        provider_fields
    )
]


for required in (
    EXPECTED_PROVIDER_IDENTITY_FIELDS
):

    assert (
        required
        in provider_field_names
    ), (
        "MorphCandidateProvider lost identity field "
        f"{required!r}."
    )


provider_methods = class_methods(
    provider_cls
)


assert (
    "build_candidates"
    in provider_methods
)


build_candidates_fn = (
    provider_methods[
        "build_candidates"
    ]
)


build_candidates_signature = (
    function_signature(
        build_candidates_fn
    )
)


print(
    "[PASS] MORPH_PROVIDER_PROTOCOL_LOCK",
    flush=True,
)

print(
    "  SHA:",
    provider_sha,
    flush=True,
)

print(
    "  identity fields:",
    EXPECTED_PROVIDER_IDENTITY_FIELDS,
    flush=True,
)

print(
    "  build_candidates signature:",
    json.dumps(
        build_candidates_signature,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Provider validator exact lock
# =============================================================================

assert (
    "validate_morph_candidate_provider"
    in functions
)


validator_fn = (
    functions[
        "validate_morph_candidate_provider"
    ]
)


validator_sha = source_sha(
    portable_text,
    validator_fn,
)


assert (
    validator_sha
    == EXPECTED_MORPH_PROVIDER_VALIDATOR_SHA
)


validator_source = node_source(
    portable_text,
    validator_fn,
)


for required in (
    EXPECTED_PROVIDER_IDENTITY_FIELDS
):

    assert (
        required
        in validator_source
    )


assert (
    "build_candidates"
    in validator_source
)


print(
    "[PASS] MORPH_PROVIDER_VALIDATOR_LOCK",
    flush=True,
)

print(
    "  SHA:",
    validator_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Morph public runtime lock
# =============================================================================

assert (
    "apply_portable_morph_div_v2"
    in functions
)


morph_fn = (
    functions[
        "apply_portable_morph_div_v2"
    ]
)


morph_sha = source_sha(
    portable_text,
    morph_fn,
)


assert (
    morph_sha
    == EXPECTED_MORPH_RUNTIME_SHA
)


morph_signature = function_signature(
    morph_fn
)


morph_source = node_source(
    portable_text,
    morph_fn,
)


morph_calls = function_calls(
    morph_fn
)


morph_parameter_names = [
    item[
        "name"
    ]

    for item in (
        morph_signature[
            "positional"
        ]
        +
        morph_signature[
            "kwonly"
        ]
    )
]


assert (
    "provider"
    in morph_parameter_names
)


assert (
    "validate_morph_candidate_provider"
    in morph_calls
)


assert (
    "provider.build_candidates"
    in morph_calls
)


print(
    "[PASS] MORPH_RUNTIME_EXPLICIT_PROVIDER_CONTRACT",
    flush=True,
)

print(
    "  SHA:",
    morph_sha,
    flush=True,
)

print(
    "  parameters:",
    morph_parameter_names,
    flush=True,
)

print(
    "  strict validator used:",
    True,
    flush=True,
)

print(
    "  provider.build_candidates used:",
    True,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Prove no hidden production Morph provider
# =============================================================================

provider_constructors = []


for call in ast.walk(
    portable_tree
):

    if not isinstance(
        call,
        ast.Call,
    ):

        continue


    name = dotted_name(
        call.func
    )


    if (
        name is not None
        and name.endswith(
            "MorphCandidateProvider"
        )
    ):

        provider_constructors.append({
            "line":
                call.lineno,

            "call":
                name,

            "source":
                node_source(
                    portable_text,
                    call,
                ),
        })


provider_factory_functions = []


for name, fn in (
    functions.items()
):

    lower = (
        name.lower()
    )


    if (
        "morph"
        not in lower
        or
        "provider"
        not in lower
    ):

        continue


    calls = function_calls(
        fn
    )


    if any(
        call.endswith(
            "MorphCandidateProvider"
        )

        for call in calls
    ):

        provider_factory_functions.append(
            name
        )


module_provider_assignments = []


for stmt in (
    portable_tree.body
):

    value = None
    names = set()


    if isinstance(
        stmt,
        ast.Assign,
    ):

        for target in (
            stmt.targets
        ):

            names |= target_names(
                target
            )


        value = stmt.value


    elif isinstance(
        stmt,
        ast.AnnAssign,
    ):

        names |= target_names(
            stmt.target
        )


        value = stmt.value


    if (
        value
        is None
    ):

        continue


    value_source = ast.unparse(
        value
    )


    if (
        "MorphCandidateProvider"
        in value_source
    ):

        module_provider_assignments.append({
            "line":
                stmt.lineno,

            "names":
                sorted(
                    names
                ),

            "value":
                value_source,
        })


assert (
    provider_constructors
    == []
)


assert (
    provider_factory_functions
    == []
)


assert (
    module_provider_assignments
    == []
)


print(
    "[PASS] NO_HIDDEN_CONCRETE_MORPH_PROVIDER",
    flush=True,
)

print(
    "  constructors:",
    len(
        provider_constructors
    ),
    flush=True,
)

print(
    "  factories:",
    provider_factory_functions,
    flush=True,
)

print(
    "  module instances:",
    module_provider_assignments,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Historical Morph authority lock — R1 REPAIR
# =============================================================================

authority_file_sha = sha256_file(
    MORPH_AUTHORITY
)


assert (
    authority_file_sha
    == EXPECTED_MORPH_AUTHORITY_FILE_SHA
), (
    "Historical Morph authority file SHA drift.\n"
    f"expected={EXPECTED_MORPH_AUTHORITY_FILE_SHA}\n"
    f"actual  ={authority_file_sha}"
)


authority = load_json(
    MORPH_AUTHORITY
)


authority_top_level_content_sha = (
    authority.get(
        "content_sha256"
    )
)


authority_contains_expected_content_sha = (
    contains_exact_value(
        authority,
        EXPECTED_MORPH_AUTHORITY_CONTENT_SHA,
    )
)


authority_canonical_full_sha = (
    canonical_json_sha(
        authority
    )
)


authority_payload_sha = None


if (
    "content_sha256"
    in authority
):

    authority_payload = dict(
        authority
    )


    authority_payload.pop(
        "content_sha256"
    )


    authority_payload_sha = (
        canonical_json_sha(
            authority_payload
        )
    )


# Determine observed historical wrapping mode.
if (
    authority_top_level_content_sha
    == EXPECTED_MORPH_AUTHORITY_CONTENT_SHA
):

    authority_lock_mode = (
        "EXACT_FILE_SHA_PLUS_TOP_LEVEL_EXPECTED_CONTENT_SHA"
    )


elif (
    authority_contains_expected_content_sha
):

    authority_lock_mode = (
        "EXACT_FILE_SHA_PLUS_EMBEDDED_EXPECTED_CONTENT_SHA"
    )


elif (
    authority_canonical_full_sha
    == EXPECTED_MORPH_AUTHORITY_CONTENT_SHA
):

    authority_lock_mode = (
        "EXACT_FILE_SHA_PLUS_CANONICAL_FULL_JSON_SHA"
    )


elif (
    authority_payload_sha
    == EXPECTED_MORPH_AUTHORITY_CONTENT_SHA
):

    authority_lock_mode = (
        "EXACT_FILE_SHA_PLUS_CANONICAL_PAYLOAD_SHA"
    )


else:

    # This is still a valid byte-exact historical lock.
    #
    # The previously recorded logical content SHA is retained as provenance
    # metadata but is not allowed to impose a newly invented JSON envelope.
    authority_lock_mode = (
        "EXACT_FILE_SHA_ONLY_"
        "EXPECTED_LOGICAL_CONTENT_SHA_NOT_EMBEDDED"
    )


# Exact file SHA is the mandatory cryptographic authority lock.
historical_authority_exact_locked = (
    authority_file_sha
    == EXPECTED_MORPH_AUTHORITY_FILE_SHA
)


assert (
    historical_authority_exact_locked
)


print(
    "[PASS] HISTORICAL_MORPH_AUTHORITY_BYTE_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    authority_file_sha,
    flush=True,
)

print(
    "  expected historical logical content SHA:",
    EXPECTED_MORPH_AUTHORITY_CONTENT_SHA,
    flush=True,
)

print(
    "  top-level content_sha256:",
    repr(
        authority_top_level_content_sha
    ),
    flush=True,
)

print(
    "  expected SHA recursively embedded:",
    authority_contains_expected_content_sha,
    flush=True,
)

print(
    "  canonical full JSON SHA:",
    authority_canonical_full_sha,
    flush=True,
)

print(
    "  canonical payload SHA:",
    authority_payload_sha,
    flush=True,
)

print(
    "  observed lock mode:",
    authority_lock_mode,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Historical semantic authority proof
#
# These are substantive Morph authority facts and remain mandatory.
# =============================================================================

assert (
    contains_exact_value(
        authority,
        EXPECTED_MORPH_AUTHORITY_TYPE,
    )
), (
    "Frozen Morph authority type missing."
)


assert (
    contains_exact_value(
        authority,
        EXPECTED_MORPH_STRUCTURAL_ORACLE_SHA,
    )
), (
    "Frozen historical Morph structural-oracle SHA missing."
)


authority_leaves = flatten_json(
    authority
)


relevant_authority_leaves = []


for leaf in (
    authority_leaves
):

    path_lower = (
        leaf[
            "path"
        ].lower()
    )


    value = (
        leaf[
            "value"
        ]
    )


    value_text = (
        str(
            value
        )
    )


    value_lower = (
        value_text.lower()
    )


    if (
        "morph"
        in path_lower
        or
        "provider"
        in path_lower
        or
        "runtime"
        in path_lower
        or
        "structural"
        in path_lower
        or
        "oracle"
        in path_lower
        or
        "authority"
        in path_lower
        or
        "morph"
        in value_lower
        or
        EXPECTED_MORPH_STRUCTURAL_ORACLE_SHA
        in value_lower
    ):

        relevant_authority_leaves.append(
            leaf
        )


print(
    "[PASS] HISTORICAL_MORPH_SEMANTIC_AUTHORITY_LOCK",
    flush=True,
)

print(
    "  authority type:",
    EXPECTED_MORPH_AUTHORITY_TYPE,
    flush=True,
)

print(
    "  structural oracle SHA:",
    EXPECTED_MORPH_STRUCTURAL_ORACLE_SHA,
    flush=True,
)

print(
    flush=True,
)


print(
    "=" * 112,
    flush=True,
)

print(
    "HISTORICAL MORPH AUTHORITY RELEVANT RECORDS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for leaf in (
    relevant_authority_leaves[
        :180
    ]
):

    print(
        " ",
        leaf[
            "path"
        ],
        "=",
        repr(
            leaf[
                "value"
            ]
        ),
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 14. Fixed post-V2 public interface
# =============================================================================

post_v2_interface = {
    "name":
        "post_v2_graph_provider",

    "positional":
        [
            "dataset_id",
        ],

    "keyword_only":
        [
            "context",
        ],

    "return":
        "graph",
}


assert (
    post_v2_interface[
        "keyword_only"
    ]
    == [
        "context",
    ]
)


assert (
    "provider"
    not in post_v2_interface[
        "keyword_only"
    ]
)


assert (
    "morph_provider"
    not in post_v2_interface[
        "keyword_only"
    ]
)


print(
    "[PASS] FIXED_POST_V2_PROVIDER_INTERFACE",
    flush=True,
)

print(
    json.dumps(
        post_v2_interface,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Dependency-seam proof
# =============================================================================

dependency_facts = {
    "post_v2_interface_has_context":
        True,

    "post_v2_interface_has_no_separate_Morph_provider":
        True,

    "Morph_runtime_requires_explicit_provider":
        (
            "provider"
            in morph_parameter_names
        ),

    "Morph_provider_protocol_exists":
        True,

    "Morph_provider_strict_validator_exists":
        True,

    "Morph_provider_self_identifies":
        all(
            field
            in provider_field_names

            for field in (
                EXPECTED_PROVIDER_IDENTITY_FIELDS
            )
        ),

    "context_currently_lacks_Morph_provider":
        (
            AUTHORIZED_CONTEXT_FIELD
            not in context_field_names
        ),

    "no_module_provider_instance":
        (
            module_provider_assignments
            == []
        ),

    "no_provider_factory":
        (
            provider_factory_functions
            == []
        ),

    "no_provider_constructor":
        (
            provider_constructors
            == []
        ),

    "historical_action_runtime_fallback_forbidden":
        True,
}


assert all(
    dependency_facts.values()
)


print(
    "[PASS] MORPH_PROVIDER_DEPENDENCY_SEAM_PROOF",
    flush=True,
)

print(
    json.dumps(
        dependency_facts,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Authorized engineering seam
# =============================================================================

seam_contract = {
    "resolution":
        "EXPLICIT_CONTEXT_DEPENDENCY_INJECTION",

    "context_class":
        "PortableCurrentBestContext",

    "new_context_field":
        AUTHORIZED_CONTEXT_FIELD,

    "required_runtime_object_contract":
        "MorphCandidateProvider",

    "duplicate_identity_field_required":
        False,

    "reason_no_duplicate_identity_field": (
        "MorphCandidateProvider already carries provider_id, "
        "provider_version, provider_source_sha256, and "
        "provider_authority_status; validate_morph_candidate_provider "
        "remains the existing fail-closed provider gate."
    ),

    "post_v2_call_contract": {
        "function":
            "apply_portable_morph_div_v2",

        "dataset_argument":
            "dataset_id",

        "graph_argument":
            "post_frozen_v9_graph",

        "context_argument":
            "context",

        "provider_argument":
            "context.morph_candidate_provider",

        "return_audit":
            False,
    },

    "forbidden_resolutions": [
        "HISTORICAL_MORPH_ACTION_RUNTIME_LOOKUP",
        "IMPLICIT_MODULE_GLOBAL_PROVIDER",
        "UNVALIDATED_PROVIDER",
        "SCIENTIFIC_POLICY_RETUNING",
    ],

    "scientific_policy_changed":
        False,

    "candidate_generator_implemented_by_this_seam":
        False,

    "concrete_provider_structural_fidelity_verified":
        False,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "AUTHORIZED MORPH PROVIDER SEAM",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        seam_contract,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 17. Readiness split
# =============================================================================

orchestration_seam_closed = (
    seam_contract[
        "resolution"
    ]
    == "EXPLICIT_CONTEXT_DEPENDENCY_INJECTION"
)


post_v2_orchestrator_implementation_authorized = (
    orchestration_seam_closed
)


# Important:
# The seam is now known, but there is still no production concrete provider.
fresh_concrete_morph_provider_resolved = (
    False
)


fresh_end_to_end_post_v2_runtime_ready = (
    post_v2_orchestrator_implementation_authorized
    and
    fresh_concrete_morph_provider_resolved
)


post_v2_structural_replay_ready = (
    fresh_concrete_morph_provider_resolved
)


readiness = {
    "Morph_provider_dependency_seam_closed":
        orchestration_seam_closed,

    "authorized_context_field":
        AUTHORIZED_CONTEXT_FIELD,

    "post_v2_orchestrator_implementation_authorized":
        post_v2_orchestrator_implementation_authorized,

    "fresh_concrete_Morph_provider_resolved":
        fresh_concrete_morph_provider_resolved,

    "fresh_end_to_end_post_v2_runtime_ready":
        fresh_end_to_end_post_v2_runtime_ready,

    "post_v2_structural_replay_ready":
        post_v2_structural_replay_ready,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "READINESS SPLIT",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        readiness,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 18. Next mutation plan
# =============================================================================

next_mutation_plan = {
    "allowed_source_changes": [
        {
            "target":
                "PortableCurrentBestContext",

            "change":
                (
                    "ADD_REQUIRED_FIELD_"
                    "morph_candidate_provider"
                ),
        },

        {
            "target":
                "post_v2_graph_provider",

            "change":
                (
                    "IMPLEMENT_EXACT_"
                    "FROZENV9_MORPH_V1_V2_ORCHESTRATION"
                ),
        },
    ],

    "required_orchestration_order": [
        "_build_portable_frozen_v9",
        "apply_portable_morph_div_v2",
        "_reconstruct_portable_mutual_cont_v1",
        "apply_portable_mutual_cont_v2",
    ],

    "FrozenV9_tuple_rule":
        "USE_ELEMENT_0_GRAPH",

    "Morph_provider_argument":
        "context.morph_candidate_provider",

    "R1P50":
        "EXCLUDED",

    "forbidden_changes": [
        "IMPLEMENT_NEW_SCIENTIFIC_MORPH_GENERATOR",
        "CHANGE_MORPH_SELECTION_POLICY",
        "CHANGE_MORPH_CONFIG",
        "READ_HISTORICAL_MORPH_ACTION_TABLE",
        "CHANGE_V1_POLICY",
        "CHANGE_V2_POLICY",
        "INCLUDE_R1P50",
    ],
}


print(
    "=" * 112,
    flush=True,
)

print(
    "NEXT SOURCE-MUTATION PLAN",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        next_mutation_plan,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Semantic gates
#
# Note:
# Expected historical logical content SHA embedding is intentionally NOT a
# gate. Exact file SHA is the byte-authoritative historical lock.
# =============================================================================

semantic_gates = {
    "parent_14_3e_a_locked":
        True,

    "portable_source_locked":
        (
            portable_sha
            == EXPECTED_PORTABLE_SHA
        ),

    "context_schema_locked":
        (
            context_sha
            == EXPECTED_CONTEXT_SHA
        ),

    "Morph_provider_protocol_locked":
        (
            provider_sha
            == EXPECTED_MORPH_PROVIDER_PROTOCOL_SHA
        ),

    "Morph_provider_identity_self_describing":
        all(
            field
            in provider_field_names

            for field in (
                EXPECTED_PROVIDER_IDENTITY_FIELDS
            )
        ),

    "Morph_provider_build_candidates_contract_present":
        (
            "build_candidates"
            in provider_methods
        ),

    "Morph_provider_validator_locked":
        (
            validator_sha
            == EXPECTED_MORPH_PROVIDER_VALIDATOR_SHA
        ),

    "Morph_runtime_locked":
        (
            morph_sha
            == EXPECTED_MORPH_RUNTIME_SHA
        ),

    "Morph_runtime_requires_explicit_provider":
        (
            "provider"
            in morph_parameter_names
        ),

    "no_hidden_concrete_Morph_provider":
        (
            provider_constructors
            == []
            and
            provider_factory_functions
            == []
            and
            module_provider_assignments
            == []
        ),

    "historical_Morph_authority_exact_file_locked":
        historical_authority_exact_locked,

    "historical_Morph_authority_type_locked":
        contains_exact_value(
            authority,
            EXPECTED_MORPH_AUTHORITY_TYPE,
        ),

    "historical_Morph_oracle_identity_locked":
        contains_exact_value(
            authority,
            EXPECTED_MORPH_STRUCTURAL_ORACLE_SHA,
        ),

    "historical_authority_wrapper_observed_not_invented":
        True,

    "fixed_post_v2_interface_has_context_only_dependency_carrier":
        (
            post_v2_interface[
                "keyword_only"
            ]
            == [
                "context",
            ]
        ),

    "explicit_context_injection_seam_closed":
        orchestration_seam_closed,

    "concrete_provider_not_falsely_claimed":
        (
            fresh_concrete_morph_provider_resolved
            is False
        ),

    "historical_action_runtime_lookup_forbidden":
        True,

    "scientific_policy_unchanged":
        True,

    "no_source_mutation":
        True,

    "GT_not_used":
        True,

    "evaluator_not_used":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3E-B-R1 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in (
    semantic_gates.items()
):

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok in (
        semantic_gates.items()
    )

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-B-R1 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 20. Next stage
# =============================================================================

next_stage = (
    "NOTEBOOK14_14_3E_C_"
    "MORPH_CONTEXT_INJECTION_AND_POST_V2_GRAPH_PROVIDER_IMPLEMENTATION"
)


# =============================================================================
# 21. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_B_R1",

    "title": (
        "MORPH_PROVIDER_SEAM_CLOSURE_"
        "HISTORICAL_AUTHORITY_CONTENT_LOCK_REPAIR"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_DEPENDENCY_INJECTION_SEAM_ONLY",

    "failed_predecessor": {
        "stage":
            "NOTEBOOK14_14_3E_B",

        "failure_point":
            "HISTORICAL_MORPH_AUTHORITY_CONTENT_LOCK",

        "failure": (
            "The historical Morph authority artifact was incorrectly "
            "required to expose the expected logical content SHA as a "
            "top-level `content_sha256` field. The exact historical file "
            "is instead byte-locked by its frozen file SHA, while wrapper "
            "and logical-content representations are recorded rather than "
            "newly imposed."
        ),

        "successful_evidence_before_failure": [
            "PARENT_14_3E_A_LOCK",
            "PORTABLE_SOURCE_LOCK",
            "CURRENT_CONTEXT_SCHEMA_LOCK",
            "MORPH_PROVIDER_PROTOCOL_LOCK",
            "MORPH_PROVIDER_VALIDATOR_LOCK",
            "MORPH_RUNTIME_EXPLICIT_PROVIDER_CONTRACT",
            "NO_HIDDEN_CONCRETE_MORPH_PROVIDER",
        ],

        "source_mutation":
            False,

        "formal_artifact_written":
            False,
    },

    "parent_14_3e_a": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "portable_source": {
        "path":
            str(
                PORTABLE_REL
            ),

        "file_sha256":
            portable_sha,
    },

    "historical_Morph_authority": {
        "path":
            str(
                MORPH_AUTHORITY_REL
            ),

        "file_sha256":
            authority_file_sha,

        "expected_historical_logical_content_sha256":
            EXPECTED_MORPH_AUTHORITY_CONTENT_SHA,

        "top_level_content_sha256":
            authority_top_level_content_sha,

        "contains_expected_logical_content_sha256":
            authority_contains_expected_content_sha,

        "canonical_full_json_sha256":
            authority_canonical_full_sha,

        "canonical_payload_sha256":
            authority_payload_sha,

        "observed_lock_mode":
            authority_lock_mode,

        "authority_type":
            EXPECTED_MORPH_AUTHORITY_TYPE,

        "structural_oracle_sha256":
            EXPECTED_MORPH_STRUCTURAL_ORACLE_SHA,

        "runtime_historical_action_lookup":
            False,
    },

    "current_context": {
        "class":
            "PortableCurrentBestContext",

        "source_sha256":
            context_sha,

        "fields":
            context_fields,

        "Morph_provider_field_currently_present":
            False,
    },

    "Morph_provider_contract": {
        "class":
            "MorphCandidateProvider",

        "source_sha256":
            provider_sha,

        "fields":
            provider_fields,

        "build_candidates_signature":
            build_candidates_signature,

        "validator_function":
            "validate_morph_candidate_provider",

        "validator_sha256":
            validator_sha,
    },

    "Morph_runtime": {
        "function":
            "apply_portable_morph_div_v2",

        "source_sha256":
            morph_sha,

        "signature":
            morph_signature,

        "explicit_provider_required":
            True,

        "provider_validation":
            True,
    },

    "hidden_provider_inventory": {
        "constructors":
            provider_constructors,

        "factories":
            provider_factory_functions,

        "module_assignments":
            module_provider_assignments,
    },

    "dependency_facts":
        dependency_facts,

    "authorized_seam":
        seam_contract,

    "readiness":
        readiness,

    "next_mutation_plan":
        next_mutation_plan,

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_provider_dependency_seam":
            "CLOSED",

        "Morph_provider_dependency_mode":
            "EXPLICIT_CONTEXT_DEPENDENCY_INJECTION",

        "authorized_context_field":
            AUTHORIZED_CONTEXT_FIELD,

        "concrete_Morph_provider":
            "NOT_YET_RESOLVED",

        "concrete_Morph_provider_structural_fidelity":
            "NOT_YET_VERIFIED",

        "post_V2_orchestrator_implementation":
            "AUTHORIZED_NOT_YET_IMPLEMENTED",

        "fresh_end_to_end_post_V2_runtime":
            "NOT_YET_READY",

        "post_V2_structural_replay":
            "NOT_YET_READY",

        "historical_Morph_action_runtime_use":
            False,

        "historical_Morph_oracle_fidelity_use":
            True,

        "R1P50_in_post_V2":
            False,

        "scientific_policy_changed":
            False,

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_read":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No concrete Morph candidate provider was implemented.",
        "No concrete Morph provider structural fidelity was verified.",
        "No historical Morph action artifact was used as runtime input.",
        "No post-V2 provider was implemented in this stage.",
        "No post-V2 structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance evidence.",
        "No scientific policy or threshold changed.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-B-R1 artifact exists but differs "
        "from the exact seam closure."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 22. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-B-R1 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "HISTORICAL MORPH AUTHORITY LOCK:",
    flush=True,
)

print(
    "  exact file SHA:",
    authority_file_sha,
    flush=True,
)

print(
    "  expected logical content SHA:",
    EXPECTED_MORPH_AUTHORITY_CONTENT_SHA,
    flush=True,
)

print(
    "  top-level content_sha256:",
    repr(
        authority_top_level_content_sha
    ),
    flush=True,
)

print(
    "  expected SHA embedded:",
    authority_contains_expected_content_sha,
    flush=True,
)

print(
    "  canonical full JSON SHA:",
    authority_canonical_full_sha,
    flush=True,
)

print(
    "  canonical payload SHA:",
    authority_payload_sha,
    flush=True,
)

print(
    "  lock mode:",
    authority_lock_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "MORPH PROVIDER SEAM:",
    flush=True,
)

print(
    "  previous:",
    "UNRESOLVED",
    flush=True,
)

print(
    "  resolved mode:",
    "EXPLICIT_CONTEXT_DEPENDENCY_INJECTION",
    flush=True,
)

print(
    "  authorized context field:",
    AUTHORIZED_CONTEXT_FIELD,
    flush=True,
)

print(
    flush=True,
)


print(
    "WHY CONTEXT INJECTION:",
    flush=True,
)

print(
    "  Morph runtime requires provider:",
    True,
    flush=True,
)

print(
    "  fixed post-V2 interface exposes context:",
    True,
    flush=True,
)

print(
    "  fixed post-V2 interface exposes separate provider:",
    False,
    flush=True,
)

print(
    "  module provider instance:",
    False,
    flush=True,
)

print(
    "  provider factory:",
    False,
    flush=True,
)

print(
    "  historical action fallback permitted:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "READINESS:",
    flush=True,
)

print(
    "  Morph dependency seam closed:",
    orchestration_seam_closed,
    flush=True,
)

print(
    "  post-V2 orchestrator implementation authorized:",
    post_v2_orchestrator_implementation_authorized,
    flush=True,
)

print(
    "  concrete fresh Morph provider resolved:",
    fresh_concrete_morph_provider_resolved,
    flush=True,
)

print(
    "  fresh end-to-end post-V2 runtime ready:",
    fresh_end_to_end_post_v2_runtime_ready,
    flush=True,
)

print(
    "  structural replay ready:",
    post_v2_structural_replay_ready,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  Morph provider dependency seam = CLOSED",
    flush=True,
)

print(
    "  concrete Morph provider = NOT YET RESOLVED",
    flush=True,
)

print(
    "  post-V2 orchestrator implementation = AUTHORIZED",
    flush=True,
)

print(
    "  post-V2 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  B12 structural fidelity = NOT YET VERIFIED",
    flush=True,
)

print(
    "  performance evidence = NONE",
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-C-R3 — POST-V2 IMPLEMENTATION
#                UNIQUE EXECUTABLE-AST HARNESS REPAIR
#
# SUCCESSOR TO
# ------------
# 14.3E-C:
#   implementation completed
#   79 targeted tests passed
#   failed on format-sensitive missing-provider string gate
#
# 14.3E-C-R1:
#   repaired missing-provider check with AST
#   failed because whole-module text scan treated documentation-only
#   "R1P50 is intentionally excluded" as a runtime dependency
#
# 14.3E-C-R2:
#   repaired R1P50 check with executable AST
#   then failed at:
#
#       len(missing_provider_guards) == 1
#
# ROOT CAUSE
# ----------
# R2 executable_nodes() did:
#
#       yield stmt
#       yield from ast.walk(stmt)
#
# But ast.walk(stmt) already includes stmt itself.
#
# Therefore a top-level ast.If was emitted twice, so the SAME
# `morph_provider is None` guard was counted twice.
#
#
# R3 REPAIR
# ---------
# - no source mutation
# - no test mutation
# - exact already-tested source state locked
# - explicitly reproduce R2 duplicate traversal
# - use object-identity-deduplicated executable AST traversal
# - re-prove all post-V2 runtime invariants
# - rerun py_compile and targeted pytest live
# - write formal implementation artifact
#
#
# NO:
#   - concrete Morph provider implementation
#   - historical action runtime lookup
#   - R1P50 runtime
#   - GT
#   - evaluator
#   - performance
#   - structural-fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import subprocess
import sys

from pathlib import Path
from typing import Any, Iterable


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_b_r1_"
    "morph_provider_seam_closure_"
    "historical_authority_content_lock_repair_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


PROVIDER_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_post_v2_graph_provider.py"
)

PROVIDER = ROOT / PROVIDER_REL


TEST_REL = Path(
    "tests/test_stage12_post_v2_graph_provider.py"
)

TEST = ROOT / TEST_REL


PRIMARY_TEST_REL = Path(
    "tests/test_stage12_portable_current_best.py"
)

PRIMARY_TEST = ROOT / PRIMARY_TEST_REL


V1_TEST_REL = Path(
    "tests/test_stage12_v1_graph_runtime.py"
)

V1_TEST = ROOT / V1_TEST_REL


B12_TEST_REL = Path(
    "tests/test_stage12_b12_portable.py"
)

B12_TEST = ROOT / B12_TEST_REL


FAILED_C_ARTIFACT = (
    ROOT
    /
    "reports/"
    "notebook14_14_3e_c_"
    "morph_context_injection_and_"
    "post_v2_graph_provider_implementation_v1.json"
)


FAILED_R1_ARTIFACT = (
    ROOT
    /
    "reports/"
    "notebook14_14_3e_c_r1_"
    "post_v2_graph_provider_implementation_"
    "harness_repair_v1.json"
)


FAILED_R2_ARTIFACT = (
    ROOT
    /
    "reports/"
    "notebook14_14_3e_c_r2_"
    "post_v2_graph_provider_implementation_"
    "executable_ast_harness_repair_v1.json"
)


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_c_r3_"
    "post_v2_graph_provider_implementation_"
    "unique_executable_ast_harness_repair_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Exact identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "9a10cdf3081d9539bf2fddc895e5b3a683c347778895597e4a453f038d17d583"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "0b3ec6b200cfc6e1e3815e73612697fe647484d463936342b0e30717bec7b12e"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)

EXPECTED_CONTEXT_SHA = (
    "c125127bb541a016a9fa78eee80f713eefb261983889ca5cfbce7489923e12d3"
)


EXPECTED_PROVIDER_FILE_SHA = (
    "3caec19d4446bd44415c3039d8e16e2f7b6ac7cdd03d4ba27941bb583055470e"
)

EXPECTED_PROVIDER_FUNCTION_SHA = (
    "dd17e3888408c7d950c4b6d5449752c695ed317b7b3bd5a0354c671110c61084"
)


EXPECTED_TEST_SHA = (
    "9278057d86ecf43d0917daa23b574a2a63578376237554f675ccfb5f5da943de"
)


EXPECTED_CONTEXT_FIELDS = [
    "b256_pred_995_root",
    "b256_pred_9975_root",
    "s9995_pred_995_root",
    "s9995_pred_9975_root",
    "dataset_root",
    "physical_scale_zyx_um",
    "frozen_v9_policy_bundle",
    "association_provider",
    "association_provider_identity_sha256",
    "morph_candidate_provider",
    "strict",
]


STAGE_CALLS = [
    "portable._build_portable_frozen_v9",
    "portable.apply_portable_morph_div_v2",
    "portable._reconstruct_portable_mutual_cont_v1",
    "portable.apply_portable_mutual_cont_v2",
]


HISTORICAL_MORPH_FILENAME = (
    "stage11_fold0_morphdiv_v2_actions_locked.pkl"
)

HISTORICAL_V1_FILENAME = (
    "stage12_fold0_mutual_cont_v1_actions_locked.pkl"
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    result = ast.get_source_segment(
        text,
        node,
    )


    assert (
        result is not None
    )


    return result


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in tree.body

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def top_level_classes(
    tree: ast.Module,
) -> dict[
    str,
    ast.ClassDef,
]:

    return {
        node.name:
            node

        for node in tree.body

        if isinstance(
            node,
            ast.ClassDef,
        )
    }


def source_sha(
    text: str,
    node: ast.AST,
) -> str:

    return sha256_text(
        node_source(
            text,
            node,
        )
    )


def class_fields(
    cls: ast.ClassDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for stmt in cls.body:

        if not isinstance(
            stmt,
            ast.AnnAssign,
        ):

            continue


        if not isinstance(
            stmt.target,
            ast.Name,
        ):

            continue


        result.append({
            "name":
                stmt.target.id,

            "annotation":
                ast.unparse(
                    stmt.annotation
                ),

            "default":
                (
                    ast.unparse(
                        stmt.value
                    )
                    if stmt.value is not None
                    else None
                ),

            "line":
                stmt.lineno,
        })


    return result


def stream_command(
    command: list[str],
    *,
    cwd: Path,
) -> int:

    print(
        "$",
        " ".join(
            command
        ),
        flush=True,
    )


    process = subprocess.Popen(
        command,
        cwd=str(
            cwd
        ),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )


    assert (
        process.stdout is not None
    )


    for line in process.stdout:

        print(
            line,
            end="",
            flush=True,
        )


    return int(
        process.wait()
    )


# =============================================================================
# 3. Parent lock
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "Morph_provider_dependency_seam"
        ]
        == "CLOSED"
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "post_V2_orchestrator_implementation"
        ]
        == "AUTHORIZED_NOT_YET_IMPLEMENTED"
    )


    return obj


# =============================================================================
# 4. Executable AST traversal
# =============================================================================

def is_docstring_stmt(
    stmt: ast.stmt,
) -> bool:

    return (
        isinstance(
            stmt,
            ast.Expr,
        )
        and
        isinstance(
            stmt.value,
            ast.Constant,
        )
        and
        isinstance(
            stmt.value.value,
            str,
        )
    )


def executable_function_statements(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[ast.stmt]:

    body = list(
        fn.body
    )


    if (
        body
        and
        is_docstring_stmt(
            body[
                0
            ]
        )
    ):

        body = body[
            1:
        ]


    return body


# -----------------------------------------------------------------------------
# R2 FLAWED traversal — retained only to reproduce the failure.
# -----------------------------------------------------------------------------

def flawed_executable_nodes(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> Iterable[ast.AST]:

    for stmt in executable_function_statements(
        fn
    ):

        # BUG:
        # this emits stmt once ...
        yield stmt

        # ... and ast.walk(stmt) emits stmt itself AGAIN.
        yield from ast.walk(
            stmt
        )


# -----------------------------------------------------------------------------
# R3 corrected traversal.
#
# ast.walk(stmt) already includes stmt itself.
#
# We additionally deduplicate by object identity so one AST object can never
# count twice even if traversal structure changes later.
# -----------------------------------------------------------------------------

def unique_executable_nodes(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> Iterable[ast.AST]:

    seen: set[int] = set()


    for stmt in executable_function_statements(
        fn
    ):

        for node in ast.walk(
            stmt
        ):

            marker = id(
                node
            )


            if marker in seen:

                continue


            seen.add(
                marker
            )


            yield node


# =============================================================================
# 5. Runtime semantic extractors
# =============================================================================

def is_name_none_compare(
    node: ast.AST,
    *,
    name: str,
) -> bool:

    if not isinstance(
        node,
        ast.Compare,
    ):

        return False


    if (
        len(
            node.ops
        )
        != 1
        or
        len(
            node.comparators
        )
        != 1
    ):

        return False


    if not isinstance(
        node.ops[
            0
        ],
        ast.Is,
    ):

        return False


    left = node.left

    right = node.comparators[
        0
    ]


    left_name_right_none = (
        isinstance(
            left,
            ast.Name,
        )
        and
        left.id
        == name
        and
        isinstance(
            right,
            ast.Constant,
        )
        and
        right.value is None
    )


    left_none_right_name = (
        isinstance(
            right,
            ast.Name,
        )
        and
        right.id
        == name
        and
        isinstance(
            left,
            ast.Constant,
        )
        and
        left.value is None
    )


    return (
        left_name_right_none
        or
        left_none_right_name
    )


def body_raises_runtime_contract_error(
    statements: list[ast.stmt],
) -> bool:

    for stmt in statements:

        for node in ast.walk(
            stmt
        ):

            if not isinstance(
                node,
                ast.Raise,
            ):

                continue


            if not isinstance(
                node.exc,
                ast.Call,
            ):

                continue


            exc_name = dotted_name(
                node.exc.func
            )


            if (
                exc_name is not None
                and
                exc_name.endswith(
                    "PortableRuntimeContractError"
                )
            ):

                return True


    return False


def provider_guard_records(
    nodes: Iterable[ast.AST],
) -> list[
    dict[str, Any]
]:

    result = []


    for node in nodes:

        if not isinstance(
            node,
            ast.If,
        ):

            continue


        if not is_name_none_compare(
            node.test,
            name="morph_provider",
        ):

            continue


        result.append({
            "object_id":
                id(
                    node
                ),

            "line":
                node.lineno,

            "column":
                node.col_offset,

            "test":
                ast.unparse(
                    node.test
                ),

            "raises_runtime_contract_error":
                body_raises_runtime_contract_error(
                    node.body
                ),
        })


    return result


def executable_calls(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for node in unique_executable_nodes(
        fn
    ):

        if not isinstance(
            node,
            ast.Call,
        ):

            continue


        result.append({
            "name":
                dotted_name(
                    node.func
                ),

            "line":
                node.lineno,

            "column":
                node.col_offset,

            "source":
                ast.unparse(
                    node
                ),
        })


    return sorted(
        result,
        key=lambda item: (
            item[
                "line"
            ],
            item[
                "column"
            ],
        ),
    )


def executable_identifiers(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[
    dict[str, Any]
]:

    result = []

    seen = set()


    for node in unique_executable_nodes(
        fn
    ):

        name = None


        if isinstance(
            node,
            ast.Name,
        ):

            name = node.id


        elif isinstance(
            node,
            ast.Attribute,
        ):

            name = dotted_name(
                node
            )


        if name is None:

            continue


        key = (
            name,
            getattr(
                node,
                "lineno",
                None,
            ),
            getattr(
                node,
                "col_offset",
                None,
            ),
        )


        if key in seen:

            continue


        seen.add(
            key
        )


        result.append({
            "name":
                name,

            "line":
                getattr(
                    node,
                    "lineno",
                    None,
                ),

            "column":
                getattr(
                    node,
                    "col_offset",
                    None,
                ),
        })


    return result


def executable_strings(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for node in unique_executable_nodes(
        fn
    ):

        if not (
            isinstance(
                node,
                ast.Constant,
            )
            and
            isinstance(
                node.value,
                str,
            )
        ):

            continue


        result.append({
            "line":
                getattr(
                    node,
                    "lineno",
                    None,
                ),

            "value":
                node.value,
        })


    return result


def module_imports(
    tree: ast.Module,
) -> list[
    dict[str, Any]
]:

    result = []


    for stmt in tree.body:

        if isinstance(
            stmt,
            ast.Import,
        ):

            for alias in stmt.names:

                result.append({
                    "kind":
                        "import",

                    "module":
                        alias.name,

                    "name":
                        None,

                    "asname":
                        alias.asname,

                    "line":
                        stmt.lineno,
                })


        elif isinstance(
            stmt,
            ast.ImportFrom,
        ):

            for alias in stmt.names:

                result.append({
                    "kind":
                        "from",

                    "module":
                        stmt.module,

                    "name":
                        alias.name,

                    "asname":
                        alias.asname,

                    "line":
                        stmt.lineno,
                })


    return result


def stage_call_records(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> list[
    dict[str, Any]
]:

    result = []


    for node in unique_executable_nodes(
        fn
    ):

        if not isinstance(
            node,
            ast.Call,
        ):

            continue


        name = dotted_name(
            node.func
        )


        if name not in STAGE_CALLS:

            continue


        result.append({
            "name":
                name,

            "line":
                node.lineno,

            "column":
                node.col_offset,

            "args": [
                ast.unparse(
                    arg
                )

                for arg in node.args
            ],

            "keywords": {
                (
                    kw.arg
                    if kw.arg is not None
                    else "**"
                ):
                    ast.unparse(
                        kw.value
                    )

                for kw in node.keywords
            },
        })


    return sorted(
        result,
        key=lambda item: (
            item[
                "line"
            ],
            item[
                "column"
            ],
        ),
    )


def frozen_graph_element_zero_present(
    fn: ast.FunctionDef | ast.AsyncFunctionDef,
) -> bool:

    for node in unique_executable_nodes(
        fn
    ):

        if not isinstance(
            node,
            ast.Assign,
        ):

            continue


        target_names = {
            target.id

            for target in node.targets

            if isinstance(
                target,
                ast.Name,
            )
        }


        if (
            "post_frozen_v9_graph"
            not in target_names
        ):

            continue


        value = node.value


        if not isinstance(
            value,
            ast.Subscript,
        ):

            continue


        if not (
            isinstance(
                value.value,
                ast.Name,
            )
            and
            value.value.id
            == "frozen_result"
        ):

            continue


        try:

            index = ast.literal_eval(
                value.slice
            )

        except Exception:

            continue


        if index == 0:

            return True


    return False


# =============================================================================
# 6. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    PROVIDER,
    TEST,
    PRIMARY_TEST,
    V1_TEST,
    B12_TEST,
):

    assert required.exists(), (
        required
    )


# All three predecessors failed before formal artifact creation.
assert (
    not FAILED_C_ARTIFACT.exists()
), (
    "Unexpected formal artifact from failed 14.3E-C."
)


assert (
    not FAILED_R1_ARTIFACT.exists()
), (
    "Unexpected formal artifact from failed 14.3E-C-R1."
)


assert (
    not FAILED_R2_ARTIFACT.exists()
), (
    "Unexpected formal artifact from failed 14.3E-C-R2."
)


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-C-R3 "
    "POST-V2 IMPLEMENTATION UNIQUE EXECUTABLE-AST HARNESS REPAIR",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "PORTABLE:",
    PORTABLE_REL,
    flush=True,
)

print(
    "PROVIDER:",
    PROVIDER_REL,
    flush=True,
)

print(
    "TEST:",
    TEST_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Parent authority lock
# =============================================================================

parent = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_B_R1_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Exact implementation-state locks
# =============================================================================

portable_sha = sha256_file(
    PORTABLE
)

provider_sha = sha256_file(
    PROVIDER
)

test_sha = sha256_file(
    TEST
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
)


assert (
    provider_sha
    == EXPECTED_PROVIDER_FILE_SHA
)


assert (
    test_sha
    == EXPECTED_TEST_SHA
)


print(
    "[PASS] IMPLEMENTATION_STATE_EXACT_LOCK",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha,
    flush=True,
)

print(
    "  provider SHA:",
    provider_sha,
    flush=True,
)

print(
    "  test SHA:",
    test_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Context seam exact lock
# =============================================================================

portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_text,
    filename=str(
        PORTABLE
    ),
)


portable_classes = top_level_classes(
    portable_tree
)


assert (
    "PortableCurrentBestContext"
    in portable_classes
)


context_cls = (
    portable_classes[
        "PortableCurrentBestContext"
    ]
)


context_sha = source_sha(
    portable_text,
    context_cls,
)


assert (
    context_sha
    == EXPECTED_CONTEXT_SHA
)


context_fields = class_fields(
    context_cls
)


context_field_names = [
    item[
        "name"
    ]

    for item in context_fields
]


assert (
    context_field_names
    == EXPECTED_CONTEXT_FIELDS
)


morph_fields = [
    item

    for item in context_fields

    if (
        item[
            "name"
        ]
        == "morph_candidate_provider"
    )
]


assert (
    len(
        morph_fields
    )
    == 1
)


assert (
    "MorphCandidateProvider"
    in morph_fields[
        0
    ][
        "annotation"
    ]
)


assert (
    morph_fields[
        0
    ][
        "default"
    ]
    == "None"
)


print(
    "[PASS] MORPH_CONTEXT_SEAM_EXACT_LOCK",
    flush=True,
)

print(
    "  context SHA:",
    context_sha,
    flush=True,
)

print(
    "  field:",
    morph_fields[
        0
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Provider exact lock
# =============================================================================

provider_text = PROVIDER.read_text(
    encoding="utf-8"
)


provider_tree = ast.parse(
    provider_text,
    filename=str(
        PROVIDER
    ),
)


provider_functions = top_level_functions(
    provider_tree
)


assert (
    list(
        provider_functions
    )
    == [
        "post_v2_graph_provider",
    ]
)


provider_fn = (
    provider_functions[
        "post_v2_graph_provider"
    ]
)


provider_function_sha = source_sha(
    provider_text,
    provider_fn,
)


assert (
    provider_function_sha
    == EXPECTED_PROVIDER_FUNCTION_SHA
)


print(
    "[PASS] POST_V2_PROVIDER_FUNCTION_EXACT_LOCK",
    flush=True,
)

print(
    "  function SHA:",
    provider_function_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Explicitly reproduce R2 duplicate-AST bug
# =============================================================================

flawed_guards = provider_guard_records(
    flawed_executable_nodes(
        provider_fn
    )
)


unique_guards = provider_guard_records(
    unique_executable_nodes(
        provider_fn
    )
)


flawed_guard_semantic_keys = [
    (
        item[
            "line"
        ],
        item[
            "column"
        ],
        item[
            "test"
        ],
    )

    for item in flawed_guards
]


unique_flawed_semantic_keys = set(
    flawed_guard_semantic_keys
)


print(
    "=" * 112,
    flush=True,
)

print(
    "R2 DUPLICATE-AST FAILURE REPRODUCTION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "flawed traversal guard count:",
    len(
        flawed_guards
    ),
    flush=True,
)

print(
    "flawed traversal unique semantic guards:",
    len(
        unique_flawed_semantic_keys
    ),
    flush=True,
)

print(
    "corrected traversal guard count:",
    len(
        unique_guards
    ),
    flush=True,
)

print(
    "flawed guards:",
    json.dumps(
        flawed_guards,
        indent=2,
    ),
    flush=True,
)

print(
    "corrected guards:",
    json.dumps(
        unique_guards,
        indent=2,
    ),
    flush=True,
)


assert (
    len(
        flawed_guards
    )
    >
    len(
        unique_flawed_semantic_keys
    )
), (
    "R2 duplicate-root traversal bug did not reproduce."
)


assert (
    len(
        unique_flawed_semantic_keys
    )
    == 1
), (
    "R2 records do not collapse to one semantic guard."
)


assert (
    len(
        unique_guards
    )
    == 1
), (
    "Corrected traversal did not yield exactly one Morph-provider guard."
)


print(
    "[PASS] R2_DUPLICATE_AST_ROOT_VISIT_REPRODUCED",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Correct missing-provider fail-closed proof
# =============================================================================

guard = unique_guards[
    0
]


assert (
    guard[
        "test"
    ]
    == "morph_provider is None"
)


assert (
    guard[
        "raises_runtime_contract_error"
    ]
    is True
)


stage_records = stage_call_records(
    provider_fn
)


actual_stage_order = [
    item[
        "name"
    ]

    for item in stage_records
]


assert (
    actual_stage_order
    == STAGE_CALLS
)


frozen_line = next(
    item[
        "line"
    ]

    for item in stage_records

    if (
        item[
            "name"
        ]
        == "portable._build_portable_frozen_v9"
    )
)


assert (
    guard[
        "line"
    ]
    <
    frozen_line
)


print(
    "[PASS] MISSING_MORPH_PROVIDER_FAIL_CLOSED_UNIQUE_AST",
    flush=True,
)

print(
    "  unique guard count:",
    len(
        unique_guards
    ),
    flush=True,
)

print(
    "  guard line:",
    guard[
        "line"
    ],
    flush=True,
)

print(
    "  expression:",
    guard[
        "test"
    ],
    flush=True,
)

print(
    "  raises PortableRuntimeContractError:",
    guard[
        "raises_runtime_contract_error"
    ],
    flush=True,
)

print(
    "  FrozenV9 first line:",
    frozen_line,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Documentation-only R1P50 reference proof
# =============================================================================

module_docstring = ast.get_docstring(
    provider_tree,
    clean=False,
)


assert isinstance(
    module_docstring,
    str,
)


module_docstring_mentions_r1p50 = (
    "r1p50"
    in module_docstring.lower()
)


function_source = node_source(
    provider_text,
    provider_fn,
)


function_source_mentions_r1p50 = (
    "r1p50"
    in function_source.lower()
)


assert (
    module_docstring_mentions_r1p50
    is True
)


assert (
    function_source_mentions_r1p50
    is False
)


print(
    "[PASS] R1P50_DOCUMENTATION_ONLY_ROOT_CAUSE_LOCK",
    flush=True,
)

print(
    "  module docstring:",
    module_docstring_mentions_r1p50,
    flush=True,
)

print(
    "  executable function source:",
    function_source_mentions_r1p50,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Executable runtime dependency inventory
# =============================================================================

runtime_calls = executable_calls(
    provider_fn
)


runtime_identifiers = executable_identifiers(
    provider_fn
)


runtime_strings = executable_strings(
    provider_fn
)


imports = module_imports(
    provider_tree
)


r1p50_identifier_hits = [
    item

    for item in runtime_identifiers

    if (
        "r1p50"
        in item[
            "name"
        ].lower()
    )
]


r1p50_call_hits = [
    item

    for item in runtime_calls

    if (
        item[
            "name"
        ]
        is not None
        and
        "r1p50"
        in item[
            "name"
        ].lower()
    )
]


r1p50_import_hits = [
    item

    for item in imports

    if (
        "r1p50"
        in str(
            item.get(
                "module"
            )
        ).lower()
        or
        "r1p50"
        in str(
            item.get(
                "name"
            )
        ).lower()
        or
        "r1p50"
        in str(
            item.get(
                "asname"
            )
        ).lower()
    )
]


r1p50_runtime_string_hits = [
    item

    for item in runtime_strings

    if (
        "r1p50"
        in item[
            "value"
        ].lower()
    )
]


assert (
    r1p50_identifier_hits
    == []
)


assert (
    r1p50_call_hits
    == []
)


assert (
    r1p50_import_hits
    == []
)


assert (
    r1p50_runtime_string_hits
    == []
)


print(
    "[PASS] R1P50_EXECUTABLE_AST_EXCLUSION",
    flush=True,
)

print(
    "  identifiers:",
    r1p50_identifier_hits,
    flush=True,
)

print(
    "  calls:",
    r1p50_call_hits,
    flush=True,
)

print(
    "  imports:",
    r1p50_import_hits,
    flush=True,
)

print(
    "  executable strings:",
    r1p50_runtime_string_hits,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Other forbidden runtime dependencies
# =============================================================================

FORBIDDEN_IDENTIFIER_TOKENS = [
    "read_pickle",
    "evaluate_graph",
    "ground_truth",
    "gt_graph",
]


forbidden_identifier_hits = {
    token: [
        item

        for item in runtime_identifiers

        if (
            token
            in item[
                "name"
            ].lower()
        )
    ]

    for token in FORBIDDEN_IDENTIFIER_TOKENS
}


forbidden_call_hits = {
    token: [
        item

        for item in runtime_calls

        if (
            item[
                "name"
            ]
            is not None
            and
            token
            in item[
                "name"
            ].lower()
        )
    ]

    for token in FORBIDDEN_IDENTIFIER_TOKENS
}


historical_runtime_string_hits = [
    item

    for item in runtime_strings

    if (
        HISTORICAL_MORPH_FILENAME.lower()
        in item[
            "value"
        ].lower()
        or
        HISTORICAL_V1_FILENAME.lower()
        in item[
            "value"
        ].lower()
    )
]


assert all(
    hits
    == []

    for hits in forbidden_identifier_hits.values()
)


assert all(
    hits
    == []

    for hits in forbidden_call_hits.values()
)


assert (
    historical_runtime_string_hits
    == []
)


print(
    "[PASS] FORBIDDEN_EXECUTABLE_DEPENDENCIES_ABSENT",
    flush=True,
)

print(
    "  identifiers:",
    forbidden_identifier_hits,
    flush=True,
)

print(
    "  calls:",
    forbidden_call_hits,
    flush=True,
)

print(
    "  historical strings:",
    historical_runtime_string_hits,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Exact stage order
# =============================================================================

assert (
    actual_stage_order
    == STAGE_CALLS
)


print(
    "[PASS] POST_V2_STAGE_ORDER_UNIQUE_AST_LOCK",
    flush=True,
)

print(
    "  order:",
    actual_stage_order,
    flush=True,
)


for item in stage_records:

    print(
        " ",
        f"L{item['line']}",
        item[
            "name"
        ],
        flush=True,
    )

    print(
        "    args:",
        item[
            "args"
        ],
        flush=True,
    )

    print(
        "    keywords:",
        item[
            "keywords"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 17. FrozenV9 element zero
# =============================================================================

frozen_element_zero = (
    frozen_graph_element_zero_present(
        provider_fn
    )
)


assert (
    frozen_element_zero
)


print(
    "[PASS] FROZEN_V9_ELEMENT_ZERO_UNIQUE_AST_LOCK",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 18. Morph / V2 orchestration arguments
# =============================================================================

morph_records = [
    item

    for item in stage_records

    if (
        item[
            "name"
        ]
        == "portable.apply_portable_morph_div_v2"
    )
]


assert (
    len(
        morph_records
    )
    == 1
)


morph_record = morph_records[
    0
]


assert (
    morph_record[
        "keywords"
    ].get(
        "context"
    )
    == "context"
)


assert (
    morph_record[
        "keywords"
    ].get(
        "provider"
    )
    == "morph_provider"
)


assert (
    morph_record[
        "keywords"
    ].get(
        "return_audit"
    )
    == "False"
)


v2_records = [
    item

    for item in stage_records

    if (
        item[
            "name"
        ]
        == "portable.apply_portable_mutual_cont_v2"
    )
]


assert (
    len(
        v2_records
    )
    == 1
)


v2_record = v2_records[
    0
]


assert (
    v2_record[
        "keywords"
    ].get(
        "context"
    )
    == "context"
)


assert (
    v2_record[
        "keywords"
    ].get(
        "return_audit"
    )
    == "False"
)


print(
    "[PASS] MORPH_AND_V2_ARGUMENT_UNIQUE_AST_LOCK",
    flush=True,
)

print(
    "  Morph provider:",
    morph_record[
        "keywords"
    ][
        "provider"
    ],
    flush=True,
)

print(
    "  Morph return_audit:",
    morph_record[
        "keywords"
    ][
        "return_audit"
    ],
    flush=True,
)

print(
    "  V2 return_audit:",
    v2_record[
        "keywords"
    ][
        "return_audit"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. py_compile — live
# =============================================================================

compile_exit = stream_command(
    [
        sys.executable,
        "-m",
        "py_compile",
        str(
            PORTABLE
        ),
        str(
            PROVIDER
        ),
        str(
            TEST
        ),
    ],
    cwd=ROOT,
)


assert (
    compile_exit
    == 0
)


print(
    "[PASS] PY_COMPILE_RECONFIRMATION",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 20. Exact targeted regression suite — live
# =============================================================================

pytest_exit = stream_command(
    [
        sys.executable,
        "-m",
        "pytest",
        "-vv",
        "-s",
        str(
            TEST_REL
        ),
        str(
            PRIMARY_TEST_REL
        ),
        str(
            V1_TEST_REL
        ),
        str(
            B12_TEST_REL
        ),
    ],
    cwd=ROOT,
)


assert (
    pytest_exit
    == 0
), (
    f"Targeted pytest failed: {pytest_exit}"
)


print(
    "[PASS] TARGETED_PYTEST_RECONFIRMATION",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 21. No mutation during R3
# =============================================================================

assert (
    sha256_file(
        PORTABLE
    )
    == EXPECTED_PORTABLE_SHA
)


assert (
    sha256_file(
        PROVIDER
    )
    == EXPECTED_PROVIDER_FILE_SHA
)


assert (
    sha256_file(
        TEST
    )
    == EXPECTED_TEST_SHA
)


print(
    "[PASS] NO_SOURCE_OR_TEST_MUTATION_IN_R3",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 22. Final runtime gates
# =============================================================================

runtime_gates = {
    "context_has_Morph_provider_seam":
        (
            "morph_candidate_provider"
            in context_field_names
        ),

    "R2_duplicate_AST_bug_reproduced":
        (
            len(
                flawed_guards
            )
            >
            len(
                unique_flawed_semantic_keys
            )
        ),

    "unique_missing_provider_guard_exactly_one":
        (
            len(
                unique_guards
            )
            == 1
        ),

    "missing_Morph_provider_fail_closed":
        (
            guard[
                "raises_runtime_contract_error"
            ]
            is True
            and
            guard[
                "line"
            ]
            <
            frozen_line
        ),

    "exact_stage_order":
        (
            actual_stage_order
            == STAGE_CALLS
        ),

    "FrozenV9_element_zero":
        frozen_element_zero,

    "Morph_provider_from_context":
        (
            morph_record[
                "keywords"
            ].get(
                "provider"
            )
            == "morph_provider"
        ),

    "Morph_non_audit_return":
        (
            morph_record[
                "keywords"
            ].get(
                "return_audit"
            )
            == "False"
        ),

    "V2_non_audit_return":
        (
            v2_record[
                "keywords"
            ].get(
                "return_audit"
            )
            == "False"
        ),

    "R1P50_executable_absent":
        (
            r1p50_identifier_hits
            == []
            and
            r1p50_call_hits
            == []
            and
            r1p50_import_hits
            == []
            and
            r1p50_runtime_string_hits
            == []
        ),

    "R1P50_reference_documentation_only":
        (
            module_docstring_mentions_r1p50
            is True
            and
            function_source_mentions_r1p50
            is False
        ),

    "historical_action_runtime_absent":
        (
            historical_runtime_string_hits
            == []
        ),

    "read_pickle_runtime_absent":
        (
            forbidden_identifier_hits[
                "read_pickle"
            ]
            == []
            and
            forbidden_call_hits[
                "read_pickle"
            ]
            == []
        ),

    "GT_runtime_absent":
        (
            forbidden_identifier_hits[
                "ground_truth"
            ]
            == []
            and
            forbidden_identifier_hits[
                "gt_graph"
            ]
            == []
            and
            forbidden_call_hits[
                "ground_truth"
            ]
            == []
            and
            forbidden_call_hits[
                "gt_graph"
            ]
            == []
        ),

    "evaluator_runtime_absent":
        (
            forbidden_identifier_hits[
                "evaluate_graph"
            ]
            == []
            and
            forbidden_call_hits[
                "evaluate_graph"
            ]
            == []
        ),
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3E-C-R3 RUNTIME GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in runtime_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_runtime_gates = [
    gate

    for gate, ok in runtime_gates.items()

    if not ok
]


assert (
    failed_runtime_gates
    == []
), (
    "14.3E-C-R3 runtime gate failures: "
    f"{failed_runtime_gates}"
)


# =============================================================================
# 23. Readiness
# =============================================================================

readiness = {
    "Morph_provider_dependency_seam":
        "IMPLEMENTED",

    "post_v2_orchestrator":
        "IMPLEMENTED",

    "post_v2_provider_interface":
        "IMPLEMENTED",

    "concrete_Morph_provider":
        "NOT_YET_RESOLVED",

    "fresh_end_to_end_post_v2_runtime_ready":
        False,

    "post_v2_structural_replay_ready":
        False,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "READINESS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    json.dumps(
        readiness,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 24. Formal semantic gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_b_r1_locked":
        True,

    "implementation_files_exact_locked":
        (
            portable_sha
            == EXPECTED_PORTABLE_SHA
            and
            provider_sha
            == EXPECTED_PROVIDER_FILE_SHA
            and
            test_sha
            == EXPECTED_TEST_SHA
        ),

    "context_seam_exact_locked":
        (
            context_sha
            == EXPECTED_CONTEXT_SHA
        ),

    "provider_function_exact_locked":
        (
            provider_function_sha
            == EXPECTED_PROVIDER_FUNCTION_SHA
        ),

    "R2_harness_failure_root_cause_proven":
        (
            len(
                flawed_guards
            )
            >
            len(
                unique_flawed_semantic_keys
            )
            and
            len(
                unique_flawed_semantic_keys
            )
            == 1
        ),

    "unique_AST_traversal_proven":
        (
            len(
                unique_guards
            )
            == 1
        ),

    "missing_provider_fail_closed":
        runtime_gates[
            "missing_Morph_provider_fail_closed"
        ],

    "exact_stage_order":
        runtime_gates[
            "exact_stage_order"
        ],

    "FrozenV9_element_zero":
        runtime_gates[
            "FrozenV9_element_zero"
        ],

    "R1P50_runtime_absent":
        runtime_gates[
            "R1P50_executable_absent"
        ],

    "historical_action_runtime_absent":
        runtime_gates[
            "historical_action_runtime_absent"
        ],

    "GT_runtime_absent":
        runtime_gates[
            "GT_runtime_absent"
        ],

    "evaluator_runtime_absent":
        runtime_gates[
            "evaluator_runtime_absent"
        ],

    "py_compile_reconfirmed":
        (
            compile_exit
            == 0
        ),

    "targeted_pytest_reconfirmed":
        (
            pytest_exit
            == 0
        ),

    "no_source_mutation_in_R3":
        True,

    "concrete_Morph_provider_not_falsely_claimed":
        True,

    "scientific_policy_unchanged":
        True,
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3E-C-R3 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_semantic_gates = [
    gate

    for gate, ok in semantic_gates.items()

    if not ok
]


assert (
    failed_semantic_gates
    == []
), (
    "14.3E-C-R3 semantic failures: "
    f"{failed_semantic_gates}"
)


# =============================================================================
# 25. Next
# =============================================================================

next_stage = (
    "NOTEBOOK14_14_3E_D_"
    "CONCRETE_MORPH_CANDIDATE_PROVIDER_"
    "AUTHORITY_AND_IMPLEMENTATION"
)


# =============================================================================
# 26. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_C_R3",

    "title": (
        "POST_V2_GRAPH_PROVIDER_IMPLEMENTATION_"
        "UNIQUE_EXECUTABLE_AST_HARNESS_REPAIR"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification": (
        "ENGINEERING_IMPLEMENTATION_VERIFICATION_"
        "UNIQUE_AST_HARNESS_REPAIR"
    ),

    "failed_predecessors": [
        {
            "stage":
                "NOTEBOOK14_14_3E_C",

            "failure_type":
                "FORMAT_SENSITIVE_STRING_FALSE_NEGATIVE",

            "failure_gate":
                "missing_Morph_provider_fail_closed",

            "source_mutation_completed":
                True,

            "targeted_tests":
                "79_PASSED",

            "formal_artifact_written":
                False,
        },

        {
            "stage":
                "NOTEBOOK14_14_3E_C_R1",

            "failure_type":
                "MODULE_DOCUMENTATION_FALSE_POSITIVE",

            "failure_gate":
                "R1P50_absence",

            "source_mutation_completed":
                False,

            "formal_artifact_written":
                False,
        },

        {
            "stage":
                "NOTEBOOK14_14_3E_C_R2",

            "failure_type":
                "DUPLICATE_AST_ROOT_VISIT",

            "failure_gate":
                "unique_missing_provider_guard",

            "root_cause": (
                "R2 executable_nodes yielded each top-level statement "
                "directly and then ast.walk(statement) yielded that same "
                "AST object again."
            ),

            "flawed_guard_count":
                len(
                    flawed_guards
                ),

            "unique_semantic_guard_count":
                len(
                    unique_flawed_semantic_keys
                ),

            "source_mutation_completed":
                False,

            "formal_artifact_written":
                False,
        },
    ],

    "parent_14_3e_b_r1": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "implementation_state": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha,

            "context_sha256":
                context_sha,

            "context_fields":
                context_fields,
        },

        "post_v2_provider": {
            "path":
                str(
                    PROVIDER_REL
                ),

            "file_sha256":
                provider_sha,

            "function_sha256":
                provider_function_sha,
        },

        "test": {
            "path":
                str(
                    TEST_REL
                ),

            "file_sha256":
                test_sha,
        },
    },

    "R2_duplicate_AST_repair": {
        "flawed_guard_records":
            flawed_guards,

        "flawed_guard_count":
            len(
                flawed_guards
            ),

        "flawed_unique_semantic_guard_count":
            len(
                unique_flawed_semantic_keys
            ),

        "corrected_guard_records":
            unique_guards,

        "corrected_guard_count":
            len(
                unique_guards
            ),

        "repair":
            "OBJECT_ID_DEDUPLICATED_AST_WALK",
    },

    "runtime_AST_contract": {
        "missing_provider_guard":
            guard,

        "stage_order":
            actual_stage_order,

        "stage_records":
            stage_records,

        "FrozenV9_element_zero":
            frozen_element_zero,

        "Morph_record":
            morph_record,

        "V2_record":
            v2_record,

        "R1P50_identifier_hits":
            r1p50_identifier_hits,

        "R1P50_call_hits":
            r1p50_call_hits,

        "R1P50_import_hits":
            r1p50_import_hits,

        "R1P50_executable_string_hits":
            r1p50_runtime_string_hits,

        "module_docstring_mentions_R1P50":
            module_docstring_mentions_r1p50,

        "runtime_function_mentions_R1P50":
            function_source_mentions_r1p50,

        "forbidden_identifier_hits":
            forbidden_identifier_hits,

        "forbidden_call_hits":
            forbidden_call_hits,

        "historical_runtime_string_hits":
            historical_runtime_string_hits,
    },

    "validation": {
        "py_compile_exit_code":
            compile_exit,

        "pytest_exit_code":
            pytest_exit,
    },

    "runtime_gates":
        runtime_gates,

    "readiness":
        readiness,

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_provider_dependency_seam":
            "IMPLEMENTED",

        "post_V2_orchestrator":
            "IMPLEMENTED",

        "post_V2_provider_interface":
            "IMPLEMENTED",

        "concrete_Morph_provider":
            "NOT_YET_RESOLVED",

        "fresh_end_to_end_post_V2_runtime":
            "NOT_YET_READY",

        "post_V2_structural_replay":
            "NOT_YET_READY",

        "post_V2_structural_fidelity":
            "NOT_YET_VERIFIED",

        "B12_structural_fidelity":
            "NOT_YET_VERIFIED",

        "R1P50_runtime_dependency":
            False,

        "historical_Morph_action_runtime_use":
            False,

        "historical_V1_action_runtime_use":
            False,

        "scientific_policy_changed":
            False,

        "source_mutation_in_14_3E_C":
            True,

        "source_mutation_in_R1":
            False,

        "source_mutation_in_R2":
            False,

        "source_mutation_in_R3":
            False,

        "GT_read":
            False,

        "evaluator_read":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No source or test file was modified by 14.3E-C-R3.",
        "No concrete Morph candidate provider was implemented.",
        "No R1P50 runtime dependency exists.",
        "No historical Morph action artifact is used at runtime.",
        "No historical V1 action artifact is used at runtime.",
        "No post-V2 structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-C-R3 artifact exists but differs "
        "from the exact verified implementation."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 27. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-C-R3 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "R2 HARNESS BUG:",
    flush=True,
)

print(
    "  flawed guard count:",
    len(
        flawed_guards
    ),
    flush=True,
)

print(
    "  unique semantic guards:",
    len(
        unique_flawed_semantic_keys
    ),
    flush=True,
)

print(
    "  corrected guard count:",
    len(
        unique_guards
    ),
    flush=True,
)

print(
    "  repair:",
    "OBJECT-ID DEDUPLICATED AST WALK",
    flush=True,
)

print(
    flush=True,
)


print(
    "POST-V2 IMPLEMENTATION:",
    flush=True,
)

print(
    "  context Morph seam:",
    "IMPLEMENTED",
    flush=True,
)

print(
    "  missing provider:",
    "FAIL CLOSED BEFORE FROZENV9",
    flush=True,
)

print(
    "  order:",
    "FrozenV9[0] -> Morph -> V1 -> V2",
    flush=True,
)

print(
    "  R1P50 runtime:",
    "ABSENT",
    flush=True,
)

print(
    "  historical action runtime:",
    "ABSENT",
    flush=True,
)

print(
    flush=True,
)


print(
    "VALIDATION:",
    flush=True,
)

print(
    "  py_compile:",
    "PASS",
    flush=True,
)

print(
    "  targeted pytest:",
    "PASS",
    flush=True,
)

print(
    "  source mutation in R3:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "READINESS:",
    flush=True,
)

print(
    "  Morph dependency seam:",
    "IMPLEMENTED",
    flush=True,
)

print(
    "  post-V2 orchestrator:",
    "IMPLEMENTED",
    flush=True,
)

print(
    "  concrete Morph provider:",
    "NOT YET RESOLVED",
    flush=True,
)

print(
    "  fresh end-to-end post-V2 runtime ready:",
    False,
    flush=True,
)

print(
    "  structural replay ready:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-A-R1
# CONCRETE MORPH PROVIDER AUTHORITY RECOVERY
# — NEGATIVE-GATE + FALSE-POSITIVE CLASSIFICATION REPAIR
#
# PREDECESSOR 14.3E-D-A
# ----------------------
# Substantive discovery succeeded, but:
#
#   source_mutation=False
#   GT_opened=False
#   evaluator_executed=False
#   performance_metric_generated=False
#
# were incorrectly inserted directly as gate booleans, producing four FAILs.
#
# More importantly, predecessor reported:
#
#   authority-backed function candidates = 0
#   historical-writer-backed functions   = 0
#   strong clean functions               = 1
#
# where the sole "strong" function was the CURRENT portable
# apply_portable_morph_div_v2 application runtime — not a concrete candidate
# generator.
#
# Therefore predecessor classification
#
#   HISTORICAL_WRITER_AND_SOURCE_CANDIDATES_FOUND
#
# is too permissive.
#
#
# THIS SUCCESSOR
# --------------
# 1. Lock 14.3E-C-R3 and current sources.
# 2. Lock historical Morph authority.
# 3. NEVER open the historical Morph .pkl.
# 4. Locate every notebook occurrence of the exact Morph action artifact.
# 5. Use notebook-global symbolic taint/dataflow to distinguish:
#
#       TARGET_WRITER
#       TARGET_READER
#       MENTION_ONLY
#
#    rather than "artifact mentioned + some .to_pickle exists".
#
# 6. Strictly scan source functions for actual candidate-generation evidence.
#    Current portable application/validator functions are explicitly NOT
#    considered concrete generator authority.
# 7. Correct negative-state gates.
# 8. Freeze a conservative recovery classification.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical Morph pickle open
#   - GT open
#   - evaluator execution
#   - performance
#   - implementation authorization unless concrete source authority is proven
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any, Iterable


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_c_r3_"
    "post_v2_graph_provider_implementation_"
    "unique_executable_ast_harness_repair_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


POST_V2_PROVIDER_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_post_v2_graph_provider.py"
)

POST_V2_PROVIDER = ROOT / POST_V2_PROVIDER_REL


MORPH_AUTHORITY_REL = Path(
    "reports/"
    "notebook13_13_3d2r3_"
    "morph_authority_correction_and_"
    "semantic_reconstruction_freeze_v1.json"
)

MORPH_AUTHORITY = ROOT / MORPH_AUTHORITY_REL


FAILED_DA_ARTIFACT = (
    ROOT
    /
    "reports/"
    "notebook14_14_3e_d_a_"
    "concrete_morph_candidate_provider_"
    "authority_recovery_v1.json"
)


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_a_r1_"
    "concrete_morph_candidate_provider_"
    "authority_recovery_gate_and_classification_repair_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "b99fedf770b292ed5c38850133e258d513442cfd70476e81008b7af2b5550a88"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "1957e9266c202a30e140f4cf11275716bc9246e2dce6c3e6ffc6a576c33c453e"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_POST_V2_PROVIDER_SHA = (
    "3caec19d4446bd44415c3039d8e16e2f7b6ac7cdd03d4ba27941bb583055470e"
)


EXPECTED_MORPH_AUTHORITY_FILE_SHA = (
    "ac0cb4959b8a9cb54784d0fd37d4da89a87b6e110298f67af00e712deebf8afb"
)


TARGET_ARTIFACT_REL = Path(
    "reports/"
    "stage11_division_probe/"
    "stage11_fold0_morphdiv_v2_actions_locked.pkl"
)

TARGET_ARTIFACT = ROOT / TARGET_ARTIFACT_REL

TARGET_BASENAME = TARGET_ARTIFACT.name


EXPECTED_TARGET_ARTIFACT_SHA = (
    "9c08f88c9824912af72a5538e6f23cd6bc7e14ba4ffc30e26ad9ad6ca37dd9eb"
)


EXPECTED_AUTHORITY_TYPE = (
    "SEMANTIC_SOURCE_LOCK_PLUS_"
    "HISTORICAL_GT_BLIND_STRUCTURAL_ORACLE"
)


# Frozen current Morph contract.
MORPH_SCHEMA = (
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
)

MORPH_CONFIG_ID = (
    "K3_R3_P100"
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def safe_rel(
    path: Path,
) -> str:

    try:

        return str(
            path.resolve().relative_to(
                ROOT
            )
        )

    except Exception:

        return str(
            path
        )


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    value = ast.get_source_segment(
        text,
        node,
    )


    assert (
        value is not None
    )


    return value


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in tree.body

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in obj.items():

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:

    return any(
        record[
            "value"
        ]
        == target

        for record in flatten_json(
            obj
        )
    )


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "post_V2_orchestrator"
        ]
        == "IMPLEMENTED"
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "concrete_Morph_provider"
        ]
        == "NOT_YET_RESOLVED"
    )


    return obj


# =============================================================================
# 4. Notebook symbolic target-path dataflow
#
# We propagate a boolean "this variable denotes / contains the exact target
# artifact path" across notebook cells.
#
# This is intentionally narrow:
#
#   - it does NOT execute notebook code;
#   - it does NOT open target pickle;
#   - it only analyzes code syntax and string/path aliases.
# =============================================================================

def target_literal_in_node(
    node: ast.AST,
) -> bool:

    for child in ast.walk(
        node
    ):

        if (
            isinstance(
                child,
                ast.Constant,
            )
            and
            isinstance(
                child.value,
                str,
            )
            and
            TARGET_BASENAME
            in child.value
        ):

            return True


    return False


def referenced_names(
    node: ast.AST,
) -> set[str]:

    return {
        child.id

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Name,
        )
    }


def expression_is_target_tainted(
    node: ast.AST,
    tainted_names: set[str],
) -> bool:

    if target_literal_in_node(
        node
    ):

        return True


    names = referenced_names(
        node
    )


    return bool(
        names
        &
        tainted_names
    )


def assignment_target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    def walk_target(
        value: ast.AST,
    ):

        if isinstance(
            value,
            ast.Name,
        ):

            result.add(
                value.id
            )


        elif isinstance(
            value,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for child in value.elts:

                walk_target(
                    child
                )


        elif isinstance(
            value,
            ast.Starred,
        ):

            walk_target(
                value.value
            )


    walk_target(
        node
    )


    return result


def update_taint_from_statement(
    stmt: ast.stmt,
    tainted_names: set[str],
):

    if isinstance(
        stmt,
        ast.Assign,
    ):

        rhs_tainted = expression_is_target_tainted(
            stmt.value,
            tainted_names,
        )


        names = set()


        for target in stmt.targets:

            names |= assignment_target_names(
                target
            )


        for name in names:

            if rhs_tainted:

                tainted_names.add(
                    name
                )


            else:

                tainted_names.discard(
                    name
                )


    elif isinstance(
        stmt,
        ast.AnnAssign,
    ):

        names = assignment_target_names(
            stmt.target
        )


        rhs_tainted = (
            stmt.value is not None
            and
            expression_is_target_tainted(
                stmt.value,
                tainted_names,
            )
        )


        for name in names:

            if rhs_tainted:

                tainted_names.add(
                    name
                )


            else:

                tainted_names.discard(
                    name
                )


def call_target_path_argument(
    call: ast.Call,
) -> ast.AST | None:

    call_name = dotted_name(
        call.func
    )


    if call_name is None:

        return None


    # pd.to_pickle(obj, path)
    if call_name in {
        "pd.to_pickle",
        "pandas.to_pickle",
    }:

        if len(
            call.args
        ) >= 2:

            return call.args[
                1
            ]


        for kw in call.keywords:

            if kw.arg in {
                "filepath_or_buffer",
                "path",
            }:

                return kw.value


        return None


    # frame.to_pickle(path)
    if call_name.endswith(
        ".to_pickle"
    ):

        if call.args:

            return call.args[
                0
            ]


        for kw in call.keywords:

            if kw.arg in {
                "path",
                "filepath_or_buffer",
            }:

                return kw.value


    return None


def read_pickle_argument(
    call: ast.Call,
) -> ast.AST | None:

    call_name = dotted_name(
        call.func
    )


    if (
        call_name is None
        or
        not call_name.endswith(
            "read_pickle"
        )
    ):

        return None


    if call.args:

        return call.args[
            0
        ]


    for kw in call.keywords:

        if kw.arg in {
            "filepath_or_buffer",
            "path",
        }:

            return kw.value


    return None


def analyze_notebook_target_io(
    path: Path,
) -> list[
    dict[str, Any]
]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    tainted_names: set[str] = set()

    records = []


    for cell_index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        if (
            cell.get(
                "cell_type"
            )
            != "code"
        ):

            continue


        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        mentions_literal = (
            TARGET_BASENAME
            in source
        )


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            if mentions_literal:

                records.append({
                    "path":
                        safe_rel(
                            path
                        ),

                    "file_sha256":
                        sha256_file(
                            path
                        ),

                    "cell_index":
                        cell_index,

                    "cell_sha256":
                        sha256_text(
                            source
                        ),

                    "classification":
                        "UNPARSEABLE_TARGET_MENTION",

                    "target_write_count":
                        0,

                    "target_read_count":
                        0,

                    "target_literal_mentioned":
                        True,

                    "tainted_names_after":
                        sorted(
                            tainted_names
                        ),

                    "source":
                        source,
                })


            continue


        write_sinks = []

        read_sinks = []


        # Process statements in order so aliases created in the cell can be
        # used by later sinks.
        for stmt in tree.body:

            update_taint_from_statement(
                stmt,
                tainted_names,
            )


            for node in ast.walk(
                stmt
            ):

                if not isinstance(
                    node,
                    ast.Call,
                ):

                    continue


                write_arg = call_target_path_argument(
                    node
                )


                if (
                    write_arg is not None
                    and
                    expression_is_target_tainted(
                        write_arg,
                        tainted_names,
                    )
                ):

                    write_sinks.append({
                        "line":
                            node.lineno,

                        "call":
                            ast.unparse(
                                node
                            ),
                    })


                read_arg = read_pickle_argument(
                    node
                )


                if (
                    read_arg is not None
                    and
                    expression_is_target_tainted(
                        read_arg,
                        tainted_names,
                    )
                ):

                    read_sinks.append({
                        "line":
                            node.lineno,

                        "call":
                            ast.unparse(
                                node
                            ),
                    })


        relevant = (
            mentions_literal
            or
            bool(
                write_sinks
            )
            or
            bool(
                read_sinks
            )
        )


        if not relevant:

            continue


        if write_sinks:

            classification = (
                "TARGET_WRITER"
            )


        elif read_sinks:

            classification = (
                "TARGET_READER"
            )


        else:

            classification = (
                "MENTION_ONLY"
            )


        records.append({
            "path":
                safe_rel(
                    path
                ),

            "file_sha256":
                sha256_file(
                    path
                ),

            "cell_index":
                cell_index,

            "cell_sha256":
                sha256_text(
                    source
                ),

            "classification":
                classification,

            "target_write_count":
                len(
                    write_sinks
                ),

            "target_read_count":
                len(
                    read_sinks
                ),

            "write_sinks":
                write_sinks,

            "read_sinks":
                read_sinks,

            "target_literal_mentioned":
                mentions_literal,

            "tainted_names_after":
                sorted(
                    tainted_names
                ),

            "source":
                source,
        })


    return records


# =============================================================================
# 5. Strict Python concrete-generator scan
#
# This deliberately excludes the CURRENT portable application layer.
#
# A function is a serious generator candidate only if it contains evidence for
# building the full candidate table, not merely consuming candidate_prob or
# adding already-locked actions to a graph.
# =============================================================================

REQUIRED_SCHEMA_SIGNALS = {
    "morph_rank",
    "candidate_rank",
    "candidate_prob",
    "source_pred_node_id",
    "target_pred_node_id",
}


GENERATOR_VERBS = (
    "candidate",
    "generate",
    "build",
    "extract",
    "recover",
    "morph",
    "rank",
)


def function_forbidden_hits(
    fn: ast.AST,
) -> dict[
    str,
    list[str],
]:

    identifiers = set()

    calls = set()

    strings = []


    for node in ast.walk(
        fn
    ):

        if isinstance(
            node,
            ast.Name,
        ):

            identifiers.add(
                node.id.lower()
            )


        elif isinstance(
            node,
            ast.Attribute,
        ):

            name = dotted_name(
                node
            )


            if name:

                identifiers.add(
                    name.lower()
                )


        elif isinstance(
            node,
            ast.Call,
        ):

            name = dotted_name(
                node.func
            )


            if name:

                calls.add(
                    name.lower()
                )


        elif (
            isinstance(
                node,
                ast.Constant,
            )
            and
            isinstance(
                node.value,
                str,
            )
        ):

            strings.append(
                node.value.lower()
            )


    forbidden_tokens = (
        "ground_truth",
        "gt_graph",
        "evaluate_graph",
        "evaluator",
        "read_pickle",
    )


    result = {}


    for token in forbidden_tokens:

        hits = []


        hits.extend(
            f"identifier:{value}"

            for value in identifiers

            if token in value
        )


        hits.extend(
            f"call:{value}"

            for value in calls

            if token in value
        )


        hits.extend(
            f"string:{value[:160]}"

            for value in strings

            if token in value
        )


        result[
            token
        ] = hits


    return result


def scan_strict_generator_candidates() -> list[
    dict[str, Any]
]:

    records = []


    scan_roots = [
        ROOT / "scripts",
        ROOT / "src",
    ]


    for scan_root in scan_roots:

        if not scan_root.exists():

            continue


        for path in scan_root.rglob(
            "*.py"
        ):

            # Current portable module is application contract authority,
            # explicitly NOT concrete generator authority.
            if (
                path.resolve()
                == PORTABLE.resolve()
            ):

                continue


            try:

                text = path.read_text(
                    encoding="utf-8"
                )


                tree = ast.parse(
                    text,
                    filename=str(
                        path
                    ),
                )

            except Exception:

                continue


            for fn in top_level_functions(
                tree
            ).values():

                source = node_source(
                    text,
                    fn,
                )


                lower = source.lower()


                schema_hits = sorted(
                    signal

                    for signal in REQUIRED_SCHEMA_SIGNALS

                    if signal in lower
                )


                config_hit = (
                    MORPH_CONFIG_ID.lower()
                    in lower
                )


                name_lower = (
                    fn.name.lower()
                )


                generator_name_signal = any(
                    verb
                    in name_lower

                    for verb in GENERATOR_VERBS
                )


                # Strict minimum: at least 4/5 schema fields plus either
                # config identity or generator-like function naming.
                candidate_like = (
                    len(
                        schema_hits
                    )
                    >= 4
                    and
                    (
                        config_hit
                        or
                        generator_name_signal
                    )
                )


                if not candidate_like:

                    continue


                forbidden = function_forbidden_hits(
                    fn
                )


                forbidden_any = any(
                    bool(
                        hits
                    )

                    for hits in forbidden.values()
                )


                # Consumption-only heuristics.
                mutation_consumer = (
                    ".add_edge"
                    in lower
                    and
                    "candidate_rank"
                    not in lower
                )


                historical_pickle_consumer = (
                    "read_pickle"
                    in lower
                )


                records.append({
                    "path":
                        safe_rel(
                            path
                        ),

                    "file_sha256":
                        sha256_file(
                            path
                        ),

                    "function":
                        fn.name,

                    "function_sha256":
                        sha256_text(
                            source
                        ),

                    "line_start":
                        fn.lineno,

                    "line_end":
                        fn.end_lineno,

                    "schema_hits":
                        schema_hits,

                    "config_hit":
                        config_hit,

                    "generator_name_signal":
                        generator_name_signal,

                    "forbidden_hits":
                        forbidden,

                    "forbidden_any":
                        forbidden_any,

                    "mutation_consumer":
                        mutation_consumer,

                    "historical_pickle_consumer":
                        historical_pickle_consumer,

                    "source":
                        source,
                })


    return records


# =============================================================================
# 6. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    POST_V2_PROVIDER,
    MORPH_AUTHORITY,
    TARGET_ARTIFACT,
):

    assert (
        required.exists()
    ), required


# Predecessor failed before artifact creation.
assert (
    not FAILED_DA_ARTIFACT.exists()
), (
    "Unexpected formal artifact from failed 14.3E-D-A."
)


# IMPORTANT:
# We hash the target artifact but DO NOT read/unpickle its contents.
assert (
    sha256_file(
        TARGET_ARTIFACT
    )
    == EXPECTED_TARGET_ARTIFACT_SHA
)


print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-A-R1 "
    "CONCRETE MORPH PROVIDER AUTHORITY RECOVERY "
    "— GATE + CLASSIFICATION REPAIR",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "TARGET ORACLE:",
    TARGET_ARTIFACT_REL,
    flush=True,
)

print(
    "TARGET ORACLE SHA:",
    EXPECTED_TARGET_ARTIFACT_SHA,
    flush=True,
)

print(
    "TARGET ORACLE CONTENT OPENED:",
    False,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Parent + source locks
# =============================================================================

parent = verify_parent(
    PARENT
)


portable_sha_before = sha256_file(
    PORTABLE
)


post_v2_sha_before = sha256_file(
    POST_V2_PROVIDER
)


authority_sha = sha256_file(
    MORPH_AUTHORITY
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    post_v2_sha_before
    == EXPECTED_POST_V2_PROVIDER_SHA
)


assert (
    authority_sha
    == EXPECTED_MORPH_AUTHORITY_FILE_SHA
)


authority = load_json(
    MORPH_AUTHORITY
)


assert (
    contains_exact_value(
        authority,
        EXPECTED_AUTHORITY_TYPE,
    )
)


assert (
    contains_exact_value(
        authority,
        EXPECTED_TARGET_ARTIFACT_SHA,
    )
)


print(
    "[PASS] AUTHORITY_AND_CURRENT_SOURCE_LOCKS",
    flush=True,
)

print(
    "  parent file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  post-V2 provider SHA:",
    post_v2_sha_before,
    flush=True,
)

print(
    "  Morph authority SHA:",
    authority_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Exact notebook target-I/O scan
# =============================================================================

notebook_records = []


NOTEBOOK_ROOT = ROOT / "notebooks"


for notebook_path in sorted(
    NOTEBOOK_ROOT.glob(
        "*.ipynb"
    )
):

    try:

        notebook_records.extend(
            analyze_notebook_target_io(
                notebook_path
            )
        )

    except Exception as exc:

        print(
            "[WARN] notebook scan failed:",
            safe_rel(
                notebook_path
            ),
            repr(
                exc
            ),
            flush=True,
        )


target_writers = [
    record

    for record in notebook_records

    if (
        record[
            "classification"
        ]
        == "TARGET_WRITER"
    )
]


target_readers = [
    record

    for record in notebook_records

    if (
        record[
            "classification"
        ]
        == "TARGET_READER"
    )
]


mention_only = [
    record

    for record in notebook_records

    if (
        record[
            "classification"
        ]
        == "MENTION_ONLY"
    )
]


print(
    "=" * 112,
    flush=True,
)

print(
    "EXACT HISTORICAL MORPH TARGET I/O",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "writers:",
    len(
        target_writers
    ),
    flush=True,
)

print(
    "readers:",
    len(
        target_readers
    ),
    flush=True,
)

print(
    "mention-only:",
    len(
        mention_only
    ),
    flush=True,
)

print(
    flush=True,
)


for record in notebook_records:

    print(
        record[
            "classification"
        ],
        record[
            "path"
        ],
        f"cell={record['cell_index']}",
        flush=True,
    )

    print(
        "  cell SHA:",
        record[
            "cell_sha256"
        ],
        flush=True,
    )

    if record.get(
        "write_sinks"
    ):

        print(
            "  write sinks:",
            record[
                "write_sinks"
            ],
            flush=True,
        )


    if record.get(
        "read_sinks"
    ):

        print(
            "  read sinks:",
            record[
                "read_sinks"
            ],
            flush=True,
        )


print(
    flush=True,
)


# =============================================================================
# 9. Print exact writer source for manual/provenance inspection
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "VERIFIED TARGET WRITER CELLS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for record in target_writers:

    print(
        "---",
        record[
            "path"
        ],
        "cell",
        record[
            "cell_index"
        ],
        "---",
        flush=True,
    )

    print(
        record[
            "source"
        ][
            :12000
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 10. Strict concrete-generator source scan
# =============================================================================

strict_candidates = (
    scan_strict_generator_candidates()
)


clean_strict_candidates = [
    record

    for record in strict_candidates

    if (
        not record[
            "forbidden_any"
        ]
        and
        not record[
            "mutation_consumer"
        ]
        and
        not record[
            "historical_pickle_consumer"
        ]
    )
]


print(
    "=" * 112,
    flush=True,
)

print(
    "STRICT CONCRETE GENERATOR SOURCE CANDIDATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "all strict candidates:",
    len(
        strict_candidates
    ),
    flush=True,
)

print(
    "clean strict candidates:",
    len(
        clean_strict_candidates
    ),
    flush=True,
)

print(
    flush=True,
)


for record in strict_candidates:

    print(
        record[
            "path"
        ],
        "::",
        record[
            "function"
        ],
        flush=True,
    )

    print(
        "  function SHA:",
        record[
            "function_sha256"
        ],
        flush=True,
    )

    print(
        "  schema hits:",
        record[
            "schema_hits"
        ],
        flush=True,
    )

    print(
        "  config hit:",
        record[
            "config_hit"
        ],
        flush=True,
    )

    print(
        "  forbidden:",
        record[
            "forbidden_any"
        ],
        flush=True,
    )

    print(
        "  mutation consumer:",
        record[
            "mutation_consumer"
        ],
        flush=True,
    )

    print(
        "  historical pickle consumer:",
        record[
            "historical_pickle_consumer"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 11. Historical writer semantics screen
#
# Writer cells are provenance evidence, but a writer cell is NOT automatically
# a deployable fresh provider.
# =============================================================================

writer_semantic_records = []


for record in target_writers:

    source = record[
        "source"
    ]


    lower = source.lower()


    schema_hits = sorted(
        signal

        for signal in REQUIRED_SCHEMA_SIGNALS

        if signal in lower
    )


    writer_semantic_records.append({
        "path":
            record[
                "path"
            ],

        "cell_index":
            record[
                "cell_index"
            ],

        "cell_sha256":
            record[
                "cell_sha256"
            ],

        "schema_hits":
            schema_hits,

        "config_id_present":
            (
                MORPH_CONFIG_ID.lower()
                in lower
            ),

        "contains_read_pickle":
            (
                "read_pickle"
                in lower
            ),

        "contains_ground_truth_token":
            (
                "ground_truth"
                in lower
                or
                "gt_graph"
                in lower
            ),

        "contains_evaluator_token":
            (
                "evaluate_graph"
                in lower
                or
                "evaluator"
                in lower
            ),

        "contains_candidate_generation_language":
            any(
                token
                in lower

                for token in (
                    "candidate_rank",
                    "morph_rank",
                    "candidate_prob",
                    "association",
                    "topk",
                    "rank",
                )
            ),
    })


print(
    "=" * 112,
    flush=True,
)

print(
    "WRITER SEMANTIC SCREEN",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for record in writer_semantic_records:

    print(
        json.dumps(
            record,
            indent=2,
        ),
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 12. Conservative authority classification
# =============================================================================

verified_writer_count = len(
    target_writers
)


clean_source_candidate_count = len(
    clean_strict_candidates
)


# Strong enough for semantic-contract freeze only if a concrete source function
# has actually been localized.
if (
    clean_source_candidate_count
    > 0
):

    recovery_status = (
        "CONCRETE_SOURCE_CANDIDATES_LOCALIZED_"
        "SEMANTIC_FREEZE_REQUIRED"
    )


elif (
    verified_writer_count
    > 0
):

    recovery_status = (
        "HISTORICAL_WRITER_CELLS_VERIFIED_"
        "CONCRETE_GENERATOR_SOURCE_UNRESOLVED"
    )


else:

    recovery_status = (
        "CONCRETE_PROVIDER_AUTHORITY_UNRESOLVED"
    )


implementation_authorized = (
    False
)


if (
    recovery_status
    ==
    "CONCRETE_SOURCE_CANDIDATES_LOCALIZED_"
    "SEMANTIC_FREEZE_REQUIRED"
):

    next_stage = (
        "NOTEBOOK14_14_3E_D_B_"
        "MORPH_CONCRETE_PROVIDER_SEMANTIC_CONTRACT_FREEZE"
    )


else:

    next_stage = (
        "NOTEBOOK14_14_3E_D_B_"
        "MORPH_HISTORICAL_WRITER_DATAFLOW_AND_"
        "CONCRETE_GENERATOR_TARGETED_RECOVERY"
    )


print(
    "=" * 112,
    flush=True,
)

print(
    "CORRECTED AUTHORITY RECOVERY CLASSIFICATION",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)

print(
    "verified target writers:",
    verified_writer_count,
    flush=True,
)

print(
    "clean concrete source candidates:",
    clean_source_candidate_count,
    flush=True,
)

print(
    "status:",
    recovery_status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "next:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Verify absolutely no source mutation occurred
# =============================================================================

portable_sha_after = sha256_file(
    PORTABLE
)


post_v2_sha_after = sha256_file(
    POST_V2_PROVIDER
)


no_source_mutation = (
    portable_sha_after
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
    and
    post_v2_sha_after
    == post_v2_sha_before
    == EXPECTED_POST_V2_PROVIDER_SHA
)


assert (
    no_source_mutation
)


# =============================================================================
# 14. Correct semantic gates
#
# Every gate expresses the PASS condition, not the raw state variable.
# =============================================================================

source_mutation = False

GT_opened = False

evaluator_executed = False

performance_metric_generated = False

historical_pickle_opened = False


semantic_gates = {
    "parent_14_3e_c_r3_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "post_v2_provider_locked":
        (
            post_v2_sha_before
            == EXPECTED_POST_V2_PROVIDER_SHA
        ),

    "historical_Morph_authority_locked":
        (
            authority_sha
            == EXPECTED_MORPH_AUTHORITY_FILE_SHA
        ),

    "historical_oracle_identity_locked_without_opening":
        (
            sha256_file(
                TARGET_ARTIFACT
            )
            == EXPECTED_TARGET_ARTIFACT_SHA
            and
            historical_pickle_opened
            is False
        ),

    "target_IO_dataflow_scan_complete":
        True,

    "strict_concrete_generator_scan_complete":
        True,

    "writer_candidate_mentions_not_automatically_treated_as_writer":
        True,

    "application_runtime_not_treated_as_concrete_generator":
        all(
            record[
                "path"
            ]
            != str(
                PORTABLE_REL
            )

            for record in clean_strict_candidates
        ),

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    # CORRECT POLARITY:
    "no_source_mutation":
        (
            source_mutation
            is False
            and
            no_source_mutation
        ),

    "GT_not_opened":
        (
            GT_opened
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),
}


print(
    "=" * 112,
    flush=True,
)

print(
    "14.3E-D-A-R1 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-A-R1 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 15. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_A_R1",

    "title": (
        "CONCRETE_MORPH_CANDIDATE_PROVIDER_AUTHORITY_RECOVERY_"
        "GATE_AND_CLASSIFICATION_REPAIR"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_SOURCE_AUTHORITY_RECOVERY_ONLY",

    "failed_predecessor": {
        "stage":
            "NOTEBOOK14_14_3E_D_A",

        "formal_artifact_written":
            False,

        "gate_polarity_bug": {
            "source_mutation":
                False,

            "GT_opened":
                False,

            "evaluator_executed":
                False,

            "performance_metric_generated":
                False,

            "problem": (
                "Raw negative states were used directly as gate booleans "
                "instead of asserting that those states were False."
            ),
        },

        "classification_bug": {
            "reported_status":
                "HISTORICAL_WRITER_AND_SOURCE_CANDIDATES_FOUND",

            "observed_authority_backed_function_candidates":
                0,

            "observed_historical_writer_backed_functions":
                0,

            "observed_strong_function_candidates":
                1,

            "sole_strong_function_kind": (
                "CURRENT_PORTABLE_MORPH_APPLICATION_RUNTIME"
            ),

            "repair": (
                "Current portable application runtime is excluded from "
                "concrete candidate-generator authority."
            ),
        },
    },

    "parent_14_3e_c_r3": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha_before,
        },

        "post_v2_provider": {
            "path":
                str(
                    POST_V2_PROVIDER_REL
                ),

            "file_sha256":
                post_v2_sha_before,
        },

        "Morph_authority": {
            "path":
                str(
                    MORPH_AUTHORITY_REL
                ),

            "file_sha256":
                authority_sha,
        },
    },

    "historical_Morph_oracle": {
        "path":
            str(
                TARGET_ARTIFACT_REL
            ),

        "file_sha256":
            EXPECTED_TARGET_ARTIFACT_SHA,

        "content_opened":
            False,

        "runtime_use":
            False,
    },

    "exact_notebook_target_IO": {
        "records":
            notebook_records,

        "writer_count":
            len(
                target_writers
            ),

        "reader_count":
            len(
                target_readers
            ),

        "mention_only_count":
            len(
                mention_only
            ),

        "verified_writers":
            target_writers,

        "verified_readers":
            target_readers,
    },

    "writer_semantic_screen":
        writer_semantic_records,

    "strict_generator_scan": {
        "required_schema_signals":
            sorted(
                REQUIRED_SCHEMA_SIGNALS
            ),

        "config_id":
            MORPH_CONFIG_ID,

        "all_candidates":
            strict_candidates,

        "clean_candidates":
            clean_strict_candidates,

        "clean_candidate_count":
            clean_source_candidate_count,

        "current_portable_application_runtime_excluded":
            True,
    },

    "resolution": {
        "status":
            recovery_status,

        "implementation_authorized":
            implementation_authorized,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "post_V2_orchestrator":
            "IMPLEMENTED",

        "Morph_provider_dependency_seam":
            "IMPLEMENTED",

        "concrete_Morph_provider_authority":
            recovery_status,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_YET",

        "fresh_end_to_end_post_V2_runtime":
            "NOT_YET_READY",

        "post_V2_structural_replay":
            "NOT_YET_READY",

        "historical_Morph_action_runtime_use":
            False,

        "historical_Morph_action_content_opened":
            False,

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No concrete Morph provider was implemented.",
        "No historical Morph action pickle was opened.",
        "A writer cell is provenance evidence, not automatically a deployable provider.",
        "The current portable Morph application runtime is not concrete candidate-generator authority.",
        "No source token scan alone proves GT-blindness.",
        "No post-V2 structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-A-R1 artifact exists but differs "
        "from the exact recovery state."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 16. Final
# =============================================================================

print(
    "=" * 112,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-A-R1 PASS",
    flush=True,
)

print(
    "=" * 112,
    flush=True,
)


print(
    "PREDECESSOR GATE BUG:",
    flush=True,
)

print(
    "  source_mutation=False ->",
    "PASS no_source_mutation",
    flush=True,
)

print(
    "  GT_opened=False ->",
    "PASS GT_not_opened",
    flush=True,
)

print(
    "  evaluator_executed=False ->",
    "PASS evaluator_not_executed",
    flush=True,
)

print(
    "  performance_metric_generated=False ->",
    "PASS performance_metric_not_generated",
    flush=True,
)

print(
    flush=True,
)


print(
    "EXACT TARGET I/O:",
    flush=True,
)

print(
    "  verified writers:",
    len(
        target_writers
    ),
    flush=True,
)

print(
    "  verified readers:",
    len(
        target_readers
    ),
    flush=True,
)

print(
    "  mention-only:",
    len(
        mention_only
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "CONCRETE GENERATOR AUTHORITY:",
    flush=True,
)

print(
    "  strict source candidates:",
    len(
        strict_candidates
    ),
    flush=True,
)

print(
    "  clean concrete candidates:",
    clean_source_candidate_count,
    flush=True,
)

print(
    "  current portable application runtime accepted as generator:",
    False,
    flush=True,
)

print(
    "  corrected status:",
    recovery_status,
    flush=True,
)

print(
    "  implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY / CLAIM BOUNDARY:",
    flush=True,
)

print(
    "  historical Morph pickle opened:",
    False,
    flush=True,
)

print(
    "  source mutation:",
    False,
    flush=True,
)

print(
    "  GT opened:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  performance generated:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-A-R2
# FALSE GENERATOR DISQUALIFICATION
# + HISTORICAL MORPH PRODUCER LINEAGE RECOVERY
#
# PARENT
# ------
# 14.3E-D-A-R1 PASS
#
# Observed state:
#
#   exact target writers = 0
#   exact target readers = 3
#
#   strict source candidates = 5
#   clean source candidates  = 1
#
# sole clean candidate:
#
#   scripts/run_stage12_single_rewire_policy_grid.py
#       :: build_v2_baseline
#
#   function SHA:
#       f805c1c3bd7f38de14f36ac27d5869bb3d24f4a392b9fd876e47ebf9a1114cc0
#
# BUT:
#
#   config hit = False
#
# Therefore it is NOT yet concrete Morph generator authority.
#
#
# PURPOSE
# -------
# A. Determine whether build_v2_baseline:
#
#       GENERATES fresh Morph candidates
#
#    or merely:
#
#       CONSUMES/APPLIES already-existing Morph action tables.
#
# B. Inspect frozen Notebook13 authority:
#
#       13.3D2-R2
#       morph_repository_writer_relocation_and_inner_block_resolution
#
#    to recover the exact historical producer lineage / writer relocation.
#
# C. Inspect referenced notebook cells without opening historical .pkl.
#
# D. Freeze a conservative next-step classification.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical Morph pickle open
#   - GT open
#   - evaluator execution
#   - performance
#   - implementation authorization
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_a_r1_"
    "concrete_morph_candidate_provider_"
    "authority_recovery_gate_and_classification_repair_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


POST_V2_REL = Path(
    "src/biohub_cell_tracking/stage12_post_v2_graph_provider.py"
)

POST_V2 = ROOT / POST_V2_REL


CANDIDATE_REL = Path(
    "scripts/run_stage12_single_rewire_policy_grid.py"
)

CANDIDATE = ROOT / CANDIDATE_REL


HIST_AUTH_REL = Path(
    "reports/"
    "notebook13_13_3d2r2_"
    "morph_repository_writer_relocation_and_"
    "inner_block_resolution_v1.json"
)

HIST_AUTH = ROOT / HIST_AUTH_REL


MORPH_AUTHORITY_REL = Path(
    "reports/"
    "notebook13_13_3d2r3_"
    "morph_authority_correction_and_"
    "semantic_reconstruction_freeze_v1.json"
)

MORPH_AUTHORITY = ROOT / MORPH_AUTHORITY_REL


TARGET_REL = Path(
    "reports/"
    "stage11_division_probe/"
    "stage11_fold0_morphdiv_v2_actions_locked.pkl"
)

TARGET = ROOT / TARGET_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_a_r2_"
    "false_generator_disqualification_and_"
    "historical_producer_lineage_recovery_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "e017cce74b3cb67f1d394909839075e89c20a4e11286f4de77bf41c6946c2d52"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "1c01b7281389f76dd02702c96a11dba647410eb975e8d9f10348774c192364ad"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_POST_V2_SHA = (
    "3caec19d4446bd44415c3039d8e16e2f7b6ac7cdd03d4ba27941bb583055470e"
)


EXPECTED_CANDIDATE_FUNCTION_SHA = (
    "f805c1c3bd7f38de14f36ac27d5869bb3d24f4a392b9fd876e47ebf9a1114cc0"
)


EXPECTED_HIST_AUTH_FILE_SHA = (
    "f3d4e400a089a9ee5c90b55f3e7d044bc853ade2d857593a8017a8f244cae610"
)


EXPECTED_MORPH_AUTHORITY_FILE_SHA = (
    "ac0cb4959b8a9cb54784d0fd37d4da89a87b6e110298f67af00e712deebf8afb"
)


EXPECTED_TARGET_SHA = (
    "9c08f88c9824912af72a5538e6f23cd6bc7e14ba4ffc30e26ad9ad6ca37dd9eb"
)


TARGET_BASENAME = (
    TARGET.name
)


MORPH_CONFIG_ID = (
    "K3_R3_P100"
)


MORPH_SCHEMA = {
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
}


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def safe_rel(
    path: Path,
) -> str:

    try:

        return str(
            path.resolve().relative_to(
                ROOT
            )
        )

    except Exception:

        return str(
            path
        )


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )


    assert (
        source is not None
    )


    return source


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in tree.body

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def source_sha(
    text: str,
    node: ast.AST,
) -> str:

    return sha256_text(
        node_source(
            text,
            node,
        )
    )


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in obj.items():

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:

    return any(
        leaf[
            "value"
        ]
        == target

        for leaf in flatten_json(
            obj
        )
    )


def function_calls(
    fn: ast.AST,
) -> list[str]:

    return sorted({
        dotted_name(
            node.func
        )
        or "<dynamic-call>"

        for node in ast.walk(
            fn
        )

        if isinstance(
            node,
            ast.Call,
        )
    })


def function_names(
    fn: ast.AST,
) -> set[str]:

    result = set()


    for node in ast.walk(
        fn
    ):

        if isinstance(
            node,
            ast.Name,
        ):

            result.add(
                node.id
            )


    return result


def assignment_columns(
    fn: ast.AST,
) -> set[str]:

    """
    Extract literal DataFrame-like column names that appear on assignment
    targets such as df["candidate_rank"] = ...
    """

    result = set()


    for node in ast.walk(
        fn
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
            ),
        ):

            continue


        targets = []


        if isinstance(
            node,
            ast.Assign,
        ):

            targets = list(
                node.targets
            )


        else:

            targets = [
                node.target
            ]


        for target in targets:

            for child in ast.walk(
                target
            ):

                if not isinstance(
                    child,
                    ast.Subscript,
                ):

                    continue


                try:

                    key = ast.literal_eval(
                        child.slice
                    )

                except Exception:

                    continue


                if isinstance(
                    key,
                    str,
                ):

                    result.add(
                        key
                    )


    return result


def string_literals(
    node: ast.AST,
) -> list[str]:

    return [
        child.value

        for child in ast.walk(
            node
        )

        if (
            isinstance(
                child,
                ast.Constant,
            )
            and
            isinstance(
                child.value,
                str,
            )
        )
    ]


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "concrete_Morph_provider_implementation"
        ]
        == "NOT_AUTHORIZED_YET"
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "historical_Morph_action_content_opened"
        ]
        is False
    )


    return obj


# =============================================================================
# 4. Candidate-module dependency discovery
# =============================================================================

def module_read_pickle_assignments(
    text: str,
    tree: ast.Module,
) -> list[
    dict[str, Any]
]:

    records = []


    for stmt in tree.body:

        value = None

        names = []


        if isinstance(
            stmt,
            ast.Assign,
        ):

            value = stmt.value


            for target in stmt.targets:

                if isinstance(
                    target,
                    ast.Name,
                ):

                    names.append(
                        target.id
                    )


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            value = stmt.value


            if isinstance(
                stmt.target,
                ast.Name,
            ):

                names.append(
                    stmt.target.id
                )


        if not isinstance(
            value,
            ast.Call,
        ):

            continue


        call_name = dotted_name(
            value.func
        )


        if (
            call_name is None
            or
            not call_name.endswith(
                "read_pickle"
            )
        ):

            continue


        path_expr = None


        if value.args:

            path_expr = ast.unparse(
                value.args[
                    0
                ]
            )


        records.append({
            "line":
                stmt.lineno,

            "assigned_names":
                names,

            "call":
                ast.unparse(
                    value
                ),

            "path_expression":
                path_expr,
        })


    return records


def module_path_assignments(
    text: str,
    tree: ast.Module,
) -> list[
    dict[str, Any]
]:

    records = []


    for stmt in tree.body:

        if not isinstance(
            stmt,
            (
                ast.Assign,
                ast.AnnAssign,
            ),
        ):

            continue


        value = (
            stmt.value
            if hasattr(
                stmt,
                "value"
            )
            else None
        )


        if value is None:

            continue


        source = ast.unparse(
            value
        )


        if not any(
            token
            in source.lower()

            for token in (
                "morph",
                ".pkl",
                "action",
                "stage11",
            )
        ):

            continue


        names = []


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in stmt.targets:

                if isinstance(
                    target,
                    ast.Name,
                ):

                    names.append(
                        target.id
                    )


        elif isinstance(
            stmt.target,
            ast.Name,
        ):

            names.append(
                stmt.target.id
            )


        records.append({
            "line":
                stmt.lineno,

            "names":
                names,

            "value":
                source,
        })


    return records


def local_helper_closure(
    text: str,
    functions: dict[str, ast.AST],
    root_name: str,
) -> list[
    dict[str, Any]
]:

    """
    Follow calls from root into same-module top-level functions.
    """

    seen = set()

    queue = [
        root_name
    ]

    result = []


    while queue:

        name = queue.pop(
            0
        )


        if name in seen:

            continue


        seen.add(
            name
        )


        if name not in functions:

            continue


        fn = functions[
            name
        ]


        calls = function_calls(
            fn
        )


        local_calls = sorted({
            call.split(
                "."
            )[
                -1
            ]

            for call in calls

            if (
                call.split(
                    "."
                )[
                    -1
                ]
                in functions
            )
        })


        result.append({
            "function":
                name,

            "function_sha256":
                source_sha(
                    text,
                    fn,
                ),

            "calls":
                calls,

            "local_calls":
                local_calls,

            "referenced_names":
                sorted(
                    function_names(
                        fn
                    )
                ),

            "assigned_columns":
                sorted(
                    assignment_columns(
                        fn
                    )
                ),
        })


        queue.extend(
            local_calls
        )


    return result


# =============================================================================
# 5. Generic historical-authority report inspection
# =============================================================================

def relevant_authority_leaves(
    obj: Any,
) -> list[
    dict[str, Any]
]:

    result = []


    tokens = (
        "writer",
        "producer",
        "relocation",
        "inner",
        "block",
        "cell",
        "notebook",
        "morph",
        "action",
        "source",
    )


    for leaf in flatten_json(
        obj
    ):

        path_lower = (
            leaf[
                "path"
            ].lower()
        )


        value_text = str(
            leaf[
                "value"
            ]
        )


        value_lower = (
            value_text.lower()
        )


        if any(
            token
            in path_lower
            or
            token
            in value_lower

            for token in tokens
        ):

            result.append(
                leaf
            )


    return result


def recursively_extract_notebook_refs(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        notebook_values = [
            (
                key,
                value,
            )

            for key, value in obj.items()

            if (
                isinstance(
                    value,
                    str,
                )
                and
                ".ipynb"
                in value
            )
        ]


        cell_values = [
            (
                key,
                value,
            )

            for key, value in obj.items()

            if (
                "cell"
                in key.lower()
                and
                isinstance(
                    value,
                    int,
                )
            )
        ]


        sha_values = [
            (
                key,
                value,
            )

            for key, value in obj.items()

            if (
                isinstance(
                    value,
                    str,
                )
                and
                re.fullmatch(
                    r"[0-9a-fA-F]{64}",
                    value,
                )
            )
        ]


        if notebook_values:

            for path_key, notebook_value in notebook_values:

                result.append({
                    "json_path":
                        prefix,

                    "path_field":
                        path_key,

                    "notebook":
                        notebook_value,

                    "cell_fields":
                        [
                            {
                                "field":
                                    key,

                                "value":
                                    value,
                            }

                            for key, value in cell_values
                        ],

                    "sha_fields":
                        [
                            {
                                "field":
                                    key,

                                "value":
                                    value,
                            }

                            for key, value in sha_values
                        ],
                })


        for key, value in obj.items():

            result.extend(
                recursively_extract_notebook_refs(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                recursively_extract_notebook_refs(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    return result


def inspect_notebook_reference(
    record: dict[str, Any],
) -> list[
    dict[str, Any]
]:

    raw = Path(
        record[
            "notebook"
        ]
    )


    path = (
        raw
        if raw.is_absolute()
        else
        ROOT / raw
    )


    if not path.exists():

        return []


    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    cells = notebook.get(
        "cells",
        []
    )


    indices = sorted({
        item[
            "value"
        ]

        for item in record[
            "cell_fields"
        ]
    })


    output = []


    for index in indices:

        # Historical reports may use either 0-based or human 1-based cell IDs.
        candidate_indices = sorted({
            index,
            index - 1,
        })


        for candidate_index in candidate_indices:

            if (
                candidate_index < 0
                or
                candidate_index
                >= len(
                    cells
                )
            ):

                continue


            cell = cells[
                candidate_index
            ]


            source = "".join(
                cell.get(
                    "source",
                    []
                )
            )


            output.append({
                "notebook":
                    safe_rel(
                        path
                    ),

                "notebook_sha256":
                    sha256_file(
                        path
                    ),

                "reported_cell":
                    index,

                "inspected_zero_based_index":
                    candidate_index,

                "cell_type":
                    cell.get(
                        "cell_type"
                    ),

                "cell_sha256":
                    sha256_text(
                        source
                    ),

                "contains_target_basename":
                    (
                        TARGET_BASENAME
                        in source
                    ),

                "contains_to_pickle":
                    (
                        ".to_pickle"
                        in source
                        or
                        "pd.to_pickle"
                        in source
                    ),

                "contains_read_pickle":
                    (
                        "read_pickle"
                        in source
                    ),

                "contains_config_id":
                    (
                        MORPH_CONFIG_ID
                        in source
                    ),

                "schema_hits":
                    sorted(
                        column

                        for column in MORPH_SCHEMA

                        if column in source
                    ),

                "source":
                    source,
            })


    return output


# =============================================================================
# 6. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    POST_V2,
    CANDIDATE,
    HIST_AUTH,
    MORPH_AUTHORITY,
    TARGET,
):

    assert (
        required.exists()
    ), required


# Target is hash-locked only. It is never opened/deserialized.
assert (
    sha256_file(
        TARGET
    )
    == EXPECTED_TARGET_SHA
)


print(
    "=" * 116,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-A-R2 "
    "FALSE GENERATOR DISQUALIFICATION "
    "+ HISTORICAL PRODUCER LINEAGE RECOVERY",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "CANDIDATE:",
    CANDIDATE_REL,
    flush=True,
)

print(
    "HISTORICAL WRITER AUTHORITY:",
    HIST_AUTH_REL,
    flush=True,
)

print(
    "TARGET ORACLE CONTENT OPENED:",
    False,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Parent / source locks
# =============================================================================

parent = verify_parent(
    PARENT
)


portable_sha_before = sha256_file(
    PORTABLE
)


post_v2_sha_before = sha256_file(
    POST_V2
)


hist_auth_sha = sha256_file(
    HIST_AUTH
)


morph_authority_sha = sha256_file(
    MORPH_AUTHORITY
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    post_v2_sha_before
    == EXPECTED_POST_V2_SHA
)


assert (
    hist_auth_sha
    == EXPECTED_HIST_AUTH_FILE_SHA
)


assert (
    morph_authority_sha
    == EXPECTED_MORPH_AUTHORITY_FILE_SHA
)


print(
    "[PASS] PARENT_AND_AUTHORITY_LOCKS",
    flush=True,
)

print(
    "  parent file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  parent content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  historical D2-R2 authority SHA:",
    hist_auth_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Exact inspection of sole D-A-R1 clean candidate
# =============================================================================

candidate_text = CANDIDATE.read_text(
    encoding="utf-8"
)


candidate_tree = ast.parse(
    candidate_text,
    filename=str(
        CANDIDATE
    ),
)


candidate_functions = top_level_functions(
    candidate_tree
)


assert (
    "build_v2_baseline"
    in candidate_functions
)


candidate_fn = candidate_functions[
    "build_v2_baseline"
]


candidate_fn_sha = source_sha(
    candidate_text,
    candidate_fn,
)


assert (
    candidate_fn_sha
    == EXPECTED_CANDIDATE_FUNCTION_SHA
)


candidate_source = node_source(
    candidate_text,
    candidate_fn,
)


candidate_calls = function_calls(
    candidate_fn
)


candidate_names = function_names(
    candidate_fn
)


candidate_assigned_columns = assignment_columns(
    candidate_fn
)


candidate_strings = string_literals(
    candidate_fn
)


candidate_schema_mentions = sorted(
    MORPH_SCHEMA
    &
    set(
        candidate_source.split()
    )
)


# More reliable substring form:
candidate_schema_hits = sorted(
    column

    for column in MORPH_SCHEMA

    if column
    in candidate_source
)


candidate_config_hit = (
    MORPH_CONFIG_ID
    in candidate_source
)


module_pickle_loads = module_read_pickle_assignments(
    candidate_text,
    candidate_tree,
)


module_path_records = module_path_assignments(
    candidate_text,
    candidate_tree,
)


helper_closure = local_helper_closure(
    candidate_text,
    candidate_functions,
    "build_v2_baseline",
)


print(
    "=" * 116,
    flush=True,
)

print(
    "SOLE CLEAN CANDIDATE EXACT INSPECTION",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)

print(
    "function SHA:",
    candidate_fn_sha,
    flush=True,
)

print(
    "signature:",
    ast.unparse(
        candidate_fn.args
    ),
    flush=True,
)

print(
    "config K3_R3_P100 present:",
    candidate_config_hit,
    flush=True,
)

print(
    "schema hits:",
    candidate_schema_hits,
    flush=True,
)

print(
    "assigned candidate columns:",
    sorted(
        candidate_assigned_columns
        &
        MORPH_SCHEMA
    ),
    flush=True,
)

print(
    "calls:",
    candidate_calls,
    flush=True,
)

print(
    "module read_pickle assignments:",
    json.dumps(
        module_pickle_loads,
        indent=2,
    ),
    flush=True,
)

print(
    "module Morph/action path assignments:",
    json.dumps(
        module_path_records,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "--- build_v2_baseline source ---",
    flush=True,
)

print(
    candidate_source,
    flush=True,
)

print(
    "--- end source ---",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Consumer/generator classification
# =============================================================================

candidate_generation_assignment_hits = sorted(
    candidate_assigned_columns
    &
    {
        "morph_rank",
        "candidate_rank",
        "candidate_prob",
    }
)


candidate_generation_ops = sorted({
    call

    for call in candidate_calls

    if any(
        token in call.lower()

        for token in (
            "rank",
            "nlargest",
            "nsmallest",
            "sort_values",
            "groupby",
            "merge",
            "concat",
            "dataframe",
            "association",
            "prob",
        )
    )
})


candidate_existing_action_signals = sorted({
    name

    for name in candidate_names

    if any(
        token
        in name.lower()

        for token in (
            "morph",
            "action",
            "locked",
            "stage11",
        )
    )
})


helper_existing_action_signals = []


for record in helper_closure:

    matching_names = [
        name

        for name in record[
            "referenced_names"
        ]

        if any(
            token
            in name.lower()

            for token in (
                "morph",
                "action",
                "locked",
                "stage11",
            )
        )
    ]


    if matching_names:

        helper_existing_action_signals.append({
            "function":
                record[
                    "function"
                ],

            "signals":
                matching_names,
        })


# Conservative rule:
#
# A concrete Morph generator should at minimum visibly construct/rank candidate
# rows, OR call a source-local helper which does.
#
# Merely mentioning schema fields while building a baseline graph is not enough.
candidate_has_generation_evidence = (
    bool(
        candidate_generation_assignment_hits
    )
    or
    bool(
        candidate_generation_ops
    )
)


candidate_uses_existing_action_context = (
    bool(
        candidate_existing_action_signals
    )
    or
    bool(
        helper_existing_action_signals
    )
    or
    bool(
        module_pickle_loads
    )
)


if (
    candidate_has_generation_evidence
    and
    candidate_config_hit
    and
    not candidate_uses_existing_action_context
):

    candidate_classification = (
        "POSSIBLE_CONCRETE_GENERATOR"
    )


elif (
    candidate_uses_existing_action_context
    or
    not candidate_config_hit
):

    candidate_classification = (
        "NOT_CONCRETE_MORPH_GENERATOR_AUTHORITY"
    )


else:

    candidate_classification = (
        "UNRESOLVED_NOT_AUTHORIZED"
    )


print(
    "=" * 116,
    flush=True,
)

print(
    "SOLE CANDIDATE CLASSIFICATION",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)

print(
    "generation assignment hits:",
    candidate_generation_assignment_hits,
    flush=True,
)

print(
    "generation operation hits:",
    candidate_generation_ops,
    flush=True,
)

print(
    "existing-action signals:",
    candidate_existing_action_signals,
    flush=True,
)

print(
    "helper existing-action signals:",
    helper_existing_action_signals,
    flush=True,
)

print(
    "generation evidence:",
    candidate_has_generation_evidence,
    flush=True,
)

print(
    "existing-action context:",
    candidate_uses_existing_action_context,
    flush=True,
)

print(
    "classification:",
    candidate_classification,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Inspect historical D2-R2 writer-relocation authority
# =============================================================================

hist_auth = load_json(
    HIST_AUTH
)


hist_auth_leaves = relevant_authority_leaves(
    hist_auth
)


notebook_refs = recursively_extract_notebook_refs(
    hist_auth
)


print(
    "=" * 116,
    flush=True,
)

print(
    "HISTORICAL 13.3D2-R2 WRITER-RELOCATION AUTHORITY",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)

print(
    "relevant leaf count:",
    len(
        hist_auth_leaves
    ),
    flush=True,
)

print(
    "notebook reference records:",
    len(
        notebook_refs
    ),
    flush=True,
)

print(
    flush=True,
)


for leaf in hist_auth_leaves[
    :220
]:

    print(
        leaf[
            "path"
        ],
        "=",
        repr(
            leaf[
                "value"
            ]
        ),
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 11. Inspect notebook cells explicitly referenced by D2-R2 authority
# =============================================================================

referenced_cell_records = []


for record in notebook_refs:

    referenced_cell_records.extend(
        inspect_notebook_reference(
            record
        )
    )


# Deduplicate.
unique_cell_records = []

seen_cell_keys = set()


for record in referenced_cell_records:

    key = (
        record[
            "notebook"
        ],
        record[
            "inspected_zero_based_index"
        ],
        record[
            "cell_sha256"
        ],
    )


    if key in seen_cell_keys:

        continue


    seen_cell_keys.add(
        key
    )


    unique_cell_records.append(
        record
    )


print(
    "=" * 116,
    flush=True,
)

print(
    "D2-R2 REFERENCED NOTEBOOK CELLS",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)

print(
    "unique referenced cells:",
    len(
        unique_cell_records
    ),
    flush=True,
)

print(
    flush=True,
)


for record in unique_cell_records:

    print(
        record[
            "notebook"
        ],
        "reported=",
        record[
            "reported_cell"
        ],
        "inspected zero-based=",
        record[
            "inspected_zero_based_index"
        ],
        flush=True,
    )

    print(
        "  cell SHA:",
        record[
            "cell_sha256"
        ],
        flush=True,
    )

    print(
        "  target basename:",
        record[
            "contains_target_basename"
        ],
        flush=True,
    )

    print(
        "  to_pickle:",
        record[
            "contains_to_pickle"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        record[
            "contains_read_pickle"
        ],
        flush=True,
    )

    print(
        "  K3_R3_P100:",
        record[
            "contains_config_id"
        ],
        flush=True,
    )

    print(
        "  schema hits:",
        record[
            "schema_hits"
        ],
        flush=True,
    )

    print(
        "  --- source ---",
        flush=True,
    )

    print(
        record[
            "source"
        ][
            :16000
        ],
        flush=True,
    )

    print(
        "  --- end source ---",
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 12. Extract authority semantics from D2-R2 report
# =============================================================================

writer_positive_leaves = [
    leaf

    for leaf in hist_auth_leaves

    if any(
        token
        in leaf[
            "path"
        ].lower()

        for token in (
            "writer",
            "producer",
            "inner_block",
            "relocation",
        )
    )
]


source_sha_leaves = [
    leaf

    for leaf in hist_auth_leaves

    if (
        isinstance(
            leaf[
                "value"
            ],
            str,
        )
        and
        re.fullmatch(
            r"[0-9a-fA-F]{64}",
            leaf[
                "value"
            ],
        )
        and
        any(
            token
            in leaf[
                "path"
            ].lower()

            for token in (
                "source",
                "writer",
                "producer",
                "cell",
                "block",
            )
        )
    )
]


print(
    "=" * 116,
    flush=True,
)

print(
    "D2-R2 PRODUCER/WIRITER AUTHORITY SUMMARY",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)

print(
    "writer/producer leaves:",
    len(
        writer_positive_leaves
    ),
    flush=True,
)

print(
    "source/cell/block SHA leaves:",
    len(
        source_sha_leaves
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Conservative resolution
# =============================================================================

false_candidate_disqualified = (
    candidate_classification
    == "NOT_CONCRETE_MORPH_GENERATOR_AUTHORITY"
)


authority_has_writer_lineage = (
    len(
        writer_positive_leaves
    )
    > 0
)


authority_has_referenced_cells = (
    len(
        unique_cell_records
    )
    > 0
)


# We still do NOT authorize implementation here.
implementation_authorized = False


if (
    false_candidate_disqualified
    and
    authority_has_writer_lineage
    and
    authority_has_referenced_cells
):

    recovery_status = (
        "FALSE_SOURCE_CANDIDATE_DISQUALIFIED_"
        "HISTORICAL_PRODUCER_LINEAGE_RECOVERED_"
        "SEMANTICS_EXTRACTION_REQUIRED"
    )


elif (
    false_candidate_disqualified
    and
    authority_has_writer_lineage
):

    recovery_status = (
        "FALSE_SOURCE_CANDIDATE_DISQUALIFIED_"
        "HISTORICAL_PRODUCER_LINEAGE_PARTIAL"
    )


elif false_candidate_disqualified:

    recovery_status = (
        "FALSE_SOURCE_CANDIDATE_DISQUALIFIED_"
        "CONCRETE_GENERATOR_AUTHORITY_STILL_UNRESOLVED"
    )


else:

    recovery_status = (
        "SOURCE_CANDIDATE_REQUIRES_FURTHER_DISAMBIGUATION"
    )


if (
    recovery_status
    ==
    "FALSE_SOURCE_CANDIDATE_DISQUALIFIED_"
    "HISTORICAL_PRODUCER_LINEAGE_RECOVERED_"
    "SEMANTICS_EXTRACTION_REQUIRED"
):

    next_stage = (
        "NOTEBOOK14_14_3E_D_B_"
        "MORPH_HISTORICAL_PRODUCER_SEMANTIC_EXTRACTION"
    )


else:

    next_stage = (
        "NOTEBOOK14_14_3E_D_B_"
        "MORPH_CONCRETE_GENERATOR_TARGETED_AUTHORITY_RECOVERY"
    )


print(
    "=" * 116,
    flush=True,
)

print(
    "14.3E-D-A-R2 RESOLUTION",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)

print(
    "D-A-R1 sole clean candidate:",
    candidate_classification,
    flush=True,
)

print(
    "historical writer-lineage evidence:",
    authority_has_writer_lineage,
    flush=True,
)

print(
    "historical referenced cells:",
    authority_has_referenced_cells,
    flush=True,
)

print(
    "recovery status:",
    recovery_status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. No mutation / no oracle open checks
# =============================================================================

source_mutation = False

historical_pickle_opened = False

GT_opened = False

evaluator_executed = False

performance_metric_generated = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    sha256_file(
        POST_V2
    )
    == post_v2_sha_before
    == EXPECTED_POST_V2_SHA
)


# =============================================================================
# 15. Formal gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_a_r1_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "post_v2_provider_locked":
        (
            post_v2_sha_before
            == EXPECTED_POST_V2_SHA
        ),

    "historical_D2_R2_authority_locked":
        (
            hist_auth_sha
            == EXPECTED_HIST_AUTH_FILE_SHA
        ),

    "historical_Morph_authority_locked":
        (
            morph_authority_sha
            == EXPECTED_MORPH_AUTHORITY_FILE_SHA
        ),

    "historical_oracle_identity_locked_without_opening":
        (
            sha256_file(
                TARGET
            )
            == EXPECTED_TARGET_SHA
            and
            historical_pickle_opened
            is False
        ),

    "sole_clean_candidate_exact_function_locked":
        (
            candidate_fn_sha
            == EXPECTED_CANDIDATE_FUNCTION_SHA
        ),

    "sole_clean_candidate_exactly_inspected":
        True,

    "historical_writer_authority_report_inspected":
        True,

    "historical_authority_notebook_refs_inspected":
        True,

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "GT_not_opened":
        (
            GT_opened
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),
}


print(
    "=" * 116,
    flush=True,
)

print(
    "14.3E-D-A-R2 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-A-R2 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 16. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_A_R2",

    "title": (
        "FALSE_GENERATOR_DISQUALIFICATION_AND_"
        "HISTORICAL_PRODUCER_LINEAGE_RECOVERY"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_AUTHORITY_DISAMBIGUATION_ONLY",

    "parent_14_3e_d_a_r1": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha_before,
        },

        "post_v2_provider": {
            "path":
                str(
                    POST_V2_REL
                ),

            "file_sha256":
                post_v2_sha_before,
        },

        "historical_D2_R2_authority": {
            "path":
                str(
                    HIST_AUTH_REL
                ),

            "file_sha256":
                hist_auth_sha,
        },

        "Morph_authority": {
            "path":
                str(
                    MORPH_AUTHORITY_REL
                ),

            "file_sha256":
                morph_authority_sha,
        },
    },

    "historical_oracle": {
        "path":
            str(
                TARGET_REL
            ),

        "file_sha256":
            EXPECTED_TARGET_SHA,

        "content_opened":
            False,

        "runtime_use":
            False,
    },

    "D_A_R1_sole_clean_candidate": {
        "path":
            str(
                CANDIDATE_REL
            ),

        "function":
            "build_v2_baseline",

        "function_sha256":
            candidate_fn_sha,

        "config_K3_R3_P100_present":
            candidate_config_hit,

        "schema_hits":
            candidate_schema_hits,

        "candidate_generation_assignment_hits":
            candidate_generation_assignment_hits,

        "candidate_generation_operation_hits":
            candidate_generation_ops,

        "existing_action_signals":
            candidate_existing_action_signals,

        "helper_existing_action_signals":
            helper_existing_action_signals,

        "module_pickle_loads":
            module_pickle_loads,

        "module_path_records":
            module_path_records,

        "helper_closure":
            helper_closure,

        "classification":
            candidate_classification,
    },

    "historical_writer_relocation_authority": {
        "path":
            str(
                HIST_AUTH_REL
            ),

        "file_sha256":
            hist_auth_sha,

        "relevant_leaves":
            hist_auth_leaves,

        "writer_positive_leaves":
            writer_positive_leaves,

        "source_sha_leaves":
            source_sha_leaves,

        "notebook_refs":
            notebook_refs,

        "referenced_cells":
            unique_cell_records,
    },

    "resolution": {
        "status":
            recovery_status,

        "false_candidate_disqualified":
            false_candidate_disqualified,

        "historical_writer_lineage_present":
            authority_has_writer_lineage,

        "historical_referenced_cells_present":
            authority_has_referenced_cells,

        "implementation_authorized":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "post_V2_orchestrator":
            "IMPLEMENTED",

        "Morph_provider_dependency_seam":
            "IMPLEMENTED",

        "D_A_R1_clean_candidate":
            candidate_classification,

        "concrete_Morph_provider_authority":
            recovery_status,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_YET",

        "historical_Morph_action_content_opened":
            False,

        "historical_Morph_action_runtime_use":
            False,

        "fresh_end_to_end_post_V2_runtime":
            "NOT_YET_READY",

        "post_V2_structural_replay":
            "NOT_YET_READY",

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "No concrete Morph candidate provider was implemented.",
        "The D-A-R1 clean candidate is not promoted merely because a heuristic scan selected it.",
        "No historical Morph pickle was opened.",
        "Historical producer lineage is not itself structural fidelity.",
        "No post-V2 structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-A-R2 artifact exists but differs "
        "from the exact authority resolution."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 17. Final
# =============================================================================

print(
    "=" * 116,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-A-R2 PASS",
    flush=True,
)

print(
    "=" * 116,
    flush=True,
)


print(
    "D-A-R1 CLEAN CANDIDATE:",
    flush=True,
)

print(
    "  function:",
    "build_v2_baseline",
    flush=True,
)

print(
    "  SHA:",
    candidate_fn_sha,
    flush=True,
)

print(
    "  K3_R3_P100:",
    candidate_config_hit,
    flush=True,
)

print(
    "  generation evidence:",
    candidate_has_generation_evidence,
    flush=True,
)

print(
    "  existing-action context:",
    candidate_uses_existing_action_context,
    flush=True,
)

print(
    "  classification:",
    candidate_classification,
    flush=True,
)

print(
    flush=True,
)


print(
    "HISTORICAL PRODUCER LINEAGE:",
    flush=True,
)

print(
    "  writer/producer authority leaves:",
    len(
        writer_positive_leaves
    ),
    flush=True,
)

print(
    "  referenced notebook cells:",
    len(
        unique_cell_records
    ),
    flush=True,
)

print(
    "  historical Morph pickle opened:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "RESOLUTION:",
    flush=True,
)

print(
    "  status:",
    recovery_status,
    flush=True,
)

print(
    "  implementation authorized:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-B
# MORPH SEMANTIC RECONSTRUCTION CONTRACT EXTRACTION
# + UPSTREAM GENERATOR RECOVERY
#
# PARENT
# ------
# 14.3E-D-A-R2 PASS
#
# IMPORTANT AUTHORITY CORRECTION
# ------------------------------
# 13.3D2-R2 proved:
#
#   exact historical writer source
#       = ABSENT IN CURRENT CHECKOUT
#
# Therefore this stage DOES NOT attempt another historical-writer archaeology
# pass for:
#
#   stage11_fold0_morphdiv_v2_actions_locked.pkl
#
# Instead it uses the already-frozen 13.3D2-R3 authority:
#
#   SEMANTIC_SOURCE_LOCK_PLUS_HISTORICAL_GT_BLIND_STRUCTURAL_ORACLE
#
# 13.3D2-R3 already froze:
#
#   - Morph semantic anchor
#   - GT-blind application core
#   - topology invariants
#   - structural oracle identity
#   - no runtime historical-action lookup
#
#
# THIS STAGE SEPARATES THREE LAYERS
# ---------------------------------
#
# A. SCIENTIFIC SELECTION POLICY
#
#   FINAL Frozen-V9 observable parents
#   W4-capable
#   current source outdegree = 1
#   temporal morphology Top 3
#   association rank <= 3
#   association probability >= 0.10
#   additive second edge
#
# B. ACTION -> GRAPH APPLICATION
#
#   source pre-outdegree = 1
#   target pre-indegree = 0
#   edge must not already exist
#   source post-outdegree = 2
#   target post-indegree = 1
#
# C. FRESH CANDIDATE GENERATION
#
#   This is the only remaining concrete MorphCandidateProvider question.
#
#
# ADDITIONAL TARGET
# -----------------
# Recover source provenance for the upstream historical table:
#
#   stage11_deployable_final_v9_policy_grid_actions.pkl
#
# This table was observed in D-A-R2 as the source consumed by later
# build_v2_baseline and filtered to config_id == K3_R3_P100.
#
# Finding its SOURCE WRITER is allowed.
# Loading its historical semantic contents at runtime is NOT allowed.
#
#
# NO:
#   - src mutation
#   - tests mutation
#   - historical Morph pickle loading
#   - upstream policy-grid pickle loading
#   - GT loading
#   - evaluator execution
#   - performance computation
#   - parameter search
#   - structural fidelity claim
#   - implementation authorization unless generation source is actually closed
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_a_r2_"
    "false_generator_disqualification_and_"
    "historical_producer_lineage_recovery_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


D2R3_REL = Path(
    "reports/"
    "notebook13_13_3d2r3_"
    "morph_authority_correction_and_"
    "semantic_reconstruction_freeze_v1.json"
)

D2R3 = ROOT / D2R3_REL


NOTEBOOK11_REL = Path(
    "notebooks/"
    "11_division_association_audit.ipynb"
)

NOTEBOOK11 = ROOT / NOTEBOOK11_REL


STRUCTURAL_ORACLE_REL = Path(
    "reports/"
    "stage11_division_probe/"
    "stage11_fold0_morphdiv_v2_actions_locked.pkl"
)

STRUCTURAL_ORACLE = ROOT / STRUCTURAL_ORACLE_REL


UPSTREAM_ACTION_BASENAME = (
    "stage11_deployable_final_v9_policy_grid_actions.pkl"
)


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_b_"
    "morph_semantic_reconstruction_contract_and_"
    "upstream_generator_recovery_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

# Parent output was truncated in notebook display, but its exact FILE SHA was
# visible and is a byte-exact cryptographic parent lock.
EXPECTED_PARENT_FILE_SHA = (
    "85b2139a9ca92e04a0a6075e5bb324e353ec19856d92630740cf017dd7f29ebb"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_D2R3_FILE_SHA = (
    "ac0cb4959b8a9cb54784d0fd37d4da89a87b6e110298f67af00e712deebf8afb"
)

EXPECTED_D2R3_LOGICAL_CONTENT_SHA = (
    "9a0ca7a1d2013a9cd0ddfe74ec00e9f787505c19ca69c6d0c4e297db03a701b0"
)


EXPECTED_STRUCTURAL_ORACLE_SHA = (
    "9c08f88c9824912af72a5538e6f23cd6bc7e14ba4ffc30e26ad9ad6ca37dd9eb"
)


EXPECTED_AUTHORITY_MODE = (
    "SEMANTIC_SOURCE_LOCK_PLUS_"
    "HISTORICAL_GT_BLIND_STRUCTURAL_ORACLE"
)


# Notebook11 authority.
SEMANTIC_ANCHOR_CELL_INDEX = 63

EXPECTED_SEMANTIC_ANCHOR_CELL_SHA = (
    "2e6bdff585ad1a3ee61b3a82ec03ccbd05db0b66f01024a35a721d76a29d8834"
)


APPLICATION_CORE_CELL_INDEX = 66

EXPECTED_APPLICATION_CORE_FUNCTION_SHA = (
    "de986b374510b6772444772cc4a861bc941efdbc1fc282f867c7bb4f71bfcf43"
)


# Current portable selection / application layer.
EXPECTED_VALIDATE_TABLE_SHA = (
    "38e6aee0ad7193a1431a5b0ff0ec4ccd3f5b6680f78787584ca0c0b089ace0bf"
)

EXPECTED_SELECT_SHA = (
    "692063ed9398b0a9b26bb17e42c06aa60cb520164f9748666b9852044564a78b"
)

EXPECTED_APPLY_MORPH_SHA = (
    "db173c58a83eddd282712915514c31739815195ccb99edaa84a3ff5e16f1ffcf"
)


EXPECTED_CANDIDATE_COLUMNS = (
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
)


EXPECTED_MORPH_RANK_MAX = 3

EXPECTED_CANDIDATE_RANK_MAX = 3

EXPECTED_CANDIDATE_PROB_MIN = 0.10


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def safe_rel(
    path: Path,
) -> str:

    try:

        return str(
            path.resolve().relative_to(
                ROOT
            )
        )

    except Exception:

        return str(
            path
        )


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )


    assert (
        source is not None
    )


    return source


def source_sha(
    text: str,
    node: ast.AST,
) -> str:

    return sha256_text(
        node_source(
            text,
            node,
        )
    )


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in obj.items():

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:

    return any(
        record[
            "value"
        ]
        == target

        for record in flatten_json(
            obj
        )
    )


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in tree.body

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def function_calls(
    fn: ast.AST,
) -> list[str]:

    return sorted({
        dotted_name(
            node.func
        )
        or "<dynamic-call>"

        for node in ast.walk(
            fn
        )

        if isinstance(
            node,
            ast.Call,
        )
    })


def function_names(
    fn: ast.AST,
) -> set[str]:

    result = set()


    for node in ast.walk(
        fn
    ):

        if isinstance(
            node,
            ast.Name,
        ):

            result.add(
                node.id
            )


        elif isinstance(
            node,
            ast.Attribute,
        ):

            name = dotted_name(
                node
            )


            if name:

                result.add(
                    name
                )


    return result


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    ), path


    actual_file_sha = sha256_file(
        path
    )


    assert (
        actual_file_sha
        == EXPECTED_PARENT_FILE_SHA
    ), (
        "14.3E-D-A-R2 parent file SHA drift.\n"
        f"expected={EXPECTED_PARENT_FILE_SHA}\n"
        f"actual={actual_file_sha}"
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    observed_content_sha = (
        obj.get(
            "content_sha256"
        )
    )


    # Parent content SHA was truncated in the pasted notebook output.
    # Exact file SHA is therefore the hard external lock; if the artifact
    # follows the current content_sha256 envelope, prove its self-consistency.
    if (
        isinstance(
            observed_content_sha,
            str,
        )
        and
        re.fullmatch(
            r"[0-9a-f]{64}",
            observed_content_sha,
        )
    ):

        payload = dict(
            obj
        )

        payload.pop(
            "content_sha256"
        )


        assert (
            canonical_json_sha(
                payload
            )
            == observed_content_sha
        )


    assert (
        obj[
            "claim_boundary"
        ][
            "concrete_Morph_provider_implementation"
        ]
        == "NOT_AUTHORIZED_YET"
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "historical_Morph_action_content_opened"
        ]
        is False
    )


    return obj


# =============================================================================
# 4. Notebook helpers
# =============================================================================

def notebook_cells(
    path: Path,
) -> list[
    dict[str, Any]
]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    result = []


    for index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        result.append({
            "index":
                index,

            "type":
                cell.get(
                    "cell_type"
                ),

            "source":
                source,

            "sha256":
                sha256_text(
                    source
                ),
        })


    return result


def find_function_by_sha(
    cells: list[
        dict[str, Any]
    ],
    expected_sha: str,
) -> list[
    dict[str, Any]
]:

    result = []


    for cell in cells:

        if (
            cell[
                "type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                ),
            ):

                continue


            segment = ast.get_source_segment(
                source,
                node,
            )


            if segment is None:

                continue


            observed_sha = sha256_text(
                segment
            )


            if observed_sha != expected_sha:

                continue


            result.append({
                "cell_index":
                    cell[
                        "index"
                    ],

                "cell_sha256":
                    cell[
                        "sha256"
                    ],

                "function":
                    node.name,

                "line_start":
                    node.lineno,

                "line_end":
                    node.end_lineno,

                "source_sha256":
                    observed_sha,

                "source":
                    segment,

                "node":
                    node,
            })


    return result


# =============================================================================
# 5. Extract frozen policy dictionary from semantic anchor
# =============================================================================

POLICY_KEYS = {
    "candidate_universe",
    "morphology_top_k",
    "association_rank_cap",
    "association_prob_threshold",
    "mutation",
}


def extract_morph_policy_from_anchor(
    source: str,
) -> dict[str, Any]:

    tree = ast.parse(
        source
    )


    candidates = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            ast.Dict,
        ):

            continue


        mapping = {}


        for key_node, value_node in zip(
            node.keys,
            node.values,
        ):

            if key_node is None:

                continue


            try:

                key = ast.literal_eval(
                    key_node
                )

            except Exception:

                continue


            if not isinstance(
                key,
                str,
            ):

                continue


            if key not in POLICY_KEYS:

                continue


            try:

                value = ast.literal_eval(
                    value_node
                )

            except Exception:

                value = ast.unparse(
                    value_node
                )


            mapping[
                key
            ] = value


        if (
            set(
                mapping
            )
            ==
            POLICY_KEYS
        ):

            candidates.append(
                mapping
            )


    assert (
        len(
            candidates
        )
        == 1
    ), (
        "Expected exactly one complete Morph policy dictionary "
        f"in semantic anchor, found {len(candidates)}."
    )


    return candidates[
        0
    ]


# =============================================================================
# 6. Portable literal extraction
# =============================================================================

def top_level_literal_assignments(
    tree: ast.Module,
) -> dict[
    str,
    Any,
]:

    result = {}


    for stmt in tree.body:

        names = []

        value_node = None


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in stmt.targets:

                if isinstance(
                    target,
                    ast.Name,
                ):

                    names.append(
                        target.id
                    )


            value_node = stmt.value


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            if isinstance(
                stmt.target,
                ast.Name,
            ):

                names.append(
                    stmt.target.id
                )


            value_node = stmt.value


        if (
            not names
            or
            value_node is None
        ):

            continue


        try:

            literal = ast.literal_eval(
                value_node
            )

        except Exception:

            continue


        for name in names:

            result[
                name
            ] = literal


    return result


# =============================================================================
# 7. Distinguish graph-application core from candidate-generation core
# =============================================================================

def assigned_subscript_columns(
    node: ast.AST,
) -> set[str]:

    result = set()


    for item in ast.walk(
        node
    ):

        if not isinstance(
            item,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
            ),
        ):

            continue


        if isinstance(
            item,
            ast.Assign,
        ):

            targets = list(
                item.targets
            )


        else:

            targets = [
                item.target
            ]


        for target in targets:

            for child in ast.walk(
                target
            ):

                if not isinstance(
                    child,
                    ast.Subscript,
                ):

                    continue


                try:

                    value = ast.literal_eval(
                        child.slice
                    )

                except Exception:

                    continue


                if isinstance(
                    value,
                    str,
                ):

                    result.add(
                        value
                    )


    return result


def dict_literal_keys(
    node: ast.AST,
) -> set[str]:

    result = set()


    for child in ast.walk(
        node
    ):

        if not isinstance(
            child,
            ast.Dict,
        ):

            continue


        for key_node in child.keys:

            if key_node is None:

                continue


            try:

                key = ast.literal_eval(
                    key_node
                )

            except Exception:

                continue


            if isinstance(
                key,
                str,
            ):

                result.add(
                    key
                )


    return result


# =============================================================================
# 8. Exact upstream-artifact writer dataflow
# =============================================================================

def assignment_target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    def visit(
        value: ast.AST,
    ):

        if isinstance(
            value,
            ast.Name,
        ):

            result.add(
                value.id
            )


        elif isinstance(
            value,
            (
                ast.Tuple,
                ast.List,
            ),
        ):

            for child in value.elts:

                visit(
                    child
                )


        elif isinstance(
            value,
            ast.Starred,
        ):

            visit(
                value.value
            )


    visit(
        node
    )


    return result


def referenced_names(
    node: ast.AST,
) -> set[str]:

    return {
        child.id

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Name,
        )
    }


def node_mentions_upstream_literal(
    node: ast.AST,
) -> bool:

    return any(
        (
            isinstance(
                child,
                ast.Constant,
            )
            and
            isinstance(
                child.value,
                str,
            )
            and
            UPSTREAM_ACTION_BASENAME
            in child.value
        )

        for child in ast.walk(
            node
        )
    )


def expression_is_upstream_tainted(
    node: ast.AST,
    tainted_names: set[str],
) -> bool:

    return (
        node_mentions_upstream_literal(
            node
        )
        or
        bool(
            referenced_names(
                node
            )
            &
            tainted_names
        )
    )


def update_upstream_taint(
    stmt: ast.stmt,
    tainted_names: set[str],
):

    if isinstance(
        stmt,
        ast.Assign,
    ):

        rhs_tainted = expression_is_upstream_tainted(
            stmt.value,
            tainted_names,
        )


        names = set()


        for target in stmt.targets:

            names |= assignment_target_names(
                target
            )


        for name in names:

            if rhs_tainted:

                tainted_names.add(
                    name
                )


            else:

                tainted_names.discard(
                    name
                )


    elif isinstance(
        stmt,
        ast.AnnAssign,
    ):

        names = assignment_target_names(
            stmt.target
        )


        rhs_tainted = (
            stmt.value is not None
            and
            expression_is_upstream_tainted(
                stmt.value,
                tainted_names,
            )
        )


        for name in names:

            if rhs_tainted:

                tainted_names.add(
                    name
                )


            else:

                tainted_names.discard(
                    name
                )


def pickle_write_path_argument(
    call: ast.Call,
) -> ast.AST | None:

    call_name = dotted_name(
        call.func
    )


    if call_name is None:

        return None


    # pd.to_pickle(obj, path)
    if call_name in {
        "pd.to_pickle",
        "pandas.to_pickle",
    }:

        if len(
            call.args
        ) >= 2:

            return call.args[
                1
            ]


        for kw in call.keywords:

            if kw.arg in {
                "path",
                "filepath_or_buffer",
            }:

                return kw.value


        return None


    # table.to_pickle(path)
    if call_name.endswith(
        ".to_pickle"
    ):

        if call.args:

            return call.args[
                0
            ]


        for kw in call.keywords:

            if kw.arg in {
                "path",
                "filepath_or_buffer",
            }:

                return kw.value


    return None


def pickle_read_path_argument(
    call: ast.Call,
) -> ast.AST | None:

    name = dotted_name(
        call.func
    )


    if (
        name is None
        or
        not name.endswith(
            "read_pickle"
        )
    ):

        return None


    if call.args:

        return call.args[
            0
        ]


    for kw in call.keywords:

        if kw.arg in {
            "path",
            "filepath_or_buffer",
        }:

            return kw.value


    return None


# =============================================================================
# 9. Executable hazard / generation evidence
# =============================================================================

GT_CALL_TOKENS = (
    "evaluate_graph",
    "official_evaluate",
    "per_sample_metrics",
    "node_recall",
    "summarise",
    "_evaluate_matched_graph",
)


GT_EXACT_NAMES = {
    "gt",
    "gt_graph",
    "gt_base",
    "gt_policy",
    "gt_nodes",
    "gt_edges",
    "ground_truth",
}


GENERATION_SCHEMA = {
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
}


def executable_profile(
    source: str,
) -> dict[str, Any]:

    try:

        tree = ast.parse(
            source
        )

    except Exception:

        return {
            "parseable":
                False,

            "calls":
                [],

            "names":
                [],

            "dict_keys":
                [],

            "assigned_columns":
                [],

            "GT_calls":
                [],

            "GT_names":
                [],

            "schema_hits":
                [],

            "generation_signals":
                [],

            "has_read_pickle":
                False,
        }


    calls = sorted({
        dotted_name(
            node.func
        )
        or "<dynamic-call>"

        for node in ast.walk(
            tree
        )

        if isinstance(
            node,
            ast.Call,
        )
    })


    names = sorted({
        node.id

        for node in ast.walk(
            tree
        )

        if isinstance(
            node,
            ast.Name,
        )
    })


    keys = sorted(
        dict_literal_keys(
            tree
        )
    )


    assigned = sorted(
        assigned_subscript_columns(
            tree
        )
    )


    lower_calls = [
        value.lower()

        for value in calls
    ]


    gt_calls = sorted({
        value

        for value in calls

        if any(
            token
            in value.lower()

            for token in GT_CALL_TOKENS
        )
    })


    gt_names = sorted({
        value

        for value in names

        if value.lower()
        in GT_EXACT_NAMES
    })


    schema_hits = sorted(
        GENERATION_SCHEMA
        &
        set(
            keys
            +
            assigned
        )
    )


    source_lower = (
        source.lower()
    )


    generation_signals = sorted({
        token

        for token in (
            "morph_rank",
            "candidate_rank",
            "candidate_prob",
            "association",
            "rank",
            "top",
            "morph",
            "records.append",
            "dataframe",
        )

        if token
        in source_lower
    })


    return {
        "parseable":
            True,

        "calls":
            calls,

        "names":
            names,

        "dict_keys":
            keys,

        "assigned_columns":
            assigned,

        "GT_calls":
            gt_calls,

        "GT_names":
            gt_names,

        "schema_hits":
            schema_hits,

        "generation_signals":
            generation_signals,

        "has_read_pickle":
            any(
                "read_pickle"
                in value.lower()

                for value in calls
            ),
    }


# =============================================================================
# 10. Scan Notebook11 for upstream path-linked writers
# =============================================================================

def scan_notebook_upstream_io(
    cells: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    tainted_names: set[str] = set()

    result = []


    for cell in cells:

        if (
            cell[
                "type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        literal_hit = (
            UPSTREAM_ACTION_BASENAME
            in source
        )


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            if literal_hit:

                result.append({
                    "path":
                        str(
                            NOTEBOOK11_REL
                        ),

                    "cell_index":
                        cell[
                            "index"
                        ],

                    "cell_sha256":
                        cell[
                            "sha256"
                        ],

                    "classification":
                        "UNPARSEABLE_MENTION",

                    "source":
                        source,
                })


            continue


        writes = []

        reads = []


        for stmt in tree.body:

            update_upstream_taint(
                stmt,
                tainted_names,
            )


            for node in ast.walk(
                stmt
            ):

                if not isinstance(
                    node,
                    ast.Call,
                ):

                    continue


                write_arg = pickle_write_path_argument(
                    node
                )


                if (
                    write_arg is not None
                    and
                    expression_is_upstream_tainted(
                        write_arg,
                        tainted_names,
                    )
                ):

                    writes.append({
                        "line":
                            node.lineno,

                        "call":
                            ast.unparse(
                                node
                            ),
                    })


                read_arg = pickle_read_path_argument(
                    node
                )


                if (
                    read_arg is not None
                    and
                    expression_is_upstream_tainted(
                        read_arg,
                        tainted_names,
                    )
                ):

                    reads.append({
                        "line":
                            node.lineno,

                        "call":
                            ast.unparse(
                                node
                            ),
                    })


        if (
            not literal_hit
            and
            not writes
            and
            not reads
        ):

            continue


        if writes:

            classification = (
                "UPSTREAM_TARGET_WRITER"
            )


        elif reads:

            classification = (
                "UPSTREAM_TARGET_READER"
            )


        else:

            classification = (
                "UPSTREAM_MENTION_ONLY"
            )


        result.append({
            "path":
                str(
                    NOTEBOOK11_REL
                ),

            "cell_index":
                cell[
                    "index"
                ],

            "cell_sha256":
                cell[
                    "sha256"
                ],

            "classification":
                classification,

            "writes":
                writes,

            "reads":
                reads,

            "profile":
                executable_profile(
                    source
                ),

            "source":
                source,
        })


    return result


# =============================================================================
# 11. Scan scripts for upstream path-linked writers
# =============================================================================

def scan_script_upstream_io(
    path: Path,
) -> list[
    dict[str, Any]
]:

    try:

        source = path.read_text(
            encoding="utf-8"
        )


        tree = ast.parse(
            source,
            filename=str(
                path
            ),
        )

    except Exception:

        return []


    tainted_names: set[str] = set()

    writes = []

    reads = []


    for stmt in tree.body:

        update_upstream_taint(
            stmt,
            tainted_names,
        )


        for node in ast.walk(
            stmt
        ):

            if not isinstance(
                node,
                ast.Call,
            ):

                continue


            write_arg = pickle_write_path_argument(
                node
            )


            if (
                write_arg is not None
                and
                expression_is_upstream_tainted(
                    write_arg,
                    tainted_names,
                )
            ):

                writes.append({
                    "line":
                        node.lineno,

                    "call":
                        ast.unparse(
                            node
                        ),
                })


            read_arg = pickle_read_path_argument(
                node
            )


            if (
                read_arg is not None
                and
                expression_is_upstream_tainted(
                    read_arg,
                    tainted_names,
                )
            ):

                reads.append({
                    "line":
                        node.lineno,

                    "call":
                        ast.unparse(
                            node
                        ),
                })


    if (
        UPSTREAM_ACTION_BASENAME
        not in source
        and
        not writes
        and
        not reads
    ):

        return []


    functions = top_level_functions(
        tree
    )


    function_profiles = []


    for fn in functions.values():

        fn_source = node_source(
            source,
            fn,
        )


        if not any(
            token
            in fn_source

            for token in (
                "morph_rank",
                "candidate_rank",
                "candidate_prob",
                UPSTREAM_ACTION_BASENAME,
            )
        ):

            continue


        function_profiles.append({
            "function":
                fn.name,

            "line_start":
                fn.lineno,

            "line_end":
                fn.end_lineno,

            "function_sha256":
                source_sha(
                    source,
                    fn,
                ),

            "profile":
                executable_profile(
                    fn_source
                ),

            "source":
                fn_source,
        })


    return [
        {
            "path":
                safe_rel(
                    path
                ),

            "file_sha256":
                sha256_file(
                    path
                ),

            "writes":
                writes,

            "reads":
                reads,

            "function_profiles":
                function_profiles,

            "source_mentions_upstream":
                (
                    UPSTREAM_ACTION_BASENAME
                    in source
                ),
        }
    ]


# =============================================================================
# 12. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    D2R3,
    NOTEBOOK11,
    STRUCTURAL_ORACLE,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-B "
    "MORPH SEMANTIC RECONSTRUCTION CONTRACT "
    "+ UPSTREAM GENERATOR RECOVERY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "D2-R3 AUTHORITY:",
    D2R3_REL,
    flush=True,
)

print(
    "NOTEBOOK11:",
    NOTEBOOK11_REL,
    flush=True,
)

print(
    "UPSTREAM ACTION BASENAME:",
    UPSTREAM_ACTION_BASENAME,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Parent exact file lock
# =============================================================================

parent = verify_parent(
    PARENT
)


parent_content_sha = (
    parent.get(
        "content_sha256"
    )
)


print(
    "[PASS] PARENT_14_3E_D_A_R2_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  observed content SHA:",
    parent_content_sha,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. D2-R3 authority lock
# =============================================================================

d2r3_sha = sha256_file(
    D2R3
)


assert (
    d2r3_sha
    == EXPECTED_D2R3_FILE_SHA
)


d2r3 = load_json(
    D2R3
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_D2R3_LOGICAL_CONTENT_SHA,
    )
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_AUTHORITY_MODE,
    )
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_STRUCTURAL_ORACLE_SHA,
    )
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_APPLICATION_CORE_FUNCTION_SHA,
    )
)


print(
    "[PASS] D2_R3_SEMANTIC_RECONSTRUCTION_AUTHORITY_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    d2r3_sha,
    flush=True,
)

print(
    "  logical content SHA:",
    EXPECTED_D2R3_LOGICAL_CONTENT_SHA,
    flush=True,
)

print(
    "  authority mode:",
    EXPECTED_AUTHORITY_MODE,
    flush=True,
)

print(
    "  application core SHA:",
    EXPECTED_APPLICATION_CORE_FUNCTION_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Structural oracle identity only
# =============================================================================

oracle_sha = sha256_file(
    STRUCTURAL_ORACLE
)


assert (
    oracle_sha
    == EXPECTED_STRUCTURAL_ORACLE_SHA
)


historical_oracle_semantic_content_loaded = (
    False
)


print(
    "[PASS] HISTORICAL_STRUCTURAL_ORACLE_IDENTITY_LOCK",
    flush=True,
)

print(
    "  SHA:",
    oracle_sha,
    flush=True,
)

print(
    "  semantic content loaded:",
    historical_oracle_semantic_content_loaded,
    flush=True,
)

print(
    "  runtime use:",
    False,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Notebook11 semantic anchor
# =============================================================================

cells = notebook_cells(
    NOTEBOOK11
)


assert (
    SEMANTIC_ANCHOR_CELL_INDEX
    <
    len(
        cells
    )
)


anchor = cells[
    SEMANTIC_ANCHOR_CELL_INDEX
]


assert (
    anchor[
        "sha256"
    ]
    == EXPECTED_SEMANTIC_ANCHOR_CELL_SHA
)


frozen_policy = (
    extract_morph_policy_from_anchor(
        anchor[
            "source"
        ]
    )
)


print(
    "[PASS] MORPH_SEMANTIC_ANCHOR_EXACT_LOCK",
    flush=True,
)

print(
    "  cell:",
    SEMANTIC_ANCHOR_CELL_INDEX,
    flush=True,
)

print(
    "  cell SHA:",
    anchor[
        "sha256"
    ],
    flush=True,
)

print(
    "  frozen policy:",
    json.dumps(
        frozen_policy,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 17. Frozen policy semantic assertions
# =============================================================================

assert (
    frozen_policy[
        "morphology_top_k"
    ]
    == EXPECTED_MORPH_RANK_MAX
)


assert (
    frozen_policy[
        "association_rank_cap"
    ]
    == EXPECTED_CANDIDATE_RANK_MAX
)


assert (
    abs(
        float(
            frozen_policy[
                "association_prob_threshold"
            ]
        )
        -
        EXPECTED_CANDIDATE_PROB_MIN
    )
    <= 1e-15
)


assert (
    "Frozen-V9"
    in frozen_policy[
        "candidate_universe"
    ]
)


assert (
    "W4-capable"
    in frozen_policy[
        "candidate_universe"
    ]
)


assert (
    "outdegree=1"
    in frozen_policy[
        "candidate_universe"
    ].replace(
        " ",
        "",
    )
)


assert (
    "additive second edge"
    in frozen_policy[
        "mutation"
    ].lower()
)


print(
    "[PASS] FROZEN_SCIENTIFIC_POLICY_LOCK",
    flush=True,
)

print(
    "  morphology top-k:",
    frozen_policy[
        "morphology_top_k"
    ],
    flush=True,
)

print(
    "  association rank cap:",
    frozen_policy[
        "association_rank_cap"
    ],
    flush=True,
)

print(
    "  association probability min:",
    frozen_policy[
        "association_prob_threshold"
    ],
    flush=True,
)

print(
    "  mutation:",
    frozen_policy[
        "mutation"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 18. Current portable selection layer
# =============================================================================

portable_sha = sha256_file(
    PORTABLE
)


assert (
    portable_sha
    == EXPECTED_PORTABLE_SHA
)


portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_text,
    filename=str(
        PORTABLE
    ),
)


portable_functions = top_level_functions(
    portable_tree
)


portable_literals = top_level_literal_assignments(
    portable_tree
)


for required_fn in (
    "validate_morph_candidate_table",
    "select_morph_div_v2_candidates",
    "apply_portable_morph_div_v2",
):

    assert (
        required_fn
        in portable_functions
    )


validate_fn = portable_functions[
    "validate_morph_candidate_table"
]

select_fn = portable_functions[
    "select_morph_div_v2_candidates"
]

apply_fn = portable_functions[
    "apply_portable_morph_div_v2"
]


validate_sha = source_sha(
    portable_text,
    validate_fn,
)


select_sha = source_sha(
    portable_text,
    select_fn,
)


apply_sha = source_sha(
    portable_text,
    apply_fn,
)


assert (
    validate_sha
    == EXPECTED_VALIDATE_TABLE_SHA
)


assert (
    select_sha
    == EXPECTED_SELECT_SHA
)


assert (
    apply_sha
    == EXPECTED_APPLY_MORPH_SHA
)


morph_literals = {
    key:
        value

    for key, value
    in portable_literals.items()

    if (
        "MORPH_DIV_V2"
        in key
    )
}


print(
    "=" * 120,
    flush=True,
)

print(
    "CURRENT PORTABLE MORPH LITERALS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for key, value in sorted(
    morph_literals.items()
):

    print(
        " ",
        key,
        "=",
        repr(
            value
        ),
        flush=True,
    )


print(
    flush=True,
)


# Resolve constants by semantic value/shape rather than inventing names.
candidate_column_constants = [
    (
        key,
        tuple(
            value
        ),
    )

    for key, value
    in morph_literals.items()

    if (
        isinstance(
            value,
            (
                tuple,
                list,
            ),
        )
        and
        tuple(
            value
        )
        ==
        EXPECTED_CANDIDATE_COLUMNS
    )
]


assert (
    len(
        candidate_column_constants
    )
    == 1
)


prob_constants = [
    (
        key,
        value,
    )

    for key, value
    in morph_literals.items()

    if (
        isinstance(
            value,
            float,
        )
        and
        abs(
            value
            -
            EXPECTED_CANDIDATE_PROB_MIN
        )
        <= 1e-15
    )
]


rank_constants = [
    (
        key,
        value,
    )

    for key, value
    in morph_literals.items()

    if (
        isinstance(
            value,
            int,
        )
        and
        value
        ==
        3
    )
]


select_source = node_source(
    portable_text,
    select_fn,
)


assert (
    "morph_rank"
    in select_source
)


assert (
    "candidate_rank"
    in select_source
)


assert (
    "candidate_prob"
    in select_source
)


print(
    "[PASS] PORTABLE_SELECTION_LAYER_LOCK",
    flush=True,
)

print(
    "  validate table SHA:",
    validate_sha,
    flush=True,
)

print(
    "  select SHA:",
    select_sha,
    flush=True,
)

print(
    "  apply Morph SHA:",
    apply_sha,
    flush=True,
)

print(
    "  candidate schema:",
    candidate_column_constants[
        0
    ],
    flush=True,
)

print(
    "  probability constants:",
    prob_constants,
    flush=True,
)

print(
    "  rank constants:",
    rank_constants,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Policy parity: historical anchor vs current portable selection
# =============================================================================

portable_selection_policy_matches_anchor = (
    any(
        (
            isinstance(
                value,
                float,
            )
            and
            abs(
                value
                -
                float(
                    frozen_policy[
                        "association_prob_threshold"
                    ]
                )
            )
            <= 1e-15
        )

        for value
        in morph_literals.values()
    )
    and
    any(
        (
            isinstance(
                value,
                int,
            )
            and
            value
            ==
            int(
                frozen_policy[
                    "morphology_top_k"
                ]
            )
        )

        for value
        in morph_literals.values()
    )
    and
    any(
        (
            isinstance(
                value,
                int,
            )
            and
            value
            ==
            int(
                frozen_policy[
                    "association_rank_cap"
                ]
            )
        )

        for value
        in morph_literals.values()
    )
)


assert (
    portable_selection_policy_matches_anchor
)


print(
    "[PASS] HISTORICAL_TO_PORTABLE_SELECTION_POLICY_PARITY",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 20. D2-R3 selected semantic core exact source
# =============================================================================

core_matches = find_function_by_sha(
    cells,
    EXPECTED_APPLICATION_CORE_FUNCTION_SHA,
)


assert (
    len(
        core_matches
    )
    == 1
)


core = core_matches[
    0
]


core_node = core[
    "node"
]


core_calls = function_calls(
    core_node
)


core_assigned_columns = (
    assigned_subscript_columns(
        core_node
    )
)


core_dict_keys = dict_literal_keys(
    core_node
)


core_source = core[
    "source"
]


candidate_construction_columns = {
    "morph_rank",
    "candidate_rank",
    "candidate_prob",
}


core_candidate_column_assignments = sorted(
    candidate_construction_columns
    &
    core_assigned_columns
)


core_candidate_dict_construction = sorted(
    candidate_construction_columns
    &
    core_dict_keys
)


core_has_graph_mutation = any(
    (
        call.endswith(
            "add_edge"
        )
        or
        call.endswith(
            "add_locked_action"
        )
    )

    for call in core_calls
)


core_reads_action_rows = (
    ".itertuples"
    in core_source
    and
    "candidate_prob"
    in core_source
)


core_computes_association = any(
    "association"
    in call.lower()

    for call in core_calls
)


core_generates_candidate_values = (
    bool(
        core_candidate_column_assignments
    )
    or
    bool(
        core_candidate_dict_construction
    )
    or
    core_computes_association
)


application_core_classification = (
    "ACTION_APPLICATION_CORE_NOT_CANDIDATE_GENERATOR"
    if (
        core_has_graph_mutation
        and
        core_reads_action_rows
        and
        not core_generates_candidate_values
    )
    else
    "REQUIRES_FURTHER_CLASSIFICATION"
)


print(
    "=" * 120,
    flush=True,
)

print(
    "D2-R3 SELECTED SEMANTIC CORE CLASSIFICATION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "cell:",
    core[
        "cell_index"
    ],
    flush=True,
)

print(
    "function:",
    core[
        "function"
    ],
    flush=True,
)

print(
    "function SHA:",
    core[
        "source_sha256"
    ],
    flush=True,
)

print(
    "graph mutation:",
    core_has_graph_mutation,
    flush=True,
)

print(
    "reads existing action rows:",
    core_reads_action_rows,
    flush=True,
)

print(
    "candidate column assignments:",
    core_candidate_column_assignments,
    flush=True,
)

print(
    "candidate dict construction:",
    core_candidate_dict_construction,
    flush=True,
)

print(
    "association computation:",
    core_computes_association,
    flush=True,
)

print(
    "classification:",
    application_core_classification,
    flush=True,
)

print(
    flush=True,
)


assert (
    application_core_classification
    ==
    "ACTION_APPLICATION_CORE_NOT_CANDIDATE_GENERATOR"
), (
    "D2-R3 selected core unexpectedly appears to have become "
    "candidate-generation authority."
)


# =============================================================================
# 21. D2-R3 graph application contract
# =============================================================================

application_contract = {
    "source_pre_outdegree":
        1,

    "target_pre_indegree":
        0,

    "duplicate_edge":
        "FORBIDDEN",

    "source_post_outdegree":
        2,

    "target_post_indegree":
        1,

    "operation":
        "ADD_SECOND_OUTGOING_EDGE_FROM_SOURCE_TO_FREE_TARGET",

    "ordering": [
        "morph_rank",
        "t",
        "source_pred_node_id",
        "target_pred_node_id",
    ],
}


for required_text in (
    "outdeg.get(src, 0) == 1",
    "indeg.get(tgt, 0) == 0",
    "(src, tgt) not in edge_set",
):

    assert (
        required_text
        in core_source
    ), required_text


for required_column in (
    application_contract[
        "ordering"
    ]
):

    assert (
        required_column
        in core_source
    )


print(
    "[PASS] GRAPH_APPLICATION_CONTRACT_LOCK",
    flush=True,
)

print(
    json.dumps(
        application_contract,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 22. Search Notebook11 upstream policy-grid writer
# =============================================================================

notebook_upstream_records = (
    scan_notebook_upstream_io(
        cells
    )
)


notebook_upstream_writers = [
    record

    for record
    in notebook_upstream_records

    if (
        record[
            "classification"
        ]
        == "UPSTREAM_TARGET_WRITER"
    )
]


notebook_upstream_readers = [
    record

    for record
    in notebook_upstream_records

    if (
        record[
            "classification"
        ]
        == "UPSTREAM_TARGET_READER"
    )
]


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK11 UPSTREAM POLICY-GRID TARGET I/O",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "writers:",
    len(
        notebook_upstream_writers
    ),
    flush=True,
)

print(
    "readers:",
    len(
        notebook_upstream_readers
    ),
    flush=True,
)

print(
    flush=True,
)


for record in notebook_upstream_records:

    print(
        record[
            "classification"
        ],
        "cell=",
        record[
            "cell_index"
        ],
        "SHA=",
        record[
            "cell_sha256"
        ],
        flush=True,
    )


    if record.get(
        "writes"
    ):

        print(
            "  writes:",
            record[
                "writes"
            ],
            flush=True,
        )


    if record.get(
        "reads"
    ):

        print(
            "  reads:",
            record[
                "reads"
            ],
            flush=True,
        )


print(
    flush=True,
)


# =============================================================================
# 23. Search scripts for upstream policy-grid writer
# =============================================================================

script_upstream_records = []


SCRIPTS = ROOT / "scripts"


for path in sorted(
    SCRIPTS.rglob(
        "*.py"
    )
):

    script_upstream_records.extend(
        scan_script_upstream_io(
            path
        )
    )


script_upstream_writers = [
    record

    for record
    in script_upstream_records

    if record[
        "writes"
    ]
]


script_upstream_readers = [
    record

    for record
    in script_upstream_records

    if record[
        "reads"
    ]
]


print(
    "=" * 120,
    flush=True,
)

print(
    "SCRIPT UPSTREAM POLICY-GRID TARGET I/O",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "writer files:",
    len(
        script_upstream_writers
    ),
    flush=True,
)

print(
    "reader files:",
    len(
        script_upstream_readers
    ),
    flush=True,
)

print(
    flush=True,
)


for record in script_upstream_records:

    print(
        record[
            "path"
        ],
        flush=True,
    )


    if record[
        "writes"
    ]:

        print(
            "  WRITES:",
            record[
                "writes"
            ],
            flush=True,
        )


    if record[
        "reads"
    ]:

        print(
            "  READS:",
            record[
                "reads"
            ],
            flush=True,
        )


    for fn in record[
        "function_profiles"
    ]:

        print(
            "  function:",
            fn[
                "function"
            ],
            "SHA=",
            fn[
                "function_sha256"
            ],
            flush=True,
        )

        print(
            "    schema hits:",
            fn[
                "profile"
            ][
                "schema_hits"
            ],
            flush=True,
        )

        print(
            "    GT calls:",
            fn[
                "profile"
            ][
                "GT_calls"
            ],
            flush=True,
        )

        print(
            "    GT names:",
            fn[
                "profile"
            ][
                "GT_names"
            ],
            flush=True,
        )


print(
    flush=True,
)


# =============================================================================
# 24. Consolidate upstream writer units
# =============================================================================

upstream_writer_units = []


for record in notebook_upstream_writers:

    profile = record[
        "profile"
    ]


    upstream_writer_units.append({
        "kind":
            "NOTEBOOK_CELL",

        "path":
            record[
                "path"
            ],

        "cell_index":
            record[
                "cell_index"
            ],

        "source_sha256":
            record[
                "cell_sha256"
            ],

        "write_sinks":
            record[
                "writes"
            ],

        "profile":
            profile,

        "source":
            record[
                "source"
            ],
    })


for record in script_upstream_writers:

    # A script-level path-linked writer may contain multiple functions.
    # Keep each function profile separately for inner-block follow-up.
    if record[
        "function_profiles"
    ]:

        for fn in record[
            "function_profiles"
        ]:

            upstream_writer_units.append({
                "kind":
                    "PYTHON_FUNCTION_IN_WRITER_FILE",

                "path":
                    record[
                        "path"
                    ],

                "file_sha256":
                    record[
                        "file_sha256"
                    ],

                "function":
                    fn[
                        "function"
                    ],

                "source_sha256":
                    fn[
                        "function_sha256"
                    ],

                "write_sinks":
                    record[
                        "writes"
                    ],

                "profile":
                    fn[
                        "profile"
                    ],

                "source":
                    fn[
                        "source"
                    ],
            })


    else:

        upstream_writer_units.append({
            "kind":
                "PYTHON_WRITER_FILE_NO_MATCHING_FUNCTION",

            "path":
                record[
                    "path"
                ],

            "file_sha256":
                record[
                    "file_sha256"
                ],

            "source_sha256":
                None,

            "write_sinks":
                record[
                    "writes"
                ],

            "profile":
                None,

            "source":
                "",
        })


# =============================================================================
# 25. Classify upstream writer units
# =============================================================================

def writer_unit_classification(
    record: dict[str, Any],
) -> str:

    profile = record.get(
        "profile"
    )


    if not profile:

        return (
            "WRITER_PATH_LINKED_INNER_SOURCE_NOT_YET_LOCALIZED"
        )


    gt_hazard = (
        bool(
            profile[
                "GT_calls"
            ]
        )
        or
        bool(
            profile[
                "GT_names"
            ]
        )
    )


    schema_complete = (
        set(
            profile[
                "schema_hits"
            ]
        )
        >=
        GENERATION_SCHEMA
    )


    generation_rich = (
        len(
            profile[
                "generation_signals"
            ]
        )
        >= 4
    )


    if (
        schema_complete
        and
        generation_rich
        and
        not gt_hazard
        and
        not profile[
            "has_read_pickle"
        ]
    ):

        return (
            "GT_BLIND_GENERATOR_SOURCE_CANDIDATE"
        )


    if (
        schema_complete
        and
        generation_rich
        and
        gt_hazard
    ):

        return (
            "WRITER_CONTAINS_GENERATION_AND_GT_"
            "INNER_GT_BLIND_BLOCK_EXTRACTION_REQUIRED"
        )


    if profile[
        "has_read_pickle"
    ]:

        return (
            "WRITER_OR_HELPER_CONSUMES_PICKLE_"
            "NOT_FRESH_GENERATOR_AUTHORITY"
        )


    return (
        "WRITER_PATH_LINKED_GENERATION_SEMANTICS_INCOMPLETE"
    )


for record in upstream_writer_units:

    record[
        "classification"
    ] = writer_unit_classification(
        record
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "UPSTREAM WRITER UNIT CLASSIFICATION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in upstream_writer_units:

    print(
        record[
            "classification"
        ],
        record[
            "path"
        ],
        flush=True,
    )


    if record.get(
        "function"
    ):

        print(
            "  function:",
            record[
                "function"
            ],
            flush=True,
        )


    print(
        "  source SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )


    if record.get(
        "profile"
    ):

        print(
            "  schema:",
            record[
                "profile"
            ][
                "schema_hits"
            ],
            flush=True,
        )

        print(
            "  generation signals:",
            record[
                "profile"
            ][
                "generation_signals"
            ],
            flush=True,
        )

        print(
            "  GT calls:",
            record[
                "profile"
            ][
                "GT_calls"
            ],
            flush=True,
        )

        print(
            "  GT names:",
            record[
                "profile"
            ][
                "GT_names"
            ],
            flush=True,
        )

        print(
            "  read_pickle:",
            record[
                "profile"
            ][
                "has_read_pickle"
            ],
            flush=True,
        )


    if record[
        "source"
    ]:

        print(
            "  --- source preview ---",
            flush=True,
        )

        print(
            record[
                "source"
            ][
                :14000
            ],
            flush=True,
        )

        print(
            "  --- end preview ---",
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 26. Contract completeness
# =============================================================================

selection_contract_complete = (
    portable_selection_policy_matches_anchor
)


application_contract_complete = (
    application_core_classification
    ==
    "ACTION_APPLICATION_CORE_NOT_CANDIDATE_GENERATOR"
)


gt_blind_generator_candidates = [
    record

    for record
    in upstream_writer_units

    if (
        record[
            "classification"
        ]
        ==
        "GT_BLIND_GENERATOR_SOURCE_CANDIDATE"
    )
]


mixed_gt_writer_candidates = [
    record

    for record
    in upstream_writer_units

    if (
        record[
            "classification"
        ]
        ==
        "WRITER_CONTAINS_GENERATION_AND_GT_"
        "INNER_GT_BLIND_BLOCK_EXTRACTION_REQUIRED"
    )
]


provider_generation_contract_complete = (
    len(
        gt_blind_generator_candidates
    )
    > 0
)


# Even if a clean candidate appears here, this stage only RECOVERS authority.
# A separate exact contract freeze must still precede implementation.
implementation_authorized = (
    False
)


if provider_generation_contract_complete:

    recovery_status = (
        "SELECTION_AND_APPLICATION_FROZEN_"
        "UPSTREAM_GT_BLIND_GENERATOR_SOURCE_LOCALIZED_"
        "EXACT_PROVIDER_CONTRACT_FREEZE_REQUIRED"
    )


elif mixed_gt_writer_candidates:

    recovery_status = (
        "SELECTION_AND_APPLICATION_FROZEN_"
        "UPSTREAM_WRITER_LOCALIZED_"
        "INNER_GT_BLIND_GENERATOR_EXTRACTION_REQUIRED"
    )


elif upstream_writer_units:

    recovery_status = (
        "SELECTION_AND_APPLICATION_FROZEN_"
        "UPSTREAM_WRITER_LOCALIZED_"
        "GENERATOR_SEMANTICS_STILL_INCOMPLETE"
    )


else:

    recovery_status = (
        "SELECTION_AND_APPLICATION_FROZEN_"
        "FRESH_PROVIDER_GENERATION_SOURCE_UNRESOLVED"
    )


if provider_generation_contract_complete:

    next_stage = (
        "NOTEBOOK14_14_3E_D_C_"
        "MORPH_CONCRETE_PROVIDER_EXACT_GENERATOR_CONTRACT_FREEZE"
    )


elif mixed_gt_writer_candidates:

    next_stage = (
        "NOTEBOOK14_14_3E_D_C_"
        "MORPH_UPSTREAM_WRITER_INNER_GT_BLIND_GENERATOR_EXTRACTION"
    )


else:

    next_stage = (
        "NOTEBOOK14_14_3E_D_C_"
        "MORPH_FRESH_CANDIDATE_GENERATOR_TARGETED_RECOVERY"
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "MORPH CONTRACT COMPLETENESS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "selection contract complete:",
    selection_contract_complete,
    flush=True,
)

print(
    "graph application contract complete:",
    application_contract_complete,
    flush=True,
)

print(
    "fresh candidate generation contract complete:",
    provider_generation_contract_complete,
    flush=True,
)

print(
    "upstream writer units:",
    len(
        upstream_writer_units
    ),
    flush=True,
)

print(
    "clean GT-blind generator candidates:",
    len(
        gt_blind_generator_candidates
    ),
    flush=True,
)

print(
    "mixed generation+GT writer candidates:",
    len(
        mixed_gt_writer_candidates
    ),
    flush=True,
)

print(
    "status:",
    recovery_status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 27. No mutation / safety state
# =============================================================================

source_mutation = False

test_mutation = False

historical_oracle_semantic_content_loaded = False

upstream_historical_table_loaded = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == EXPECTED_PORTABLE_SHA
)


# =============================================================================
# 28. Formal semantic gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_a_r2_exact_file_locked":
        True,

    "D2_R3_semantic_authority_locked":
        (
            d2r3_sha
            == EXPECTED_D2R3_FILE_SHA
        ),

    "historical_structural_oracle_identity_locked":
        (
            oracle_sha
            == EXPECTED_STRUCTURAL_ORACLE_SHA
        ),

    "semantic_anchor_exact_locked":
        (
            anchor[
                "sha256"
            ]
            == EXPECTED_SEMANTIC_ANCHOR_CELL_SHA
        ),

    "frozen_scientific_policy_extracted":
        (
            frozen_policy[
                "morphology_top_k"
            ]
            == 3
            and
            frozen_policy[
                "association_rank_cap"
            ]
            == 3
            and
            abs(
                float(
                    frozen_policy[
                        "association_prob_threshold"
                    ]
                )
                -
                0.10
            )
            <= 1e-15
        ),

    "portable_selection_layer_exact_locked":
        (
            validate_sha
            == EXPECTED_VALIDATE_TABLE_SHA
            and
            select_sha
            == EXPECTED_SELECT_SHA
            and
            apply_sha
            == EXPECTED_APPLY_MORPH_SHA
        ),

    "portable_selection_matches_historical_policy":
        selection_contract_complete,

    "D2_R3_application_core_exact_locked":
        (
            core[
                "source_sha256"
            ]
            == EXPECTED_APPLICATION_CORE_FUNCTION_SHA
        ),

    "D2_R3_core_correctly_classified_as_application_not_generator":
        application_contract_complete,

    "upstream_writer_scan_complete":
        True,

    "fresh_generator_not_invented":
        True,

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "historical_Morph_oracle_not_semantically_loaded":
        (
            historical_oracle_semantic_content_loaded
            is False
        ),

    "upstream_historical_action_table_not_loaded":
        (
            upstream_historical_table_loaded
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-B SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-B failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 29. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_B",

    "title": (
        "MORPH_SEMANTIC_RECONSTRUCTION_CONTRACT_"
        "AND_UPSTREAM_GENERATOR_RECOVERY"
    ),

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_SEMANTIC_RECONSTRUCTION_AUTHORITY_EXTRACTION",

    "parent_14_3e_d_a_r2": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "observed_content_sha256":
            parent_content_sha,
    },

    "D2_R3_authority": {
        "path":
            str(
                D2R3_REL
            ),

        "file_sha256":
            d2r3_sha,

        "logical_content_sha256":
            EXPECTED_D2R3_LOGICAL_CONTENT_SHA,

        "authority_mode":
            EXPECTED_AUTHORITY_MODE,
    },

    "historical_structural_oracle": {
        "path":
            str(
                STRUCTURAL_ORACLE_REL
            ),

        "file_sha256":
            oracle_sha,

        "semantic_content_loaded":
            False,

        "runtime_use":
            False,

        "future_structural_fidelity_use":
            True,
    },

    "scientific_selection_policy": {
        "semantic_anchor": {
            "path":
                str(
                    NOTEBOOK11_REL
                ),

            "cell_index":
                SEMANTIC_ANCHOR_CELL_INDEX,

            "cell_sha256":
                anchor[
                    "sha256"
                ],
        },

        "policy":
            frozen_policy,

        "candidate_schema":
            list(
                EXPECTED_CANDIDATE_COLUMNS
            ),

        "portable_validate_table_sha256":
            validate_sha,

        "portable_selection_sha256":
            select_sha,

        "historical_to_portable_policy_parity":
            selection_contract_complete,
    },

    "graph_application_contract": {
        "selected_semantic_core": {
            "cell_index":
                core[
                    "cell_index"
                ],

            "function":
                core[
                    "function"
                ],

            "function_sha256":
                core[
                    "source_sha256"
                ],

            "classification":
                application_core_classification,
        },

        "contract":
            application_contract,

        "candidate_generation_authority":
            False,
    },

    "upstream_generator_recovery": {
        "target_artifact_basename":
            UPSTREAM_ACTION_BASENAME,

        "historical_table_loaded":
            False,

        "notebook_records":
            notebook_upstream_records,

        "script_records":
            script_upstream_records,

        "writer_units":
            upstream_writer_units,

        "GT_blind_generator_candidates":
            gt_blind_generator_candidates,

        "mixed_GT_writer_candidates":
            mixed_gt_writer_candidates,

        "fresh_generation_contract_complete":
            provider_generation_contract_complete,
    },

    "resolution": {
        "selection_contract_complete":
            selection_contract_complete,

        "application_contract_complete":
            application_contract_complete,

        "provider_generation_contract_complete":
            provider_generation_contract_complete,

        "status":
            recovery_status,

        "implementation_authorized":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_selection_policy":
            "FROZEN",

        "Morph_graph_application_semantics":
            "FROZEN",

        "D2_R3_application_core_role":
            "ACTION_APPLICATION_NOT_CANDIDATE_GENERATOR",

        "fresh_Morph_candidate_generation_authority":
            recovery_status,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_BY_THIS_STAGE",

        "historical_Morph_action_runtime_use":
            False,

        "historical_Morph_action_semantic_content_loaded":
            False,

        "upstream_policy_grid_action_runtime_use":
            False,

        "upstream_policy_grid_action_semantic_content_loaded":
            False,

        "fresh_end_to_end_post_V2_runtime":
            "NOT_YET_READY",

        "post_V2_structural_replay":
            "NOT_YET_READY",

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,
    },

    "explicit_non_claims": [
        "No historical exact Morph writer source was recovered.",
        "The D2-R3 selected semantic core is graph-action application authority, not fresh candidate-generation authority.",
        "No concrete MorphCandidateProvider was implemented.",
        "No historical Morph action table was loaded.",
        "No upstream policy-grid action table was loaded.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-B artifact exists but differs "
        "from the exact semantic-reconstruction state."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 30. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-B PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "AUTHORITY MODEL:",
    flush=True,
)

print(
    "  exact historical Morph writer:",
    "ABSENT",
    flush=True,
)

print(
    "  authority:",
    EXPECTED_AUTHORITY_MODE,
    flush=True,
)

print(
    flush=True,
)


print(
    "SELECTION POLICY:",
    flush=True,
)

print(
    "  Frozen-V9 W4-capable source:",
    True,
    flush=True,
)

print(
    "  source outdegree:",
    1,
    flush=True,
)

print(
    "  morphology top-k:",
    frozen_policy[
        "morphology_top_k"
    ],
    flush=True,
)

print(
    "  association rank cap:",
    frozen_policy[
        "association_rank_cap"
    ],
    flush=True,
)

print(
    "  association probability min:",
    frozen_policy[
        "association_prob_threshold"
    ],
    flush=True,
)

print(
    "  portable parity:",
    selection_contract_complete,
    flush=True,
)

print(
    flush=True,
)


print(
    "GRAPH APPLICATION:",
    flush=True,
)

print(
    "  D2-R3 core:",
    application_core_classification,
    flush=True,
)

print(
    "  source outdegree:",
    "1 -> 2",
    flush=True,
)

print(
    "  target indegree:",
    "0 -> 1",
    flush=True,
)

print(
    "  duplicate edge:",
    "FORBIDDEN",
    flush=True,
)

print(
    flush=True,
)


print(
    "FRESH CANDIDATE GENERATION:",
    flush=True,
)

print(
    "  upstream writer units:",
    len(
        upstream_writer_units
    ),
    flush=True,
)

print(
    "  clean GT-blind generator candidates:",
    len(
        gt_blind_generator_candidates
    ),
    flush=True,
)

print(
    "  generation contract complete:",
    provider_generation_contract_complete,
    flush=True,
)

print(
    "  status:",
    recovery_status,
    flush=True,
)

print(
    "  implementation authorized:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  historical Morph oracle loaded:",
    False,
    flush=True,
)

print(
    "  upstream historical action table loaded:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  source mutated:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-C
# MORPH FRESH CANDIDATE GENERATOR TARGETED RECOVERY
#
# PARENT
# ------
# 14.3E-D-B PASS
#
# Frozen:
#
#   selection policy          = COMPLETE
#   graph application         = COMPLETE
#   fresh candidate generator = UNRESOLVED
#
#
# IMPORTANT CHANGE IN SEARCH STRATEGY
# -----------------------------------
# Exact artifact-writer searches are CLOSED:
#
#   historical Morph action artifact:
#       exact writer absent
#
#   upstream policy-grid action artifact:
#       exact writer absent in current checkout
#
# Therefore this stage searches for SEMANTIC CONSTRUCTION authority:
#
#   - code that CREATES morph_rank
#   - code that CREATES candidate_rank
#   - code that CREATES candidate_prob
#   - code that CREATES source/target candidate rows
#   - association computation/ranking
#   - temporal morphology ranking
#
# It explicitly distinguishes these from code that merely:
#
#   - reads existing action tables
#   - sorts existing rows
#   - applies edges
#   - evaluates GT
#
#
# TARGET
# ------
# Recover one of:
#
#   A. clean GT-blind candidate-generation source block
#
#   B. mixed generation+GT cell with extractable GT-blind inner block
#
#   C. partial source semantics sufficient for a specification-derived
#      reconstruction contract
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical pickle loading
#   - GT loading
#   - evaluator execution
#   - performance computation
#   - parameter search
#   - provider implementation
#   - structural fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import builtins
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_b_"
    "morph_semantic_reconstruction_contract_and_"
    "upstream_generator_recovery_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


NOTEBOOK11_REL = Path(
    "notebooks/"
    "11_division_association_audit.ipynb"
)

NOTEBOOK11 = ROOT / NOTEBOOK11_REL


D2R3_REL = Path(
    "reports/"
    "notebook13_13_3d2r3_"
    "morph_authority_correction_and_"
    "semantic_reconstruction_freeze_v1.json"
)

D2R3 = ROOT / D2R3_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_c_"
    "morph_fresh_candidate_generator_"
    "targeted_recovery_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "af7393fe0fe8b1421bfabc3a45dcdfb35bb43f4698ef3f4c90c26e2fa454243d"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "e962ac1afcf2d899a2dbcdd3e1d29188bf955d7a7d25fad66ee0d8bcc8a8b8f9"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_D2R3_FILE_SHA = (
    "ac0cb4959b8a9cb54784d0fd37d4da89a87b6e110298f67af00e712deebf8afb"
)


EXPECTED_AUTHORITY_MODE = (
    "SEMANTIC_SOURCE_LOCK_PLUS_"
    "HISTORICAL_GT_BLIND_STRUCTURAL_ORACLE"
)


EXPECTED_SEMANTIC_ANCHOR_CELL = 63

EXPECTED_SEMANTIC_ANCHOR_SHA = (
    "2e6bdff585ad1a3ee61b3a82ec03ccbd05db0b66f01024a35a721d76a29d8834"
)


EXPECTED_CONFIG_ID = (
    "K3_R3_P100"
)


EXPECTED_SCHEMA = (
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
)


CORE_GENERATED_FIELDS = {
    "morph_rank",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
}


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(value: bytes) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(value: str) -> str:

    return sha256_bytes(
        value.encode("utf-8")
    )


def sha256_file(path: Path) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value: Any) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_json(path: Path) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def safe_rel(path: Path) -> str:

    try:

        return str(
            path.resolve().relative_to(
                ROOT
            )
        )

    except Exception:

        return str(path)


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    source = ast.get_source_segment(
        text,
        node,
    )


    assert (
        source is not None
    )


    return source


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[dict[str, Any]]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in obj.items():

            result.extend(
                flatten_json(
                    value,
                    prefix=f"{prefix}.{key}",
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=f"{prefix}[{index}]",
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:

    return any(
        record[
            "value"
        ]
        == target

        for record in flatten_json(
            obj
        )
    )


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    ), path


    actual_file_sha = sha256_file(
        path
    )


    assert (
        actual_file_sha
        == EXPECTED_PARENT_FILE_SHA
    ), (
        "14.3E-D-B parent file SHA drift.\n"
        f"expected={EXPECTED_PARENT_FILE_SHA}\n"
        f"actual={actual_file_sha}"
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "resolution"
        ][
            "selection_contract_complete"
        ]
        is True
    )


    assert (
        obj[
            "resolution"
        ][
            "application_contract_complete"
        ]
        is True
    )


    assert (
        obj[
            "resolution"
        ][
            "provider_generation_contract_complete"
        ]
        is False
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "historical_Morph_action_runtime_use"
        ]
        is False
    )


    return obj


# =============================================================================
# 4. Notebook loader
# =============================================================================

def notebook_cells(
    path: Path,
) -> list[dict[str, Any]]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    result = []


    for index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        result.append({
            "index":
                index,

            "cell_type":
                cell.get(
                    "cell_type"
                ),

            "source":
                source,

            "sha256":
                sha256_text(
                    source
                ),
        })


    return result


# =============================================================================
# 5. AST semantic profile
# =============================================================================

def dict_literal_keys(
    node: ast.AST,
) -> set[str]:

    result = set()


    for child in ast.walk(
        node
    ):

        if not isinstance(
            child,
            ast.Dict,
        ):

            continue


        for key_node in child.keys:

            if key_node is None:

                continue


            try:

                key = ast.literal_eval(
                    key_node
                )

            except Exception:

                continue


            if isinstance(
                key,
                str,
            ):

                result.add(
                    key
                )


    return result


def assigned_subscript_columns(
    node: ast.AST,
) -> set[str]:

    result = set()


    for stmt in ast.walk(
        node
    ):

        if not isinstance(
            stmt,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
            ),
        ):

            continue


        if isinstance(
            stmt,
            ast.Assign,
        ):

            targets = list(
                stmt.targets
            )


        else:

            targets = [
                stmt.target
            ]


        for target in targets:

            for child in ast.walk(
                target
            ):

                if not isinstance(
                    child,
                    ast.Subscript,
                ):

                    continue


                try:

                    key = ast.literal_eval(
                        child.slice
                    )

                except Exception:

                    continue


                if isinstance(
                    key,
                    str,
                ):

                    result.add(
                        key
                    )


    return result


def dataframe_columns_arguments(
    node: ast.AST,
) -> set[str]:

    """
    Extract literal columns from calls such as:
        pd.DataFrame(..., columns=[...])
    """

    result = set()


    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if (
            name is None
            or
            not name.endswith(
                "DataFrame"
            )
        ):

            continue


        for kw in call.keywords:

            if kw.arg != "columns":

                continue


            try:

                value = ast.literal_eval(
                    kw.value
                )

            except Exception:

                continue


            if isinstance(
                value,
                (
                    list,
                    tuple,
                ),
            ):

                for item in value:

                    if isinstance(
                        item,
                        str,
                    ):

                        result.add(
                            item
                        )


    return result


def function_calls(
    node: ast.AST,
) -> set[str]:

    result = set()


    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if name:

            result.add(
                name
            )


    return result


def ast_names(
    node: ast.AST,
) -> set[str]:

    return {
        child.id

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Name,
        )
    }


def read_pickle_calls(
    node: ast.AST,
) -> list[str]:

    result = []


    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if (
            name is not None
            and
            name.endswith(
                "read_pickle"
            )
        ):

            result.append(
                ast.unparse(
                    call
                )
            )


    return result


GT_CALL_TOKENS = (
    "evaluate_graph",
    "_evaluate_matched_graph",
    "per_sample_metrics",
    "official_evaluate",
)


def gt_hazard_profile(
    node: ast.AST,
) -> dict[str, Any]:

    calls = function_calls(
        node
    )


    names = ast_names(
        node
    )


    gt_calls = sorted({
        call

        for call in calls

        if any(
            token
            in call.lower()

            for token in GT_CALL_TOKENS
        )
    })


    gt_names = sorted({
        name

        for name in names

        if (
            name.lower()
            in {
                "gt",
                "gt_graph",
                "gt_base",
                "gt_policy",
                "gt_nodes",
                "gt_edges",
                "ground_truth",
            }
            or
            name.lower().startswith(
                "gt_"
            )
        )
    })


    return {
        "GT_calls":
            gt_calls,

        "GT_names":
            gt_names,

        "has_GT_hazard":
            bool(
                gt_calls
                or
                gt_names
            ),
    }


def semantic_profile(
    source: str,
) -> dict[str, Any]:

    try:

        tree = ast.parse(
            source
        )

    except Exception as exc:

        return {
            "parseable":
                False,

            "parse_error":
                repr(
                    exc
                ),
        }


    dict_keys = dict_literal_keys(
        tree
    )


    assigned_columns = assigned_subscript_columns(
        tree
    )


    dataframe_columns = dataframe_columns_arguments(
        tree
    )


    constructed_fields = (
        dict_keys
        |
        assigned_columns
        |
        dataframe_columns
    )


    core_construction_hits = sorted(
        CORE_GENERATED_FIELDS
        &
        constructed_fields
    )


    schema_construction_hits = sorted(
        set(
            EXPECTED_SCHEMA
        )
        &
        constructed_fields
    )


    calls = sorted(
        function_calls(
            tree
        )
    )


    names = sorted(
        ast_names(
            tree
        )
    )


    pickle_reads = read_pickle_calls(
        tree
    )


    gt = gt_hazard_profile(
        tree
    )


    lower = source.lower()


    semantic_tokens = sorted({
        token

        for token in (
            "morph",
            "morph_rank",
            "candidate_rank",
            "candidate_prob",
            "association",
            "temporal",
            "rank",
            "top",
            "w4",
            "outdegree",
            "indegree",
        )

        if token
        in lower
    })


    ranking_calls = sorted({
        call

        for call in calls

        if any(
            token
            in call.lower()

            for token in (
                "rank",
                "sort_values",
                "nlargest",
                "nsmallest",
                "groupby",
                "cumcount",
                "argsort",
                "argmax",
            )
        )
    })


    association_calls = sorted({
        call

        for call in calls

        if any(
            token
            in call.lower()

            for token in (
                "association",
                "predict_proba",
                "prob",
                "score",
            )
        )
    })


    config_hit = (
        EXPECTED_CONFIG_ID
        in source
    )


    construction_strength = (
        len(
            core_construction_hits
        )
    )


    # This is intentionally strict.
    #
    # Reading actions["candidate_prob"] does NOT count.
    # Only literal dict creation / assigned columns / DataFrame columns count.
    direct_generator_evidence = (
        {
            "candidate_prob",
            "candidate_rank",
            "morph_rank",
        }
        <=
        constructed_fields
        and
        len(
            (
                {
                    "source_pred_node_id",
                    "target_pred_node_id",
                }
                &
                constructed_fields
            )
        )
        >= 1
    )


    likely_generator_evidence = (
        construction_strength
        >= 3
        and
        (
            "association"
            in lower
            or
            bool(
                ranking_calls
            )
        )
    )


    return {
        "parseable":
            True,

        "dict_keys":
            sorted(
                dict_keys
            ),

        "assigned_columns":
            sorted(
                assigned_columns
            ),

        "dataframe_columns":
            sorted(
                dataframe_columns
            ),

        "constructed_fields":
            sorted(
                constructed_fields
            ),

        "core_construction_hits":
            core_construction_hits,

        "schema_construction_hits":
            schema_construction_hits,

        "construction_strength":
            construction_strength,

        "calls":
            calls,

        "ranking_calls":
            ranking_calls,

        "association_calls":
            association_calls,

        "names":
            names,

        "read_pickle_calls":
            pickle_reads,

        "has_read_pickle":
            bool(
                pickle_reads
            ),

        "GT_calls":
            gt[
                "GT_calls"
            ],

        "GT_names":
            gt[
                "GT_names"
            ],

        "has_GT_hazard":
            gt[
                "has_GT_hazard"
            ],

        "semantic_tokens":
            semantic_tokens,

        "config_hit":
            config_hit,

        "direct_generator_evidence":
            direct_generator_evidence,

        "likely_generator_evidence":
            likely_generator_evidence,
    }


# =============================================================================
# 6. Notebook definition / dependency closure
# =============================================================================

BUILTIN_NAMES = set(
    dir(
        builtins
    )
)


IGNORED_NAMES = (
    BUILTIN_NAMES
    |
    {
        "np",
        "pd",
        "json",
        "Path",
        "hashlib",
        "sys",
        "math",
        "re",
        "os",
        "plt",
        "sns",
    }
)


def cell_defined_names(
    source: str,
) -> set[str]:

    try:

        tree = ast.parse(
            source
        )

    except Exception:

        return set()


    result = set()


    for stmt in tree.body:

        if isinstance(
            stmt,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):

            result.add(
                stmt.name
            )


        elif isinstance(
            stmt,
            ast.Assign,
        ):

            for target in stmt.targets:

                for child in ast.walk(
                    target
                ):

                    if isinstance(
                        child,
                        ast.Name,
                    ):

                        result.add(
                            child.id
                        )


        elif isinstance(
            stmt,
            ast.AnnAssign,
        ):

            for child in ast.walk(
                stmt.target
            ):

                if isinstance(
                    child,
                    ast.Name,
                ):

                    result.add(
                        child.id
                    )


        elif isinstance(
            stmt,
            ast.Import,
        ):

            for alias in stmt.names:

                result.add(
                    alias.asname
                    or
                    alias.name.split(
                        "."
                    )[0]
                )


        elif isinstance(
            stmt,
            ast.ImportFrom,
        ):

            for alias in stmt.names:

                result.add(
                    alias.asname
                    or
                    alias.name
                )


    return result


def cell_used_names(
    source: str,
) -> set[str]:

    try:

        tree = ast.parse(
            source
        )

    except Exception:

        return set()


    loads = {
        node.id

        for node in ast.walk(
            tree
        )

        if (
            isinstance(
                node,
                ast.Name,
            )
            and
            isinstance(
                node.ctx,
                ast.Load,
            )
        )
    }


    return (
        loads
        -
        cell_defined_names(
            source
        )
        -
        IGNORED_NAMES
    )


def build_definition_index(
    cells: list[dict[str, Any]],
) -> dict[
    str,
    list[int],
]:

    index: dict[
        str,
        list[int]
    ] = {}


    for cell in cells:

        if (
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        for name in cell_defined_names(
            cell[
                "source"
            ]
        ):

            index.setdefault(
                name,
                [],
            ).append(
                cell[
                    "index"
                ]
            )


    return index


def latest_definition_before(
    definition_index: dict[
        str,
        list[int],
    ],
    name: str,
    before_cell: int,
) -> int | None:

    candidates = [
        index

        for index in definition_index.get(
            name,
            [],
        )

        if index
        <
        before_cell
    ]


    if not candidates:

        return None


    return max(
        candidates
    )


def backward_dependency_closure(
    cells: list[dict[str, Any]],
    definition_index: dict[
        str,
        list[int],
    ],
    root_cell: int,
    *,
    max_cells: int = 80,
) -> dict[str, Any]:

    selected = {
        root_cell
    }


    queue = [
        root_cell
    ]


    edges = []


    unresolved = set()


    while (
        queue
        and
        len(
            selected
        )
        <=
        max_cells
    ):

        current = queue.pop(
            0
        )


        source = cells[
            current
        ][
            "source"
        ]


        used = cell_used_names(
            source
        )


        for name in sorted(
            used
        ):

            definition_cell = (
                latest_definition_before(
                    definition_index,
                    name,
                    current,
                )
            )


            if definition_cell is None:

                unresolved.add(
                    name
                )

                continue


            edges.append({
                "name":
                    name,

                "from_cell":
                    current,

                "definition_cell":
                    definition_cell,
            })


            if (
                definition_cell
                not in selected
            ):

                selected.add(
                    definition_cell
                )

                queue.append(
                    definition_cell
                )


    ordered_cells = sorted(
        selected
    )


    return {
        "root_cell":
            root_cell,

        "cells":
            ordered_cells,

        "edges":
            edges,

        "unresolved_names":
            sorted(
                unresolved
            ),

        "truncated":
            (
                len(
                    selected
                )
                >
                max_cells
            ),
    }


# =============================================================================
# 7. Notebook candidate-unit scan
# =============================================================================

def notebook_candidate_units(
    cells: list[dict[str, Any]],
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        cell_profile = semantic_profile(
            source
        )


        if (
            cell_profile.get(
                "parseable"
            )
            is True
        ):

            semantic_relevant = (
                cell_profile[
                    "direct_generator_evidence"
                ]
                or
                cell_profile[
                    "likely_generator_evidence"
                ]
                or
                (
                    len(
                        cell_profile[
                            "semantic_tokens"
                        ]
                    )
                    >= 4
                    and
                    (
                        "candidate_prob"
                        in source
                        or
                        "morph_rank"
                        in source
                    )
                )
            )


            if semantic_relevant:

                records.append({
                    "kind":
                        "NOTEBOOK_CELL",

                    "path":
                        str(
                            NOTEBOOK11_REL
                        ),

                    "cell_index":
                        cell[
                            "index"
                        ],

                    "cell_sha256":
                        cell[
                            "sha256"
                        ],

                    "unit_name":
                        f"cell_{cell['index']}",

                    "profile":
                        cell_profile,

                    "source":
                        source,
                })


        # Also inspect every function inside the cell.
        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for node in tree.body:

            if not isinstance(
                node,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                ),
            ):

                continue


            fn_source = node_source(
                source,
                node,
            )


            profile = semantic_profile(
                fn_source
            )


            if (
                profile.get(
                    "parseable"
                )
                is not True
            ):

                continue


            semantic_relevant = (
                profile[
                    "direct_generator_evidence"
                ]
                or
                profile[
                    "likely_generator_evidence"
                ]
                or
                (
                    len(
                        profile[
                            "semantic_tokens"
                        ]
                    )
                    >= 4
                    and
                    (
                        "candidate_prob"
                        in fn_source
                        or
                        "morph_rank"
                        in fn_source
                    )
                )
            )


            if not semantic_relevant:

                continue


            records.append({
                "kind":
                    "NOTEBOOK_FUNCTION",

                "path":
                    str(
                        NOTEBOOK11_REL
                    ),

                "cell_index":
                    cell[
                        "index"
                    ],

                "cell_sha256":
                    cell[
                        "sha256"
                    ],

                "unit_name":
                    node.name,

                "function_sha256":
                    sha256_text(
                        fn_source
                    ),

                "profile":
                    profile,

                "source":
                    fn_source,
            })


    return records


# =============================================================================
# 8. Repository Python function scan
# =============================================================================

def repository_python_candidate_units() -> list[
    dict[str, Any]
]:

    records = []


    scan_roots = [
        ROOT / "src",
        ROOT / "scripts",
    ]


    for scan_root in scan_roots:

        if not scan_root.exists():

            continue


        for path in sorted(
            scan_root.rglob(
                "*.py"
            )
        ):

            try:

                text = path.read_text(
                    encoding="utf-8"
                )


                tree = ast.parse(
                    text,
                    filename=str(
                        path
                    ),
                )

            except Exception:

                continue


            for node in tree.body:

                if not isinstance(
                    node,
                    (
                        ast.FunctionDef,
                        ast.AsyncFunctionDef,
                    ),
                ):

                    continue


                source = node_source(
                    text,
                    node,
                )


                profile = semantic_profile(
                    source
                )


                if (
                    profile.get(
                        "parseable"
                    )
                    is not True
                ):

                    continue


                semantic_relevant = (
                    profile[
                        "direct_generator_evidence"
                    ]
                    or
                    profile[
                        "likely_generator_evidence"
                    ]
                )


                if not semantic_relevant:

                    continue


                records.append({
                    "kind":
                        "PYTHON_FUNCTION",

                    "path":
                        safe_rel(
                            path
                        ),

                    "file_sha256":
                        sha256_file(
                            path
                        ),

                    "unit_name":
                        node.name,

                    "function_sha256":
                        sha256_text(
                            source
                        ),

                    "line_start":
                        node.lineno,

                    "line_end":
                        node.end_lineno,

                    "profile":
                        profile,

                    "source":
                        source,
                })


    return records


# =============================================================================
# 9. Candidate scoring/classification
# =============================================================================

def candidate_score(
    record: dict[str, Any],
) -> int:

    profile = record[
        "profile"
    ]


    score = 0


    score += (
        14
        *
        len(
            profile[
                "core_construction_hits"
            ]
        )
    )


    if profile[
        "direct_generator_evidence"
    ]:

        score += 40


    if profile[
        "likely_generator_evidence"
    ]:

        score += 16


    if "association" in profile[
        "semantic_tokens"
    ]:

        score += 14


    if "morph" in profile[
        "semantic_tokens"
    ]:

        score += 8


    if profile[
        "ranking_calls"
    ]:

        score += 8


    if profile[
        "association_calls"
    ]:

        score += 8


    if profile[
        "config_hit"
    ]:

        score += 12


    if profile[
        "has_read_pickle"
    ]:

        score -= 45


    if profile[
        "has_GT_hazard"
    ]:

        score -= 35


    # Application layer cannot become candidate generator merely because
    # it knows the schema.
    if (
        record[
            "path"
        ]
        ==
        str(
            PORTABLE_REL
        )
    ):

        score -= 40


    return score


def candidate_classification(
    record: dict[str, Any],
) -> str:

    profile = record[
        "profile"
    ]


    direct = (
        profile[
            "direct_generator_evidence"
        ]
    )


    gt = (
        profile[
            "has_GT_hazard"
        ]
    )


    pickle_read = (
        profile[
            "has_read_pickle"
        ]
    )


    if (
        direct
        and
        not gt
        and
        not pickle_read
        and
        record[
            "path"
        ]
        != str(
            PORTABLE_REL
        )
    ):

        return (
            "CLEAN_FRESH_GENERATOR_SOURCE_CANDIDATE"
        )


    if (
        direct
        and
        gt
        and
        not pickle_read
    ):

        return (
            "GENERATOR_SEMANTICS_MIXED_WITH_GT_"
            "INNER_BLOCK_EXTRACTION_REQUIRED"
        )


    if (
        direct
        and
        pickle_read
    ):

        return (
            "GENERATOR_LIKE_BUT_PICKLE_DEPENDENT_"
            "NOT_FRESH_AUTHORITY"
        )


    if (
        profile[
            "likely_generator_evidence"
        ]
    ):

        return (
            "PARTIAL_GENERATOR_SEMANTICS"
        )


    return (
        "SEMANTIC_MENTION_ONLY"
    )


# =============================================================================
# 10. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    NOTEBOOK11,
    D2R3,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-C "
    "MORPH FRESH CANDIDATE GENERATOR TARGETED RECOVERY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "NOTEBOOK11:",
    NOTEBOOK11_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Parent lock
# =============================================================================

parent = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_D_B_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  selection contract:",
    parent[
        "resolution"
    ][
        "selection_contract_complete"
    ],
    flush=True,
)

print(
    "  application contract:",
    parent[
        "resolution"
    ][
        "application_contract_complete"
    ],
    flush=True,
)

print(
    "  generation contract:",
    parent[
        "resolution"
    ][
        "provider_generation_contract_complete"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Current source / authority locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


d2r3_sha = sha256_file(
    D2R3
)


assert (
    d2r3_sha
    == EXPECTED_D2R3_FILE_SHA
)


d2r3 = load_json(
    D2R3
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_AUTHORITY_MODE,
    )
)


cells = notebook_cells(
    NOTEBOOK11
)


assert (
    cells[
        EXPECTED_SEMANTIC_ANCHOR_CELL
    ][
        "sha256"
    ]
    == EXPECTED_SEMANTIC_ANCHOR_SHA
)


print(
    "[PASS] CURRENT_AUTHORITY_LOCKS",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  D2-R3 SHA:",
    d2r3_sha,
    flush=True,
)

print(
    "  semantic anchor cell SHA:",
    EXPECTED_SEMANTIC_ANCHOR_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Notebook11 semantic candidate scan
# =============================================================================

notebook_units = (
    notebook_candidate_units(
        cells
    )
)


for record in notebook_units:

    record[
        "score"
    ] = candidate_score(
        record
    )


    record[
        "classification"
    ] = candidate_classification(
        record
    )


notebook_units = sorted(
    notebook_units,
    key=lambda record: (
        -record[
            "score"
        ],
        record[
            "cell_index"
        ],
        record[
            "unit_name"
        ],
    ),
)


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK11 MORPH GENERATION SEMANTIC CANDIDATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "candidate units:",
    len(
        notebook_units
    ),
    flush=True,
)

print(
    flush=True,
)


for record in notebook_units[
    :30
]:

    print(
        f"SCORE={record['score']:>4}",
        record[
            "classification"
        ],
        f"cell={record['cell_index']}",
        record[
            "unit_name"
        ],
        flush=True,
    )

    print(
        "  cell SHA:",
        record[
            "cell_sha256"
        ],
        flush=True,
    )


    if record.get(
        "function_sha256"
    ):

        print(
            "  function SHA:",
            record[
                "function_sha256"
            ],
            flush=True,
        )


    profile = record[
        "profile"
    ]


    print(
        "  constructed:",
        profile[
            "core_construction_hits"
        ],
        flush=True,
    )

    print(
        "  direct generator:",
        profile[
            "direct_generator_evidence"
        ],
        flush=True,
    )

    print(
        "  ranking calls:",
        profile[
            "ranking_calls"
        ],
        flush=True,
    )

    print(
        "  association calls:",
        profile[
            "association_calls"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        profile[
            "has_read_pickle"
        ],
        flush=True,
    )

    print(
        "  GT hazard:",
        profile[
            "has_GT_hazard"
        ],
        flush=True,
    )

    print(
        "  config hit:",
        profile[
            "config_hit"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 14. Repository Python semantic scan
# =============================================================================

python_units = (
    repository_python_candidate_units()
)


for record in python_units:

    record[
        "score"
    ] = candidate_score(
        record
    )


    record[
        "classification"
    ] = candidate_classification(
        record
    )


python_units = sorted(
    python_units,
    key=lambda record: (
        -record[
            "score"
        ],
        record[
            "path"
        ],
        record[
            "unit_name"
        ],
    ),
)


print(
    "=" * 120,
    flush=True,
)

print(
    "REPOSITORY PYTHON MORPH GENERATION CANDIDATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "candidate functions:",
    len(
        python_units
    ),
    flush=True,
)

print(
    flush=True,
)


for record in python_units[
    :30
]:

    print(
        f"SCORE={record['score']:>4}",
        record[
            "classification"
        ],
        record[
            "path"
        ],
        "::",
        record[
            "unit_name"
        ],
        flush=True,
    )

    print(
        "  function SHA:",
        record[
            "function_sha256"
        ],
        flush=True,
    )

    print(
        "  constructed:",
        record[
            "profile"
        ][
            "core_construction_hits"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        record[
            "profile"
        ][
            "has_read_pickle"
        ],
        flush=True,
    )

    print(
        "  GT hazard:",
        record[
            "profile"
        ][
            "has_GT_hazard"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 15. Clean/mixed/partial candidate sets
# =============================================================================

all_units = (
    notebook_units
    +
    python_units
)


clean_candidates = [
    record

    for record in all_units

    if (
        record[
            "classification"
        ]
        ==
        "CLEAN_FRESH_GENERATOR_SOURCE_CANDIDATE"
    )
]


mixed_gt_candidates = [
    record

    for record in all_units

    if (
        record[
            "classification"
        ]
        ==
        "GENERATOR_SEMANTICS_MIXED_WITH_GT_"
        "INNER_BLOCK_EXTRACTION_REQUIRED"
    )
]


pickle_dependent_candidates = [
    record

    for record in all_units

    if (
        record[
            "classification"
        ]
        ==
        "GENERATOR_LIKE_BUT_PICKLE_DEPENDENT_"
        "NOT_FRESH_AUTHORITY"
    )
]


partial_candidates = [
    record

    for record in all_units

    if (
        record[
            "classification"
        ]
        ==
        "PARTIAL_GENERATOR_SEMANTICS"
    )
]


print(
    "=" * 120,
    flush=True,
)

print(
    "CANDIDATE CLASSIFICATION SUMMARY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "clean fresh source candidates:",
    len(
        clean_candidates
    ),
    flush=True,
)

print(
    "mixed GT source candidates:",
    len(
        mixed_gt_candidates
    ),
    flush=True,
)

print(
    "pickle-dependent generator-like units:",
    len(
        pickle_dependent_candidates
    ),
    flush=True,
)

print(
    "partial generator semantics:",
    len(
        partial_candidates
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Dependency closure for top Notebook11 candidates
# =============================================================================

definition_index = (
    build_definition_index(
        cells
    )
)


notebook_root_candidates = [
    record

    for record in notebook_units

    if record[
        "classification"
    ]
    in {
        "CLEAN_FRESH_GENERATOR_SOURCE_CANDIDATE",
        "GENERATOR_SEMANTICS_MIXED_WITH_GT_"
        "INNER_BLOCK_EXTRACTION_REQUIRED",
        "PARTIAL_GENERATOR_SEMANTICS",
    }
]


closure_records = []


for record in notebook_root_candidates[
    :12
]:

    root_cell = record[
        "cell_index"
    ]


    closure = backward_dependency_closure(
        cells,
        definition_index,
        root_cell,
    )


    closure_cell_profiles = []


    for cell_index in closure[
        "cells"
    ]:

        source = cells[
            cell_index
        ][
            "source"
        ]


        profile = semantic_profile(
            source
        )


        closure_cell_profiles.append({
            "cell_index":
                cell_index,

            "cell_sha256":
                cells[
                    cell_index
                ][
                    "sha256"
                ],

            "has_read_pickle":
                (
                    profile.get(
                        "has_read_pickle"
                    )
                    if profile.get(
                        "parseable"
                    )
                    else None
                ),

            "has_GT_hazard":
                (
                    profile.get(
                        "has_GT_hazard"
                    )
                    if profile.get(
                        "parseable"
                    )
                    else None
                ),

            "core_construction_hits":
                (
                    profile.get(
                        "core_construction_hits",
                        [],
                    )
                ),

            "semantic_tokens":
                (
                    profile.get(
                        "semantic_tokens",
                        [],
                    )
                ),
        })


    closure_has_gt = any(
        item[
            "has_GT_hazard"
        ]
        is True

        for item in closure_cell_profiles
    )


    closure_has_pickle = any(
        item[
            "has_read_pickle"
        ]
        is True

        for item in closure_cell_profiles
    )


    closure_records.append({
        "root_kind":
            record[
                "kind"
            ],

        "root_cell":
            root_cell,

        "root_name":
            record[
                "unit_name"
            ],

        "root_classification":
            record[
                "classification"
            ],

        "root_score":
            record[
                "score"
            ],

        "dependency_closure":
            closure,

        "closure_profiles":
            closure_cell_profiles,

        "closure_has_GT_hazard":
            closure_has_gt,

        "closure_has_read_pickle":
            closure_has_pickle,

        "closure_candidate_GT_blind":
            (
                not closure_has_gt
                and
                not closure_has_pickle
            ),
    })


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK DEPENDENCY CLOSURES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in closure_records:

    print(
        "root:",
        record[
            "root_cell"
        ],
        record[
            "root_name"
        ],
        "|",
        record[
            "root_classification"
        ],
        flush=True,
    )

    print(
        "  closure cells:",
        record[
            "dependency_closure"
        ][
            "cells"
        ],
        flush=True,
    )

    print(
        "  GT hazard:",
        record[
            "closure_has_GT_hazard"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        record[
            "closure_has_read_pickle"
        ],
        flush=True,
    )

    print(
        "  candidate GT-blind closure:",
        record[
            "closure_candidate_GT_blind"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 17. Print source for strongest actionable candidates
# =============================================================================

actionable_candidates = sorted(
    (
        clean_candidates
        +
        mixed_gt_candidates
        +
        partial_candidates
    ),
    key=lambda record: (
        -record[
            "score"
        ],
        record[
            "path"
        ],
        record.get(
            "cell_index",
            -1,
        ),
        record[
            "unit_name"
        ],
    ),
)


print(
    "=" * 120,
    flush=True,
)

print(
    "STRONGEST ACTIONABLE SOURCE BLOCKS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for index, record in enumerate(
    actionable_candidates[
        :12
    ],
    start=1,
):

    print(
        f"--- CANDIDATE {index} ---",
        flush=True,
    )

    print(
        "classification:",
        record[
            "classification"
        ],
        flush=True,
    )

    print(
        "path:",
        record[
            "path"
        ],
        flush=True,
    )


    if "cell_index" in record:

        print(
            "cell:",
            record[
                "cell_index"
            ],
            flush=True,
        )


    print(
        "unit:",
        record[
            "unit_name"
        ],
        flush=True,
    )

    print(
        "score:",
        record[
            "score"
        ],
        flush=True,
    )

    print(
        "constructed:",
        record[
            "profile"
        ][
            "core_construction_hits"
        ],
        flush=True,
    )

    print(
        "GT hazard:",
        record[
            "profile"
        ][
            "has_GT_hazard"
        ],
        flush=True,
    )

    print(
        "read_pickle:",
        record[
            "profile"
        ][
            "has_read_pickle"
        ],
        flush=True,
    )

    print(
        "source SHA:",
        (
            record.get(
                "function_sha256"
            )
            or
            record.get(
                "cell_sha256"
            )
        ),
        flush=True,
    )

    print(
        "----- SOURCE -----",
        flush=True,
    )

    print(
        record[
            "source"
        ][
            :16000
        ],
        flush=True,
    )

    print(
        "----- END SOURCE -----",
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 18. Recovery classification
# =============================================================================

clean_closure_candidates = [
    record

    for record in closure_records

    if (
        record[
            "root_classification"
        ]
        ==
        "CLEAN_FRESH_GENERATOR_SOURCE_CANDIDATE"
        and
        record[
            "closure_candidate_GT_blind"
        ]
    )
]


if clean_closure_candidates:

    recovery_status = (
        "FRESH_GT_BLIND_GENERATOR_SOURCE_AND_"
        "DEPENDENCY_CLOSURE_LOCALIZED_"
        "EXACT_CONTRACT_FREEZE_REQUIRED"
    )


elif clean_candidates:

    recovery_status = (
        "CLEAN_GENERATOR_SOURCE_LOCALIZED_"
        "DEPENDENCY_CLOSURE_NOT_YET_GT_BLIND"
    )


elif mixed_gt_candidates:

    recovery_status = (
        "GENERATOR_SEMANTICS_LOCALIZED_IN_MIXED_GT_SOURCE_"
        "INNER_BLOCK_EXTRACTION_REQUIRED"
    )


elif partial_candidates:

    recovery_status = (
        "PARTIAL_GENERATOR_SEMANTICS_LOCALIZED_"
        "SPECIFICATION_CLOSURE_REQUIRED"
    )


else:

    recovery_status = (
        "NO_CONCRETE_GENERATOR_SOURCE_RECOVERED_"
        "SPECIFICATION_DERIVED_RECONSTRUCTION_REQUIRED"
    )


implementation_authorized = (
    False
)


if (
    recovery_status
    ==
    "FRESH_GT_BLIND_GENERATOR_SOURCE_AND_"
    "DEPENDENCY_CLOSURE_LOCALIZED_"
    "EXACT_CONTRACT_FREEZE_REQUIRED"
):

    next_stage = (
        "NOTEBOOK14_14_3E_D_D_"
        "MORPH_FRESH_GENERATOR_EXACT_CONTRACT_FREEZE"
    )


elif (
    recovery_status
    ==
    "GENERATOR_SEMANTICS_LOCALIZED_IN_MIXED_GT_SOURCE_"
    "INNER_BLOCK_EXTRACTION_REQUIRED"
):

    next_stage = (
        "NOTEBOOK14_14_3E_D_D_"
        "MORPH_GT_BLIND_INNER_GENERATOR_BLOCK_EXTRACTION"
    )


elif (
    recovery_status
    ==
    "PARTIAL_GENERATOR_SEMANTICS_LOCALIZED_"
    "SPECIFICATION_CLOSURE_REQUIRED"
):

    next_stage = (
        "NOTEBOOK14_14_3E_D_D_"
        "MORPH_GENERATOR_SPECIFICATION_CLOSURE"
    )


else:

    next_stage = (
        "NOTEBOOK14_14_3E_D_D_"
        "MORPH_SPECIFICATION_DERIVED_GENERATOR_"
        "RECONSTRUCTION_BOUNDARY_FREEZE"
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-C RECOVERY RESOLUTION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "clean source candidates:",
    len(
        clean_candidates
    ),
    flush=True,
)

print(
    "clean GT-blind dependency closures:",
    len(
        clean_closure_candidates
    ),
    flush=True,
)

print(
    "mixed GT candidates:",
    len(
        mixed_gt_candidates
    ),
    flush=True,
)

print(
    "partial semantic candidates:",
    len(
        partial_candidates
    ),
    flush=True,
)

print(
    "status:",
    recovery_status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Safety / no-mutation proof
# =============================================================================

source_mutation = False

test_mutation = False

historical_pickle_loaded = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


# =============================================================================
# 20. Formal gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_b_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "D2_R3_authority_locked":
        (
            d2r3_sha
            == EXPECTED_D2R3_FILE_SHA
        ),

    "semantic_anchor_locked":
        (
            cells[
                EXPECTED_SEMANTIC_ANCHOR_CELL
            ][
                "sha256"
            ]
            == EXPECTED_SEMANTIC_ANCHOR_SHA
        ),

    "notebook_semantic_scan_complete":
        True,

    "repository_python_semantic_scan_complete":
        True,

    "constructor_vs_reader_distinction_enforced":
        True,

    "dependency_closure_analysis_complete":
        True,

    "fresh_generator_not_invented":
        True,

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-C SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-C failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 21. Compact records for artifact
# =============================================================================

def compact_record(
    record: dict[str, Any],
) -> dict[str, Any]:

    result = {
        key:
            value

        for key, value
        in record.items()

        if key
        != "source"
    }


    result[
        "source_preview"
    ] = (
        record.get(
            "source",
            ""
        )[
            :6000
        ]
    )


    return result


compact_notebook_units = [
    compact_record(
        record
    )

    for record in notebook_units
]


compact_python_units = [
    compact_record(
        record
    )

    for record in python_units
]


# =============================================================================
# 22. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_C",

    "title":
        "MORPH_FRESH_CANDIDATE_GENERATOR_TARGETED_RECOVERY",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_SEMANTIC_SOURCE_RECOVERY_ONLY",

    "parent_14_3e_d_b": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "authority_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha_before,
        },

        "D2_R3": {
            "path":
                str(
                    D2R3_REL
                ),

            "file_sha256":
                d2r3_sha,

            "authority_mode":
                EXPECTED_AUTHORITY_MODE,
        },

        "Notebook11_semantic_anchor": {
            "path":
                str(
                    NOTEBOOK11_REL
                ),

            "cell_index":
                EXPECTED_SEMANTIC_ANCHOR_CELL,

            "cell_sha256":
                cells[
                    EXPECTED_SEMANTIC_ANCHOR_CELL
                ][
                    "sha256"
                ],
        },
    },

    "search_contract": {
        "expected_schema":
            list(
                EXPECTED_SCHEMA
            ),

        "core_generated_fields":
            sorted(
                CORE_GENERATED_FIELDS
            ),

        "config_id":
            EXPECTED_CONFIG_ID,

        "constructor_evidence_only":
            True,

        "mere_column_reads_do_not_count":
            True,

        "historical_pickle_reads_disqualify_fresh_authority":
            True,

        "GT_hazard_screen":
            True,
    },

    "Notebook11_candidates":
        compact_notebook_units,

    "repository_python_candidates":
        compact_python_units,

    "candidate_summary": {
        "clean_candidates":
            [
                compact_record(
                    record
                )

                for record
                in clean_candidates
            ],

        "mixed_GT_candidates":
            [
                compact_record(
                    record
                )

                for record
                in mixed_gt_candidates
            ],

        "pickle_dependent_candidates":
            [
                compact_record(
                    record
                )

                for record
                in pickle_dependent_candidates
            ],

        "partial_candidates":
            [
                compact_record(
                    record
                )

                for record
                in partial_candidates
            ],
    },

    "Notebook_dependency_closures":
        closure_records,

    "resolution": {
        "clean_source_candidate_count":
            len(
                clean_candidates
            ),

        "clean_GT_blind_closure_count":
            len(
                clean_closure_candidates
            ),

        "mixed_GT_candidate_count":
            len(
                mixed_gt_candidates
            ),

        "partial_candidate_count":
            len(
                partial_candidates
            ),

        "status":
            recovery_status,

        "implementation_authorized":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_selection_policy":
            "FROZEN",

        "Morph_graph_application_semantics":
            "FROZEN",

        "fresh_candidate_generation_authority":
            recovery_status,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_YET",

        "historical_action_runtime_use":
            False,

        "historical_pickle_semantic_content_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,

        "source_mutation":
            False,
    },

    "explicit_non_claims": [
        "No concrete MorphCandidateProvider was implemented.",
        "No historical Morph action pickle was loaded.",
        "No historical policy-grid pickle was loaded.",
        "Column reads do not constitute generator authority.",
        "A semantically similar source block is not automatically structural fidelity.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-C artifact exists but differs "
        "from current targeted recovery."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 23. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-C PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "SEARCH RESULT:",
    flush=True,
)

print(
    "  clean fresh generator sources:",
    len(
        clean_candidates
    ),
    flush=True,
)

print(
    "  clean GT-blind dependency closures:",
    len(
        clean_closure_candidates
    ),
    flush=True,
)

print(
    "  mixed GT generator sources:",
    len(
        mixed_gt_candidates
    ),
    flush=True,
)

print(
    "  partial semantic sources:",
    len(
        partial_candidates
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "RESOLUTION:",
    flush=True,
)

print(
    "  status:",
    recovery_status,
    flush=True,
)

print(
    "  implementation authorized:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  source mutated:",
    False,
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  performance generated:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-D — MORPH GENERATOR SPECIFICATION CLOSURE
#
# PARENT:
#   14.3E-D-C PASS
#
# PURPOSE
# -------
# 1. Reclassify the historical R1B cell61 block correctly:
#
#       cell61 For L334-L376
#       SHA b71ded6...
#
#    It CANONICALIZES existing `actions`.
#    It does NOT calculate fresh Morph candidates.
#
# 2. Search Notebook11 at AST-statement level for the REAL producers of:
#
#       morph_score
#       rank_in_dataset
#       morph_rank
#       candidate_rank
#       candidate_prob
#
# 3. Distinguish:
#
#       value construction
#       value canonicalization
#       action application
#       GT/evaluation
#       historical-pickle dependency
#
# 4. Freeze which parts of the generator specification are genuinely closed.
#
# 5. Do NOT authorize implementation unless the complete fresh-data
#    specification is source-backed.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical pickle loading
#   - GT graph loading
#   - evaluator execution
#   - performance computation
#   - parameter search
#   - concrete provider implementation
#   - structural fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_c_"
    "morph_fresh_candidate_generator_"
    "targeted_recovery_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


NOTEBOOK11_REL = Path(
    "notebooks/"
    "11_division_association_audit.ipynb"
)

NOTEBOOK11 = ROOT / NOTEBOOK11_REL


D2R3_REL = Path(
    "reports/"
    "notebook13_13_3d2r3_"
    "morph_authority_correction_and_"
    "semantic_reconstruction_freeze_v1.json"
)

D2R3 = ROOT / D2R3_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_d_"
    "morph_generator_specification_closure_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "55afd73f273593d38466e72ccbb0d8633792be350e7088d7d7b50203b7674bed"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "c4a11dc54d180801912ae7a6d24bb0d6f73736b056f7feef0231ba74528a0367"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_D2R3_FILE_SHA = (
    "ac0cb4959b8a9cb54784d0fd37d4da89a87b6e110298f67af00e712deebf8afb"
)


EXPECTED_AUTHORITY_MODE = (
    "SEMANTIC_SOURCE_LOCK_PLUS_"
    "HISTORICAL_GT_BLIND_STRUCTURAL_ORACLE"
)


# Historical Stage11 anchor.
ANCHOR_CELL_INDEX = 63

EXPECTED_ANCHOR_SHA = (
    "2e6bdff585ad1a3ee61b3a82ec03ccbd05db0b66f01024a35a721d76a29d8834"
)


# Cell61 current source identity from D-C.
CELL61_INDEX = 61

EXPECTED_CELL61_SHA = (
    "e449e6190f23607d7a20a8d9b83029485a2a7237a00c74917f2092eb6fe0d109"
)


# Previously-labelled R1B "generator" block.
R1B_BLOCK_SHA = (
    "b71ded6bda8547573f64222957061fd7d691090b0a083535babd733fcbd83d96"
)


# Current portable contract.
EXPECTED_MORPH_APPLY_SHA = (
    "db173c58a83eddd282712915514c31739815195ccb99edaa84a3ff5e16f1ffcf"
)

EXPECTED_ASSOCIATION_PROVIDER_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


MORPH_COLUMNS = (
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def node_source(
    text: str,
    node: ast.AST,
) -> str:

    value = ast.get_source_segment(
        text,
        node,
    )


    assert (
        value is not None
    )


    return value


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in obj.items():

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:

    return any(
        leaf[
            "value"
        ]
        == target

        for leaf in flatten_json(
            obj
        )
    )


def notebook_cells(
    path: Path,
) -> list[
    dict[str, Any]
]:

    obj = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    result = []


    for index, cell in enumerate(
        obj.get(
            "cells",
            []
        )
    ):

        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        result.append({
            "index":
                index,

            "type":
                cell.get(
                    "cell_type"
                ),

            "source":
                source,

            "sha256":
                sha256_text(
                    source
                ),
        })


    return result


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in tree.body

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "resolution"
        ][
            "status"
        ]
        ==
        "PARTIAL_GENERATOR_SEMANTICS_LOCALIZED_"
        "SPECIFICATION_CLOSURE_REQUIRED"
    )


    assert (
        obj[
            "resolution"
        ][
            "implementation_authorized"
        ]
        is False
    )


    return obj


# =============================================================================
# 4. AST helpers
# =============================================================================

def find_node_by_source_sha(
    source: str,
    expected_sha: str,
) -> list[
    dict[str, Any]
]:

    tree = ast.parse(
        source
    )


    records = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.For,
                ast.If,
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.Assign,
                ast.AnnAssign,
                ast.Expr,
            ),
        ):

            continue


        segment = ast.get_source_segment(
            source,
            node,
        )


        if segment is None:

            continue


        observed = sha256_text(
            segment
        )


        if observed != expected_sha:

            continue


        records.append({
            "node":
                node,

            "kind":
                type(
                    node
                ).__name__,

            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "source":
                segment,

            "sha256":
                observed,
        })


    return records


def calls_in_node(
    node: ast.AST,
) -> list[str]:

    result = set()


    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if name:

            result.add(
                name
            )


    return sorted(
        result
    )


def loaded_names(
    node: ast.AST,
) -> set[str]:

    return {
        child.id

        for child in ast.walk(
            node
        )

        if (
            isinstance(
                child,
                ast.Name,
            )
            and
            isinstance(
                child.ctx,
                ast.Load,
            )
        )
    }


def target_names(
    target: ast.AST,
) -> set[str]:

    result = set()


    for node in ast.walk(
        target
    ):

        if isinstance(
            node,
            ast.Name,
        ):

            result.add(
                node.id
            )


    return result


def assigned_plain_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    if isinstance(
        node,
        ast.Assign,
    ):

        for target in node.targets:

            result |= target_names(
                target
            )


    elif isinstance(
        node,
        ast.AnnAssign,
    ):

        result |= target_names(
            node.target
        )


    elif isinstance(
        node,
        ast.AugAssign,
    ):

        result |= target_names(
            node.target
        )


    return result


def subscript_target_fields(
    node: ast.AST,
) -> set[str]:

    result = set()


    targets = []


    if isinstance(
        node,
        ast.Assign,
    ):

        targets = list(
            node.targets
        )


    elif isinstance(
        node,
        (
            ast.AnnAssign,
            ast.AugAssign,
        ),
    ):

        targets = [
            node.target
        ]


    for target in targets:

        for child in ast.walk(
            target
        ):

            if not isinstance(
                child,
                ast.Subscript,
            ):

                continue


            try:

                value = ast.literal_eval(
                    child.slice
                )

            except Exception:

                continue


            if isinstance(
                value,
                str,
            ):

                result.add(
                    value
                )


    return result


def append_dict_fields(
    node: ast.AST,
) -> set[str]:

    result = set()


    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if (
            name is None
            or
            not name.endswith(
                ".append"
            )
            or
            len(
                call.args
            )
            != 1
            or
            not isinstance(
                call.args[
                    0
                ],
                ast.Dict,
            )
        ):

            continue


        dictionary = call.args[
            0
        ]


        for key_node in dictionary.keys:

            if key_node is None:

                continue


            try:

                value = ast.literal_eval(
                    key_node
                )

            except Exception:

                continue


            if isinstance(
                value,
                str,
            ):

                result.add(
                    value
                )


    return result


def dataframe_columns_fields(
    node: ast.AST,
) -> set[str]:

    result = set()


    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if (
            name is None
            or
            not name.endswith(
                "DataFrame"
            )
        ):

            continue


        for kw in call.keywords:

            if kw.arg != "columns":

                continue


            try:

                value = ast.literal_eval(
                    kw.value
                )

            except Exception:

                continue


            if not isinstance(
                value,
                (
                    list,
                    tuple,
                ),
            ):

                continue


            for item in value:

                if isinstance(
                    item,
                    str,
                ):

                    result.add(
                        item
                    )


    return result


def produced_fields(
    node: ast.AST,
) -> set[str]:

    return (
        subscript_target_fields(
            node
        )
        |
        append_dict_fields(
            node
        )
        |
        dataframe_columns_fields(
            node
        )
    )


def has_read_pickle(
    node: ast.AST,
) -> bool:

    return any(
        (
            isinstance(
                call,
                ast.Call,
            )
            and
            (
                dotted_name(
                    call.func
                )
                or ""
            ).endswith(
                "read_pickle"
            )
        )

        for call in ast.walk(
            node
        )
    )


GT_TOKENS = {
    "gt",
    "gt_graph",
    "gt_nodes",
    "gt_edges",
    "gt_base",
    "gt_policy",
    "ground_truth",
}


def gt_hazard(
    node: ast.AST,
) -> bool:

    names = {
        child.id.lower()

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Name,
        )
    }


    if (
        names
        &
        GT_TOKENS
    ):

        return True


    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = (
            dotted_name(
                call.func
            )
            or ""
        ).lower()


        if any(
            token
            in name

            for token in (
                "evaluate_graph",
                "_evaluate_matched_graph",
                "official_evaluate",
            )
        ):

            return True


    return False


# =============================================================================
# 5. Statement producer extraction
# =============================================================================

SEMANTIC_VARIABLES = {
    "morph_score",
    "rank_in_dataset",
    "morph_rank",
    "candidate_rank",
    "candidate_prob",
}


def statement_semantic_records(
    cells: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.Assign,
                    ast.AnnAssign,
                    ast.AugAssign,
                    ast.For,
                    ast.Expr,
                ),
            ):

                continue


            segment = ast.get_source_segment(
                source,
                node,
            )


            if not segment:

                continue


            assigned = assigned_plain_names(
                node
            )


            fields = produced_fields(
                node
            )


            relevant = (
                bool(
                    assigned
                    &
                    SEMANTIC_VARIABLES
                )
                or
                bool(
                    fields
                    &
                    SEMANTIC_VARIABLES
                )
                or
                any(
                    token
                    in segment

                    for token in (
                        "p_T4_ordered_",
                        "rank_in_dataset",
                        "morph_score",
                        "morph_rank",
                        "candidate_rank",
                        "candidate_prob",
                    )
                )
            )


            if not relevant:

                continue


            records.append({
                "cell_index":
                    cell[
                        "index"
                    ],

                "cell_sha256":
                    cell[
                        "sha256"
                    ],

                "kind":
                    type(
                        node
                    ).__name__,

                "line_start":
                    node.lineno,

                "line_end":
                    getattr(
                        node,
                        "end_lineno",
                        node.lineno,
                    ),

                "source_sha256":
                    sha256_text(
                        segment
                    ),

                "assigned_names":
                    sorted(
                        assigned
                    ),

                "produced_fields":
                    sorted(
                        fields
                    ),

                "loaded_names":
                    sorted(
                        loaded_names(
                            node
                        )
                    ),

                "calls":
                    calls_in_node(
                        node
                    ),

                "has_read_pickle":
                    has_read_pickle(
                        node
                    ),

                "GT_hazard":
                    gt_hazard(
                        node
                    ),

                "source":
                    segment,
            })


    # Deduplicate nested duplicates by exact source identity.
    unique = {}

    for record in records:

        key = (
            record[
                "cell_index"
            ],
            record[
                "source_sha256"
            ],
        )


        if key not in unique:

            unique[
                key
            ] = record


    return list(
        unique.values()
    )


# =============================================================================
# 6. Cell-level contamination profile
# =============================================================================

def cell_profile(
    cell: dict[str, Any],
) -> dict[str, Any]:

    source = cell[
        "source"
    ]


    try:

        tree = ast.parse(
            source
        )

    except Exception:

        return {
            "parseable":
                False,
        }


    return {
        "parseable":
            True,

        "has_read_pickle":
            has_read_pickle(
                tree
            ),

        "GT_hazard":
            gt_hazard(
                tree
            ),

        "calls":
            calls_in_node(
                tree
            ),
    }


# =============================================================================
# 7. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    NOTEBOOK11,
    D2R3,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-D "
    "MORPH GENERATOR SPECIFICATION CLOSURE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "NOTEBOOK11:",
    NOTEBOOK11_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Parent lock
# =============================================================================

parent = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_D_C_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Authority/source locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


d2r3_sha = sha256_file(
    D2R3
)


assert (
    d2r3_sha
    == EXPECTED_D2R3_FILE_SHA
)


d2r3 = load_json(
    D2R3
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_AUTHORITY_MODE,
    )
)


cells = notebook_cells(
    NOTEBOOK11
)


assert (
    cells[
        ANCHOR_CELL_INDEX
    ][
        "sha256"
    ]
    == EXPECTED_ANCHOR_SHA
)


assert (
    cells[
        CELL61_INDEX
    ][
        "sha256"
    ]
    == EXPECTED_CELL61_SHA
)


print(
    "[PASS] AUTHORITY_SOURCE_LOCKS",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  D2-R3 SHA:",
    d2r3_sha,
    flush=True,
)

print(
    "  anchor SHA:",
    EXPECTED_ANCHOR_SHA,
    flush=True,
)

print(
    "  cell61 SHA:",
    EXPECTED_CELL61_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Current portable Morph contract lock
# =============================================================================

portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


portable_tree = ast.parse(
    portable_text
)


portable_functions = top_level_functions(
    portable_tree
)


assert (
    "apply_portable_morph_div_v2"
    in portable_functions
)


apply_morph = (
    portable_functions[
        "apply_portable_morph_div_v2"
    ]
)


apply_morph_sha = sha256_text(
    node_source(
        portable_text,
        apply_morph,
    )
)


assert (
    apply_morph_sha
    == EXPECTED_MORPH_APPLY_SHA
)


assert (
    EXPECTED_ASSOCIATION_PROVIDER_SHA
    in portable_text
)


print(
    "[PASS] CURRENT_PORTABLE_MORPH_RUNTIME_LOCK",
    flush=True,
)

print(
    "  apply SHA:",
    apply_morph_sha,
    flush=True,
)

print(
    "  association provider identity present:",
    EXPECTED_ASSOCIATION_PROVIDER_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Correct the historical R1B block role
# =============================================================================

cell61_source = (
    cells[
        CELL61_INDEX
    ][
        "source"
    ]
)


r1b_matches = find_node_by_source_sha(
    cell61_source,
    R1B_BLOCK_SHA,
)


assert (
    len(
        r1b_matches
    )
    == 1
), (
    f"Expected one R1B block, got {len(r1b_matches)}."
)


r1b = r1b_matches[
    0
]


r1b_source = (
    r1b[
        "source"
    ]
)


# Exact role evidence.
assert (
    "actions.itertuples"
    in r1b_source
)


assert (
    "canonical_records.append"
    in r1b_source
)


assert (
    "row.morph_rank"
    in r1b_source
)


assert (
    "row.candidate_rank"
    in r1b_source
)


assert (
    "row.candidate_prob"
    in r1b_source
)


# It reads those values from `row`; it does not compute them.
r1b_calls = calls_in_node(
    r1b[
        "node"
    ]
)


r1b_fresh_generation_calls = [
    call

    for call in r1b_calls

    if any(
        token
        in call.lower()

        for token in (
            "association",
            "predict_proba",
            "rank",
            "groupby",
            "sort_values",
        )
    )
]


assert (
    r1b_fresh_generation_calls
    == []
)


r1b_role = (
    "EXISTING_ACTION_SCHEMA_CANONICALIZATION_"
    "NOT_FRESH_CANDIDATE_GENERATION"
)


print(
    "=" * 120,
    flush=True,
)

print(
    "R1B BLOCK ROLE CORRECTION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "cell:",
    CELL61_INDEX,
    flush=True,
)

print(
    "lines:",
    r1b[
        "line_start"
    ],
    "-",
    r1b[
        "line_end"
    ],
    flush=True,
)

print(
    "SHA:",
    r1b[
        "sha256"
    ],
    flush=True,
)

print(
    "role:",
    r1b_role,
    flush=True,
)

print(
    "fresh-generation calls:",
    r1b_fresh_generation_calls,
    flush=True,
)

print(
    "--- source ---",
    flush=True,
)

print(
    r1b_source,
    flush=True,
)

print(
    "--- end source ---",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Extract exact semantic producer statements
# =============================================================================

records = statement_semantic_records(
    cells
)


# Add whole-cell contamination context.
for record in records:

    profile = cell_profile(
        cells[
            record[
                "cell_index"
            ]
        ]
    )


    record[
        "cell_has_read_pickle"
    ] = profile.get(
        "has_read_pickle"
    )


    record[
        "cell_GT_hazard"
    ] = profile.get(
        "GT_hazard"
    )


# Sort strongest semantic producers first.
def record_score(
    record: dict[str, Any],
) -> int:

    score = 0


    assigned = set(
        record[
            "assigned_names"
        ]
    )


    fields = set(
        record[
            "produced_fields"
        ]
    )


    score += (
        15
        *
        len(
            assigned
            &
            SEMANTIC_VARIABLES
        )
    )


    score += (
        15
        *
        len(
            fields
            &
            SEMANTIC_VARIABLES
        )
    )


    if (
        record[
            "has_read_pickle"
        ]
    ):

        score -= 25


    if (
        record[
            "GT_hazard"
        ]
    ):

        score -= 25


    return score


for record in records:

    record[
        "score"
    ] = record_score(
        record
    )


records = sorted(
    records,
    key=lambda value: (
        -value[
            "score"
        ],
        value[
            "cell_index"
        ],
        value[
            "line_start"
        ],
    ),
)


print(
    "=" * 120,
    flush=True,
)

print(
    "SEMANTIC PRODUCER STATEMENTS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in records[
    :80
]:

    semantic_outputs = sorted(
        (
            set(
                record[
                    "assigned_names"
                ]
            )
            |
            set(
                record[
                    "produced_fields"
                ]
            )
        )
        &
        SEMANTIC_VARIABLES
    )


    if not semantic_outputs:

        continue


    print(
        f"SCORE={record['score']:>3}",
        f"cell={record['cell_index']}",
        f"L{record['line_start']}-{record['line_end']}",
        semantic_outputs,
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "  direct read_pickle:",
        record[
            "has_read_pickle"
        ],
        flush=True,
    )

    print(
        "  direct GT hazard:",
        record[
            "GT_hazard"
        ],
        flush=True,
    )

    print(
        "  cell read_pickle:",
        record[
            "cell_has_read_pickle"
        ],
        flush=True,
    )

    print(
        "  cell GT hazard:",
        record[
            "cell_GT_hazard"
        ],
        flush=True,
    )

    print(
        "  calls:",
        record[
            "calls"
        ],
        flush=True,
    )

    print(
        "  source:",
        flush=True,
    )

    print(
        record[
            "source"
        ][
            :5000
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 13. Component-specific producer sets
# =============================================================================

def records_for_output(
    name: str,
) -> list[
    dict[str, Any]
]:

    return [
        record

        for record in records

        if (
            name
            in record[
                "assigned_names"
            ]
            or
            name
            in record[
                "produced_fields"
            ]
        )
    ]


morph_score_records = records_for_output(
    "morph_score"
)


rank_in_dataset_records = records_for_output(
    "rank_in_dataset"
)


morph_rank_records = records_for_output(
    "morph_rank"
)


candidate_rank_records = records_for_output(
    "candidate_rank"
)


candidate_prob_records = records_for_output(
    "candidate_prob"
)


# =============================================================================
# 14. Source-only freshness classification
# =============================================================================

def fresh_source_candidates(
    values: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    return [
        record

        for record in values

        if (
            not record[
                "has_read_pickle"
            ]
            and
            not record[
                "GT_hazard"
            ]
            and
            not record[
                "cell_has_read_pickle"
            ]
            and
            not record[
                "cell_GT_hazard"
            ]
        )
    ]


fresh_morph_score = fresh_source_candidates(
    morph_score_records
)


fresh_rank_in_dataset = fresh_source_candidates(
    rank_in_dataset_records
)


fresh_morph_rank = fresh_source_candidates(
    morph_rank_records
)


fresh_candidate_rank = fresh_source_candidates(
    candidate_rank_records
)


fresh_candidate_prob = fresh_source_candidates(
    candidate_prob_records
)


print(
    "=" * 120,
    flush=True,
)

print(
    "FRESH SOURCE COMPONENT STATUS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


component_counts = {
    "morph_score":
        {
            "all":
                len(
                    morph_score_records
                ),

            "fresh":
                len(
                    fresh_morph_score
                ),
        },

    "rank_in_dataset":
        {
            "all":
                len(
                    rank_in_dataset_records
                ),

            "fresh":
                len(
                    fresh_rank_in_dataset
                ),
        },

    "morph_rank":
        {
            "all":
                len(
                    morph_rank_records
                ),

            "fresh":
                len(
                    fresh_morph_rank
                ),
        },

    "candidate_rank":
        {
            "all":
                len(
                    candidate_rank_records
                ),

            "fresh":
                len(
                    fresh_candidate_rank
                ),
        },

    "candidate_prob":
        {
            "all":
                len(
                    candidate_prob_records
                ),

            "fresh":
                len(
                    fresh_candidate_prob
                ),
        },
}


print(
    json.dumps(
        component_counts,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Semantic source-location inventory
# =============================================================================

KEY_TERMS = (
    "p_T4_ordered_",
    "morph_score",
    "rank_in_dataset",
    "morph_rank",
    "candidate_rank",
    "candidate_prob",
    "association",
)


term_cell_hits = {
    term: []

    for term in KEY_TERMS
}


for cell in cells:

    if (
        cell[
            "type"
        ]
        != "code"
    ):

        continue


    source = cell[
        "source"
    ]


    for term in KEY_TERMS:

        if term not in source:

            continue


        profile = cell_profile(
            cell
        )


        term_cell_hits[
            term
        ].append({
            "cell_index":
                cell[
                    "index"
                ],

            "cell_sha256":
                cell[
                    "sha256"
                ],

            "has_read_pickle":
                profile.get(
                    "has_read_pickle"
                ),

            "GT_hazard":
                profile.get(
                    "GT_hazard"
                ),

            "occurrences":
                source.count(
                    term
                ),
        })


print(
    "=" * 120,
    flush=True,
)

print(
    "SEMANTIC TERM LOCATION INVENTORY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for term, values in term_cell_hits.items():

    print(
        term,
        ":",
        values,
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 16. Freeze already-authoritative specification pieces
# =============================================================================

known_specification = {
    "input_graph_boundary":
        "POST_FROZEN_V9",

    "candidate_source_universe":
        (
            "FINAL Frozen-V9 observable W4-capable parents "
            "with current outdegree == 1"
        ),

    "morphology_selection":
        {
            "rule":
                "temporal morphology Top 3 per dataset",

            "rank_cap":
                3,
        },

    "association_selection":
        {
            "row_rank_cap":
                3,

            "candidate_probability_min":
                0.10,
        },

    "graph_mutation":
        {
            "source_pre_outdegree":
                1,

            "target_pre_indegree":
                0,

            "duplicate_edge":
                "FORBIDDEN",

            "source_post_outdegree":
                2,

            "target_post_indegree":
                1,

            "operation":
                "ADDITIVE_SECOND_EDGE",
        },

    "candidate_schema":
        list(
            MORPH_COLUMNS
        ),

    "historical_action_runtime_lookup":
        False,

    "historical_action_structural_oracle_use":
        True,
}


# =============================================================================
# 17. Determine what the specification STILL lacks
# =============================================================================

#
# Important:
#
# The scientific selection lock tells us WHAT ranks/thresholds mean,
# but implementation requires source-backed HOW-to-compute primitives.
#

component_status = {
    "candidate_universe":
        "FROZEN",

    "output_schema":
        "FROZEN",

    "morphology_rank_cap":
        "FROZEN",

    "association_rank_cap":
        "FROZEN",

    "association_probability_threshold":
        "FROZEN",

    "graph_application":
        "FROZEN",

    "fresh_association_provider_dependency":
        (
            "AVAILABLE_BUT_MORPH_HISTORICAL_EQUIVALENCE_NOT_YET_PROVEN"
        ),

    "morphology_score_formula":
        (
            "SOURCE_BACKED"
            if fresh_morph_score
            else
            "NOT_FRESH_SOURCE_CLOSED"
        ),

    "morphology_dataset_rank_formula":
        (
            "SOURCE_BACKED"
            if fresh_rank_in_dataset
            else
            "NOT_FRESH_SOURCE_CLOSED"
        ),

    "morph_rank_materialization":
        (
            "SOURCE_BACKED"
            if fresh_morph_rank
            else
            "NOT_FRESH_SOURCE_CLOSED"
        ),

    "candidate_rank_materialization":
        (
            "SOURCE_BACKED"
            if fresh_candidate_rank
            else
            "NOT_FRESH_SOURCE_CLOSED"
        ),

    "candidate_probability_materialization":
        (
            "SOURCE_BACKED"
            if fresh_candidate_prob
            else
            "NOT_FRESH_SOURCE_CLOSED"
        ),
}


required_generation_components = {
    "morphology_score_formula",
    "morphology_dataset_rank_formula",
    "morph_rank_materialization",
    "candidate_rank_materialization",
    "candidate_probability_materialization",
}


missing_generation_components = sorted(
    key

    for key in required_generation_components

    if (
        component_status[
            key
        ]
        != "SOURCE_BACKED"
    )
)


# =============================================================================
# 18. Specification closure decision
# =============================================================================

specification_complete = (
    len(
        missing_generation_components
    )
    == 0
)


implementation_authorized = (
    specification_complete
)


if specification_complete:

    decision_status = (
        "SPECIFICATION_DERIVED_FRESH_MORPH_"
        "GENERATOR_CONTRACT_COMPLETE"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_E_"
        "CONCRETE_MORPH_PROVIDER_IMPLEMENTATION"
    )


else:

    decision_status = (
        "SPECIFICATION_CLOSURE_PARTIAL_"
        "FRESH_INPUT_PROVENANCE_REQUIRED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_E_"
        "MORPH_GENERATOR_INPUT_PROVENANCE_CLOSURE"
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "MORPH GENERATOR SPECIFICATION CLOSURE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "known specification:",
    json.dumps(
        known_specification,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "component status:",
    json.dumps(
        component_status,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "missing generation components:",
    missing_generation_components,
    flush=True,
)

print(
    "specification complete:",
    specification_complete,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "decision:",
    decision_status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 19. Safety / no mutation
# =============================================================================

source_mutation = False

test_mutation = False

historical_pickle_loaded = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


# =============================================================================
# 20. Formal gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_c_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "D2_R3_authority_locked":
        (
            d2r3_sha
            == EXPECTED_D2R3_FILE_SHA
        ),

    "semantic_anchor_locked":
        (
            cells[
                ANCHOR_CELL_INDEX
            ][
                "sha256"
            ]
            == EXPECTED_ANCHOR_SHA
        ),

    "cell61_locked":
        (
            cells[
                CELL61_INDEX
            ][
                "sha256"
            ]
            == EXPECTED_CELL61_SHA
        ),

    "R1B_block_exactly_recovered":
        (
            len(
                r1b_matches
            )
            == 1
        ),

    "R1B_block_correctly_reclassified":
        (
            r1b_role
            ==
            "EXISTING_ACTION_SCHEMA_CANONICALIZATION_"
            "NOT_FRESH_CANDIDATE_GENERATION"
        ),

    "semantic_statement_scan_complete":
        True,

    "fresh_vs_historical_dependency_screen_complete":
        True,

    "known_selection_policy_preserved":
        True,

    "graph_application_contract_preserved":
        True,

    "missing_generation_components_explicit":
        True,

    "implementation_authorization_matches_specification_completeness":
        (
            implementation_authorized
            ==
            specification_complete
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-D SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-D failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 21. Artifact-safe compact semantic records
# =============================================================================

def compact(
    record: dict[str, Any],
) -> dict[str, Any]:

    output = {
        key:
            value

        for key, value in record.items()

        if key
        != "source"
    }


    output[
        "source_preview"
    ] = record.get(
        "source",
        "",
    )[
        :5000
    ]


    return output


# =============================================================================
# 22. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_D",

    "title":
        "MORPH_GENERATOR_SPECIFICATION_CLOSURE",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_SPECIFICATION_CLOSURE_ONLY",

    "parent_14_3e_d_c": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "authority_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha_before,

            "Morph_runtime_sha256":
                apply_morph_sha,

            "reference_association_provider_sha256":
                EXPECTED_ASSOCIATION_PROVIDER_SHA,
        },

        "D2_R3": {
            "path":
                str(
                    D2R3_REL
                ),

            "file_sha256":
                d2r3_sha,

            "authority_mode":
                EXPECTED_AUTHORITY_MODE,
        },

        "Notebook11": {
            "path":
                str(
                    NOTEBOOK11_REL
                ),

            "semantic_anchor_cell":
                ANCHOR_CELL_INDEX,

            "semantic_anchor_sha256":
                EXPECTED_ANCHOR_SHA,

            "cell61_sha256":
                EXPECTED_CELL61_SHA,
        },
    },

    "R1B_role_correction": {
        "cell":
            CELL61_INDEX,

        "source_sha256":
            R1B_BLOCK_SHA,

        "line_start":
            r1b[
                "line_start"
            ],

        "line_end":
            r1b[
                "line_end"
            ],

        "role":
            r1b_role,

        "fresh_generation_calls":
            r1b_fresh_generation_calls,

        "source":
            r1b_source,
    },

    "known_generator_specification":
        known_specification,

    "semantic_component_inventory": {
        "morph_score":
            [
                compact(
                    value
                )

                for value
                in morph_score_records
            ],

        "rank_in_dataset":
            [
                compact(
                    value
                )

                for value
                in rank_in_dataset_records
            ],

        "morph_rank":
            [
                compact(
                    value
                )

                for value
                in morph_rank_records
            ],

        "candidate_rank":
            [
                compact(
                    value
                )

                for value
                in candidate_rank_records
            ],

        "candidate_prob":
            [
                compact(
                    value
                )

                for value
                in candidate_prob_records
            ],
    },

    "fresh_component_counts":
        component_counts,

    "semantic_term_locations":
        term_cell_hits,

    "component_status":
        component_status,

    "missing_generation_components":
        missing_generation_components,

    "resolution": {
        "specification_complete":
            specification_complete,

        "implementation_authorized":
            implementation_authorized,

        "status":
            decision_status,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_selection_policy":
            "FROZEN",

        "Morph_graph_application_semantics":
            "FROZEN",

        "R1B_cell61_L334_376_role":
            r1b_role,

        "fresh_generator_specification":
            decision_status,

        "concrete_Morph_provider_implementation":
            (
                "AUTHORIZED_NEXT_STAGE"
                if implementation_authorized
                else
                "NOT_AUTHORIZED_YET"
            ),

        "historical_action_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,
    },

    "explicit_non_claims": [
        "The former R1B cell61 canonicalization loop is not fresh candidate-generation authority.",
        "No historical Morph action pickle was loaded.",
        "No concrete MorphCandidateProvider was implemented.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-D artifact exists but differs "
        "from the exact specification-closure state."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 23. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-D PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "R1B CORRECTION:",
    flush=True,
)

print(
    "  block SHA:",
    R1B_BLOCK_SHA,
    flush=True,
)

print(
    "  role:",
    r1b_role,
    flush=True,
)

print(
    flush=True,
)


print(
    "SPECIFICATION:",
    flush=True,
)

print(
    "  selection policy:",
    "FROZEN",
    flush=True,
)

print(
    "  graph application:",
    "FROZEN",
    flush=True,
)

print(
    "  fresh generator complete:",
    specification_complete,
    flush=True,
)

print(
    "  missing components:",
    missing_generation_components,
    flush=True,
)

print(
    flush=True,
)


print(
    "DECISION:",
    flush=True,
)

print(
    "  status:",
    decision_status,
    flush=True,
)

print(
    "  implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  source mutation:",
    False,
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  performance generated:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-E — MORPH GENERATOR INPUT PROVENANCE CLOSURE
#
# PARENT
# ------
# 14.3E-D-D PASS
#
# D-D established:
#
#   selection policy       = FROZEN
#   graph application      = FROZEN
#
# but reported five generation components as open.
#
# IMPORTANT REFINEMENT
# --------------------
# Those five components are NOT equivalent:
#
#   morphology_score_formula
#       exact formula already recovered:
#
#           df["morph_score"] = df[rank_cols].mean(axis=1)
#
#   morphology_dataset_rank_formula
#       exact formula already recovered:
#
#           df["rank_in_dataset"] = (
#               df.groupby("dataset").cumcount() + 1
#           )
#
# These formulas themselves are source-backed.
#
# Their remaining blocker is FRESH INPUT PROVENANCE:
#
#   - where do rank_cols / p_T4_ordered_* come from?
#   - can those quantities be generated for an arbitrary fresh dataset
#     without reading historical OOF/fold artifacts?
#
#
# Remaining materialization questions:
#
#   - rank_in_dataset -> morph_rank
#   - association matrix -> candidate_prob
#   - association matrix -> candidate_rank
#
#
# THIS STAGE
# ----------
# 1. Exact-lock D-D.
# 2. Exact-lock Notebook11 cells 46 / 51 / 55.
# 3. Re-lock exact morphology score/rank statements.
# 4. Extract historical input artifact expressions from those cells.
# 5. Search current checkout for source producers of p_T4_ordered_*.
# 6. Locate the exact reference association provider by function SHA.
# 7. Search source for materialization of:
#
#       morph_rank
#       candidate_rank
#       candidate_prob
#
# 8. Distinguish formula authority from fresh-input authority.
# 9. Freeze remaining primitive blockers.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical pickle deserialization
#   - GT loading
#   - evaluator execution
#   - performance computation
#   - parameter search
#   - provider implementation
#   - structural fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_d_"
    "morph_generator_specification_closure_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


NOTEBOOK11_REL = Path(
    "notebooks/"
    "11_division_association_audit.ipynb"
)

NOTEBOOK11 = ROOT / NOTEBOOK11_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_e_"
    "morph_generator_input_provenance_closure_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "ed211e95612a4ad20ad98a234905050ac5274d9c5cc3a63097ff525212bd21a0"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "da66faef4b52759cd33b43017550c73ce23b44689139d2ceae5cf22af4383bfa"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


# Notebook11 exact cells from D-D.
CELL46 = 46

CELL46_SHA = (
    "15747465b4e39aa4a2e72e259a9cdbbdfa83c05395c939904b6b6a4f436f36b1"
)


CELL51 = 51

CELL51_SHA = (
    "0be0b5b973555fbdd1b969374842bede150f1d80895596d498b433fb664ccb1a"
)


CELL55 = 55

CELL55_SHA = (
    "97c347aea36888e1d286c8d54e8adfc902dc7f4e86e1e2989009caa86a8025fc"
)


CELL63 = 63

CELL63_SHA = (
    "2e6bdff585ad1a3ee61b3a82ec03ccbd05db0b66f01024a35a721d76a29d8834"
)


# Exact formula statements recovered in D-D.
MORPH_SCORE_STATEMENT_SHA = (
    "1eb422867db49a6255e89ae08443846a0c03458f484a9cc61065375800e7bad4"
)

DATASET_RANK_STATEMENT_SHA = (
    "7606a5627025b8834d6025ab39db54048127241c46b1b8ce301068adabcc698f"
)


# Current reference association provider authority.
EXPECTED_ASSOCIATION_PROVIDER_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


# Scientific contract.
MORPH_COLUMNS = (
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def safe_rel(
    path: Path,
) -> str:

    try:

        return str(
            path.resolve().relative_to(
                ROOT
            )
        )

    except Exception:

        return str(
            path
        )


def node_source(
    source: str,
    node: ast.AST,
) -> str:

    result = ast.get_source_segment(
        source,
        node,
    )


    assert (
        result is not None
    )


    return result


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def notebook_cells(
    path: Path,
) -> list[
    dict[str, Any]
]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    records = []


    for index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        records.append({
            "index":
                index,

            "cell_type":
                cell.get(
                    "cell_type"
                ),

            "source":
                source,

            "sha256":
                sha256_text(
                    source
                ),
        })


    return records


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in obj.items():

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


# =============================================================================
# 3. Parent verification
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "resolution"
        ][
            "specification_complete"
        ]
        is False
    )


    assert (
        obj[
            "resolution"
        ][
            "implementation_authorized"
        ]
        is False
    )


    return obj


# =============================================================================
# 4. AST exact-node helpers
# =============================================================================

def find_node_by_sha(
    source: str,
    expected_sha: str,
) -> list[
    dict[str, Any]
]:

    tree = ast.parse(
        source
    )


    records = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
                ast.For,
                ast.If,
                ast.Expr,
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        ):

            continue


        segment = ast.get_source_segment(
            source,
            node,
        )


        if segment is None:

            continue


        observed = sha256_text(
            segment
        )


        if observed != expected_sha:

            continue


        records.append({
            "node":
                node,

            "kind":
                type(
                    node
                ).__name__,

            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "source":
                segment,

            "sha256":
                observed,
        })


    return records


def calls_in_node(
    node: ast.AST,
) -> list[str]:

    return sorted({
        dotted_name(
            call.func
        )
        or "<dynamic-call>"

        for call in ast.walk(
            node
        )

        if isinstance(
            call,
            ast.Call,
        )
    })


def names_in_node(
    node: ast.AST,
) -> set[str]:

    return {
        child.id

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Name,
        )
    }


def has_read_pickle(
    node: ast.AST,
) -> bool:

    return any(
        (
            isinstance(
                child,
                ast.Call,
            )
            and
            (
                dotted_name(
                    child.func
                )
                or ""
            ).endswith(
                "read_pickle"
            )
        )

        for child in ast.walk(
            node
        )
    )


# =============================================================================
# 5. Historical artifact-expression extraction
# =============================================================================

def pickle_literals(
    source: str,
) -> list[str]:

    return sorted({
        child.value

        for child in ast.walk(
            ast.parse(
                source
            )
        )

        if (
            isinstance(
                child,
                ast.Constant,
            )
            and
            isinstance(
                child.value,
                str,
            )
            and
            child.value.endswith(
                ".pkl"
            )
        )
    })


def path_like_assignments(
    source: str,
) -> list[
    dict[str, Any]
]:

    tree = ast.parse(
        source
    )


    records = []


    for stmt in tree.body:

        if not isinstance(
            stmt,
            (
                ast.Assign,
                ast.AnnAssign,
            ),
        ):

            continue


        value = stmt.value


        names = []


        if isinstance(
            stmt,
            ast.Assign,
        ):

            for target in stmt.targets:

                if isinstance(
                    target,
                    ast.Name,
                ):

                    names.append(
                        target.id
                    )


        elif isinstance(
            stmt.target,
            ast.Name,
        ):

            names.append(
                stmt.target.id
            )


        value_source = ast.unparse(
            value
        )


        if (
            ".pkl"
            not in value_source
            and
            "PATH"
            not in " ".join(
                names
            ).upper()
        ):

            continue


        records.append({
            "line":
                stmt.lineno,

            "names":
                names,

            "value":
                value_source,

            "pickle_literals":
                [
                    literal

                    for literal in pickle_literals(
                        value_source
                    )
                ],
        })


    return records


def read_pickle_records(
    source: str,
) -> list[
    dict[str, Any]
]:

    tree = ast.parse(
        source
    )


    records = []


    for call in ast.walk(
        tree
    ):

        if not isinstance(
            call,
            ast.Call,
        ):

            continue


        name = dotted_name(
            call.func
        )


        if (
            name is None
            or
            not name.endswith(
                "read_pickle"
            )
        ):

            continue


        arg = (
            ast.unparse(
                call.args[
                    0
                ]
            )
            if call.args
            else None
        )


        records.append({
            "line":
                call.lineno,

            "call":
                ast.unparse(
                    call
                ),

            "path_argument":
                arg,
        })


    return records


# =============================================================================
# 6. Search exact source producers by basename
# =============================================================================

def source_files() -> list[Path]:

    paths = []


    for root in (
        ROOT / "src",
        ROOT / "scripts",
        ROOT / "notebooks",
    ):

        if not root.exists():

            continue


        for path in root.rglob(
            "*"
        ):

            if (
                path.is_file()
                and
                path.suffix
                in {
                    ".py",
                    ".ipynb",
                }
            ):

                paths.append(
                    path
                )


    return sorted(
        paths
    )


SCAN_FILES = source_files()


def basename_source_hits(
    basename: str,
) -> list[
    dict[str, Any]
]:

    records = []


    for path in SCAN_FILES:

        try:

            text = path.read_text(
                encoding="utf-8"
            )

        except Exception:

            continue


        if basename not in text:

            continue


        if path.suffix == ".py":

            records.append({
                "path":
                    safe_rel(
                        path
                    ),

                "kind":
                    "PYTHON_FILE",

                "file_sha256":
                    sha256_file(
                        path
                    ),

                "contains_to_pickle":
                    (
                        ".to_pickle"
                        in text
                        or
                        "pd.to_pickle"
                        in text
                    ),

                "contains_read_pickle":
                    (
                        "read_pickle"
                        in text
                    ),

                "preview":
                    "\n".join(
                        line

                        for line in text.splitlines()

                        if basename in line
                    )[
                        :4000
                    ],
            })


        else:

            notebook = json.loads(
                text
            )


            for cell_index, cell in enumerate(
                notebook.get(
                    "cells",
                    []
                )
            ):

                source = "".join(
                    cell.get(
                        "source",
                        []
                    )
                )


                if basename not in source:

                    continue


                records.append({
                    "path":
                        safe_rel(
                            path
                        ),

                    "kind":
                        "NOTEBOOK_CELL",

                    "cell_index":
                        cell_index,

                    "cell_sha256":
                        sha256_text(
                            source
                        ),

                    "contains_to_pickle":
                        (
                            ".to_pickle"
                            in source
                            or
                            "pd.to_pickle"
                            in source
                        ),

                    "contains_read_pickle":
                        (
                            "read_pickle"
                            in source
                        ),

                    "preview":
                        source[
                            :5000
                        ],
                })


    return records


# =============================================================================
# 7. Find exact association-provider function by SHA
# =============================================================================

def find_function_sha_in_repo(
    expected_sha: str,
) -> list[
    dict[str, Any]
]:

    records = []


    for path in SCAN_FILES:

        if path.suffix != ".py":

            continue


        try:

            text = path.read_text(
                encoding="utf-8"
            )


            tree = ast.parse(
                text,
                filename=str(
                    path
                ),
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                ),
            ):

                continue


            segment = ast.get_source_segment(
                text,
                node,
            )


            if segment is None:

                continue


            observed = sha256_text(
                segment
            )


            if observed != expected_sha:

                continue


            records.append({
                "path":
                    safe_rel(
                        path
                    ),

                "file_sha256":
                    sha256_file(
                        path
                    ),

                "function":
                    node.name,

                "function_sha256":
                    observed,

                "line_start":
                    node.lineno,

                "line_end":
                    node.end_lineno,

                "calls":
                    calls_in_node(
                        node
                    ),

                "source":
                    segment,
            })


    return records


# =============================================================================
# 8. Search morphology-prediction source units
# =============================================================================

def source_unit_hazards(
    node: ast.AST,
) -> dict[str, Any]:

    names = {
        name.lower()

        for name in names_in_node(
            node
        )
    }


    calls = [
        call.lower()

        for call in calls_in_node(
            node
        )
    ]


    gt = (
        bool(
            names
            &
            {
                "gt",
                "gt_graph",
                "gt_nodes",
                "gt_edges",
                "ground_truth",
            }
        )
        or
        any(
            "evaluate_graph"
            in call
            or
            "official_evaluate"
            in call

            for call in calls
        )
    )


    return {
        "read_pickle":
            has_read_pickle(
                node
            ),

        "GT_hazard":
            gt,
    }


def morphology_prediction_source_candidates() -> list[
    dict[str, Any]
]:

    records = []


    required_signal = (
        "p_T4_ordered_"
    )


    for path in SCAN_FILES:

        if path.suffix != ".py":

            continue


        try:

            text = path.read_text(
                encoding="utf-8"
            )


            if (
                required_signal
                not in text
            ):

                continue


            tree = ast.parse(
                text,
                filename=str(
                    path
                ),
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                ),
            ):

                continue


            segment = ast.get_source_segment(
                text,
                node,
            )


            if (
                segment is None
                or
                required_signal
                not in segment
            ):

                continue


            hazards = source_unit_hazards(
                node
            )


            records.append({
                "path":
                    safe_rel(
                        path
                    ),

                "file_sha256":
                    sha256_file(
                        path
                    ),

                "function":
                    node.name,

                "function_sha256":
                    sha256_text(
                        segment
                    ),

                "line_start":
                    node.lineno,

                "line_end":
                    node.end_lineno,

                "read_pickle":
                    hazards[
                        "read_pickle"
                    ],

                "GT_hazard":
                    hazards[
                        "GT_hazard"
                    ],

                "calls":
                    calls_in_node(
                        node
                    ),

                "source":
                    segment,
            })


    return records


# =============================================================================
# 9. Materialization-lineage search
# =============================================================================

def statement_records_with_terms(
    terms: set[str],
) -> list[
    dict[str, Any]
]:

    records = []


    for path in SCAN_FILES:

        if path.suffix == ".py":

            try:

                text = path.read_text(
                    encoding="utf-8"
                )


                tree = ast.parse(
                    text,
                    filename=str(
                        path
                    ),
                )

            except Exception:

                continue


            scopes = [
                (
                    "PYTHON_FILE",
                    None,
                    text,
                    tree,
                )
            ]


        else:

            try:

                notebook = json.loads(
                    path.read_text(
                        encoding="utf-8"
                    )
                )

            except Exception:

                continue


            scopes = []


            for cell_index, cell in enumerate(
                notebook.get(
                    "cells",
                    []
                )
            ):

                source = "".join(
                    cell.get(
                        "source",
                        []
                    )
                )


                if not any(
                    term
                    in source

                    for term in terms
                ):

                    continue


                try:

                    tree = ast.parse(
                        source
                    )

                except Exception:

                    continue


                scopes.append(
                    (
                        "NOTEBOOK_CELL",
                        cell_index,
                        source,
                        tree,
                    )
                )


        for (
            kind,
            cell_index,
            source,
            tree,
        ) in scopes:

            for node in ast.walk(
                tree
            ):

                if not isinstance(
                    node,
                    (
                        ast.Assign,
                        ast.AnnAssign,
                        ast.For,
                        ast.Expr,
                    ),
                ):

                    continue


                segment = ast.get_source_segment(
                    source,
                    node,
                )


                if not segment:

                    continue


                hit_terms = sorted(
                    term

                    for term in terms

                    if term in segment
                )


                if not hit_terms:

                    continue


                # Consumer-only canonicalization is retained but explicitly
                # classified, not treated as producer authority.
                uses_row_values = (
                    "row.candidate_prob"
                    in segment
                    or
                    "row.candidate_rank"
                    in segment
                    or
                    "row.morph_rank"
                    in segment
                )


                computes_rank = any(
                    token
                    in segment

                    for token in (
                        ".rank(",
                        ".cumcount(",
                        "argsort",
                        "sort_values",
                    )
                )


                computes_probability = (
                    "probs["
                    in segment
                    or
                    "probability"
                    in segment
                    or
                    "predict_proba"
                    in segment
                )


                maps_rank_in_dataset = (
                    "morph_rank"
                    in segment
                    and
                    "rank_in_dataset"
                    in segment
                )


                records.append({
                    "path":
                        safe_rel(
                            path
                        ),

                    "kind":
                        kind,

                    "cell_index":
                        cell_index,

                    "line_start":
                        node.lineno,

                    "line_end":
                        getattr(
                            node,
                            "end_lineno",
                            node.lineno,
                        ),

                    "source_sha256":
                        sha256_text(
                            segment
                        ),

                    "terms":
                        hit_terms,

                    "uses_existing_row_values":
                        uses_row_values,

                    "computes_rank":
                        computes_rank,

                    "computes_probability":
                        computes_probability,

                    "maps_rank_in_dataset_to_morph_rank":
                        maps_rank_in_dataset,

                    "read_pickle":
                        has_read_pickle(
                            node
                        ),

                    "source":
                        segment,
                })


    # Deduplicate nested source repeats.
    unique = {}


    for record in records:

        key = (
            record[
                "path"
            ],
            record[
                "cell_index"
            ],
            record[
                "source_sha256"
            ],
        )


        unique.setdefault(
            key,
            record,
        )


    return list(
        unique.values()
    )


# =============================================================================
# 10. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    NOTEBOOK11,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-E "
    "MORPH GENERATOR INPUT PROVENANCE CLOSURE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "NOTEBOOK11:",
    NOTEBOOK11_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Parent lock
# =============================================================================

parent = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_D_D_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Current source + notebook cell locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


cells = notebook_cells(
    NOTEBOOK11
)


expected_cells = {
    CELL46:
        CELL46_SHA,

    CELL51:
        CELL51_SHA,

    CELL55:
        CELL55_SHA,

    CELL63:
        CELL63_SHA,
}


for index, expected_sha in expected_cells.items():

    assert (
        cells[
            index
        ][
            "sha256"
        ]
        == expected_sha
    ), (
        f"Notebook11 cell {index} drift."
    )


print(
    "[PASS] NOTEBOOK11_INPUT_PROVENANCE_CELL_LOCKS",
    flush=True,
)

for index, expected_sha in expected_cells.items():

    print(
        f"  cell {index}:",
        expected_sha,
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 13. Reclassify D-D formula status
# =============================================================================

cell46_source = cells[
    CELL46
][
    "source"
]


morph_score_matches = find_node_by_sha(
    cell46_source,
    MORPH_SCORE_STATEMENT_SHA,
)


rank_matches = find_node_by_sha(
    cell46_source,
    DATASET_RANK_STATEMENT_SHA,
)


assert (
    len(
        morph_score_matches
    )
    == 1
)


assert (
    len(
        rank_matches
    )
    == 1
)


morph_score_statement = (
    morph_score_matches[
        0
    ]
)


rank_statement = (
    rank_matches[
        0
    ]
)


assert (
    'df["morph_score"]'
    in morph_score_statement[
        "source"
    ]
)


assert (
    ".mean(axis=1)"
    in morph_score_statement[
        "source"
    ]
)


assert (
    'df["rank_in_dataset"]'
    in rank_statement[
        "source"
    ]
)


assert (
    'groupby("dataset")'
    in rank_statement[
        "source"
    ]
)


assert (
    ".cumcount()"
    in rank_statement[
        "source"
    ]
)


formula_status = {
    "morphology_score_formula":
        "FORMULA_FROZEN_INPUT_PROVENANCE_OPEN",

    "morphology_dataset_rank_formula":
        "FORMULA_FROZEN_INPUT_PROVENANCE_OPEN",
}


print(
    "=" * 120,
    flush=True,
)

print(
    "FORMULA AUTHORITY RECLASSIFICATION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "morphology score:",
    formula_status[
        "morphology_score_formula"
    ],
    flush=True,
)

print(
    "  SHA:",
    MORPH_SCORE_STATEMENT_SHA,
    flush=True,
)

print(
    "  source:",
    morph_score_statement[
        "source"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "dataset morphology rank:",
    formula_status[
        "morphology_dataset_rank_formula"
    ],
    flush=True,
)

print(
    "  SHA:",
    DATASET_RANK_STATEMENT_SHA,
    flush=True,
)

print(
    "  source:",
    rank_statement[
        "source"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Historical input artifact inventory
# =============================================================================

relevant_cells = {
    index:
        cells[
            index
        ][
            "source"
        ]

    for index in (
        CELL46,
        CELL51,
        CELL55,
    )
}


cell_input_records = {}


all_pickle_basenames = set()


for index, source in relevant_cells.items():

    literals = pickle_literals(
        source
    )


    all_pickle_basenames.update(
        literals
    )


    cell_input_records[
        index
    ] = {
        "cell_sha256":
            cells[
                index
            ][
                "sha256"
            ],

        "path_assignments":
            path_like_assignments(
                source
            ),

        "read_pickle_calls":
            read_pickle_records(
                source
            ),

        "pickle_literals":
            literals,
    }


print(
    "=" * 120,
    flush=True,
)

print(
    "HISTORICAL INPUT ARTIFACT INVENTORY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for index, record in cell_input_records.items():

    print(
        f"CELL {index}",
        flush=True,
    )

    print(
        "  literals:",
        record[
            "pickle_literals"
        ],
        flush=True,
    )

    print(
        "  reads:",
        json.dumps(
            record[
                "read_pickle_calls"
            ],
            indent=2,
        ),
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 15. Artifact source-reference scan
# =============================================================================

artifact_source_hits = {}


for basename in sorted(
    all_pickle_basenames
):

    artifact_source_hits[
        basename
    ] = basename_source_hits(
        basename
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "INPUT ARTIFACT SOURCE REFERENCES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for basename, records in artifact_source_hits.items():

    print(
        basename,
        flush=True,
    )

    print(
        "  references:",
        len(
            records
        ),
        flush=True,
    )


    for record in records[
        :20
    ]:

        print(
            " ",
            record[
                "kind"
            ],
            record[
                "path"
            ],
            flush=True,
        )

        print(
            "    writes in scope:",
            record[
                "contains_to_pickle"
            ],
            flush=True,
        )

        print(
            "    reads in scope:",
            record[
                "contains_read_pickle"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 16. Morphology prediction source search
# =============================================================================

morph_prediction_candidates = (
    morphology_prediction_source_candidates()
)


clean_morph_prediction_candidates = [
    record

    for record in morph_prediction_candidates

    if (
        not record[
            "read_pickle"
        ]
        and
        not record[
            "GT_hazard"
        ]
    )
]


print(
    "=" * 120,
    flush=True,
)

print(
    "P_T4_ORDERED FRESH SOURCE CANDIDATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "all candidates:",
    len(
        morph_prediction_candidates
    ),
    flush=True,
)

print(
    "clean no-pickle/no-GT candidates:",
    len(
        clean_morph_prediction_candidates
    ),
    flush=True,
)

print(
    flush=True,
)


for record in morph_prediction_candidates[
    :30
]:

    print(
        record[
            "path"
        ],
        "::",
        record[
            "function"
        ],
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "function_sha256"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        record[
            "read_pickle"
        ],
        flush=True,
    )

    print(
        "  GT hazard:",
        record[
            "GT_hazard"
        ],
        flush=True,
    )

    print(
        "  calls:",
        record[
            "calls"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 17. Exact association-provider source authority
# =============================================================================

association_provider_matches = (
    find_function_sha_in_repo(
        EXPECTED_ASSOCIATION_PROVIDER_SHA
    )
)


assert (
    len(
        association_provider_matches
    )
    >= 1
), (
    "Reference association provider SHA not found in current source checkout."
)


print(
    "=" * 120,
    flush=True,
)

print(
    "REFERENCE ASSOCIATION PROVIDER SOURCE AUTHORITY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in association_provider_matches:

    print(
        record[
            "path"
        ],
        "::",
        record[
            "function"
        ],
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "function_sha256"
        ],
        flush=True,
    )

    print(
        "  calls:",
        record[
            "calls"
        ],
        flush=True,
    )

    print(
        "  --- source ---",
        flush=True,
    )

    print(
        record[
            "source"
        ][
            :14000
        ],
        flush=True,
    )

    print(
        "  --- end source ---",
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 18. Materialization lineage
# =============================================================================

materialization_records = (
    statement_records_with_terms(
        {
            "morph_rank",
            "rank_in_dataset",
            "candidate_rank",
            "candidate_prob",
        }
    )
)


morph_rank_mapping_records = [
    record

    for record in materialization_records

    if record[
        "maps_rank_in_dataset_to_morph_rank"
    ]
]


candidate_rank_compute_records = [
    record

    for record in materialization_records

    if (
        "candidate_rank"
        in record[
            "terms"
        ]
        and
        record[
            "computes_rank"
        ]
        and
        not record[
            "uses_existing_row_values"
        ]
    )
]


candidate_prob_compute_records = [
    record

    for record in materialization_records

    if (
        "candidate_prob"
        in record[
            "terms"
        ]
        and
        record[
            "computes_probability"
        ]
        and
        not record[
            "uses_existing_row_values"
        ]
    )
]


print(
    "=" * 120,
    flush=True,
)

print(
    "MATERIALIZATION LINEAGE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "rank_in_dataset -> morph_rank records:",
    len(
        morph_rank_mapping_records
    ),
    flush=True,
)

print(
    "candidate_rank compute records:",
    len(
        candidate_rank_compute_records
    ),
    flush=True,
)

print(
    "candidate_prob compute records:",
    len(
        candidate_prob_compute_records
    ),
    flush=True,
)

print(
    flush=True,
)


for title, records in (
    (
        "MORPH RANK MAPPINGS",
        morph_rank_mapping_records,
    ),
    (
        "CANDIDATE RANK COMPUTATION",
        candidate_rank_compute_records,
    ),
    (
        "CANDIDATE PROB COMPUTATION",
        candidate_prob_compute_records,
    ),
):

    print(
        "---",
        title,
        "---",
        flush=True,
    )


    for record in records[
        :30
    ]:

        print(
            record[
                "path"
            ],
            (
                f"cell={record['cell_index']}"
                if record[
                    "cell_index"
                ]
                is not None
                else ""
            ),
            f"L{record['line_start']}-{record['line_end']}",
            flush=True,
        )

        print(
            "  SHA:",
            record[
                "source_sha256"
            ],
            flush=True,
        )

        print(
            "  read_pickle:",
            record[
                "read_pickle"
            ],
            flush=True,
        )

        print(
            "  source:",
            flush=True,
        )

        print(
            record[
                "source"
            ][
                :5000
            ],
            flush=True,
        )

        print(
            flush=True,
        )


# =============================================================================
# 19. Provenance component resolution
# =============================================================================

#
# Formula authority is now separated from input authority.
#

morphology_formula_closed = (
    len(
        morph_score_matches
    )
    == 1
)


dataset_rank_formula_closed = (
    len(
        rank_matches
    )
    == 1
)


fresh_morphology_prediction_source_localized = (
    len(
        clean_morph_prediction_candidates
    )
    > 0
)


association_provider_source_closed = (
    len(
        association_provider_matches
    )
    >= 1
)


morph_rank_materialization_source_localized = (
    len(
        morph_rank_mapping_records
    )
    > 0
)


candidate_rank_materialization_source_localized = (
    len(
        candidate_rank_compute_records
    )
    > 0
)


candidate_probability_materialization_source_localized = (
    len(
        candidate_prob_compute_records
    )
    > 0
)


component_status = {
    "morphology_score_formula":
        (
            "SOURCE_FORMULA_FROZEN"
            if morphology_formula_closed
            else
            "UNRESOLVED"
        ),

    "morphology_score_fresh_input":
        (
            "SOURCE_CANDIDATE_LOCALIZED_CONTRACT_FREEZE_REQUIRED"
            if fresh_morphology_prediction_source_localized
            else
            "FRESH_INPUT_SOURCE_UNRESOLVED"
        ),

    "morphology_dataset_rank_formula":
        (
            "SOURCE_FORMULA_FROZEN"
            if dataset_rank_formula_closed
            else
            "UNRESOLVED"
        ),

    "morph_rank_materialization":
        (
            "SOURCE_LINEAGE_LOCALIZED"
            if morph_rank_materialization_source_localized
            else
            "SPECIFICATION_MAPPING_STILL_REQUIRED"
        ),

    "association_provider":
        (
            "EXACT_SOURCE_AUTHORITY_AVAILABLE"
            if association_provider_source_closed
            else
            "UNRESOLVED"
        ),

    "candidate_rank_materialization":
        (
            "SOURCE_LINEAGE_LOCALIZED"
            if candidate_rank_materialization_source_localized
            else
            "ASSOCIATION_OUTPUT_TO_RANK_MAPPING_STILL_REQUIRED"
        ),

    "candidate_probability_materialization":
        (
            "SOURCE_LINEAGE_LOCALIZED"
            if candidate_probability_materialization_source_localized
            else
            "ASSOCIATION_OUTPUT_TO_PROBABILITY_MAPPING_STILL_REQUIRED"
        ),
}


remaining_blockers = [
    key

    for key, value in component_status.items()

    if (
        value
        in {
            "UNRESOLVED",
            "FRESH_INPUT_SOURCE_UNRESOLVED",
            "SPECIFICATION_MAPPING_STILL_REQUIRED",
            "ASSOCIATION_OUTPUT_TO_RANK_MAPPING_STILL_REQUIRED",
            "ASSOCIATION_OUTPUT_TO_PROBABILITY_MAPPING_STILL_REQUIRED",
        }
    )
]


input_provenance_complete = (
    remaining_blockers
    == []
)


implementation_authorized = (
    False
)


if input_provenance_complete:

    status = (
        "INPUT_PROVENANCE_COMPONENTS_LOCALIZED_"
        "EXACT_GENERATOR_CONTRACT_FREEZE_REQUIRED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_F_"
        "MORPH_GENERATOR_EXACT_INPUT_CONTRACT_FREEZE"
    )


else:

    status = (
        "INPUT_PROVENANCE_PARTIAL_"
        "MISSING_PRIMITIVE_CLOSURE_REQUIRED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_F_"
        "MORPH_GENERATOR_MISSING_PRIMITIVE_CLOSURE"
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "INPUT PROVENANCE RESOLUTION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    json.dumps(
        component_status,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "remaining blockers:",
    remaining_blockers,
    flush=True,
)

print(
    "input provenance complete:",
    input_provenance_complete,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "status:",
    status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 20. Safety
# =============================================================================

source_mutation = False

test_mutation = False

historical_pickle_loaded = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


# =============================================================================
# 21. Semantic gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_d_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "Notebook11_relevant_cells_locked":
        all(
            cells[
                index
            ][
                "sha256"
            ]
            == expected_sha

            for index, expected_sha
            in expected_cells.items()
        ),

    "morphology_score_formula_exactly_frozen":
        morphology_formula_closed,

    "dataset_rank_formula_exactly_frozen":
        dataset_rank_formula_closed,

    "historical_input_inventory_complete":
        True,

    "input_artifact_source_reference_scan_complete":
        True,

    "morphology_prediction_source_scan_complete":
        True,

    "reference_association_provider_exact_source_found":
        association_provider_source_closed,

    "materialization_lineage_scan_complete":
        True,

    "formula_vs_input_provenance_distinction_enforced":
        True,

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-E SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-E failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 22. Compact artifact records
# =============================================================================

def compact_record(
    record: dict[str, Any],
) -> dict[str, Any]:

    result = {
        key:
            value

        for key, value
        in record.items()

        if key
        != "source"
    }


    if (
        "source"
        in record
    ):

        result[
            "source_preview"
        ] = record[
            "source"
        ][
            :6000
        ]


    return result


# =============================================================================
# 23. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_E",

    "title":
        "MORPH_GENERATOR_INPUT_PROVENANCE_CLOSURE",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_INPUT_PROVENANCE_RECOVERY_ONLY",

    "parent_14_3e_d_d": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "formula_authority": {
        "morphology_score": {
            "cell":
                CELL46,

            "cell_sha256":
                CELL46_SHA,

            "statement_sha256":
                MORPH_SCORE_STATEMENT_SHA,

            "source":
                morph_score_statement[
                    "source"
                ],

            "status":
                component_status[
                    "morphology_score_formula"
                ],
        },

        "dataset_rank": {
            "cell":
                CELL46,

            "cell_sha256":
                CELL46_SHA,

            "statement_sha256":
                DATASET_RANK_STATEMENT_SHA,

            "source":
                rank_statement[
                    "source"
                ],

            "status":
                component_status[
                    "morphology_dataset_rank_formula"
                ],
        },
    },

    "historical_input_inventory":
        cell_input_records,

    "historical_input_source_references":
        artifact_source_hits,

    "morphology_prediction_source_candidates":
        [
            compact_record(
                record
            )

            for record
            in morph_prediction_candidates
        ],

    "clean_morphology_prediction_source_candidates":
        [
            compact_record(
                record
            )

            for record
            in clean_morph_prediction_candidates
        ],

    "association_provider_authority":
        [
            compact_record(
                record
            )

            for record
            in association_provider_matches
        ],

    "materialization_lineage": {
        "morph_rank_mapping":
            [
                compact_record(
                    record
                )

                for record
                in morph_rank_mapping_records
            ],

        "candidate_rank_computation":
            [
                compact_record(
                    record
                )

                for record
                in candidate_rank_compute_records
            ],

        "candidate_probability_computation":
            [
                compact_record(
                    record
                )

                for record
                in candidate_prob_compute_records
            ],
    },

    "component_status":
        component_status,

    "resolution": {
        "remaining_blockers":
            remaining_blockers,

        "input_provenance_complete":
            input_provenance_complete,

        "implementation_authorized":
            False,

        "status":
            status,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_selection_policy":
            "FROZEN",

        "Morph_graph_application_semantics":
            "FROZEN",

        "morphology_score_formula":
            component_status[
                "morphology_score_formula"
            ],

        "morphology_dataset_rank_formula":
            component_status[
                "morphology_dataset_rank_formula"
            ],

        "fresh_input_provenance":
            status,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_YET",

        "historical_action_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,

        "source_mutation":
            False,
    },

    "explicit_non_claims": [
        "Historical pickle inputs were not opened.",
        "A source formula is not treated as fresh-input provenance.",
        "No candidate materialization mapping is invented when source evidence is absent.",
        "No concrete MorphCandidateProvider was implemented.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-E artifact exists but differs "
        "from current input-provenance state."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 24. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-E PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "FORMULA AUTHORITY:",
    flush=True,
)

print(
    "  morphology score formula:",
    component_status[
        "morphology_score_formula"
    ],
    flush=True,
)

print(
    "  dataset rank formula:",
    component_status[
        "morphology_dataset_rank_formula"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "FRESH INPUT AUTHORITY:",
    flush=True,
)

print(
    "  clean p_T4 source candidates:",
    len(
        clean_morph_prediction_candidates
    ),
    flush=True,
)

print(
    "  association provider source:",
    component_status[
        "association_provider"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "MATERIALIZATION:",
    flush=True,
)

print(
    "  morph_rank:",
    component_status[
        "morph_rank_materialization"
    ],
    flush=True,
)

print(
    "  candidate_rank:",
    component_status[
        "candidate_rank_materialization"
    ],
    flush=True,
)

print(
    "  candidate_prob:",
    component_status[
        "candidate_probability_materialization"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "RESOLUTION:",
    flush=True,
)

print(
    "  remaining blockers:",
    remaining_blockers,
    flush=True,
)

print(
    "  input provenance complete:",
    input_provenance_complete,
    flush=True,
)

print(
    "  implementation authorized:",
    False,
    flush=True,
)

print(
    "  status:",
    status,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  source mutated:",
    False,
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  performance generated:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-F — MORPH GENERATOR MISSING PRIMITIVE CLOSURE
#
# PARENT
# ------
# 14.3E-D-E PASS
#
# D-E correctly established:
#
#   morphology score formula:
#       df["morph_score"] = df[rank_cols].mean(axis=1)
#
#   dataset rank formula:
#       df["rank_in_dataset"] =
#           df.groupby("dataset").cumcount() + 1
#
#   reference association provider:
#       fast_association_pair
#       SHA 7aae212...
#
# BUT D-E broad materialization scanning was intentionally exploratory and
# produced false-positive "lineage" hits from:
#
#   - Notebook13 provenance-analysis code
#   - Notebook14 provenance-analysis code
#   - MUTUAL-CONT-V1/V2
#
# Therefore:
#
#   19 morph-rank records
#   1 candidate-rank record
#   43 candidate-prob records
#
# are NOT accepted as Morph generator authority.
#
#
# THIS STAGE
# ----------
# Restrict authority discovery to:
#
#   notebooks/11_division_association_audit.ipynb
#   code cells <= semantic freeze cell 63
#
# Recover exact source semantics for:
#
#   1. rank_cols construction
#   2. p_T4_ordered_* production status
#   3. morphology score ordering / dataset ranking
#   4. rank_in_dataset -> morph_rank mapping
#   5. association probs -> candidate_prob mapping
#   6. association probs -> candidate_rank mapping
#   7. action-row source / target identity mapping
#
# It distinguishes:
#
#   VALUE_PRODUCER
#   VALUE_MAPPING
#   EXISTING_VALUE_CONSUMER
#   PROVENANCE_META_CODE
#
#
# IMPORTANT
# ---------
# No historical pickle is deserialized.
#
# Merely finding a `.to_pickle(...)` statement does not make its surrounding
# cell a fresh runtime authority.
#
# No implementation is authorized by this stage.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical pickle loading
#   - GT loading
#   - evaluator execution
#   - performance computation
#   - parameter search
#   - provider implementation
#   - structural fidelity claim
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_e_"
    "morph_generator_input_provenance_closure_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


NOTEBOOK11_REL = Path(
    "notebooks/"
    "11_division_association_audit.ipynb"
)

NOTEBOOK11 = ROOT / NOTEBOOK11_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_f_"
    "morph_generator_missing_primitive_closure_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_DD_FILE_SHA = (
    "ed211e95612a4ad20ad98a234905050ac5274d9c5cc3a63097ff525212bd21a0"
)

EXPECTED_DD_CONTENT_SHA = (
    "da66faef4b52759cd33b43017550c73ce23b44689139d2ceae5cf22af4383bfa"
)


# D-E output was truncated after artifact creation, so its exact file/content
# SHA were not visible in the returned log.
#
# This stage therefore:
#   - exact-locks D-E by internal content_sha self-consistency
#   - exact-locks its D-D parent identity
#   - exact-locks its stage/decision/next-stage
#   - captures D-E's actual file SHA into THIS formal artifact
#
# This is deliberate; we do not invent a SHA that was not observed.


CELL46 = 46

CELL46_SHA = (
    "15747465b4e39aa4a2e72e259a9cdbbdfa83c05395c939904b6b6a4f436f36b1"
)


CELL51 = 51

CELL51_SHA = (
    "0be0b5b973555fbdd1b969374842bede150f1d80895596d498b433fb664ccb1a"
)


CELL55 = 55

CELL55_SHA = (
    "97c347aea36888e1d286c8d54e8adfc902dc7f4e86e1e2989009caa86a8025fc"
)


FREEZE_CELL = 63

FREEZE_CELL_SHA = (
    "2e6bdff585ad1a3ee61b3a82ec03ccbd05db0b66f01024a35a721d76a29d8834"
)


MORPH_SCORE_STATEMENT_SHA = (
    "1eb422867db49a6255e89ae08443846a0c03458f484a9cc61065375800e7bad4"
)

DATASET_RANK_STATEMENT_SHA = (
    "7606a5627025b8834d6025ab39db54048127241c46b1b8ce301068adabcc698f"
)


EXPECTED_ASSOCIATION_PROVIDER_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


EXPECTED_NEXT_FROM_DE = (
    "NOTEBOOK14_14_3E_D_F_"
    "MORPH_GENERATOR_MISSING_PRIMITIVE_CLOSURE"
)


MORPH_SCHEMA = (
    "dataset",
    "morph_rank",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def node_source(
    source: str,
    node: ast.AST,
) -> str:

    value = ast.get_source_segment(
        source,
        node,
    )


    assert (
        value is not None
    )


    return value


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def notebook_cells(
    path: Path,
) -> list[
    dict[str, Any]
]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    records = []


    for index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        records.append({
            "index":
                index,

            "cell_type":
                cell.get(
                    "cell_type"
                ),

            "source":
                source,

            "sha256":
                sha256_text(
                    source
                ),
        })


    return records


def top_level_functions(
    tree: ast.Module,
) -> dict[
    str,
    ast.FunctionDef | ast.AsyncFunctionDef,
]:

    return {
        node.name:
            node

        for node in tree.body

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        )
    }


def calls_in_node(
    node: ast.AST,
) -> list[str]:

    return sorted({
        dotted_name(
            child.func
        )
        or "<dynamic-call>"

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Call,
        )
    })


def has_read_pickle(
    node: ast.AST,
) -> bool:

    return any(
        (
            isinstance(
                child,
                ast.Call,
            )
            and
            (
                dotted_name(
                    child.func
                )
                or ""
            ).endswith(
                "read_pickle"
            )
        )

        for child in ast.walk(
            node
        )
    )


# =============================================================================
# 3. D-E parent self-consistency lock
# =============================================================================

def verify_parent(
    path: Path,
) -> tuple[
    dict[str, Any],
    str,
    str,
]:

    assert (
        path.exists()
    ), path


    file_sha = sha256_file(
        path
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "stage"
        )
        == "NOTEBOOK14_14_3E_D_E"
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    content_sha = obj.get(
        "content_sha256"
    )


    assert (
        isinstance(
            content_sha,
            str,
        )
        and
        re.fullmatch(
            r"[0-9a-f]{64}",
            content_sha,
        )
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == content_sha
    ), (
        "D-E internal content SHA mismatch."
    )


    assert (
        obj[
            "parent_14_3e_d_d"
        ][
            "file_sha256"
        ]
        == EXPECTED_DD_FILE_SHA
    )


    assert (
        obj[
            "parent_14_3e_d_d"
        ][
            "content_sha256"
        ]
        == EXPECTED_DD_CONTENT_SHA
    )


    assert (
        obj[
            "resolution"
        ][
            "implementation_authorized"
        ]
        is False
    )


    assert (
        obj.get(
            "next_stage"
        )
        == EXPECTED_NEXT_FROM_DE
    )


    assert (
        obj[
            "claim_boundary"
        ][
            "historical_pickle_loaded"
        ]
        is False
    )


    return (
        obj,
        file_sha,
        content_sha,
    )


# =============================================================================
# 4. Exact statement finder
# =============================================================================

def find_node_by_sha(
    source: str,
    expected_sha: str,
) -> list[
    dict[str, Any]
]:

    tree = ast.parse(
        source
    )


    records = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
                ast.For,
                ast.If,
                ast.Expr,
            ),
        ):

            continue


        segment = ast.get_source_segment(
            source,
            node,
        )


        if segment is None:

            continue


        observed = sha256_text(
            segment
        )


        if observed != expected_sha:

            continue


        records.append({
            "node":
                node,

            "kind":
                type(
                    node
                ).__name__,

            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "source":
                segment,

            "sha256":
                observed,
        })


    return records


# =============================================================================
# 5. Parent map / enclosing statement helpers
# =============================================================================

def parent_map(
    tree: ast.AST,
) -> dict[
    ast.AST,
    ast.AST,
]:

    result = {}


    for parent in ast.walk(
        tree
    ):

        for child in ast.iter_child_nodes(
            parent
        ):

            result[
                child
            ] = parent


    return result


def nearest_ancestor(
    node: ast.AST,
    parents: dict[
        ast.AST,
        ast.AST,
    ],
    classes: tuple[
        type,
        ...,
    ],
) -> ast.AST | None:

    current = node


    while current in parents:

        current = parents[
            current
        ]


        if isinstance(
            current,
            classes,
        ):

            return current


    return None


# =============================================================================
# 6. Dict action-construction extraction
# =============================================================================

ACTION_KEYS = {
    "dataset",
    "morph_rank",
    "rank_in_dataset",
    "t",
    "source_pred_node_id",
    "target_pred_node_id",
    "candidate_rank",
    "candidate_prob",
}


def literal_dict_mapping(
    node: ast.Dict,
) -> dict[
    str,
    ast.AST,
]:

    result = {}


    for key_node, value_node in zip(
        node.keys,
        node.values,
    ):

        if key_node is None:

            continue


        try:

            key = ast.literal_eval(
                key_node
            )

        except Exception:

            continue


        if not isinstance(
            key,
            str,
        ):

            continue


        result[
            key
        ] = value_node


    return result


def action_dict_records(
    cells: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            FREEZE_CELL
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        parents = parent_map(
            tree
        )


        for dictionary in ast.walk(
            tree
        ):

            if not isinstance(
                dictionary,
                ast.Dict,
            ):

                continue


            mapping = literal_dict_mapping(
                dictionary
            )


            keys = set(
                mapping
            )


            morph_identity_hits = (
                keys
                &
                {
                    "morph_rank",
                    "candidate_rank",
                    "candidate_prob",
                    "rank_in_dataset",
                }
            )


            endpoint_hits = (
                keys
                &
                {
                    "source_pred_node_id",
                    "target_pred_node_id",
                }
            )


            if (
                not morph_identity_hits
                or
                not endpoint_hits
            ):

                continue


            enclosing_for = nearest_ancestor(
                dictionary,
                parents,
                (
                    ast.For,
                    ast.AsyncFor,
                ),
            )


            enclosing_stmt = nearest_ancestor(
                dictionary,
                parents,
                (
                    ast.Assign,
                    ast.AnnAssign,
                    ast.Expr,
                    ast.For,
                    ast.AsyncFor,
                ),
            )


            if enclosing_for is not None:

                context_node = (
                    enclosing_for
                )


            elif enclosing_stmt is not None:

                context_node = (
                    enclosing_stmt
                )


            else:

                context_node = (
                    dictionary
                )


            dict_source = node_source(
                source,
                dictionary,
            )


            context_source = node_source(
                source,
                context_node,
            )


            value_expressions = {
                key:
                    ast.unparse(
                        value
                    )

                for key, value
                in mapping.items()

                if key
                in ACTION_KEYS
            }


            uses_existing_action_row = any(
                expression.startswith(
                    "row."
                )
                or
                ".candidate_prob"
                in expression
                or
                ".candidate_rank"
                in expression
                or
                ".morph_rank"
                in expression

                for expression
                in value_expressions.values()
            )


            records.append({
                "cell_index":
                    cell[
                        "index"
                    ],

                "cell_sha256":
                    cell[
                        "sha256"
                    ],

                "dict_line":
                    dictionary.lineno,

                "dict_sha256":
                    sha256_text(
                        dict_source
                    ),

                "keys":
                    sorted(
                        keys
                    ),

                "value_expressions":
                    value_expressions,

                "uses_existing_action_row":
                    uses_existing_action_row,

                "context_kind":
                    type(
                        context_node
                    ).__name__,

                "context_line_start":
                    context_node.lineno,

                "context_line_end":
                    getattr(
                        context_node,
                        "end_lineno",
                        context_node.lineno,
                    ),

                "context_sha256":
                    sha256_text(
                        context_source
                    ),

                "context_calls":
                    calls_in_node(
                        context_node
                    ),

                "context_read_pickle":
                    has_read_pickle(
                        context_node
                    ),

                "context_source":
                    context_source,
            })


    # Exact-source dedup.
    unique = {}


    for record in records:

        key = (
            record[
                "cell_index"
            ],
            record[
                "dict_sha256"
            ],
            record[
                "context_sha256"
            ],
        )


        unique.setdefault(
            key,
            record,
        )


    return list(
        unique.values()
    )


# =============================================================================
# 7. Relevant statement extraction from exact Stage11 cells
# =============================================================================

def relevant_statement_records(
    cell: dict[str, Any],
    terms: tuple[str, ...],
) -> list[
    dict[str, Any]
]:

    source = cell[
        "source"
    ]


    tree = ast.parse(
        source
    )


    records = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
                ast.For,
                ast.If,
                ast.Expr,
            ),
        ):

            continue


        segment = ast.get_source_segment(
            source,
            node,
        )


        if not segment:

            continue


        hits = [
            term

            for term in terms

            if term
            in segment
        ]


        if not hits:

            continue


        records.append({
            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "kind":
                type(
                    node
                ).__name__,

            "source_sha256":
                sha256_text(
                    segment
                ),

            "hits":
                hits,

            "calls":
                calls_in_node(
                    node
                ),

            "read_pickle":
                has_read_pickle(
                    node
                ),

            "source":
                segment,
        })


    # Prefer smallest exact statements over huge enclosing loops.
    records = sorted(
        records,
        key=lambda record: (
            (
                record[
                    "line_end"
                ]
                -
                record[
                    "line_start"
                ]
            ),
            record[
                "line_start"
            ],
        ),
    )


    unique = {}


    for record in records:

        unique.setdefault(
            record[
                "source_sha256"
            ],
            record,
        )


    return list(
        unique.values()
    )


# =============================================================================
# 8. Detect production of p_T4_ordered_* before Morph freeze
# =============================================================================

def p_t4_producer_records(
    cells: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            FREEZE_CELL
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.Assign,
                    ast.AnnAssign,
                    ast.AugAssign,
                ),
            ):

                continue


            segment = ast.get_source_segment(
                source,
                node,
            )


            if not segment:

                continue


            # Production means p_T4_ordered appears on assignment target,
            # not merely in a reader-side rank_cols list.
            targets = []


            if isinstance(
                node,
                ast.Assign,
            ):

                targets = list(
                    node.targets
                )


            else:

                targets = [
                    node.target
                ]


            target_source = " ".join(
                ast.unparse(
                    target
                )

                for target in targets
            )


            if (
                "p_T4_ordered_"
                not in target_source
            ):

                continue


            records.append({
                "cell_index":
                    cell[
                        "index"
                    ],

                "cell_sha256":
                    cell[
                        "sha256"
                    ],

                "line_start":
                    node.lineno,

                "line_end":
                    getattr(
                        node,
                        "end_lineno",
                        node.lineno,
                    ),

                "source_sha256":
                    sha256_text(
                        segment
                    ),

                "target":
                    target_source,

                "calls":
                    calls_in_node(
                        node
                    ),

                "read_pickle":
                    has_read_pickle(
                        node
                    ),

                "source":
                    segment,
            })


    return records


# =============================================================================
# 9. Detect rank_cols exact definitions
# =============================================================================

def rank_cols_records(
    cell: dict[str, Any],
) -> list[
    dict[str, Any]
]:

    source = cell[
        "source"
    ]


    tree = ast.parse(
        source
    )


    records = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
            ),
        ):

            continue


        target_names = []


        if isinstance(
            node,
            ast.Assign,
        ):

            for target in node.targets:

                if isinstance(
                    target,
                    ast.Name,
                ):

                    target_names.append(
                        target.id
                    )


        elif isinstance(
            node.target,
            ast.Name,
        ):

            target_names.append(
                node.target.id
            )


        if (
            "rank_cols"
            not in target_names
        ):

            continue


        segment = node_source(
            source,
            node,
        )


        records.append({
            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "source_sha256":
                sha256_text(
                    segment
                ),

            "expression":
                ast.unparse(
                    node.value
                ),

            "source":
                segment,
        })


    return records


# =============================================================================
# 10. Exact association provider source lock
# =============================================================================

def repo_python_files() -> list[Path]:

    result = []


    for root in (
        ROOT / "src",
        ROOT / "scripts",
    ):

        if not root.exists():

            continue


        result.extend(
            path

            for path in root.rglob(
                "*.py"
            )

            if path.is_file()
        )


    return sorted(
        result
    )


def find_function_by_sha(
    expected_sha: str,
) -> list[
    dict[str, Any]
]:

    records = []


    for path in repo_python_files():

        try:

            source = path.read_text(
                encoding="utf-8"
            )


            tree = ast.parse(
                source,
                filename=str(
                    path
                ),
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                ),
            ):

                continue


            segment = ast.get_source_segment(
                source,
                node,
            )


            if segment is None:

                continue


            observed = sha256_text(
                segment
            )


            if observed != expected_sha:

                continue


            records.append({
                "path":
                    str(
                        path.relative_to(
                            ROOT
                        )
                    ),

                "file_sha256":
                    sha256_file(
                        path
                    ),

                "function":
                    node.name,

                "function_sha256":
                    observed,

                "line_start":
                    node.lineno,

                "line_end":
                    node.end_lineno,

                "source":
                    segment,
            })


    return records


# =============================================================================
# 11. Association rank / probability semantics in Stage11 only
# =============================================================================

def association_semantic_blocks(
    cells: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    records = []


    signals = (
        "candidate_rank",
        "candidate_prob",
        "argsort",
        "argmax",
        "probs",
        "association",
        "source_pred_node_id",
        "target_pred_node_id",
    )


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            FREEZE_CELL
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.For,
                    ast.Assign,
                    ast.AnnAssign,
                    ast.Expr,
                ),
            ):

                continue


            segment = ast.get_source_segment(
                source,
                node,
            )


            if not segment:

                continue


            hits = sorted({
                signal

                for signal in signals

                if signal
                in segment
            })


            # Require real association/candidate semantic density.
            if (
                len(
                    hits
                )
                <
                3
            ):

                continue


            consumer_only = any(
                token
                in segment

                for token in (
                    "row.candidate_prob",
                    "row.candidate_rank",
                )
            )


            ranking_evidence = any(
                token
                in segment

                for token in (
                    "argsort",
                    "argmax",
                    "sort_values",
                    "enumerate",
                    "rank(",
                )
            )


            probability_index_evidence = (
                "probs["
                in segment
            )


            records.append({
                "cell_index":
                    cell[
                        "index"
                    ],

                "cell_sha256":
                    cell[
                        "sha256"
                    ],

                "line_start":
                    node.lineno,

                "line_end":
                    getattr(
                        node,
                        "end_lineno",
                        node.lineno,
                    ),

                "source_sha256":
                    sha256_text(
                        segment
                    ),

                "hits":
                    hits,

                "consumer_only":
                    consumer_only,

                "ranking_evidence":
                    ranking_evidence,

                "probability_index_evidence":
                    probability_index_evidence,

                "read_pickle":
                    has_read_pickle(
                        node
                    ),

                "calls":
                    calls_in_node(
                        node
                    ),

                "source":
                    segment,
            })


    # Deduplicate.
    unique = {}


    for record in records:

        key = (
            record[
                "cell_index"
            ],
            record[
                "source_sha256"
            ],
        )


        unique.setdefault(
            key,
            record,
        )


    return sorted(
        unique.values(),
        key=lambda record: (
            record[
                "cell_index"
            ],
            record[
                "line_start"
            ],
        ),
    )


# =============================================================================
# 12. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    NOTEBOOK11,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-F "
    "MORPH GENERATOR MISSING PRIMITIVE CLOSURE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "NOTEBOOK11 AUTHORITY WINDOW:",
    f"cells 0..{FREEZE_CELL}",
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Parent lock
# =============================================================================

(
    parent,
    parent_file_sha,
    parent_content_sha,
) = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_D_E_SELF_CONSISTENT_EXACT_CAPTURE",
    flush=True,
)

print(
    "  actual D-E file SHA:",
    parent_file_sha,
    flush=True,
)

print(
    "  actual D-E content SHA:",
    parent_content_sha,
    flush=True,
)

print(
    "  D-E parent D-D file SHA:",
    EXPECTED_DD_FILE_SHA,
    flush=True,
)

print(
    "  D-E parent D-D content SHA:",
    EXPECTED_DD_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Current source / Notebook11 locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


cells = notebook_cells(
    NOTEBOOK11
)


for index, expected_sha in (
    (
        CELL46,
        CELL46_SHA,
    ),
    (
        CELL51,
        CELL51_SHA,
    ),
    (
        CELL55,
        CELL55_SHA,
    ),
    (
        FREEZE_CELL,
        FREEZE_CELL_SHA,
    ),
):

    assert (
        cells[
            index
        ][
            "sha256"
        ]
        == expected_sha
    )


print(
    "[PASS] STAGE11_AUTHORITY_WINDOW_LOCK",
    flush=True,
)

print(
    "  cell46:",
    CELL46_SHA,
    flush=True,
)

print(
    "  cell51:",
    CELL51_SHA,
    flush=True,
)

print(
    "  cell55:",
    CELL55_SHA,
    flush=True,
)

print(
    "  freeze cell63:",
    FREEZE_CELL_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Re-lock formulas from D-D / D-E
# =============================================================================

score_nodes = find_node_by_sha(
    cells[
        CELL46
    ][
        "source"
    ],
    MORPH_SCORE_STATEMENT_SHA,
)


rank_nodes = find_node_by_sha(
    cells[
        CELL46
    ][
        "source"
    ],
    DATASET_RANK_STATEMENT_SHA,
)


assert (
    len(
        score_nodes
    )
    == 1
)


assert (
    len(
        rank_nodes
    )
    == 1
)


print(
    "[PASS] MORPHOLOGY_FORMULA_AUTHORITY_RELOCK",
    flush=True,
)

print(
    "  morph score SHA:",
    MORPH_SCORE_STATEMENT_SHA,
    flush=True,
)

print(
    "  dataset rank SHA:",
    DATASET_RANK_STATEMENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. rank_cols exact authority
# =============================================================================

rank_cols = rank_cols_records(
    cells[
        CELL46
    ]
)


print(
    "=" * 120,
    flush=True,
)

print(
    "CELL46 RANK_COLS AUTHORITY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "definitions:",
    len(
        rank_cols
    ),
    flush=True,
)


for record in rank_cols:

    print(
        "  SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "  expression:",
        record[
            "expression"
        ],
        flush=True,
    )

    print(
        "  source:",
        record[
            "source"
        ],
        flush=True,
    )


print(
    flush=True,
)


assert (
    len(
        rank_cols
    )
    >= 1
), (
    "No rank_cols definition found in exact cell46."
)


# =============================================================================
# 17. p_T4 production status
# =============================================================================

p_t4_producers = p_t4_producer_records(
    cells
)


print(
    "=" * 120,
    flush=True,
)

print(
    "P_T4_ORDERED PRODUCTION BEFORE MORPH FREEZE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "producer assignments:",
    len(
        p_t4_producers
    ),
    flush=True,
)


for record in p_t4_producers:

    print(
        f"  cell={record['cell_index']}",
        f"L{record['line_start']}-{record['line_end']}",
        flush=True,
    )

    print(
        "    SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "    target:",
        record[
            "target"
        ],
        flush=True,
    )

    print(
        "    read_pickle:",
        record[
            "read_pickle"
        ],
        flush=True,
    )

    print(
        "    source:",
        record[
            "source"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 18. Exact relevant source from cells 46 / 51 / 55
# =============================================================================

CELL46_TERMS = (
    "rank_cols",
    "p_T4_ordered_",
    "morph_score",
    "rank_in_dataset",
    "sort_values",
)


CELL51_TERMS = (
    "rank_in_dataset",
    "candidate_rank",
    "candidate_prob",
    "source_pred_node_id",
    "target_pred_node_id",
    "action",
    "prob_threshold",
)


CELL55_TERMS = (
    "rank_in_dataset",
    "morph_score",
    "p_T4_ordered_",
    "candidate_rank",
    "candidate_prob",
    "predcenter",
)


relevant46 = relevant_statement_records(
    cells[
        CELL46
    ],
    CELL46_TERMS,
)


relevant51 = relevant_statement_records(
    cells[
        CELL51
    ],
    CELL51_TERMS,
)


relevant55 = relevant_statement_records(
    cells[
        CELL55
    ],
    CELL55_TERMS,
)


for label, records in (
    (
        "CELL46",
        relevant46,
    ),
    (
        "CELL51",
        relevant51,
    ),
    (
        "CELL55",
        relevant55,
    ),
):

    print(
        "=" * 120,
        flush=True,
    )

    print(
        label,
        "RELEVANT EXACT STATEMENTS",
        flush=True,
    )

    print(
        "=" * 120,
        flush=True,
    )


    for record in records[
        :40
    ]:

        print(
            f"L{record['line_start']}-{record['line_end']}",
            record[
                "hits"
            ],
            flush=True,
        )

        print(
            "  SHA:",
            record[
                "source_sha256"
            ],
            flush=True,
        )

        print(
            "  read_pickle:",
            record[
                "read_pickle"
            ],
            flush=True,
        )

        print(
            "  calls:",
            record[
                "calls"
            ],
            flush=True,
        )

        print(
            "  source:",
            flush=True,
        )

        print(
            record[
                "source"
            ][
                :7000
            ],
            flush=True,
        )

        print(
            flush=True,
        )


# =============================================================================
# 19. Exact Morph-like action dicts BEFORE freeze
# =============================================================================

action_dicts = action_dict_records(
    cells
)


print(
    "=" * 120,
    flush=True,
)

print(
    "STAGE11 MORPH-LIKE ACTION CONSTRUCTION DICTS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "records:",
    len(
        action_dicts
    ),
    flush=True,
)

print(
    flush=True,
)


for index, record in enumerate(
    action_dicts,
    start=1,
):

    print(
        f"[{index}] cell={record['cell_index']}",
        f"L{record['context_line_start']}-{record['context_line_end']}",
        flush=True,
    )

    print(
        "  dict SHA:",
        record[
            "dict_sha256"
        ],
        flush=True,
    )

    print(
        "  context SHA:",
        record[
            "context_sha256"
        ],
        flush=True,
    )

    print(
        "  values:",
        json.dumps(
            record[
                "value_expressions"
            ],
            indent=2,
        ),
        flush=True,
    )

    print(
        "  existing-action-row consumer:",
        record[
            "uses_existing_action_row"
        ],
        flush=True,
    )

    print(
        "  context read_pickle:",
        record[
            "context_read_pickle"
        ],
        flush=True,
    )

    print(
        "  context calls:",
        record[
            "context_calls"
        ],
        flush=True,
    )

    print(
        "  --- context ---",
        flush=True,
    )

    print(
        record[
            "context_source"
        ][
            :10000
        ],
        flush=True,
    )

    print(
        "  --- end context ---",
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 20. Extract concrete mapping evidence
# =============================================================================

morph_rank_mapping_evidence = []

candidate_rank_mapping_evidence = []

candidate_prob_mapping_evidence = []


for record in action_dicts:

    expressions = record[
        "value_expressions"
    ]


    morph_expr = expressions.get(
        "morph_rank"
    )


    candidate_rank_expr = expressions.get(
        "candidate_rank"
    )


    candidate_prob_expr = expressions.get(
        "candidate_prob"
    )


    if (
        morph_expr is not None
        and
        not record[
            "uses_existing_action_row"
        ]
    ):

        morph_rank_mapping_evidence.append({
            "cell_index":
                record[
                    "cell_index"
                ],

            "context_sha256":
                record[
                    "context_sha256"
                ],

            "expression":
                morph_expr,

            "context_source":
                record[
                    "context_source"
                ],
        })


    if (
        candidate_rank_expr is not None
        and
        not record[
            "uses_existing_action_row"
        ]
    ):

        candidate_rank_mapping_evidence.append({
            "cell_index":
                record[
                    "cell_index"
                ],

            "context_sha256":
                record[
                    "context_sha256"
                ],

            "expression":
                candidate_rank_expr,

            "context_source":
                record[
                    "context_source"
                ],
        })


    if (
        candidate_prob_expr is not None
        and
        not record[
            "uses_existing_action_row"
        ]
    ):

        candidate_prob_mapping_evidence.append({
            "cell_index":
                record[
                    "cell_index"
                ],

            "context_sha256":
                record[
                    "context_sha256"
                ],

            "expression":
                candidate_prob_expr,

            "context_source":
                record[
                    "context_source"
                ],
        })


print(
    "=" * 120,
    flush=True,
)

print(
    "DIRECT ACTION-FIELD MAPPING EVIDENCE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "morph_rank mappings:",
    len(
        morph_rank_mapping_evidence
    ),
    flush=True,
)

print(
    "candidate_rank mappings:",
    len(
        candidate_rank_mapping_evidence
    ),
    flush=True,
)

print(
    "candidate_prob mappings:",
    len(
        candidate_prob_mapping_evidence
    ),
    flush=True,
)

print(
    flush=True,
)


for name, evidence in (
    (
        "MORPH_RANK",
        morph_rank_mapping_evidence,
    ),
    (
        "CANDIDATE_RANK",
        candidate_rank_mapping_evidence,
    ),
    (
        "CANDIDATE_PROB",
        candidate_prob_mapping_evidence,
    ),
):

    print(
        name,
        flush=True,
    )


    for record in evidence:

        print(
            "  cell:",
            record[
                "cell_index"
            ],
            flush=True,
        )

        print(
            "  expression:",
            record[
                "expression"
            ],
            flush=True,
        )

        print(
            "  context SHA:",
            record[
                "context_sha256"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 21. Association-specific source blocks in Stage11 authority window
# =============================================================================

association_blocks = association_semantic_blocks(
    cells
)


print(
    "=" * 120,
    flush=True,
)

print(
    "STAGE11 ASSOCIATION RANK/PROB SEMANTIC BLOCKS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "records:",
    len(
        association_blocks
    ),
    flush=True,
)

print(
    flush=True,
)


for record in association_blocks[
    :40
]:

    print(
        f"cell={record['cell_index']}",
        f"L{record['line_start']}-{record['line_end']}",
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "  hits:",
        record[
            "hits"
        ],
        flush=True,
    )

    print(
        "  consumer only:",
        record[
            "consumer_only"
        ],
        flush=True,
    )

    print(
        "  ranking evidence:",
        record[
            "ranking_evidence"
        ],
        flush=True,
    )

    print(
        "  probability-index evidence:",
        record[
            "probability_index_evidence"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        record[
            "read_pickle"
        ],
        flush=True,
    )

    print(
        "  source:",
        flush=True,
    )

    print(
        record[
            "source"
        ][
            :8000
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 22. Reference association provider exact lock
# =============================================================================

association_provider_matches = (
    find_function_by_sha(
        EXPECTED_ASSOCIATION_PROVIDER_SHA
    )
)


assert (
    len(
        association_provider_matches
    )
    >= 1
)


print(
    "[PASS] REFERENCE_ASSOCIATION_PROVIDER_EXACT_SOURCE_LOCK",
    flush=True,
)

for record in association_provider_matches:

    print(
        " ",
        record[
            "path"
        ],
        "::",
        record[
            "function"
        ],
        flush=True,
    )

    print(
        "    SHA:",
        record[
            "function_sha256"
        ],
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 23. Strict primitive classification
# =============================================================================

#
# p_T4 fresh producer:
#
# If there is no assignment target for p_T4_ordered_* in the Stage11 authority
# window, its values are reader-side historical features, not fresh authority.
#

fresh_p_t4_producer_closed = (
    len(
        [
            record

            for record
            in p_t4_producers

            if not record[
                "read_pickle"
            ]
        ]
    )
    > 0
)


#
# morph_rank mapping:
#
# Strong source evidence = an action-construction expression before freeze,
# not a Notebook13/14 provenance marker.
#

morph_rank_mapping_closed = (
    len(
        morph_rank_mapping_evidence
    )
    > 0
)


#
# candidate fields:
#
# We require direct action-row mapping evidence plus Stage11 ranking/probability
# blocks. This prevents a random later `candidate_prob` policy from qualifying.
#

stage11_rank_generation_blocks = [
    record

    for record
    in association_blocks

    if (
        not record[
            "consumer_only"
        ]
        and
        record[
            "ranking_evidence"
        ]
    )
]


stage11_prob_generation_blocks = [
    record

    for record
    in association_blocks

    if (
        not record[
            "consumer_only"
        ]
        and
        record[
            "probability_index_evidence"
        ]
    )
]


candidate_rank_mapping_closed = (
    bool(
        candidate_rank_mapping_evidence
    )
    and
    bool(
        stage11_rank_generation_blocks
    )
)


candidate_prob_mapping_closed = (
    bool(
        candidate_prob_mapping_evidence
    )
    and
    bool(
        stage11_prob_generation_blocks
    )
)


component_status = {
    "morphology_score_formula":
        "FROZEN",

    "dataset_rank_formula":
        "FROZEN",

    "rank_cols_definition":
        (
            "SOURCE_BACKED"
            if rank_cols
            else
            "UNRESOLVED"
        ),

    "p_T4_ordered_fresh_producer":
        (
            "SOURCE_BACKED"
            if fresh_p_t4_producer_closed
            else
            "UNRESOLVED_HISTORICAL_FEATURE_INPUT_ONLY"
        ),

    "morph_rank_materialization":
        (
            "SOURCE_BACKED"
            if morph_rank_mapping_closed
            else
            "UNRESOLVED"
        ),

    "reference_association_provider":
        "EXACT_SOURCE_AUTHORITY",

    "candidate_rank_materialization":
        (
            "SOURCE_BACKED"
            if candidate_rank_mapping_closed
            else
            "UNRESOLVED"
        ),

    "candidate_probability_materialization":
        (
            "SOURCE_BACKED"
            if candidate_prob_mapping_closed
            else
            "UNRESOLVED"
        ),
}


remaining_blockers = [
    key

    for key, value
    in component_status.items()

    if (
        value.startswith(
            "UNRESOLVED"
        )
    )
]


# =============================================================================
# 24. Resolution
# =============================================================================

if (
    remaining_blockers
    ==
    [
        "p_T4_ordered_fresh_producer"
    ]
):

    status = (
        "ONLY_FRESH_MORPHOLOGY_PROBABILITY_VECTOR_"
        "PRODUCER_REMAINS_UNRESOLVED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_G_"
        "MORPHOLOGY_PROBABILITY_VECTOR_SOURCE_AUTHORITY_CLOSURE"
    )


elif not remaining_blockers:

    status = (
        "ALL_GENERATOR_PRIMITIVES_SOURCE_CLOSED_"
        "EXACT_PROVIDER_CONTRACT_FREEZE_REQUIRED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_G_"
        "MORPH_CONCRETE_PROVIDER_EXACT_CONTRACT_FREEZE"
    )


else:

    status = (
        "MORPH_GENERATOR_PRIMITIVES_PARTIALLY_CLOSED_"
        "TARGETED_CLOSURE_REQUIRED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_G_"
        "MORPH_REMAINING_PRIMITIVE_TARGETED_CLOSURE"
    )


implementation_authorized = (
    False
)


print(
    "=" * 120,
    flush=True,
)

print(
    "MISSING PRIMITIVE CLOSURE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    json.dumps(
        component_status,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)

print(
    "remaining blockers:",
    remaining_blockers,
    flush=True,
)

print(
    "status:",
    status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 25. Safety / no mutation
# =============================================================================

source_mutation = False

test_mutation = False

historical_pickle_loaded = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


# =============================================================================
# 26. Formal gates
# =============================================================================

semantic_gates = {
    "parent_D_E_self_consistent_and_exactly_captured":
        True,

    "parent_D_E_has_correct_D_D_parent":
        (
            parent[
                "parent_14_3e_d_d"
            ][
                "file_sha256"
            ]
            == EXPECTED_DD_FILE_SHA
            and
            parent[
                "parent_14_3e_d_d"
            ][
                "content_sha256"
            ]
            == EXPECTED_DD_CONTENT_SHA
        ),

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "Stage11_authority_window_locked":
        True,

    "morphology_score_formula_locked":
        (
            len(
                score_nodes
            )
            == 1
        ),

    "dataset_rank_formula_locked":
        (
            len(
                rank_nodes
            )
            == 1
        ),

    "rank_cols_definition_inspected":
        (
            len(
                rank_cols
            )
            >= 1
        ),

    "p_T4_target_producer_scan_complete":
        True,

    "Stage11_action_dict_scan_complete":
        True,

    "Stage11_association_semantic_scan_complete":
        True,

    "association_provider_exact_source_locked":
        (
            len(
                association_provider_matches
            )
            >= 1
        ),

    "Notebook13_14_meta_code_excluded_from_materialization_authority":
        True,

    "post_freeze_cells_excluded_from_Morph_generator_authority":
        True,

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-F SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-F failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 27. Compact artifact helpers
# =============================================================================

def compact(
    record: dict[str, Any],
) -> dict[str, Any]:

    result = {
        key:
            value

        for key, value
        in record.items()

        if key
        not in {
            "node",
            "context_source",
            "source",
        }
    }


    if (
        "context_source"
        in record
    ):

        result[
            "context_source_preview"
        ] = (
            record[
                "context_source"
            ][
                :8000
            ]
        )


    if (
        "source"
        in record
    ):

        result[
            "source_preview"
        ] = (
            record[
                "source"
            ][
                :6000
            ]
        )


    return result


# =============================================================================
# 28. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_F",

    "title":
        "MORPH_GENERATOR_MISSING_PRIMITIVE_CLOSURE",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_MORPH_PRIMITIVE_SOURCE_CLOSURE_ONLY",

    "parent_14_3e_d_e": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            parent_file_sha,

        "content_sha256":
            parent_content_sha,

        "lock_mode":
            (
                "SELF_CONSISTENT_CONTENT_SHA_PLUS_"
                "KNOWN_DD_PARENT_PLUS_STAGE_DECISION_NEXT_LOCK"
            ),
    },

    "source_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha_before,
        },

        "Notebook11": {
            "path":
                str(
                    NOTEBOOK11_REL
                ),

            "cell46_sha256":
                CELL46_SHA,

            "cell51_sha256":
                CELL51_SHA,

            "cell55_sha256":
                CELL55_SHA,

            "freeze_cell63_sha256":
                FREEZE_CELL_SHA,
        },
    },

    "formula_authority": {
        "morphology_score_statement_sha256":
            MORPH_SCORE_STATEMENT_SHA,

        "dataset_rank_statement_sha256":
            DATASET_RANK_STATEMENT_SHA,

        "rank_cols":
            rank_cols,
    },

    "p_T4_ordered_producer_records":
        [
            compact(
                record
            )

            for record
            in p_t4_producers
        ],

    "Stage11_action_construction_records":
        [
            compact(
                record
            )

            for record
            in action_dicts
        ],

    "direct_field_mapping_evidence": {
        "morph_rank":
            morph_rank_mapping_evidence,

        "candidate_rank":
            candidate_rank_mapping_evidence,

        "candidate_prob":
            candidate_prob_mapping_evidence,
    },

    "Stage11_association_semantic_blocks":
        [
            compact(
                record
            )

            for record
            in association_blocks
        ],

    "reference_association_provider": {
        "identity_sha256":
            EXPECTED_ASSOCIATION_PROVIDER_SHA,

        "matches":
            [
                compact(
                    record
                )

                for record
                in association_provider_matches
            ],
    },

    "broad_D_E_materialization_scan_repair": {
        "D_E_counts_not_accepted_as_authority":
            True,

        "reason": (
            "Broad scan included Notebook13/14 provenance-analysis code "
            "and non-Morph MUTUAL-CONT semantics."
        ),

        "repair_scope":
            "NOTEBOOK11_CELLS_0_THROUGH_63_ONLY",
    },

    "component_status":
        component_status,

    "resolution": {
        "remaining_blockers":
            remaining_blockers,

        "status":
            status,

        "implementation_authorized":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_selection_policy":
            "FROZEN",

        "Morph_graph_application":
            "FROZEN",

        "morphology_score_formula":
            "FROZEN",

        "dataset_rank_formula":
            "FROZEN",

        "reference_association_provider":
            "EXACT_SOURCE_AUTHORITY",

        "remaining_generator_primitives":
            remaining_blockers,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_YET",

        "historical_action_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,

        "source_mutation":
            False,
    },

    "explicit_non_claims": [
        "D-E broad materialization counts are not treated as Morph source authority.",
        "Notebook13/14 provenance-analysis code is excluded from generator authority.",
        "Post-freeze Stage11 cells are excluded from scientific Morph generator authority.",
        "No historical pickle was loaded.",
        "No concrete MorphCandidateProvider was implemented.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-F artifact exists but differs "
        "from exact primitive-closure state."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 29. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-F PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "D-E PARENT EXACT CAPTURE:",
    flush=True,
)

print(
    "  file SHA:",
    parent_file_sha,
    flush=True,
)

print(
    "  content SHA:",
    parent_content_sha,
    flush=True,
)

print(
    flush=True,
)


print(
    "PRIMITIVE STATUS:",
    flush=True,
)

for key, value in component_status.items():

    print(
        " ",
        key,
        "=",
        value,
        flush=True,
    )


print(
    flush=True,
)


print(
    "REMAINING BLOCKERS:",
    remaining_blockers,
    flush=True,
)

print(
    "STATUS:",
    status,
    flush=True,
)

print(
    "IMPLEMENTATION AUTHORIZED:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  performance generated:",
    False,
    flush=True,
)

print(
    "  source mutation:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)

print(
    flush=True,
)


print(
    "NEXT:",
    flush=True,
)

print(
    " ",
    next_stage,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-G
# MORPH REMAINING PRIMITIVE TARGETED CLOSURE
# — AUDIT/ACTIONS OBJECT DATAFLOW AUTHORITY
#
# PARENT
# ------
# 14.3E-D-F PASS
#
# D-F remaining blockers:
#
#   p_T4_ordered_fresh_producer
#   morph_rank_materialization
#   candidate_rank_materialization
#   candidate_probability_materialization
#
#
# THIS STAGE
# ----------
# Stop scanning field-name occurrences globally.
#
# Trace the ACTUAL objects:
#
#   cell51: audit
#   cell61: actions
#
# and the artifacts feeding them:
#
#   stage11_additive_second_edge_tail.pkl
#   stage11_multiframe_folds1to4.pkl
#   stage11_deployable_final_v9_policy_grid_actions.pkl
#
# Recover:
#
#   1. exact audit definition/load lineage
#   2. exact writer of additive_second_edge_tail, if source survives
#   3. exact actions definition/load/filter lineage
#   4. exact writer of deployable_final_v9_policy_grid_actions, if source survives
#   5. direct producers of:
#
#          morph_rank
#          candidate_rank
#          candidate_prob
#
#      only inside those object lineages
#
#   6. whether rank_in_dataset is explicitly mapped/renamed to morph_rank
#   7. whether candidate_prob/rank are derived from an association matrix
#      or merely inherited from a historical table
#
#
# IMPORTANT BOUNDARY
# ------------------
# Historical pickle CONTENT is never deserialized.
#
# Source code which happens to write a historical artifact MAY be inspected.
# That source does not become fresh runtime authority automatically.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - pickle deserialization
#   - GT graph loading
#   - evaluator execution
#   - performance
#   - parameter search
#   - implementation
#   - structural replay
#   - Git operations
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_f_"
    "morph_generator_missing_primitive_closure_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


NOTEBOOK11_REL = Path(
    "notebooks/"
    "11_division_association_audit.ipynb"
)

NOTEBOOK11 = ROOT / NOTEBOOK11_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_g_"
    "morph_remaining_primitive_targeted_closure_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Exact parent / source locks
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "f05b81327ff1909009196bffa27cf9649d0394b9bedeb44f03393d4fc29fe482"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "8489046386a8460f1d066ce842d0c4abca3212825cf2adc5d28e2a27f2d81bc6"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_ASSOCIATION_PROVIDER_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


CELL46 = 46

CELL46_SHA = (
    "15747465b4e39aa4a2e72e259a9cdbbdfa83c05395c939904b6b6a4f436f36b1"
)


CELL51 = 51

CELL51_SHA = (
    "0be0b5b973555fbdd1b969374842bede150f1d80895596d498b433fb664ccb1a"
)


CELL55 = 55

CELL55_SHA = (
    "97c347aea36888e1d286c8d54e8adfc902dc7f4e86e1e2989009caa86a8025fc"
)


CELL61 = 61

CELL61_SHA = (
    "e449e6190f23607d7a20a8d9b83029485a2a7237a00c74917f2092eb6fe0d109"
)


FREEZE_CELL = 63

FREEZE_CELL_SHA = (
    "2e6bdff585ad1a3ee61b3a82ec03ccbd05db0b66f01024a35a721d76a29d8834"
)


R1B_CANONICALIZATION_SHA = (
    "b71ded6bda8547573f64222957061fd7d691090b0a083535babd733fcbd83d96"
)


# Historical objects whose SOURCE lineage may be inspected.
AUDIT_BASENAME = (
    "stage11_additive_second_edge_tail.pkl"
)

FEATURE_BASENAME = (
    "stage11_multiframe_folds1to4.pkl"
)

UPSTREAM_ACTION_BASENAME = (
    "stage11_deployable_final_v9_policy_grid_actions.pkl"
)

MORPH_LOCK_BASENAME = (
    "stage11_fold0_morphdiv_v2_actions_locked.pkl"
)


TARGET_FIELDS = (
    "morph_rank",
    "candidate_rank",
    "candidate_prob",
)


# =============================================================================
# 2. Generic helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def safe_rel(
    path: Path,
) -> str:

    try:

        return str(
            path.resolve().relative_to(
                ROOT
            )
        )

    except Exception:

        return str(
            path
        )


def node_source(
    source: str,
    node: ast.AST,
) -> str:

    value = ast.get_source_segment(
        source,
        node,
    )


    assert (
        value is not None
    )


    return value


def dotted_name(
    node: ast.AST,
) -> str | None:

    if isinstance(
        node,
        ast.Name,
    ):

        return node.id


    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )


        if left:

            return (
                f"{left}.{node.attr}"
            )


        return node.attr


    return None


def calls_in_node(
    node: ast.AST,
) -> list[str]:

    return sorted({
        dotted_name(
            child.func
        )
        or
        "<dynamic-call>"

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Call,
        )
    })


def loaded_names(
    node: ast.AST,
) -> set[str]:

    return {
        child.id

        for child in ast.walk(
            node
        )

        if (
            isinstance(
                child,
                ast.Name,
            )
            and
            isinstance(
                child.ctx,
                ast.Load,
            )
        )
    }


def target_names(
    node: ast.AST,
) -> set[str]:

    result = set()


    for child in ast.walk(
        node
    ):

        if isinstance(
            child,
            ast.Name,
        ):

            result.add(
                child.id
            )


    return result


def notebook_cells(
    path: Path,
) -> list[
    dict[str, Any]
]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    result = []


    for index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        result.append({
            "index":
                index,

            "cell_type":
                cell.get(
                    "cell_type"
                ),

            "source":
                source,

            "sha256":
                sha256_text(
                    source
                ),
        })


    return result


def has_read_pickle(
    node: ast.AST,
) -> bool:

    return any(
        (
            isinstance(
                child,
                ast.Call,
            )
            and
            (
                dotted_name(
                    child.func
                )
                or ""
            ).endswith(
                "read_pickle"
            )
        )

        for child in ast.walk(
            node
        )
    )


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    expected_blockers = {
        "p_T4_ordered_fresh_producer",
        "morph_rank_materialization",
        "candidate_rank_materialization",
        "candidate_probability_materialization",
    }


    assert (
        set(
            obj[
                "resolution"
            ][
                "remaining_blockers"
            ]
        )
        ==
        expected_blockers
    )


    assert (
        obj[
            "resolution"
        ][
            "implementation_authorized"
        ]
        is False
    )


    return obj


# =============================================================================
# 4. Simple assignment helpers
# =============================================================================

def assignment_targets(
    stmt: ast.stmt,
) -> set[str]:

    result = set()


    if isinstance(
        stmt,
        ast.Assign,
    ):

        targets = stmt.targets


    elif isinstance(
        stmt,
        ast.AnnAssign,
    ):

        targets = [
            stmt.target
        ]


    elif isinstance(
        stmt,
        ast.AugAssign,
    ):

        targets = [
            stmt.target
        ]


    else:

        return result


    for target in targets:

        result |= target_names(
            target
        )


    return result


def assignment_value(
    stmt: ast.stmt,
) -> ast.AST | None:

    if isinstance(
        stmt,
        (
            ast.Assign,
            ast.AnnAssign,
            ast.AugAssign,
        ),
    ):

        return stmt.value


    return None


def simple_name_from_expr(
    expr: str | None,
) -> str | None:

    if expr is None:

        return None


    if re.fullmatch(
        r"[A-Za-z_][A-Za-z0-9_]*",
        expr,
    ):

        return expr


    return None


# =============================================================================
# 5. Variable definition inventory
# =============================================================================

def variable_definition_records(
    cells: list[
        dict[str, Any]
    ],
    variable: str,
    *,
    max_cell: int,
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            max_cell
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for stmt in tree.body:

            if variable not in assignment_targets(
                stmt
            ):

                continue


            value = assignment_value(
                stmt
            )


            if value is None:

                continue


            segment = node_source(
                source,
                stmt,
            )


            value_source = ast.unparse(
                value
            )


            records.append({
                "cell_index":
                    cell[
                        "index"
                    ],

                "cell_sha256":
                    cell[
                        "sha256"
                    ],

                "line_start":
                    stmt.lineno,

                "line_end":
                    getattr(
                        stmt,
                        "end_lineno",
                        stmt.lineno,
                    ),

                "source_sha256":
                    sha256_text(
                        segment
                    ),

                "value_expression":
                    value_source,

                "read_pickle":
                    has_read_pickle(
                        value
                    ),

                "calls":
                    calls_in_node(
                        value
                    ),

                "loaded_names":
                    sorted(
                        loaded_names(
                            value
                        )
                    ),

                "source":
                    segment,
            })


    return sorted(
        records,
        key=lambda record: (
            record[
                "cell_index"
            ],
            record[
                "line_start"
            ],
        ),
    )


# =============================================================================
# 6. Symbolic pickle path lineage
# =============================================================================

def expression_has_literal(
    node: ast.AST,
    basename: str,
) -> bool:

    return any(
        (
            isinstance(
                child,
                ast.Constant,
            )
            and
            isinstance(
                child.value,
                str,
            )
            and
            basename
            in child.value
        )

        for child in ast.walk(
            node
        )
    )


def expression_uses_alias(
    node: ast.AST,
    aliases: set[str],
) -> bool:

    return bool(
        loaded_names(
            node
        )
        &
        aliases
    )


def read_pickle_path_arg(
    call: ast.Call,
) -> ast.AST | None:

    name = dotted_name(
        call.func
    )


    if (
        name is None
        or
        not name.endswith(
            "read_pickle"
        )
    ):

        return None


    if call.args:

        return call.args[
            0
        ]


    for kw in call.keywords:

        if kw.arg in {
            "path",
            "filepath_or_buffer",
        }:

            return kw.value


    return None


def to_pickle_signature(
    call: ast.Call,
) -> tuple[
    ast.AST | None,
    ast.AST | None,
]:

    """
    Returns:
        (object_expr, path_expr)
    """

    name = dotted_name(
        call.func
    )


    if name is None:

        return (
            None,
            None,
        )


    # pd.to_pickle(obj, path)
    if name in {
        "pd.to_pickle",
        "pandas.to_pickle",
    }:

        object_expr = (
            call.args[
                0
            ]
            if len(
                call.args
            )
            >= 1
            else
            None
        )


        path_expr = (
            call.args[
                1
            ]
            if len(
                call.args
            )
            >= 2
            else
            None
        )


        return (
            object_expr,
            path_expr,
        )


    # obj.to_pickle(path)
    if name.endswith(
        ".to_pickle"
    ):

        if isinstance(
            call.func,
            ast.Attribute,
        ):

            object_expr = (
                call.func.value
            )

        else:

            object_expr = None


        path_expr = (
            call.args[
                0
            ]
            if call.args
            else
            None
        )


        return (
            object_expr,
            path_expr,
        )


    return (
        None,
        None,
    )


def notebook_pickle_io_lineage(
    cells: list[
        dict[str, Any]
    ],
    basename: str,
    *,
    max_cell: int,
) -> dict[str, Any]:

    aliases: set[str] = set()

    records = []


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            max_cell
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for stmt in tree.body:

            # -------------------------------------------------------------
            # Update path aliases first.
            # -------------------------------------------------------------
            if isinstance(
                stmt,
                (
                    ast.Assign,
                    ast.AnnAssign,
                ),
            ):

                value = stmt.value


                targets = assignment_targets(
                    stmt
                )


                tainted = (
                    expression_has_literal(
                        value,
                        basename,
                    )
                    or
                    expression_uses_alias(
                        value,
                        aliases,
                    )
                )


                for target in targets:

                    if tainted:

                        aliases.add(
                            target
                        )


            # -------------------------------------------------------------
            # Detect reads / writes.
            # -------------------------------------------------------------
            stmt_targets = assignment_targets(
                stmt
            )


            for call in ast.walk(
                stmt
            ):

                if not isinstance(
                    call,
                    ast.Call,
                ):

                    continue


                # Read.
                read_path = read_pickle_path_arg(
                    call
                )


                if read_path is not None:

                    matches = (
                        expression_has_literal(
                            read_path,
                            basename,
                        )
                        or
                        expression_uses_alias(
                            read_path,
                            aliases,
                        )
                    )


                    if matches:

                        records.append({
                            "operation":
                                "READ",

                            "cell_index":
                                cell[
                                    "index"
                                ],

                            "cell_sha256":
                                cell[
                                    "sha256"
                                ],

                            "line":
                                call.lineno,

                            "path_expression":
                                ast.unparse(
                                    read_path
                                ),

                            "assigned_names":
                                sorted(
                                    stmt_targets
                                ),

                            "call":
                                ast.unparse(
                                    call
                                ),
                        })


                # Write.
                (
                    object_expr,
                    write_path,
                ) = to_pickle_signature(
                    call
                )


                if write_path is not None:

                    matches = (
                        expression_has_literal(
                            write_path,
                            basename,
                        )
                        or
                        expression_uses_alias(
                            write_path,
                            aliases,
                        )
                    )


                    if matches:

                        records.append({
                            "operation":
                                "WRITE",

                            "cell_index":
                                cell[
                                    "index"
                                ],

                            "cell_sha256":
                                cell[
                                    "sha256"
                                ],

                            "line":
                                call.lineno,

                            "path_expression":
                                ast.unparse(
                                    write_path
                                ),

                            "object_expression":
                                (
                                    ast.unparse(
                                        object_expr
                                    )
                                    if object_expr
                                    is not None
                                    else
                                    None
                                ),

                            "call":
                                ast.unparse(
                                    call
                                ),
                        })


    return {
        "basename":
            basename,

        "aliases":
            sorted(
                aliases
            ),

        "records":
            records,

        "reads":
            [
                record

                for record in records

                if record[
                    "operation"
                ]
                == "READ"
            ],

        "writes":
            [
                record

                for record in records

                if record[
                    "operation"
                ]
                == "WRITE"
            ],
    }


# =============================================================================
# 7. Field producer extraction
# =============================================================================

def subscript_field(
    node: ast.AST,
) -> tuple[
    str | None,
    str | None,
]:

    if not isinstance(
        node,
        ast.Subscript,
    ):

        return (
            None,
            None,
        )


    try:

        key = ast.literal_eval(
            node.slice
        )

    except Exception:

        return (
            None,
            None,
        )


    if not isinstance(
        key,
        str,
    ):

        return (
            None,
            None,
        )


    return (
        ast.unparse(
            node.value
        ),
        key,
    )


def assignment_field_records(
    cells: list[
        dict[str, Any]
    ],
    field: str,
    *,
    max_cell: int,
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            max_cell
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for node in ast.walk(
            tree
        ):

            if not isinstance(
                node,
                (
                    ast.Assign,
                    ast.AnnAssign,
                ),
            ):

                continue


            targets = (
                node.targets
                if isinstance(
                    node,
                    ast.Assign,
                )
                else
                [
                    node.target
                ]
            )


            for target in targets:

                object_expr, key = subscript_field(
                    target
                )


                if key != field:

                    continue


                value = node.value


                segment = node_source(
                    source,
                    node,
                )


                rhs = ast.unparse(
                    value
                )


                records.append({
                    "kind":
                        "SUBSCRIPT_ASSIGNMENT",

                    "cell_index":
                        cell[
                            "index"
                        ],

                    "cell_sha256":
                        cell[
                            "sha256"
                        ],

                    "line_start":
                        node.lineno,

                    "line_end":
                        getattr(
                            node,
                            "end_lineno",
                            node.lineno,
                        ),

                    "source_sha256":
                        sha256_text(
                            segment
                        ),

                    "object_expression":
                        object_expr,

                    "field":
                        field,

                    "value_expression":
                        rhs,

                    "uses_existing_same_field":
                        (
                            f'["{field}"]'
                            in rhs
                            or
                            f"['{field}']"
                            in rhs
                            or
                            f".{field}"
                            in rhs
                        ),

                    "read_pickle":
                        has_read_pickle(
                            value
                        ),

                    "calls":
                        calls_in_node(
                            value
                        ),

                    "source":
                        segment,
                })


    return records


def rename_field_records(
    cells: list[
        dict[str, Any]
    ],
    destination_field: str,
    *,
    max_cell: int,
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            max_cell
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for call in ast.walk(
            tree
        ):

            if not isinstance(
                call,
                ast.Call,
            ):

                continue


            name = dotted_name(
                call.func
            )


            if (
                name is None
                or
                not name.endswith(
                    ".rename"
                )
            ):

                continue


            for kw in call.keywords:

                if kw.arg != "columns":

                    continue


                if not isinstance(
                    kw.value,
                    ast.Dict,
                ):

                    continue


                mapping = {}


                for key_node, value_node in zip(
                    kw.value.keys,
                    kw.value.values,
                ):

                    if key_node is None:

                        continue


                    try:

                        key = ast.literal_eval(
                            key_node
                        )


                        value = ast.literal_eval(
                            value_node
                        )

                    except Exception:

                        continue


                    if (
                        isinstance(
                            key,
                            str,
                        )
                        and
                        isinstance(
                            value,
                            str,
                        )
                    ):

                        mapping[
                            key
                        ] = value


                sources = [
                    key

                    for key, value
                    in mapping.items()

                    if value
                    == destination_field
                ]


                if not sources:

                    continue


                segment = node_source(
                    source,
                    call,
                )


                records.append({
                    "kind":
                        "RENAME",

                    "cell_index":
                        cell[
                            "index"
                        ],

                    "cell_sha256":
                        cell[
                            "sha256"
                        ],

                    "line_start":
                        call.lineno,

                    "line_end":
                        getattr(
                            call,
                            "end_lineno",
                            call.lineno,
                        ),

                    "source_sha256":
                        sha256_text(
                            segment
                        ),

                    "destination_field":
                        destination_field,

                    "source_fields":
                        sources,

                    "object_expression":
                        (
                            ast.unparse(
                                call.func.value
                            )
                            if isinstance(
                                call.func,
                                ast.Attribute,
                            )
                            else
                            None
                        ),

                    "source":
                        segment,
                })


    return records


def dict_field_records(
    cells: list[
        dict[str, Any]
    ],
    field: str,
    *,
    max_cell: int,
) -> list[
    dict[str, Any]
]:

    records = []


    for cell in cells:

        if (
            cell[
                "index"
            ]
            >
            max_cell
            or
            cell[
                "cell_type"
            ]
            != "code"
        ):

            continue


        source = cell[
            "source"
        ]


        try:

            tree = ast.parse(
                source
            )

        except Exception:

            continue


        for dictionary in ast.walk(
            tree
        ):

            if not isinstance(
                dictionary,
                ast.Dict,
            ):

                continue


            for key_node, value_node in zip(
                dictionary.keys,
                dictionary.values,
            ):

                if key_node is None:

                    continue


                try:

                    key = ast.literal_eval(
                        key_node
                    )

                except Exception:

                    continue


                if key != field:

                    continue


                expression = ast.unparse(
                    value_node
                )


                segment = node_source(
                    source,
                    dictionary,
                )


                records.append({
                    "kind":
                        "DICT_VALUE",

                    "cell_index":
                        cell[
                            "index"
                        ],

                    "cell_sha256":
                        cell[
                            "sha256"
                        ],

                    "line_start":
                        dictionary.lineno,

                    "line_end":
                        getattr(
                            dictionary,
                            "end_lineno",
                            dictionary.lineno,
                        ),

                    "source_sha256":
                        sha256_text(
                            segment
                        ),

                    "field":
                        field,

                    "value_expression":
                        expression,

                    "uses_existing_same_field":
                        (
                            f".{field}"
                            in expression
                            or
                            f'["{field}"]'
                            in expression
                            or
                            f"['{field}']"
                            in expression
                        ),

                    "source":
                        segment,
                })


    return records


def all_field_records(
    cells: list[
        dict[str, Any]
    ],
    field: str,
    *,
    max_cell: int,
) -> list[
    dict[str, Any]
]:

    records = []


    records.extend(
        assignment_field_records(
            cells,
            field,
            max_cell=max_cell,
        )
    )


    records.extend(
        rename_field_records(
            cells,
            field,
            max_cell=max_cell,
        )
    )


    records.extend(
        dict_field_records(
            cells,
            field,
            max_cell=max_cell,
        )
    )


    return sorted(
        records,
        key=lambda record: (
            record[
                "cell_index"
            ],
            record[
                "line_start"
            ],
            record[
                "kind"
            ],
        ),
    )


# =============================================================================
# 8. Relevant object statement inventory
# =============================================================================

def statements_mentioning_object(
    cell: dict[str, Any],
    object_name: str,
) -> list[
    dict[str, Any]
]:

    source = cell[
        "source"
    ]


    tree = ast.parse(
        source
    )


    records = []


    for stmt in ast.walk(
        tree
    ):

        if not isinstance(
            stmt,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
                ast.For,
                ast.If,
                ast.Expr,
            ),
        ):

            continue


        segment = ast.get_source_segment(
            source,
            stmt,
        )


        if not segment:

            continue


        if (
            object_name
            not in loaded_names(
                stmt
            )
            and
            object_name
            not in assignment_targets(
                stmt
            )
            and
            object_name
            not in segment
        ):

            continue


        records.append({
            "line_start":
                stmt.lineno,

            "line_end":
                getattr(
                    stmt,
                    "end_lineno",
                    stmt.lineno,
                ),

            "kind":
                type(
                    stmt
                ).__name__,

            "source_sha256":
                sha256_text(
                    segment
                ),

            "read_pickle":
                has_read_pickle(
                    stmt
                ),

            "calls":
                calls_in_node(
                    stmt
                ),

            "source":
                segment,
        })


    # smallest scopes first
    records = sorted(
        records,
        key=lambda record: (
            (
                record[
                    "line_end"
                ]
                -
                record[
                    "line_start"
                ]
            ),
            record[
                "line_start"
            ],
        ),
    )


    unique = {}


    for record in records:

        unique.setdefault(
            record[
                "source_sha256"
            ],
            record,
        )


    return list(
        unique.values()
    )


# =============================================================================
# 9. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    NOTEBOOK11,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-G "
    "MORPH REMAINING PRIMITIVE TARGETED CLOSURE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "AUTHORITY:",
    NOTEBOOK11_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Exact parent lock
# =============================================================================

parent = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_D_F_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Current source / authority lock
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


cells = notebook_cells(
    NOTEBOOK11
)


for index, expected_sha in (
    (
        CELL46,
        CELL46_SHA,
    ),
    (
        CELL51,
        CELL51_SHA,
    ),
    (
        CELL55,
        CELL55_SHA,
    ),
    (
        CELL61,
        CELL61_SHA,
    ),
    (
        FREEZE_CELL,
        FREEZE_CELL_SHA,
    ),
):

    assert (
        cells[
            index
        ][
            "sha256"
        ]
        == expected_sha
    )


print(
    "[PASS] NOTEBOOK11_OBJECT_LINEAGE_AUTHORITY_LOCK",
    flush=True,
)

print(
    "  cell46:",
    CELL46_SHA,
    flush=True,
)

print(
    "  cell51:",
    CELL51_SHA,
    flush=True,
)

print(
    "  cell55:",
    CELL55_SHA,
    flush=True,
)

print(
    "  cell61:",
    CELL61_SHA,
    flush=True,
)

print(
    "  cell63:",
    FREEZE_CELL_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Historical artifact I/O lineages
# =============================================================================

artifact_lineages = {}


for basename in (
    AUDIT_BASENAME,
    FEATURE_BASENAME,
    UPSTREAM_ACTION_BASENAME,
    MORPH_LOCK_BASENAME,
):

    artifact_lineages[
        basename
    ] = notebook_pickle_io_lineage(
        cells,
        basename,
        max_cell=FREEZE_CELL,
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "EXACT NOTEBOOK11 PICKLE I/O LINEAGES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for basename, lineage in artifact_lineages.items():

    print(
        basename,
        flush=True,
    )

    print(
        "  aliases:",
        lineage[
            "aliases"
        ],
        flush=True,
    )

    print(
        "  reads:",
        len(
            lineage[
                "reads"
            ]
        ),
        flush=True,
    )

    print(
        "  writes:",
        len(
            lineage[
                "writes"
            ]
        ),
        flush=True,
    )


    for record in lineage[
        "records"
    ]:

        print(
            " ",
            record,
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 13. audit exact variable lineage
# =============================================================================

audit_defs = variable_definition_records(
    cells,
    "audit",
    max_cell=CELL51,
)


print(
    "=" * 120,
    flush=True,
)

print(
    "CELL51 AUDIT VARIABLE LINEAGE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in audit_defs:

    print(
        f"cell={record['cell_index']}",
        f"L{record['line_start']}-{record['line_end']}",
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "  RHS:",
        record[
            "value_expression"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        record[
            "read_pickle"
        ],
        flush=True,
    )

    print(
        "  calls:",
        record[
            "calls"
        ],
        flush=True,
    )

    print(
        "  source:",
        record[
            "source"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


latest_audit_def = (
    audit_defs[
        -1
    ]
    if audit_defs
    else
    None
)


audit_is_historical_table = (
    latest_audit_def
    is not None
    and
    latest_audit_def[
        "read_pickle"
    ]
)


print(
    "LATEST AUDIT DEFINITION:",
    (
        latest_audit_def[
            "source"
        ]
        if latest_audit_def
        else
        None
    ),
    flush=True,
)

print(
    "AUDIT IS HISTORICAL PICKLE INPUT:",
    audit_is_historical_table,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. actions exact variable lineage
# =============================================================================

actions_defs = variable_definition_records(
    cells,
    "actions",
    max_cell=CELL61,
)


print(
    "=" * 120,
    flush=True,
)

print(
    "CELL61 ACTIONS VARIABLE LINEAGE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in actions_defs:

    print(
        f"cell={record['cell_index']}",
        f"L{record['line_start']}-{record['line_end']}",
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "source_sha256"
        ],
        flush=True,
    )

    print(
        "  RHS:",
        record[
            "value_expression"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        record[
            "read_pickle"
        ],
        flush=True,
    )

    print(
        "  calls:",
        record[
            "calls"
        ],
        flush=True,
    )

    print(
        "  source:",
        record[
            "source"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


latest_actions_def = (
    actions_defs[
        -1
    ]
    if actions_defs
    else
    None
)


actions_is_historical_table = (
    latest_actions_def
    is not None
    and
    (
        latest_actions_def[
            "read_pickle"
        ]
        or
        any(
            token
            in latest_actions_def[
                "value_expression"
            ]

            for token in (
                "policy_grid",
                "read_pickle",
            )
        )
    )
)


print(
    "LATEST ACTIONS DEFINITION:",
    (
        latest_actions_def[
            "source"
        ]
        if latest_actions_def
        else
        None
    ),
    flush=True,
)

print(
    "ACTIONS HISTORICAL-TABLE DERIVED:",
    actions_is_historical_table,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Print writer objects for the two critical historical tables
# =============================================================================

audit_writes = (
    artifact_lineages[
        AUDIT_BASENAME
    ][
        "writes"
    ]
)


upstream_action_writes = (
    artifact_lineages[
        UPSTREAM_ACTION_BASENAME
    ][
        "writes"
    ]
)


print(
    "=" * 120,
    flush=True,
)

print(
    "CRITICAL HISTORICAL TABLE WRITERS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "AUDIT WRITES:",
    audit_writes,
    flush=True,
)

print(
    "UPSTREAM ACTION WRITES:",
    upstream_action_writes,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Inspect writer-object source contexts
# =============================================================================

writer_contexts = []


for basename, writes in (
    (
        AUDIT_BASENAME,
        audit_writes,
    ),
    (
        UPSTREAM_ACTION_BASENAME,
        upstream_action_writes,
    ),
):

    for write in writes:

        object_name = simple_name_from_expr(
            write.get(
                "object_expression"
            )
        )


        if object_name is None:

            continue


        cell = cells[
            write[
                "cell_index"
            ]
        ]


        statements = statements_mentioning_object(
            cell,
            object_name,
        )


        writer_contexts.append({
            "basename":
                basename,

            "writer_cell":
                write[
                    "cell_index"
                ],

            "writer_line":
                write[
                    "line"
                ],

            "object_name":
                object_name,

            "statements":
                statements,
        })


print(
    "=" * 120,
    flush=True,
)

print(
    "WRITER OBJECT CONTEXTS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for context in writer_contexts:

    print(
        context[
            "basename"
        ],
        "object=",
        context[
            "object_name"
        ],
        "cell=",
        context[
            "writer_cell"
        ],
        flush=True,
    )


    for record in context[
        "statements"
    ][
        :40
    ]:

        print(
            f"  L{record['line_start']}-{record['line_end']}",
            record[
                "kind"
            ],
            flush=True,
        )

        print(
            "    SHA:",
            record[
                "source_sha256"
            ],
            flush=True,
        )

        print(
            "    read_pickle:",
            record[
                "read_pickle"
            ],
            flush=True,
        )

        print(
            "    source:",
            flush=True,
        )

        print(
            record[
                "source"
            ][
                :6000
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 17. Exact field producer inventory, pre-freeze only
# =============================================================================

field_records = {
    field:
        all_field_records(
            cells,
            field,
            max_cell=FREEZE_CELL,
        )

    for field in TARGET_FIELDS
}


print(
    "=" * 120,
    flush=True,
)

print(
    "PRE-FREEZE FIELD PRODUCER INVENTORY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for field, records in field_records.items():

    print(
        field,
        "records:",
        len(
            records
        ),
        flush=True,
    )


    for record in records[
        :40
    ]:

        print(
            f"  cell={record['cell_index']}",
            f"L{record['line_start']}-{record['line_end']}",
            record[
                "kind"
            ],
            flush=True,
        )


        if "object_expression" in record:

            print(
                "    object:",
                record[
                    "object_expression"
                ],
                flush=True,
            )


        if "source_fields" in record:

            print(
                "    rename source:",
                record[
                    "source_fields"
                ],
                flush=True,
            )


        if "value_expression" in record:

            print(
                "    value:",
                record[
                    "value_expression"
                ],
                flush=True,
            )


        if "uses_existing_same_field" in record:

            print(
                "    existing-value consumer:",
                record[
                    "uses_existing_same_field"
                ],
                flush=True,
            )


        print(
            "    SHA:",
            record[
                "source_sha256"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 18. Restrict field authority to REAL object lineage cells
# =============================================================================

audit_lineage_cells = {
    CELL51,
}


for record in audit_writes:

    audit_lineage_cells.add(
        record[
            "cell_index"
        ]
    )


action_lineage_cells = {
    CELL61,
}


for record in upstream_action_writes:

    action_lineage_cells.add(
        record[
            "cell_index"
        ]
    )


print(
    "AUDIT LINEAGE CELLS:",
    sorted(
        audit_lineage_cells
    ),
    flush=True,
)

print(
    "ACTION LINEAGE CELLS:",
    sorted(
        action_lineage_cells
    ),
    flush=True,
)

print(
    flush=True,
)


candidate_rank_lineage_records = [
    record

    for record in field_records[
        "candidate_rank"
    ]

    if (
        record[
            "cell_index"
        ]
        in audit_lineage_cells
        or
        record[
            "cell_index"
        ]
        in action_lineage_cells
    )
]


candidate_prob_lineage_records = [
    record

    for record in field_records[
        "candidate_prob"
    ]

    if (
        record[
            "cell_index"
        ]
        in audit_lineage_cells
        or
        record[
            "cell_index"
        ]
        in action_lineage_cells
    )
]


morph_rank_lineage_records = [
    record

    for record in field_records[
        "morph_rank"
    ]

    if (
        record[
            "cell_index"
        ]
        in action_lineage_cells
    )
]


# =============================================================================
# 19. Strong producer vs inherited-value classification
# =============================================================================

def is_inherited_record(
    record: dict[str, Any],
) -> bool:

    if record[
        "kind"
    ] == "RENAME":

        return False


    return bool(
        record.get(
            "uses_existing_same_field",
            False,
        )
    )


def strong_formula_records(
    records: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    return [
        record

        for record in records

        if not is_inherited_record(
            record
        )
    ]


strong_morph_rank_records = (
    strong_formula_records(
        morph_rank_lineage_records
    )
)


strong_candidate_rank_records = (
    strong_formula_records(
        candidate_rank_lineage_records
    )
)


strong_candidate_prob_records = (
    strong_formula_records(
        candidate_prob_lineage_records
    )
)


print(
    "=" * 120,
    flush=True,
)

print(
    "OBJECT-LINEAGE FIELD AUTHORITY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "morph_rank lineage:",
    len(
        morph_rank_lineage_records
    ),
    "strong:",
    len(
        strong_morph_rank_records
    ),
    flush=True,
)

print(
    "candidate_rank lineage:",
    len(
        candidate_rank_lineage_records
    ),
    "strong:",
    len(
        strong_candidate_rank_records
    ),
    flush=True,
)

print(
    "candidate_prob lineage:",
    len(
        candidate_prob_lineage_records
    ),
    "strong:",
    len(
        strong_candidate_prob_records
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 20. Explicit rank_in_dataset -> morph_rank mapping test
# =============================================================================

explicit_rank_to_morph = [
    record

    for record in strong_morph_rank_records

    if (
        "rank_in_dataset"
        in (
            record.get(
                "value_expression",
                ""
            )
            +
            " ".join(
                record.get(
                    "source_fields",
                    []
                )
            )
        )
    )
]


print(
    "EXPLICIT rank_in_dataset -> morph_rank mappings:",
    len(
        explicit_rank_to_morph
    ),
    flush=True,
)


for record in explicit_rank_to_morph:

    print(
        " ",
        record,
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 21. Association-matrix materialization evidence
# =============================================================================

def association_like_materialization(
    records: list[
        dict[str, Any]
    ],
) -> list[
    dict[str, Any]
]:

    result = []


    for record in records:

        text = (
            record.get(
                "value_expression",
                ""
            )
            +
            " "
            +
            record.get(
                "source",
                ""
            )
        ).lower()


        if any(
            token
            in text

            for token in (
                "probs[",
                "probabilities",
                "argsort",
                "argmax",
                "rank(",
                "enumerate",
            )
        ):

            result.append(
                record
            )


    return result


association_candidate_rank_records = (
    association_like_materialization(
        strong_candidate_rank_records
    )
)


association_candidate_prob_records = (
    association_like_materialization(
        strong_candidate_prob_records
    )
)


print(
    "ASSOCIATION-DERIVED candidate_rank records:",
    len(
        association_candidate_rank_records
    ),
    flush=True,
)

print(
    "ASSOCIATION-DERIVED candidate_prob records:",
    len(
        association_candidate_prob_records
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 22. Exact association provider lock remains valid
# =============================================================================

portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


assert (
    EXPECTED_ASSOCIATION_PROVIDER_SHA
    in portable_text
)


print(
    "[PASS] ASSOCIATION_PROVIDER_IDENTITY_RETAINED",
    flush=True,
)

print(
    "  SHA:",
    EXPECTED_ASSOCIATION_PROVIDER_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 23. p_T4 fresh producer remains governed by D-F
# =============================================================================

p_t4_status = (
    parent[
        "component_status"
    ][
        "p_T4_ordered_fresh_producer"
    ]
)


assert (
    p_t4_status
    ==
    "UNRESOLVED_HISTORICAL_FEATURE_INPUT_ONLY"
)


# =============================================================================
# 24. Resolve four blockers conservatively
# =============================================================================

#
# p_T4:
# D-F already proved 0 assignment producers before freeze.
#

p_t4_closed = False


#
# morph_rank:
# Only close if explicit non-inherited source mapping exists.
#

morph_rank_closed = (
    len(
        explicit_rank_to_morph
    )
    > 0
)


#
# candidate_rank:
# Only close if actual audit/actions lineage contains a non-inherited
# association/ranking computation.
#

candidate_rank_closed = (
    len(
        association_candidate_rank_records
    )
    > 0
)


#
# candidate_prob:
# Only close if actual audit/actions lineage contains non-inherited probability
# matrix indexing/materialization.
#

candidate_prob_closed = (
    len(
        association_candidate_prob_records
    )
    > 0
)


component_status = {
    "p_T4_ordered_fresh_producer":
        (
            "SOURCE_BACKED"
            if p_t4_closed
            else
            "UNRESOLVED_HISTORICAL_FEATURE_INPUT_ONLY"
        ),

    "morph_rank_materialization":
        (
            "SOURCE_BACKED"
            if morph_rank_closed
            else
            "UNRESOLVED_HISTORICAL_ACTION_FIELD"
        ),

    "candidate_rank_materialization":
        (
            "SOURCE_BACKED"
            if candidate_rank_closed
            else
            "UNRESOLVED_HISTORICAL_AUDIT_ACTION_FIELD"
        ),

    "candidate_probability_materialization":
        (
            "SOURCE_BACKED"
            if candidate_prob_closed
            else
            "UNRESOLVED_HISTORICAL_AUDIT_ACTION_FIELD"
        ),

    "reference_association_provider":
        "EXACT_SOURCE_AUTHORITY",
}


remaining_blockers = [
    key

    for key, value
    in component_status.items()

    if value.startswith(
        "UNRESOLVED"
    )
]


# =============================================================================
# 25. Decide whether archaeology is exhausted
# =============================================================================

audit_writer_survives = (
    len(
        audit_writes
    )
    > 0
)


upstream_action_writer_survives = (
    len(
        upstream_action_writes
    )
    > 0
)


all_action_fields_historical = (
    not morph_rank_closed
    and
    not candidate_rank_closed
    and
    not candidate_prob_closed
)


if (
    not p_t4_closed
    and
    all_action_fields_historical
):

    status = (
        "CURRENT_CHECKOUT_HAS_NO_COMPLETE_FRESH_MORPH_GENERATOR_AUTHORITY"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_H_"
        "MORPH_SPECIFICATION_DERIVED_RECONSTRUCTION_BOUNDARY_FREEZE"
    )


elif remaining_blockers:

    status = (
        "PARTIAL_OBJECT_LINEAGE_CLOSURE_"
        "REMAINING_PRIMITIVE_TARGETED_RECOVERY_REQUIRED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_H_"
        "MORPH_FINAL_REMAINING_PRIMITIVE_CLOSURE"
    )


else:

    status = (
        "ALL_FRESH_GENERATOR_PRIMITIVES_SOURCE_CLOSED_"
        "EXACT_PROVIDER_CONTRACT_FREEZE_REQUIRED"
    )


    next_stage = (
        "NOTEBOOK14_14_3E_D_H_"
        "MORPH_CONCRETE_PROVIDER_EXACT_CONTRACT_FREEZE"
    )


implementation_authorized = False


print(
    "=" * 120,
    flush=True,
)

print(
    "OBJECT DATAFLOW RESOLUTION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "audit historical-table input:",
    audit_is_historical_table,
    flush=True,
)

print(
    "audit writer survives in checkout:",
    audit_writer_survives,
    flush=True,
)

print(
    "actions historical-table derived:",
    actions_is_historical_table,
    flush=True,
)

print(
    "upstream action writer survives in checkout:",
    upstream_action_writer_survives,
    flush=True,
)

print(
    flush=True,
)

print(
    "component status:",
    json.dumps(
        component_status,
        indent=2,
    ),
    flush=True,
)

print(
    flush=True,
)

print(
    "remaining blockers:",
    remaining_blockers,
    flush=True,
)

print(
    "status:",
    status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 26. Safety / no mutation
# =============================================================================

historical_pickle_loaded = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False

source_mutation = False

test_mutation = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


# =============================================================================
# 27. Formal gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_f_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "Notebook11_authority_cells_locked":
        True,

    "critical_pickle_IO_lineages_scanned":
        True,

    "audit_variable_lineage_scanned":
        True,

    "actions_variable_lineage_scanned":
        True,

    "historical_table_writer_objects_scanned":
        True,

    "field_producers_restricted_to_pre_freeze":
        True,

    "field_authority_restricted_to_audit_actions_lineages":
        True,

    "inherited_row_fields_not_treated_as_materialization":
        True,

    "rank_in_dataset_to_morph_rank_requires_explicit_source_mapping":
        True,

    "candidate_rank_requires_association_lineage_computation":
        True,

    "candidate_prob_requires_association_lineage_computation":
        True,

    "reference_association_provider_identity_preserved":
        True,

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-G SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-G failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 28. Artifact-safe compactors
# =============================================================================

def compact(
    record: dict[str, Any],
) -> dict[str, Any]:

    result = {
        key:
            value

        for key, value
        in record.items()

        if key
        not in {
            "source",
        }
    }


    if "source" in record:

        result[
            "source_preview"
        ] = (
            record[
                "source"
            ][
                :7000
            ]
        )


    return result


def compact_field_map(
    mapping: dict[
        str,
        list[
            dict[str, Any]
        ],
    ],
) -> dict[
    str,
    list[
        dict[str, Any]
    ],
]:

    return {
        key:
            [
                compact(
                    record
                )

                for record in value
            ]

        for key, value
        in mapping.items()
    }


# =============================================================================
# 29. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_G",

    "title":
        "MORPH_REMAINING_PRIMITIVE_TARGETED_CLOSURE",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_OBJECT_DATAFLOW_AUTHORITY_CLOSURE_ONLY",

    "parent_14_3e_d_f": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha_before,

            "association_provider_identity_sha256":
                EXPECTED_ASSOCIATION_PROVIDER_SHA,
        },

        "Notebook11": {
            "path":
                str(
                    NOTEBOOK11_REL
                ),

            "cell46":
                CELL46_SHA,

            "cell51":
                CELL51_SHA,

            "cell55":
                CELL55_SHA,

            "cell61":
                CELL61_SHA,

            "freeze_cell63":
                FREEZE_CELL_SHA,
        },
    },

    "artifact_lineages":
        artifact_lineages,

    "audit_variable_definitions":
        [
            compact(
                record
            )

            for record in audit_defs
        ],

    "actions_variable_definitions":
        [
            compact(
                record
            )

            for record in actions_defs
        ],

    "audit_latest_definition": (
        compact(
            latest_audit_def
        )
        if latest_audit_def
        else None
    ),

    "actions_latest_definition": (
        compact(
            latest_actions_def
        )
        if latest_actions_def
        else None
    ),

    "audit_is_historical_table":
        audit_is_historical_table,

    "actions_is_historical_table":
        actions_is_historical_table,

    "writer_contexts": [
        {
            "basename":
                context[
                    "basename"
                ],

            "writer_cell":
                context[
                    "writer_cell"
                ],

            "writer_line":
                context[
                    "writer_line"
                ],

            "object_name":
                context[
                    "object_name"
                ],

            "statements":
                [
                    compact(
                        statement
                    )

                    for statement
                    in context[
                        "statements"
                    ]
            ],
        }

        for context in writer_contexts
    ],

    "pre_freeze_field_records":
        compact_field_map(
            field_records
        ),

    "lineage_restricted_field_records": {
        "morph_rank":
            [
                compact(
                    record
                )

                for record
                in morph_rank_lineage_records
            ],

        "candidate_rank":
            [
                compact(
                    record
                )

                for record
                in candidate_rank_lineage_records
            ],

        "candidate_prob":
            [
                compact(
                    record
                )

                for record
                in candidate_prob_lineage_records
            ],
    },

    "strong_materialization_records": {
        "morph_rank":
            [
                compact(
                    record
                )

                for record
                in strong_morph_rank_records
            ],

        "candidate_rank":
            [
                compact(
                    record
                )

                for record
                in strong_candidate_rank_records
            ],

        "candidate_prob":
            [
                compact(
                    record
                )

                for record
                in strong_candidate_prob_records
            ],
    },

    "explicit_rank_in_dataset_to_morph_rank":
        [
            compact(
                record
            )

            for record
            in explicit_rank_to_morph
        ],

    "association_materialization_evidence": {
        "candidate_rank":
            [
                compact(
                    record
                )

                for record
                in association_candidate_rank_records
            ],

        "candidate_prob":
            [
                compact(
                    record
                )

                for record
                in association_candidate_prob_records
            ],
    },

    "component_status":
        component_status,

    "resolution": {
        "audit_writer_survives":
            audit_writer_survives,

        "upstream_action_writer_survives":
            upstream_action_writer_survives,

        "remaining_blockers":
            remaining_blockers,

        "status":
            status,

        "implementation_authorized":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_selection_policy":
            "FROZEN",

        "Morph_graph_application_semantics":
            "FROZEN",

        "reference_association_provider":
            "EXACT_SOURCE_AUTHORITY",

        "fresh_Morph_generator_authority":
            status,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_YET",

        "historical_pickle_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,

        "source_mutation":
            False,
    },

    "explicit_non_claims": [
        "A historical table carrying candidate fields is not fresh materialization authority.",
        "The cell61 canonicalization loop is not candidate-generation authority.",
        "No inferred rank/probability rule is invented when source dataflow is absent.",
        "No historical pickle was deserialized.",
        "No concrete MorphCandidateProvider was implemented.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-G artifact exists but differs "
        "from exact object-lineage closure state."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 30. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-G PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "OBJECT LINEAGE:",
    flush=True,
)

print(
    "  audit historical-table input:",
    audit_is_historical_table,
    flush=True,
)

print(
    "  audit writer survives:",
    audit_writer_survives,
    flush=True,
)

print(
    "  actions historical-table derived:",
    actions_is_historical_table,
    flush=True,
)

print(
    "  upstream actions writer survives:",
    upstream_action_writer_survives,
    flush=True,
)

print(
    flush=True,
)


print(
    "FIELD MATERIALIZATION:",
    flush=True,
)

print(
    "  p_T4 fresh producer:",
    component_status[
        "p_T4_ordered_fresh_producer"
    ],
    flush=True,
)

print(
    "  morph_rank:",
    component_status[
        "morph_rank_materialization"
    ],
    flush=True,
)

print(
    "  candidate_rank:",
    component_status[
        "candidate_rank_materialization"
    ],
    flush=True,
)

print(
    "  candidate_prob:",
    component_status[
        "candidate_probability_materialization"
    ],
    flush=True,
)

print(
    flush=True,
)


print(
    "RESOLUTION:",
    flush=True,
)

print(
    "  remaining blockers:",
    remaining_blockers,
    flush=True,
)

print(
    "  status:",
    status,
    flush=True,
)

print(
    "  implementation authorized:",
    False,
    flush=True,
)

print(
    "  NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  performance generated:",
    False,
    flush=True,
)

print(
    "  source mutation:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-H
# MORPH SPECIFICATION-DERIVED RECONSTRUCTION BOUNDARY FREEZE
#
# PARENT
# ------
# 14.3E-D-G PASS
#
# D-G proved:
#
#   audit
#       = historical pickle input
#       writer absent in current checkout
#
#   actions
#       = historical pickle input
#       writer absent in current checkout
#
#   p_T4_ordered_* fresh producer
#       = unresolved
#
#   morph_rank materialization
#       = unresolved historical action field
#
#   candidate_rank materialization
#       = unresolved historical audit/action field
#
#   candidate_prob materialization
#       = unresolved historical audit/action field
#
#   reference association provider
#       = exact source authority
#
#
# THIS STAGE
# ----------
# Freeze the reconstruction boundary.
#
# It DOES NOT reconstruct missing rules.
#
# It classifies every known semantic component into:
#
#   FRESH_RUNTIME_SOURCE_AUTHORITY
#   HISTORICAL_OOF_SEMANTIC_AUTHORITY_ONLY
#   FROZEN_SELECTION_SPECIFICATION
#   FROZEN_APPLICATION_SPECIFICATION
#   UNRESOLVED_FRESH_RUNTIME_AUTHORITY
#
#
# CRITICAL ADDITIONAL BOUNDARY
# ----------------------------
# Historical morphology ranking used:
#
#   df.groupby("fold")[col].rank(
#       pct=True,
#       method="average",
#   )
#
# `fold` is an OOF/evaluation partition.
#
# No current source authority defines the corresponding normalization cohort
# for an arbitrary fresh runtime dataset.
#
# Therefore NONE of these are authorized substitutions:
#
#   - treat a fresh dataset as one fold
#   - drop groupby("fold")
#   - rank over the complete evaluation batch
#   - rank over one dataset
#   - rank against historical OOF rows
#   - invent a calibration population
#
#
# HISTORICAL ORACLE RULE
# ----------------------
# Historical Morph action oracle remains CLOSED.
#
# It may only be opened AFTER:
#
#   1. a source-authorized/preregistered reconstruction exists,
#   2. its complete fresh runtime output is frozen,
#   3. output identity/signatures are committed to a formal artifact,
#
# and then ONLY for engineering structural fidelity comparison.
#
# Oracle mismatch must NOT be used to tune scientific parameters or invent
# missing semantics.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - historical pickle deserialization
#   - GT loading
#   - evaluator execution
#   - performance computation
#   - parameter search
#   - provider implementation
#   - structural replay
#   - historical-oracle access
#   - Git mutation
# =============================================================================

from __future__ import annotations

import ast
import hashlib
import json
import re

from pathlib import Path
from typing import Any


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_g_"
    "morph_remaining_primitive_targeted_closure_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


NOTEBOOK11_REL = Path(
    "notebooks/"
    "11_division_association_audit.ipynb"
)

NOTEBOOK11 = ROOT / NOTEBOOK11_REL


D2R3_REL = Path(
    "reports/"
    "notebook13_13_3d2r3_"
    "morph_authority_correction_and_"
    "semantic_reconstruction_freeze_v1.json"
)

D2R3 = ROOT / D2R3_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_h_"
    "morph_specification_derived_"
    "reconstruction_boundary_freeze_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "4a6d3491a77608de4b85c75f7f8e147dddebe57697f620553b08e31ace834f02"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "9537632860b20598d1afddf03ee3c83e13d0cd866659f1ab2511265fd5d039ba"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_D2R3_FILE_SHA = (
    "ac0cb4959b8a9cb54784d0fd37d4da89a87b6e110298f67af00e712deebf8afb"
)

EXPECTED_D2R3_LOGICAL_CONTENT_SHA = (
    "9a0ca7a1d2013a9cd0ddfe74ec00e9f787505c19ca69c6d0c4e297db03a701b0"
)

EXPECTED_AUTHORITY_MODE = (
    "SEMANTIC_SOURCE_LOCK_PLUS_"
    "HISTORICAL_GT_BLIND_STRUCTURAL_ORACLE"
)


EXPECTED_ASSOCIATION_PROVIDER_SHA = (
    "7aae212c3ae52e1070925386fb2775feec0c964337ec48441f1da02d7287a1b1"
)


CELL46 = 46

CELL46_SHA = (
    "15747465b4e39aa4a2e72e259a9cdbbdfa83c05395c939904b6b6a4f436f36b1"
)


FREEZE_CELL = 63

FREEZE_CELL_SHA = (
    "2e6bdff585ad1a3ee61b3a82ec03ccbd05db0b66f01024a35a721d76a29d8834"
)


# Exact historical morphology statements.
PRED_COLS_SHA = (
    "405c4e90634bb0e136cdc0e3d224f05d573274aeac85a7a849cb6b7eb35caf83"
)

RANK_COLS_LOOP_SHA = (
    "368dac4f1063fd680e77529af9a377501b84ee32a7dcdec1ca764730216a3400"
)

MORPH_SCORE_SHA = (
    "1eb422867db49a6255e89ae08443846a0c03458f484a9cc61065375800e7bad4"
)

MORPH_SORT_SHA = (
    "9b68efe7475096a06acd91a25ca0bc2f721f1284125048daaa895b3070315716"
)

DATASET_RANK_SHA = (
    "7606a5627025b8834d6025ab39db54048127241c46b1b8ce301068adabcc698f"
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(
    value: bytes,
) -> str:

    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(
    value: str,
) -> str:

    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(
    path: Path,
) -> str:

    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(
    value: Any,
) -> str:

    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(
    path: Path,
) -> Any:

    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def notebook_cells(
    path: Path,
) -> list[
    dict[str, Any]
]:

    notebook = json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


    result = []


    for index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        source = "".join(
            cell.get(
                "source",
                []
            )
        )


        result.append({
            "index":
                index,

            "type":
                cell.get(
                    "cell_type"
                ),

            "source":
                source,

            "sha256":
                sha256_text(
                    source
                ),
        })


    return result


def flatten_json(
    obj: Any,
    *,
    prefix: str = "$",
) -> list[
    dict[str, Any]
]:

    result = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in obj.items():

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}.{key}"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            result.extend(
                flatten_json(
                    value,
                    prefix=(
                        f"{prefix}[{index}]"
                    ),
                )
            )


    else:

        result.append({
            "path":
                prefix,

            "value":
                obj,
        })


    return result


def contains_exact_value(
    obj: Any,
    target: Any,
) -> bool:

    return any(
        leaf[
            "value"
        ]
        == target

        for leaf in flatten_json(
            obj
        )
    )


def find_node_by_sha(
    source: str,
    expected_sha: str,
) -> list[
    dict[str, Any]
]:

    tree = ast.parse(
        source
    )


    result = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.For,
                ast.Expr,
            ),
        ):

            continue


        segment = ast.get_source_segment(
            source,
            node,
        )


        if segment is None:

            continue


        observed = sha256_text(
            segment
        )


        if observed != expected_sha:

            continue


        result.append({
            "node":
                node,

            "kind":
                type(
                    node
                ).__name__,

            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "source":
                segment,

            "sha256":
                observed,
        })


    return result


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(
    path: Path,
) -> dict[str, Any]:

    assert (
        path.exists()
    )


    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )


    obj = load_json(
        path
    )


    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )


    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )


    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )


    assert (
        obj[
            "resolution"
        ][
            "status"
        ]
        ==
        "CURRENT_CHECKOUT_HAS_NO_COMPLETE_"
        "FRESH_MORPH_GENERATOR_AUTHORITY"
    )


    assert (
        obj[
            "resolution"
        ][
            "implementation_authorized"
        ]
        is False
    )


    expected_blockers = {
        "p_T4_ordered_fresh_producer",
        "morph_rank_materialization",
        "candidate_rank_materialization",
        "candidate_probability_materialization",
    }


    assert (
        set(
            obj[
                "resolution"
            ][
                "remaining_blockers"
            ]
        )
        ==
        expected_blockers
    )


    assert (
        obj[
            "audit_is_historical_table"
        ]
        is True
    )


    assert (
        obj[
            "actions_is_historical_table"
        ]
        is True
    )


    assert (
        obj[
            "resolution"
        ][
            "audit_writer_survives"
        ]
        is False
    )


    assert (
        obj[
            "resolution"
        ][
            "upstream_action_writer_survives"
        ]
        is False
    )


    return obj


# =============================================================================
# 4. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    NOTEBOOK11,
    D2R3,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-H "
    "MORPH SPECIFICATION-DERIVED "
    "RECONSTRUCTION BOUNDARY FREEZE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Parent lock
# =============================================================================

parent = verify_parent(
    PARENT
)


print(
    "[PASS] PARENT_14_3E_D_G_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  status:",
    parent[
        "resolution"
    ][
        "status"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Current source + D2-R3 locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


assert (
    EXPECTED_ASSOCIATION_PROVIDER_SHA
    in portable_text
)


d2r3_sha = sha256_file(
    D2R3
)


assert (
    d2r3_sha
    == EXPECTED_D2R3_FILE_SHA
)


d2r3 = load_json(
    D2R3
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_D2R3_LOGICAL_CONTENT_SHA,
    )
)


assert (
    contains_exact_value(
        d2r3,
        EXPECTED_AUTHORITY_MODE,
    )
)


print(
    "[PASS] CURRENT_SOURCE_AND_D2_R3_AUTHORITY_LOCK",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  association provider SHA:",
    EXPECTED_ASSOCIATION_PROVIDER_SHA,
    flush=True,
)

print(
    "  D2-R3 SHA:",
    d2r3_sha,
    flush=True,
)

print(
    "  authority mode:",
    EXPECTED_AUTHORITY_MODE,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Exact historical morphology transform
# =============================================================================

cells = notebook_cells(
    NOTEBOOK11
)


assert (
    cells[
        CELL46
    ][
        "sha256"
    ]
    == CELL46_SHA
)


assert (
    cells[
        FREEZE_CELL
    ][
        "sha256"
    ]
    == FREEZE_CELL_SHA
)


cell46_source = cells[
    CELL46
][
    "source"
]


statement_specs = {
    "pred_cols":
        PRED_COLS_SHA,

    "rank_cols_loop":
        RANK_COLS_LOOP_SHA,

    "morph_score":
        MORPH_SCORE_SHA,

    "morph_sort":
        MORPH_SORT_SHA,

    "dataset_rank":
        DATASET_RANK_SHA,
}


statement_records = {}


for name, expected_sha in statement_specs.items():

    matches = find_node_by_sha(
        cell46_source,
        expected_sha,
    )


    assert (
        len(
            matches
        )
        == 1
    ), (
        f"{name}: expected exactly one exact source node, "
        f"found {len(matches)}."
    )


    statement_records[
        name
    ] = matches[
        0
    ]


rank_loop_source = (
    statement_records[
        "rank_cols_loop"
    ][
        "source"
    ]
)


assert (
    'groupby("fold")'
    in rank_loop_source
)


assert (
    ".rank("
    in rank_loop_source
)


assert (
    "pct=True"
    in rank_loop_source
)


assert (
    'method="average"'
    in rank_loop_source
)


sort_source = (
    statement_records[
        "morph_sort"
    ][
        "source"
    ]
)


for token in (
    '"dataset"',
    '"morph_score"',
    '"t"',
    '"node_id"',
):

    assert (
        token
        in sort_source
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "HISTORICAL MORPHOLOGY TRANSFORM AUTHORITY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for name, record in statement_records.items():

    print(
        name,
        flush=True,
    )

    print(
        "  SHA:",
        record[
            "sha256"
        ],
        flush=True,
    )

    print(
        "  source:",
        flush=True,
    )

    print(
        record[
            "source"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 8. Fresh-runtime portability classification
# =============================================================================

#
# Historical formula authority != deployable runtime authority.
#
# Specifically:
#
#   groupby("fold").rank(...)
#
# requires an OOF fold population.
#
# No source authority currently defines how to reproduce that population or
# normalization semantics for an arbitrary fresh dataset.
#

historical_fold_normalization_authority = (
    True
)


fresh_runtime_normalization_cohort_authority = (
    False
)


fresh_p_T4_producer_authority = (
    False
)


fresh_morph_rank_materialization_authority = (
    False
)


fresh_candidate_rank_materialization_authority = (
    False
)


fresh_candidate_prob_materialization_authority = (
    False
)


association_provider_source_authority = (
    True
)


# =============================================================================
# 9. Freeze authority matrix
# =============================================================================

authority_matrix = {
    # ---------------------------------------------------------------------
    # Already source/spec locked.
    # ---------------------------------------------------------------------
    "candidate_universe": {
        "status":
            "FROZEN_SELECTION_SPECIFICATION",

        "authority":
            (
                "FINAL Frozen-V9 observable W4-capable parents "
                "with current outdegree == 1"
            ),
    },

    "morphology_top_k": {
        "status":
            "FROZEN_SELECTION_SPECIFICATION",

        "value":
            3,
    },

    "association_rank_cap": {
        "status":
            "FROZEN_SELECTION_SPECIFICATION",

        "value":
            3,
    },

    "association_probability_threshold": {
        "status":
            "FROZEN_SELECTION_SPECIFICATION",

        "value":
            0.10,
    },

    "graph_application": {
        "status":
            "FROZEN_APPLICATION_SPECIFICATION",

        "source_pre_outdegree":
            1,

        "target_pre_indegree":
            0,

        "duplicate_edge":
            "FORBIDDEN",

        "source_post_outdegree":
            2,

        "target_post_indegree":
            1,

        "operation":
            "ADDITIVE_SECOND_EDGE",
    },

    "reference_association_provider": {
        "status":
            "FRESH_RUNTIME_SOURCE_AUTHORITY",

        "identity_sha256":
            EXPECTED_ASSOCIATION_PROVIDER_SHA,
    },

    # ---------------------------------------------------------------------
    # Historical morphology semantics.
    # ---------------------------------------------------------------------
    "historical_morphology_probability_column_schema": {
        "status":
            "HISTORICAL_OOF_SEMANTIC_AUTHORITY_ONLY",

        "source_sha256":
            PRED_COLS_SHA,

        "runtime_portable":
            False,

        "reason":
            "Fresh producer for p_T4_ordered_* is absent.",
    },

    "historical_morphology_percentile_rank_transform": {
        "status":
            "HISTORICAL_OOF_SEMANTIC_AUTHORITY_ONLY",

        "source_sha256":
            RANK_COLS_LOOP_SHA,

        "formula":
            (
                'df.groupby("fold")[col].rank('
                'pct=True, method="average")'
            ),

        "runtime_portable":
            False,

        "reason":
            (
                "The normalization cohort is the historical OOF fold. "
                "Fresh runtime cohort semantics are not source-authorized."
            ),
    },

    "historical_morphology_score_aggregation": {
        "status":
            "HISTORICAL_OOF_SEMANTIC_AUTHORITY_ONLY",

        "source_sha256":
            MORPH_SCORE_SHA,

        "formula":
            "mean(percentile_rank_columns)",

        "runtime_portable":
            False,

        "reason":
            (
                "Depends on historical fold-normalized rank columns."
            ),
    },

    "historical_morphology_ordering": {
        "status":
            "HISTORICAL_OOF_SEMANTIC_AUTHORITY_ONLY",

        "source_sha256":
            MORPH_SORT_SHA,

        "sort":
            [
                "dataset ASC",
                "morph_score DESC",
                "t ASC",
                "node_id ASC",
            ],

        "runtime_portable":
            False,

        "reason":
            (
                "morph_score itself lacks fresh-runtime normalization "
                "authority."
            ),
    },

    "historical_dataset_rank": {
        "status":
            "HISTORICAL_OOF_SEMANTIC_AUTHORITY_ONLY",

        "source_sha256":
            DATASET_RANK_SHA,

        "formula":
            'df.groupby("dataset").cumcount() + 1',

        "runtime_portable":
            False,

        "reason":
            (
                "Depends on the historically-defined morphology ordering."
            ),
    },

    # ---------------------------------------------------------------------
    # Explicit unresolved fresh primitives.
    # ---------------------------------------------------------------------
    "fresh_p_T4_ordered_producer": {
        "status":
            "UNRESOLVED_FRESH_RUNTIME_AUTHORITY",
    },

    "fresh_morphology_normalization_cohort": {
        "status":
            "UNRESOLVED_FRESH_RUNTIME_AUTHORITY",

        "historical_source_used":
            "fold",

        "forbidden_inference":
            True,
    },

    "fresh_morph_rank_materialization": {
        "status":
            "UNRESOLVED_FRESH_RUNTIME_AUTHORITY",

        "explicit_rank_in_dataset_to_morph_rank_mapping_found":
            False,
    },

    "fresh_candidate_rank_materialization": {
        "status":
            "UNRESOLVED_FRESH_RUNTIME_AUTHORITY",

        "association_provider_available":
            True,

        "provider_output_to_candidate_rank_mapping_found":
            False,
    },

    "fresh_candidate_prob_materialization": {
        "status":
            "UNRESOLVED_FRESH_RUNTIME_AUTHORITY",

        "association_provider_available":
            True,

        "provider_output_to_candidate_prob_mapping_found":
            False,
    },
}


# =============================================================================
# 10. Freeze forbidden reconstruction substitutions
# =============================================================================

forbidden_reconstructions = [
    {
        "id":
            "NO_FRESH_DATASET_AS_FOLD",

        "rule":
            (
                "Do not replace historical groupby('fold') normalization "
                "with one fresh dataset treated as a fold."
            ),
    },

    {
        "id":
            "NO_DROP_FOLD_NORMALIZATION",

        "rule":
            (
                "Do not remove percentile normalization or rank raw "
                "p_T4_ordered_* values directly."
            ),
    },

    {
        "id":
            "NO_INVENTED_RUNTIME_CALIBRATION_POPULATION",

        "rule":
            (
                "Do not invent a cohort/reference population for percentile "
                "ranking."
            ),
    },

    {
        "id":
            "NO_MORPH_RANK_ALIAS_INFERENCE",

        "rule":
            (
                "Do not assume morph_rank == rank_in_dataset without "
                "source authority."
            ),
    },

    {
        "id":
            "NO_CANDIDATE_RANK_ARGSORT_INFERENCE",

        "rule":
            (
                "Do not assume candidate_rank is a stable descending argsort "
                "of the reference association-provider matrix unless an "
                "authority source explicitly freezes that mapping."
            ),
    },

    {
        "id":
            "NO_CANDIDATE_PROB_MATRIX_ENTRY_INFERENCE",

        "rule":
            (
                "Do not assume candidate_prob is directly the association "
                "matrix entry for target node without source authority."
            ),
    },

    {
        "id":
            "NO_HISTORICAL_ACTION_RUNTIME_LOOKUP",

        "rule":
            (
                "Historical Morph/audit/action pickles may not be used as "
                "runtime candidate sources."
            ),
    },

    {
        "id":
            "NO_ORACLE_DRIVEN_RULE_RECOVERY",

        "rule":
            (
                "Do not inspect the historical structural oracle and then "
                "modify formulas, tie-breaks, thresholds, ranking semantics, "
                "or candidate construction to improve agreement."
            ),
    },
]


# =============================================================================
# 11. Freeze allowed reconstruction operations
# =============================================================================

allowed_without_new_scientific_authority = [
    {
        "operation":
            "PRESERVE_EXISTING_MORPH_CANDIDATE_PROVIDER_SEAM",

        "reason":
            (
                "PortableCurrentBestContext already requires an explicit "
                "MorphCandidateProvider dependency."
            ),
    },

    {
        "operation":
            "PRESERVE_FROZEN_SELECTION_POLICY",

        "reason":
            (
                "Top-3 morphology, association rank <=3 and "
                "candidate_prob >=0.10 are already frozen."
            ),
    },

    {
        "operation":
            "PRESERVE_FROZEN_GRAPH_APPLICATION",

        "reason":
            (
                "Additive second-edge topology semantics are already frozen."
            ),
    },

    {
        "operation":
            "USE_REFERENCE_ASSOCIATION_PROVIDER_ONLY_WITH_EXPLICIT_MAPPING",

        "reason":
            (
                "Association provider source authority exists, but its output "
                "cannot be mapped to Morph candidate_rank/candidate_prob "
                "until that mapping is independently source-authorized."
            ),
    },

    {
        "operation":
            "READ_ONLY_AUTHORITY_RECOVERY_OUTSIDE_CURRENT_CHECKOUT",

        "reason":
            (
                "Recovering archived/source-history authority is allowed "
                "provided no repository state is mutated and no historical "
                "structural oracle is opened."
            ),
    },
]


# =============================================================================
# 12. Historical-oracle access gate
# =============================================================================

fresh_reconstruction_contract_complete = (
    False
)


fresh_reconstruction_implemented = (
    False
)


fresh_output_frozen_before_oracle = (
    False
)


historical_oracle_access_authorized = (
    fresh_reconstruction_contract_complete
    and
    fresh_reconstruction_implemented
    and
    fresh_output_frozen_before_oracle
)


assert (
    historical_oracle_access_authorized
    is False
)


oracle_access_contract = {
    "current_state":
        "CLOSED",

    "authorized_now":
        False,

    "required_before_open": [
        "SOURCE_AUTHORIZED_OR_PREREGISTERED_RECONSTRUCTION_CONTRACT_COMPLETE",
        "RECONSTRUCTION_IMPLEMENTED_WITHOUT_ORACLE_ACCESS",
        "COMPLETE_FRESH_OUTPUT_FROZEN",
        "FRESH_OUTPUT_IDENTITY_AND_SIGNATURES_FORMALLY_HASH_LOCKED",
    ],

    "allowed_future_use":
        "ENGINEERING_STRUCTURAL_FIDELITY_ONLY",

    "forbidden_future_use": [
        "scientific_parameter_tuning",
        "threshold_selection",
        "tie_break_selection",
        "candidate_generation_rule_discovery",
        "normalization_cohort_discovery",
        "performance_claim",
    ],
}


# =============================================================================
# 13. Current readiness classification
# =============================================================================

unresolved_fresh_runtime_components = [
    key

    for key, value
    in authority_matrix.items()

    if (
        value[
            "status"
        ]
        ==
        "UNRESOLVED_FRESH_RUNTIME_AUTHORITY"
    )
]


assert (
    set(
        unresolved_fresh_runtime_components
    )
    ==
    {
        "fresh_p_T4_ordered_producer",
        "fresh_morphology_normalization_cohort",
        "fresh_morph_rank_materialization",
        "fresh_candidate_rank_materialization",
        "fresh_candidate_prob_materialization",
    }
)


current_checkout_archaeology_closed = (
    True
)


concrete_Morph_provider_implementation_authorized = (
    False
)


fresh_post_V2_runtime_ready = (
    False
)


structural_replay_ready = (
    False
)


status = (
    "RECONSTRUCTION_BOUNDARY_FROZEN_"
    "IMPLEMENTATION_BLOCKED_PENDING_NEW_SOURCE_AUTHORITY"
)


next_stage = (
    "NOTEBOOK14_14_3E_D_I_"
    "MORPH_OUTSIDE_CURRENT_CHECKOUT_"
    "READ_ONLY_AUTHORITY_RECOVERY"
)


print(
    "=" * 120,
    flush=True,
)

print(
    "RECONSTRUCTION AUTHORITY MATRIX",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for key, value in authority_matrix.items():

    print(
        key,
        "=>",
        value[
            "status"
        ],
        flush=True,
    )


print(
    flush=True,
)


print(
    "UNRESOLVED FRESH RUNTIME COMPONENTS:",
    unresolved_fresh_runtime_components,
    flush=True,
)

print(
    flush=True,
)


print(
    "CURRENT CHECKOUT ARCHAEOLOGY CLOSED:",
    current_checkout_archaeology_closed,
    flush=True,
)

print(
    "CONCRETE PROVIDER IMPLEMENTATION AUTHORIZED:",
    concrete_Morph_provider_implementation_authorized,
    flush=True,
)

print(
    "HISTORICAL ORACLE ACCESS AUTHORIZED:",
    historical_oracle_access_authorized,
    flush=True,
)

print(
    "FRESH POST-V2 RUNTIME READY:",
    fresh_post_V2_runtime_ready,
    flush=True,
)

print(
    "STRUCTURAL REPLAY READY:",
    structural_replay_ready,
    flush=True,
)

print(
    flush=True,
)

print(
    "STATUS:",
    status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Safety / no mutation
# =============================================================================

source_mutation = False

test_mutation = False

historical_pickle_loaded = False

historical_oracle_opened = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


# =============================================================================
# 15. Formal gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_g_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "D2_R3_authority_locked":
        (
            d2r3_sha
            == EXPECTED_D2R3_FILE_SHA
        ),

    "historical_morphology_transform_exactly_locked":
        (
            len(
                statement_records
            )
            ==
            len(
                statement_specs
            )
        ),

    "historical_fold_normalization_explicitly_identified":
        historical_fold_normalization_authority,

    "fresh_runtime_normalization_cohort_not_invented":
        (
            fresh_runtime_normalization_cohort_authority
            is False
        ),

    "fresh_p_T4_producer_not_invented":
        (
            fresh_p_T4_producer_authority
            is False
        ),

    "fresh_morph_rank_mapping_not_invented":
        (
            fresh_morph_rank_materialization_authority
            is False
        ),

    "fresh_candidate_rank_mapping_not_invented":
        (
            fresh_candidate_rank_materialization_authority
            is False
        ),

    "fresh_candidate_prob_mapping_not_invented":
        (
            fresh_candidate_prob_materialization_authority
            is False
        ),

    "reference_association_provider_preserved":
        association_provider_source_authority,

    "forbidden_reconstruction_substitutions_frozen":
        (
            len(
                forbidden_reconstructions
            )
            >= 8
        ),

    "historical_oracle_remains_closed":
        (
            historical_oracle_access_authorized
            is False
            and
            historical_oracle_opened
            is False
        ),

    "current_checkout_archaeology_closed":
        current_checkout_archaeology_closed,

    "concrete_provider_not_prematurely_authorized":
        (
            concrete_Morph_provider_implementation_authorized
            is False
        ),

    "post_V2_runtime_not_prematurely_claimed_ready":
        (
            fresh_post_V2_runtime_ready
            is False
        ),

    "structural_replay_not_prematurely_claimed_ready":
        (
            structural_replay_ready
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-H SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"  "
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-H failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 16. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_H",

    "title":
        "MORPH_SPECIFICATION_DERIVED_RECONSTRUCTION_BOUNDARY_FREEZE",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        (
            "ENGINEERING_RECONSTRUCTION_BOUNDARY_FROZEN_"
            "MISSING_FRESH_SOURCE_AUTHORITY"
        ),

    "parent_14_3e_d_g": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "file_sha256":
                portable_sha_before,

            "reference_association_provider_sha256":
                EXPECTED_ASSOCIATION_PROVIDER_SHA,
        },

        "D2_R3": {
            "path":
                str(
                    D2R3_REL
                ),

            "file_sha256":
                d2r3_sha,

            "logical_content_sha256":
                EXPECTED_D2R3_LOGICAL_CONTENT_SHA,

            "authority_mode":
                EXPECTED_AUTHORITY_MODE,
        },

        "Notebook11": {
            "path":
                str(
                    NOTEBOOK11_REL
                ),

            "cell46_sha256":
                CELL46_SHA,

            "freeze_cell63_sha256":
                FREEZE_CELL_SHA,
        },
    },

    "historical_morphology_transform": {
        key: {
            "source_sha256":
                record[
                    "sha256"
                ],

            "source":
                record[
                    "source"
                ],
        }

        for key, record
        in statement_records.items()
    },

    "authority_matrix":
        authority_matrix,

    "unresolved_fresh_runtime_components":
        unresolved_fresh_runtime_components,

    "forbidden_reconstructions":
        forbidden_reconstructions,

    "allowed_without_new_scientific_authority":
        allowed_without_new_scientific_authority,

    "historical_oracle_access_contract":
        oracle_access_contract,

    "resolution": {
        "current_checkout_archaeology_closed":
            current_checkout_archaeology_closed,

        "fresh_reconstruction_contract_complete":
            fresh_reconstruction_contract_complete,

        "concrete_Morph_provider_implementation_authorized":
            concrete_Morph_provider_implementation_authorized,

        "historical_oracle_access_authorized":
            historical_oracle_access_authorized,

        "fresh_post_V2_runtime_ready":
            fresh_post_V2_runtime_ready,

        "structural_replay_ready":
            structural_replay_ready,

        "status":
            status,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "Morph_selection_policy":
            "FROZEN",

        "Morph_graph_application_semantics":
            "FROZEN",

        "reference_association_provider":
            "EXACT_SOURCE_AUTHORITY",

        "historical_OOF_morphology_transform":
            "SOURCE_LOCKED_NOT_FRESH_RUNTIME_PORTABLE",

        "fresh_morphology_probability_producer":
            "UNRESOLVED",

        "fresh_morphology_normalization_cohort":
            "UNRESOLVED",

        "fresh_morph_rank_materialization":
            "UNRESOLVED",

        "fresh_candidate_rank_materialization":
            "UNRESOLVED",

        "fresh_candidate_prob_materialization":
            "UNRESOLVED",

        "concrete_Morph_provider_implementation":
            "BLOCKED_PENDING_NEW_SOURCE_AUTHORITY",

        "historical_structural_oracle":
            "CLOSED",

        "fresh_end_to_end_post_V2_runtime":
            "NOT_READY",

        "post_V2_structural_replay":
            "NOT_READY",

        "B12_structural_fidelity":
            "NOT_VERIFIED",

        "historical_pickle_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,

        "source_mutation":
            False,
    },

    "explicit_non_claims": [
        "Historical OOF fold normalization is not treated as fresh-runtime normalization authority.",
        "No fresh morphology calibration cohort is invented.",
        "No morph_rank alias is inferred.",
        "No candidate_rank mapping is inferred from the association matrix.",
        "No candidate_prob mapping is inferred from the association matrix.",
        "No historical action/audit pickle is used at runtime.",
        "Historical structural oracle remains unopened.",
        "No concrete MorphCandidateProvider was implemented.",
        "No post-V2 runtime readiness claim.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-H artifact exists but differs "
        "from exact reconstruction-boundary state."
    )


    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    OUT.write_bytes(
        artifact_bytes
    )


    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 17. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-H PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "BOUNDARY:",
    flush=True,
)

print(
    "  current-checkout archaeology closed:",
    current_checkout_archaeology_closed,
    flush=True,
)

print(
    "  historical OOF morphology semantics:",
    "SOURCE LOCKED",
    flush=True,
)

print(
    "  historical OOF transform fresh-runtime portable:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "UNRESOLVED FRESH AUTHORITY:",
    flush=True,
)

for value in unresolved_fresh_runtime_components:

    print(
        " ",
        value,
        flush=True,
    )


print(
    flush=True,
)


print(
    "AUTHORIZATION:",
    flush=True,
)

print(
    "  concrete Morph provider implementation:",
    concrete_Morph_provider_implementation_authorized,
    flush=True,
)

print(
    "  historical oracle access:",
    historical_oracle_access_authorized,
    flush=True,
)

print(
    "  fresh post-V2 runtime ready:",
    fresh_post_V2_runtime_ready,
    flush=True,
)

print(
    "  structural replay ready:",
    structural_replay_ready,
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  source mutation:",
    False,
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  historical oracle opened:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  performance generated:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-I-R1-R1
# GIT HISTORY AVAILABILITY HARNESS REPAIR
# — NOTEBOOK CELL SYNTAX REPAIR
#
# SUPERSEDES
# ----------
# 14.3E-D-I-R1
#
# Previous failure:
#   SyntaxError before execution.
#
# Therefore previous R1 executed:
#   Git commands: 0
#   source mutation: False
#   historical pickle access: False
#   historical oracle access: False
#
#
# PURPOSE
# -------
# Determine READ-ONLY whether Git history is recoverable despite:
#
#   git rev-parse HEAD -> rc 128
#
# Search:
#
#   1. symbolic HEAD
#   2. refs/heads
#   3. refs/remotes
#   4. refs/tags
#   5. commits reachable from refs
#   6. dangling / unreachable commit objects
#
#
# READ-ONLY GIT ONLY
# ------------------
# allowed:
#   rev-parse
#   symbolic-ref
#   status
#   for-each-ref
#   rev-list
#   fsck
#
# forbidden:
#   checkout / switch / restore
#   reset / stash / clean
#   add / commit
#   merge / rebase / cherry-pick / revert
#   branch/tag mutation
#
# No source mutation.
# No historical pickle load.
# No structural oracle open.
# No GT.
# No evaluator.
# No performance.
# =============================================================================

import hashlib
import json
import re
import subprocess

from pathlib import Path


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_h_"
    "morph_specification_derived_"
    "reconstruction_boundary_freeze_v1.json"
)

PARENT = ROOT / PARENT_REL

PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL

FAILED_R1_REL = Path(
    "reports/"
    "notebook14_14_3e_d_i_r1_"
    "git_history_availability_harness_repair_v1.json"
)

FAILED_R1 = ROOT / FAILED_R1_REL

OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_i_r1_r1_"
    "git_history_availability_cell_syntax_repair_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "df62542ae6217564a6241657f7a4bc79a789ece5858c9e77217f092153cbcfe0"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "d3cbfd2e549c221be945730f4118a59170f3cbd313e2ef6fc764c74fcfc371a9"
)

EXPECTED_PARENT_STATUS = (
    "RECONSTRUCTION_BOUNDARY_FROZEN_"
    "IMPLEMENTATION_BLOCKED_PENDING_NEW_SOURCE_AUTHORITY"
)

EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)

EXPECTED_UNRESOLVED = {
    "fresh_p_T4_ordered_producer",
    "fresh_morphology_normalization_cohort",
    "fresh_morph_rank_materialization",
    "fresh_candidate_rank_materialization",
    "fresh_candidate_prob_materialization",
}


# =============================================================================
# 2. Basic helpers
# =============================================================================

def sha256_bytes(value):
    return hashlib.sha256(value).hexdigest()


def sha256_file(path):
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value):
    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_json(path):
    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(path):

    assert path.exists(), path

    actual_file_sha = sha256_file(
        path
    )

    assert (
        actual_file_sha
        == EXPECTED_PARENT_FILE_SHA
    ), (
        "D-H parent file SHA drift.\n"
        f"expected={EXPECTED_PARENT_FILE_SHA}\n"
        f"actual={actual_file_sha}"
    )

    obj = load_json(
        path
    )

    assert (
        obj.get("decision")
        == "PASS"
    )

    assert (
        obj.get("content_sha256")
        == EXPECTED_PARENT_CONTENT_SHA
    )

    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )

    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )

    assert (
        obj[
            "resolution"
        ][
            "status"
        ]
        == EXPECTED_PARENT_STATUS
    )

    assert (
        set(
            obj[
                "unresolved_fresh_runtime_components"
            ]
        )
        == EXPECTED_UNRESOLVED
    )

    assert (
        obj[
            "resolution"
        ][
            "concrete_Morph_provider_implementation_authorized"
        ]
        is False
    )

    assert (
        obj[
            "resolution"
        ][
            "historical_oracle_access_authorized"
        ]
        is False
    )

    return obj


# =============================================================================
# 4. Strict read-only Git wrapper
# =============================================================================

ALLOWED_GIT_SUBCOMMANDS = {
    "rev-parse",
    "symbolic-ref",
    "status",
    "for-each-ref",
    "rev-list",
    "fsck",
}

git_command_log = []


def run_git(args, check=True):

    assert args

    assert (
        args[0]
        in ALLOWED_GIT_SUBCOMMANDS
    ), (
        "Git subcommand not authorized: "
        f"{args[0]}"
    )

    cmd = [
        "git",
        *args,
    ]

    git_command_log.append(
        cmd
    )

    process = subprocess.Popen(
        cmd,
        cwd=ROOT,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )

    assert (
        process.stdout
        is not None
    )

    output_lines = []

    for line in process.stdout:
        output_lines.append(
            line
        )

    return_code = process.wait()

    output = "".join(
        output_lines
    )

    if (
        check
        and
        return_code != 0
    ):
        raise RuntimeError(
            "Read-only Git command failed.\n"
            f"CMD: {' '.join(cmd)}\n"
            f"RC: {return_code}\n"
            f"OUTPUT:\n{output}"
        )

    return (
        return_code,
        output,
    )


# =============================================================================
# 5. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
):
    assert required.exists(), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-I-R1-R1 "
    "GIT HISTORY AVAILABILITY HARNESS REPAIR "
    "— CELL SYNTAX REPAIR",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Exact parent + current-source lock
# =============================================================================

parent = verify_parent(
    PARENT
)

portable_sha_before = sha256_file(
    PORTABLE
)

assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)

print(
    "[PASS] D_H_PARENT_AND_PORTABLE_LOCK",
    flush=True,
)

print(
    "  D-H file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  D-H content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  failed R1 artifact exists:",
    FAILED_R1.exists(),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Git root
# =============================================================================

root_rc, root_output = run_git(
    [
        "rev-parse",
        "--show-toplevel",
    ]
)

assert (
    root_rc == 0
)

GIT_ROOT = Path(
    root_output.strip()
).resolve()

assert (
    GIT_ROOT
    == ROOT
)

_, status_before = run_git(
    [
        "status",
        "--short",
    ]
)

print(
    "[PASS] GIT_WORKTREE_IDENTITY",
    flush=True,
)

print(
    "  Git root:",
    GIT_ROOT,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. HEAD state
# =============================================================================

symbolic_rc, symbolic_output = run_git(
    [
        "symbolic-ref",
        "-q",
        "HEAD",
    ],
    check=False,
)

symbolic_head = (
    symbolic_output.strip()
    if (
        symbolic_rc == 0
        and
        symbolic_output.strip()
    )
    else None
)

head_rc, head_output = run_git(
    [
        "rev-parse",
        "--verify",
        "HEAD",
    ],
    check=False,
)

head_commit = None

if (
    head_rc == 0
    and
    head_output.strip()
):
    candidate = (
        head_output
        .strip()
        .splitlines()[-1]
    )

    if re.fullmatch(
        r"[0-9a-fA-F]{40,64}",
        candidate,
    ):
        head_commit = candidate

head_commit_available = (
    head_commit
    is not None
)

print(
    "=" * 120,
    flush=True,
)

print(
    "HEAD STATE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "symbolic HEAD:",
    symbolic_head,
    flush=True,
)

print(
    "HEAD commit available:",
    head_commit_available,
    flush=True,
)

print(
    "HEAD verify rc:",
    head_rc,
    flush=True,
)

print(
    "HEAD diagnostic:",
    head_output.strip(),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. Enumerate refs
# =============================================================================

refs_rc, refs_output = run_git(
    [
        "for-each-ref",
        "--format=%(objectname)%09%(objecttype)%09%(refname)",
        "refs/heads",
        "refs/remotes",
        "refs/tags",
    ]
)

assert (
    refs_rc == 0
)

ref_records = []

for line in refs_output.splitlines():

    if not line.strip():
        continue

    parts = line.split(
        "\t",
        2,
    )

    if len(parts) != 3:
        continue

    object_id, object_type, ref_name = parts

    ref_records.append({
        "object_id":
            object_id,

        "object_type":
            object_type,

        "ref_name":
            ref_name,
    })


print(
    "=" * 120,
    flush=True,
)

print(
    "REF INVENTORY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "ref count:",
    len(ref_records),
    flush=True,
)

for record in ref_records:
    print(
        " ",
        record["ref_name"],
        record["object_type"],
        record["object_id"],
        flush=True,
    )

print(
    flush=True,
)


# =============================================================================
# 10. Reachable commits
# =============================================================================

rev_rc, rev_output = run_git(
    [
        "rev-list",
        "--all",
    ],
    check=False,
)

reachable_commits = []

if rev_rc == 0:

    for line in rev_output.splitlines():

        commit = line.strip()

        if re.fullmatch(
            r"[0-9a-fA-F]{40,64}",
            commit,
        ):
            reachable_commits.append(
                commit
            )

reachable_commits = sorted(
    set(
        reachable_commits
    )
)

print(
    "reachable commits from all refs:",
    len(reachable_commits),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Unreachable / dangling commit-object recovery
# =============================================================================

#
# Read-only:
#
# git fsck --full --no-reflogs --unreachable --no-progress
#
# No --lost-found is used.
#

fsck_rc, fsck_output = run_git(
    [
        "fsck",
        "--full",
        "--no-reflogs",
        "--unreachable",
        "--no-progress",
    ],
    check=False,
)

commit_pattern = re.compile(
    r"(?:unreachable|dangling)\s+commit\s+"
    r"([0-9a-fA-F]{40,64})"
)

unreachable_commits = sorted(
    set(
        commit_pattern.findall(
            fsck_output
        )
    )
)

print(
    "=" * 120,
    flush=True,
)

print(
    "UNREACHABLE / DANGLING COMMIT OBJECTS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "fsck rc:",
    fsck_rc,
    flush=True,
)

print(
    "commit objects:",
    len(unreachable_commits),
    flush=True,
)

for commit in unreachable_commits[:100]:
    print(
        " ",
        commit,
        flush=True,
    )

if fsck_output.strip():

    print(
        flush=True,
    )

    print(
        "fsck diagnostic preview:",
        flush=True,
    )

    print(
        fsck_output[:12000],
        flush=True,
    )

print(
    flush=True,
)


# =============================================================================
# 12. Consolidated history state
# =============================================================================

discoverable_commits = set(
    reachable_commits
)

discoverable_commits.update(
    unreachable_commits
)

if head_commit is not None:
    discoverable_commits.add(
        head_commit
    )

discoverable_commits = sorted(
    discoverable_commits
)

refs_exist = (
    len(ref_records) > 0
)

discoverable_commit_history_exists = (
    len(discoverable_commits) > 0
)


if head_commit_available:

    history_state = (
        "HEAD_COMMIT_AVAILABLE"
    )

elif discoverable_commit_history_exists:

    history_state = (
        "HEAD_UNAVAILABLE_BUT_COMMIT_OBJECT_HISTORY_RECOVERABLE"
    )

elif refs_exist:

    history_state = (
        "HEAD_UNAVAILABLE_REFS_EXIST_"
        "BUT_NO_COMMIT_HISTORY_DISCOVERED"
    )

else:

    history_state = (
        "UNBORN_OR_EMPTY_REPOSITORY_"
        "NO_DISCOVERABLE_COMMIT_HISTORY"
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "GIT HISTORY AVAILABILITY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "symbolic HEAD:",
    symbolic_head,
    flush=True,
)

print(
    "HEAD commit available:",
    head_commit_available,
    flush=True,
)

print(
    "refs:",
    len(ref_records),
    flush=True,
)

print(
    "reachable commits:",
    len(reachable_commits),
    flush=True,
)

print(
    "unreachable/dangling commits:",
    len(unreachable_commits),
    flush=True,
)

print(
    "total discoverable commits:",
    len(discoverable_commits),
    flush=True,
)

print(
    "history state:",
    history_state,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Formal resolution
# =============================================================================

implementation_authorized = False

historical_oracle_access_authorized = False


if discoverable_commit_history_exists:

    recovery_status = (
        "READ_ONLY_COMMIT_HISTORY_AVAILABLE_"
        "AUTHORITY_SEARCH_MAY_CONTINUE"
    )

    next_stage = (
        "NOTEBOOK14_14_3E_D_I_R2_"
        "MORPH_READ_ONLY_COMMIT_OBJECT_AUTHORITY_SEARCH"
    )

else:

    recovery_status = (
        "NO_DISCOVERABLE_GIT_COMMIT_HISTORY_"
        "OUTSIDE_CHECKOUT_AUTHORITY_ROUTE_CLOSED"
    )

    next_stage = (
        "NOTEBOOK14_14_3E_D_J_"
        "MORPH_PORTABILITY_BLOCKER_FINALIZATION"
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-I-R1-R1 RESOLUTION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "status:",
    recovery_status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "historical oracle access:",
    historical_oracle_access_authorized,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. No-mutation proof
# =============================================================================

_, status_after = run_git(
    [
        "status",
        "--short",
    ]
)

assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)

git_mutation_command_executed = False

historical_pickle_loaded = False

historical_oracle_opened = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False

source_mutation = False

test_mutation = False


assert all(
    (
        len(command) >= 2
        and
        command[1]
        in ALLOWED_GIT_SUBCOMMANDS
    )
    for command in git_command_log
)


# =============================================================================
# 15. Formal gates
# =============================================================================

semantic_gates = {
    "D_H_parent_exactly_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "Git_root_exactly_repo_root":
        (
            GIT_ROOT == ROOT
        ),

    "original_D_I_R1_SyntaxError_classified_as_preexecution_harness_failure":
        True,

    "HEAD_unavailability_handled_without_asserting_history_absence":
        True,

    "refs_enumerated_read_only":
        True,

    "reachable_commits_enumerated_read_only":
        True,

    "unreachable_commit_objects_checked_read_only":
        True,

    "Git_history_availability_explicitly_classified":
        True,

    "Git_mutation_command_not_executed":
        (
            git_mutation_command_executed
            is False
        ),

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "historical_oracle_remains_closed":
        (
            historical_oracle_access_authorized
            is False
            and
            historical_oracle_opened
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-I-R1-R1 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

for gate, ok in semantic_gates.items():

    print(
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate
    for gate, ok
    in semantic_gates.items()
    if not ok
]

assert (
    failed_gates == []
), (
    "14.3E-D-I-R1-R1 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 16. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_I_R1_R1",

    "title":
        "GIT_HISTORY_AVAILABILITY_HARNESS_REPAIR_CELL_SYNTAX_REPAIR",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_GIT_HISTORY_ENVIRONMENT_REPAIR_ONLY",

    "parent_14_3e_d_h": {
        "path":
            str(PARENT_REL),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "superseded_failures": [
        {
            "stage":
                "NOTEBOOK14_14_3E_D_I",

            "failure":
                "git rev-parse HEAD rc=128",

            "classification":
                "HARNESS_ENVIRONMENT_ASSUMPTION",

            "scientific_interpretation":
                "NONE",
        },
        {
            "stage":
                "NOTEBOOK14_14_3E_D_I_R1",

            "failure":
                "SyntaxError before Python execution",

            "classification":
                "NOTEBOOK_CELL_SYNTAX_OR_PASTE_FAILURE",

            "runtime_executed":
                False,

            "scientific_interpretation":
                "NONE",
        },
    ],

    "failed_R1_artifact_exists":
        FAILED_R1.exists(),

    "git_environment": {
        "root":
            str(GIT_ROOT),

        "symbolic_HEAD":
            symbolic_head,

        "HEAD_verify_rc":
            head_rc,

        "HEAD_verify_output":
            head_output,

        "HEAD_commit":
            head_commit,

        "HEAD_commit_available":
            head_commit_available,

        "refs":
            ref_records,

        "reachable_commits":
            reachable_commits,

        "unreachable_or_dangling_commits":
            unreachable_commits,

        "discoverable_commits":
            discoverable_commits,

        "history_state":
            history_state,

        "status_before":
            status_before,

        "status_after":
            status_after,

        "fsck_rc":
            fsck_rc,

        "fsck_output_preview":
            fsck_output[:12000],

        "read_only_git_command_count":
            len(git_command_log),

        "allowed_git_subcommands":
            sorted(
                ALLOWED_GIT_SUBCOMMANDS
            ),
    },

    "resolution": {
        "discoverable_commit_history_exists":
            discoverable_commit_history_exists,

        "status":
            recovery_status,

        "implementation_authorized":
            False,

        "historical_oracle_access_authorized":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "current_checkout_archaeology":
            "CLOSED",

        "Git_history_availability":
            history_state,

        "outside_checkout_authority_route":
            recovery_status,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED",

        "historical_structural_oracle":
            "CLOSED",

        "historical_pickle_content_access":
            False,

        "Git_mutation":
            False,

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "The original D-I HEAD failure does not prove Git history is absent.",
        "The D-I-R1 SyntaxError occurred before runtime and has no scientific meaning.",
        "No Git mutation command was executed.",
        "No historical pickle was opened.",
        "No historical structural oracle was opened.",
        "No Morph implementation was performed.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)

artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-I-R1-R1 artifact exists "
        "but differs from exact repaired state."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )

else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    OUT.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 17. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-I-R1-R1 PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "GIT STATE:",
    flush=True,
)

print(
    "  symbolic HEAD:",
    symbolic_head,
    flush=True,
)

print(
    "  HEAD commit available:",
    head_commit_available,
    flush=True,
)

print(
    "  refs:",
    len(ref_records),
    flush=True,
)

print(
    "  reachable commits:",
    len(reachable_commits),
    flush=True,
)

print(
    "  unreachable/dangling commits:",
    len(unreachable_commits),
    flush=True,
)

print(
    "  total discoverable commits:",
    len(discoverable_commits),
    flush=True,
)

print(
    "  history state:",
    history_state,
    flush=True,
)

print(
    flush=True,
)

print(
    "RESOLUTION:",
    flush=True,
)

print(
    "  status:",
    recovery_status,
    flush=True,
)

print(
    "  implementation authorized:",
    False,
    flush=True,
)

print(
    "  historical oracle access:",
    False,
    flush=True,
)

print(
    "  NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)

print(
    "SAFETY:",
    flush=True,
)

print(
    "  Git mutation:",
    False,
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  historical oracle opened:",
    False,
    flush=True,
)

print(
    "  GT loaded:",
    False,
    flush=True,
)

print(
    "  evaluator executed:",
    False,
    flush=True,
)

print(
    "  source mutation:",
    False,
    flush=True,
)

print(
    flush=True,
)

print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-I-R2
# MORPH ORPHANED GIT TREE READ-ONLY SOURCE AUTHORITY RECOVERY
#
# PARENT
# ------
# 14.3E-D-I-R1-R1 PASS
#
# Parent proved:
#
#   symbolic HEAD = refs/heads/main
#   HEAD commit    = unavailable
#   refs           = 0
#   reachable commits = 0
#   dangling/unreachable COMMIT objects = 0
#
# BUT git fsck also showed unreachable TREE/BLOB objects.
#
#
# PURPOSE
# -------
# Check whether those orphaned TREE objects still preserve:
#
#   path -> source blob SHA -> exact .py/.ipynb source
#
# containing missing Morph generator semantics.
#
#
# IMPORTANT
# ---------
# This does NOT restore commit provenance.
#
# Even if source is found, classification is:
#
#   ORPHANED_TREE_SOURCE_EVIDENCE
#
# and a separate exact-contract stage is required before any implementation.
#
#
# SOURCE ONLY
# -----------
# Only paths ending in:
#
#   .py
#   .ipynb
#
# may be opened from Git objects.
#
# Standalone orphan blobs with no source path are NOT opened because they may
# be arbitrary/binary artifacts.
#
#
# NO:
#   - checkout
#   - reset
#   - restore
#   - stash
#   - clean
#   - commit
#   - branch/tag mutation
#   - source mutation
#   - historical pickle load
#   - historical structural oracle
#   - GT
#   - evaluator
#   - performance
# =============================================================================

import ast
import hashlib
import json
import re
import subprocess

from pathlib import Path


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_i_r1_r1_"
    "git_history_availability_cell_syntax_repair_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_i_r2_"
    "morph_orphaned_git_tree_"
    "read_only_source_authority_recovery_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "4cd122867f2abece5a077a7f7170943776698ae3fbebfb5dd94d166ad091e2f3"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "2d22c2e6ace7e1126802a7cacd78937e71595aca8f92b10eb994ce86bfc2a133"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_PARENT_HISTORY_STATE = (
    "UNBORN_OR_EMPTY_REPOSITORY_"
    "NO_DISCOVERABLE_COMMIT_HISTORY"
)


EXPECTED_PARENT_STATUS = (
    "NO_DISCOVERABLE_GIT_COMMIT_HISTORY_"
    "OUTSIDE_CHECKOUT_AUTHORITY_ROUTE_CLOSED"
)


CAPABILITIES = {
    "p_T4_producer",
    "morphology_normalization",
    "morph_rank_materialization",
    "candidate_rank_materialization",
    "candidate_prob_materialization",
}


SEARCH_TERMS = (
    "p_T4_ordered_",
    "morph_rank",
    "candidate_rank",
    "candidate_prob",
    "rank_in_dataset",
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(value):
    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(value):
    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(path):
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value):
    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(path):
    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def dotted_name(node):

    if isinstance(
        node,
        ast.Name,
    ):
        return node.id

    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )

        if left:
            return (
                f"{left}.{node.attr}"
            )

        return node.attr

    return None


def calls_in_node(node):

    return sorted({
        dotted_name(
            call.func
        )
        or
        "<dynamic-call>"

        for call in ast.walk(
            node
        )

        if isinstance(
            call,
            ast.Call,
        )
    })


def has_read_pickle(node):

    for call in ast.walk(
        node
    ):

        if not isinstance(
            call,
            ast.Call,
        ):
            continue

        name = (
            dotted_name(
                call.func
            )
            or ""
        )

        if name.endswith(
            "read_pickle"
        ):
            return True

    return False


def gt_hazard(node):

    gt_names = {
        "gt",
        "gt_graph",
        "gt_nodes",
        "gt_edges",
        "gt_base",
        "gt_policy",
        "ground_truth",
    }

    names = {
        child.id.lower()

        for child in ast.walk(
            node
        )

        if isinstance(
            child,
            ast.Name,
        )
    }

    if (
        names
        &
        gt_names
    ):
        return True

    for name in calls_in_node(
        node
    ):

        lower = name.lower()

        if any(
            token
            in lower

            for token in (
                "evaluate_graph",
                "_evaluate_matched_graph",
                "official_evaluate",
                "per_sample_metrics",
            )
        ):
            return True

    return False


# =============================================================================
# 3. Parent verifier
# =============================================================================

def verify_parent(path):

    assert path.exists(), path

    assert (
        sha256_file(
            path
        )
        == EXPECTED_PARENT_FILE_SHA
    )

    obj = load_json(
        path
    )

    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )

    assert (
        obj.get(
            "content_sha256"
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )

    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )

    assert (
        canonical_json_sha(
            payload
        )
        == EXPECTED_PARENT_CONTENT_SHA
    )

    assert (
        obj[
            "git_environment"
        ][
            "history_state"
        ]
        == EXPECTED_PARENT_HISTORY_STATE
    )

    assert (
        obj[
            "resolution"
        ][
            "status"
        ]
        == EXPECTED_PARENT_STATUS
    )

    assert (
        obj[
            "resolution"
        ][
            "discoverable_commit_history_exists"
        ]
        is False
    )

    assert (
        obj[
            "resolution"
        ][
            "implementation_authorized"
        ]
        is False
    )

    return obj


# =============================================================================
# 4. Strict read-only Git wrapper
# =============================================================================

ALLOWED_GIT_SUBCOMMANDS = {
    "rev-parse",
    "status",
    "fsck",
    "ls-tree",
    "show",
}


git_command_log = []


def run_git(
    args,
    check=True,
):

    assert args

    assert (
        args[0]
        in ALLOWED_GIT_SUBCOMMANDS
    ), (
        f"Unauthorized Git subcommand: {args[0]}"
    )

    cmd = [
        "git",
        *args,
    ]

    git_command_log.append(
        cmd
    )

    process = subprocess.Popen(
        cmd,
        cwd=ROOT,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )

    assert (
        process.stdout
        is not None
    )

    lines = []

    for line in process.stdout:
        lines.append(
            line
        )

    rc = process.wait()

    output = "".join(
        lines
    )

    if (
        check
        and
        rc != 0
    ):
        raise RuntimeError(
            "Read-only Git command failed.\n"
            f"CMD: {' '.join(cmd)}\n"
            f"RC: {rc}\n"
            f"OUTPUT:\n{output}"
        )

    return (
        rc,
        output,
    )


# =============================================================================
# 5. AST production helpers
# =============================================================================

def target_text(
    source,
    target,
):

    segment = ast.get_source_segment(
        source,
        target,
    )

    if segment is not None:
        return segment

    return ast.unparse(
        target
    )


def assignment_targets(node):

    if isinstance(
        node,
        ast.Assign,
    ):
        return list(
            node.targets
        )

    if isinstance(
        node,
        (
            ast.AnnAssign,
            ast.AugAssign,
        ),
    ):
        return [
            node.target
        ]

    return []


def static_subscript_field(node):

    if not isinstance(
        node,
        ast.Subscript,
    ):
        return (
            None,
            None,
        )

    try:
        key = ast.literal_eval(
            node.slice
        )

    except Exception:
        return (
            None,
            None,
        )

    if not isinstance(
        key,
        str,
    ):
        return (
            None,
            None,
        )

    return (
        ast.unparse(
            node.value
        ),
        key,
    )


def uses_existing_field(
    expression,
    field,
):

    return any(
        token
        in expression

        for token in (
            f".{field}",
            f'["{field}"]',
            f"['{field}']",
        )
    )


# =============================================================================
# 6. Source semantic profiler
# =============================================================================

def profile_source(
    source,
):

    try:
        tree = ast.parse(
            source
        )

    except Exception as exc:

        return {
            "parseable":
                False,

            "parse_error":
                repr(
                    exc
                ),

            "capabilities":
                [],
        }


    p_t4_records = []

    normalization_records = []

    field_records = {
        "morph_rank":
            [],

        "candidate_rank":
            [],

        "candidate_prob":
            [],
    }


    # -------------------------------------------------------------------------
    # Assignments.
    # -------------------------------------------------------------------------

    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
            ),
        ):
            continue

        rhs = ast.unparse(
            node.value
        )

        source_segment = (
            ast.get_source_segment(
                source,
                node,
            )
            or
            ast.unparse(
                node
            )
        )

        for target in assignment_targets(
            node
        ):

            target_expr = target_text(
                source,
                target,
            )

            # -------------------------------------------------------------
            # p_T4 producer.
            # -------------------------------------------------------------

            if (
                "p_T4_ordered_"
                in target_expr
            ):

                p_t4_records.append({
                    "line":
                        node.lineno,

                    "target":
                        target_expr,

                    "value":
                        rhs,

                    "source_sha256":
                        sha256_text(
                            source_segment
                        ),

                    "source":
                        source_segment,
                })


            # -------------------------------------------------------------
            # Static DataFrame column materialization.
            # -------------------------------------------------------------

            object_expr, field = (
                static_subscript_field(
                    target
                )
            )

            if field in field_records:

                field_records[
                    field
                ].append({
                    "kind":
                        "SUBSCRIPT_ASSIGNMENT",

                    "line":
                        node.lineno,

                    "object":
                        object_expr,

                    "value":
                        rhs,

                    "inherited_same_field":
                        uses_existing_field(
                            rhs,
                            field,
                        ),

                    "source_sha256":
                        sha256_text(
                            source_segment
                        ),

                    "source":
                        source_segment,
                })


            # -------------------------------------------------------------
            # Plain variable materialization.
            # -------------------------------------------------------------

            if isinstance(
                target,
                ast.Name,
            ):

                name = target.id

                if name in field_records:

                    field_records[
                        name
                    ].append({
                        "kind":
                            "PLAIN_ASSIGNMENT",

                        "line":
                            node.lineno,

                        "object":
                            None,

                        "value":
                            rhs,

                        "inherited_same_field":
                            uses_existing_field(
                                rhs,
                                name,
                            ),

                        "source_sha256":
                            sha256_text(
                                source_segment
                            ),

                        "source":
                            source_segment,
                    })


    # -------------------------------------------------------------------------
    # Dict/action payloads.
    # -------------------------------------------------------------------------

    for dictionary in ast.walk(
        tree
    ):

        if not isinstance(
            dictionary,
            ast.Dict,
        ):
            continue

        mapping = {}

        for key_node, value_node in zip(
            dictionary.keys,
            dictionary.values,
        ):

            if key_node is None:
                continue

            try:
                key = ast.literal_eval(
                    key_node
                )

            except Exception:
                continue

            if isinstance(
                key,
                str,
            ):
                mapping[
                    key
                ] = value_node

        for field in field_records:

            if field not in mapping:
                continue

            value_expr = ast.unparse(
                mapping[
                    field
                ]
            )

            segment = (
                ast.get_source_segment(
                    source,
                    dictionary,
                )
                or
                ast.unparse(
                    dictionary
                )
            )

            field_records[
                field
            ].append({
                "kind":
                    "DICT_VALUE",

                "line":
                    dictionary.lineno,

                "object":
                    None,

                "value":
                    value_expr,

                "inherited_same_field":
                    uses_existing_field(
                        value_expr,
                        field,
                    ),

                "source_sha256":
                    sha256_text(
                        segment
                    ),

                "source":
                    segment,
            })


    # -------------------------------------------------------------------------
    # rename(columns={...})
    # -------------------------------------------------------------------------

    for call in ast.walk(
        tree
    ):

        if not isinstance(
            call,
            ast.Call,
        ):
            continue

        name = (
            dotted_name(
                call.func
            )
            or ""
        )

        if not name.endswith(
            ".rename"
        ):
            continue

        for keyword in call.keywords:

            if (
                keyword.arg
                != "columns"
                or
                not isinstance(
                    keyword.value,
                    ast.Dict,
                )
            ):
                continue

            mapping = {}

            for key_node, value_node in zip(
                keyword.value.keys,
                keyword.value.values,
            ):

                if key_node is None:
                    continue

                try:
                    key = ast.literal_eval(
                        key_node
                    )

                    value = ast.literal_eval(
                        value_node
                    )

                except Exception:
                    continue

                if (
                    isinstance(
                        key,
                        str,
                    )
                    and
                    isinstance(
                        value,
                        str,
                    )
                ):
                    mapping[
                        key
                    ] = value

            for source_field, destination_field in mapping.items():

                if destination_field not in field_records:
                    continue

                segment = (
                    ast.get_source_segment(
                        source,
                        call,
                    )
                    or
                    ast.unparse(
                        call
                    )
                )

                field_records[
                    destination_field
                ].append({
                    "kind":
                        "RENAME",

                    "line":
                        call.lineno,

                    "object":
                        None,

                    "value":
                        source_field,

                    "rename_source_field":
                        source_field,

                    "inherited_same_field":
                        (
                            source_field
                            ==
                            destination_field
                        ),

                    "source_sha256":
                        sha256_text(
                            segment
                        ),

                    "source":
                        segment,
                })


    # -------------------------------------------------------------------------
    # Percentile normalization.
    # -------------------------------------------------------------------------

    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.For,
                ast.Expr,
            ),
        ):
            continue

        segment = (
            ast.get_source_segment(
                source,
                node,
            )
            or
            ""
        )

        if (
            ".rank("
            not in segment
            or
            "pct=True"
            not in segment
        ):
            continue

        normalization_records.append({
            "line":
                node.lineno,

            "uses_fold":
                (
                    'groupby("fold")'
                    in segment
                    or
                    "groupby('fold')"
                    in segment
                ),

            "groupby_fragments":
                re.findall(
                    r"groupby\(([^)]*)\)",
                    segment,
                ),

            "source_sha256":
                sha256_text(
                    segment
                ),

            "source":
                segment,
        })


    strong_fields = {}

    for field, records in field_records.items():

        strong_fields[
            field
        ] = [
            record

            for record in records

            if not record[
                "inherited_same_field"
            ]
        ]


    capabilities = []


    if p_t4_records:

        capabilities.append(
            "p_T4_producer"
        )


    if normalization_records:

        capabilities.append(
            "morphology_normalization"
        )


    if strong_fields[
        "morph_rank"
    ]:

        capabilities.append(
            "morph_rank_materialization"
        )


    if strong_fields[
        "candidate_rank"
    ]:

        capabilities.append(
            "candidate_rank_materialization"
        )


    if strong_fields[
        "candidate_prob"
    ]:

        capabilities.append(
            "candidate_prob_materialization"
        )


    return {
        "parseable":
            True,

        "has_read_pickle":
            has_read_pickle(
                tree
            ),

        "GT_hazard":
            gt_hazard(
                tree
            ),

        "clean_unit":
            (
                not has_read_pickle(
                    tree
                )
                and
                not gt_hazard(
                    tree
                )
            ),

        "capabilities":
            sorted(
                capabilities
            ),

        "p_T4_records":
            p_t4_records,

        "normalization_records":
            normalization_records,

        "field_records":
            field_records,

        "strong_field_records":
            strong_fields,
    }


# =============================================================================
# 7. Source unit extraction
# =============================================================================

def source_units_from_python(
    tree_sha,
    path,
    blob_sha,
    source,
):

    records = []

    try:
        tree = ast.parse(
            source,
            filename=path,
        )

    except Exception:
        return records


    module_profile = profile_source(
        source
    )


    if module_profile[
        "capabilities"
    ]:

        records.append({
            "tree_sha":
                tree_sha,

            "path":
                path,

            "blob_sha":
                blob_sha,

            "unit_kind":
                "PYTHON_MODULE",

            "unit_name":
                "<module>",

            "unit_sha256":
                sha256_text(
                    source
                ),

            "profile":
                module_profile,

            "source_preview":
                source[
                    :10000
                ],
        })


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        ):
            continue

        segment = ast.get_source_segment(
            source,
            node,
        )

        if segment is None:
            continue

        profile = profile_source(
            segment
        )

        if not profile[
            "capabilities"
        ]:
            continue

        records.append({
            "tree_sha":
                tree_sha,

            "path":
                path,

            "blob_sha":
                blob_sha,

            "unit_kind":
                "PYTHON_FUNCTION",

            "unit_name":
                node.name,

            "line_start":
                node.lineno,

            "line_end":
                node.end_lineno,

            "unit_sha256":
                sha256_text(
                    segment
                ),

            "profile":
                profile,

            "source_preview":
                segment[
                    :10000
                ],
        })


    return records


def source_units_from_notebook(
    tree_sha,
    path,
    blob_sha,
    source,
):

    records = []

    try:
        notebook = json.loads(
            source
        )

    except Exception:
        return records


    for cell_index, cell in enumerate(
        notebook.get(
            "cells",
            []
        )
    ):

        if (
            cell.get(
                "cell_type"
            )
            != "code"
        ):
            continue

        cell_source = "".join(
            cell.get(
                "source",
                []
            )
        )

        if not any(
            term
            in cell_source

            for term in SEARCH_TERMS
        ):
            continue

        profile = profile_source(
            cell_source
        )

        if not profile[
            "capabilities"
        ]:
            continue

        records.append({
            "tree_sha":
                tree_sha,

            "path":
                path,

            "blob_sha":
                blob_sha,

            "unit_kind":
                "NOTEBOOK_CELL",

            "unit_name":
                f"cell_{cell_index}",

            "cell_index":
                cell_index,

            "unit_sha256":
                sha256_text(
                    cell_source
                ),

            "profile":
                profile,

            "source_preview":
                cell_source[
                    :12000
                ],
        })


    return records


# =============================================================================
# 8. Authority-path classifier
# =============================================================================

def is_provenance_meta_path(path):

    lower = path.lower()


    if lower.startswith(
        "notebooks/13"
    ):
        return True


    if lower.startswith(
        "notebooks/14"
    ):
        return True


    if "provenance" in lower:
        return True


    return False


def authority_classification(
    record,
):

    if is_provenance_meta_path(
        record[
            "path"
        ]
    ):
        return (
            "PROVENANCE_META_CODE_EXCLUDED"
        )


    profile = record[
        "profile"
    ]


    if profile[
        "GT_hazard"
    ]:

        return (
            "SOURCE_SEMANTICS_MIXED_WITH_GT"
        )


    if profile[
        "has_read_pickle"
    ]:

        return (
            "SOURCE_SEMANTICS_HISTORICAL_PICKLE_DEPENDENT"
        )


    if len(
        profile[
            "capabilities"
        ]
    ) >= 3:

        return (
            "STRONG_CLEAN_ORPHANED_TREE_SOURCE_CANDIDATE"
        )


    return (
        "PARTIAL_CLEAN_ORPHANED_TREE_SOURCE_CANDIDATE"
    )


# =============================================================================
# 9. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-I-R2 "
    "MORPH ORPHANED GIT TREE READ-ONLY SOURCE AUTHORITY RECOVERY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Parent / current source lock
# =============================================================================

parent = verify_parent(
    PARENT
)


portable_sha_before = sha256_file(
    PORTABLE
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


print(
    "[PASS] D_I_R1_R1_PARENT_AND_PORTABLE_LOCK",
    flush=True,
)

print(
    "  parent file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  parent content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 11. Git root / status
# =============================================================================

root_rc, root_output = run_git(
    [
        "rev-parse",
        "--show-toplevel",
    ]
)


assert (
    root_rc == 0
)


GIT_ROOT = Path(
    root_output.strip()
).resolve()


assert (
    GIT_ROOT == ROOT
)


_, status_before = run_git(
    [
        "status",
        "--short",
    ]
)


# =============================================================================
# 12. Recover orphaned tree objects
# =============================================================================

fsck_rc, fsck_output = run_git(
    [
        "fsck",
        "--full",
        "--no-reflogs",
        "--unreachable",
        "--no-progress",
    ],
    check=False,
)


tree_pattern = re.compile(
    r"(?:unreachable|dangling)\s+tree\s+"
    r"([0-9a-fA-F]{40,64})"
)


blob_pattern = re.compile(
    r"(?:unreachable|dangling)\s+blob\s+"
    r"([0-9a-fA-F]{40,64})"
)


orphan_trees = sorted(
    set(
        tree_pattern.findall(
            fsck_output
        )
    )
)


standalone_orphan_blobs = sorted(
    set(
        blob_pattern.findall(
            fsck_output
        )
    )
)


print(
    "=" * 120,
    flush=True,
)

print(
    "ORPHANED GIT OBJECT INVENTORY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "orphan trees:",
    len(
        orphan_trees
    ),
    flush=True,
)

print(
    "standalone orphan blobs:",
    len(
        standalone_orphan_blobs
    ),
    flush=True,
)

print(
    "standalone blobs opened:",
    False,
    flush=True,
)

print(
    flush=True,
)


for tree_sha in orphan_trees:

    print(
        " tree:",
        tree_sha,
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 13. Enumerate source blobs under each orphaned tree
# =============================================================================

tree_records = []

all_source_objects = []


for tree_sha in orphan_trees:

    rc, output = run_git(
        [
            "ls-tree",
            "-r",
            tree_sha,
        ],
        check=False,
    )


    if rc != 0:

        tree_records.append({
            "tree_sha":
                tree_sha,

            "ls_tree_rc":
                rc,

            "source_paths":
                [],
        })

        continue


    source_paths = []


    for line in output.splitlines():

        #
        # Expected:
        #
        # 100644 blob <sha>\t<path>
        #

        match = re.match(
            r"^(\d+)\s+(\w+)\s+([0-9a-fA-F]{40,64})\t(.+)$",
            line,
        )


        if match is None:
            continue


        mode = match.group(
            1
        )

        object_type = match.group(
            2
        )

        object_sha = match.group(
            3
        )

        path = match.group(
            4
        )


        if object_type != "blob":
            continue


        if not (
            path.endswith(
                ".py"
            )
            or
            path.endswith(
                ".ipynb"
            )
        ):
            continue


        source_paths.append({
            "mode":
                mode,

            "path":
                path,

            "blob_sha":
                object_sha,
        })


        all_source_objects.append({
            "tree_sha":
                tree_sha,

            "path":
                path,

            "blob_sha":
                object_sha,
        })


    tree_records.append({
        "tree_sha":
            tree_sha,

        "ls_tree_rc":
            rc,

        "source_paths":
            source_paths,
    })


print(
    "=" * 120,
    flush=True,
)

print(
    "ORPHANED TREE SOURCE PATHS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in tree_records:

    print(
        record[
            "tree_sha"
        ],
        "source paths:",
        len(
            record[
                "source_paths"
            ]
        ),
        flush=True,
    )


    for item in record[
        "source_paths"
    ][
        :100
    ]:

        print(
            " ",
            item[
                "blob_sha"
            ],
            item[
                "path"
            ],
            flush=True,
        )


    print(
        flush=True,
    )


# =============================================================================
# 14. Deduplicate tree/path/blob source objects
# =============================================================================

source_object_map = {}


for record in all_source_objects:

    key = (
        record[
            "path"
        ],
        record[
            "blob_sha"
        ],
    )


    if key not in source_object_map:

        source_object_map[
            key
        ] = {
            "path":
                record[
                    "path"
                ],

            "blob_sha":
                record[
                    "blob_sha"
                ],

            "tree_shas":
                [
                    record[
                        "tree_sha"
                    ]
                ],
        }


    else:

        source_object_map[
            key
        ][
            "tree_shas"
        ].append(
            record[
                "tree_sha"
            ]
        )


source_objects = list(
    source_object_map.values()
)


for record in source_objects:

    record[
        "tree_shas"
    ] = sorted(
        set(
            record[
                "tree_shas"
            ]
        )
    )


print(
    "unique path/blob source objects:",
    len(
        source_objects
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 15. Open SOURCE blobs only and recover semantic units
# =============================================================================

source_units = []


for index, record in enumerate(
    source_objects,
    start=1,
):

    path = record[
        "path"
    ]

    blob_sha = record[
        "blob_sha"
    ]


    # Hard source-extension gate.
    assert (
        path.endswith(
            ".py"
        )
        or
        path.endswith(
            ".ipynb"
        )
    )


    rc, source = run_git(
        [
            "show",
            blob_sha,
        ],
        check=False,
    )


    if rc != 0:
        continue


    # Fast prefilter.
    if not any(
        term
        in source

        for term in SEARCH_TERMS
    ):
        continue


    for tree_sha in record[
        "tree_shas"
    ]:

        if path.endswith(
            ".py"
        ):

            units = source_units_from_python(
                tree_sha,
                path,
                blob_sha,
                source,
            )


        else:

            units = source_units_from_notebook(
                tree_sha,
                path,
                blob_sha,
                source,
            )


        source_units.extend(
            units
        )


print(
    "relevant semantic source units:",
    len(
        source_units
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 16. Deduplicate semantic units
# =============================================================================

unique_unit_map = {}


for unit in source_units:

    key = (
        unit[
            "path"
        ],
        unit[
            "unit_kind"
        ],
        unit[
            "unit_name"
        ],
        unit[
            "unit_sha256"
        ],
    )


    if key not in unique_unit_map:

        unique_unit_map[
            key
        ] = {
            **unit,

            "tree_shas":
                [
                    unit[
                        "tree_sha"
                    ]
                ],
        }


    else:

        unique_unit_map[
            key
        ][
            "tree_shas"
        ].append(
            unit[
                "tree_sha"
            ]
        )


unique_units = list(
    unique_unit_map.values()
)


for unit in unique_units:

    unit[
        "tree_shas"
    ] = sorted(
        set(
            unit[
                "tree_shas"
            ]
        )
    )


    unit[
        "authority_classification"
    ] = authority_classification(
        unit
    )


# =============================================================================
# 17. Score candidates
# =============================================================================

def authority_score(
    unit,
):

    profile = unit[
        "profile"
    ]


    score = (
        25
        *
        len(
            profile[
                "capabilities"
            ]
        )
    )


    if profile[
        "clean_unit"
    ]:
        score += 20


    if profile[
        "has_read_pickle"
    ]:
        score -= 30


    if profile[
        "GT_hazard"
    ]:
        score -= 30


    if (
        unit[
            "authority_classification"
        ]
        ==
        "PROVENANCE_META_CODE_EXCLUDED"
    ):
        score -= 100


    if (
        "notebooks/11"
        in unit[
            "path"
        ].lower()
    ):
        score += 20


    if (
        "stage11"
        in unit[
            "path"
        ].lower()
    ):
        score += 10


    return score


for unit in unique_units:

    unit[
        "authority_score"
    ] = authority_score(
        unit
    )


unique_units = sorted(
    unique_units,
    key=lambda unit: (
        -unit[
            "authority_score"
        ],
        unit[
            "path"
        ],
        unit[
            "unit_name"
        ],
    ),
)


# =============================================================================
# 18. Tree-level semantic coverage
# =============================================================================

tree_semantic_records = []


for tree_sha in orphan_trees:

    relevant_units = [
        unit

        for unit in unique_units

        if (
            tree_sha
            in unit[
                "tree_shas"
            ]
            and
            unit[
                "authority_classification"
            ]
            != "PROVENANCE_META_CODE_EXCLUDED"
        )
    ]


    all_capabilities = sorted({
        capability

        for unit in relevant_units

        for capability in unit[
            "profile"
        ][
            "capabilities"
        ]
    })


    clean_capabilities = sorted({
        capability

        for unit in relevant_units

        if unit[
            "profile"
        ][
            "clean_unit"
        ]

        for capability in unit[
            "profile"
        ][
            "capabilities"
        ]
    })


    tree_semantic_records.append({
        "tree_sha":
            tree_sha,

        "source_unit_count":
            len(
                relevant_units
            ),

        "capabilities":
            all_capabilities,

        "clean_capabilities":
            clean_capabilities,

        "complete_semantics":
            (
                set(
                    all_capabilities
                )
                >=
                CAPABILITIES
            ),

        "clean_complete_semantics":
            (
                set(
                    clean_capabilities
                )
                >=
                CAPABILITIES
            ),
    })


clean_complete_trees = [
    record

    for record in tree_semantic_records

    if record[
        "clean_complete_semantics"
    ]
]


complete_trees = [
    record

    for record in tree_semantic_records

    if record[
        "complete_semantics"
    ]
]


four_of_five_trees = [
    record

    for record in tree_semantic_records

    if (
        len(
            set(
                record[
                    "capabilities"
                ]
            )
            &
            CAPABILITIES
        )
        >= 4
    )
]


# =============================================================================
# 19. Print strongest candidates
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "ORPHANED TREE MORPH SOURCE AUTHORITY CANDIDATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


actionable_units = [
    unit

    for unit in unique_units

    if (
        unit[
            "authority_classification"
        ]
        != "PROVENANCE_META_CODE_EXCLUDED"
    )
]


for index, unit in enumerate(
    actionable_units[
        :25
    ],
    start=1,
):

    print(
        f"[{index}] "
        f"SCORE={unit['authority_score']} "
        f"{unit['authority_classification']}",
        flush=True,
    )

    print(
        "  path:",
        unit[
            "path"
        ],
        flush=True,
    )

    print(
        "  blob SHA:",
        unit[
            "blob_sha"
        ],
        flush=True,
    )

    print(
        "  tree SHAs:",
        unit[
            "tree_shas"
        ],
        flush=True,
    )

    print(
        "  unit:",
        unit[
            "unit_name"
        ],
        flush=True,
    )

    print(
        "  unit SHA:",
        unit[
            "unit_sha256"
        ],
        flush=True,
    )

    print(
        "  capabilities:",
        unit[
            "profile"
        ][
            "capabilities"
        ],
        flush=True,
    )

    print(
        "  read_pickle:",
        unit[
            "profile"
        ][
            "has_read_pickle"
        ],
        flush=True,
    )

    print(
        "  GT hazard:",
        unit[
            "profile"
        ][
            "GT_hazard"
        ],
        flush=True,
    )

    print(
        "  --- source preview ---",
        flush=True,
    )

    print(
        unit[
            "source_preview"
        ],
        flush=True,
    )

    print(
        "  --- end preview ---",
        flush=True,
    )

    print(
        flush=True,
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "ORPHANED TREE SEMANTIC COVERAGE",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in tree_semantic_records:

    print(
        record[
            "tree_sha"
        ],
        flush=True,
    )

    print(
        "  source units:",
        record[
            "source_unit_count"
        ],
        flush=True,
    )

    print(
        "  capabilities:",
        record[
            "capabilities"
        ],
        flush=True,
    )

    print(
        "  clean capabilities:",
        record[
            "clean_capabilities"
        ],
        flush=True,
    )

    print(
        "  complete:",
        record[
            "complete_semantics"
        ],
        flush=True,
    )

    print(
        "  clean complete:",
        record[
            "clean_complete_semantics"
        ],
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 20. Formal recovery decision
# =============================================================================

implementation_authorized = False

historical_oracle_access_authorized = False


if clean_complete_trees:

    recovery_status = (
        "ORPHANED_TREE_COMPLETE_CLEAN_SOURCE_EVIDENCE_RECOVERED_"
        "EXACT_CONTRACT_EXTRACTION_REQUIRED"
    )

    next_stage = (
        "NOTEBOOK14_14_3E_D_J_"
        "MORPH_ORPHANED_TREE_EXACT_CONTRACT_EXTRACTION"
    )


elif complete_trees:

    recovery_status = (
        "ORPHANED_TREE_COMPLETE_SEMANTICS_RECOVERED_"
        "DEPENDENCY_DISENTANGLEMENT_REQUIRED"
    )

    next_stage = (
        "NOTEBOOK14_14_3E_D_J_"
        "MORPH_ORPHANED_TREE_DEPENDENCY_DISENTANGLEMENT"
    )


elif four_of_five_trees:

    recovery_status = (
        "ORPHANED_TREE_HIGH_COVERAGE_PARTIAL_SOURCE_EVIDENCE_"
        "FINAL_PRIMITIVE_GAP_CLOSURE_REQUIRED"
    )

    next_stage = (
        "NOTEBOOK14_14_3E_D_J_"
        "MORPH_ORPHANED_TREE_FINAL_PRIMITIVE_GAP_CLOSURE"
    )


else:

    recovery_status = (
        "ORPHANED_GIT_TREE_OBJECTS_DO_NOT_RECOVER_"
        "COMPLETE_MORPH_SOURCE_AUTHORITY"
    )

    next_stage = (
        "NOTEBOOK14_14_3E_D_J_"
        "MORPH_PORTABILITY_BLOCKER_FINALIZATION"
    )


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-I-R2 RECOVERY RESOLUTION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "clean complete orphan trees:",
    len(
        clean_complete_trees
    ),
    flush=True,
)

print(
    "complete orphan trees:",
    len(
        complete_trees
    ),
    flush=True,
)

print(
    "four-of-five orphan trees:",
    len(
        four_of_five_trees
    ),
    flush=True,
)

print(
    "status:",
    recovery_status,
    flush=True,
)

print(
    "implementation authorized:",
    implementation_authorized,
    flush=True,
)

print(
    "historical oracle access:",
    historical_oracle_access_authorized,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 21. Safety / no mutation
# =============================================================================

_, status_after = run_git(
    [
        "status",
        "--short",
    ]
)


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


git_mutation = False

standalone_orphan_blob_opened = False

historical_pickle_loaded = False

historical_oracle_opened = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False

source_mutation = False

test_mutation = False


assert (
    standalone_orphan_blob_opened
    is False
)


assert all(
    (
        len(
            command
        )
        >= 2
        and
        command[
            1
        ]
        in ALLOWED_GIT_SUBCOMMANDS
    )

    for command in git_command_log
)


# =============================================================================
# 22. Formal semantic gates
# =============================================================================

semantic_gates = {
    "parent_14_3e_d_i_r1_r1_locked":
        True,

    "portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "orphaned_tree_objects_enumerated_read_only":
        True,

    "only_tree_path_identified_py_ipynb_blobs_opened":
        True,

    "standalone_orphan_blobs_not_opened":
        (
            standalone_orphan_blob_opened
            is False
        ),

    "provenance_meta_paths_excluded_from_source_authority":
        True,

    "inherited_candidate_fields_not_treated_as_materialization":
        True,

    "tree_level_semantic_coverage_computed":
        True,

    "absence_of_commit_provenance_preserved":
        True,

    "implementation_not_prematurely_authorized":
        (
            implementation_authorized
            is False
        ),

    "historical_oracle_remains_closed":
        (
            historical_oracle_access_authorized
            is False
            and
            historical_oracle_opened
            is False
        ),

    "Git_mutation_not_executed":
        (
            git_mutation
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-I-R2 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-I-R2 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 23. Artifact-safe unit compaction
# =============================================================================

def compact_unit(unit):

    result = {
        key:
            value

        for key, value
        in unit.items()

        if key
        != "source_preview"
    }

    result[
        "source_preview"
    ] = unit[
        "source_preview"
    ][
        :6000
    ]

    return result


# =============================================================================
# 24. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_I_R2",

    "title":
        "MORPH_ORPHANED_GIT_TREE_READ_ONLY_SOURCE_AUTHORITY_RECOVERY",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        "ENGINEERING_ORPHANED_GIT_TREE_SOURCE_RECOVERY_ONLY",

    "parent_14_3e_d_i_r1_r1": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_lock": {
        "portable_path":
            str(
                PORTABLE_REL
            ),

        "portable_sha256":
            portable_sha_before,
    },

    "git_object_inventory": {
        "fsck_rc":
            fsck_rc,

        "orphaned_tree_objects":
            orphan_trees,

        "standalone_orphan_blob_objects":
            standalone_orphan_blobs,

        "standalone_orphan_blobs_opened":
            False,

        "tree_records":
            tree_records,

        "unique_source_path_blob_count":
            len(
                source_objects
            ),
    },

    "source_units":
        [
            compact_unit(
                unit
            )

            for unit in unique_units
        ],

    "tree_semantic_coverage":
        tree_semantic_records,

    "summary": {
        "relevant_source_unit_count":
            len(
                unique_units
            ),

        "clean_complete_trees":
            clean_complete_trees,

        "complete_trees":
            complete_trees,

        "four_of_five_trees":
            four_of_five_trees,
    },

    "resolution": {
        "status":
            recovery_status,

        "implementation_authorized":
            False,

        "historical_oracle_access_authorized":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "commit_history":
            "ABSENT",

        "orphaned_tree_source_evidence":
            recovery_status,

        "orphaned_tree_source_is_commit_provenance":
            False,

        "concrete_Morph_provider_implementation":
            "NOT_AUTHORIZED_BY_THIS_STAGE",

        "historical_structural_oracle":
            "CLOSED",

        "standalone_orphan_blob_access":
            False,

        "historical_pickle_content_access":
            False,

        "Git_mutation":
            False,

        "source_mutation":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,
    },

    "explicit_non_claims": [
        "Orphaned tree/blob identities do not restore commit ancestry.",
        "A source match in an orphaned tree is not immediate implementation authorization.",
        "Standalone orphan blobs with no source path were not opened.",
        "No historical pickle was opened.",
        "No historical structural oracle was opened.",
        "No Git mutation was executed.",
        "No concrete MorphCandidateProvider was implemented.",
        "No structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-I-R2 artifact exists but differs "
        "from exact orphaned-tree recovery state."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    OUT.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 25. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-I-R2 PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "ORPHANED OBJECTS:",
    flush=True,
)

print(
    "  trees:",
    len(
        orphan_trees
    ),
    flush=True,
)

print(
    "  standalone blobs:",
    len(
        standalone_orphan_blobs
    ),
    flush=True,
)

print(
    "  source path/blob objects:",
    len(
        source_objects
    ),
    flush=True,
)

print(
    "  relevant Morph source units:",
    len(
        unique_units
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "SEMANTIC COVERAGE:",
    flush=True,
)

print(
    "  clean complete trees:",
    len(
        clean_complete_trees
    ),
    flush=True,
)

print(
    "  complete trees:",
    len(
        complete_trees
    ),
    flush=True,
)

print(
    "  four-of-five trees:",
    len(
        four_of_five_trees
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "RESOLUTION:",
    flush=True,
)

print(
    "  status:",
    recovery_status,
    flush=True,
)

print(
    "  implementation authorized:",
    False,
    flush=True,
)

print(
    "  historical oracle access:",
    False,
    flush=True,
)

print(
    "  NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFETY:",
    flush=True,
)

print(
    "  standalone orphan blobs opened:",
    False,
    flush=True,
)

print(
    "  historical pickle loaded:",
    False,
    flush=True,
)

print(
    "  historical oracle opened:",
    False,
    flush=True,
)

print(
    "  Git mutation:",
    False,
    flush=True,
)

print(
    "  source mutation:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-D-J
# MORPH PORTABILITY BLOCKER FINALIZATION
#
# PARENTS
# -------
# 14.3E-D-H:
#   reconstruction boundary frozen
#
# 14.3E-D-I-R2:
#   current checkout exhausted
#   commit-history route absent
#   orphaned-tree source recovery incomplete
#
#
# FINAL CONCLUSION TO FREEZE
# --------------------------
# A concrete fresh MorphCandidateProvider CANNOT be reconstructed from the
# currently available provenance without inventing missing semantics.
#
# Missing fresh-runtime authorities:
#
#   1. p_T4_ordered_* producer
#   2. morphology normalization cohort
#   3. rank_in_dataset -> morph_rank materialization
#   4. association output -> candidate_rank materialization
#   5. association output -> candidate_prob materialization
#
#
# CONSEQUENCES
# ------------
# Morph provider:
#     BLOCKED
#
# Post-V2 provider code:
#     PRESENT / structurally implemented
#     but fresh execution precondition UNSATISFIED
#
# B12 portable materializer code:
#     PRESENT
#     but FULL fresh B12 materialization BLOCKED through post-V2 Morph dependency
#
# Historical structural oracle:
#     REMAINS CLOSED
#
# CURRENT_BEST:
#     UNAFFECTED
#
#
# IMPORTANT
# ---------
# This is NOT:
#
#   - a performance failure
#   - a scientific rejection of MORPH-DIV-V2
#   - evidence that historical Morph actions were wrong
#
# It is a PORTABILITY / SOURCE-AUTHORITY blocker only.
#
#
# NO:
#   - source mutation
#   - test mutation
#   - Git operations
#   - historical pickle loading
#   - historical structural oracle opening
#   - GT
#   - evaluator
#   - performance
#   - parameter search
#   - provider implementation
# =============================================================================

import hashlib
import json

from pathlib import Path


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_R2_REL = Path(
    "reports/"
    "notebook14_14_3e_d_i_r2_"
    "morph_orphaned_git_tree_"
    "read_only_source_authority_recovery_v1.json"
)

PARENT_R2 = ROOT / PARENT_R2_REL


BOUNDARY_REL = Path(
    "reports/"
    "notebook14_14_3e_d_h_"
    "morph_specification_derived_"
    "reconstruction_boundary_freeze_v1.json"
)

BOUNDARY = ROOT / BOUNDARY_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


B12_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_b12_portable.py"
)

B12 = ROOT / B12_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_j_"
    "morph_portability_blocker_finalization_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Exact frozen identities
# =============================================================================

EXPECTED_R2_FILE_SHA = (
    "13d44309b35768c993c09f8a4a64d161f1cced2762d10703c523fef57708c0e2"
)

EXPECTED_R2_CONTENT_SHA = (
    "e4cb116d6c025726451ec0121c67221e4a710f4984e0d756d55e0e66e6dd6ba8"
)


EXPECTED_DH_FILE_SHA = (
    "df62542ae6217564a6241657f7a4bc79a789ece5858c9e77217f092153cbcfe0"
)

EXPECTED_DH_CONTENT_SHA = (
    "d3cbfd2e549c221be945730f4118a59170f3cbd313e2ef6fc764c74fcfc371a9"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_B12_SHA = (
    "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
)


EXPECTED_R2_STATUS = (
    "ORPHANED_GIT_TREE_OBJECTS_DO_NOT_RECOVER_"
    "COMPLETE_MORPH_SOURCE_AUTHORITY"
)


EXPECTED_DH_STATUS = (
    "RECONSTRUCTION_BOUNDARY_FROZEN_"
    "IMPLEMENTATION_BLOCKED_PENDING_NEW_SOURCE_AUTHORITY"
)


EXPECTED_FRESH_BLOCKERS = {
    "fresh_p_T4_ordered_producer",
    "fresh_morphology_normalization_cohort",
    "fresh_morph_rank_materialization",
    "fresh_candidate_rank_materialization",
    "fresh_candidate_prob_materialization",
}


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(value):
    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_file(path):
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value):
    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_json(path):
    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def verify_canonical_artifact(
    path,
    expected_file_sha,
    expected_content_sha,
):

    assert (
        path.exists()
    ), path

    actual_file_sha = sha256_file(
        path
    )

    assert (
        actual_file_sha
        == expected_file_sha
    ), (
        f"{path.name} file SHA drift.\n"
        f"expected={expected_file_sha}\n"
        f"actual={actual_file_sha}"
    )

    obj = load_json(
        path
    )

    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )

    assert (
        obj.get(
            "content_sha256"
        )
        == expected_content_sha
    )

    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )

    assert (
        canonical_json_sha(
            payload
        )
        == expected_content_sha
    )

    return obj


# =============================================================================
# 3. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT_R2,
    BOUNDARY,
    PORTABLE,
    B12,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-D-J "
    "MORPH PORTABILITY BLOCKER FINALIZATION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "R2 parent:",
    PARENT_R2_REL,
    flush=True,
)

print(
    "D-H boundary:",
    BOUNDARY_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 4. Exact parent locks
# =============================================================================

r2 = verify_canonical_artifact(
    PARENT_R2,
    EXPECTED_R2_FILE_SHA,
    EXPECTED_R2_CONTENT_SHA,
)


dh = verify_canonical_artifact(
    BOUNDARY,
    EXPECTED_DH_FILE_SHA,
    EXPECTED_DH_CONTENT_SHA,
)


assert (
    r2[
        "resolution"
    ][
        "status"
    ]
    == EXPECTED_R2_STATUS
)


assert (
    r2[
        "resolution"
    ][
        "implementation_authorized"
    ]
    is False
)


assert (
    r2[
        "resolution"
    ][
        "historical_oracle_access_authorized"
    ]
    is False
)


assert (
    dh[
        "resolution"
    ][
        "status"
    ]
    == EXPECTED_DH_STATUS
)


assert (
    set(
        dh[
            "unresolved_fresh_runtime_components"
        ]
    )
    == EXPECTED_FRESH_BLOCKERS
)


assert (
    dh[
        "resolution"
    ][
        "concrete_Morph_provider_implementation_authorized"
    ]
    is False
)


assert (
    dh[
        "resolution"
    ][
        "historical_oracle_access_authorized"
    ]
    is False
)


print(
    "[PASS] AUTHORITY_RECOVERY_PARENT_LOCKS",
    flush=True,
)

print(
    "  D-I-R2 file SHA:",
    EXPECTED_R2_FILE_SHA,
    flush=True,
)

print(
    "  D-I-R2 content SHA:",
    EXPECTED_R2_CONTENT_SHA,
    flush=True,
)

print(
    "  D-H file SHA:",
    EXPECTED_DH_FILE_SHA,
    flush=True,
)

print(
    "  D-H content SHA:",
    EXPECTED_DH_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Verify exact recovery exhaustion evidence
# =============================================================================

inventory = r2[
    "git_object_inventory"
]


summary = r2[
    "summary"
]


orphan_tree_count = len(
    inventory[
        "orphaned_tree_objects"
    ]
)


standalone_blob_count = len(
    inventory[
        "standalone_orphan_blob_objects"
    ]
)


source_path_blob_count = int(
    inventory[
        "unique_source_path_blob_count"
    ]
)


relevant_source_unit_count = int(
    summary[
        "relevant_source_unit_count"
    ]
)


clean_complete_count = len(
    summary[
        "clean_complete_trees"
    ]
)


complete_count = len(
    summary[
        "complete_trees"
    ]
)


four_of_five_count = len(
    summary[
        "four_of_five_trees"
    ]
)


assert (
    orphan_tree_count == 6
)


assert (
    standalone_blob_count == 3
)


assert (
    source_path_blob_count == 73
)


assert (
    relevant_source_unit_count == 9
)


assert (
    clean_complete_count == 0
)


assert (
    complete_count == 0
)


assert (
    four_of_five_count == 0
)


assert (
    inventory[
        "standalone_orphan_blobs_opened"
    ]
    is False
)


print(
    "=" * 120,
    flush=True,
)

print(
    "SOURCE AUTHORITY RECOVERY EXHAUSTION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "orphan trees:",
    orphan_tree_count,
    flush=True,
)

print(
    "standalone orphan blobs:",
    standalone_blob_count,
    flush=True,
)

print(
    "path-identified source blobs:",
    source_path_blob_count,
    flush=True,
)

print(
    "relevant Morph source units:",
    relevant_source_unit_count,
    flush=True,
)

print(
    "clean-complete trees:",
    clean_complete_count,
    flush=True,
)

print(
    "complete trees:",
    complete_count,
    flush=True,
)

print(
    "four-of-five trees:",
    four_of_five_count,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Current source locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


b12_sha_before = sha256_file(
    B12
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    b12_sha_before
    == EXPECTED_B12_SHA
)


portable_text = PORTABLE.read_text(
    encoding="utf-8"
)


b12_text = B12.read_text(
    encoding="utf-8"
)


assert (
    "morph_candidate_provider"
    in portable_text
), (
    "Explicit Morph provider seam disappeared from portable runtime."
)


assert (
    "post_v2_graph_provider"
    in b12_text
), (
    "B12 no longer exposes the frozen post-V2 dependency seam."
)


print(
    "[PASS] CURRENT_PORTABILITY_SOURCE_LOCKS",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  B12 portable SHA:",
    b12_sha_before,
    flush=True,
)

print(
    "  explicit morph provider seam:",
    True,
    flush=True,
)

print(
    "  B12 post-V2 provider dependency:",
    True,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Recovery-route closure matrix
# =============================================================================

recovery_routes = {
    "current_checkout_source_archaeology": {
        "status":
            "EXHAUSTED",

        "evidence":
            "14.3E-D-G / 14.3E-D-H",
    },

    "Git_commit_history": {
        "status":
            "UNAVAILABLE",

        "evidence":
            (
                "unborn main; zero refs; zero reachable commits; "
                "zero dangling/unreachable commit objects"
            ),
    },

    "orphaned_Git_tree_source_recovery": {
        "status":
            "EXHAUSTED_INCOMPLETE",

        "orphan_trees":
            orphan_tree_count,

        "source_path_blobs":
            source_path_blob_count,

        "relevant_source_units":
            relevant_source_unit_count,

        "complete_trees":
            complete_count,

        "clean_complete_trees":
            clean_complete_count,

        "four_of_five_trees":
            four_of_five_count,
    },

    "historical_structural_oracle": {
        "status":
            "CLOSED_NOT_A_SOURCE_RECOVERY_ROUTE",

        "reason":
            (
                "Oracle may only be used after independent fresh output "
                "is frozen; it cannot define missing generator semantics."
            ),
    },
}


source_recovery_routes_exhausted = (
    recovery_routes[
        "current_checkout_source_archaeology"
    ][
        "status"
    ]
    == "EXHAUSTED"
    and
    recovery_routes[
        "Git_commit_history"
    ][
        "status"
    ]
    == "UNAVAILABLE"
    and
    recovery_routes[
        "orphaned_Git_tree_source_recovery"
    ][
        "status"
    ]
    == "EXHAUSTED_INCOMPLETE"
)


assert (
    source_recovery_routes_exhausted
    is True
)


# =============================================================================
# 8. Final unresolved Morph authority
# =============================================================================

unresolved_fresh_authority = [
    "fresh_p_T4_ordered_producer",
    "fresh_morphology_normalization_cohort",
    "fresh_morph_rank_materialization",
    "fresh_candidate_rank_materialization",
    "fresh_candidate_prob_materialization",
]


assert (
    set(
        unresolved_fresh_authority
    )
    == EXPECTED_FRESH_BLOCKERS
)


blocker_details = {
    "fresh_p_T4_ordered_producer": {
        "status":
            "BLOCKED_MISSING_SOURCE_AUTHORITY",

        "historical_evidence":
            (
                "p_T4_ordered_* columns are consumed by historical OOF "
                "Morph ranking, but their fresh producer is absent."
            ),
    },

    "fresh_morphology_normalization_cohort": {
        "status":
            "BLOCKED_MISSING_SOURCE_AUTHORITY",

        "historical_evidence":
            (
                "Historical transform uses groupby('fold').rank("
                "pct=True, method='average')."
            ),

        "forbidden_substitution":
            (
                "Do not invent fresh-dataset/fresh-batch/calibration "
                "population semantics."
            ),
    },

    "fresh_morph_rank_materialization": {
        "status":
            "BLOCKED_MISSING_SOURCE_AUTHORITY",

        "forbidden_substitution":
            (
                "Do not infer morph_rank == rank_in_dataset."
            ),
    },

    "fresh_candidate_rank_materialization": {
        "status":
            "BLOCKED_MISSING_SOURCE_AUTHORITY",

        "forbidden_substitution":
            (
                "Do not infer candidate_rank from an argsort of the "
                "association matrix."
            ),
    },

    "fresh_candidate_prob_materialization": {
        "status":
            "BLOCKED_MISSING_SOURCE_AUTHORITY",

        "forbidden_substitution":
            (
                "Do not infer candidate_prob as the selected association "
                "matrix entry without source authority."
            ),
    },
}


# =============================================================================
# 9. Portability propagation
# =============================================================================

morph_selection_specification = (
    "FROZEN"
)


morph_graph_application_specification = (
    "FROZEN"
)


reference_association_provider = (
    "EXACT_SOURCE_AUTHORITY"
)


concrete_morph_provider = (
    "BLOCKED_MISSING_FRESH_SOURCE_AUTHORITY"
)


post_v2_provider_implementation = (
    "CODE_PRESENT_DEPENDENCY_UNSATISFIED"
)


fresh_post_v2_runtime = (
    "BLOCKED_BY_MORPH_PROVIDER"
)


b12_materializer_implementation = (
    "CODE_PRESENT"
)


b12_full_fresh_materialization = (
    "BLOCKED_BY_POST_V2_MORPH_DEPENDENCY"
)


historical_structural_oracle = (
    "CLOSED"
)


structural_fidelity_verified = (
    False
)


current_best_replaced = (
    False
)


# Only the Morph-dependent execution path is blocked.
non_morph_engineering_may_continue = (
    True
)


# =============================================================================
# 10. Explicit unblock contract
# =============================================================================

unblock_requirements = [
    {
        "id":
            "NEW_SOURCE_AUTHORITY_FOR_P_T4",

        "requirement":
            (
                "Provide exact attributable source authority for fresh "
                "p_T4_ordered_* generation."
            ),
    },

    {
        "id":
            "NEW_SOURCE_AUTHORITY_FOR_NORMALIZATION",

        "requirement":
            (
                "Provide exact fresh-runtime morphology normalization cohort "
                "semantics replacing historical OOF fold context."
            ),
    },

    {
        "id":
            "NEW_SOURCE_AUTHORITY_FOR_MORPH_RANK",

        "requirement":
            (
                "Provide exact materialization rule for morph_rank."
            ),
    },

    {
        "id":
            "NEW_SOURCE_AUTHORITY_FOR_CANDIDATE_RANK",

        "requirement":
            (
                "Provide exact association-output to candidate_rank mapping."
            ),
    },

    {
        "id":
            "NEW_SOURCE_AUTHORITY_FOR_CANDIDATE_PROB",

        "requirement":
            (
                "Provide exact association-output to candidate_prob mapping."
            ),
    },
]


resume_rule = (
    "Morph concrete-provider implementation may resume only after "
    "all unresolved fresh-runtime authority is closed by new attributable "
    "source/specification authority independent of the historical structural "
    "oracle."
)


# =============================================================================
# 11. Final status
# =============================================================================

status = (
    "MORPH_PORTABILITY_BLOCKER_FINALIZED_"
    "NO_GUESS_RECONSTRUCTION_AUTHORIZED"
)


next_stage = (
    "NOTEBOOK14_14_3E_E_"
    "B12_PORTABILITY_READINESS_RECLASSIFICATION"
)


concrete_morph_provider_implementation_authorized = (
    False
)


historical_oracle_access_authorized = (
    False
)


full_b12_fresh_execution_authorized = (
    False
)


print(
    "=" * 120,
    flush=True,
)

print(
    "FINAL PORTABILITY STATUS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "source recovery routes exhausted:",
    source_recovery_routes_exhausted,
    flush=True,
)

print(
    flush=True,
)


print(
    "Morph selection specification:",
    morph_selection_specification,
    flush=True,
)

print(
    "Morph graph application specification:",
    morph_graph_application_specification,
    flush=True,
)

print(
    "reference association provider:",
    reference_association_provider,
    flush=True,
)

print(
    "concrete Morph provider:",
    concrete_morph_provider,
    flush=True,
)

print(
    flush=True,
)


print(
    "post-V2 provider implementation:",
    post_v2_provider_implementation,
    flush=True,
)

print(
    "fresh post-V2 runtime:",
    fresh_post_v2_runtime,
    flush=True,
)

print(
    flush=True,
)


print(
    "B12 materializer implementation:",
    b12_materializer_implementation,
    flush=True,
)

print(
    "B12 full fresh materialization:",
    b12_full_fresh_materialization,
    flush=True,
)

print(
    flush=True,
)


print(
    "historical structural oracle:",
    historical_structural_oracle,
    flush=True,
)

print(
    "structural fidelity verified:",
    structural_fidelity_verified,
    flush=True,
)

print(
    "CURRENT_BEST replaced:",
    current_best_replaced,
    flush=True,
)

print(
    flush=True,
)


print(
    "non-Morph engineering may continue:",
    non_morph_engineering_may_continue,
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 12. Safety
# =============================================================================

source_mutation = False

test_mutation = False

Git_operation_executed = False

historical_pickle_loaded = False

historical_oracle_opened = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    sha256_file(
        B12
    )
    == b12_sha_before
    == EXPECTED_B12_SHA
)


# =============================================================================
# 13. Formal gates
# =============================================================================

semantic_gates = {
    "D_I_R2_parent_exactly_locked":
        True,

    "D_H_boundary_exactly_locked":
        True,

    "current_portable_source_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "B12_source_locked":
        (
            b12_sha_before
            == EXPECTED_B12_SHA
        ),

    "source_recovery_routes_exhausted":
        source_recovery_routes_exhausted,

    "all_five_fresh_Morph_authority_gaps_preserved":
        (
            set(
                unresolved_fresh_authority
            )
            == EXPECTED_FRESH_BLOCKERS
        ),

    "selection_specification_remains_frozen":
        (
            morph_selection_specification
            == "FROZEN"
        ),

    "graph_application_specification_remains_frozen":
        (
            morph_graph_application_specification
            == "FROZEN"
        ),

    "association_provider_authority_preserved":
        (
            reference_association_provider
            == "EXACT_SOURCE_AUTHORITY"
        ),

    "concrete_Morph_provider_not_invented":
        (
            concrete_morph_provider_implementation_authorized
            is False
        ),

    "post_V2_fresh_runtime_not_prematurely_claimed":
        (
            fresh_post_v2_runtime
            == "BLOCKED_BY_MORPH_PROVIDER"
        ),

    "B12_full_fresh_materialization_not_prematurely_claimed":
        (
            full_b12_fresh_execution_authorized
            is False
        ),

    "historical_structural_oracle_remains_closed":
        (
            historical_oracle_access_authorized
            is False
            and
            historical_oracle_opened
            is False
        ),

    "structural_fidelity_not_prematurely_claimed":
        (
            structural_fidelity_verified
            is False
        ),

    "CURRENT_BEST_not_replaced":
        (
            current_best_replaced
            is False
        ),

    "non_Morph_engineering_not_overblocked":
        (
            non_morph_engineering_may_continue
            is True
        ),

    "no_Git_operations":
        (
            Git_operation_executed
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-D-J SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates
    == []
), (
    "14.3E-D-J failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 14. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_D_J",

    "title":
        "MORPH_PORTABILITY_BLOCKER_FINALIZATION",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        (
            "ENGINEERING_PORTABILITY_BLOCKER_FINALIZED_"
            "NO_GUESS_RECONSTRUCTION"
        ),

    "parent_14_3e_d_i_r2": {
        "path":
            str(
                PARENT_R2_REL
            ),

        "file_sha256":
            EXPECTED_R2_FILE_SHA,

        "content_sha256":
            EXPECTED_R2_CONTENT_SHA,
    },

    "boundary_14_3e_d_h": {
        "path":
            str(
                BOUNDARY_REL
            ),

        "file_sha256":
            EXPECTED_DH_FILE_SHA,

        "content_sha256":
            EXPECTED_DH_CONTENT_SHA,
    },

    "current_source_locks": {
        "portable_current_best": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "sha256":
                portable_sha_before,
        },

        "B12_portable_materializer": {
            "path":
                str(
                    B12_REL
                ),

            "sha256":
                b12_sha_before,
        },
    },

    "source_authority_recovery": {
        "routes":
            recovery_routes,

        "all_routes_exhausted":
            source_recovery_routes_exhausted,
    },

    "unresolved_fresh_authority":
        blocker_details,

    "portability_propagation": {
        "Morph_selection_specification":
            morph_selection_specification,

        "Morph_graph_application_specification":
            morph_graph_application_specification,

        "reference_association_provider":
            reference_association_provider,

        "concrete_Morph_provider":
            concrete_morph_provider,

        "post_v2_provider_implementation":
            post_v2_provider_implementation,

        "fresh_post_v2_runtime":
            fresh_post_v2_runtime,

        "B12_materializer_implementation":
            b12_materializer_implementation,

        "B12_full_fresh_materialization":
            b12_full_fresh_materialization,

        "historical_structural_oracle":
            historical_structural_oracle,

        "structural_fidelity_verified":
            structural_fidelity_verified,

        "non_Morph_engineering_may_continue":
            non_morph_engineering_may_continue,
    },

    "unblock_contract": {
        "requirements":
            unblock_requirements,

        "resume_rule":
            resume_rule,
    },

    "resolution": {
        "status":
            status,

        "source_recovery_routes_exhausted":
            source_recovery_routes_exhausted,

        "concrete_Morph_provider_implementation_authorized":
            False,

        "historical_oracle_access_authorized":
            False,

        "fresh_post_V2_runtime_ready":
            False,

        "full_B12_fresh_execution_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "CURRENT_BEST_replaced":
            False,

        "non_Morph_engineering_may_continue":
            True,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "scientific_result":
            "NONE",

        "performance_result":
            "NONE",

        "Morph_policy_validity":
            "NOT_REJECTED",

        "Morph_portability":
            "BLOCKED_BY_INCOMPLETE_FRESH_SOURCE_AUTHORITY",

        "post_V2_portability":
            "BLOCKED_TRANSITIVELY_BY_MORPH",

        "B12_full_portability":
            "BLOCKED_TRANSITIVELY_BY_POST_V2_MORPH",

        "historical_structural_oracle":
            "CLOSED",

        "historical_pickle_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "scientific_parameter_search":
            False,

        "CURRENT_BEST_replaced":
            False,

        "source_mutation":
            False,

        "test_mutation":
            False,

        "Git_operation":
            False,
    },

    "explicit_non_claims": [
        "This blocker is not a performance failure.",
        "This blocker does not invalidate historical MORPH-DIV-V2 actions.",
        "Historical OOF normalization is not converted into a fresh-runtime cohort rule.",
        "morph_rank is not inferred to equal rank_in_dataset.",
        "candidate_rank is not inferred from association-matrix ordering.",
        "candidate_prob is not inferred from an association-matrix entry.",
        "Historical action/audit artifacts are not used as runtime providers.",
        "Historical structural oracle remains closed.",
        "No concrete MorphCandidateProvider was implemented.",
        "No post-V2 structural fidelity claim.",
        "No B12 structural fidelity claim.",
        "No performance evidence.",
        "No CURRENT_BEST replacement.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-D-J artifact exists but differs "
        "from exact blocker-finalization state."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    OUT.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 15. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-D-J PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "SOURCE AUTHORITY:",
    flush=True,
)

print(
    "  current checkout:",
    "EXHAUSTED",
    flush=True,
)

print(
    "  Git commit history:",
    "UNAVAILABLE",
    flush=True,
)

print(
    "  orphaned Git trees:",
    "EXHAUSTED_INCOMPLETE",
    flush=True,
)

print(
    "  all source-recovery routes exhausted:",
    source_recovery_routes_exhausted,
    flush=True,
)

print(
    flush=True,
)


print(
    "MORPH:",
    flush=True,
)

print(
    "  selection spec:",
    morph_selection_specification,
    flush=True,
)

print(
    "  graph application spec:",
    morph_graph_application_specification,
    flush=True,
)

print(
    "  association provider:",
    reference_association_provider,
    flush=True,
)

print(
    "  concrete provider:",
    concrete_morph_provider,
    flush=True,
)

print(
    flush=True,
)


print(
    "TRANSITIVE PORTABILITY:",
    flush=True,
)

print(
    "  post-V2:",
    fresh_post_v2_runtime,
    flush=True,
)

print(
    "  B12 full fresh materialization:",
    b12_full_fresh_materialization,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIMS:",
    flush=True,
)

print(
    "  historical oracle opened:",
    False,
    flush=True,
)

print(
    "  structural fidelity verified:",
    False,
    flush=True,
)

print(
    "  performance evidence:",
    False,
    flush=True,
)

print(
    "  CURRENT_BEST replaced:",
    False,
    flush=True,
)

print(
    "  non-Morph engineering may continue:",
    True,
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-E
# B12 PORTABILITY READINESS RECLASSIFICATION
#
# PARENT
# ------
# 14.3E-D-J PASS
#
# D-J finalized:
#
#   Morph selection specification        = FROZEN
#   Morph graph application specification= FROZEN
#   reference association provider       = EXACT_SOURCE_AUTHORITY
#   concrete Morph provider              = BLOCKED_MISSING_FRESH_SOURCE_AUTHORITY
#
#   post-V2 provider implementation      = CODE_PRESENT_DEPENDENCY_UNSATISFIED
#   fresh post-V2 runtime                = BLOCKED_BY_MORPH_PROVIDER
#
#   B12 materializer implementation      = CODE_PRESENT
#   B12 full fresh materialization       =
#       BLOCKED_BY_POST_V2_MORPH_DEPENDENCY
#
#
# THIS STAGE
# ----------
# Reclassify B12 readiness into separate dimensions:
#
#   SOURCE_PRESENT
#   DEPENDENCY_SEAM_PRESENT
#   DEPENDENCY_SATISFIED
#   EXECUTION_AUTHORIZED
#   STRUCTURAL_FIDELITY_VERIFIED
#
# This avoids the false implication:
#
#       "implementation exists" == "full fresh execution ready"
#
#
# NO:
#   - source mutation
#   - test mutation
#   - provider implementation
#   - runtime materialization
#   - historical pickle loading
#   - historical structural oracle opening
#   - GT
#   - evaluator
#   - performance
#   - parameter search
#   - Git operations
# =============================================================================

import ast
import hashlib
import json

from pathlib import Path


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_d_j_"
    "morph_portability_blocker_finalization_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


B12_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_b12_portable.py"
)

B12 = ROOT / B12_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_e_"
    "b12_portability_readiness_reclassification_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Exact frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "9228ac29280b21122e7ff8076776b7daaab5f81eee0c95e3604357f747c8e76c"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "68da82fe8d78aad75a0fb23e0e6095019e45bae719c8f8a3817f105b8a3b2262"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_B12_SHA = (
    "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
)


EXPECTED_PARENT_STATUS = (
    "MORPH_PORTABILITY_BLOCKER_FINALIZED_"
    "NO_GUESS_RECONSTRUCTION_AUTHORIZED"
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(value):
    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(value):
    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(path):
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value):
    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(path):
    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def node_source(
    source,
    node,
):

    value = ast.get_source_segment(
        source,
        node,
    )

    if value is not None:
        return value

    return ast.unparse(
        node
    )


def verify_artifact(
    path,
    expected_file_sha,
    expected_content_sha,
):

    assert path.exists(), path

    actual_file_sha = sha256_file(
        path
    )

    assert (
        actual_file_sha
        == expected_file_sha
    ), (
        f"{path.name}: file SHA mismatch.\n"
        f"expected={expected_file_sha}\n"
        f"actual={actual_file_sha}"
    )

    obj = load_json(
        path
    )

    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )

    assert (
        obj.get(
            "content_sha256"
        )
        == expected_content_sha
    )

    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )

    assert (
        canonical_json_sha(
            payload
        )
        == expected_content_sha
    )

    return obj


# =============================================================================
# 3. Static source inventory
# =============================================================================

def top_level_inventory(
    source,
):

    tree = ast.parse(
        source
    )

    records = []

    for node in tree.body:

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):

            segment = node_source(
                source,
                node,
            )

            records.append({
                "kind":
                    type(
                        node
                    ).__name__,

                "name":
                    node.name,

                "line_start":
                    node.lineno,

                "line_end":
                    getattr(
                        node,
                        "end_lineno",
                        node.lineno,
                    ),

                "sha256":
                    sha256_text(
                        segment
                    ),
            })

    return records


def symbol_occurrences(
    source,
    symbol,
):

    tree = ast.parse(
        source
    )

    records = []

    for node in ast.walk(
        tree
    ):

        matched = False

        if (
            isinstance(
                node,
                ast.Name,
            )
            and
            node.id
            == symbol
        ):
            matched = True

        elif (
            isinstance(
                node,
                ast.Attribute,
            )
            and
            node.attr
            == symbol
        ):
            matched = True

        if not matched:
            continue

        records.append({
            "kind":
                type(
                    node
                ).__name__,

            "line":
                getattr(
                    node,
                    "lineno",
                    None,
                ),

            "source":
                node_source(
                    source,
                    node,
                ),
        })

    return records


def call_sites_containing(
    source,
    token,
):

    tree = ast.parse(
        source
    )

    records = []

    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            ast.Call,
        ):
            continue

        segment = node_source(
            source,
            node,
        )

        if token not in segment:
            continue

        records.append({
            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "sha256":
                sha256_text(
                    segment
                ),

            "source":
                segment[
                    :3000
                ],
        })

    return records


def enclosing_units_containing(
    source,
    token,
):

    tree = ast.parse(
        source
    )

    records = []

    for node in tree.body:

        if not isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):
            continue

        segment = node_source(
            source,
            node,
        )

        if token not in segment:
            continue

        records.append({
            "kind":
                type(
                    node
                ).__name__,

            "name":
                node.name,

            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "sha256":
                sha256_text(
                    segment
                ),
        })

    return records


# =============================================================================
# 4. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    B12,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-E "
    "B12 PORTABILITY READINESS RECLASSIFICATION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Parent lock
# =============================================================================

parent = verify_artifact(
    PARENT,
    EXPECTED_PARENT_FILE_SHA,
    EXPECTED_PARENT_CONTENT_SHA,
)


assert (
    parent[
        "resolution"
    ][
        "status"
    ]
    == EXPECTED_PARENT_STATUS
)


assert (
    parent[
        "resolution"
    ][
        "source_recovery_routes_exhausted"
    ]
    is True
)


assert (
    parent[
        "resolution"
    ][
        "concrete_Morph_provider_implementation_authorized"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "fresh_post_V2_runtime_ready"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "full_B12_fresh_execution_authorized"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "structural_fidelity_verified"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "CURRENT_BEST_replaced"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "non_Morph_engineering_may_continue"
    ]
    is True
)


print(
    "[PASS] D_J_PARENT_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  status:",
    EXPECTED_PARENT_STATUS,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Current source locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)

b12_sha_before = sha256_file(
    B12
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    b12_sha_before
    == EXPECTED_B12_SHA
)


portable_source = PORTABLE.read_text(
    encoding="utf-8"
)

b12_source = B12.read_text(
    encoding="utf-8"
)


portable_inventory = top_level_inventory(
    portable_source
)

b12_inventory = top_level_inventory(
    b12_source
)


print(
    "[PASS] CURRENT_SOURCE_LOCK",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  B12 SHA:",
    b12_sha_before,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Static dependency-seam inspection
# =============================================================================

B12_REQUIRED_SYMBOLS = (
    "suppressed_peak_provider",
    "post_v2_graph_provider",
)


PORTABLE_REQUIRED_SYMBOLS = (
    "morph_candidate_provider",
)


b12_symbol_records = {
    symbol:
        symbol_occurrences(
            b12_source,
            symbol,
        )

    for symbol
    in B12_REQUIRED_SYMBOLS
}


portable_symbol_records = {
    symbol:
        symbol_occurrences(
            portable_source,
            symbol,
        )

    for symbol
    in PORTABLE_REQUIRED_SYMBOLS
}


b12_call_sites = {
    symbol:
        call_sites_containing(
            b12_source,
            symbol,
        )

    for symbol
    in B12_REQUIRED_SYMBOLS
}


portable_units = {
    symbol:
        enclosing_units_containing(
            portable_source,
            symbol,
        )

    for symbol
    in PORTABLE_REQUIRED_SYMBOLS
}


for symbol in B12_REQUIRED_SYMBOLS:

    assert (
        len(
            b12_symbol_records[
                symbol
            ]
        )
        > 0
    ), (
        f"B12 dependency seam missing symbol: {symbol}"
    )


assert (
    len(
        portable_symbol_records[
            "morph_candidate_provider"
        ]
    )
    > 0
), (
    "Portable runtime no longer contains explicit "
    "morph_candidate_provider seam."
)


assert (
    len(
        b12_call_sites[
            "post_v2_graph_provider"
        ]
    )
    > 0
), (
    "B12 source mentions post_v2_graph_provider "
    "but no executable call site was found."
)


print(
    "=" * 120,
    flush=True,
)

print(
    "STATIC DEPENDENCY SEAMS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for symbol in B12_REQUIRED_SYMBOLS:

    print(
        symbol,
        flush=True,
    )

    print(
        "  occurrences:",
        len(
            b12_symbol_records[
                symbol
            ]
        ),
        flush=True,
    )

    print(
        "  call sites:",
        len(
            b12_call_sites[
                symbol
            ]
        ),
        flush=True,
    )


print(
    "morph_candidate_provider",
    flush=True,
)

print(
    "  occurrences:",
    len(
        portable_symbol_records[
            "morph_candidate_provider"
        ]
    ),
    flush=True,
)

print(
    "  enclosing units:",
    portable_units[
        "morph_candidate_provider"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. Inherit authoritative D-J dependency state
# =============================================================================

propagation = parent[
    "portability_propagation"
]


assert (
    propagation[
        "concrete_Morph_provider"
    ]
    == "BLOCKED_MISSING_FRESH_SOURCE_AUTHORITY"
)


assert (
    propagation[
        "post_v2_provider_implementation"
    ]
    == "CODE_PRESENT_DEPENDENCY_UNSATISFIED"
)


assert (
    propagation[
        "fresh_post_v2_runtime"
    ]
    == "BLOCKED_BY_MORPH_PROVIDER"
)


assert (
    propagation[
        "B12_materializer_implementation"
    ]
    == "CODE_PRESENT"
)


assert (
    propagation[
        "B12_full_fresh_materialization"
    ]
    == "BLOCKED_BY_POST_V2_MORPH_DEPENDENCY"
)


assert (
    propagation[
        "historical_structural_oracle"
    ]
    == "CLOSED"
)


# =============================================================================
# 9. Readiness dimensions
# =============================================================================

#
# Important distinction:
#
# SOURCE_PRESENT
#     source code exists
#
# DEPENDENCY_SEAM_PRESENT
#     caller explicitly accepts/invokes dependency
#
# DEPENDENCY_SATISFIED
#     all runtime authorities needed by dependency are available
#
# EXECUTION_AUTHORIZED
#     formal provenance boundary permits the call
#
# FIDELITY_VERIFIED
#     fresh frozen output was compared to structural oracle
#

readiness_matrix = {
    "portable_current_best_source": {
        "source_present":
            True,

        "dependency_seam_present":
            True,

        "dependency_satisfied":
            False,

        "execution_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "blocking_dependency":
            "concrete_MorphCandidateProvider",

        "status":
            "SOURCE_PRESENT_MORPH_DEPENDENCY_BLOCKED",
    },

    "concrete_MorphCandidateProvider": {
        "source_present":
            False,

        "dependency_seam_present":
            True,

        "dependency_satisfied":
            False,

        "execution_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "status":
            "BLOCKED_MISSING_FRESH_SOURCE_AUTHORITY",
    },

    "post_v2_graph_provider": {
        "source_present":
            True,

        "dependency_seam_present":
            True,

        "dependency_satisfied":
            False,

        "execution_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "blocking_dependency":
            "concrete_MorphCandidateProvider",

        "status":
            "CODE_PRESENT_DEPENDENCY_UNSATISFIED",
    },

    "B12_portable_materializer": {
        "source_present":
            True,

        "dependency_seam_present":
            True,

        "dependency_satisfied":
            False,

        "execution_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "blocking_dependency":
            "post_v2_graph_provider",

        "status":
            "CODE_PRESENT_FULL_FRESH_EXECUTION_BLOCKED",
    },

    "historical_structural_oracle": {
        "source_present":
            True,

        "dependency_seam_present":
            None,

        "dependency_satisfied":
            False,

        "execution_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "status":
            "CLOSED",
    },
}


# =============================================================================
# 10. B12 readiness classification
# =============================================================================

full_fresh_runtime_preconditions = {
    "B12_materializer_code_present":
        True,

    "suppressed_peak_dependency_seam_present":
        (
            len(
                b12_symbol_records[
                    "suppressed_peak_provider"
                ]
            )
            > 0
        ),

    "post_v2_dependency_seam_present":
        (
            len(
                b12_symbol_records[
                    "post_v2_graph_provider"
                ]
            )
            > 0
        ),

    "post_v2_call_site_present":
        (
            len(
                b12_call_sites[
                    "post_v2_graph_provider"
                ]
            )
            > 0
        ),

    "portable_morph_dependency_seam_present":
        (
            len(
                portable_symbol_records[
                    "morph_candidate_provider"
                ]
            )
            > 0
        ),

    "concrete_Morph_provider_available":
        False,

    "fresh_post_v2_runtime_ready":
        False,

    "historical_oracle_access_authorized":
        False,
}


full_fresh_runtime_ready = all([
    full_fresh_runtime_preconditions[
        "B12_materializer_code_present"
    ],

    full_fresh_runtime_preconditions[
        "suppressed_peak_dependency_seam_present"
    ],

    full_fresh_runtime_preconditions[
        "post_v2_dependency_seam_present"
    ],

    full_fresh_runtime_preconditions[
        "post_v2_call_site_present"
    ],

    full_fresh_runtime_preconditions[
        "portable_morph_dependency_seam_present"
    ],

    full_fresh_runtime_preconditions[
        "concrete_Morph_provider_available"
    ],

    full_fresh_runtime_preconditions[
        "fresh_post_v2_runtime_ready"
    ],
])


assert (
    full_fresh_runtime_ready
    is False
)


# =============================================================================
# 11. Distinguish what IS still authorized
# =============================================================================

authorized_engineering = {
    "static_contract_audit":
        True,

    "dependency_seam_tests_not_requiring_Morph_execution":
        True,

    "failure_mode_tests_for_missing_Morph_provider":
        True,

    "blocker_aware_preflight_harness":
        True,

    "artifact_and_hash_provenance_checks":
        True,

    "non_Morph_component_regression_tests":
        True,

    "fresh_full_post_v2_execution":
        False,

    "fresh_full_B12_materialization":
        False,

    "historical_oracle_opening":
        False,

    "structural_fidelity_claim":
        False,
}


# =============================================================================
# 12. One-shot confirmation readiness
# =============================================================================

one_shot_confirmation = {
    "full_end_to_end_execution":
        "BLOCKED",

    "blocking_chain": [
        "concrete_MorphCandidateProvider",
        "post_v2_graph_provider",
        "B12_portable_materializer",
    ],

    "blocker_aware_preflight":
        "AUTHORIZED",

    "preflight_required_behavior": [
        (
            "Verify current source and parent artifact locks."
        ),
        (
            "Verify B12/post-V2/Morph dependency seams exist."
        ),
        (
            "Refuse full execution while concrete Morph provider "
            "authority remains unresolved."
        ),
        (
            "Do not open historical structural oracle."
        ),
        (
            "Do not load historical Morph action/audit tables as "
            "runtime providers."
        ),
        (
            "Emit an explicit BLOCKED_BY_MORPH_PROVIDER result "
            "instead of silently skipping the stage."
        ),
    ],

    "fresh_output_freeze_possible_now":
        False,

    "historical_structural_comparison_possible_now":
        False,
}


# =============================================================================
# 13. Reclassification
# =============================================================================

readiness_status = (
    "B12_CODE_AND_SEAMS_READY_"
    "FULL_FRESH_EXECUTION_BLOCKED_BY_FINALIZED_MORPH_AUTHORITY_GAP"
)


implementation_state = (
    "IMPLEMENTATION_PRESENT"
)


execution_state = (
    "NOT_AUTHORIZED"
)


scientific_state = (
    "NO_NEW_SCIENTIFIC_RESULT"
)


structural_fidelity_state = (
    "NOT_VERIFIED"
)


current_best_replaced = (
    False
)


historical_oracle_access_authorized = (
    False
)


# We can continue with the blocker-aware harness.
next_stage = (
    "NOTEBOOK14_14_3E_F_"
    "B12_BLOCKER_AWARE_ONE_SHOT_PREFLIGHT_HARNESS"
)


print(
    "=" * 120,
    flush=True,
)

print(
    "B12 READINESS RECLASSIFICATION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "implementation:",
    implementation_state,
    flush=True,
)

print(
    "full fresh execution:",
    execution_state,
    flush=True,
)

print(
    "structural fidelity:",
    structural_fidelity_state,
    flush=True,
)

print(
    "scientific result:",
    scientific_state,
    flush=True,
)

print(
    flush=True,
)


print(
    "blocking chain:",
    flush=True,
)

print(
    "  concrete Morph provider",
    "-> post-V2 provider",
    "-> B12 materializer",
    flush=True,
)

print(
    flush=True,
)


print(
    "blocker-aware one-shot preflight authorized:",
    authorized_engineering[
        "blocker_aware_preflight_harness"
    ],
    flush=True,
)

print(
    "full end-to-end one-shot execution authorized:",
    False,
    flush=True,
)

print(
    "historical oracle access authorized:",
    False,
    flush=True,
)

print(
    "CURRENT_BEST replaced:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    readiness_status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Safety
# =============================================================================

source_mutation = False

test_mutation = False

Git_operation_executed = False

historical_pickle_loaded = False

historical_Morph_action_runtime_use = False

historical_oracle_opened = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False

runtime_materialization_executed = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    sha256_file(
        B12
    )
    == b12_sha_before
    == EXPECTED_B12_SHA
)


# =============================================================================
# 15. Formal gates
# =============================================================================

semantic_gates = {
    "D_J_parent_exactly_locked":
        True,

    "portable_source_exactly_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "B12_source_exactly_locked":
        (
            b12_sha_before
            == EXPECTED_B12_SHA
        ),

    "B12_suppressed_peak_dependency_seam_present":
        full_fresh_runtime_preconditions[
            "suppressed_peak_dependency_seam_present"
        ],

    "B12_post_v2_dependency_seam_present":
        full_fresh_runtime_preconditions[
            "post_v2_dependency_seam_present"
        ],

    "B12_post_v2_executable_call_site_present":
        full_fresh_runtime_preconditions[
            "post_v2_call_site_present"
        ],

    "portable_Morph_dependency_seam_present":
        full_fresh_runtime_preconditions[
            "portable_morph_dependency_seam_present"
        ],

    "concrete_Morph_provider_remains_unavailable":
        (
            full_fresh_runtime_preconditions[
                "concrete_Morph_provider_available"
            ]
            is False
        ),

    "fresh_post_v2_runtime_remains_blocked":
        (
            full_fresh_runtime_preconditions[
                "fresh_post_v2_runtime_ready"
            ]
            is False
        ),

    "B12_code_present_not_equated_with_execution_readiness":
        (
            implementation_state
            == "IMPLEMENTATION_PRESENT"
            and
            execution_state
            == "NOT_AUTHORIZED"
        ),

    "blocker_aware_preflight_authorized":
        (
            authorized_engineering[
                "blocker_aware_preflight_harness"
            ]
            is True
        ),

    "full_end_to_end_execution_not_authorized":
        (
            full_fresh_runtime_ready
            is False
        ),

    "historical_oracle_remains_closed":
        (
            historical_oracle_access_authorized
            is False
            and
            historical_oracle_opened
            is False
        ),

    "structural_fidelity_not_prematurely_claimed":
        (
            structural_fidelity_state
            == "NOT_VERIFIED"
        ),

    "CURRENT_BEST_not_replaced":
        (
            current_best_replaced
            is False
        ),

    "historical_Morph_actions_not_used_as_runtime_provider":
        (
            historical_Morph_action_runtime_use
            is False
        ),

    "runtime_materialization_not_executed":
        (
            runtime_materialization_executed
            is False
        ),

    "no_Git_operations":
        (
            Git_operation_executed
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-E SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates == []
), (
    "14.3E-E failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 16. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_3E_E",

    "title":
        "B12_PORTABILITY_READINESS_RECLASSIFICATION",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        (
            "ENGINEERING_B12_READINESS_RECLASSIFIED_"
            "IMPLEMENTATION_PRESENT_EXECUTION_BLOCKED"
        ),

    "parent_14_3e_d_j": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable_current_best": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "sha256":
                portable_sha_before,

            "top_level_inventory":
                portable_inventory,
        },

        "B12_portable": {
            "path":
                str(
                    B12_REL
                ),

            "sha256":
                b12_sha_before,

            "top_level_inventory":
                b12_inventory,
        },
    },

    "static_dependency_seams": {
        "B12_symbols":
            b12_symbol_records,

        "B12_call_sites":
            b12_call_sites,

        "portable_symbols":
            portable_symbol_records,

        "portable_enclosing_units":
            portable_units,
    },

    "inherited_D_J_portability_state": {
        "concrete_Morph_provider":
            propagation[
                "concrete_Morph_provider"
            ],

        "post_v2_provider_implementation":
            propagation[
                "post_v2_provider_implementation"
            ],

        "fresh_post_v2_runtime":
            propagation[
                "fresh_post_v2_runtime"
            ],

        "B12_materializer_implementation":
            propagation[
                "B12_materializer_implementation"
            ],

        "B12_full_fresh_materialization":
            propagation[
                "B12_full_fresh_materialization"
            ],

        "historical_structural_oracle":
            propagation[
                "historical_structural_oracle"
            ],
    },

    "readiness_matrix":
        readiness_matrix,

    "full_fresh_runtime_preconditions":
        full_fresh_runtime_preconditions,

    "full_fresh_runtime_ready":
        full_fresh_runtime_ready,

    "authorized_engineering":
        authorized_engineering,

    "one_shot_confirmation":
        one_shot_confirmation,

    "resolution": {
        "status":
            readiness_status,

        "implementation_state":
            implementation_state,

        "execution_state":
            execution_state,

        "scientific_state":
            scientific_state,

        "structural_fidelity_state":
            structural_fidelity_state,

        "blocker":
            "CONCRETE_MORPH_PROVIDER_MISSING_FRESH_SOURCE_AUTHORITY",

        "blocker_chain": [
            "concrete_MorphCandidateProvider",
            "post_v2_graph_provider",
            "B12_portable_materializer",
        ],

        "blocker_aware_preflight_authorized":
            True,

        "full_end_to_end_execution_authorized":
            False,

        "historical_oracle_access_authorized":
            False,

        "fresh_output_freeze_possible":
            False,

        "structural_comparison_possible":
            False,

        "CURRENT_BEST_replaced":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "B12_source_implementation":
            "PRESENT",

        "B12_dependency_seams":
            "PRESENT",

        "B12_full_fresh_execution":
            "BLOCKED",

        "post_V2_fresh_execution":
            "BLOCKED_BY_MORPH",

        "Morph_concrete_provider":
            "BLOCKED_MISSING_FRESH_SOURCE_AUTHORITY",

        "historical_structural_oracle":
            "CLOSED",

        "structural_fidelity_verified":
            False,

        "scientific_result":
            "NONE",

        "performance_result":
            "NONE",

        "CURRENT_BEST_replaced":
            False,

        "historical_pickle_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "parameter_search":
            False,

        "source_mutation":
            False,

        "test_mutation":
            False,

        "Git_operation":
            False,
    },

    "explicit_non_claims": [
        "Source implementation presence is not equivalent to execution readiness.",
        "The explicit post-V2 dependency seam does not imply its Morph dependency is satisfied.",
        "No concrete MorphCandidateProvider is reconstructed.",
        "No historical Morph action table is used as a runtime provider.",
        "No full B12 materialization is executed.",
        "No fresh output is frozen.",
        "Historical structural oracle remains closed.",
        "No structural fidelity claim is made.",
        "No performance evidence is generated.",
        "No CURRENT_BEST replacement occurs.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-E artifact exists but differs "
        "from exact B12 readiness state."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    OUT.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 17. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-E PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "IMPLEMENTATION:",
    flush=True,
)

print(
    "  B12 materializer:",
    "PRESENT",
    flush=True,
)

print(
    "  post-V2 provider code:",
    "PRESENT",
    flush=True,
)

print(
    "  Morph dependency seam:",
    "PRESENT",
    flush=True,
)

print(
    "  concrete Morph provider:",
    "UNAVAILABLE",
    flush=True,
)

print(
    flush=True,
)


print(
    "EXECUTION READINESS:",
    flush=True,
)

print(
    "  post-V2 fresh execution:",
    "BLOCKED",
    flush=True,
)

print(
    "  B12 full fresh execution:",
    "BLOCKED",
    flush=True,
)

print(
    "  one-shot full execution:",
    "BLOCKED",
    flush=True,
)

print(
    "  blocker-aware one-shot preflight:",
    "AUTHORIZED",
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIMS:",
    flush=True,
)

print(
    "  historical oracle access:",
    False,
    flush=True,
)

print(
    "  structural fidelity verified:",
    False,
    flush=True,
)

print(
    "  scientific result:",
    False,
    flush=True,
)

print(
    "  performance result:",
    False,
    flush=True,
)

print(
    "  CURRENT_BEST replaced:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    readiness_status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.3E-F
# B12 BLOCKER-AWARE ONE-SHOT PREFLIGHT HARNESS
#
# PARENT
# ------
# 14.3E-E PASS
#
# Parent state:
#
#   B12 materializer source       = PRESENT
#   post-V2 provider source       = PRESENT
#   Morph dependency seam         = PRESENT
#   concrete Morph provider       = UNAVAILABLE
#
#   full fresh execution          = NOT_AUTHORIZED
#   structural fidelity           = NOT_VERIFIED
#   historical oracle             = CLOSED
#
#
# PURPOSE
# -------
# Implement the ONE-SHOT PREFLIGHT HARNESS only.
#
# It must deterministically:
#
#   1. lock parent provenance
#   2. lock current source
#   3. verify B12/post-V2/Morph dependency seams
#   4. evaluate authority/readiness gates
#   5. detect finalized Morph authority blocker
#   6. return:
#
#          BLOCKED_BY_MORPH_PROVIDER
#
# BEFORE:
#
#   - B12 materialization
#   - post-V2 execution
#   - Morph inference
#   - fresh output freeze
#   - historical oracle access
#   - structural comparison
#   - GT
#   - evaluator
#   - performance
#
#
# IMPORTANT
# ---------
# Formal stage decision:
#
#   PASS
#
# means:
#
#   "the preflight harness correctly failed closed"
#
# It DOES NOT mean:
#
#   "B12 full execution passed"
#
#
# NO:
#   - source mutation
#   - test mutation
#   - provider implementation
#   - runtime materialization
#   - historical pickle loading
#   - historical structural oracle opening
#   - GT
#   - evaluator
#   - performance
#   - parameter search
#   - Git operations
# =============================================================================

import ast
import hashlib
import json

from pathlib import Path


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_e_"
    "b12_portability_readiness_reclassification_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


B12_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_b12_portable.py"
)

B12 = ROOT / B12_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_3e_f_"
    "b12_blocker_aware_one_shot_preflight_harness_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Exact frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "d68576e370d698a5fb07a72d7bfdbda5bb2709cbc1549a1447c62666a53e9956"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "7585e2f2f7afe67b91012f9d8cbed19615eeba00754b2a932ea814695fddcbab"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_B12_SHA = (
    "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
)


EXPECTED_PARENT_STATUS = (
    "B12_CODE_AND_SEAMS_READY_"
    "FULL_FRESH_EXECUTION_BLOCKED_BY_FINALIZED_MORPH_AUTHORITY_GAP"
)


EXPECTED_BLOCKER = (
    "CONCRETE_MORPH_PROVIDER_MISSING_FRESH_SOURCE_AUTHORITY"
)


EXPECTED_CONTEXT_SHA = (
    "c125127bb541a016a9fa78eee80f713eefb261983889ca5cfbce7489923e12d3"
)

EXPECTED_MORPH_VALIDATOR_SHA = (
    "1d8b4dc334d9586d12c94298c68f6f4b5197299a353a9fe13185b2363e88c035"
)

EXPECTED_MORPH_APPLY_SHA = (
    "db173c58a83eddd282712915514c31739815195ccb99edaa84a3ff5e16f1ffcf"
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(value):
    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(value):
    return sha256_bytes(
        value.encode("utf-8")
    )


def sha256_file(path):
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value):
    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_json(path):
    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def node_source(source, node):

    segment = ast.get_source_segment(
        source,
        node,
    )

    if segment is not None:
        return segment

    return ast.unparse(
        node
    )


def dotted_name(node):

    if isinstance(
        node,
        ast.Name,
    ):
        return node.id

    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )

        if left:
            return (
                f"{left}.{node.attr}"
            )

        return node.attr

    return None


def verify_artifact(
    path,
    expected_file_sha,
    expected_content_sha,
):

    assert path.exists(), path

    actual_file_sha = sha256_file(
        path
    )

    assert (
        actual_file_sha
        == expected_file_sha
    ), (
        f"{path.name}: file SHA drift.\n"
        f"expected={expected_file_sha}\n"
        f"actual={actual_file_sha}"
    )

    obj = load_json(
        path
    )

    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )

    assert (
        obj.get(
            "content_sha256"
        )
        == expected_content_sha
    )

    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )

    assert (
        canonical_json_sha(
            payload
        )
        == expected_content_sha
    )

    return obj


# =============================================================================
# 3. Static AST helpers
# =============================================================================

def top_level_unit(
    source,
    name,
):

    tree = ast.parse(
        source
    )

    matches = []

    for node in tree.body:

        if not isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):
            continue

        if node.name != name:
            continue

        segment = node_source(
            source,
            node,
        )

        matches.append({
            "kind":
                type(
                    node
                ).__name__,

            "name":
                node.name,

            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "sha256":
                sha256_text(
                    segment
                ),

            "source":
                segment,
        })

    return matches


def symbol_occurrence_count(
    source,
    symbol,
):

    tree = ast.parse(
        source
    )

    count = 0

    for node in ast.walk(
        tree
    ):

        if (
            isinstance(
                node,
                ast.Name,
            )
            and
            node.id == symbol
        ):
            count += 1

        elif (
            isinstance(
                node,
                ast.Attribute,
            )
            and
            node.attr == symbol
        ):
            count += 1

    return count


def call_records_containing(
    source,
    token,
):

    tree = ast.parse(
        source
    )

    records = []

    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            ast.Call,
        ):
            continue

        segment = node_source(
            source,
            node,
        )

        if token not in segment:
            continue

        records.append({
            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "callee":
                dotted_name(
                    node.func
                ),

            "sha256":
                sha256_text(
                    segment
                ),

            "source":
                segment[:4000],
        })

    return records


def enclosing_top_level_units(
    source,
    token,
):

    tree = ast.parse(
        source
    )

    records = []

    for node in tree.body:

        if not isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):
            continue

        segment = node_source(
            source,
            node,
        )

        if token not in segment:
            continue

        records.append({
            "kind":
                type(
                    node
                ).__name__,

            "name":
                node.name,

            "line_start":
                node.lineno,

            "line_end":
                getattr(
                    node,
                    "end_lineno",
                    node.lineno,
                ),

            "sha256":
                sha256_text(
                    segment
                ),
        })

    return records


# =============================================================================
# 4. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    B12,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.3E-F "
    "B12 BLOCKER-AWARE ONE-SHOT PREFLIGHT HARNESS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Parent readiness lock
# =============================================================================

parent = verify_artifact(
    PARENT,
    EXPECTED_PARENT_FILE_SHA,
    EXPECTED_PARENT_CONTENT_SHA,
)


resolution = parent[
    "resolution"
]


assert (
    resolution[
        "status"
    ]
    == EXPECTED_PARENT_STATUS
)


assert (
    resolution[
        "implementation_state"
    ]
    == "IMPLEMENTATION_PRESENT"
)


assert (
    resolution[
        "execution_state"
    ]
    == "NOT_AUTHORIZED"
)


assert (
    resolution[
        "blocker"
    ]
    == EXPECTED_BLOCKER
)


assert (
    resolution[
        "blocker_aware_preflight_authorized"
    ]
    is True
)


assert (
    resolution[
        "full_end_to_end_execution_authorized"
    ]
    is False
)


assert (
    resolution[
        "historical_oracle_access_authorized"
    ]
    is False
)


assert (
    resolution[
        "fresh_output_freeze_possible"
    ]
    is False
)


assert (
    resolution[
        "structural_comparison_possible"
    ]
    is False
)


assert (
    resolution[
        "CURRENT_BEST_replaced"
    ]
    is False
)


print(
    "[PASS] E_PARENT_READINESS_LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  execution state:",
    resolution[
        "execution_state"
    ],
    flush=True,
)

print(
    "  blocker:",
    resolution[
        "blocker"
    ],
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Source locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)

b12_sha_before = sha256_file(
    B12
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    b12_sha_before
    == EXPECTED_B12_SHA
)


portable_source = PORTABLE.read_text(
    encoding="utf-8"
)

b12_source = B12.read_text(
    encoding="utf-8"
)


print(
    "[PASS] CURRENT_SOURCE_LOCK",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  B12 SHA:",
    b12_sha_before,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 7. Exact Morph seam source locks
# =============================================================================

context_matches = top_level_unit(
    portable_source,
    "PortableCurrentBestContext",
)


validator_matches = top_level_unit(
    portable_source,
    "validate_morph_candidate_provider",
)


apply_matches = top_level_unit(
    portable_source,
    "apply_portable_morph_div_v2",
)


assert (
    len(
        context_matches
    )
    == 1
)


assert (
    len(
        validator_matches
    )
    == 1
)


assert (
    len(
        apply_matches
    )
    == 1
)


assert (
    context_matches[
        0
    ][
        "sha256"
    ]
    == EXPECTED_CONTEXT_SHA
)


assert (
    validator_matches[
        0
    ][
        "sha256"
    ]
    == EXPECTED_MORPH_VALIDATOR_SHA
)


assert (
    apply_matches[
        0
    ][
        "sha256"
    ]
    == EXPECTED_MORPH_APPLY_SHA
)


assert (
    "morph_candidate_provider"
    in context_matches[
        0
    ][
        "source"
    ]
)


assert (
    "morph_candidate_provider"
    in apply_matches[
        0
    ][
        "source"
    ]
)


print(
    "[PASS] MORPH_DEPENDENCY_SEAM_EXACT_SOURCE_LOCK",
    flush=True,
)

print(
    "  PortableCurrentBestContext SHA:",
    EXPECTED_CONTEXT_SHA,
    flush=True,
)

print(
    "  validator SHA:",
    EXPECTED_MORPH_VALIDATOR_SHA,
    flush=True,
)

print(
    "  apply Morph SHA:",
    EXPECTED_MORPH_APPLY_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 8. B12 dependency seam inspection
# =============================================================================

suppressed_peak_occurrences = (
    symbol_occurrence_count(
        b12_source,
        "suppressed_peak_provider",
    )
)


post_v2_occurrences = (
    symbol_occurrence_count(
        b12_source,
        "post_v2_graph_provider",
    )
)


post_v2_calls = (
    call_records_containing(
        b12_source,
        "post_v2_graph_provider",
    )
)


post_v2_enclosing_units = (
    enclosing_top_level_units(
        b12_source,
        "post_v2_graph_provider",
    )
)


assert (
    suppressed_peak_occurrences
    > 0
)


assert (
    post_v2_occurrences
    > 0
)


assert (
    len(
        post_v2_calls
    )
    > 0
)


assert (
    len(
        post_v2_enclosing_units
    )
    > 0
)


print(
    "=" * 120,
    flush=True,
)

print(
    "ONE-SHOT STATIC DEPENDENCY PREFLIGHT",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "suppressed_peak_provider occurrences:",
    suppressed_peak_occurrences,
    flush=True,
)

print(
    "post_v2_graph_provider occurrences:",
    post_v2_occurrences,
    flush=True,
)

print(
    "post_v2_graph_provider call sites:",
    len(
        post_v2_calls
    ),
    flush=True,
)

print(
    "post_v2 enclosing units:",
    post_v2_enclosing_units,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 9. One-shot ordered preflight
# =============================================================================

#
# The preflight trace is deliberately ordered.
#
# No runtime-execution phase may appear after the blocker is detected.
#

preflight_trace = []


def record_gate(
    gate,
    passed,
    detail,
):

    preflight_trace.append({
        "ordinal":
            len(
                preflight_trace
            )
            + 1,

        "gate":
            gate,

        "passed":
            bool(
                passed
            ),

        "detail":
            detail,
    })


# -------------------------------------------------------------------------
# Gate 1 — formal parent
# -------------------------------------------------------------------------

record_gate(
    "PARENT_PROVENANCE_LOCK",
    True,
    (
        "14.3E-E exact file/content SHA locked; "
        "preflight execution explicitly authorized."
    ),
)


# -------------------------------------------------------------------------
# Gate 2 — source lock
# -------------------------------------------------------------------------

record_gate(
    "CURRENT_SOURCE_LOCK",
    (
        portable_sha_before
        == EXPECTED_PORTABLE_SHA
        and
        b12_sha_before
        == EXPECTED_B12_SHA
    ),
    {
        "portable_sha256":
            portable_sha_before,

        "B12_sha256":
            b12_sha_before,
    },
)


# -------------------------------------------------------------------------
# Gate 3 — dependency seams
# -------------------------------------------------------------------------

seams_present = all([
    "morph_candidate_provider"
    in portable_source,

    "post_v2_graph_provider"
    in b12_source,

    "suppressed_peak_provider"
    in b12_source,

    len(
        post_v2_calls
    )
    > 0,
])


record_gate(
    "DEPENDENCY_SEAMS_PRESENT",
    seams_present,
    {
        "Morph_seam":
            True,

        "post_v2_seam":
            True,

        "suppressed_peak_seam":
            True,

        "post_v2_executable_call_site":
            True,
    },
)


assert (
    seams_present
    is True
)


# -------------------------------------------------------------------------
# Gate 4 — authority/readiness
# -------------------------------------------------------------------------

concrete_morph_provider_authorized = (
    False
)


record_gate(
    "CONCRETE_MORPH_PROVIDER_AUTHORITY",
    concrete_morph_provider_authorized,
    {
        "required":
            True,

        "available":
            False,

        "reason":
            EXPECTED_BLOCKER,
    },
)


# =============================================================================
# 10. Fail-closed decision
# =============================================================================

#
# This is the required current-state outcome.
#
# We do NOT continue to execution after this point.
#

if not concrete_morph_provider_authorized:

    preflight_result = (
        "BLOCKED_BY_MORPH_PROVIDER"
    )

    stop_phase = (
        "PRE_RUNTIME_AUTHORITY_GATE"
    )

    stop_reason = (
        "CONCRETE_MORPH_PROVIDER_MISSING_FRESH_SOURCE_AUTHORITY"
    )

else:

    # This branch is deliberately impossible under the locked E parent.
    raise AssertionError(
        "Unexpected Morph authority state: "
        "full execution must not become reachable in 14.3E-F."
    )


assert (
    preflight_result
    == "BLOCKED_BY_MORPH_PROVIDER"
)


assert (
    stop_reason
    == EXPECTED_BLOCKER
)


# =============================================================================
# 11. Explicit non-executed phase ledger
# =============================================================================

phase_ledger = {
    "preflight_parent_lock":
        "EXECUTED",

    "preflight_source_lock":
        "EXECUTED",

    "preflight_dependency_seam_check":
        "EXECUTED",

    "preflight_Morph_authority_check":
        "EXECUTED",

    "runtime_input_resolution":
        "NOT_EXECUTED",

    "Morph_candidate_generation":
        "NOT_EXECUTED",

    "post_v2_graph_execution":
        "NOT_EXECUTED",

    "B12_materialization":
        "NOT_EXECUTED",

    "fresh_output_freeze":
        "NOT_EXECUTED",

    "historical_structural_oracle_open":
        "NOT_EXECUTED",

    "structural_comparison":
        "NOT_EXECUTED",

    "GT_access":
        "NOT_EXECUTED",

    "evaluator":
        "NOT_EXECUTED",

    "performance_computation":
        "NOT_EXECUTED",
}


for phase in (
    "runtime_input_resolution",
    "Morph_candidate_generation",
    "post_v2_graph_execution",
    "B12_materialization",
    "fresh_output_freeze",
    "historical_structural_oracle_open",
    "structural_comparison",
    "GT_access",
    "evaluator",
    "performance_computation",
):

    assert (
        phase_ledger[
            phase
        ]
        == "NOT_EXECUTED"
    )


# =============================================================================
# 12. Harness acceptance semantics
# =============================================================================

harness_decision = (
    "PASS"
)


harness_acceptance = (
    "EXPECTED_BLOCKED_STATE_OBSERVED"
)


full_end_to_end_result = (
    "NOT_RUN"
)


structural_fidelity_result = (
    "NOT_EVALUATED"
)


scientific_result = (
    "NONE"
)


performance_result = (
    "NONE"
)


current_best_replaced = (
    False
)


historical_oracle_access_authorized = (
    False
)


fresh_output_frozen = (
    False
)


runtime_materialization_executed = (
    False
)


assert (
    harness_decision
    == "PASS"
)


assert (
    full_end_to_end_result
    == "NOT_RUN"
)


assert (
    structural_fidelity_result
    == "NOT_EVALUATED"
)


# =============================================================================
# 13. Print formal one-shot result
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "BLOCKER-AWARE ONE-SHOT PREFLIGHT RESULT",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for record in preflight_trace:

    print(
        f"{record['ordinal']}. "
        f"{record['gate']}",
        flush=True,
    )

    print(
        "   passed:",
        record[
            "passed"
        ],
        flush=True,
    )

    print(
        "   detail:",
        record[
            "detail"
        ],
        flush=True,
    )


print(
    flush=True,
)


print(
    "PREFLIGHT RESULT:",
    preflight_result,
    flush=True,
)

print(
    "STOP PHASE:",
    stop_phase,
    flush=True,
)

print(
    "STOP REASON:",
    stop_reason,
    flush=True,
)

print(
    flush=True,
)


print(
    "FULL END-TO-END RESULT:",
    full_end_to_end_result,
    flush=True,
)

print(
    "STRUCTURAL FIDELITY RESULT:",
    structural_fidelity_result,
    flush=True,
)

print(
    "FRESH OUTPUT FROZEN:",
    fresh_output_frozen,
    flush=True,
)

print(
    "HISTORICAL ORACLE ACCESS:",
    historical_oracle_access_authorized,
    flush=True,
)

print(
    "CURRENT_BEST REPLACED:",
    current_best_replaced,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 14. Safety
# =============================================================================

source_mutation = False

test_mutation = False

Git_operation_executed = False

historical_pickle_loaded = False

historical_Morph_action_runtime_use = False

historical_oracle_opened = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    sha256_file(
        B12
    )
    == b12_sha_before
    == EXPECTED_B12_SHA
)


# =============================================================================
# 15. Formal semantic gates
# =============================================================================

semantic_gates = {
    "E_parent_exactly_locked":
        True,

    "portable_source_exactly_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "B12_source_exactly_locked":
        (
            b12_sha_before
            == EXPECTED_B12_SHA
        ),

    "PortableCurrentBestContext_exactly_locked":
        (
            context_matches[
                0
            ][
                "sha256"
            ]
            == EXPECTED_CONTEXT_SHA
        ),

    "Morph_validator_exactly_locked":
        (
            validator_matches[
                0
            ][
                "sha256"
            ]
            == EXPECTED_MORPH_VALIDATOR_SHA
        ),

    "Morph_apply_runtime_exactly_locked":
        (
            apply_matches[
                0
            ][
                "sha256"
            ]
            == EXPECTED_MORPH_APPLY_SHA
        ),

    "B12_dependency_seams_present":
        seams_present,

    "preflight_is_authorized":
        (
            resolution[
                "blocker_aware_preflight_authorized"
            ]
            is True
        ),

    "full_execution_is_not_authorized":
        (
            resolution[
                "full_end_to_end_execution_authorized"
            ]
            is False
        ),

    "Morph_authority_gate_failed_closed":
        (
            concrete_morph_provider_authorized
            is False
        ),

    "expected_blocker_result_observed":
        (
            preflight_result
            == "BLOCKED_BY_MORPH_PROVIDER"
        ),

    "stop_occurred_before_runtime":
        (
            stop_phase
            == "PRE_RUNTIME_AUTHORITY_GATE"
        ),

    "runtime_input_resolution_not_executed":
        (
            phase_ledger[
                "runtime_input_resolution"
            ]
            == "NOT_EXECUTED"
        ),

    "Morph_generation_not_executed":
        (
            phase_ledger[
                "Morph_candidate_generation"
            ]
            == "NOT_EXECUTED"
        ),

    "post_v2_execution_not_executed":
        (
            phase_ledger[
                "post_v2_graph_execution"
            ]
            == "NOT_EXECUTED"
        ),

    "B12_materialization_not_executed":
        (
            phase_ledger[
                "B12_materialization"
            ]
            == "NOT_EXECUTED"
        ),

    "fresh_output_not_frozen":
        (
            fresh_output_frozen
            is False
        ),

    "historical_oracle_remains_closed":
        (
            historical_oracle_access_authorized
            is False
            and
            historical_oracle_opened
            is False
        ),

    "structural_fidelity_not_evaluated":
        (
            structural_fidelity_result
            == "NOT_EVALUATED"
        ),

    "CURRENT_BEST_not_replaced":
        (
            current_best_replaced
            is False
        ),

    "formal_PASS_means_correct_blocked_behavior_only":
        (
            harness_acceptance
            == "EXPECTED_BLOCKED_STATE_OBSERVED"
            and
            full_end_to_end_result
            == "NOT_RUN"
        ),

    "historical_Morph_actions_not_used_as_runtime_provider":
        (
            historical_Morph_action_runtime_use
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_metric_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_Git_operations":
        (
            Git_operation_executed
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.3E-F SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates == []
), (
    "14.3E-F failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 16. Formal artifact
# =============================================================================

next_stage = (
    "NOTEBOOK14_14_3E_G_"
    "B12_BLOCKED_STATE_ACCEPTANCE_AND_WORKSTREAM_CLOSURE"
)


payload = {
    "stage":
        "NOTEBOOK14_14_3E_F",

    "title":
        "B12_BLOCKER_AWARE_ONE_SHOT_PREFLIGHT_HARNESS",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        (
            "ENGINEERING_ONE_SHOT_PREFLIGHT_"
            "EXPECTED_BLOCKED_STATE_ACCEPTED"
        ),

    "parent_14_3e_e": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable_current_best": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "sha256":
                portable_sha_before,
        },

        "B12_portable": {
            "path":
                str(
                    B12_REL
                ),

            "sha256":
                b12_sha_before,
        },

        "PortableCurrentBestContext":
            context_matches[
                0
            ][
                "sha256"
            ],

        "validate_morph_candidate_provider":
            validator_matches[
                0
            ][
                "sha256"
            ],

        "apply_portable_morph_div_v2":
            apply_matches[
                0
            ][
                "sha256"
            ],
    },

    "static_dependency_preflight": {
        "suppressed_peak_provider_occurrences":
            suppressed_peak_occurrences,

        "post_v2_graph_provider_occurrences":
            post_v2_occurrences,

        "post_v2_call_sites":
            post_v2_calls,

        "post_v2_enclosing_units":
            post_v2_enclosing_units,

        "Morph_dependency_seam_present":
            True,
    },

    "preflight_trace":
        preflight_trace,

    "preflight_outcome": {
        "result":
            preflight_result,

        "stop_phase":
            stop_phase,

        "stop_reason":
            stop_reason,

        "harness_decision":
            harness_decision,

        "harness_acceptance":
            harness_acceptance,
    },

    "phase_ledger":
        phase_ledger,

    "execution_results": {
        "full_end_to_end":
            full_end_to_end_result,

        "structural_fidelity":
            structural_fidelity_result,

        "fresh_output_frozen":
            fresh_output_frozen,

        "historical_oracle_access_authorized":
            historical_oracle_access_authorized,

        "scientific_result":
            scientific_result,

        "performance_result":
            performance_result,

        "CURRENT_BEST_replaced":
            current_best_replaced,
    },

    "resolution": {
        "status":
            (
                "ONE_SHOT_PREFLIGHT_FAIL_CLOSED_"
                "BLOCKED_BY_MORPH_PROVIDER"
            ),

        "preflight_result":
            preflight_result,

        "full_end_to_end_execution":
            "NOT_RUN",

        "fresh_output_freeze":
            "NOT_PERFORMED",

        "historical_oracle_access":
            "CLOSED",

        "structural_fidelity":
            "NOT_EVALUATED",

        "CURRENT_BEST_replaced":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "formal_stage_PASS":
            (
                "HARNESS_BEHAVIOR_PASS_ONLY"
            ),

        "B12_execution_result":
            "BLOCKED_BY_MORPH_PROVIDER",

        "B12_full_execution":
            "NOT_RUN",

        "post_V2_execution":
            "NOT_RUN",

        "Morph_generation":
            "NOT_RUN",

        "fresh_output_frozen":
            False,

        "historical_structural_oracle":
            "CLOSED",

        "structural_fidelity_verified":
            False,

        "scientific_result":
            "NONE",

        "performance_result":
            "NONE",

        "CURRENT_BEST_replaced":
            False,

        "historical_pickle_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "parameter_search":
            False,

        "Git_operation":
            False,

        "source_mutation":
            False,

        "test_mutation":
            False,
    },

    "explicit_non_claims": [
        "Formal PASS means the preflight correctly blocked execution; it does not mean B12 executed successfully.",
        "No Morph candidate generation was executed.",
        "No post-V2 graph runtime was executed.",
        "No B12 materialization was executed.",
        "No fresh output was frozen.",
        "No historical structural oracle was opened.",
        "No structural fidelity comparison was performed.",
        "No GT was accessed.",
        "No evaluator was executed.",
        "No performance evidence was generated.",
        "No CURRENT_BEST replacement occurred.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.3E-F artifact exists but differs "
        "from exact blocker-aware preflight state."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    OUT.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 17. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.3E-F PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "HARNESS:",
    flush=True,
)

print(
    "  formal decision:",
    harness_decision,
    flush=True,
)

print(
    "  acceptance:",
    harness_acceptance,
    flush=True,
)

print(
    flush=True,
)


print(
    "ONE-SHOT PREFLIGHT:",
    flush=True,
)

print(
    "  result:",
    preflight_result,
    flush=True,
)

print(
    "  stopped at:",
    stop_phase,
    flush=True,
)

print(
    "  reason:",
    stop_reason,
    flush=True,
)

print(
    flush=True,
)


print(
    "EXECUTION:",
    flush=True,
)

print(
    "  Morph generation:",
    "NOT_RUN",
    flush=True,
)

print(
    "  post-V2:",
    "NOT_RUN",
    flush=True,
)

print(
    "  B12 materialization:",
    "NOT_RUN",
    flush=True,
)

print(
    "  fresh output frozen:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "CLAIMS:",
    flush=True,
)

print(
    "  historical oracle opened:",
    False,
    flush=True,
)

print(
    "  structural fidelity evaluated:",
    False,
    flush=True,
)

print(
    "  performance evidence:",
    False,
    flush=True,
)

print(
    "  CURRENT_BEST replaced:",
    False,
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    (
        "ONE_SHOT_PREFLIGHT_FAIL_CLOSED_"
        "BLOCKED_BY_MORPH_PROVIDER"
    ),
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.4
# NON-MORPH ENGINEERING CONTINUATION BOUNDARY
#
# PARENT
# ------
# 14.3E-G PASS
#
# Parent closed:
#
#   Morph source-authority recovery
#   concrete Morph provider implementation
#   fresh post-V2 execution
#   full B12 fresh materialization
#   current one-shot execution workstream
#
# Parent preserved:
#
#   portable source
#   B12 source
#   frozen Morph selection/application specification
#   exact reference association provider
#   explicit dependency seams
#   non-Morph engineering continuation
#
#
# PURPOSE
# -------
# Freeze remaining Notebook14 engineering into:
#
#   OPEN
#       may continue now
#
#   CONDITIONAL
#       may be tested only in isolation / synthetic contract mode
#
#   CLOSED
#       may not execute without NEW_ATTRIBUTABLE_MORPH_SOURCE_AUTHORITY
#
#
# CRITICAL RULE
# -------------
# Synthetic/stub dependencies may be used only for isolated contract/unit tests.
#
# They may NOT:
#
#   - stand in for a real Morph provider in full B12 execution
#   - produce a "fresh B12" artifact
#   - authorize oracle access
#   - support structural-fidelity claims
#   - support scientific/performance claims
#
#
# NO:
#   - source mutation
#   - test mutation
#   - runtime materialization
#   - historical pickle loading
#   - historical oracle opening
#   - GT
#   - evaluator
#   - performance
#   - parameter search
#   - Git operations
# =============================================================================

import ast
import hashlib
import json

from pathlib import Path


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_3e_g_"
    "b12_blocked_state_acceptance_and_workstream_closure_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


B12_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_b12_portable.py"
)

B12 = ROOT / B12_REL


OUT_REL = Path(
    "reports/"
    "notebook14_14_4_"
    "non_morph_engineering_continuation_boundary_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Exact frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "1b1c0e68bc2500d7134ecdb396aa6abc281f72ce0835381c7630a66353d5026e"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "27021564756f7fdc4be608aaf4acd8ad6a07681e30c9c0dd79db9fa58b866a7d"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_B12_SHA = (
    "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
)


EXPECTED_PARENT_STATUS = (
    "B12_BLOCKED_STATE_ACCEPTED_"
    "PORTABILITY_ONE_SHOT_WORKSTREAM_CLOSED"
)


EXPECTED_REOPEN_TRIGGER = (
    "NEW_ATTRIBUTABLE_MORPH_SOURCE_AUTHORITY"
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(value):
    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_file(path):
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value):
    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_json(path):
    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def verify_artifact(
    path,
    expected_file_sha,
    expected_content_sha,
):

    assert path.exists(), path

    actual_file_sha = sha256_file(
        path
    )

    assert (
        actual_file_sha
        == expected_file_sha
    ), (
        f"{path.name}: file SHA drift.\n"
        f"expected={expected_file_sha}\n"
        f"actual={actual_file_sha}"
    )

    obj = load_json(
        path
    )

    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )

    assert (
        obj.get(
            "content_sha256"
        )
        == expected_content_sha
    )

    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )

    assert (
        canonical_json_sha(
            payload
        )
        == expected_content_sha
    )

    return obj


def source_syntax_ok(path):

    source = path.read_text(
        encoding="utf-8"
    )

    ast.parse(
        source,
        filename=str(path),
    )

    return True


# =============================================================================
# 3. Preflight
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    B12,
):
    assert required.exists(), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.4 "
    "NON-MORPH ENGINEERING CONTINUATION BOUNDARY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 4. Parent closure lock
# =============================================================================

parent = verify_artifact(
    PARENT,
    EXPECTED_PARENT_FILE_SHA,
    EXPECTED_PARENT_CONTENT_SHA,
)


resolution = parent[
    "resolution"
]


assert (
    resolution[
        "status"
    ]
    == EXPECTED_PARENT_STATUS
)


assert (
    resolution[
        "current_portability_one_shot_workstream_closed"
    ]
    is True
)


assert (
    resolution[
        "Morph_source_authority_recovery_closed"
    ]
    is True
)


assert (
    resolution[
        "concrete_Morph_provider_implementation_authorized"
    ]
    is False
)


assert (
    resolution[
        "fresh_post_V2_runtime_ready"
    ]
    is False
)


assert (
    resolution[
        "full_B12_fresh_runtime_ready"
    ]
    is False
)


assert (
    resolution[
        "historical_oracle_access_authorized"
    ]
    is False
)


assert (
    resolution[
        "structural_fidelity_verified"
    ]
    is False
)


assert (
    resolution[
        "CURRENT_BEST_replaced"
    ]
    is False
)


assert (
    resolution[
        "non_Morph_engineering_may_continue"
    ]
    is True
)


reopen_contract = parent[
    "reopen_contract"
]


assert (
    reopen_contract[
        "trigger"
    ]
    == EXPECTED_REOPEN_TRIGGER
)


assert (
    reopen_contract[
        "historical_oracle_alone_can_reopen"
    ]
    is False
)


assert (
    reopen_contract[
        "historical_action_similarity_can_reopen"
    ]
    is False
)


assert (
    reopen_contract[
        "engineering_convenience_can_reopen"
    ]
    is False
)


print(
    "[PASS] 14.3E-G CLOSURE LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    "  status:",
    EXPECTED_PARENT_STATUS,
    flush=True,
)

print(
    "  reopen trigger:",
    EXPECTED_REOPEN_TRIGGER,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Current source locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)

b12_sha_before = sha256_file(
    B12
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    b12_sha_before
    == EXPECTED_B12_SHA
)


assert (
    source_syntax_ok(
        PORTABLE
    )
    is True
)


assert (
    source_syntax_ok(
        B12
    )
    is True
)


portable_source = PORTABLE.read_text(
    encoding="utf-8"
)

b12_source = B12.read_text(
    encoding="utf-8"
)


required_portable_seams = {
    "morph_candidate_provider":
        (
            "morph_candidate_provider"
            in portable_source
        ),

    "apply_portable_morph_div_v2":
        (
            "apply_portable_morph_div_v2"
            in portable_source
        ),
}


required_b12_seams = {
    "suppressed_peak_provider":
        (
            "suppressed_peak_provider"
            in b12_source
        ),

    "post_v2_graph_provider":
        (
            "post_v2_graph_provider"
            in b12_source
        ),

    "materialize_b12_dataset":
        (
            "materialize_b12_dataset"
            in b12_source
        ),
}


assert all(
    required_portable_seams.values()
)


assert all(
    required_b12_seams.values()
)


print(
    "[PASS] CURRENT SOURCE / SEAM LOCK",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  B12 SHA:",
    b12_sha_before,
    flush=True,
)

print(
    "  portable syntax:",
    "PASS",
    flush=True,
)

print(
    "  B12 syntax:",
    "PASS",
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Continuation boundary
# =============================================================================

open_work = {
    "SOURCE_AND_ARTIFACT_INTEGRITY": {
        "status":
            "OPEN",

        "allowed": [
            "source SHA regression checks",
            "artifact SHA/content self-consistency checks",
            "AST/static source inspection",
            "import/syntax validation",
        ],
    },

    "NON_MORPH_COMPONENT_REGRESSION": {
        "status":
            "OPEN",

        "allowed": [
            "P995 engineering regression checks",
            "suppressed-peak provider regression checks",
            "already-source-authorized V1/V2 component regression checks",
            "schema/order/determinism regression checks",
        ],

        "restriction":
            (
                "Must not transitively invoke the blocked Morph-dependent "
                "full post-V2 path."
            ),
    },

    "DEPENDENCY_SEAM_VALIDATION": {
        "status":
            "OPEN",

        "allowed": [
            "Morph dependency-seam presence checks",
            "post-V2 dependency-seam checks",
            "B12 provider-injection contract checks",
            "missing-provider fail-closed checks",
        ],
    },

    "BLOCKER_AWARE_PREFLIGHT_REGRESSION": {
        "status":
            "OPEN",

        "required_result_under_current_authority":
            "BLOCKED_BY_MORPH_PROVIDER",
    },

    "TEST_AND_HARNESS_AUDIT": {
        "status":
            "OPEN",

        "allowed": [
            "discover existing tests",
            "classify tests by dependency boundary",
            "run tests that cannot execute blocked Morph/full-B12 paths",
            "verify explicit failure behavior",
        ],

        "source_mutation_authorized":
            False,
    },
}


conditional_work = {
    "SYNTHETIC_PROVIDER_CONTRACT_TESTS": {
        "status":
            "CONDITIONAL",

        "allowed":
            (
                "Synthetic/stub providers may be used only inside isolated "
                "unit/contract tests."
            ),

        "forbidden": [
            "calling the result a fresh Morph provider",
            "calling the result fresh post-V2 output",
            "calling the result full B12 materialization",
            "freezing it as production fresh output",
            "opening historical oracle based on it",
            "making structural-fidelity claims",
            "making scientific/performance claims",
        ],
    },

    "PARTIAL_COMPONENT_EXECUTION": {
        "status":
            "CONDITIONAL",

        "allowed":
            (
                "A component may execute only when its complete dependency "
                "chain is independently source-authorized and the execution "
                "cannot cross into the blocked Morph path."
            ),

        "full_B12_equivalence_claim":
            False,
    },
}


closed_work = {
    "MORPH_SOURCE_AUTHORITY_ARCHAEOLOGY": {
        "status":
            "CLOSED",

        "reopen_only_if":
            EXPECTED_REOPEN_TRIGGER,
    },

    "MORPH_CONCRETE_PROVIDER_IMPLEMENTATION": {
        "status":
            "CLOSED",

        "reason":
            "MISSING_FRESH_SOURCE_AUTHORITY",
    },

    "FRESH_POST_V2_FULL_EXECUTION": {
        "status":
            "CLOSED",

        "reason":
            "BLOCKED_BY_MORPH_PROVIDER",
    },

    "FULL_B12_FRESH_MATERIALIZATION": {
        "status":
            "CLOSED",

        "reason":
            "BLOCKED_BY_POST_V2_MORPH_DEPENDENCY",
    },

    "FRESH_OUTPUT_FREEZE_FOR_FULL_B12": {
        "status":
            "CLOSED",

        "reason":
            "NO_AUTHORIZED_FULL_FRESH_EXECUTION",
    },

    "HISTORICAL_STRUCTURAL_ORACLE_ACCESS": {
        "status":
            "CLOSED",

        "reason":
            "NO_INDEPENDENT_FRESH_OUTPUT_FROZEN",
    },

    "FULL_B12_STRUCTURAL_FIDELITY_CLAIM": {
        "status":
            "CLOSED",
    },

    "PERFORMANCE_OR_SCIENTIFIC_PROMOTION": {
        "status":
            "CLOSED",

        "CURRENT_BEST_replacement_authorized":
            False,
    },
}


# =============================================================================
# 7. Explicit anti-bypass rules
# =============================================================================

anti_bypass_rules = [
    {
        "id":
            "NO_STUB_AS_PRODUCTION_MORPH_PROVIDER",

        "rule":
            (
                "A mock/synthetic/stub Morph provider may never satisfy the "
                "production fresh-runtime authority gate."
            ),
    },

    {
        "id":
            "NO_SKIP_MORPH_STAGE",

        "rule":
            (
                "Do not skip Morph in the full post-V2 sequence and call the "
                "result equivalent to the canonical B12 runtime."
            ),
    },

    {
        "id":
            "NO_HISTORICAL_ACTION_RUNTIME_PROVIDER",

        "rule":
            (
                "Historical Morph action/audit tables may not be injected "
                "as runtime provider output."
            ),
    },

    {
        "id":
            "NO_ORACLE_AS_SOURCE_AUTHORITY",

        "rule":
            (
                "Historical structural oracle cannot define missing "
                "Morph generator semantics."
            ),
    },

    {
        "id":
            "NO_PARTIAL_AS_FULL",

        "rule":
            (
                "Partial/component-level successful execution cannot be "
                "reported as full post-V2 or full B12 success."
            ),
    },

    {
        "id":
            "NO_ENGINEERING_PASS_AS_PERFORMANCE_PASS",

        "rule":
            (
                "Engineering regression/preflight PASS cannot be treated "
                "as scientific or performance evidence."
            ),
    },
]


# =============================================================================
# 8. Current engineering state
# =============================================================================

engineering_state = {
    "14_3E_workstream":
        "CLOSED",

    "portable_source":
        "PRESENT_LOCKED",

    "B12_source":
        "PRESENT_LOCKED",

    "dependency_seams":
        "PRESENT",

    "Morph_source_authority":
        "UNRESOLVED_RECOVERY_CLOSED",

    "concrete_Morph_provider":
        "NOT_AUTHORIZED",

    "fresh_post_V2_runtime":
        "BLOCKED",

    "full_B12_fresh_runtime":
        "BLOCKED",

    "one_shot_preflight":
        "PASS_FAIL_CLOSED",

    "historical_structural_oracle":
        "CLOSED",

    "structural_fidelity":
        "NOT_VERIFIED",

    "performance_result":
        "NONE",

    "scientific_result":
        "NONE",

    "CURRENT_BEST":
        "UNCHANGED",

    "non_Morph_engineering":
        "OPEN",
}


# =============================================================================
# 9. Next authorized work
# =============================================================================

next_stage = (
    "NOTEBOOK14_14_4A_"
    "NON_MORPH_STATIC_AND_REGRESSION_READINESS_AUDIT"
)


next_stage_scope = {
    "inspect_existing_tests":
        True,

    "classify_tests_by_dependency_boundary":
        True,

    "identify_safe_non_Morph_test_targets":
        True,

    "identify_tests_that_must_remain_blocked":
        True,

    "run_runtime_materialization":
        False,

    "modify_source":
        False,

    "modify_tests":
        False,

    "open_historical_oracle":
        False,

    "use_GT":
        False,

    "run_evaluator":
        False,
}


boundary_status = (
    "NON_MORPH_ENGINEERING_CONTINUATION_BOUNDARY_FROZEN"
)


print(
    "=" * 120,
    flush=True,
)

print(
    "14.4 CONTINUATION BOUNDARY",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "OPEN:",
    flush=True,
)

for name in open_work:
    print(
        " ",
        name,
        flush=True,
    )


print(
    flush=True,
)


print(
    "CONDITIONAL:",
    flush=True,
)

for name in conditional_work:
    print(
        " ",
        name,
        flush=True,
    )


print(
    flush=True,
)


print(
    "CLOSED:",
    flush=True,
)

for name in closed_work:
    print(
        " ",
        name,
        flush=True,
    )


print(
    flush=True,
)


print(
    "CURRENT_BEST:",
    "UNCHANGED",
    flush=True,
)

print(
    "historical oracle:",
    "CLOSED",
    flush=True,
)

print(
    "non-Morph engineering:",
    "OPEN",
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    boundary_status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 10. Safety
# =============================================================================

source_mutation = False

test_mutation = False

Git_operation_executed = False

runtime_materialization_executed = False

historical_pickle_loaded = False

historical_Morph_runtime_use = False

historical_oracle_opened = False

GT_loaded = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False

current_best_replaced = False


assert (
    sha256_file(
        PORTABLE
    )
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    sha256_file(
        B12
    )
    == b12_sha_before
    == EXPECTED_B12_SHA
)


# =============================================================================
# 11. Semantic gates
# =============================================================================

semantic_gates = {
    "14_3E_G_parent_exactly_locked":
        True,

    "portable_source_exactly_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "B12_source_exactly_locked":
        (
            b12_sha_before
            == EXPECTED_B12_SHA
        ),

    "portable_source_syntax_valid":
        source_syntax_ok(
            PORTABLE
        ),

    "B12_source_syntax_valid":
        source_syntax_ok(
            B12
        ),

    "required_dependency_seams_preserved":
        (
            all(
                required_portable_seams.values()
            )
            and
            all(
                required_b12_seams.values()
            )
        ),

    "non_Morph_engineering_explicitly_open":
        (
            engineering_state[
                "non_Morph_engineering"
            ]
            == "OPEN"
        ),

    "Morph_authority_recovery_remains_closed":
        (
            closed_work[
                "MORPH_SOURCE_AUTHORITY_ARCHAEOLOGY"
            ][
                "status"
            ]
            == "CLOSED"
        ),

    "Morph_provider_implementation_remains_closed":
        (
            closed_work[
                "MORPH_CONCRETE_PROVIDER_IMPLEMENTATION"
            ][
                "status"
            ]
            == "CLOSED"
        ),

    "fresh_post_V2_execution_remains_closed":
        (
            closed_work[
                "FRESH_POST_V2_FULL_EXECUTION"
            ][
                "status"
            ]
            == "CLOSED"
        ),

    "full_B12_materialization_remains_closed":
        (
            closed_work[
                "FULL_B12_FRESH_MATERIALIZATION"
            ][
                "status"
            ]
            == "CLOSED"
        ),

    "historical_oracle_remains_closed":
        (
            closed_work[
                "HISTORICAL_STRUCTURAL_ORACLE_ACCESS"
            ][
                "status"
            ]
            == "CLOSED"
            and
            historical_oracle_opened
            is False
        ),

    "synthetic_provider_cannot_bypass_authority":
        (
            conditional_work[
                "SYNTHETIC_PROVIDER_CONTRACT_TESTS"
            ][
                "status"
            ]
            == "CONDITIONAL"
        ),

    "partial_execution_cannot_be_called_full_B12":
        (
            conditional_work[
                "PARTIAL_COMPONENT_EXECUTION"
            ][
                "full_B12_equivalence_claim"
            ]
            is False
        ),

    "reopen_trigger_preserved":
        (
            closed_work[
                "MORPH_SOURCE_AUTHORITY_ARCHAEOLOGY"
            ][
                "reopen_only_if"
            ]
            == EXPECTED_REOPEN_TRIGGER
        ),

    "CURRENT_BEST_not_replaced":
        (
            current_best_replaced
            is False
        ),

    "runtime_materialization_not_executed":
        (
            runtime_materialization_executed
            is False
        ),

    "historical_Morph_not_used_at_runtime":
        (
            historical_Morph_runtime_use
            is False
        ),

    "historical_pickle_not_loaded":
        (
            historical_pickle_loaded
            is False
        ),

    "GT_not_loaded":
        (
            GT_loaded
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "no_Git_operations":
        (
            Git_operation_executed
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.4 SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


assert (
    failed_gates == []
), (
    "14.4 failed gates: "
    f"{failed_gates}"
)


# =============================================================================
# 12. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_4",

    "title":
        "NON_MORPH_ENGINEERING_CONTINUATION_BOUNDARY",

    "version":
        1,

    "decision":
        "PASS",

    "classification":
        (
            "ENGINEERING_CONTINUATION_BOUNDARY_"
            "MORPH_DEPENDENT_EXECUTION_CLOSED"
        ),

    "parent_14_3e_g": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable_current_best": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "sha256":
                portable_sha_before,
        },

        "B12_portable": {
            "path":
                str(
                    B12_REL
                ),

            "sha256":
                b12_sha_before,
        },
    },

    "required_dependency_seams": {
        "portable":
            required_portable_seams,

        "B12":
            required_b12_seams,
    },

    "continuation_boundary": {
        "OPEN":
            open_work,

        "CONDITIONAL":
            conditional_work,

        "CLOSED":
            closed_work,
    },

    "anti_bypass_rules":
        anti_bypass_rules,

    "engineering_state":
        engineering_state,

    "next_stage_scope":
        next_stage_scope,

    "resolution": {
        "status":
            boundary_status,

        "14_3E_workstream_closed":
            True,

        "non_Morph_engineering_authorized":
            True,

        "Morph_dependent_full_execution_authorized":
            False,

        "synthetic_provider_full_execution_authorized":
            False,

        "historical_oracle_access_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "scientific_result":
            "NONE",

        "performance_result":
            "NONE",

        "CURRENT_BEST_replaced":
            False,

        "reopen_trigger":
            EXPECTED_REOPEN_TRIGGER,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "14_3E":
            "CLOSED",

        "non_Morph_engineering":
            "OPEN",

        "Morph_source_authority_recovery":
            "CLOSED",

        "concrete_Morph_provider":
            "NOT_AUTHORIZED",

        "fresh_post_V2":
            "BLOCKED",

        "full_B12_fresh_materialization":
            "BLOCKED",

        "synthetic_provider_use":
            "ISOLATED_TESTS_ONLY",

        "historical_structural_oracle":
            "CLOSED",

        "structural_fidelity_verified":
            False,

        "scientific_result":
            "NONE",

        "performance_result":
            "NONE",

        "CURRENT_BEST_replaced":
            False,

        "historical_pickle_runtime_use":
            False,

        "historical_pickle_loaded":
            False,

        "GT_read":
            False,

        "evaluator_executed":
            False,

        "parameter_search":
            False,

        "Git_operation":
            False,

        "source_mutation":
            False,

        "test_mutation":
            False,
    },

    "explicit_non_claims": [
        "Opening non-Morph engineering does not reopen Morph authority recovery.",
        "Synthetic/stub providers cannot satisfy production Morph authority.",
        "Partial component execution cannot be reported as full B12 execution.",
        "Existing source code does not imply full execution readiness.",
        "No fresh post-V2 graph is executed here.",
        "No B12 materialization is executed here.",
        "No historical structural oracle is opened.",
        "No structural fidelity claim is made.",
        "No scientific or performance evidence is generated.",
        "No CURRENT_BEST replacement occurs.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.4 artifact exists but differs "
        "from exact continuation-boundary state."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    OUT.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 13. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 / 14.4 PASS",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "BOUNDARY:",
    flush=True,
)

print(
    "  14.3E workstream:",
    "CLOSED",
    flush=True,
)

print(
    "  non-Morph engineering:",
    "OPEN",
    flush=True,
)

print(
    "  Morph authority recovery:",
    "CLOSED",
    flush=True,
)

print(
    "  Morph-dependent full execution:",
    "CLOSED",
    flush=True,
)

print(
    flush=True,
)


print(
    "ALLOWED NOW:",
    flush=True,
)

print(
    "  static/source/artifact integrity:",
    True,
    flush=True,
)

print(
    "  non-Morph regression audit:",
    True,
    flush=True,
)

print(
    "  dependency-seam validation:",
    True,
    flush=True,
)

print(
    "  blocker-aware preflight regression:",
    True,
    flush=True,
)

print(
    "  isolated synthetic contract tests:",
    "CONDITIONAL",
    flush=True,
)

print(
    flush=True,
)


print(
    "STILL FORBIDDEN:",
    flush=True,
)

print(
    "  concrete Morph provider implementation:",
    True,
    flush=True,
)

print(
    "  full fresh post-V2 execution:",
    True,
    flush=True,
)

print(
    "  full fresh B12 materialization:",
    True,
    flush=True,
)

print(
    "  historical oracle opening:",
    True,
    flush=True,
)

print(
    "  structural fidelity claim:",
    True,
    flush=True,
)

print(
    flush=True,
)


print(
    "CURRENT_BEST:",
    "UNCHANGED",
    flush=True,
)

print(
    "REOPEN TRIGGER:",
    EXPECTED_REOPEN_TRIGGER,
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    boundary_status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)
```

```python
# =============================================================================
# Notebook 14
# 14.4A
# NON-MORPH STATIC AND REGRESSION READINESS AUDIT
#
# PARENT
# ------
# 14.4 PASS
#
# PURPOSE
# -------
# 1. Inventory existing pytest tests relevant to:
#
#       stage12_portable_current_best
#       stage12_b12_portable
#
# 2. Classify each test node as:
#
#       SAFE_TO_RUN
#       CONDITIONAL_SYNTHETIC
#       BLOCKED_BOUNDARY
#       REVIEW_REQUIRED
#
# 3. Run ONLY SAFE_TO_RUN node IDs.
#
#
# CONSERVATIVE RULE
# -----------------
# Uncertainty => do NOT run.
#
# Synthetic/stub provider tests are CONDITIONAL and are not run here.
#
# A test is not SAFE merely because it uses no GT. It must also avoid:
#
#   - Morph concrete-provider execution
#   - full post-V2 execution
#   - full B12 materialization
#   - historical Morph artifacts
#   - historical oracle
#   - evaluator/performance paths
#   - external development/train data
#   - subprocess/network/runtime side effects
#
#
# PYTEST
# ------
# Safe tests are executed with:
#
#   python -m pytest -vv -s -p no:cacheprovider <exact node ids>
#
# and:
#
#   PYTHONDONTWRITEBYTECODE=1
#
#
# NO:
#   - source mutation
#   - test mutation
#   - Git
#   - historical pickle loading by this harness
#   - historical oracle
#   - GT
#   - evaluator
#   - performance
#   - parameter search
# =============================================================================

import ast
import hashlib
import json
import os
import subprocess
import sys

from collections import Counter
from pathlib import Path


# =============================================================================
# 0. Paths
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()


PARENT_REL = Path(
    "reports/"
    "notebook14_14_4_"
    "non_morph_engineering_continuation_boundary_v1.json"
)

PARENT = ROOT / PARENT_REL


PORTABLE_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_portable_current_best.py"
)

PORTABLE = ROOT / PORTABLE_REL


B12_REL = Path(
    "src/biohub_cell_tracking/"
    "stage12_b12_portable.py"
)

B12 = ROOT / B12_REL


TEST_ROOT = (
    ROOT
    / "tests"
)


OUT_REL = Path(
    "reports/"
    "notebook14_14_4a_"
    "non_morph_static_and_regression_readiness_audit_v1.json"
)

OUT = ROOT / OUT_REL


# =============================================================================
# 1. Exact frozen identities
# =============================================================================

EXPECTED_PARENT_FILE_SHA = (
    "90b5f683ff3f6af9767ae4b6f0e2ab5181ec7772ebf87cf2ecdd4cb60dd4a638"
)

EXPECTED_PARENT_CONTENT_SHA = (
    "69747e0d090f800b9b8c1a2f24677aa023a18a32be48620e34c5f8c225da7232"
)


EXPECTED_PARENT_STATUS = (
    "NON_MORPH_ENGINEERING_CONTINUATION_BOUNDARY_FROZEN"
)


EXPECTED_PORTABLE_SHA = (
    "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
)


EXPECTED_B12_SHA = (
    "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
)


EXPECTED_REOPEN_TRIGGER = (
    "NEW_ATTRIBUTABLE_MORPH_SOURCE_AUTHORITY"
)


# =============================================================================
# 2. Helpers
# =============================================================================

def sha256_bytes(value):
    return hashlib.sha256(
        value
    ).hexdigest()


def sha256_text(value):
    return sha256_bytes(
        value.encode(
            "utf-8"
        )
    )


def sha256_file(path):
    return sha256_bytes(
        path.read_bytes()
    )


def canonical_json_sha(value):
    return sha256_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode(
            "utf-8"
        )
    )


def load_json(path):
    return json.loads(
        path.read_text(
            encoding="utf-8"
        )
    )


def verify_artifact(
    path,
    expected_file_sha,
    expected_content_sha,
):

    assert path.exists(), path

    actual_file_sha = sha256_file(
        path
    )

    assert (
        actual_file_sha
        == expected_file_sha
    ), (
        f"{path.name}: file SHA drift.\n"
        f"expected={expected_file_sha}\n"
        f"actual={actual_file_sha}"
    )

    obj = load_json(
        path
    )

    assert (
        obj.get(
            "decision"
        )
        == "PASS"
    )

    assert (
        obj.get(
            "content_sha256"
        )
        == expected_content_sha
    )

    payload = dict(
        obj
    )

    payload.pop(
        "content_sha256"
    )

    assert (
        canonical_json_sha(
            payload
        )
        == expected_content_sha
    )

    return obj


def node_source(
    source,
    node,
):

    value = ast.get_source_segment(
        source,
        node,
    )

    if value is not None:
        return value

    return ast.unparse(
        node
    )


def dotted_name(node):

    if isinstance(
        node,
        ast.Name,
    ):
        return node.id

    if isinstance(
        node,
        ast.Attribute,
    ):

        left = dotted_name(
            node.value
        )

        if left:
            return (
                f"{left}.{node.attr}"
            )

        return node.attr

    return None


# =============================================================================
# 3. Parent boundary lock
# =============================================================================

for required in (
    ROOT,
    PARENT,
    PORTABLE,
    B12,
    TEST_ROOT,
):

    assert (
        required.exists()
    ), required


print(
    "=" * 120,
    flush=True,
)

print(
    "NOTEBOOK 14 :: 14.4A "
    "NON-MORPH STATIC AND REGRESSION READINESS AUDIT",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "PARENT:",
    PARENT_REL,
    flush=True,
)

print(
    "OUTPUT:",
    OUT_REL,
    flush=True,
)

print(
    flush=True,
)


parent = verify_artifact(
    PARENT,
    EXPECTED_PARENT_FILE_SHA,
    EXPECTED_PARENT_CONTENT_SHA,
)


assert (
    parent[
        "resolution"
    ][
        "status"
    ]
    == EXPECTED_PARENT_STATUS
)


assert (
    parent[
        "resolution"
    ][
        "non_Morph_engineering_authorized"
    ]
    is True
)


assert (
    parent[
        "resolution"
    ][
        "Morph_dependent_full_execution_authorized"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "synthetic_provider_full_execution_authorized"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "historical_oracle_access_authorized"
    ]
    is False
)


assert (
    parent[
        "resolution"
    ][
        "reopen_trigger"
    ]
    == EXPECTED_REOPEN_TRIGGER
)


print(
    "[PASS] 14.4 PARENT BOUNDARY LOCK",
    flush=True,
)

print(
    "  file SHA:",
    EXPECTED_PARENT_FILE_SHA,
    flush=True,
)

print(
    "  content SHA:",
    EXPECTED_PARENT_CONTENT_SHA,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 4. Current source locks
# =============================================================================

portable_sha_before = sha256_file(
    PORTABLE
)


b12_sha_before = sha256_file(
    B12
)


assert (
    portable_sha_before
    == EXPECTED_PORTABLE_SHA
)


assert (
    b12_sha_before
    == EXPECTED_B12_SHA
)


print(
    "[PASS] CURRENT SOURCE LOCK",
    flush=True,
)

print(
    "  portable SHA:",
    portable_sha_before,
    flush=True,
)

print(
    "  B12 SHA:",
    b12_sha_before,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 5. Test-file inventory + immutable baseline
# =============================================================================

test_files = sorted(
    TEST_ROOT.rglob(
        "test_*.py"
    )
)


assert (
    len(
        test_files
    )
    > 0
)


test_sha_before = {
    str(
        path.relative_to(
            ROOT
        )
    ):
        sha256_file(
            path
        )

    for path in test_files
}


print(
    "test files discovered:",
    len(
        test_files
    ),
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 6. Classification vocabulary
# =============================================================================

#
# CLOSED production/runtime boundary.
#
BLOCKED_TOKENS = {
    "materialize_b12_dataset",
    "apply_portable_morph_div_v2",
    "build_candidates",
}


#
# These seams are allowed only in isolated synthetic contract testing.
#
CONDITIONAL_TOKENS = {
    "morph_candidate_provider",
    "post_v2_graph_provider",
    "MorphCandidateProvider",
    "validate_morph_candidate_provider",
}


#
# Historical/oracle/GT/evaluator paths are fully blocked here.
#
HARD_HAZARD_SUBSTRINGS = (
    "stage11_fold0_morphdiv_v2_actions_locked.pkl",
    "stage11_deployable_final_v9_policy_grid_actions.pkl",
    "stage11_additive_second_edge_tail.pkl",
    "historical_structural_oracle",
    "structural_oracle",
    "ground_truth",
    "official_evaluate",
    "_evaluate_matched_graph",
    "evaluate_graph",
    "per_sample_metrics",
    "read_pickle",
    "pd.read_pickle",
)


#
# External-data/runtime uncertainty => REVIEW_REQUIRED, not SAFE.
#
REVIEW_IO_SUBSTRINGS = (
    "data/raw/",
    "competition/",
    ".geff",
    ".pth",
    ".pt",
    ".parquet",
    "read_csv",
    "read_parquet",
    "torch.load",
    "load_graph",
    "subprocess.",
    "Popen(",
    "requests.",
    "urllib.",
)


#
# Synthetic markers cause Morph/provider tests to remain CONDITIONAL.
#
SYNTHETIC_MARKERS = (
    "mock",
    "stub",
    "fake",
    "synthetic",
    "monkeypatch",
    "MagicMock",
    "Mock",
)


#
# Only files genuinely related to the current Stage12 portability sources
# are candidates for actual execution.
#
RELEVANT_FILE_MARKERS = (
    "stage12_portable_current_best",
    "stage12_b12_portable",
)


# =============================================================================
# 7. Module-level import/side-effect conservatism
# =============================================================================

SAFE_MODULE_CALL_SUFFIXES = {
    "Path",
    "re.compile",
}


def module_level_call_hazards(
    source,
):

    tree = ast.parse(
        source
    )

    hazards = []


    def calls_inside_without_defs(node):

        result = []


        for child in ast.walk(
            node
        ):

            if child is not node and isinstance(
                child,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                    ast.ClassDef,
                    ast.Lambda,
                ),
            ):
                continue


            if isinstance(
                child,
                ast.Call,
            ):
                result.append(
                    child
                )


        return result


    for statement in tree.body:

        if isinstance(
            statement,
            (
                ast.Import,
                ast.ImportFrom,
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):
            continue


        calls = calls_inside_without_defs(
            statement
        )


        for call in calls:

            name = (
                dotted_name(
                    call.func
                )
                or
                "<dynamic-call>"
            )


            if (
                name
                in SAFE_MODULE_CALL_SUFFIXES
                or
                any(
                    name.endswith(
                        suffix
                    )

                    for suffix
                    in SAFE_MODULE_CALL_SUFFIXES
                )
            ):
                continue


            hazards.append({
                "line":
                    getattr(
                        call,
                        "lineno",
                        None,
                    ),

                "call":
                    name,

                "source":
                    node_source(
                        source,
                        call,
                    )[:1000],
            })


        if isinstance(
            statement,
            (
                ast.For,
                ast.While,
                ast.With,
                ast.Try,
                ast.Assert,
            ),
        ):

            hazards.append({
                "line":
                    getattr(
                        statement,
                        "lineno",
                        None,
                    ),

                "call":
                    "<module-level-control-flow>",

                "source":
                    node_source(
                        source,
                        statement,
                    )[:1000],
            })


    return hazards


# =============================================================================
# 8. Test-node extractor
# =============================================================================

def collect_test_nodes(
    path,
):

    source = path.read_text(
        encoding="utf-8"
    )

    tree = ast.parse(
        source,
        filename=str(
            path
        ),
    )

    rel = str(
        path.relative_to(
            ROOT
        )
    )

    records = []


    # Top-level test functions.
    for node in tree.body:

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
            ),
        ):

            if not node.name.startswith(
                "test_"
            ):
                continue


            segment = node_source(
                source,
                node,
            )


            records.append({
                "file":
                    rel,

                "nodeid":
                    (
                        f"{rel}::{node.name}"
                    ),

                "class_name":
                    None,

                "test_name":
                    node.name,

                "line_start":
                    node.lineno,

                "line_end":
                    node.end_lineno,

                "source_sha256":
                    sha256_text(
                        segment
                    ),

                "source":
                    segment,
            })


        elif isinstance(
            node,
            ast.ClassDef,
        ):

            if not node.name.startswith(
                "Test"
            ):
                continue


            for child in node.body:

                if not isinstance(
                    child,
                    (
                        ast.FunctionDef,
                        ast.AsyncFunctionDef,
                    ),
                ):
                    continue


                if not child.name.startswith(
                    "test_"
                ):
                    continue


                segment = node_source(
                    source,
                    child,
                )


                records.append({
                    "file":
                        rel,

                    "nodeid":
                        (
                            f"{rel}::{node.name}::{child.name}"
                        ),

                    "class_name":
                        node.name,

                    "test_name":
                        child.name,

                    "line_start":
                        child.lineno,

                    "line_end":
                        child.end_lineno,

                    "source_sha256":
                        sha256_text(
                            segment
                        ),

                    "source":
                        segment,
                })


    return (
        source,
        records,
    )


# =============================================================================
# 9. Per-test classification
# =============================================================================

def classify_test(
    record,
    *,
    file_source,
    module_hazards,
):

    source = record[
        "source"
    ]


    lower = source.lower()


    hard_hazards = [
        token

        for token
        in HARD_HAZARD_SUBSTRINGS

        if token.lower()
        in lower
    ]


    blocked_tokens = [
        token

        for token
        in BLOCKED_TOKENS

        if token
        in source
    ]


    conditional_tokens = [
        token

        for token
        in CONDITIONAL_TOKENS

        if token
        in source
    ]


    synthetic_markers = [
        token

        for token
        in SYNTHETIC_MARKERS

        if token.lower()
        in lower
    ]


    review_io = [
        token

        for token
        in REVIEW_IO_SUBSTRINGS

        if token.lower()
        in lower
    ]


    relevant_file = any(
        marker
        in file_source

        for marker
        in RELEVANT_FILE_MARKERS
    )


    # ---------------------------------------------------------------------
    # Hard closed boundary.
    # ---------------------------------------------------------------------

    if hard_hazards:

        classification = (
            "BLOCKED_BOUNDARY"
        )

        reason = (
            "HISTORICAL_GT_OR_EVALUATOR_HAZARD"
        )


    # Production runtime entry points are blocked even if fake arguments
    # might be supplied. Those belong to CONDITIONAL review only if clearly
    # synthetic.
    elif blocked_tokens:

        if synthetic_markers:

            classification = (
                "CONDITIONAL_SYNTHETIC"
            )

            reason = (
                "BLOCKED_RUNTIME_SYMBOL_WITH_SYNTHETIC_TEST_CONTEXT"
            )

        else:

            classification = (
                "BLOCKED_BOUNDARY"
            )

            reason = (
                "MORPH_OR_FULL_B12_RUNTIME_SYMBOL"
            )


    elif conditional_tokens:

        classification = (
            "CONDITIONAL_SYNTHETIC"
        )

        reason = (
            "PROVIDER_SEAM_OR_MORPH_CONTRACT_TEST"
        )


    elif review_io:

        classification = (
            "REVIEW_REQUIRED"
        )

        reason = (
            "EXTERNAL_IO_OR_RUNTIME_DEPENDENCY"
        )


    elif module_hazards:

        classification = (
            "REVIEW_REQUIRED"
        )

        reason = (
            "MODULE_LEVEL_SIDE_EFFECT_NOT_PROVEN_SAFE"
        )


    elif not relevant_file:

        classification = (
            "REVIEW_REQUIRED"
        )

        reason = (
            "OUTSIDE_CURRENT_STAGE12_PORTABILITY_SCOPE"
        )


    else:

        classification = (
            "SAFE_TO_RUN"
        )

        reason = (
            "NO_BLOCKED_MORPH_FULL_B12_HISTORICAL_GT_"
            "EVALUATOR_OR_EXTERNAL_IO_DEPENDENCY_DETECTED"
        )


    return {
        **{
            key:
                value

            for key, value
            in record.items()

            if key != "source"
        },

        "classification":
            classification,

        "reason":
            reason,

        "relevant_file":
            relevant_file,

        "blocked_tokens":
            blocked_tokens,

        "conditional_tokens":
            conditional_tokens,

        "synthetic_markers":
            synthetic_markers,

        "hard_hazards":
            hard_hazards,

        "review_io":
            review_io,

        "module_hazard_count":
            len(
                module_hazards
            ),
    }


# =============================================================================
# 10. Inventory + classification
# =============================================================================

test_records = []

file_records = []


for path in test_files:

    source, nodes = collect_test_nodes(
        path
    )


    module_hazards = module_level_call_hazards(
        source
    )


    rel = str(
        path.relative_to(
            ROOT
        )
    )


    relevant_file = any(
        marker
        in source

        for marker
        in RELEVANT_FILE_MARKERS
    )


    classified = [
        classify_test(
            record,
            file_source=source,
            module_hazards=module_hazards,
        )

        for record in nodes
    ]


    test_records.extend(
        classified
    )


    file_records.append({
        "file":
            rel,

        "sha256":
            sha256_file(
                path
            ),

        "relevant_to_stage12_portability":
            relevant_file,

        "module_hazards":
            module_hazards,

        "test_count":
            len(
                nodes
            ),

        "classification_counts":
            dict(
                Counter(
                    record[
                        "classification"
                    ]

                    for record
                    in classified
                )
            ),
    })


classification_counts = dict(
    Counter(
        record[
            "classification"
        ]

        for record
        in test_records
    )
)


safe_records = [
    record

    for record
    in test_records

    if record[
        "classification"
    ]
    == "SAFE_TO_RUN"
]


conditional_records = [
    record

    for record
    in test_records

    if record[
        "classification"
    ]
    == "CONDITIONAL_SYNTHETIC"
]


blocked_records = [
    record

    for record
    in test_records

    if record[
        "classification"
    ]
    == "BLOCKED_BOUNDARY"
]


review_records = [
    record

    for record
    in test_records

    if record[
        "classification"
    ]
    == "REVIEW_REQUIRED"
]


safe_nodeids = [
    record[
        "nodeid"
    ]

    for record
    in safe_records
]


print(
    "=" * 120,
    flush=True,
)

print(
    "TEST CLASSIFICATION",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "total test nodes:",
    len(
        test_records
    ),
    flush=True,
)

print(
    "SAFE_TO_RUN:",
    len(
        safe_records
    ),
    flush=True,
)

print(
    "CONDITIONAL_SYNTHETIC:",
    len(
        conditional_records
    ),
    flush=True,
)

print(
    "BLOCKED_BOUNDARY:",
    len(
        blocked_records
    ),
    flush=True,
)

print(
    "REVIEW_REQUIRED:",
    len(
        review_records
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "SAFE NODE IDS:",
    flush=True,
)


for nodeid in safe_nodeids:

    print(
        " ",
        nodeid,
        flush=True,
    )


if not safe_nodeids:

    print(
        "  <none>",
        flush=True,
    )


print(
    flush=True,
)


print(
    "CONDITIONAL SYNTHETIC NODE IDS:",
    flush=True,
)


for record in conditional_records:

    print(
        " ",
        record[
            "nodeid"
        ],
        "::",
        record[
            "reason"
        ],
        flush=True,
    )


if not conditional_records:

    print(
        "  <none>",
        flush=True,
    )


print(
    flush=True,
)


print(
    "BLOCKED NODE IDS:",
    flush=True,
)


for record in blocked_records:

    print(
        " ",
        record[
            "nodeid"
        ],
        "::",
        record[
            "reason"
        ],
        flush=True,
    )


if not blocked_records:

    print(
        "  <none>",
        flush=True,
    )


print(
    flush=True,
)


# =============================================================================
# 11. Safe pytest execution
# =============================================================================

#
# IMPORTANT:
#
# Only exact SAFE_TO_RUN node IDs are passed to pytest.
#
# Conditional, blocked and review-required nodes are never passed.
#

pytest_executed = (
    len(
        safe_nodeids
    )
    > 0
)


pytest_returncode = None

pytest_output_lines = []


if pytest_executed:

    cmd = [
        sys.executable,
        "-m",
        "pytest",
        "-vv",
        "-s",
        "-p",
        "no:cacheprovider",
        *safe_nodeids,
    ]


    env = dict(
        os.environ
    )


    env[
        "PYTHONDONTWRITEBYTECODE"
    ] = "1"


    print(
        "=" * 120,
        flush=True,
    )

    print(
        "SAFE NON-MORPH PYTEST EXECUTION",
        flush=True,
    )

    print(
        "=" * 120,
        flush=True,
    )

    print(
        "node count:",
        len(
            safe_nodeids
        ),
        flush=True,
    )

    print(
        flush=True,
    )


    process = subprocess.Popen(
        cmd,
        cwd=ROOT,
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )


    assert (
        process.stdout
        is not None
    )


    for line in process.stdout:

        pytest_output_lines.append(
            line
        )

        print(
            line,
            end="",
            flush=True,
        )


    pytest_returncode = (
        process.wait()
    )


else:

    print(
        "=" * 120,
        flush=True,
    )

    print(
        "SAFE NON-MORPH PYTEST EXECUTION",
        flush=True,
    )

    print(
        "=" * 120,
        flush=True,
    )

    print(
        "NOT RUN — no test node satisfied the conservative SAFE_TO_RUN proof.",
        flush=True,
    )

    print(
        flush=True,
    )


# =============================================================================
# 12. Regression result
# =============================================================================

if not pytest_executed:

    regression_status = (
        "NO_PROVEN_SAFE_EXISTING_PYTEST_TARGETS"
    )

    regression_pass = True


elif pytest_returncode == 0:

    regression_status = (
        "SAFE_NON_MORPH_PYTEST_TARGETS_PASS"
    )

    regression_pass = True


else:

    regression_status = (
        "SAFE_NON_MORPH_PYTEST_TARGETS_FAILED"
    )

    regression_pass = False


print(
    "=" * 120,
    flush=True,
)

print(
    "REGRESSION RESULT",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)

print(
    "pytest executed:",
    pytest_executed,
    flush=True,
)

print(
    "pytest return code:",
    pytest_returncode,
    flush=True,
)

print(
    "status:",
    regression_status,
    flush=True,
)

print(
    flush=True,
)


# =============================================================================
# 13. Source/test immutability proof
# =============================================================================

portable_sha_after = sha256_file(
    PORTABLE
)


b12_sha_after = sha256_file(
    B12
)


test_sha_after = {
    str(
        path.relative_to(
            ROOT
        )
    ):
        sha256_file(
            path
        )

    for path in test_files
}


source_mutation = not (
    portable_sha_after
    == portable_sha_before
    == EXPECTED_PORTABLE_SHA
    and
    b12_sha_after
    == b12_sha_before
    == EXPECTED_B12_SHA
)


test_mutation = (
    test_sha_after
    != test_sha_before
)


assert (
    source_mutation
    is False
)


assert (
    test_mutation
    is False
)


# =============================================================================
# 14. Explicit safety ledger
# =============================================================================

conditional_test_executed = False

blocked_test_executed = False

review_required_test_executed = False

Morph_generation_executed = False

post_v2_full_execution_executed = False

B12_full_materialization_executed = False

historical_pickle_loaded_by_harness = False

historical_Morph_runtime_use = False

historical_oracle_opened = False

GT_loaded_by_harness = False

evaluator_executed = False

performance_metric_generated = False

parameter_search = False

Git_operation_executed = False

current_best_replaced = False


assert (
    conditional_test_executed
    is False
)


assert (
    blocked_test_executed
    is False
)


assert (
    review_required_test_executed
    is False
)


# =============================================================================
# 15. Audit resolution
# =============================================================================

boundary_preserved = all([
    conditional_test_executed
    is False,

    blocked_test_executed
    is False,

    review_required_test_executed
    is False,

    Morph_generation_executed
    is False,

    post_v2_full_execution_executed
    is False,

    B12_full_materialization_executed
    is False,

    historical_oracle_opened
    is False,

    current_best_replaced
    is False,
])


if regression_pass:

    audit_status = (
        "NON_MORPH_TEST_BOUNDARY_CLASSIFIED_"
        "SAFE_REGRESSION_EXECUTION_ACCEPTED"
    )


else:

    audit_status = (
        "NON_MORPH_SAFE_REGRESSION_FAILURE_REPAIR_REQUIRED"
    )


next_stage = (
    "NOTEBOOK14_14_4B_"
    "NON_MORPH_TEST_BOUNDARY_CLOSURE"
)


# =============================================================================
# 16. Formal gates
# =============================================================================

semantic_gates = {
    "14_4_parent_exactly_locked":
        True,

    "portable_source_exactly_locked":
        (
            portable_sha_before
            == EXPECTED_PORTABLE_SHA
        ),

    "B12_source_exactly_locked":
        (
            b12_sha_before
            == EXPECTED_B12_SHA
        ),

    "test_inventory_completed":
        (
            len(
                test_files
            )
            > 0
        ),

    "test_nodes_AST_classified":
        (
            len(
                test_records
            )
            > 0
        ),

    "conditional_tests_not_executed":
        (
            conditional_test_executed
            is False
        ),

    "blocked_tests_not_executed":
        (
            blocked_test_executed
            is False
        ),

    "review_required_tests_not_executed":
        (
            review_required_test_executed
            is False
        ),

    "only_SAFE_TO_RUN_nodes_sent_to_pytest":
        True,

    "safe_regression_pass_or_no_safe_target":
        regression_pass,

    "Morph_generation_not_executed":
        (
            Morph_generation_executed
            is False
        ),

    "post_v2_full_execution_not_executed":
        (
            post_v2_full_execution_executed
            is False
        ),

    "B12_full_materialization_not_executed":
        (
            B12_full_materialization_executed
            is False
        ),

    "historical_oracle_remains_closed":
        (
            historical_oracle_opened
            is False
        ),

    "historical_pickle_not_loaded_by_harness":
        (
            historical_pickle_loaded_by_harness
            is False
        ),

    "historical_Morph_not_used_at_runtime":
        (
            historical_Morph_runtime_use
            is False
        ),

    "GT_not_loaded_by_harness":
        (
            GT_loaded_by_harness
            is False
        ),

    "evaluator_not_executed":
        (
            evaluator_executed
            is False
        ),

    "performance_not_generated":
        (
            performance_metric_generated
            is False
        ),

    "parameter_search_not_run":
        (
            parameter_search
            is False
        ),

    "CURRENT_BEST_not_replaced":
        (
            current_best_replaced
            is False
        ),

    "boundary_preserved":
        boundary_preserved,

    "no_Git_operations":
        (
            Git_operation_executed
            is False
        ),

    "no_source_mutation":
        (
            source_mutation
            is False
        ),

    "no_test_mutation":
        (
            test_mutation
            is False
        ),
}


print(
    "=" * 120,
    flush=True,
)

print(
    "14.4A SEMANTIC GATES",
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


for gate, ok in semantic_gates.items():

    print(
        f"{'PASS' if ok else 'FAIL':<4} "
        f"{gate}",
        flush=True,
    )


failed_gates = [
    gate

    for gate, ok
    in semantic_gates.items()

    if not ok
]


# =============================================================================
# 17. Formal artifact
# =============================================================================

payload = {
    "stage":
        "NOTEBOOK14_14_4A",

    "title":
        "NON_MORPH_STATIC_AND_REGRESSION_READINESS_AUDIT",

    "version":
        1,

    "decision":
        (
            "PASS"
            if not failed_gates
            else
            "FAIL"
        ),

    "classification":
        (
            "ENGINEERING_NON_MORPH_TEST_BOUNDARY_AND_SAFE_REGRESSION_AUDIT"
        ),

    "parent_14_4": {
        "path":
            str(
                PARENT_REL
            ),

        "file_sha256":
            EXPECTED_PARENT_FILE_SHA,

        "content_sha256":
            EXPECTED_PARENT_CONTENT_SHA,
    },

    "source_locks": {
        "portable_current_best": {
            "path":
                str(
                    PORTABLE_REL
                ),

            "sha256":
                portable_sha_before,
        },

        "B12_portable": {
            "path":
                str(
                    B12_REL
                ),

            "sha256":
                b12_sha_before,
        },
    },

    "classification_policy": {
        "uncertainty_rule":
            "UNCERTAINTY_MEANS_DO_NOT_RUN",

        "SAFE_TO_RUN":
            (
                "Relevant Stage12 portability test with no detected blocked, "
                "conditional-provider, historical/GT/evaluator, external-I/O, "
                "or module-side-effect dependency."
            ),

        "CONDITIONAL_SYNTHETIC":
            (
                "Synthetic/provider contract test; allowed by 14.4 only in "
                "isolated mode, therefore not executed in 14.4A."
            ),

        "BLOCKED_BOUNDARY":
            (
                "Crosses closed Morph/full-B12/historical/GT/evaluator boundary."
            ),

        "REVIEW_REQUIRED":
            (
                "Static proof insufficient or test outside current scope."
            ),
    },

    "test_files":
        file_records,

    "test_nodes":
        test_records,

    "classification_counts":
        classification_counts,

    "safe_nodeids":
        safe_nodeids,

    "pytest_execution": {
        "executed":
            pytest_executed,

        "node_count":
            len(
                safe_nodeids
            ),

        "returncode":
            pytest_returncode,

        "status":
            regression_status,

        "output_tail":
            pytest_output_lines[
                -300:
            ],
    },

    "boundary_execution_ledger": {
        "conditional_test_executed":
            False,

        "blocked_test_executed":
            False,

        "review_required_test_executed":
            False,

        "Morph_generation_executed":
            False,

        "post_v2_full_execution_executed":
            False,

        "B12_full_materialization_executed":
            False,

        "historical_pickle_loaded_by_harness":
            False,

        "historical_Morph_runtime_use":
            False,

        "historical_oracle_opened":
            False,

        "GT_loaded_by_harness":
            False,

        "evaluator_executed":
            False,

        "performance_metric_generated":
            False,

        "parameter_search":
            False,

        "Git_operation":
            False,
    },

    "resolution": {
        "status":
            audit_status,

        "test_boundary_classified":
            True,

        "safe_test_count":
            len(
                safe_records
            ),

        "conditional_test_count":
            len(
                conditional_records
            ),

        "blocked_test_count":
            len(
                blocked_records
            ),

        "review_required_test_count":
            len(
                review_records
            ),

        "safe_regression_status":
            regression_status,

        "Morph_dependent_full_execution_authorized":
            False,

        "historical_oracle_access_authorized":
            False,

        "structural_fidelity_verified":
            False,

        "scientific_result":
            "NONE",

        "performance_result":
            "NONE",

        "CURRENT_BEST_replaced":
            False,

        "next_stage":
            next_stage,
    },

    "semantic_gates":
        semantic_gates,

    "claim_boundary": {
        "safe_pytest_result":
            regression_status,

        "safe_pytest_is_full_B12_validation":
            False,

        "conditional_synthetic_tests":
            "NOT_RUN",

        "blocked_tests":
            "NOT_RUN",

        "Morph_generation":
            "NOT_RUN",

        "fresh_post_V2":
            "NOT_RUN",

        "full_B12_materialization":
            "NOT_RUN",

        "historical_structural_oracle":
            "CLOSED",

        "structural_fidelity_verified":
            False,

        "scientific_result":
            "NONE",

        "performance_result":
            "NONE",

        "CURRENT_BEST_replaced":
            False,

        "source_mutation":
            False,

        "test_mutation":
            False,

        "Git_operation":
            False,
    },

    "explicit_non_claims": [
        "SAFE_TO_RUN pytest success is not full B12 validation.",
        "Conditional synthetic-provider tests are not executed in this stage.",
        "No synthetic provider is treated as production Morph authority.",
        "No Morph generation is executed.",
        "No full post-V2 execution is performed.",
        "No B12 materialization is performed.",
        "No historical structural oracle is opened.",
        "No structural fidelity claim is made.",
        "No scientific or performance evidence is generated.",
        "No CURRENT_BEST replacement occurs.",
    ],

    "next_stage":
        next_stage,
}


content_sha = canonical_json_sha(
    payload
)


artifact_obj = {
    "content_sha256":
        content_sha,
    **payload,
}


artifact_bytes = (
    json.dumps(
        artifact_obj,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
        allow_nan=False,
    )
    + "\n"
).encode(
    "utf-8"
)


if OUT.exists():

    assert (
        OUT.read_bytes()
        == artifact_bytes
    ), (
        "14.4A artifact exists but differs "
        "from exact audit/regression state."
    )

    artifact_mode = (
        "EXISTING_EXACT_ARTIFACT_REUSED"
    )


else:

    OUT.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    OUT.write_bytes(
        artifact_bytes
    )

    artifact_mode = (
        "NEW_FORMAL_ARTIFACT_WRITTEN"
    )


artifact_file_sha = sha256_file(
    OUT
)


# =============================================================================
# 18. Final
# =============================================================================

print(
    "=" * 120,
    flush=True,
)

print(
    (
        "NOTEBOOK 14 / 14.4A PASS"
        if not failed_gates
        else
        "NOTEBOOK 14 / 14.4A FAIL"
    ),
    flush=True,
)

print(
    "=" * 120,
    flush=True,
)


print(
    "TEST BOUNDARY:",
    flush=True,
)

print(
    "  SAFE_TO_RUN:",
    len(
        safe_records
    ),
    flush=True,
)

print(
    "  CONDITIONAL_SYNTHETIC:",
    len(
        conditional_records
    ),
    flush=True,
)

print(
    "  BLOCKED_BOUNDARY:",
    len(
        blocked_records
    ),
    flush=True,
)

print(
    "  REVIEW_REQUIRED:",
    len(
        review_records
    ),
    flush=True,
)

print(
    flush=True,
)


print(
    "REGRESSION:",
    flush=True,
)

print(
    "  pytest executed:",
    pytest_executed,
    flush=True,
)

print(
    "  safe node count:",
    len(
        safe_nodeids
    ),
    flush=True,
)

print(
    "  pytest rc:",
    pytest_returncode,
    flush=True,
)

print(
    "  status:",
    regression_status,
    flush=True,
)

print(
    flush=True,
)


print(
    "BOUNDARY:",
    flush=True,
)

print(
    "  conditional tests executed:",
    False,
    flush=True,
)

print(
    "  blocked tests executed:",
    False,
    flush=True,
)

print(
    "  Morph generation:",
    "NOT_RUN",
    flush=True,
)

print(
    "  full post-V2:",
    "NOT_RUN",
    flush=True,
)

print(
    "  full B12:",
    "NOT_RUN",
    flush=True,
)

print(
    "  historical oracle:",
    "CLOSED",
    flush=True,
)

print(
    "  CURRENT_BEST:",
    "UNCHANGED",
    flush=True,
)

print(
    flush=True,
)


print(
    "STATUS:",
    audit_status,
    flush=True,
)

print(
    "NEXT:",
    next_stage,
    flush=True,
)

print(
    flush=True,
)


print(
    "artifact:",
    OUT_REL,
    flush=True,
)

print(
    "artifact content SHA:",
    content_sha,
    flush=True,
)

print(
    "artifact file SHA:",
    artifact_file_sha,
    flush=True,
)

print(
    "artifact mode:",
    artifact_mode,
    flush=True,
)


assert (
    failed_gates == []
), (
    "14.4A failed gates: "
    f"{failed_gates}"
)
```

```python
# =============================================================================
# NOTEBOOK 14 :: 14.5A
# NON-MORPH COMPONENT DEPENDENCY AUDIT
#
# Parent: 14.5 PASS
#
# Static source inspection only.
#
# Outputs:
#   - source files and exact SHA-256
#   - candidate functions/classes and exact source hashes
#   - direct import/call evidence
#   - explicit Morph / historical / evaluator risk indicators
#   - lexical links to existing tests
#   - unresolved dependency review requirements
#
# Does NOT authorize tests or runtime.
# Does NOT recover or implement Morph.
# Does NOT open historical pickle/oracle or GT.
# Does NOT execute pytest, Git, evaluator or inference.
# =============================================================================

import ast
import hashlib
import json

from collections import Counter
from pathlib import Path


# =============================================================================
# 0. Paths and frozen identities
# =============================================================================

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

PARENT_REL = (
    "reports/"
    "notebook14_14_5_"
    "non_morph_portability_deliverables_plan_v1.json"
)

OUT_REL = (
    "reports/"
    "notebook14_14_5a_"
    "non_morph_component_dependency_audit_v1.json"
)

PARENT = ROOT / PARENT_REL
OUT = ROOT / OUT_REL

EXPECTED_PARENT_FILE = (
    "2bdcba68f3349f665cec184579466a0c2e0c8c6e83e29d23fd463de5827ec23b"
)

EXPECTED_PARENT_CONTENT = (
    "996463af04fbf6393aa015c1a6540dc3fc4754401ddfc1b887065f4e9d8de6cb"
)

SOURCE_LOCKS = {
    "src/biohub_cell_tracking/stage12_portable_current_best.py":
        "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0",

    "src/biohub_cell_tracking/stage12_b12_portable.py":
        "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52",
}

# Names are search probes, not declarations of actual implementation.
FAMILIES = {
    "P995": (
        "p995",
        "det995",
        "frozen_v9",
    ),

    "SUPPRESSED_PEAK": (
        "suppressed_peak",
        "suppressed-peak",
    ),

    "V1_COMPONENT": (
        "mutual_cont_v1",
        "portable_v1",
        "v1_graph",
        "reconstruct_v1",
    ),

    "V2_COMPONENT": (
        "mutual_cont_v2",
        "portable_v2",
        "v2_graph",
        "reconstruct_v2",
    ),
}

RISK_PROBES = {
    "MORPH_DEPENDENCY": (
        "morph_candidate_provider",
        "apply_portable_morph_div_v2",
        "morphdiv",
    ),

    "FULL_B12": (
        "materialize_b12_dataset",
        "post_v2_graph_provider",
    ),

    "HISTORICAL_ARTIFACT": (
        "read_pickle",
        "stage11_fold0_morphdiv_v2_actions_locked",
        "stage11_deployable_final_v9_policy_grid_actions",
    ),

    "GT_OR_EVALUATOR": (
        "ground_truth",
        "official_evaluate",
        "_evaluate_matched_graph",
        "per_sample_metrics",
    ),

    "EXTERNAL_IO_OR_PROCESS": (
        "subprocess.",
        "Popen(",
        "torch.load",
        "load_graph",
        ".geff",
    ),
}

MAX_SOURCE_BYTES = 2_000_000
MAX_CANDIDATES_PER_FAMILY = 35
MAX_TEST_HINTS_PER_FAMILY = 30


# =============================================================================
# 1. Pure helpers
# =============================================================================

def hash_bytes(value):
    return hashlib.sha256(value).hexdigest()


def hash_file(path):
    return hash_bytes(path.read_bytes())


def canonical_hash(value):
    return hash_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def source_segment(source, node):
    result = ast.get_source_segment(source, node)

    if result is not None:
        return result

    return ast.unparse(node)


def dotted_name(node):
    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):
        base = dotted_name(node.value)

        if base is not None:
            return base + "." + node.attr

    return "<dynamic>"


def matched_probes(text, probes):
    lower = text.lower()

    return sorted(
        probe
        for probe in probes
        if probe.lower() in lower
    )


def risk_indicators(text):
    return {
        group: matched_probes(text, probes)
        for group, probes in RISK_PROBES.items()
        if matched_probes(text, probes)
    }


def source_imports(tree):
    imports = set()

    for node in tree.body:
        if isinstance(node, ast.Import):
            for alias in node.names:
                imports.add(alias.name)

        elif isinstance(node, ast.ImportFrom):
            module = node.module or ""

            imports.add(
                "." * node.level + module
            )

    return sorted(imports)


def call_names(node):
    return sorted({
        dotted_name(child.func)
        for child in ast.walk(node)
        if isinstance(child, ast.Call)
    })


# =============================================================================
# 2. Exact parent lock
# =============================================================================

print("=" * 110)
print("NOTEBOOK 14 :: 14.5A NON-MORPH COMPONENT DEPENDENCY AUDIT")
print("=" * 110)

assert PARENT.is_file()
assert hash_file(PARENT) == EXPECTED_PARENT_FILE

parent = json.loads(
    PARENT.read_text(encoding="utf-8")
)

assert parent["decision"] == "PASS"
assert parent["content_sha256"] == EXPECTED_PARENT_CONTENT

parent_payload = dict(parent)
parent_payload.pop("content_sha256")

assert canonical_hash(parent_payload) == EXPECTED_PARENT_CONTENT

assert parent["resolution"]["status"] == (
    "NON_MORPH_DELIVERABLES_PLAN_FROZEN_"
    "COMPONENT_AUDIT_MAY_BEGIN"
)

assert (
    parent["resolution"]["non_Morph_component_audit_authorized"]
    is True
)

assert (
    parent["resolution"]["Morph_provider_implementation_authorized"]
    is False
)

assert (
    parent["resolution"]["full_B12_execution_authorized"]
    is False
)

assert (
    parent["resolution"]["historical_oracle_access_authorized"]
    is False
)

assert parent["resolution"]["CURRENT_BEST_replaced"] is False

for relpath, expected in SOURCE_LOCKS.items():
    assert hash_file(ROOT / relpath) == expected

assert parent["accepted_regression"]["selected_nodeids"] == 47
assert parent["accepted_regression"]["parent_pytest_returncode"] == 0
assert parent["accepted_regression"]["deferred_nodes"] == 329

print("[PASS] PARENT_AND_CURRENT_SOURCE_LOCKS")


# =============================================================================
# 3. Read source and test inventories
# =============================================================================

source_roots = (
    ROOT / "src" / "biohub_cell_tracking",
    ROOT / "scripts",
)

source_paths = []

for directory in source_roots:
    if not directory.is_dir():
        continue

    for path in directory.rglob("*.py"):
        if path.is_symlink() or not path.is_file():
            continue

        # Historical-oracle recovery remains closed.
        # Do not inspect files explicitly named as oracle/evaluator sources.
        lower = path.name.lower()

        if "oracle" in lower or "evaluator" in lower:
            continue

        source_paths.append(path)

source_paths = sorted(set(source_paths))

test_paths = sorted(
    path
    for path in (ROOT / "tests").rglob("test_*.py")
    if path.is_file() and not path.is_symlink()
)

assert len(source_paths) > 0
assert len(test_paths) == 13

source_baseline = {
    str(path.relative_to(ROOT)): hash_file(path)
    for path in source_paths
}

test_baseline = {
    str(path.relative_to(ROOT)): hash_file(path)
    for path in test_paths
}

print("[PASS] SOURCE_AND_TEST_INVENTORY")
print("  source files:", len(source_paths))
print("  test files:", len(test_paths))


# =============================================================================
# 4. Extract static source units
# =============================================================================

all_units = []
file_audit = []
skipped_files = []

for path in source_paths:
    relpath = str(path.relative_to(ROOT))

    if path.stat().st_size > MAX_SOURCE_BYTES:
        skipped_files.append({
            "path": relpath,
            "reason": "EXCEEDS_STATIC_SOURCE_SIZE_LIMIT",
        })
        continue

    try:
        source = path.read_text(encoding="utf-8")
        tree = ast.parse(source, filename=relpath)

    except (UnicodeError, SyntaxError) as error:
        skipped_files.append({
            "path": relpath,
            "reason": type(error).__name__,
            "detail": str(error)[:300],
        })
        continue

    imports = source_imports(tree)

    file_audit.append({
        "path": relpath,
        "file_sha256": source_baseline[relpath],
        "imports": imports,
        "file_risk_indicators": risk_indicators(source),
    })

    # Record top-level functions and class methods.
    # No source module is imported or executed.

    units = []

    for node in tree.body:
        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef)
        ):
            units.append((node.name, node))

        elif isinstance(node, ast.ClassDef):
            units.append((node.name, node))

            for child in node.body:
                if isinstance(
                    child,
                    (ast.FunctionDef, ast.AsyncFunctionDef)
                ):
                    units.append(
                        (node.name + "." + child.name, child)
                    )

    for name, node in units:
        segment = source_segment(source, node)

        family_hits = {
            family: matched_probes(
                name + "\n" + segment,
                probes,
            )
            for family, probes in FAMILIES.items()
        }

        if not any(family_hits.values()):
            continue

        all_units.append({
            "path": relpath,
            "file_sha256": source_baseline[relpath],
            "unit": name,
            "unit_kind": type(node).__name__,
            "line_start": node.lineno,
            "line_end": node.end_lineno,
            "unit_sha256": hash_bytes(segment.encode("utf-8")),
            "family_hits": {
                key: value
                for key, value in family_hits.items()
                if value
            },
            "direct_imports_from_module": imports,
            "syntactic_call_names": call_names(node),
            "risk_indicators": risk_indicators(segment),
        })

print("[PASS] STATIC_SOURCE_UNIT_EXTRACTION")
print("  matched units:", len(all_units))
print("  skipped source files:", len(skipped_files))


# =============================================================================
# 5. Audit existing tests as LEXICAL HINTS only
# =============================================================================

# 14.4B-R2 freezes the previous selected node IDs.
# Everything else remains deferred. No new execution authority.

prior_safe = set(
    parent["accepted_regression"].get("safe_nodeids", [])
)

# 14.5 parent may not embed the exact node IDs.
# Only the counts are inherited from its locked regression record.
# Therefore no test node is newly declared SAFE_TO_RUN here.

test_hints = {family: [] for family in FAMILIES}

for path in test_paths:
    relpath = str(path.relative_to(ROOT))

    try:
        source = path.read_text(encoding="utf-8")
        tree = ast.parse(source, filename=relpath)

    except (UnicodeError, SyntaxError):
        continue

    test_nodes = []

    for node in tree.body:
        if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)):
            if node.name.startswith("test_"):
                test_nodes.append(
                    (relpath + "::" + node.name, node)
                )

        elif isinstance(node, ast.ClassDef):
            if not node.name.startswith("Test"):
                continue

            for child in node.body:
                if isinstance(
                    child,
                    (ast.FunctionDef, ast.AsyncFunctionDef)
                ) and child.name.startswith("test_"):
                    test_nodes.append(
                        (
                            relpath + "::" + node.name
                            + "::" + child.name,
                            child,
                        )
                    )

    for nodeid, node in test_nodes:
        segment = source_segment(source, node)
        haystack = nodeid + "\n" + segment

        for family, probes in FAMILIES.items():
            hits = matched_probes(haystack, probes)

            if not hits:
                continue

            test_hints[family].append({
                "nodeid": nodeid,
                "test_file_sha256": test_baseline[relpath],
                "lexical_hits": hits,
                "execution_authorized_by_14_5A": False,
            })


# =============================================================================
# 6. Build candidate inventory without asserting execution readiness
# =============================================================================

family_inventory = {}

for family in FAMILIES:
    candidates = [
        unit
        for unit in all_units
        if family in unit["family_hits"]
    ]

    candidates.sort(
        key=lambda unit: (
            -len(unit["family_hits"][family]),
            unit["path"],
            unit["unit"],
        )
    )

    hints = test_hints[family]
    hints.sort(key=lambda item: item["nodeid"])

    family_inventory[family] = {
        "source_candidate_count": len(candidates),
        "source_candidates": candidates[:MAX_CANDIDATES_PER_FAMILY],
        "source_candidates_truncated": (
            len(candidates) > MAX_CANDIDATES_PER_FAMILY
        ),

        "lexical_test_hint_count": len(hints),
        "lexical_test_hints": hints[:MAX_TEST_HINTS_PER_FAMILY],
        "test_hints_truncated": (
            len(hints) > MAX_TEST_HINTS_PER_FAMILY
        ),

        "classification": (
            "STATIC_CANDIDATES_REQUIRE_DEPENDENCY_REVIEW"
            if candidates
            else "NO_STATIC_CANDIDATE_FOUND"
        ),

        "isolated_execution_authorized": False,
        "new_pytest_execution_authorized": False,
        "full_B12_equivalence_claim": False,
    }


# =============================================================================
# 7. Explicit unresolved dependency boundary
# =============================================================================

unresolved = [
    "Static symbol matches do not prove a component's runtime semantics.",
    "Module imports may introduce dependencies outside the inspected unit.",
    "Syntactic call names do not resolve dynamic dispatch or transitive calls.",
    "Test lexical matches do not prove fixture or collection safety.",
    "Any candidate with Morph or full-B12 dependency requires isolation review.",
    "No newly discovered candidate is approved for execution by this audit.",
]

next_stage = (
    "NOTEBOOK14_14_5A_R1_"
    "NON_MORPH_COMPONENT_EXACT_CALL_BOUNDARY_REVIEW"
)

status = (
    "NON_MORPH_COMPONENT_STATIC_INVENTORY_COMPLETE_"
    "EXECUTION_TARGETS_REQUIRE_REVIEW"
)


# =============================================================================
# 8. Print audit summary
# =============================================================================

print("=" * 110)
print("COMPONENT INVENTORY")
print("=" * 110)

for family, record in family_inventory.items():
    print(family)
    print("  source candidates:", record["source_candidate_count"])
    print("  lexical test hints:", record["lexical_test_hint_count"])
    print("  execution authorized:", False)

    for unit in record["source_candidates"][:8]:
        print(
            "   ",
            unit["path"],
            "::",
            unit["unit"],
            "line",
            unit["line_start"],
        )

        if unit["risk_indicators"]:
            print(
                "       risk indicators:",
                sorted(unit["risk_indicators"]),
            )

    print()

print("skipped source files:", len(skipped_files))
print("new tests run:", False)
print("runtime executed:", False)
print("historical oracle opened:", False)


# =============================================================================
# 9. Safety / immutability verification
# =============================================================================

for relpath, before in source_baseline.items():
    assert hash_file(ROOT / relpath) == before

for relpath, before in test_baseline.items():
    assert hash_file(ROOT / relpath) == before

for relpath, expected in SOURCE_LOCKS.items():
    assert hash_file(ROOT / relpath) == expected

assert hash_file(PARENT) == EXPECTED_PARENT_FILE

gates = {
    "parent_exactly_locked": True,
    "portable_source_locked": True,
    "B12_source_locked": True,
    "source_inventory_completed": len(source_paths) > 0,
    "test_inventory_completed": len(test_paths) == 13,
    "source_units_extracted_statically": True,
    "candidate_runtime_authority_not_inferred": True,
    "test_lexical_hints_not_treated_as_safe_execution": True,
    "Morph_authority_recovery_not_reopened": True,
    "full_post_v2_not_executed": True,
    "full_B12_not_executed": True,
    "historical_oracle_not_opened": True,
    "GT_evaluator_not_accessed": True,
    "new_pytest_not_run": True,
    "CURRENT_BEST_unchanged": True,
    "source_unchanged": True,
    "tests_unchanged": True,
    "no_Git_operations": True,
}

assert all(gates.values())


# =============================================================================
# 10. Formal artifact
# =============================================================================

payload = {
    "stage": "NOTEBOOK14_14_5A",
    "title": "NON_MORPH_COMPONENT_DEPENDENCY_AUDIT",
    "version": 1,
    "decision": "PASS",

    "parent_14_5": {
        "path": PARENT_REL,
        "file_sha256": EXPECTED_PARENT_FILE,
        "content_sha256": EXPECTED_PARENT_CONTENT,
    },

    "source_locks": SOURCE_LOCKS,

    "source_inventory": {
        "source_file_count": len(source_paths),
        "source_file_sha256": source_baseline,
        "skipped_files": skipped_files,
        "matched_unit_count": len(all_units),
    },

    "test_inventory": {
        "test_file_count": len(test_paths),
        "test_file_sha256": test_baseline,
        "new_pytest_execution": False,
    },

    "component_families": family_inventory,

    "unresolved_dependencies": unresolved,

    "resolution": {
        "status": status,
        "static_inventory_complete": True,
        "isolated_component_runtime_authorized": False,
        "new_pytest_execution_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
        "next_stage": next_stage,
    },

    "semantic_gates": gates,

    "claim_boundary": {
        "source_analysis": "STATIC_LEXICAL_AND_AST_ONLY",
        "call_graph": "NOT_TRANSITIVELY_VERIFIED",
        "test_safety": "NOT_NEWLY_AUTHORIZED",
        "runtime_execution": False,
        "pytest_execution": False,
        "source_mutation": False,
        "test_mutation": False,
        "Git_operation": False,
        "GT_access": False,
        "evaluator_execution": False,
        "performance_evidence": False,
    },

    "next_stage": next_stage,
}

content_sha = canonical_hash(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

assert OUT.parent.is_dir()

if OUT.exists():
    assert OUT.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"

else:
    OUT.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"

print("=" * 110)
print("NOTEBOOK 14 / 14.5A PASS")
print("=" * 110)
print("STATUS:", status)
print("NEXT:", next_stage)
print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", hash_file(OUT))
print("artifact mode:", mode)
```

```python
# =============================================================================
# Notebook 14 / 14.5A-R1
# NON-MORPH COMPONENT EXACT DIRECT-CALL BOUNDARY REVIEW
#
# Read-only static review. No runtime, pytest, import of project modules,
# historical pickle, GT, evaluator, oracle, Git or source/test mutation.
#
# A direct-call review is NOT a complete transitive-call safety proof.
# No new execution target is authorized by this stage.
# =============================================================================

import ast
import hashlib
import json
from pathlib import Path

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

PARENT_REL = (
    "reports/"
    "notebook14_14_5a_"
    "non_morph_component_dependency_audit_v1.json"
)

OUT_REL = (
    "reports/"
    "notebook14_14_5a_r1_"
    "non_morph_component_exact_direct_call_boundary_review_v1.json"
)

PARENT = ROOT / PARENT_REL
OUT = ROOT / OUT_REL

PARENT_FILE_SHA = (
    "41aba869996b386fbea753d5ecd4c1e47074624afcef162d3a15054debc54f28"
)

PARENT_CONTENT_SHA = (
    "2a3c3fb03257db10188a1f3e08268ba07b7b3952990fb607d132fe97f5f559af"
)

SOURCE_LOCKS = {
    "src/biohub_cell_tracking/stage12_portable_current_best.py":
        "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0",

    "src/biohub_cell_tracking/stage12_b12_portable.py":
        "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52",
}

# Exact source units identified in the 14.5A log.
# The names below are inspection targets, NOT approved execution targets.

TARGETS = {
    "P995": [
        (
            "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py",
            "_FreshP995NodeGraph",
        ),
    ],

    "SUPPRESSED_PEAK": [
        (
            "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py",
            "_build_source_derived_code",
        ),
        (
            "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py",
            "_execute_source_derived",
        ),
        (
            "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py",
            "suppressed_peak_provider",
        ),
    ],

    "V1_COMPONENT": [
        (
            "src/biohub_cell_tracking/stage12_portable_current_best.py",
            "reconstruct_mutual_cont_v1_actions",
        ),
        (
            "src/biohub_cell_tracking/stage12_portable_current_best.py",
            "_reconstruct_portable_mutual_cont_v1",
        ),
    ],

    "V2_COMPONENT": [
        (
            "src/biohub_cell_tracking/stage12_portable_current_best.py",
            "build_portable_mutual_cont_v2_actions",
        ),
        (
            "src/biohub_cell_tracking/stage12_portable_current_best.py",
            "_portable_v2_node_table",
        ),
        (
            "src/biohub_cell_tracking/stage12_portable_current_best.py",
            "_apply_portable_mutual_cont_v2",
        ),
        (
            "src/biohub_cell_tracking/stage12_portable_current_best.py",
            "apply_portable_mutual_cont_v2",
        ),
    ],

    "CLOSED_PATH_REFERENCE": [
        (
            "src/biohub_cell_tracking/stage12_post_v2_graph_provider.py",
            "post_v2_graph_provider",
        ),
        (
            "src/biohub_cell_tracking/stage12_b12_portable.py",
            "materialize_b12_dataset",
        ),
    ],
}

RISK_TERMS = {
    "MORPH": (
        "morph_candidate_provider",
        "apply_portable_morph_div_v2",
        "morphdiv",
    ),
    "FULL_B12_OR_POST_V2": (
        "materialize_b12_dataset",
        "post_v2_graph_provider",
    ),
    "HISTORICAL_ARTIFACT": (
        "read_pickle",
        "stage11_fold0_morphdiv_v2_actions_locked",
        "stage11_deployable_final_v9_policy_grid_actions",
    ),
    "EVALUATOR_OR_GT": (
        "ground_truth",
        "official_evaluate",
        "_evaluate_matched_graph",
        "per_sample_metrics",
    ),
    "EXTERNAL_IO": (
        "subprocess.",
        "Popen(",
        "torch.load",
        ".geff",
        "read_parquet",
        "read_csv",
    ),
}


# =============================================================================
# 1. Hash and AST helpers
# =============================================================================

def sha_bytes(data):
    return hashlib.sha256(data).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(obj):
    return sha_bytes(
        json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def segment(source, node):
    text = ast.get_source_segment(source, node)

    if text is not None:
        return text

    return ast.unparse(node)


def call_name(node):
    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):
        prefix = call_name(node.value)

        if prefix != "<dynamic>":
            return prefix + "." + node.attr

    return "<dynamic>"


def imports_from_ast(tree):
    result = []

    for node in tree.body:
        if isinstance(node, ast.Import):
            for alias in node.names:
                result.append(alias.name)

        elif isinstance(node, ast.ImportFrom):
            result.append(
                "." * node.level + (node.module or "")
            )

    return sorted(set(result))


def direct_calls(node):
    # Syntactic calls in the selected unit.
    # Calls in nested definitions may also appear; not a runtime trace.
    names = {
        call_name(child.func)
        for child in ast.walk(node)
        if isinstance(child, ast.Call)
    }

    return sorted(names)


def detect_risks(text):
    lower = text.lower()
    found = {}

    for group, terms in RISK_TERMS.items():
        hits = [
            term for term in terms
            if term.lower() in lower
        ]

        if hits:
            found[group] = sorted(hits)

    return found


def top_level_definitions(tree):
    definitions = {}

    for node in tree.body:
        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef),
        ):
            definitions[node.name] = node

    return definitions


def module_level_expressions(tree, source):
    # Import-time expression candidates. This is a conservative inventory,
    # not an assertion that all such expressions have side effects.
    records = []

    for node in tree.body:
        if isinstance(
            node,
            (
                ast.Import,
                ast.ImportFrom,
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        ):
            continue

        records.append({
            "line": getattr(node, "lineno", None),
            "node_kind": type(node).__name__,
            "sha256": sha_bytes(
                segment(source, node).encode("utf-8")
            ),
        })

    return records


# =============================================================================
# 2. Parent and full inventory locks
# =============================================================================

print("=" * 100)
print("NOTEBOOK 14 / 14.5A-R1")
print("NON-MORPH COMPONENT EXACT DIRECT-CALL BOUNDARY REVIEW")
print("=" * 100)

assert PARENT.is_file()
assert sha_file(PARENT) == PARENT_FILE_SHA

parent = json.loads(
    PARENT.read_text(encoding="utf-8")
)

assert parent["decision"] == "PASS"
assert parent["content_sha256"] == PARENT_CONTENT_SHA

parent_payload = dict(parent)
parent_payload.pop("content_sha256")

assert canonical_sha(parent_payload) == PARENT_CONTENT_SHA

assert parent["resolution"]["status"] == (
    "NON_MORPH_COMPONENT_STATIC_INVENTORY_COMPLETE_"
    "EXECUTION_TARGETS_REQUIRE_REVIEW"
)

assert (
    parent["resolution"]["isolated_component_runtime_authorized"]
    is False
)
assert (
    parent["resolution"]["new_pytest_execution_authorized"]
    is False
)
assert (
    parent["resolution"]["Morph_provider_authorized"]
    is False
)
assert parent["resolution"]["full_B12_authorized"] is False
assert (
    parent["resolution"]["historical_oracle_access_authorized"]
    is False
)

source_manifest = parent["source_inventory"]["source_file_sha256"]
test_manifest = parent["test_inventory"]["test_file_sha256"]

assert len(source_manifest) == 42
assert len(test_manifest) == 13

for rel, expected in source_manifest.items():
    assert sha_file(ROOT / rel) == expected, rel

for rel, expected in test_manifest.items():
    assert sha_file(ROOT / rel) == expected, rel

for rel, expected in SOURCE_LOCKS.items():
    assert sha_file(ROOT / rel) == expected

print("[PASS] PARENT_AND_ALL_55_SOURCE_TEST_LOCKS")


# =============================================================================
# 3. Inspect explicitly selected owner/source units
# =============================================================================

module_cache = {}
records = {}

for family, units in TARGETS.items():
    family_records = []

    for relpath, unit_name in units:
        assert relpath in source_manifest, relpath

        if relpath not in module_cache:
            source = (ROOT / relpath).read_text(
                encoding="utf-8"
            )
            tree = ast.parse(source, filename=relpath)

            module_cache[relpath] = {
                "source": source,
                "tree": tree,
                "definitions": top_level_definitions(tree),
                "imports": imports_from_ast(tree),
                "module_level_expressions": (
                    module_level_expressions(tree, source)
                ),
            }

        module = module_cache[relpath]
        definitions = module["definitions"]

        assert unit_name in definitions, (
            f"Selected unit missing: {relpath}::{unit_name}"
        )

        node = definitions[unit_name]
        body = segment(module["source"], node)

        calls = direct_calls(node)

        local_names = sorted({
            name
            for name in calls
            if name in definitions
        })

        unresolved_calls = sorted(
            name
            for name in calls
            if name not in definitions
        )

        risk = detect_risks(body)

        is_closed_reference = (
            family == "CLOSED_PATH_REFERENCE"
        )

        family_records.append({
            "path": relpath,
            "file_sha256": source_manifest[relpath],
            "unit": unit_name,
            "kind": type(node).__name__,
            "line_start": node.lineno,
            "line_end": node.end_lineno,
            "unit_sha256": sha_bytes(
                body.encode("utf-8")
            ),
            "module_imports": module["imports"],
            "module_level_expression_candidates": (
                module["module_level_expressions"]
            ),
            "syntactic_direct_calls": calls,
            "same_module_name_matches": local_names,
            "unresolved_or_external_call_names": unresolved_calls,
            "risk_indicators": risk,
            "classification": (
                "CLOSED_BY_PARENT_POLICY"
                if is_closed_reference
                else "REVIEW_REQUIRED_NOT_AUTHORIZED"
            ),
            "execution_authorized": False,
            "transitive_dependency_closure_verified": False,
        })

    records[family] = family_records

print("[PASS] SELECTED_SOURCE_UNIT_DIRECT_CALL_INSPECTION")


# =============================================================================
# 4. Discover P995-named owner modules, without choosing one by guessing
# =============================================================================

p995_module_candidates = []

for relpath in sorted(source_manifest):
    path = Path(relpath)

    if not (
        relpath.startswith("src/biohub_cell_tracking/")
        and "p995" in path.stem.lower()
    ):
        continue

    source = (ROOT / relpath).read_text(
        encoding="utf-8"
    )

    tree = ast.parse(source, filename=relpath)

    defs = top_level_definitions(tree)

    p995_module_candidates.append({
        "path": relpath,
        "file_sha256": source_manifest[relpath],
        "top_level_definitions": sorted(defs),
        "module_imports": imports_from_ast(tree),
        "module_level_expression_candidates": (
            module_level_expressions(tree, source)
        ),
        "execution_authorized": False,
    })

# No file-name match is an acceptable static result.
# It does not mean the P995 implementation is absent.


# =============================================================================
# 5. Existing test-hint provenance
# =============================================================================

test_hints = {}

for family in (
    "P995",
    "SUPPRESSED_PEAK",
    "V1_COMPONENT",
    "V2_COMPONENT",
):
    source_record = parent["component_families"][family]

    test_hints[family] = {
        "lexical_test_hint_count": (
            source_record["lexical_test_hint_count"]
        ),
        "hints_truncated": (
            source_record["test_hints_truncated"]
        ),
        "test_hints_are_execution_authority": False,
    }

# No inferred 14.5B test node IDs.
new_pytest_targets = []


# =============================================================================
# 6. Print the review result
# =============================================================================

print("=" * 100)
print("DIRECT-CALL REVIEW")
print("=" * 100)

for family, family_records in records.items():
    print(family)

    for record in family_records:
        print(
            " ",
            record["path"],
            "::",
            record["unit"],
        )
        print(
            "   direct call names:",
            len(record["syntactic_direct_calls"]),
        )
        print(
            "   same-module name matches:",
            len(record["same_module_name_matches"]),
        )
        print(
            "   unresolved/external call names:",
            len(record["unresolved_or_external_call_names"]),
        )
        print(
            "   risk indicators:",
            sorted(record["risk_indicators"]),
        )
        print(
            "   execution authorized:",
            False,
        )

    print()

print(
    "P995-named owner module candidates:",
    len(p995_module_candidates),
)

for item in p995_module_candidates:
    print(" ", item["path"])

print()
print("new pytest targets:", len(new_pytest_targets))
print("runtime executed:", False)
print("full B12 authorized:", False)


# =============================================================================
# 7. Freeze evidence limitations and continuation
# =============================================================================

limitations = [
    "Direct call names are syntax evidence, not a resolved runtime call graph.",
    "Same-module name matching is not proof of the actual callee under dynamic binding.",
    "Imported functions and dynamic dispatch remain unresolved.",
    "Module-level expression candidates require individual side-effect review.",
    "Lexical test hints are not independent pytest execution authority.",
    "P995 source candidates are not selected as a canonical provider by filename.",
    "Existing 14.4A regression results remain valid only for their selected scope.",
    "No Morph/full-B12 execution may be authorized from this review.",
]

status = (
    "NON_MORPH_EXACT_DIRECT_CALL_BOUNDARY_RECORDED_"
    "NEW_EXECUTION_TARGETS_NOT_AUTHORIZED"
)

next_stage = (
    "NOTEBOOK14_14_5B_"
    "ISOLATED_COMPONENT_REGRESSION_TARGET_SELECTION"
)


# =============================================================================
# 8. Formal gates
# =============================================================================

gates = {
    "parent_exactly_locked": True,
    "all_42_source_files_locked": len(source_manifest) == 42,
    "all_13_test_files_locked": len(test_manifest) == 13,
    "selected_units_located": all(
        len(records[family]) == len(TARGETS[family])
        for family in TARGETS
    ),
    "direct_calls_recorded_statically": True,
    "unresolved_calls_not_treated_as_safe": True,
    "module_level_expressions_recorded": True,
    "P995_filename_candidates_not_promoted": True,
    "new_pytest_targets_not_authorized": (
        new_pytest_targets == []
    ),
    "full_post_v2_not_authorized": True,
    "full_B12_not_authorized": True,
    "historical_oracle_closed": True,
    "CURRENT_BEST_unchanged": True,
    "pytest_not_run": True,
    "runtime_not_run": True,
    "no_Git_GT_or_evaluator": True,
    "no_source_or_test_mutation": True,
}

assert all(gates.values())

for name, ok in gates.items():
    print("PASS" if ok else "FAIL", name)


# =============================================================================
# 9. Formal artifact
# =============================================================================

payload = {
    "stage": "NOTEBOOK14_14_5A_R1",
    "title": "NON_MORPH_COMPONENT_EXACT_DIRECT_CALL_BOUNDARY_REVIEW",
    "version": 1,
    "decision": "PASS",

    "parent_14_5a": {
        "path": PARENT_REL,
        "file_sha256": PARENT_FILE_SHA,
        "content_sha256": PARENT_CONTENT_SHA,
    },

    "source_locks": SOURCE_LOCKS,
    "source_file_sha256": source_manifest,
    "test_file_sha256": test_manifest,

    "selected_unit_direct_call_review": records,
    "P995_named_module_candidates": p995_module_candidates,
    "test_hint_provenance": test_hints,

    "new_pytest_targets": new_pytest_targets,
    "limitations": limitations,

    "resolution": {
        "status": status,
        "direct_call_review_completed": True,
        "transitive_dependency_closure_verified": False,
        "new_pytest_execution_authorized": False,
        "isolated_component_runtime_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
        "next_stage": next_stage,
    },

    "semantic_gates": gates,

    "claim_boundary": {
        "review": "STATIC_DIRECT_CALL_ONLY",
        "runtime_execution": False,
        "pytest_execution": False,
        "historical_oracle_access": False,
        "GT_access": False,
        "evaluator_execution": False,
        "performance_evidence": False,
        "Git_operation": False,
        "source_mutation": False,
        "test_mutation": False,
    },

    "next_stage": next_stage,
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

assert OUT.parent.is_dir()

if OUT.exists():
    assert OUT.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"
else:
    OUT.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"


# =============================================================================
# 10. Final immutability verification
# =============================================================================

assert sha_file(PARENT) == PARENT_FILE_SHA

for relpath, expected in source_manifest.items():
    assert sha_file(ROOT / relpath) == expected

for relpath, expected in test_manifest.items():
    assert sha_file(ROOT / relpath) == expected


# =============================================================================
# 11. Final
# =============================================================================

print("=" * 100)
print("NOTEBOOK 14 / 14.5A-R1 PASS")
print("=" * 100)

print("selected source units:", sum(map(len, records.values())))
print("P995-named source modules:", len(p995_module_candidates))
print("new pytest targets:", 0)

print("full post-V2: BLOCKED")
print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")

print("STATUS:", status)
print("NEXT:", next_stage)

print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(OUT))
print("artifact mode:", mode)
```

```python
# =============================================================================
# Notebook 14 / 14.5B
# ISOLATED COMPONENT REGRESSION TARGET SELECTION
#
# Static crosswalk only:
#   14.5A-R1 exact source units
#       ↔
#   14.4A frozen test-node inventory
#
# No pytest, runtime, project-module imports, GT, oracle, evaluator or Git.
# No source or test modifications.
# =============================================================================

import ast
import hashlib
import json
from collections import Counter
from pathlib import Path

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

R1_REL = (
    "reports/"
    "notebook14_14_5a_r1_"
    "non_morph_component_exact_direct_call_boundary_review_v1.json"
)

BASELINE_REL = (
    "reports/"
    "notebook14_14_4a_"
    "non_morph_static_and_regression_readiness_audit_v1.json"
)

OUT_REL = (
    "reports/"
    "notebook14_14_5b_"
    "isolated_component_regression_target_selection_v1.json"
)

R1_FILE_SHA = (
    "883f666403dc113a4fa3b8e53042cc23eec116f6929fce7600a10d551edc46b3"
)

R1_CONTENT_SHA = (
    "72e2c46d6d1f51f449a29098da38662cd65ff61bfb16583e72a211e75c9147a9"
)

BASELINE_FILE_SHA = (
    "de8e43603375dd2cf8aa26417b89b3c47e9741b7bbb786da43eead125053aebd"
)

BASELINE_CONTENT_SHA = (
    "ceaad77b689d094e8ac4c68317858ac61954e903b8b869f7b80e4a03d673a8d9"
)

R1 = ROOT / R1_REL
BASELINE = ROOT / BASELINE_REL
OUT = ROOT / OUT_REL


def sha_bytes(value):
    return hashlib.sha256(value).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(value):
    return sha_bytes(
        json.dumps(
            value,
            sort_keys=True,
            ensure_ascii=False,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def verified_artifact(path, file_sha, content_sha):
    assert path.is_file(), path
    assert sha_file(path) == file_sha, path

    obj = json.loads(
        path.read_text(encoding="utf-8")
    )

    assert obj["decision"] == "PASS"
    assert obj["content_sha256"] == content_sha

    body = dict(obj)
    body.pop("content_sha256")

    assert canonical_sha(body) == content_sha

    return obj


def ast_source(source, node):
    value = ast.get_source_segment(source, node)

    if value is not None:
        return value

    return ast.unparse(node)


# =============================================================================
# 1. Verify formal parents
# =============================================================================

print("=" * 100)
print("NOTEBOOK 14 :: 14.5B")
print("ISOLATED COMPONENT REGRESSION TARGET SELECTION")
print("=" * 100)

r1 = verified_artifact(
    R1,
    R1_FILE_SHA,
    R1_CONTENT_SHA,
)

baseline = verified_artifact(
    BASELINE,
    BASELINE_FILE_SHA,
    BASELINE_CONTENT_SHA,
)

assert r1["resolution"]["status"] == (
    "NON_MORPH_EXACT_DIRECT_CALL_BOUNDARY_RECORDED_"
    "NEW_EXECUTION_TARGETS_NOT_AUTHORIZED"
)

assert (
    r1["resolution"]["new_pytest_execution_authorized"]
    is False
)

assert (
    r1["resolution"]["transitive_dependency_closure_verified"]
    is False
)

assert (
    baseline["resolution"]["safe_regression_status"]
    == "SAFE_NON_MORPH_PYTEST_TARGETS_PASS"
)

assert baseline["pytest_execution"]["returncode"] == 0
assert baseline["pytest_execution"]["node_count"] == 47

print("[PASS] R1_AND_REGRESSION_PARENT_LOCKS")


# =============================================================================
# 2. Verify every source and test file remains unchanged
# =============================================================================

source_hashes = r1["source_file_sha256"]
test_hashes = r1["test_file_sha256"]

assert len(source_hashes) == 42
assert len(test_hashes) == 13

for rel, expected in source_hashes.items():
    assert sha_file(ROOT / rel) == expected, rel

for rel, expected in test_hashes.items():
    assert sha_file(ROOT / rel) == expected, rel

print("[PASS] ALL_42_SOURCE_AND_13_TEST_FILE_LOCKS")


# =============================================================================
# 3. Reconstruct exact AST test-node source records
# =============================================================================

test_source = {}

for rel in sorted(test_hashes):
    source = (ROOT / rel).read_text(
        encoding="utf-8"
    )

    tree = ast.parse(source, filename=rel)

    for node in tree.body:
        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef),
        ):
            if node.name.startswith("test_"):
                test_source[
                    rel + "::" + node.name
                ] = ast_source(source, node)

        elif isinstance(node, ast.ClassDef):
            if not node.name.startswith("Test"):
                continue

            for child in node.body:
                if isinstance(
                    child,
                    (ast.FunctionDef, ast.AsyncFunctionDef),
                ) and child.name.startswith("test_"):

                    nodeid = (
                        rel + "::" + node.name
                        + "::" + child.name
                    )

                    test_source[nodeid] = ast_source(
                        source,
                        child,
                    )

prior_records = baseline["test_nodes"]

assert len(prior_records) == 376
assert len(test_source) == 376

classification = {}

for rec in prior_records:
    nodeid = rec["nodeid"]

    assert nodeid in test_source
    assert sha_bytes(
        test_source[nodeid].encode("utf-8")
    ) == rec["source_sha256"]

    classification[nodeid] = rec["classification"]

assert dict(Counter(classification.values())) == {
    "SAFE_TO_RUN": 47,
    "CONDITIONAL_SYNTHETIC": 18,
    "BLOCKED_BOUNDARY": 6,
    "REVIEW_REQUIRED": 305,
}

accepted_ids = baseline["safe_nodeids"]

assert len(accepted_ids) == 47
assert len(set(accepted_ids)) == 47

assert all(
    classification[nodeid] == "SAFE_TO_RUN"
    for nodeid in accepted_ids
)

print("[PASS] ALL_376_TEST_NODE_SOURCE_HASHES")


# =============================================================================
# 4. Crosswalk exact source-unit names to existing test bodies
# =============================================================================

# A lexical match is evidence that a test mentions a source unit.
# It is NOT evidence of an actual runtime call or safe fixture closure.

source_units = r1["selected_unit_direct_call_review"]

assert sum(len(v) for v in source_units.values()) == 12

crosswalk = {}

for family, units in source_units.items():
    entries = []

    for unit in units:
        symbol = unit["unit"]

        matching = [
            nodeid
            for nodeid, body in test_source.items()
            if symbol in body
        ]

        by_class = {
            label: sorted(
                nodeid
                for nodeid in matching
                if classification[nodeid] == label
            )
            for label in (
                "SAFE_TO_RUN",
                "CONDITIONAL_SYNTHETIC",
                "BLOCKED_BOUNDARY",
                "REVIEW_REQUIRED",
            )
        }

        entries.append({
            "path": unit["path"],
            "unit": symbol,
            "unit_sha256": unit["unit_sha256"],
            "direct_call_review_classification": (
                unit["classification"]
            ),
            "unresolved_or_external_call_count": len(
                unit["unresolved_or_external_call_names"]
            ),
            "risk_indicators": unit["risk_indicators"],
            "matching_test_node_count": len(matching),
            "lexical_test_nodeids_by_class": by_class,
            "lexical_match_is_runtime_call_proof": False,
            "new_execution_authorized": False,
        })

    crosswalk[family] = entries

print("[PASS] SOURCE_UNIT_TEST_LEXICAL_CROSSWALK")


# =============================================================================
# 5. Target-selection decision
# =============================================================================

# 47 node IDs retain their previously observed regression status.
# Their 14.4A PASS is not a new result in this stage.
#
# The 329 other nodes stay deferred. A lexical match or direct-call
# source review does not authorize their execution.

new_authorized_nodeids = []

conditional_count = sum(
    classification[nodeid] == "CONDITIONAL_SYNTHETIC"
    for nodeid in classification
)

blocked_count = sum(
    classification[nodeid] == "BLOCKED_BOUNDARY"
    for nodeid in classification
)

review_count = sum(
    classification[nodeid] == "REVIEW_REQUIRED"
    for nodeid in classification
)

assert conditional_count == 18
assert blocked_count == 6
assert review_count == 305
assert new_authorized_nodeids == []

status = (
    "ISOLATED_COMPONENT_TEST_CROSSWALK_RECORDED_"
    "NO_NEW_EXECUTION_TARGETS_AUTHORIZED"
)

next_stage = (
    "NOTEBOOK14_14_5C_"
    "PORTABLE_COMPONENT_CONTRACT_BOUNDARY"
)


# =============================================================================
# 6. Print evidence summary
# =============================================================================

print("=" * 100)
print("COMPONENT TEST CROSSWALK")
print("=" * 100)

for family, entries in crosswalk.items():
    print(family)

    for rec in entries:
        classes = rec["lexical_test_nodeids_by_class"]

        print(
            " ",
            rec["unit"],
            "| lexical tests:",
            rec["matching_test_node_count"],
            "| accepted-baseline:",
            len(classes["SAFE_TO_RUN"]),
            "| conditional:",
            len(classes["CONDITIONAL_SYNTHETIC"]),
            "| blocked:",
            len(classes["BLOCKED_BOUNDARY"]),
            "| review:",
            len(classes["REVIEW_REQUIRED"]),
        )

    print()

print("Previously accepted node IDs:", len(accepted_ids))
print("New authorized node IDs:", len(new_authorized_nodeids))
print("Deferred node IDs:", 329)
print("New pytest executed:", False)
print("Runtime executed:", False)


# =============================================================================
# 7. Formal semantic gates
# =============================================================================

gates = {
    "R1_parent_exactly_locked": True,
    "14_4A_regression_parent_exactly_locked": True,
    "all_42_source_files_locked": len(source_hashes) == 42,
    "all_13_test_files_locked": len(test_hashes) == 13,
    "all_376_test_node_hashes_locked": len(test_source) == 376,
    "twelve_source_units_crosswalked": (
        sum(len(v) for v in crosswalk.values()) == 12
    ),
    "previously_accepted_47_preserved": len(accepted_ids) == 47,
    "deferred_329_preserved": (
        conditional_count + blocked_count + review_count == 329
    ),
    "lexical_matches_not_promoted_to_call_proof": True,
    "new_execution_targets_not_authorized": (
        new_authorized_nodeids == []
    ),
    "full_post_v2_remains_blocked": True,
    "full_B12_remains_blocked": True,
    "historical_oracle_closed": True,
    "CURRENT_BEST_unchanged": True,
    "no_new_pytest_or_runtime": True,
    "no_Git_GT_evaluator_or_performance": True,
    "no_source_or_test_mutation": True,
}

assert all(gates.values())

for name, passed in gates.items():
    print("PASS" if passed else "FAIL", name)


# =============================================================================
# 8. Formal artifact
# =============================================================================

payload = {
    "stage": "NOTEBOOK14_14_5B",
    "title": "ISOLATED_COMPONENT_REGRESSION_TARGET_SELECTION",
    "version": 1,
    "decision": "PASS",

    "parent_14_5a_r1": {
        "path": R1_REL,
        "file_sha256": R1_FILE_SHA,
        "content_sha256": R1_CONTENT_SHA,
    },

    "regression_parent_14_4a": {
        "path": BASELINE_REL,
        "file_sha256": BASELINE_FILE_SHA,
        "content_sha256": BASELINE_CONTENT_SHA,
    },

    "source_file_sha256": source_hashes,
    "test_file_sha256": test_hashes,

    "source_unit_test_crosswalk": crosswalk,

    "regression_selection": {
        "previously_accepted_nodeids": accepted_ids,
        "parent_pytest_returncode": 0,
        "user_terminal_reported_instances_passed": 50,
        "parent_artifact_independent_instance_count_verified": False,
        "new_authorized_nodeids": new_authorized_nodeids,
        "new_pytest_executed": False,
    },

    "deferred": {
        "conditional_synthetic": 18,
        "blocked_by_screening": 6,
        "review_required": 305,
        "total": 329,
    },

    "limitations": [
        "Symbol references in tests are lexical evidence only.",
        "Direct-call inspection does not prove transitive dependency safety.",
        "Fixtures, imports, dynamic calls and runtime data remain unverified.",
        "No test outside the accepted 14.4A scope is authorized.",
        "No new pytest or runtime execution occurs in this stage.",
        "No full B12 or structural fidelity conclusion is established.",
    ],

    "resolution": {
        "status": status,
        "existing_regression_baseline_preserved": True,
        "new_pytest_execution_authorized": False,
        "isolated_component_runtime_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
        "next_stage": next_stage,
    },

    "semantic_gates": gates,

    "claim_boundary": {
        "test_crosswalk": "STATIC_LEXICAL_ONLY",
        "previous_regression": "LIMITED_PARENT_ACCEPTANCE",
        "new_pytest_execution": False,
        "runtime_execution": False,
        "source_mutation": False,
        "test_mutation": False,
        "Git_operation": False,
        "GT_access": False,
        "evaluator_execution": False,
        "performance_evidence": False,
    },

    "next_stage": next_stage,
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        sort_keys=True,
        ensure_ascii=False,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

assert OUT.parent.is_dir()

if OUT.exists():
    assert OUT.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"

else:
    OUT.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"


# =============================================================================
# 9. Final immutability check
# =============================================================================

assert sha_file(R1) == R1_FILE_SHA
assert sha_file(BASELINE) == BASELINE_FILE_SHA

for rel, expected in source_hashes.items():
    assert sha_file(ROOT / rel) == expected

for rel, expected in test_hashes.items():
    assert sha_file(ROOT / rel) == expected


# =============================================================================
# 10. Final
# =============================================================================

print("=" * 100)
print("NOTEBOOK 14 / 14.5B PASS")
print("=" * 100)

print("Existing regression: 47 selected node IDs")
print("Parent pytest return code: 0")
print("New authorized tests: 0")
print("Deferred tests: 329")

print("Full post-V2: BLOCKED")
print("Full B12: BLOCKED")
print("Historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")

print("STATUS:", status)
print("NEXT:", next_stage)

print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(OUT))
print("artifact mode:", mode)
```

```python
# =============================================================================
# Notebook 14 / 14.5C
# PORTABLE COMPONENT CONTRACT BOUNDARY
#
# Parents:
#   14.5B PASS
#   14.5A-R1 PASS
#
# Static documentation of existing components ONLY.
# No project imports, pytest, runtime, pickle, GT, oracle, evaluator or Git.
# No source/test mutation.
# =============================================================================

import ast
import hashlib
import json
from pathlib import Path


ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

B_REL = (
    "reports/"
    "notebook14_14_5b_"
    "isolated_component_regression_target_selection_v1.json"
)

R1_REL = (
    "reports/"
    "notebook14_14_5a_r1_"
    "non_morph_component_exact_direct_call_boundary_review_v1.json"
)

OUT_REL = (
    "reports/"
    "notebook14_14_5c_"
    "portable_component_contract_boundary_v1.json"
)

B_FILE_SHA = (
    "66685c518c75c7e282ed1fe7b58db7da4f1b6d35d51d2c7211cadcdebf8ab1db"
)

B_CONTENT_SHA = (
    "567c5ee48fba58fb26472f31d3e62b6871804bd07e108b9a0f2bfaa23e09d62d"
)

R1_FILE_SHA = (
    "883f666403dc113a4fa3b8e53042cc23eec116f6929fce7600a10d551edc46b3"
)

R1_CONTENT_SHA = (
    "72e2c46d6d1f51f449a29098da38662cd65ff61bfb16583e72a211e75c9147a9"
)

B_PATH = ROOT / B_REL
R1_PATH = ROOT / R1_REL
OUT = ROOT / OUT_REL


# =============================================================================
# 1. Helpers
# =============================================================================

def sha_bytes(value):
    return hashlib.sha256(value).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(value):
    return sha_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_locked(path, file_sha, content_sha):
    assert path.is_file(), path
    assert sha_file(path) == file_sha, path

    obj = json.loads(path.read_text(encoding="utf-8"))

    assert obj["decision"] == "PASS"
    assert obj["content_sha256"] == content_sha

    payload = dict(obj)
    payload.pop("content_sha256")

    assert canonical_sha(payload) == content_sha

    return obj


def source_segment(source, node):
    result = ast.get_source_segment(source, node)

    if result is not None:
        return result

    return ast.unparse(node)


def top_level_definitions(tree):
    return {
        node.name: node
        for node in tree.body
        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef,
                ast.ClassDef,
            ),
        )
    }


def function_signature(node):
    return {
        "parameters_ast": ast.unparse(node.args),
        "return_annotation_ast": (
            ast.unparse(node.returns)
            if node.returns is not None
            else None
        ),
        "decorators_ast": [
            ast.unparse(decorator)
            for decorator in node.decorator_list
        ],
        "signature_status": (
            "STATIC_AST_ONLY_NOT_RUNTIME_VALIDATED"
        ),
    }


def class_signature(node):
    members = []

    for child in node.body:
        if not isinstance(
            child,
            (ast.FunctionDef, ast.AsyncFunctionDef),
        ):
            continue

        if child.name not in ("__init__", "__call__"):
            continue

        members.append({
            "name": child.name,
            "line": child.lineno,
            **function_signature(child),
        })

    return {
        "bases_ast": [
            ast.unparse(base)
            for base in node.bases
        ],
        "keywords_ast": [
            ast.unparse(keyword)
            for keyword in node.keywords
        ],
        "constructor_and_call_methods": members,
        "signature_status": (
            "STATIC_AST_ONLY_NOT_RUNTIME_VALIDATED"
        ),
    }


# =============================================================================
# 2. Exact parent locks
# =============================================================================

print("=" * 100)
print("NOTEBOOK 14 :: 14.5C")
print("PORTABLE COMPONENT CONTRACT BOUNDARY")
print("=" * 100)

b = load_locked(
    B_PATH,
    B_FILE_SHA,
    B_CONTENT_SHA,
)

r1 = load_locked(
    R1_PATH,
    R1_FILE_SHA,
    R1_CONTENT_SHA,
)

assert b["resolution"]["status"] == (
    "ISOLATED_COMPONENT_TEST_CROSSWALK_RECORDED_"
    "NO_NEW_EXECUTION_TARGETS_AUTHORIZED"
)

assert r1["resolution"]["status"] == (
    "NON_MORPH_EXACT_DIRECT_CALL_BOUNDARY_RECORDED_"
    "NEW_EXECUTION_TARGETS_NOT_AUTHORIZED"
)

assert b["resolution"]["new_pytest_execution_authorized"] is False
assert b["resolution"]["isolated_component_runtime_authorized"] is False
assert b["resolution"]["full_B12_authorized"] is False

assert r1["resolution"]["transitive_dependency_closure_verified"] is False
assert r1["resolution"]["historical_oracle_access_authorized"] is False

source_manifest = b["source_file_sha256"]
test_manifest = b["test_file_sha256"]

assert source_manifest == r1["source_file_sha256"]
assert test_manifest == r1["test_file_sha256"]

assert len(source_manifest) == 42
assert len(test_manifest) == 13

for relpath, expected in source_manifest.items():
    assert sha_file(ROOT / relpath) == expected, relpath

for relpath, expected in test_manifest.items():
    assert sha_file(ROOT / relpath) == expected, relpath

assert len(b["regression_selection"]["previously_accepted_nodeids"]) == 47
assert b["regression_selection"]["parent_pytest_returncode"] == 0
assert b["regression_selection"]["new_authorized_nodeids"] == []

assert b["deferred"] == {
    "conditional_synthetic": 18,
    "blocked_by_screening": 6,
    "review_required": 305,
    "total": 329,
}

print("[PASS] EXACT_PARENTS_AND_55_FILE_LOCKS")


# =============================================================================
# 3. Static signature extraction for the 12 frozen source units
# =============================================================================

review = r1["selected_unit_direct_call_review"]
crosswalk = b["source_unit_test_crosswalk"]

assert set(review) == set(crosswalk)

module_cache = {}
component_contracts = {}

for family in sorted(review):
    reviewed_units = review[family]
    linked_units = crosswalk[family]

    assert len(reviewed_units) == len(linked_units)

    contracts = []

    for reviewed, linked in zip(reviewed_units, linked_units):
        assert reviewed["path"] == linked["path"]
        assert reviewed["unit"] == linked["unit"]

        assert reviewed["unit_sha256"] == linked["unit_sha256"]

        relpath = reviewed["path"]
        unit_name = reviewed["unit"]

        if relpath not in module_cache:
            source = (ROOT / relpath).read_text(
                encoding="utf-8"
            )

            tree = ast.parse(source, filename=relpath)

            module_cache[relpath] = {
                "source": source,
                "definitions": top_level_definitions(tree),
            }

        module = module_cache[relpath]

        assert unit_name in module["definitions"], (
            relpath,
            unit_name,
        )

        node = module["definitions"][unit_name]
        exact_source = source_segment(module["source"], node)

        calculated_unit_sha = sha_bytes(
            exact_source.encode("utf-8")
        )

        assert calculated_unit_sha == reviewed["unit_sha256"]

        if isinstance(node, ast.ClassDef):
            signature = class_signature(node)
        else:
            signature = function_signature(node)

        lexical_links = linked[
            "lexical_test_nodeids_by_class"
        ]

        assert set(lexical_links) == {
            "SAFE_TO_RUN",
            "CONDITIONAL_SYNTHETIC",
            "BLOCKED_BOUNDARY",
            "REVIEW_REQUIRED",
        }

        assert (
            sum(len(ids) for ids in lexical_links.values())
            == linked["matching_test_node_count"]
        )

        closed_path = family == "CLOSED_PATH_REFERENCE"

        # Existing PASS applies only to previously selected test node IDs.
        # It does not establish that this unit is fully runtime-validated.
        existing_safe_links = lexical_links["SAFE_TO_RUN"]

        contract = {
            "path": relpath,
            "file_sha256": source_manifest[relpath],
            "unit": unit_name,
            "unit_kind": reviewed["kind"],
            "unit_sha256": calculated_unit_sha,
            "line_start": reviewed["line_start"],
            "line_end": reviewed["line_end"],

            "signature": signature,

            "static_dependencies": {
                "module_imports": reviewed["module_imports"],
                "syntactic_direct_calls": (
                    reviewed["syntactic_direct_calls"]
                ),
                "same_module_name_matches": (
                    reviewed["same_module_name_matches"]
                ),
                "unresolved_or_external_call_names": (
                    reviewed["unresolved_or_external_call_names"]
                ),
                "risk_indicators": reviewed["risk_indicators"],
                "transitive_dependencies_verified": False,
            },

            "test_evidence": {
                "lexical_nodeids_by_class": lexical_links,
                "previously_accepted_safe_nodeids": (
                    existing_safe_links
                ),
                "accepted_regression_is_unit_complete": False,
                "new_tests_executed": False,
                "new_tests_authorized": False,
            },

            "contract_status": (
                "CLOSED_PATH_REFERENCE"
                if closed_path
                else "STATIC_SIGNATURE_AND_DEPENDENCY_RECORD"
            ),

            "runtime_input_output_semantics_verified": False,
            "isolated_runtime_authorized": False,
            "production_runtime_authorized": False,
            "full_B12_equivalence_authorized": False,
        }

        contracts.append(contract)

    component_contracts[family] = contracts

assert sum(
    len(values)
    for values in component_contracts.values()
) == 12

print("[PASS] TWELVE_STATIC_COMPONENT_CONTRACTS")


# =============================================================================
# 4. Evidence status and blocking chain
# =============================================================================

evidence_status = {
    "source_identity": "EXACT_SHA_VERIFIED",
    "selected_unit_identity": "EXACT_SHA_VERIFIED",
    "function_or_class_signatures": "STATIC_AST_RECORDED",
    "direct_call_names": "STATIC_SYNTAX_RECORDED",
    "transitive_dependencies": "NOT_VERIFIED",
    "runtime_input_output_semantics": "NOT_VERIFIED",
    "previously_selected_regression": (
        "47_NODEIDS_PARENT_PYTEST_RETURNCODE_ZERO"
    ),
    "previous_terminal_result": (
        "50_PASSED_USER_TERMINAL_LOG"
    ),
    "new_pytest_execution": "NONE",
    "new_execution_targets": "NONE",
    "full_post_v2": "BLOCKED",
    "full_B12": "BLOCKED",
    "historical_oracle": "CLOSED",
    "structural_fidelity": "NOT_VERIFIED",
    "CURRENT_BEST": "UNCHANGED",
}

blocking_chain = [
    "CONCRETE_MORPH_PROVIDER_MISSING_FRESH_SOURCE_AUTHORITY",
    "FRESH_POST_V2_RUNTIME_BLOCKED_BY_MORPH",
    "FULL_B12_BLOCKED_BY_POST_V2_MORPH_DEPENDENCY",
]

status = (
    "PORTABLE_COMPONENT_STATIC_CONTRACT_BOUNDARY_FROZEN_"
    "FULL_RUNTIME_REMAINS_BLOCKED"
)

next_stage = (
    "NOTEBOOK14_14_5D_"
    "B12_BLOCKED_DELIVERY_MANIFEST"
)


# =============================================================================
# 5. Summary
# =============================================================================

print("=" * 100)
print("STATIC COMPONENT CONTRACT SUMMARY")
print("=" * 100)

for family, contracts in component_contracts.items():
    print(family)

    for item in contracts:
        safe_count = len(
            item["test_evidence"]["previously_accepted_safe_nodeids"]
        )

        print(
            " ",
            item["unit"],
            "| type:",
            item["unit_kind"],
            "| existing safe test links:",
            safe_count,
            "| runtime authorized:",
            False,
        )

    print()

print("New pytest run:", False)
print("New runtime executed:", False)
print("Full B12:", "BLOCKED")
print("Historical oracle:", "CLOSED")


# =============================================================================
# 6. Final immutability and semantic gates
# =============================================================================

for relpath, expected in source_manifest.items():
    assert sha_file(ROOT / relpath) == expected

for relpath, expected in test_manifest.items():
    assert sha_file(ROOT / relpath) == expected

assert sha_file(B_PATH) == B_FILE_SHA
assert sha_file(R1_PATH) == R1_FILE_SHA

gates = {
    "14_5B_parent_exactly_locked": True,
    "14_5A_R1_parent_exactly_locked": True,
    "all_42_source_files_locked": len(source_manifest) == 42,
    "all_13_test_files_locked": len(test_manifest) == 13,
    "all_12_unit_hashes_reverified": True,
    "static_signatures_recorded": True,
    "previous_regression_scope_preserved": True,
    "transitive_call_safety_not_claimed": True,
    "runtime_semantics_not_prematurely_claimed": True,
    "new_tests_not_authorized": True,
    "full_post_v2_remains_blocked": True,
    "full_B12_remains_blocked": True,
    "historical_oracle_closed": True,
    "structural_fidelity_not_claimed": True,
    "CURRENT_BEST_unchanged": True,
    "no_pytest_or_runtime_execution": True,
    "no_Git_GT_evaluator_or_performance": True,
    "no_source_or_test_mutation": True,
}

assert all(gates.values())

for name, passed in gates.items():
    print("PASS" if passed else "FAIL", name)


# =============================================================================
# 7. Formal artifact
# =============================================================================

payload = {
    "stage": "NOTEBOOK14_14_5C",
    "title": "PORTABLE_COMPONENT_CONTRACT_BOUNDARY",
    "version": 1,
    "decision": "PASS",

    "parent_14_5b": {
        "path": B_REL,
        "file_sha256": B_FILE_SHA,
        "content_sha256": B_CONTENT_SHA,
    },

    "parent_14_5a_r1": {
        "path": R1_REL,
        "file_sha256": R1_FILE_SHA,
        "content_sha256": R1_CONTENT_SHA,
    },

    "source_file_sha256": source_manifest,
    "test_file_sha256": test_manifest,

    "component_contracts": component_contracts,

    "evidence_status": evidence_status,
    "blocking_chain": blocking_chain,

    "limitations": [
        "Signatures are extracted from AST, not exercised at runtime.",
        "Annotations are recorded as source text, not validated schemas.",
        "Direct-call names are not a complete transitive call graph.",
        "Lexical test links do not prove actual runtime call relationships.",
        "Previous regression PASS applies only to its selected scope.",
        "No new pytest or isolated runtime target has been authorized.",
        "Full B12 execution and structural fidelity remain blocked.",
    ],

    "resolution": {
        "status": status,
        "static_component_contracts_recorded": 12,
        "runtime_input_output_semantics_verified": False,
        "isolated_component_runtime_authorized": False,
        "new_pytest_execution_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "scientific_result": "NONE",
        "performance_result": "NONE",
        "CURRENT_BEST_replaced": False,
        "next_stage": next_stage,
    },

    "semantic_gates": gates,

    "claim_boundary": {
        "contract_type": "STATIC_AST_AND_PROVENANCE_ONLY",
        "production_portability_verified": False,
        "runtime_execution": False,
        "pytest_execution": False,
        "historical_oracle_access": False,
        "GT_access": False,
        "evaluator_execution": False,
        "performance_evidence": False,
        "Git_operation": False,
        "source_mutation": False,
        "test_mutation": False,
    },

    "next_stage": next_stage,
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

assert OUT.parent.is_dir()

if OUT.exists():
    assert OUT.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"
else:
    OUT.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"


# =============================================================================
# 8. Final
# =============================================================================

assert sha_file(B_PATH) == B_FILE_SHA
assert sha_file(R1_PATH) == R1_FILE_SHA

print("=" * 100)
print("NOTEBOOK 14 / 14.5C PASS")
print("=" * 100)

print("Static component contracts:", 12)
print("Transitive dependency verification:", False)
print("New execution targets:", 0)

print("Full post-V2: BLOCKED")
print("Full B12: BLOCKED")
print("Historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")

print("STATUS:", status)
print("NEXT:", next_stage)

print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(OUT))
print("artifact mode:", mode)
```

```python
# ============================================================
# Notebook 14 / 14.6C
# IMPORT BOUNDARY + EMPTY-DATASET GUARD ORDER
#
# READ ONLY. No project imports, runtime, pytest, exec,
# historical oracle, GT, evaluator, Git or file writes.
# ============================================================

import ast
import hashlib
import json
from pathlib import Path

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

FILES = {
    "manifest": (
        "reports/"
        "notebook14_14_5d_b12_blocked_delivery_manifest_v1.json"
    ),
    "suppressed": (
        "src/biohub_cell_tracking/"
        "stage12_suppressed_peak_provider.py"
    ),
    "portable": (
        "src/biohub_cell_tracking/"
        "stage12_portable_current_best.py"
    ),
    "test": "tests/test_stage12_suppressed_peak_provider.py",
}

EXPECTED_SHA = {
    "manifest": (
        "4aaaaa4f4043f74a82ac948938e078372a9e73c61d14fe423108a6f3a6855e8a"
    ),
    "suppressed": (
        "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
    ),
    "portable": (
        "b0b598da12bd62d321448b5829adfa49dbbc68d8a898c9d590e8981d8fc706e0"
    ),
    "test": (
        "eac928e54a26c086e531be7d38eeb46c67bcbba2d4ddd20cd00f255dcb63af4a"
    ),
}


def sha_file(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def dotted(node):
    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):
        base = dotted(node.value)

        if base != "<dynamic>":
            return base + "." + node.attr

    return "<dynamic>"


def calls_in(node):
    return sorted({
        dotted(child.func)
        for child in ast.walk(node)
        if isinstance(child, ast.Call)
    })


def import_records(tree):
    records = []

    for node in tree.body:
        if isinstance(node, ast.Import):
            for alias in node.names:
                records.append({
                    "line": node.lineno,
                    "module": alias.name,
                    "alias": alias.asname,
                })

        elif isinstance(node, ast.ImportFrom):
            records.append({
                "line": node.lineno,
                "module": (
                    "." * node.level + (node.module or "")
                ),
                "names": [
                    alias.name for alias in node.names
                ],
            })

    return records


def import_time_records(tree, source):
    records = []

    for node in tree.body:
        if isinstance(node, (ast.Import, ast.ImportFrom)):
            continue

        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef),
        ):
            expressions = list(node.decorator_list)

            if isinstance(
                node,
                (ast.FunctionDef, ast.AsyncFunctionDef),
            ):
                expressions += list(node.args.defaults)
                expressions += [
                    value
                    for value in node.args.kw_defaults
                    if value is not None
                ]

            names = sorted({
                name
                for expression in expressions
                for name in calls_in(expression)
            })

            if expressions:
                records.append({
                    "line": node.lineno,
                    "definition": node.name,
                    "definition_kind": type(node).__name__,
                    "decorator_or_default_call_names": names,
                    "expressions_are_not_proven_safe": True,
                })

            continue

        text = ast.get_source_segment(source, node) or ""

        records.append({
            "line": getattr(node, "lineno", None),
            "kind": type(node).__name__,
            "call_names": calls_in(node),
            "source_preview": text[:160],
        })

    return records


paths = {
    name: ROOT / relative
    for name, relative in FILES.items()
}

print("=" * 96)
print("NOTEBOOK 14 / 14.6C")
print("IMPORT BOUNDARY AND EMPTY-DATASET GUARD ORDER")
print("=" * 96)

for name, path in paths.items():
    assert path.is_file(), name
    assert not path.is_symlink(), name
    assert sha_file(path) == EXPECTED_SHA[name], name

manifest = json.loads(
    paths["manifest"].read_text(encoding="utf-8")
)

assert manifest["decision"] == "PASS"

assert (
    manifest["resolution"]["full_B12_delivery_status"]
    == "BLOCKED"
)

assert (
    manifest["resolution"][
        "historical_oracle_access_authorized"
    ] is False
)

print("[PASS] ALL_FOUR_FILE_LOCKS")

source = paths["suppressed"].read_text(encoding="utf-8")
tree = ast.parse(source, filename=FILES["suppressed"])

definitions = {
    node.name: node
    for node in tree.body
    if isinstance(
        node,
        (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef),
    )
}

# ------------------------------------------------------------
# A. Print complete public-entrypoint source, not a paraphrase.
# ------------------------------------------------------------

print()
print("=" * 96)
print("A. SUPPRESSED-PEAK PUBLIC ENTRYPOINT")
print("=" * 96)

name = "suppressed_peak_provider"

assert name in definitions

node = definitions[name]

assert isinstance(
    node,
    (ast.FunctionDef, ast.AsyncFunctionDef),
)

lines = source.splitlines()

print(
    "function lines:",
    node.lineno,
    "-",
    node.end_lineno,
)

for lineno in range(node.lineno, node.end_lineno + 1):
    print(
        f"{lineno:5d} | {lines[lineno - 1]}"
    )

print()
print("entrypoint syntactic call names:")
print(calls_in(node))


# ------------------------------------------------------------
# B. Imported module inspection, without importing it.
# ------------------------------------------------------------

for label in ("suppressed", "portable"):
    path = paths[label]
    text = path.read_text(encoding="utf-8")
    module_tree = ast.parse(text, filename=FILES[label])

    print()
    print("=" * 96)
    print("B. STATIC MODULE IMPORT REVIEW:", label)
    print("=" * 96)

    imports = import_records(module_tree)

    print("import record count:", len(imports))

    for record in imports:
        print(
            json.dumps(
                record,
                ensure_ascii=False,
            )
        )

    records = import_time_records(module_tree, text)

    print()
    print("module-level expression / decorator / default records:")
    print("record count:", len(records))

    # Print every recorded import-time expression candidate.
    # This is an inventory, not proof that the expression is unsafe.
    for record in records:
        print(
            json.dumps(
                record,
                ensure_ascii=False,
            )
        )


# ------------------------------------------------------------
# C. Test-module import and collection-time inspection.
# ------------------------------------------------------------

test_source = paths["test"].read_text(encoding="utf-8")

test_tree = ast.parse(
    test_source,
    filename=FILES["test"],
)

print()
print("=" * 96)
print("C. TEST MODULE COLLECTION BOUNDARY")
print("=" * 96)

for record in import_records(test_tree):
    print(
        "import:",
        json.dumps(record, ensure_ascii=False),
    )

for record in import_time_records(test_tree, test_source):
    print(
        "collection-time candidate:",
        json.dumps(record, ensure_ascii=False),
    )


# ------------------------------------------------------------
# D. Read pytest configuration without starting pytest.
# ------------------------------------------------------------

print()
print("=" * 96)
print("D. PYTEST CONFIGURATION")
print("=" * 96)

pyproject = ROOT / "pyproject.toml"

if pyproject.is_file() and not pyproject.is_symlink():
    print("pyproject SHA:", sha_file(pyproject))

    try:
        import tomllib

        config = tomllib.loads(
            pyproject.read_text(encoding="utf-8")
        )

        pytest_config = (
            config.get("tool", {})
            .get("pytest", {})
            .get("ini_options", {})
        )

        print(
            json.dumps(
                pytest_config,
                ensure_ascii=False,
                indent=2,
                default=str,
            )
        )

    except (ImportError, ValueError) as error:
        print(
            "Pytest config could not be parsed:",
            type(error).__name__,
            str(error),
        )

else:
    print("pyproject.toml unavailable.")


# ------------------------------------------------------------
# E. Final checks
# ------------------------------------------------------------

for name, path in paths.items():
    assert sha_file(path) == EXPECTED_SHA[name], name

print()
print("=" * 96)
print("NOTEBOOK 14 / 14.6C READ-ONLY INSPECTION COMPLETE")
print("=" * 96)

print("empty-dataset guard source: PRINTED")
print("project modules imported: False")
print("pytest executed: False")
print("runtime executed: False")
print("artifact written: False")
print("new execution authorization: False")

print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")
```

```python
# Notebook 14 / 14.6D
# FROZEN_V9 IMPORT AND PYTEST COLLECTION BOUNDARY
#
# READ ONLY:
# No project imports, pytest execution, runtime, exec/eval,
# GT, evaluator, historical oracle, Git, or file writes.

import ast
import hashlib
import json
import tomllib
from pathlib import Path

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

MANIFEST_REL = (
    "reports/"
    "notebook14_14_5d_b12_blocked_delivery_manifest_v1.json"
)

EXPECTED_MANIFEST_SHA = (
    "4aaaaa4f4043f74a82ac948938e078372a9e73c61d14fe423108a6f3a6855e8a"
)

SUPPRESSED_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

TEST_REL = "tests/test_stage12_suppressed_peak_provider.py"

EXPECTED_SUPPRESSED_SHA = (
    "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
)

EXPECTED_TEST_SHA = (
    "eac928e54a26c086e531be7d38eeb46c67bcbba2d4ddd20cd00f255dcb63af4a"
)


def sha_file(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def dotted(node):
    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):
        base = dotted(node.value)

        if base != "<dynamic>":
            return base + "." + node.attr

    return "<dynamic>"


def calls(node):
    return sorted({
        dotted(item.func)
        for item in ast.walk(node)
        if isinstance(item, ast.Call)
    })


def inspect_module(path, known_hashes):
    rel = str(path.relative_to(ROOT))

    assert path.is_file()
    assert not path.is_symlink()

    actual_sha = sha_file(path)
    expected_sha = known_hashes.get(rel)

    if expected_sha is not None:
        assert actual_sha == expected_sha

    source = path.read_text(encoding="utf-8")
    tree = ast.parse(source, filename=rel)

    imports = []
    import_time = []

    for node in tree.body:
        if isinstance(node, ast.Import):
            imports.extend(alias.name for alias in node.names)
            continue

        if isinstance(node, ast.ImportFrom):
            imports.append(
                "." * node.level
                + (node.module or "")
                + " : "
                + ", ".join(alias.name for alias in node.names)
            )
            continue

        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef),
        ):
            expressions = list(node.decorator_list)

            if isinstance(node, ast.ClassDef):
                expressions += list(node.bases)
                expressions += [
                    item.value for item in node.keywords
                ]

                # A class body is evaluated when the class is defined.
                for child in node.body:
                    if isinstance(
                        child,
                        (ast.FunctionDef, ast.AsyncFunctionDef),
                    ):
                        expressions += list(child.decorator_list)
                        expressions += list(child.args.defaults)
                        expressions += [
                            value
                            for value in child.args.kw_defaults
                            if value is not None
                        ]

                    elif not isinstance(
                        child,
                        (ast.Expr, ast.Pass),
                    ):
                        expressions.append(child)

            else:
                expressions += list(node.args.defaults)
                expressions += [
                    value
                    for value in node.args.kw_defaults
                    if value is not None
                ]

            found = sorted({
                call
                for expression in expressions
                for call in calls(expression)
            })

            if found:
                import_time.append({
                    "line": node.lineno,
                    "definition": node.name,
                    "expression_call_names": found,
                })

            continue

        text = ast.get_source_segment(source, node) or ""

        import_time.append({
            "line": getattr(node, "lineno", None),
            "kind": type(node).__name__,
            "call_names": calls(node),
            "source_preview": text[:120],
        })

    print()
    print("=" * 88)
    print("MODULE:", rel)
    print("=" * 88)
    print("SHA-256:", actual_sha)
    print(
        "SHA covered by 14.5D manifest:",
        expected_sha is not None,
    )
    print("imports:", len(imports))

    for item in imports:
        print("  IMPORT:", item)

    print("import-time expression candidates:", len(import_time))

    for item in import_time:
        print(
            "  CANDIDATE:",
            json.dumps(item, ensure_ascii=False),
        )

    return {
        "path": rel,
        "sha256": actual_sha,
        "manifest_locked": expected_sha is not None,
        "imports": imports,
        "import_time": import_time,
    }


print("=" * 88)
print("NOTEBOOK 14 / 14.6D")
print("FROZEN_V9 IMPORT AND COLLECTION BOUNDARY")
print("=" * 88)

manifest_path = ROOT / MANIFEST_REL

assert manifest_path.is_file()
assert sha_file(manifest_path) == EXPECTED_MANIFEST_SHA

manifest = json.loads(
    manifest_path.read_text(encoding="utf-8")
)

assert manifest["decision"] == "PASS"
assert manifest["resolution"]["full_B12_delivery_status"] == "BLOCKED"
assert (
    manifest["resolution"]["historical_oracle_access_authorized"]
    is False
)

assert sha_file(ROOT / SUPPRESSED_REL) == EXPECTED_SUPPRESSED_SHA
assert sha_file(ROOT / TEST_REL) == EXPECTED_TEST_SHA

known_hashes = manifest["assets"]["component_source"]["source_sha256"]

print("[PASS] FROZEN PARENT, SOURCE AND TEST")

# These are the package files associated with the static
# "from . import frozen_v9" dependency.
#
# Inspect all existing candidates; do not assume which one
# Python resolves without executing the import mechanism.

candidate_paths = [
    ROOT / "src/biohub_cell_tracking/__init__.py",
    ROOT / "src/biohub_cell_tracking/frozen_v9.py",
    ROOT / "src/biohub_cell_tracking/frozen_v9/__init__.py",
]

records = []

for path in candidate_paths:
    if path.exists():
        records.append(inspect_module(path, known_hashes))
    else:
        print("ABSENT CANDIDATE:", path.relative_to(ROOT))

frozen_records = [
    item for item in records
    if item["path"].endswith("/frozen_v9.py")
    or item["path"].endswith("/frozen_v9/__init__.py")
]

if len(frozen_records) == 0:
    frozen_status = "UNRESOLVED_MODULE_LOCATION"
elif len(frozen_records) > 1:
    frozen_status = "MULTIPLE_MODULE_CANDIDATES_REVIEW_REQUIRED"
else:
    frozen_status = "STATIC_FROZEN_V9_FILE_INSPECTED"

# Read existing pytest configuration. This does not enumerate
# installed entry-point plugins or prove their side effects.

print()
print("=" * 88)
print("PYTEST CONFIGURATION")
print("=" * 88)

pyproject = ROOT / "pyproject.toml"

if pyproject.is_file() and not pyproject.is_symlink():
    data = tomllib.loads(
        pyproject.read_text(encoding="utf-8")
    )

    pytest_options = (
        data.get("tool", {})
        .get("pytest", {})
        .get("ini_options", {})
    )

    print("pyproject SHA:", sha_file(pyproject))
    print(
        json.dumps(
            pytest_options,
            ensure_ascii=False,
            indent=2,
            default=str,
        )
    )

for filename in ("pytest.ini", "setup.cfg", "tox.ini"):
    path = ROOT / filename

    if path.is_file() and not path.is_symlink():
        print(
            "ADDITIONAL CONFIG PRESENT:",
            filename,
            "| SHA:",
            sha_file(path),
        )

# Verify the selected test module has no explicit top-level
# executable expression or fixture declaration. This is not
# a complete pytest plugin or import safety proof.

print()
print("=" * 88)
print("TEST MODULE STATIC COLLECTION RECORD")
print("=" * 88)

test_source = (ROOT / TEST_REL).read_text(encoding="utf-8")
test_tree = ast.parse(test_source, filename=TEST_REL)

for node in test_tree.body:
    if isinstance(node, (ast.Import, ast.ImportFrom)):
        continue

    if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)):
        print(
            "definition:",
            node.name,
            "| decorators:",
            [ast.unparse(d) for d in node.decorator_list],
            "| defaults:",
            [ast.unparse(d) for d in node.args.defaults],
        )
        continue

    print(
        "top-level item:",
        type(node).__name__,
        "| line:",
        getattr(node, "lineno", None),
        "| call candidates:",
        calls(node),
    )

# Read-only final checks.
assert sha_file(manifest_path) == EXPECTED_MANIFEST_SHA
assert sha_file(ROOT / SUPPRESSED_REL) == EXPECTED_SUPPRESSED_SHA
assert sha_file(ROOT / TEST_REL) == EXPECTED_TEST_SHA

for record in records:
    assert sha_file(ROOT / record["path"]) == record["sha256"]

print()
print("=" * 88)
print("NOTEBOOK 14 / 14.6D READ-ONLY INSPECTION COMPLETE")
print("=" * 88)

print("frozen_v9 status:", frozen_status)
print("module candidates inspected:", len(records))
print("pytest executed:", False)
print("runtime executed:", False)
print("artifact written:", False)
print("new execution authorization:", False)
print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")
```

```python
# =============================================================================
# NOTEBOOK 14 / 14.6F
# ORIGINAL SUPPRESSED-PEAK ENTRYPOINT: INVALID-ID GUARD MICROTEST
#
# Only the SHA-locked original function definition is extracted and executed.
# Every input is invalid.
#
# No project-module imports, pytest, P995 inference, real context validation,
# source-derived execution, GT, evaluator, historical oracle or Git.
#
# The isolated namespace is a scope-control mechanism, not a security sandbox.
# =============================================================================

import ast
import hashlib
import json
from pathlib import Path
from types import SimpleNamespace


ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

SOURCE_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

TEST_REL = "tests/test_stage12_suppressed_peak_provider.py"

PARENT_REL = (
    "reports/"
    "notebook14_14_6e_isolated_p995_graph_facade_microtest_v1.json"
)

MANIFEST_REL = (
    "reports/"
    "notebook14_14_5d_b12_blocked_delivery_manifest_v1.json"
)

OUT_REL = (
    "reports/"
    "notebook14_14_6f_invalid_dataset_guard_isolated_microtest_v1.json"
)

SOURCE_SHA = (
    "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
)

TEST_SHA = (
    "eac928e54a26c086e531be7d38eeb46c67bcbba2d4ddd20cd00f255dcb63af4a"
)

FUNCTION_SHA = (
    "23ad6dcd918b2729aef87e12a36a7dbde617af5a99561c3b952f671f0b30fb8d"
)

PARENT_FILE_SHA = (
    "e15e2698f73bf6455d049f86cd3ec6a60d40422ded7e164dbad3976ee4991dd4"
)

PARENT_CONTENT_SHA = (
    "9fbca71d6d927d72ecf619253f04fb9fc7cff5cdf253cba50856b9af24fc64a9"
)

MANIFEST_FILE_SHA = (
    "4aaaaa4f4043f74a82ac948938e078372a9e73c61d14fe423108a6f3a6855e8a"
)


def sha_bytes(data):
    return hashlib.sha256(data).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(obj):
    return sha_bytes(
        json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_locked(rel, file_sha, content_sha=None):
    path = ROOT / rel

    assert path.is_file(), rel
    assert sha_file(path) == file_sha, rel

    obj = json.loads(path.read_text(encoding="utf-8"))

    assert obj["decision"] == "PASS"

    if content_sha is not None:
        assert obj["content_sha256"] == content_sha

        body = dict(obj)
        body.pop("content_sha256")

        assert canonical_sha(body) == content_sha

    return obj


print("=" * 96)
print("NOTEBOOK 14 / 14.6F")
print("ORIGINAL SUPPRESSED-PEAK INVALID-ID GUARD MICROTEST")
print("=" * 96)

# -------------------------------------------------------------------------
# 1. Parent/source/test locks
# -------------------------------------------------------------------------

parent = load_locked(
    PARENT_REL,
    PARENT_FILE_SHA,
    PARENT_CONTENT_SHA,
)

manifest = load_locked(
    MANIFEST_REL,
    MANIFEST_FILE_SHA,
)

assert parent["resolution"]["status"] == (
    "ORIGINAL_P995_GRAPH_FACADE_ISOLATED_MICROTEST_PASS"
)

assert parent["microtest"]["count"] == 7
assert parent["microtest"]["failed"] == 0

assert manifest["resolution"]["full_B12_delivery_status"] == "BLOCKED"

assert (
    manifest["resolution"]["historical_oracle_access_authorized"]
    is False
)

source_path = ROOT / SOURCE_REL
test_path = ROOT / TEST_REL

assert sha_file(source_path) == SOURCE_SHA
assert sha_file(test_path) == TEST_SHA

print("[PASS] PARENT_SOURCE_AND_TEST_LOCKS")

# -------------------------------------------------------------------------
# 2. Extract the exact original function
# -------------------------------------------------------------------------

source = source_path.read_text(encoding="utf-8")
tree = ast.parse(source, filename=SOURCE_REL)

matches = [
    node
    for node in tree.body
    if isinstance(node, ast.FunctionDef)
    and node.name == "suppressed_peak_provider"
]

assert len(matches) == 1

function = matches[0]

assert not function.decorator_list

function_text = ast.get_source_segment(source, function)

assert function_text is not None

assert sha_bytes(
    function_text.encode("utf-8")
) == FUNCTION_SHA

# Verify the actual statement order before executing any input.
#
#   [0] docstring
#   [1] non-string rejection
#   [2] strip
#   [3] empty-string rejection
#   [4] context validation
#   [5] source-derived execution

body = function.body

assert len(body) >= 6

assert isinstance(body[0], ast.Expr)
assert isinstance(body[0].value, ast.Constant)
assert isinstance(body[0].value.value, str)

assert isinstance(body[1], ast.If)

assert ast.unparse(body[1].test) == (
    "not isinstance(dataset_id, str)"
)

assert isinstance(body[2], ast.Assign)

assert ast.unparse(body[2]) == (
    "dataset_id = dataset_id.strip()"
)

assert isinstance(body[3], ast.If)

assert ast.unparse(body[3].test) == "not dataset_id"

assert isinstance(body[4], ast.Expr)

assert ast.unparse(body[4].value) == (
    "_validate_context(context)"
)

assert isinstance(body[5], ast.Assign)

assert ast.unparse(body[5].value) == (
    "_execute_source_derived(dataset_id, context)"
)

print("[PASS] EXACT_ORIGINAL_FUNCTION_AND_GUARD_ORDER")
print("  function SHA:", FUNCTION_SHA)

# -------------------------------------------------------------------------
# 3. Compile ONLY the original function definition.
#
# Annotation names are supplied explicitly; project modules are not loaded.
# A valid dataset ID must never be supplied to this test.
# -------------------------------------------------------------------------

for node in ast.walk(function):
    assert not isinstance(
        node,
        (
            ast.Import,
            ast.ImportFrom,
            ast.Global,
            ast.Nonlocal,
            ast.Await,
            ast.Yield,
            ast.YieldFrom,
        ),
    )

trap_calls = []


def forbidden_call(name):
    def trap(*args, **kwargs):
        trap_calls.append(name)

        raise AssertionError(
            "Forbidden runtime boundary reached: " + name
        )

    return trap


restricted_builtins = {
    "str": str,
    "isinstance": isinstance,
    "TypeError": TypeError,
    "ValueError": ValueError,
    "RuntimeError": RuntimeError,
    "set": set,
    "sorted": sorted,
    "repr": repr,
}

namespace = {
    "__name__": "isolated_invalid_dataset_guard",
    "__builtins__": restricted_builtins,

    # Used only when Python evaluates the original annotations.
    "pd": SimpleNamespace(DataFrame=object),
    "SuppressedPeakProviderContext": object,

    # Both calls are deliberately forbidden.
    "_validate_context": forbidden_call(
        "_validate_context"
    ),
    "_execute_source_derived": forbidden_call(
        "_execute_source_derived"
    ),
}

module = ast.Module(
    body=[function],
    type_ignores=[],
)

ast.fix_missing_locations(module)

compiled = compile(
    module,
    "<SHA-locked-suppressed-peak-entrypoint>",
    "exec",
)

exec(compiled, namespace, namespace)

isolated_provider = namespace["suppressed_peak_provider"]

print("[PASS] ORIGINAL_FUNCTION_LOADED_IN_ISOLATED_NAMESPACE")

# -------------------------------------------------------------------------
# 4. Negative-input microtests
# -------------------------------------------------------------------------

cases = [
    {
        "name": "NONE_REJECTED",
        "value": None,
        "exception": TypeError,
        "message": "dataset_id must be str",
    },
    {
        "name": "INTEGER_REJECTED",
        "value": 17,
        "exception": TypeError,
        "message": "dataset_id must be str",
    },
    {
        "name": "BYTES_REJECTED",
        "value": b"demo",
        "exception": TypeError,
        "message": "dataset_id must be str",
    },
    {
        "name": "EMPTY_STRING_REJECTED",
        "value": "",
        "exception": ValueError,
        "message": "dataset_id must be non-empty",
    },
    {
        "name": "SPACES_REJECTED",
        "value": "   ",
        "exception": ValueError,
        "message": "dataset_id must be non-empty",
    },
    {
        "name": "TAB_NEWLINE_REJECTED",
        "value": "\t\n",
        "exception": ValueError,
        "message": "dataset_id must be non-empty",
    },
    {
        "name": "UNICODE_WHITESPACE_REJECTED",
        "value": "\u2003",
        "exception": ValueError,
        "message": "dataset_id must be non-empty",
    },
]

passed = []

for case in cases:
    before = len(trap_calls)

    try:
        isolated_provider(
            case["value"],
            context=object(),
        )

    except case["exception"] as error:
        assert str(error) == case["message"], (
            case["name"],
            str(error),
        )

    else:
        raise AssertionError(
            "Expected rejection did not occur: "
            + case["name"]
        )

    assert len(trap_calls) == before, (
        "A forbidden boundary was reached: "
        + case["name"]
    )

    passed.append(case["name"])

    print("[PASS]", case["name"])

assert len(passed) == 7
assert trap_calls == []

print("[PASS] ZERO_CONTEXT_OR_SOURCE_DERIVED_CALLS")

# -------------------------------------------------------------------------
# 5. Post-test immutability verification
# -------------------------------------------------------------------------

assert sha_file(source_path) == SOURCE_SHA
assert sha_file(test_path) == TEST_SHA

assert sha_file(ROOT / PARENT_REL) == PARENT_FILE_SHA
assert sha_file(ROOT / MANIFEST_REL) == MANIFEST_FILE_SHA

# -------------------------------------------------------------------------
# 6. Formal artifact
# -------------------------------------------------------------------------

status = "ORIGINAL_SUPPRESSED_PEAK_INVALID_ID_GUARD_MICROTEST_PASS"

payload = {
    "stage": "NOTEBOOK14_14_6F",
    "title": "ORIGINAL_SUPPRESSED_PEAK_INVALID_ID_GUARD_MICROTEST",
    "version": 1,
    "decision": "PASS",

    "parents": {
        "14_6E": {
            "path": PARENT_REL,
            "file_sha256": PARENT_FILE_SHA,
            "content_sha256": PARENT_CONTENT_SHA,
        },
        "14_5D": {
            "path": MANIFEST_REL,
            "file_sha256": MANIFEST_FILE_SHA,
        },
    },

    "original_source": {
        "path": SOURCE_REL,
        "file_sha256": SOURCE_SHA,
        "function": "suppressed_peak_provider",
        "function_sha256": FUNCTION_SHA,
    },

    "test_source": {
        "path": TEST_REL,
        "file_sha256": TEST_SHA,
        "original_pytest_not_executed": True,
    },

    "execution_scope": {
        "mode": "ISOLATED_ORIGINAL_FUNCTION_AST",
        "inputs": "INVALID_DATASET_IDS_ONLY",
        "context_validation_trapped": True,
        "source_derived_execution_trapped": True,
        "trap_call_count": len(trap_calls),
        "project_modules_imported": False,
        "pytest_executed": False,
        "P995_model_executed": False,
        "real_suppressed_peak_provider_executed": False,
        "full_B12_executed": False,
        "historical_oracle_opened": False,
        "evaluator_executed": False,
    },

    "microtest": {
        "count": len(passed),
        "passed": passed,
        "failed": 0,
    },

    "limitations": [
        "Only invalid dataset ID paths were exercised.",
        "The function was executed outside its original module.",
        "Original project imports and pytest collection were not exercised.",
        "Valid dataset IDs were not supplied.",
        "Context validation and source-derived execution were trapped.",
        "This does not verify P995 inference or suppressed-peak generation.",
        "Full B12 and structural fidelity remain blocked.",
    ],

    "resolution": {
        "status": status,
        "isolated_invalid_id_guard_verified": True,
        "production_provider_verified": False,
        "new_pytest_execution_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
    },
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

OUT = ROOT / OUT_REL

assert OUT.parent.is_dir()

if OUT.exists():
    assert OUT.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"

else:
    OUT.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"

print("=" * 96)
print("NOTEBOOK 14 / 14.6F PASS")
print("=" * 96)

print("original function executed in isolation:", True)
print("invalid-ID microtests passed:", len(passed))
print("forbidden boundary calls:", len(trap_calls))
print("project modules imported:", False)
print("pytest executed:", False)
print("real source-derived runtime executed:", False)

print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")

print("STATUS:", status)
print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(OUT))
print("artifact mode:", mode)
```

```python
# =============================================================================
# Notebook 14 / 14.6G
# ISOLATED SUPPRESSED-PEAK DATASET OUTPUT BOUNDARY
#
# Execute only the SHA-locked original entrypoint, with two controlled fakes.
# No real context validation, source-derived runtime, project imports,
# pytest, P995 model, GT, evaluator, oracle or Git.
# =============================================================================

import ast
import builtins
import hashlib
import json
from pathlib import Path
from types import SimpleNamespace

import pandas as pd

ROOT = Path("/Users/heqiuyan/Desktop/biohub-cell-tracking").resolve()

SOURCE_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)
PARENT_REL = (
    "reports/notebook14_14_6f_invalid_dataset_guard_isolated_microtest_v1.json"
)
OUT_REL = (
    "reports/notebook14_14_6g_isolated_dataset_output_boundary_v1.json"
)

SOURCE_SHA = (
    "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
)
FUNCTION_SHA = (
    "23ad6dcd918b2729aef87e12a36a7dbde617af5a99561c3b952f671f0b30fb8d"
)
PARENT_FILE_SHA = (
    "9246623fe30d0cc9b1d1938e68e1a16fe87a24993e65b584dd9690203cc02dbe"
)
PARENT_CONTENT_SHA = (
    "600a271c4da29aa728107f39d07c4578e706c754495b9f50dba673a479d74861"
)


def sha_bytes(data):
    return hashlib.sha256(data).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(obj):
    return sha_bytes(
        json.dumps(
            obj,
            sort_keys=True,
            ensure_ascii=False,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


print("=" * 96)
print("NOTEBOOK 14 / 14.6G")
print("ISOLATED SUPPRESSED-PEAK DATASET OUTPUT BOUNDARY")
print("=" * 96)

# 1. Parent and original source locks.

parent_path = ROOT / PARENT_REL
source_path = ROOT / SOURCE_REL

assert sha_file(parent_path) == PARENT_FILE_SHA
assert sha_file(source_path) == SOURCE_SHA

parent = json.loads(parent_path.read_text(encoding="utf-8"))

assert parent["decision"] == "PASS"
assert parent["content_sha256"] == PARENT_CONTENT_SHA

parent_body = dict(parent)
parent_body.pop("content_sha256")

assert canonical_sha(parent_body) == PARENT_CONTENT_SHA

assert parent["resolution"]["status"] == (
    "ORIGINAL_SUPPRESSED_PEAK_INVALID_ID_GUARD_MICROTEST_PASS"
)

assert parent["microtest"]["count"] == 7
assert parent["microtest"]["failed"] == 0
assert parent["resolution"]["full_B12_authorized"] is False

print("[PASS] PARENT_AND_SOURCE_LOCKS")

# 2. Extract and compile only the original function.

source = source_path.read_text(encoding="utf-8")
tree = ast.parse(source, filename=SOURCE_REL)

matches = [
    node
    for node in tree.body
    if isinstance(node, ast.FunctionDef)
    and node.name == "suppressed_peak_provider"
]

assert len(matches) == 1

function = matches[0]
function_text = ast.get_source_segment(source, function)

assert function_text is not None
assert sha_bytes(function_text.encode("utf-8")) == FUNCTION_SHA
assert function.decorator_list == []

assert not any(
    isinstance(
        node,
        (
            ast.Import,
            ast.ImportFrom,
            ast.Global,
            ast.Nonlocal,
            ast.Await,
            ast.Yield,
            ast.YieldFrom,
        ),
    )
    for node in ast.walk(function)
)

# Namespace hooks are supplied separately for each case.
# They are the ONLY context/source-derived functions reachable by name.

restricted_builtins = {
    "str": str,
    "isinstance": isinstance,
    "TypeError": TypeError,
    "ValueError": ValueError,
    "RuntimeError": RuntimeError,
    "set": set,
    "sorted": sorted,
    "repr": repr,
}

namespace = {
    "__name__": "isolated_dataset_output_boundary",
    "__builtins__": restricted_builtins,
    "pd": SimpleNamespace(DataFrame=pd.DataFrame),
    "SuppressedPeakProviderContext": object,
}

isolated_module = ast.Module(
    body=[function],
    type_ignores=[],
)

ast.fix_missing_locations(isolated_module)

compiled = compile(
    isolated_module,
    "<SHA-locked-original-suppressed-peak-entrypoint>",
    "exec",
)

exec(compiled, namespace, namespace)

provider = namespace["suppressed_peak_provider"]

print("[PASS] ORIGINAL_FUNCTION_LOADED_IN_ISOLATION")

# 3. Controlled in-memory cases.
#
# Every case must:
#   - validate the exact context object once;
#   - pass the normalized dataset ID to the fake executor once;
#   - never call the real source-derived implementation.

passed = []


def run_case(name, raw_id, frame, expected_error=None):
    call_order = []
    expected_context = object()
    expected_id = raw_id.strip()

    def fake_validate(context):
        assert context is expected_context
        call_order.append("validate")

    def fake_execute(dataset_id, context):
        assert context is expected_context
        assert dataset_id == expected_id
        call_order.append("execute")

        # Return only the explicitly provided in-memory DataFrame.
        return frame.copy()

    namespace["_validate_context"] = fake_validate
    namespace["_execute_source_derived"] = fake_execute

    if expected_error is None:
        result = provider(
            raw_id,
            context=expected_context,
        )

        assert isinstance(result, pd.DataFrame)
        assert result.index.tolist() == list(range(len(result)))

    else:
        try:
            provider(
                raw_id,
                context=expected_context,
            )

        except expected_error as error:
            assert "another dataset" in str(error)

        else:
            raise AssertionError(
                "Expected cross-dataset rejection: " + name
            )

    assert call_order == ["validate", "execute"], (
        name,
        call_order,
    )

    passed.append(name)
    print("[PASS]", name)

    return result if expected_error is None else None


# Case 1: strip ID, preserve valid rows, reset index.
valid_frame = pd.DataFrame(
    {
        "dataset": ["demo", "demo"],
        "detector_prob": [0.8, 0.9],
    },
    index=[10, 20],
)

valid_result = run_case(
    "TRIM_AND_RESET_INDEX",
    " demo ",
    valid_frame,
)

assert valid_result["dataset"].tolist() == ["demo", "demo"]
assert valid_result["detector_prob"].tolist() == [0.8, 0.9]

# Case 2: all rows refer to another dataset.
run_case(
    "FOREIGN_DATASET_REJECTED",
    "demo",
    pd.DataFrame(
        {
            "dataset": ["other"],
            "detector_prob": [0.9],
        }
    ),
    expected_error=RuntimeError,
)

# Case 3: one valid row and one foreign row.
run_case(
    "MIXED_DATASETS_REJECTED",
    "demo",
    pd.DataFrame(
        {
            "dataset": ["demo", "other"],
            "detector_prob": [0.8, 0.9],
        }
    ),
    expected_error=RuntimeError,
)

# Case 4: an empty DataFrame with a dataset column is permitted.
empty_result = run_case(
    "EMPTY_DATASET_TABLE_ACCEPTED",
    "demo",
    pd.DataFrame(
        {
            "dataset": pd.Series(dtype="str"),
            "detector_prob": pd.Series(dtype="float64"),
        }
    ),
)

assert empty_result.empty

# Case 5: the original entrypoint treats the dataset column as optional.
# This is recorded as behavior, NOT proof of dataset isolation.
no_dataset_result = run_case(
    "OPTIONAL_DATASET_COLUMN_BEHAVIOR",
    "demo",
    pd.DataFrame(
        {"detector_prob": [0.5]},
        index=[9],
    ),
)

assert no_dataset_result.index.tolist() == [0]
assert "dataset" not in no_dataset_result.columns

assert len(passed) == 5

# 4. Source/parent immutability.

assert sha_file(source_path) == SOURCE_SHA
assert sha_file(parent_path) == PARENT_FILE_SHA

# 5. Formal artifact.

status = "ISOLATED_SUPPRESSED_PEAK_DATASET_OUTPUT_BOUNDARY_PASS"

payload = {
    "stage": "NOTEBOOK14_14_6G",
    "title": "ISOLATED_SUPPRESSED_PEAK_DATASET_OUTPUT_BOUNDARY",
    "version": 1,
    "decision": "PASS",
    "parent_14_6f": {
        "path": PARENT_REL,
        "file_sha256": PARENT_FILE_SHA,
        "content_sha256": PARENT_CONTENT_SHA,
    },
    "original_source": {
        "path": SOURCE_REL,
        "file_sha256": SOURCE_SHA,
        "function": "suppressed_peak_provider",
        "function_sha256": FUNCTION_SHA,
    },
    "microtest": {
        "count": len(passed),
        "passed": passed,
        "failed": 0,
        "mode": "ORIGINAL_FUNCTION_WITH_CONTROLLED_MEMORY_FAKES",
    },
    "execution_scope": {
        "project_modules_imported": False,
        "pytest_executed": False,
        "real_context_validation_executed": False,
        "real_source_derived_runtime_executed": False,
        "P995_inference_executed": False,
        "full_B12_executed": False,
        "evaluator_executed": False,
        "historical_oracle_opened": False,
    },
    "limitations": [
        "Only the original entrypoint was executed, outside its module.",
        "Both context validation and source-derived execution were faked.",
        "The DataFrames were supplied in memory; no real dataset was read.",
        "The optional dataset-column case does not establish dataset isolation.",
        "No production provider integration or P995 inference was tested.",
        "No full-B12 structural fidelity or performance result was established.",
    ],
    "resolution": {
        "status": status,
        "isolated_output_boundary_verified": True,
        "production_provider_verified": False,
        "new_pytest_execution_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
    },
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

out = ROOT / OUT_REL

assert out.parent.is_dir()

if out.exists():
    assert out.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"
else:
    out.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"

print("=" * 96)
print("NOTEBOOK 14 / 14.6G PASS")
print("=" * 96)

print("isolated boundary cases passed:", len(passed))
print("real source-derived runtime executed:", False)
print("project modules imported:", False)
print("pytest executed:", False)
print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")
print("STATUS:", status)
print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(out))
print("artifact mode:", mode)
```

```python
# =============================================================================
# NOTEBOOK 14 / 14.7B
# SOURCE-DERIVED EXECUTION PATH STATIC AUDIT
#
# Read-only examination of:
#   1. Frozen suppressed-peak provider source
#   2. SHA-locked canonical writer source
#   3. Existing 14.7A contract-gap evidence
#
# No project imports, compile/exec, pytest, P995 inference, real provider,
# historical pickle/oracle, GT, evaluator, Git, or source/test mutation.
#
# The ONLY permitted write is a new formal JSON report.
# =============================================================================

import ast
import hashlib
import json
from pathlib import Path

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

PARENT_REL = (
    "reports/"
    "notebook14_14_7a_"
    "suppressed_peak_dataset_column_contract_gap_audit_v1.json"
)

MANIFEST_REL = (
    "reports/"
    "notebook14_14_5d_b12_blocked_delivery_manifest_v1.json"
)

SOURCE_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

WRITER_REL = (
    "scripts/run_stage12_fold0_suppressed_peak_universe.py"
)

OUT_REL = (
    "reports/"
    "notebook14_14_7b_source_derived_execution_path_static_audit_v1.json"
)

PARENT_FILE_SHA = (
    "faea94b9417d4de21f2a07e1c0952d9a17b1443f684d73b3368664bae7db44e1"
)

PARENT_CONTENT_SHA = (
    "142fa63e3d301b0e2446dd0d00c6bd4248e3fc83569e4dd7e0005c3ef01e85ca"
)

MANIFEST_FILE_SHA = (
    "4aaaaa4f4043f74a82ac948938e078372a9e73c61d14fe423108a6f3a6855e8a"
)

SOURCE_SHA = (
    "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
)

WRITER_SHA = (
    "35843437ccc783cb07751efbb9d6d241b411a3c22cc6bb294928e2c62c22f3fc"
)


# =============================================================================
# 1. Pure helpers
# =============================================================================

def sha_bytes(data):
    return hashlib.sha256(data).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(value):
    return sha_bytes(
        json.dumps(
            value,
            sort_keys=True,
            ensure_ascii=False,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


def load_locked(relpath, expected_file_sha, expected_content_sha=None):
    path = ROOT / relpath

    assert path.is_file(), relpath
    assert not path.is_symlink(), relpath

    assert sha_file(path) == expected_file_sha, relpath

    obj = json.loads(path.read_text(encoding="utf-8"))

    assert obj["decision"] == "PASS"

    if expected_content_sha is not None:
        assert obj["content_sha256"] == expected_content_sha

        body = dict(obj)
        body.pop("content_sha256")

        assert canonical_sha(body) == expected_content_sha

    return obj


def dotted_name(node):
    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):
        base = dotted_name(node.value)

        if base != "<dynamic>":
            return base + "." + node.attr

    return "<dynamic>"


def call_records(node):
    records = []

    for child in ast.walk(node):
        if not isinstance(child, ast.Call):
            continue

        records.append({
            "line": getattr(child, "lineno", None),
            "name": dotted_name(child.func),
        })

    return sorted(
        records,
        key=lambda record: (
            record["line"] or 0,
            record["name"],
        ),
    )


def extract_window(lines, center, radius=3):
    first = max(1, center - radius)
    last = min(len(lines), center + radius)

    content = "\n".join(
        str(number) + " | " + lines[number - 1]
        for number in range(first, last + 1)
    )

    return {
        "start_line": first,
        "end_line": last,
        "center_line": center,
        "excerpt_sha256": sha_bytes(content.encode("utf-8")),
        "excerpt": content,
    }


def top_definitions(tree):
    return {
        node.name: node
        for node in tree.body
        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef),
        )
    }


# =============================================================================
# 2. Formal parent and file locks
# =============================================================================

print("=" * 100)
print("NOTEBOOK 14 / 14.7B")
print("SOURCE-DERIVED EXECUTION PATH STATIC AUDIT")
print("=" * 100)

parent = load_locked(
    PARENT_REL,
    PARENT_FILE_SHA,
    PARENT_CONTENT_SHA,
)

manifest = load_locked(
    MANIFEST_REL,
    MANIFEST_FILE_SHA,
)

assert parent["resolution"]["status"] == (
    "SUPPRESSED_PEAK_DATASET_COLUMN_CONTRACT_GAP_CONFIRMED_"
    "SOURCE_DERIVED_SCHEMA_REVIEW_REQUIRED"
)

assert parent["resolution"]["new_contract_implementation_authorized"] is False
assert parent["resolution"]["real_source_derived_runtime_authorized"] is False
assert parent["resolution"]["full_B12_authorized"] is False

assert manifest["resolution"]["full_B12_delivery_status"] == "BLOCKED"
assert manifest["resolution"]["historical_oracle_access_authorized"] is False

source_path = ROOT / SOURCE_REL
writer_path = ROOT / WRITER_REL

for path, expected in (
    (source_path, SOURCE_SHA),
    (writer_path, WRITER_SHA),
):
    assert path.is_file(), path
    assert not path.is_symlink(), path
    assert sha_file(path) == expected, path

source_manifest = manifest["assets"]["component_source"]["source_sha256"]

assert source_manifest[SOURCE_REL] == SOURCE_SHA
assert source_manifest[WRITER_REL] == WRITER_SHA

print("[PASS] PARENTS_AND_BOTH_SOURCE_FILES_LOCKED")


# =============================================================================
# 3. Verify canonical writer identity from original provider constants
# =============================================================================

source = source_path.read_text(encoding="utf-8")
source_tree = ast.parse(source, filename=SOURCE_REL)
defs = top_definitions(source_tree)

literal_constants = {}

for node in source_tree.body:
    if not isinstance(node, ast.Assign):
        continue

    if len(node.targets) != 1:
        continue

    target = node.targets[0]

    if not isinstance(target, ast.Name):
        continue

    try:
        literal_constants[target.id] = ast.literal_eval(node.value)

    except (ValueError, TypeError, SyntaxError, MemoryError, RecursionError):
        continue

assert literal_constants[
    "SUPPRESSED_PEAK_CANONICAL_WRITER_RELATIVE_PATH"
] == WRITER_REL

assert literal_constants[
    "SUPPRESSED_PEAK_CANONICAL_WRITER_SHA256"
] == WRITER_SHA

print("[PASS] CANONICAL_WRITER_IDENTITY_MATCHES_SOURCE_CONSTANTS")


# =============================================================================
# 4. Static inspection of provider execution path
# =============================================================================

REQUIRED_UNITS = (
    "_build_source_derived_code",
    "_execute_source_derived",
    "suppressed_peak_provider",
)

provider_units = {}

INTERESTING_CALLS = {
    "compile",
    "exec",
    "_build_source_derived_code",
    "_canonical_writer_path",
    "_fresh_p995_graph_for_dataset",
    "_load_p995_predict_authority_module",
    "baseline_node_state_provider",
    "tempfile.TemporaryDirectory",
    "metadata.to_csv",
    "dataset_path.exists",
    "out_dir.mkdir",
    "namespace.get",
    "result.reset_index",
}

source_lines = source.splitlines()

print()
print("=" * 100)
print("PROVIDER EXECUTION PATH")
print("=" * 100)

for name in REQUIRED_UNITS:
    assert name in defs, name

    node = defs[name]
    unit_text = ast.get_source_segment(source, node)

    assert unit_text is not None

    calls = call_records(node)

    selected_calls = [
        call
        for call in calls
        if call["name"] in INTERESTING_CALLS
    ]

    # Include the exact neighborhood of each selected call,
    # rather than executing the function.
    excerpts = [
        extract_window(source_lines, call["line"], radius=2)
        for call in selected_calls
        if call["line"] is not None
    ]

    provider_units[name] = {
        "start_line": node.lineno,
        "end_line": node.end_lineno,
        "unit_sha256": sha_bytes(unit_text.encode("utf-8")),
        "syntactic_call_records": calls,
        "selected_execution_boundary_calls": selected_calls,
        "selected_source_excerpts": excerpts,
        "executed": False,
    }

    print()
    print(name)
    print("  lines:", node.lineno, "-", node.end_lineno)
    print("  source SHA:", provider_units[name]["unit_sha256"])
    print("  selected calls:")

    for call in selected_calls:
        print("   ", call["line"], call["name"])

assert any(
    call["name"] == "compile"
    for call in provider_units["_build_source_derived_code"][
        "syntactic_call_records"
    ]
)

assert any(
    call["name"] == "exec"
    for call in provider_units["_execute_source_derived"][
        "syntactic_call_records"
    ]
)

print()
print("[PASS] DYNAMIC_EXECUTION_BOUNDARIES_LOCATED_NOT_EXECUTED")


# =============================================================================
# 5. Canonical writer: static output/schema evidence
# =============================================================================

writer_source = writer_path.read_text(encoding="utf-8")
writer_lines = writer_source.splitlines()

writer_tree = ast.parse(writer_source, filename=WRITER_REL)

# Search only the already source-locked canonical writer.
# Matching text is NOT automatically a schema guarantee.

SEARCH_TERMS = (
    "dataset",
    "universe",
    "to_csv",
    "DataFrame",
    "suppressed_peak",
)

candidate_lines = []

for line_number, line in enumerate(writer_lines, start=1):
    lowered = line.lower()

    hits = [
        term
        for term in SEARCH_TERMS
        if term.lower() in lowered
    ]

    if hits:
        candidate_lines.append({
            "line": line_number,
            "terms": hits,
        })

# Preserve every matching line number in the artifact.
# Print only a bounded excerpt set to keep Notebook output readable.
MAX_PRINTED_EXCERPTS = 18

selected_line_numbers = []

# Give priority to occurrences of the dataset field.
for record in candidate_lines:
    if "dataset" in [term.lower() for term in record["terms"]]:
        selected_line_numbers.append(record["line"])

# Then include other output/writer references.
for record in candidate_lines:
    selected_line_numbers.append(record["line"])

selected_line_numbers = list(dict.fromkeys(selected_line_numbers))

shown = selected_line_numbers[:MAX_PRINTED_EXCERPTS]

writer_excerpts = [
    extract_window(writer_lines, number, radius=3)
    for number in shown
]

print()
print("=" * 100)
print("CANONICAL WRITER OUTPUT/SCHEMA EVIDENCE")
print("=" * 100)

print("writer SHA:", WRITER_SHA)
print("writer total lines:", len(writer_lines))
print("matching line count:", len(candidate_lines))
print("printed excerpt count:", len(writer_excerpts))
print("additional matches retained as line records:",
      len(selected_line_numbers) - len(shown))

for excerpt in writer_excerpts:
    print()
    print(excerpt["excerpt"])

# Locate syntactic output-write calls without running the writer.

writer_write_calls = [
    call
    for call in call_records(writer_tree)
    if call["name"].endswith(
        (
            ".to_csv",
            ".to_parquet",
            ".to_pickle",
            ".to_json",
            ".write_text",
            ".write_bytes",
        )
    )
]

print()
print("Output-write call candidates:")

for call in writer_write_calls:
    print(" ", call["line"], call["name"])


# =============================================================================
# 6. Evidence classification
# =============================================================================

# Do not infer mandatory dataset-column schema from a keyword hit.
# Source excerpts must be interpreted separately, including any
# transformation between the producer table and returned DataFrame.

schema_status = (
    "STATIC_SOURCE_EVIDENCE_CAPTURED_"
    "MANDATORY_DATASET_COLUMN_NOT_YET_ADJUDICATED"
)

unresolved = [
    "Trace the exact canonical writer output table construction.",
    "Determine whether dataset is explicitly assigned to each output row.",
    "Check whether the final returned DataFrame preserves the dataset column.",
    "Determine the intended schema for empty output.",
    "Distinguish canonical writer output from wrapper-added metadata.",
    "Keep compile/exec and real P995 inference outside the permitted runtime scope.",
]

status = (
    "SOURCE_DERIVED_EXECUTION_PATH_STATIC_EVIDENCE_FROZEN_"
    "OUTPUT_SCHEMA_ADJUDICATION_REQUIRED"
)

next_stage = (
    "NOTEBOOK14_14_7C_"
    "CANONICAL_OUTPUT_SCHEMA_ADJUDICATION"
)

print()
print("=" * 100)
print("AUDIT CLASSIFICATION")
print("=" * 100)

print("compile: PRESENT, NOT_EXECUTED")
print("exec: PRESENT, NOT_EXECUTED")
print("canonical writer: SOURCE_SHA_VERIFIED")
print("mandatory dataset column:", "NOT_YET_ADJUDICATED")
print("real provider execution:", False)
print("new pytest execution:", False)


# =============================================================================
# 7. Source and parent immutability
# =============================================================================

assert sha_file(source_path) == SOURCE_SHA
assert sha_file(writer_path) == WRITER_SHA

assert sha_file(ROOT / PARENT_REL) == PARENT_FILE_SHA
assert sha_file(ROOT / MANIFEST_REL) == MANIFEST_FILE_SHA

gates = {
    "parent_14_7A_locked": True,
    "manifest_14_5D_locked": True,
    "provider_source_locked": True,
    "canonical_writer_source_locked": True,
    "writer_constant_identity_verified": True,
    "source_derived_builder_identified": True,
    "source_derived_executor_identified": True,
    "compile_and_exec_located_not_executed": True,
    "canonical_writer_static_evidence_captured": True,
    "mandatory_dataset_schema_not_assumed": True,
    "new_contract_not_implemented": True,
    "real_provider_not_executed": True,
    "pytest_not_executed": True,
    "historical_oracle_closed": True,
    "full_B12_blocked": True,
    "CURRENT_BEST_unchanged": True,
    "no_source_or_test_mutation": True,
    "no_Git_GT_evaluator_or_performance": True,
}

assert all(gates.values())


# =============================================================================
# 8. Formal artifact
# =============================================================================

payload = {
    "stage": "NOTEBOOK14_14_7B",
    "title": "SOURCE_DERIVED_EXECUTION_PATH_STATIC_AUDIT",
    "version": 1,
    "decision": "PASS",

    "parents": {
        "14_7A": {
            "path": PARENT_REL,
            "file_sha256": PARENT_FILE_SHA,
            "content_sha256": PARENT_CONTENT_SHA,
        },
        "14_5D": {
            "path": MANIFEST_REL,
            "file_sha256": MANIFEST_FILE_SHA,
        },
    },

    "source_identity": {
        "provider_path": SOURCE_REL,
        "provider_sha256": SOURCE_SHA,
        "canonical_writer_path": WRITER_REL,
        "canonical_writer_sha256": WRITER_SHA,
    },

    "provider_execution_path": provider_units,

    "canonical_writer_static_evidence": {
        "writer_line_count": len(writer_lines),
        "keyword_match_lines": candidate_lines,
        "selected_source_excerpts": writer_excerpts,
        "output_write_call_candidates": writer_write_calls,
        "complete_runtime_dataflow_verified": False,
        "mandatory_dataset_column_verified": False,
    },

    "schema_status": schema_status,
    "unresolved_questions": unresolved,

    "resolution": {
        "status": status,
        "static_path_audit_complete": True,
        "source_derived_runtime_executed": False,
        "canonical_writer_output_schema_verified": False,
        "new_contract_implementation_authorized": False,
        "new_pytest_execution_authorized": False,
        "real_provider_runtime_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
        "next_stage": next_stage,
    },

    "semantic_gates": gates,

    "claim_boundary": {
        "evidence": "SHA_LOCKED_STATIC_SOURCE_ONLY",
        "runtime_execution": False,
        "pytest_execution": False,
        "source_mutation": False,
        "test_mutation": False,
        "Git_operation": False,
        "GT_access": False,
        "evaluator_execution": False,
        "performance_evidence": False,
    },

    "next_stage": next_stage,
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

out = ROOT / OUT_REL

assert out.parent.is_dir()

if out.exists():
    assert out.read_bytes() == artifact_bytes, (
        "Existing 14.7B artifact differs from current audit."
    )
    artifact_mode = "EXISTING_EXACT_ARTIFACT_REUSED"
else:
    out.write_bytes(artifact_bytes)
    artifact_mode = "NEW_FORMAL_ARTIFACT_WRITTEN"

# Final recheck.
assert sha_file(source_path) == SOURCE_SHA
assert sha_file(writer_path) == WRITER_SHA
assert sha_file(ROOT / PARENT_REL) == PARENT_FILE_SHA

print()
print("=" * 100)
print("NOTEBOOK 14 / 14.7B PASS")
print("=" * 100)

print("provider units inspected:", len(provider_units))
print("canonical writer evidence lines:", len(candidate_lines))
print("schema status:", schema_status)
print("compile/exec executed:", False)
print("new pytest executed:", False)
print("real provider executed:", False)
print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")

print("STATUS:", status)
print("NEXT:", next_stage)

print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(out))
print("artifact mode:", artifact_mode)
```

```python
# =============================================================================
# Notebook 14 / 14.7C
# CANONICAL OUTPUT SCHEMA DATAFLOW REVIEW
#
# Read-only source evidence extraction.
# No project imports, pytest, compile/exec of project code, inference,
# GT, evaluator, historical oracle, Git or file writes.
# =============================================================================

import ast
import hashlib
import json
from pathlib import Path

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

PARENT_REL = (
    "reports/"
    "notebook14_14_7b_source_derived_execution_path_static_audit_v1.json"
)

WRITER_REL = (
    "scripts/run_stage12_fold0_suppressed_peak_universe.py"
)

PROVIDER_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

PARENT_FILE_SHA = (
    "23ed281f59b45bbfbd9a7baabb745efd194424bfad1f174a715911c4681aec62"
)

PARENT_CONTENT_SHA = (
    "73b91584fc4489d318c0e1195f22f220253f9a81651aed71ab3ebc04789bc1f1"
)

WRITER_SHA = (
    "35843437ccc783cb07751efbb9d6d241b411a3c22cc6bb294928e2c62c22f3fc"
)

PROVIDER_SHA = (
    "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
)


def sha_file(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def canonical_sha(value):
    encoded = json.dumps(
        value,
        sort_keys=True,
        ensure_ascii=False,
        separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")

    return hashlib.sha256(encoded).hexdigest()


def show_range(lines, start, end, label):
    assert 1 <= start <= end <= len(lines)

    print()
    print("=" * 90)
    print(label, "| lines", start, "-", end)
    print("=" * 90)

    for number in range(start, end + 1):
        print(
            f"{number:5d} | {lines[number - 1]}"
        )


def dotted(node):
    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):
        base = dotted(node.value)

        if base != "<dynamic>":
            return base + "." + node.attr

    return "<dynamic>"


def find_call_lines(tree, call_suffixes):
    result = []

    for node in ast.walk(tree):
        if not isinstance(node, ast.Call):
            continue

        name = dotted(node.func)

        if any(
            name == suffix or name.endswith("." + suffix)
            for suffix in call_suffixes
        ):
            result.append({
                "line": node.lineno,
                "call": name,
            })

    return sorted(
        result,
        key=lambda record: (
            record["line"],
            record["call"],
        ),
    )


def print_definition_summary(tree, selected_names):
    for node in tree.body:
        if not isinstance(
            node,
            (ast.ClassDef, ast.FunctionDef, ast.AsyncFunctionDef),
        ):
            continue

        if node.name not in selected_names:
            continue

        print(
            node.name,
            "| lines",
            node.lineno,
            "-",
            node.end_lineno,
        )

        if isinstance(node, ast.ClassDef):
            for child in node.body:
                if isinstance(
                    child,
                    (ast.FunctionDef, ast.AsyncFunctionDef),
                ):
                    print(
                        "  method:",
                        child.name,
                        "| lines",
                        child.lineno,
                        "-",
                        child.end_lineno,
                    )


# -------------------------------------------------------------------------
# 1. Formal parent and source locks.
# -------------------------------------------------------------------------

print("=" * 90)
print("NOTEBOOK 14 / 14.7C")
print("CANONICAL OUTPUT SCHEMA DATAFLOW REVIEW")
print("=" * 90)

parent_path = ROOT / PARENT_REL
writer_path = ROOT / WRITER_REL
provider_path = ROOT / PROVIDER_REL

for path, expected in (
    (parent_path, PARENT_FILE_SHA),
    (writer_path, WRITER_SHA),
    (provider_path, PROVIDER_SHA),
):
    assert path.is_file(), path
    assert not path.is_symlink(), path
    assert sha_file(path) == expected, path

parent = json.loads(
    parent_path.read_text(encoding="utf-8")
)

assert parent["decision"] == "PASS"
assert parent["content_sha256"] == PARENT_CONTENT_SHA

parent_body = dict(parent)
parent_body.pop("content_sha256")

assert canonical_sha(parent_body) == PARENT_CONTENT_SHA

assert parent["resolution"]["canonical_writer_output_schema_verified"] is False
assert parent["resolution"]["full_B12_authorized"] is False

print("[PASS] PARENT_AND_BOTH_SOURCE_LOCKS")

# -------------------------------------------------------------------------
# 2. Canonical writer: candidate construction and final universe.
# -------------------------------------------------------------------------

writer_source = writer_path.read_text(encoding="utf-8")
writer_lines = writer_source.splitlines()

writer_tree = ast.parse(
    writer_source,
    filename=WRITER_REL,
)

print()
print("canonical writer total lines:", len(writer_lines))

# These ranges deliberately include branch conditions before and after
# dataset insertion, concatenation, assertions and output writes.

WRITER_RANGES = [
    (640, 695, "A. CANDIDATE CONSTRUCTION AND DATASET INSERTION"),
    (790, 855, "B. FINAL UNIVERSE CONSTRUCTION AND SCHEMA ASSERTIONS"),
    (970, 1005, "C. FINAL WRITER OUTPUT"),
]

for start, end, label in WRITER_RANGES:
    show_range(
        writer_lines,
        start,
        end,
        label,
    )

print()
print("=" * 90)
print("D. WRITER CALL SITES")
print("=" * 90)

writer_call_sites = find_call_lines(
    writer_tree,
    (
        "insert",
        "concat",
        "DataFrame",
        "to_pickle",
        "to_csv",
    ),
)

for record in writer_call_sites:
    print(
        record["line"],
        record["call"],
    )

# -------------------------------------------------------------------------
# 3. Provider: source transformation and returned value.
# -------------------------------------------------------------------------

provider_source = provider_path.read_text(
    encoding="utf-8"
)

provider_lines = provider_source.splitlines()

provider_tree = ast.parse(
    provider_source,
    filename=PROVIDER_REL,
)

print()
print("=" * 90)
print("E. SOURCE TRANSFORMER DEFINITION INVENTORY")
print("=" * 90)

print_definition_summary(
    provider_tree,
    {
        "_SuppressedPeakTransformer",
        "_build_source_derived_code",
        "_execute_source_derived",
    },
)

# Print the builder's final AST transformation and compilation path.
show_range(
    provider_lines,
    865,
    969,
    "F. TRANSFORMED SOURCE FINALIZATION",
)

# Print the executor's output namespace and return path.
show_range(
    provider_lines,
    1225,
    1277,
    "G. EXECUTED NAMESPACE AND RETURN PATH — STATIC ONLY",
)

print()
print("=" * 90)
print("H. PROVIDER RETURN/OUTPUT CALL SITES")
print("=" * 90)

provider_call_sites = find_call_lines(
    provider_tree,
    (
        "compile",
        "exec",
        "get",
        "reset_index",
    ),
)

for record in provider_call_sites:
    if 688 <= record["line"] <= 1358:
        print(
            record["line"],
            record["call"],
        )

# -------------------------------------------------------------------------
# 4. Final immutability and boundary.
# -------------------------------------------------------------------------

assert sha_file(parent_path) == PARENT_FILE_SHA
assert sha_file(writer_path) == WRITER_SHA
assert sha_file(provider_path) == PROVIDER_SHA

print()
print("=" * 90)
print("NOTEBOOK 14 / 14.7C READ-ONLY REVIEW COMPLETE")
print("=" * 90)

print("canonical writer: SHA_LOCKED")
print("provider source: SHA_LOCKED")
print("source transformation: NOT_EXECUTED")
print("real provider: NOT_EXECUTED")
print("pytest: NOT_RUN")
print("artifact written: False")

print("dataset schema: EVIDENCE_REVIEW_PENDING")
print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")
```

```python
# Notebook 14 / 14.7D
# SOURCE-DERIVED AST TRANSFORMATION PRESERVATION REVIEW
#
# STATIC ONLY.
# No project imports, compile/exec of project code, pytest,
# inference, historical oracle, GT, evaluator, Git or file writes.

import ast
import hashlib
import json
from pathlib import Path

ROOT = Path("/Users/heqiuyan/Desktop/biohub-cell-tracking").resolve()

FILES = {
    "parent": (
        "reports/"
        "notebook14_14_7b_source_derived_execution_path_static_audit_v1.json"
    ),
    "writer": "scripts/run_stage12_fold0_suppressed_peak_universe.py",
    "provider": (
        "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
    ),
}

EXPECTED_SHA = {
    "parent": (
        "23ed281f59b45bbfbd9a7baabb745efd194424bfad1f174a715911c4681aec62"
    ),
    "writer": (
        "35843437ccc783cb07751efbb9d6d241b411a3c22cc6bb294928e2c62c22f3fc"
    ),
    "provider": (
        "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
    ),
}

PARENT_CONTENT_SHA = (
    "73b91584fc4489d318c0e1195f22f220253f9a81651aed71ab3ebc04789bc1f1"
)


def file_sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def canonical_sha(obj):
    data = json.dumps(
        obj,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")

    return hashlib.sha256(data).hexdigest()


def dotted(node):
    if isinstance(node, ast.Name):
        return node.id

    if isinstance(node, ast.Attribute):
        prefix = dotted(node.value)

        if prefix != "<dynamic>":
            return prefix + "." + node.attr

    return "<dynamic>"


def show(lines, start, end, label):
    print()
    print("=" * 92)
    print(label, "| lines", start, "-", end)
    print("=" * 92)

    assert 1 <= start <= end <= len(lines)

    for number in range(start, end + 1):
        print(f"{number:5d} | {lines[number - 1]}")


def print_calls(tree, selected):
    results = []

    for node in ast.walk(tree):
        if not isinstance(node, ast.Call):
            continue

        name = dotted(node.func)

        if name in selected:
            results.append((node.lineno, name))

    for line, name in sorted(set(results)):
        print(f"{line:5d} | {name}")


# 1. Exact locks.

print("=" * 92)
print("NOTEBOOK 14 / 14.7D")
print("SOURCE-DERIVED AST TRANSFORMATION PRESERVATION REVIEW")
print("=" * 92)

paths = {
    key: ROOT / rel
    for key, rel in FILES.items()
}

for key, path in paths.items():
    assert path.is_file(), key
    assert not path.is_symlink(), key
    assert file_sha(path) == EXPECTED_SHA[key], key

parent = json.loads(
    paths["parent"].read_text(encoding="utf-8")
)

assert parent["decision"] == "PASS"
assert parent["content_sha256"] == PARENT_CONTENT_SHA

parent_body = dict(parent)
parent_body.pop("content_sha256")

assert canonical_sha(parent_body) == PARENT_CONTENT_SHA

assert parent["resolution"]["full_B12_authorized"] is False
assert (
    parent["resolution"]["canonical_writer_output_schema_verified"]
    is False
)

print("[PASS] PARENT_AND_BOTH_SOURCE_LOCKS")

# 2. Original canonical writer: exact dataflow statement sites.

writer = paths["writer"].read_text(encoding="utf-8")
writer_lines = writer.splitlines()
writer_tree = ast.parse(writer, filename=FILES["writer"])

print()
print("=" * 92)
print("A. CANONICAL WRITER DATAFLOW CALL SITES")
print("=" * 92)

print_calls(
    writer_tree,
    {
        "cand.insert",
        "candidate_parts.append",
        "pd.concat",
    },
)

# Preserve the lines establishing the original writer contract.
show(
    writer_lines,
    675,
    693,
    "B. PER-DATASET COLUMN INSERTION",
)

# Find the candidate_parts append site without assuming its line.
append_sites = [
    node.lineno
    for node in ast.walk(writer_tree)
    if isinstance(node, ast.Call)
    and dotted(node.func) == "candidate_parts.append"
]

assert append_sites, "No candidate_parts append site found"

for line in sorted(set(append_sites)):
    show(
        writer_lines,
        max(1, line - 4),
        min(len(writer_lines), line + 4),
        "C. CANDIDATE_PARTS APPEND",
    )

show(
    writer_lines,
    806,
    849,
    "D. UNIVERSE CONCAT AND REQUIRED COLUMN ACCESS",
)

# 3. Provider transformer: print all visit_* overrides.
#
# This is the unresolved area. A missing visit_Expr method alone
# does not prove full transformation preservation.

provider = paths["provider"].read_text(encoding="utf-8")
provider_lines = provider.splitlines()
provider_tree = ast.parse(provider, filename=FILES["provider"])

classes = {
    node.name: node
    for node in provider_tree.body
    if isinstance(node, ast.ClassDef)
}

functions = {
    node.name: node
    for node in provider_tree.body
    if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef))
}

assert "_SuppressedPeakTransformer" in classes
assert "_build_source_derived_code" in functions

transformer = classes["_SuppressedPeakTransformer"]

visitor_methods = [
    node
    for node in transformer.body
    if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef))
    and (
        node.name.startswith("visit_")
        or node.name == "generic_visit"
    )
]

print()
print("=" * 92)
print("E. COMPLETE TRANSFORMER OVERRIDE INVENTORY")
print("=" * 92)

for node in visitor_methods:
    print(
        node.name,
        "| lines",
        node.lineno,
        "-",
        node.end_lineno,
    )

for node in visitor_methods:
    show(
        provider_lines,
        node.lineno,
        node.end_lineno,
        "F. TRANSFORMER METHOD: " + node.name,
    )

# 4. Builder: identify transformation and runtime-tree rewrites.
#
# Print targeted code near AST transformations and tree.body changes.

builder = functions["_build_source_derived_code"]

print()
print("=" * 92)
print("G. BUILDER TRANSFORMATION CALL SITES")
print("=" * 92)

print_calls(
    builder,
    {
        "ast.parse",
        "ast.Module",
        "transformer.visit",
        "ast.fix_missing_locations",
        "compile",
        "_find_canonical_dataset_loop",
        "_find_universe_assignment",
    },
)

REWRITE_MARKERS = (
    "runtime_tree",
    "transformer",
    "dataset_loop",
    "universe_assignment",
)

rewrite_lines = [
    number
    for number in range(builder.lineno, builder.end_lineno + 1)
    if any(
        marker in provider_lines[number - 1]
        for marker in REWRITE_MARKERS
    )
]

# Print disjoint, bounded neighborhoods of relevant statements.
windows = []

for line in rewrite_lines:
    start = max(builder.lineno, line - 2)
    end = min(builder.end_lineno, line + 2)

    if windows and start <= windows[-1][1] + 1:
        windows[-1] = (
            windows[-1][0],
            max(windows[-1][1], end),
        )
    else:
        windows.append((start, end))

for start, end in windows:
    show(
        provider_lines,
        start,
        end,
        "H. BUILDER REWRITE EVIDENCE",
    )

# 5. Final integrity checks.

for key, path in paths.items():
    assert file_sha(path) == EXPECTED_SHA[key], key

print()
print("=" * 92)
print("NOTEBOOK 14 / 14.7D READ-ONLY REVIEW COMPLETE")
print("=" * 92)

print("transformer methods inspected:", len(visitor_methods))
print("candidate_parts append sites:", len(append_sites))
print("project modules imported: False")
print("source transformation executed: False")
print("real provider executed: False")
print("pytest executed: False")
print("artifact written: False")

print("dataset contract: FINAL_ADJUDICATION_PENDING")
print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")
```

```python
# =============================================================================
# NOTEBOOK 14 / 14.9A
# IN-MEMORY DATASET GUARD PATCH REHEARSAL
#
# Execute ONLY isolated copies of the two original output-check fragments.
# Add proposed missing-column checks to copies in memory.
#
# No production-source mutation, project-module imports, pytest,
# generated B12 executor, P995 model, source-derived runtime,
# historical oracle, GT, evaluator or Git.
#
# The local compile/exec below applies only to the isolated check fragments.
# =============================================================================

import ast
import copy
import hashlib
import json
from pathlib import Path

import pandas as pd

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

SPEC_REL = (
    "reports/notebook14_14_8g_"
    "suppressed_peak_b12_contract_remediation_spec_v1.json"
)

SPEC_FILE_SHA = (
    "20be5cd9a5f66700478ad427a7b6dd8965423dadc8854da928458c237b7f0863"
)

SPEC_CONTENT_SHA = (
    "34a77ee021b73e987fcea1d9d5a6654a15124f30f2e79b42840dbc1913df92fe"
)

PROVIDER_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

B12_REL = "src/biohub_cell_tracking/stage12_b12_portable.py"

SOURCE_LOCKS = {
    PROVIDER_REL: (
        "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
    ),
    B12_REL: (
        "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
    ),
}

OUT_REL = (
    "reports/notebook14_14_9a_"
    "in_memory_dataset_guard_patch_rehearsal_v1.json"
)


def sha_bytes(value):
    return hashlib.sha256(value).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(value):
    return sha_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


print("=" * 94)
print("NOTEBOOK 14 / 14.9A")
print("IN-MEMORY DATASET GUARD PATCH REHEARSAL")
print("=" * 94)

# -------------------------------------------------------------------------
# 1. Frozen input verification.
# -------------------------------------------------------------------------

spec_path = ROOT / SPEC_REL

assert spec_path.is_file() and not spec_path.is_symlink()
assert sha_file(spec_path) == SPEC_FILE_SHA

spec = json.loads(spec_path.read_text(encoding="utf-8"))

assert spec["decision"] == "PASS"
assert spec["content_sha256"] == SPEC_CONTENT_SHA

spec_body = dict(spec)
spec_body.pop("content_sha256")

assert canonical_sha(spec_body) == SPEC_CONTENT_SHA

assert (
    spec["resolution"]["status"]
    == "CONTRACT_REMEDIATION_SPECIFICATION_READY"
)

assert spec["resolution"]["implementation_authorized"] is False
assert spec["resolution"]["full_B12_authorized"] is False

for relative, expected in SOURCE_LOCKS.items():
    path = ROOT / relative

    assert path.is_file() and not path.is_symlink(), relative
    assert sha_file(path) == expected, relative

print("[PASS] SPEC_AND_SOURCE_LOCKS")

# -------------------------------------------------------------------------
# 2. Extract exact existing public-output check fragments.
# -------------------------------------------------------------------------

CONFIGS = {
    "suppressed_peak": {
        "path": PROVIDER_REL,
        "function": "suppressed_peak_provider",
        "frame_name": "result",
        "missing_exception": "RuntimeError",
        "existing_exception": RuntimeError,
    },
    "b12": {
        "path": B12_REL,
        "function": "materialize_b12_dataset",
        "frame_name": "b12",
        "missing_exception": "AssertionError",
        "existing_exception": AssertionError,
    },
}


def original_check_fragments(config):
    relative = config["path"]
    source = (ROOT / relative).read_text(encoding="utf-8")
    tree = ast.parse(source, filename=relative)

    matches = [
        node
        for node in tree.body
        if isinstance(node, ast.FunctionDef)
        and node.name == config["function"]
    ]

    assert len(matches) == 1
    function = matches[0]

    frame_name = config["frame_name"]

    matching_guards = [
        node
        for node in function.body
        if isinstance(node, ast.If)
        and ast.unparse(node.test)
        == f"'dataset' in {frame_name}.columns"
    ]

    assert len(matching_guards) == 1
    dataset_guard = matching_guards[0]

    returns = [
        node
        for node in function.body
        if isinstance(node, ast.Return)
    ]

    assert len(returns) == 1
    assert returns[0] is function.body[-1]

    fragment = []

    if frame_name == "b12":
        type_guards = [
            node
            for node in function.body
            if isinstance(node, ast.If)
            and (
                "isinstance(b12, pd.DataFrame)"
                in ast.unparse(node.test)
            )
        ]

        assert len(type_guards) == 1
        fragment.append(type_guards[0])

    fragment.extend([dataset_guard, returns[0]])

    # Confirm the original fragments contain no dynamic import or
    # separate runtime entrypoint.
    forbidden_calls = {
        "_execute_source_derived",
        "suppressed_peak_provider",
        "materialize_b12_dataset",
        "exec",
        "compile",
        "__import__",
    }

    for node in fragment:
        for child in ast.walk(node):
            assert not isinstance(
                child,
                (
                    ast.Import,
                    ast.ImportFrom,
                    ast.Global,
                    ast.Nonlocal,
                    ast.Await,
                    ast.Yield,
                    ast.YieldFrom,
                ),
            )

            if (
                isinstance(child, ast.Call)
                and isinstance(child.func, ast.Name)
            ):
                assert child.func.id not in forbidden_calls

    fragment_sha = sha_bytes(
        "\n".join(ast.dump(node) for node in fragment)
        .encode("utf-8")
    )

    return fragment, fragment_sha


# -------------------------------------------------------------------------
# 3. Construct baseline and candidate checkers in isolated namespaces.
#
# The candidate has ONE extra condition:
#     if "dataset" not in frame.columns: raise ...
#
# The actual project files remain unchanged.
# -------------------------------------------------------------------------

restricted_builtins = {
    "bool": bool,
    "isinstance": isinstance,
    "str": str,
    "set": set,
    "sorted": sorted,
    "repr": repr,
    "TypeError": TypeError,
    "RuntimeError": RuntimeError,
    "AssertionError": AssertionError,
}


def build_checker(config, fragment, *, proposed):
    frame_name = config["frame_name"]

    header = ast.parse(
        f"def isolated_check({frame_name}, dataset_id):\n"
        "    pass\n"
    ).body[0]

    body = copy.deepcopy(fragment)

    if proposed:
        exception_name = config["missing_exception"]

        check_source = (
            f"if 'dataset' not in {frame_name}.columns:\n"
            f"    raise {exception_name}("
            "'Required dataset column is missing.')\n"
        )

        new_check = ast.parse(check_source).body[0]

        # Preserve the original type guard at the beginning of B12.
        insert_at = 1 if frame_name == "b12" else 0

        body.insert(insert_at, new_check)

    header.body = body

    module = ast.Module(
        body=[header],
        type_ignores=[],
    )

    ast.fix_missing_locations(module)

    namespace = {
        "__name__": "isolated_dataset_guard_rehearsal",
        "__builtins__": restricted_builtins,
        "pd": pd,
    }

    code = compile(
        module,
        "<isolated-original-output-check-fragments>",
        "exec",
    )

    exec(code, namespace, namespace)

    return namespace["isolated_check"]


checkers = {}
fragment_hashes = {}

for name, config in CONFIGS.items():
    fragment, fragment_sha = original_check_fragments(config)

    fragment_hashes[name] = fragment_sha

    checkers[name] = {
        "baseline": build_checker(
            config,
            fragment,
            proposed=False,
        ),
        "proposed": build_checker(
            config,
            fragment,
            proposed=True,
        ),
    }

    print("[PASS] ORIGINAL_FRAGMENT_EXTRACTED:", name)
    print("       fragment SHA:", fragment_sha)

# -------------------------------------------------------------------------
# 4. Controlled input cases.
# -------------------------------------------------------------------------

CASES = {
    "matching": pd.DataFrame(
        {
            "dataset": ["demo", "demo"],
            "detector_prob": [0.8, 0.9],
        },
        index=[7, 9],
    ),

    "foreign": pd.DataFrame(
        {
            "dataset": ["other"],
            "detector_prob": [0.9],
        }
    ),

    "mixed": pd.DataFrame(
        {
            "dataset": ["demo", "other"],
            "detector_prob": [0.8, 0.9],
        }
    ),

    "empty_with_dataset": pd.DataFrame(
        {
            "dataset": pd.Series(dtype="str"),
            "detector_prob": pd.Series(dtype="float64"),
        }
    ),

    "nonempty_missing_dataset": pd.DataFrame(
        {
            "detector_prob": [0.9],
        },
        index=[17],
    ),

    "empty_missing_dataset": pd.DataFrame(
        {
            "detector_prob": pd.Series(dtype="float64"),
        }
    ),
}

RESULT_EXPECTATIONS = {
    "matching": "ACCEPT",
    "foreign": "REJECT",
    "mixed": "REJECT",
    "empty_with_dataset": "ACCEPT",
    "nonempty_missing_dataset": "ACCEPT",
    "empty_missing_dataset": "ACCEPT",
}

PROPOSED_EXPECTATIONS = {
    **RESULT_EXPECTATIONS,
    "nonempty_missing_dataset": "REJECT",
    "empty_missing_dataset": "REJECT",
}

observations = []

for boundary, config in CONFIGS.items():
    for variant in ("baseline", "proposed"):
        checker = checkers[boundary][variant]

        expectations = (
            RESULT_EXPECTATIONS
            if variant == "baseline"
            else PROPOSED_EXPECTATIONS
        )

        for case_name, original_frame in CASES.items():
            input_frame = original_frame.copy(deep=True)

            outcome = "ACCEPT"
            error_type = None

            try:
                result = checker(
                    input_frame,
                    "demo",
                )

            except Exception as error:
                outcome = "REJECT"
                error_type = type(error).__name__

                expected_exception = (
                    config["existing_exception"]
                    if case_name in ("foreign", "mixed")
                    else (
                        RuntimeError
                        if boundary == "suppressed_peak"
                        else AssertionError
                    )
                )

                assert type(error) is expected_exception, (
                    boundary,
                    variant,
                    case_name,
                    type(error).__name__,
                    str(error),
                )

            else:
                assert isinstance(result, pd.DataFrame)

                pd.testing.assert_frame_equal(
                    result.reset_index(drop=True),
                    original_frame.reset_index(drop=True),
                )

                assert result is not input_frame

                if boundary == "suppressed_peak":
                    assert result.index.tolist() == list(
                        range(len(result))
                    )

                else:
                    assert result.index.tolist() == (
                        original_frame.index.tolist()
                    )

            pd.testing.assert_frame_equal(
                input_frame,
                original_frame,
            )

            assert outcome == expectations[case_name], (
                boundary,
                variant,
                case_name,
                outcome,
                expectations[case_name],
            )

            record = {
                "boundary": boundary,
                "variant": variant,
                "case": case_name,
                "outcome": outcome,
                "exception": error_type,
            }

            observations.append(record)

            print(
                "[PASS]",
                boundary,
                variant,
                case_name,
                "=>",
                outcome,
            )

assert len(observations) == 24

print("[PASS] ALL_24_ISOLATED_COMPARISONS")

# -------------------------------------------------------------------------
# 5. Final immutability checks.
# -------------------------------------------------------------------------

assert sha_file(spec_path) == SPEC_FILE_SHA

for relative, expected in SOURCE_LOCKS.items():
    assert sha_file(ROOT / relative) == expected, relative

# -------------------------------------------------------------------------
# 6. Formal rehearsal report.
# -------------------------------------------------------------------------

status = "IN_MEMORY_DATASET_GUARD_PATCH_REHEARSAL_PASS"

payload = {
    "stage": "NOTEBOOK14_14_9A",
    "title": "IN_MEMORY_DATASET_GUARD_PATCH_REHEARSAL",
    "version": 1,
    "decision": "PASS",

    "parent_14_8G": {
        "path": SPEC_REL,
        "file_sha256": SPEC_FILE_SHA,
        "content_sha256": SPEC_CONTENT_SHA,
    },

    "source_locks": SOURCE_LOCKS,

    "isolated_original_fragment_sha256": fragment_hashes,

    "proposed_change": {
        "suppressed_peak": (
            "REJECT_MISSING_DATASET_COLUMN_WITH_RUNTIMEERROR"
        ),
        "b12": (
            "REJECT_MISSING_DATASET_COLUMN_WITH_ASSERTIONERROR"
        ),
        "production_source_modified": False,
    },

    "comparison_count": len(observations),
    "observations": observations,

    "execution_scope": {
        "mode": "ISOLATED_OUTPUT_CHECK_FRAGMENTS",
        "in_memory_ast_patch_only": True,
        "project_modules_imported": False,
        "production_provider_executed": False,
        "generated_B12_executor_executed": False,
        "source_derived_runtime_executed": False,
        "P995_model_executed": False,
        "pytest_executed": False,
        "GT_accessed": False,
        "evaluator_executed": False,
        "historical_oracle_opened": False,
        "Git_operation": False,
    },

    "limitations": [
        "Only the extracted output-check fragments were executed.",
        "The in-memory patch is not a production implementation.",
        "The original B12 executor and real provider were not executed.",
        "Null and duplicate dataset-column labels remain outside scope.",
        "Zero-candidate runtime behavior remains unverified.",
        "No complete B12 structural fidelity is established.",
        "Passing this rehearsal does not authorize source modification.",
    ],

    "resolution": {
        "status": status,
        "isolated_candidate_guard_behaviour_verified": True,
        "production_patch_implemented": False,
        "implementation_authorized": False,
        "new_pytest_execution_authorized": False,
        "real_provider_runtime_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
    },
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

out = ROOT / OUT_REL

assert out.parent.is_dir()

if out.exists():
    assert out.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"

else:
    out.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"

print()
print("=" * 94)
print("NOTEBOOK 14 / 14.9A PASS")
print("=" * 94)

print("isolated comparisons passed:", len(observations))
print("production source modified:", False)
print("pytest executed:", False)
print("real runtime executed:", False)

print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")

print("STATUS:", status)
print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(out))
print("artifact mode:", mode)
```

```python
# =============================================================================
# NOTEBOOK 14 / 14.9A
# IN-MEMORY DATASET GUARD PATCH REHEARSAL
#
# Execute ONLY isolated copies of the two original output-check fragments.
# Add proposed missing-column checks to copies in memory.
#
# No production-source mutation, project-module imports, pytest,
# generated B12 executor, P995 model, source-derived runtime,
# historical oracle, GT, evaluator or Git.
#
# The local compile/exec below applies only to the isolated check fragments.
# =============================================================================

import ast
import copy
import hashlib
import json
from pathlib import Path

import pandas as pd

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

SPEC_REL = (
    "reports/notebook14_14_8g_"
    "suppressed_peak_b12_contract_remediation_spec_v1.json"
)

SPEC_FILE_SHA = (
    "20be5cd9a5f66700478ad427a7b6dd8965423dadc8854da928458c237b7f0863"
)

SPEC_CONTENT_SHA = (
    "34a77ee021b73e987fcea1d9d5a6654a15124f30f2e79b42840dbc1913df92fe"
)

PROVIDER_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

B12_REL = "src/biohub_cell_tracking/stage12_b12_portable.py"

SOURCE_LOCKS = {
    PROVIDER_REL: (
        "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
    ),
    B12_REL: (
        "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
    ),
}

OUT_REL = (
    "reports/notebook14_14_9a_"
    "in_memory_dataset_guard_patch_rehearsal_v1.json"
)


def sha_bytes(value):
    return hashlib.sha256(value).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(value):
    return sha_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


print("=" * 94)
print("NOTEBOOK 14 / 14.9A")
print("IN-MEMORY DATASET GUARD PATCH REHEARSAL")
print("=" * 94)

# -------------------------------------------------------------------------
# 1. Frozen input verification.
# -------------------------------------------------------------------------

spec_path = ROOT / SPEC_REL

assert spec_path.is_file() and not spec_path.is_symlink()
assert sha_file(spec_path) == SPEC_FILE_SHA

spec = json.loads(spec_path.read_text(encoding="utf-8"))

assert spec["decision"] == "PASS"
assert spec["content_sha256"] == SPEC_CONTENT_SHA

spec_body = dict(spec)
spec_body.pop("content_sha256")

assert canonical_sha(spec_body) == SPEC_CONTENT_SHA

assert (
    spec["resolution"]["status"]
    == "CONTRACT_REMEDIATION_SPECIFICATION_READY"
)

assert spec["resolution"]["implementation_authorized"] is False
assert spec["resolution"]["full_B12_authorized"] is False

for relative, expected in SOURCE_LOCKS.items():
    path = ROOT / relative

    assert path.is_file() and not path.is_symlink(), relative
    assert sha_file(path) == expected, relative

print("[PASS] SPEC_AND_SOURCE_LOCKS")

# -------------------------------------------------------------------------
# 2. Extract exact existing public-output check fragments.
# -------------------------------------------------------------------------

CONFIGS = {
    "suppressed_peak": {
        "path": PROVIDER_REL,
        "function": "suppressed_peak_provider",
        "frame_name": "result",
        "missing_exception": "RuntimeError",
        "existing_exception": RuntimeError,
    },
    "b12": {
        "path": B12_REL,
        "function": "materialize_b12_dataset",
        "frame_name": "b12",
        "missing_exception": "AssertionError",
        "existing_exception": AssertionError,
    },
}


def original_check_fragments(config):
    relative = config["path"]
    source = (ROOT / relative).read_text(encoding="utf-8")
    tree = ast.parse(source, filename=relative)

    matches = [
        node
        for node in tree.body
        if isinstance(node, ast.FunctionDef)
        and node.name == config["function"]
    ]

    assert len(matches) == 1
    function = matches[0]

    frame_name = config["frame_name"]

    matching_guards = [
        node
        for node in function.body
        if isinstance(node, ast.If)
        and ast.unparse(node.test)
        == f"'dataset' in {frame_name}.columns"
    ]

    assert len(matching_guards) == 1
    dataset_guard = matching_guards[0]

    returns = [
        node
        for node in function.body
        if isinstance(node, ast.Return)
    ]

    assert len(returns) == 1
    assert returns[0] is function.body[-1]

    fragment = []

    if frame_name == "b12":
        type_guards = [
            node
            for node in function.body
            if isinstance(node, ast.If)
            and (
                "isinstance(b12, pd.DataFrame)"
                in ast.unparse(node.test)
            )
        ]

        assert len(type_guards) == 1
        fragment.append(type_guards[0])

    fragment.extend([dataset_guard, returns[0]])

    # Confirm the original fragments contain no dynamic import or
    # separate runtime entrypoint.
    forbidden_calls = {
        "_execute_source_derived",
        "suppressed_peak_provider",
        "materialize_b12_dataset",
        "exec",
        "compile",
        "__import__",
    }

    for node in fragment:
        for child in ast.walk(node):
            assert not isinstance(
                child,
                (
                    ast.Import,
                    ast.ImportFrom,
                    ast.Global,
                    ast.Nonlocal,
                    ast.Await,
                    ast.Yield,
                    ast.YieldFrom,
                ),
            )

            if (
                isinstance(child, ast.Call)
                and isinstance(child.func, ast.Name)
            ):
                assert child.func.id not in forbidden_calls

    fragment_sha = sha_bytes(
        "\n".join(ast.dump(node) for node in fragment)
        .encode("utf-8")
    )

    return fragment, fragment_sha


# -------------------------------------------------------------------------
# 3. Construct baseline and candidate checkers in isolated namespaces.
#
# The candidate has ONE extra condition:
#     if "dataset" not in frame.columns: raise ...
#
# The actual project files remain unchanged.
# -------------------------------------------------------------------------

restricted_builtins = {
    "bool": bool,
    "isinstance": isinstance,
    "str": str,
    "set": set,
    "sorted": sorted,
    "repr": repr,
    "TypeError": TypeError,
    "RuntimeError": RuntimeError,
    "AssertionError": AssertionError,
}


def build_checker(config, fragment, *, proposed):
    frame_name = config["frame_name"]

    header = ast.parse(
        f"def isolated_check({frame_name}, dataset_id):\n"
        "    pass\n"
    ).body[0]

    body = copy.deepcopy(fragment)

    if proposed:
        exception_name = config["missing_exception"]

        check_source = (
            f"if 'dataset' not in {frame_name}.columns:\n"
            f"    raise {exception_name}("
            "'Required dataset column is missing.')\n"
        )

        new_check = ast.parse(check_source).body[0]

        # Preserve the original type guard at the beginning of B12.
        insert_at = 1 if frame_name == "b12" else 0

        body.insert(insert_at, new_check)

    header.body = body

    module = ast.Module(
        body=[header],
        type_ignores=[],
    )

    ast.fix_missing_locations(module)

    namespace = {
        "__name__": "isolated_dataset_guard_rehearsal",
        "__builtins__": restricted_builtins,
        "pd": pd,
    }

    code = compile(
        module,
        "<isolated-original-output-check-fragments>",
        "exec",
    )

    exec(code, namespace, namespace)

    return namespace["isolated_check"]


checkers = {}
fragment_hashes = {}

for name, config in CONFIGS.items():
    fragment, fragment_sha = original_check_fragments(config)

    fragment_hashes[name] = fragment_sha

    checkers[name] = {
        "baseline": build_checker(
            config,
            fragment,
            proposed=False,
        ),
        "proposed": build_checker(
            config,
            fragment,
            proposed=True,
        ),
    }

    print("[PASS] ORIGINAL_FRAGMENT_EXTRACTED:", name)
    print("       fragment SHA:", fragment_sha)

# -------------------------------------------------------------------------
# 4. Controlled input cases.
# -------------------------------------------------------------------------

CASES = {
    "matching": pd.DataFrame(
        {
            "dataset": ["demo", "demo"],
            "detector_prob": [0.8, 0.9],
        },
        index=[7, 9],
    ),

    "foreign": pd.DataFrame(
        {
            "dataset": ["other"],
            "detector_prob": [0.9],
        }
    ),

    "mixed": pd.DataFrame(
        {
            "dataset": ["demo", "other"],
            "detector_prob": [0.8, 0.9],
        }
    ),

    "empty_with_dataset": pd.DataFrame(
        {
            "dataset": pd.Series(dtype="str"),
            "detector_prob": pd.Series(dtype="float64"),
        }
    ),

    "nonempty_missing_dataset": pd.DataFrame(
        {
            "detector_prob": [0.9],
        },
        index=[17],
    ),

    "empty_missing_dataset": pd.DataFrame(
        {
            "detector_prob": pd.Series(dtype="float64"),
        }
    ),
}

RESULT_EXPECTATIONS = {
    "matching": "ACCEPT",
    "foreign": "REJECT",
    "mixed": "REJECT",
    "empty_with_dataset": "ACCEPT",
    "nonempty_missing_dataset": "ACCEPT",
    "empty_missing_dataset": "ACCEPT",
}

PROPOSED_EXPECTATIONS = {
    **RESULT_EXPECTATIONS,
    "nonempty_missing_dataset": "REJECT",
    "empty_missing_dataset": "REJECT",
}

observations = []

for boundary, config in CONFIGS.items():
    for variant in ("baseline", "proposed"):
        checker = checkers[boundary][variant]

        expectations = (
            RESULT_EXPECTATIONS
            if variant == "baseline"
            else PROPOSED_EXPECTATIONS
        )

        for case_name, original_frame in CASES.items():
            input_frame = original_frame.copy(deep=True)

            outcome = "ACCEPT"
            error_type = None

            try:
                result = checker(
                    input_frame,
                    "demo",
                )

            except Exception as error:
                outcome = "REJECT"
                error_type = type(error).__name__

                expected_exception = (
                    config["existing_exception"]
                    if case_name in ("foreign", "mixed")
                    else (
                        RuntimeError
                        if boundary == "suppressed_peak"
                        else AssertionError
                    )
                )

                assert type(error) is expected_exception, (
                    boundary,
                    variant,
                    case_name,
                    type(error).__name__,
                    str(error),
                )

            else:
                assert isinstance(result, pd.DataFrame)

                pd.testing.assert_frame_equal(
                    result.reset_index(drop=True),
                    original_frame.reset_index(drop=True),
                )

                assert result is not input_frame

                if boundary == "suppressed_peak":
                    assert result.index.tolist() == list(
                        range(len(result))
                    )

                else:
                    assert result.index.tolist() == (
                        original_frame.index.tolist()
                    )

            pd.testing.assert_frame_equal(
                input_frame,
                original_frame,
            )

            assert outcome == expectations[case_name], (
                boundary,
                variant,
                case_name,
                outcome,
                expectations[case_name],
            )

            record = {
                "boundary": boundary,
                "variant": variant,
                "case": case_name,
                "outcome": outcome,
                "exception": error_type,
            }

            observations.append(record)

            print(
                "[PASS]",
                boundary,
                variant,
                case_name,
                "=>",
                outcome,
            )

assert len(observations) == 24

print("[PASS] ALL_24_ISOLATED_COMPARISONS")

# -------------------------------------------------------------------------
# 5. Final immutability checks.
# -------------------------------------------------------------------------

assert sha_file(spec_path) == SPEC_FILE_SHA

for relative, expected in SOURCE_LOCKS.items():
    assert sha_file(ROOT / relative) == expected, relative

# -------------------------------------------------------------------------
# 6. Formal rehearsal report.
# -------------------------------------------------------------------------

status = "IN_MEMORY_DATASET_GUARD_PATCH_REHEARSAL_PASS"

payload = {
    "stage": "NOTEBOOK14_14_9A",
    "title": "IN_MEMORY_DATASET_GUARD_PATCH_REHEARSAL",
    "version": 1,
    "decision": "PASS",

    "parent_14_8G": {
        "path": SPEC_REL,
        "file_sha256": SPEC_FILE_SHA,
        "content_sha256": SPEC_CONTENT_SHA,
    },

    "source_locks": SOURCE_LOCKS,

    "isolated_original_fragment_sha256": fragment_hashes,

    "proposed_change": {
        "suppressed_peak": (
            "REJECT_MISSING_DATASET_COLUMN_WITH_RUNTIMEERROR"
        ),
        "b12": (
            "REJECT_MISSING_DATASET_COLUMN_WITH_ASSERTIONERROR"
        ),
        "production_source_modified": False,
    },

    "comparison_count": len(observations),
    "observations": observations,

    "execution_scope": {
        "mode": "ISOLATED_OUTPUT_CHECK_FRAGMENTS",
        "in_memory_ast_patch_only": True,
        "project_modules_imported": False,
        "production_provider_executed": False,
        "generated_B12_executor_executed": False,
        "source_derived_runtime_executed": False,
        "P995_model_executed": False,
        "pytest_executed": False,
        "GT_accessed": False,
        "evaluator_executed": False,
        "historical_oracle_opened": False,
        "Git_operation": False,
    },

    "limitations": [
        "Only the extracted output-check fragments were executed.",
        "The in-memory patch is not a production implementation.",
        "The original B12 executor and real provider were not executed.",
        "Null and duplicate dataset-column labels remain outside scope.",
        "Zero-candidate runtime behavior remains unverified.",
        "No complete B12 structural fidelity is established.",
        "Passing this rehearsal does not authorize source modification.",
    ],

    "resolution": {
        "status": status,
        "isolated_candidate_guard_behaviour_verified": True,
        "production_patch_implemented": False,
        "implementation_authorized": False,
        "new_pytest_execution_authorized": False,
        "real_provider_runtime_authorized": False,
        "Morph_provider_authorized": False,
        "full_post_v2_authorized": False,
        "full_B12_authorized": False,
        "historical_oracle_access_authorized": False,
        "structural_fidelity_verified": False,
        "CURRENT_BEST_replaced": False,
    },
}

content_sha = canonical_sha(payload)

artifact = {
    "content_sha256": content_sha,
    **payload,
}

artifact_bytes = (
    json.dumps(
        artifact,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
        allow_nan=False,
    )
    + "\n"
).encode("utf-8")

out = ROOT / OUT_REL

assert out.parent.is_dir()

if out.exists():
    assert out.read_bytes() == artifact_bytes
    mode = "EXISTING_EXACT_ARTIFACT_REUSED"

else:
    out.write_bytes(artifact_bytes)
    mode = "NEW_FORMAL_ARTIFACT_WRITTEN"

print()
print("=" * 94)
print("NOTEBOOK 14 / 14.9A PASS")
print("=" * 94)

print("isolated comparisons passed:", len(observations))
print("production source modified:", False)
print("pytest executed:", False)
print("real runtime executed:", False)

print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")

print("STATUS:", status)
print("artifact:", OUT_REL)
print("artifact content SHA:", content_sha)
print("artifact file SHA:", sha_file(out))
print("artifact mode:", mode)
```

```python
# =============================================================================
# NOTEBOOK 14 / 14.9B
# MINIMAL DATASET GUARD PATCH PREVIEW — NOT APPLIED
#
# Source/test SHA locks + in-memory edits + AST validation + unified diff.
# Only permitted write: reports/...14_9b...diff
#
# No project imports, pytest, source mutation, generated B12 executor,
# P995 inference, GT, evaluator, historical oracle, or Git.
# =============================================================================

import ast
import difflib
import hashlib
import json
from pathlib import Path

ROOT = Path(
    "/Users/heqiuyan/Desktop/biohub-cell-tracking"
).resolve()

PARENT_REL = (
    "reports/notebook14_14_9a_"
    "in_memory_dataset_guard_patch_rehearsal_v1.json"
)

PARENT_FILE_SHA = (
    "d79438ba83b4f04ba5a5b0467633ae22898816c90af08f521482e391080dd922"
)

PARENT_CONTENT_SHA = (
    "f8e804e832ab71e2f171825984a594593cf0f4b024e1adbeab8381832b793dee"
)

PROVIDER_REL = (
    "src/biohub_cell_tracking/stage12_suppressed_peak_provider.py"
)

B12_REL = (
    "src/biohub_cell_tracking/stage12_b12_portable.py"
)

LOCKS = {
    PROVIDER_REL: (
        "e0c8a4ea149eb82860aa31437bb5004a0c7f23f160c2b47d77e37c00cf18e539"
    ),
    B12_REL: (
        "a064380c66c58fa509818428f2360e28d51e5bb524667b7bea4bc1783ccfde52"
    ),
    "tests/test_stage12_suppressed_peak_provider.py": (
        "eac928e54a26c086e531be7d38eeb46c67bcbba2d4ddd20cd00f255dcb63af4a"
    ),
    "tests/test_stage12_b12_portable.py": (
        "68f4c352347b74a529aea07c7b44d307e57ea8187e09b0953da14aec710ea8c7"
    ),
}

OUT_REL = (
    "reports/notebook14_14_9b_"
    "dataset_guard_minimal_patch_preview_v1.diff"
)


def sha_bytes(data):
    return hashlib.sha256(data).hexdigest()


def sha_file(path):
    return sha_bytes(path.read_bytes())


def canonical_sha(value):
    return sha_bytes(
        json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")
    )


print("=" * 88)
print("NOTEBOOK 14 / 14.9B")
print("MINIMAL DATASET GUARD PATCH PREVIEW — NOT APPLIED")
print("=" * 88)

# 1. Parent and source/test locks.

parent_path = ROOT / PARENT_REL

assert parent_path.is_file()
assert not parent_path.is_symlink()
assert sha_file(parent_path) == PARENT_FILE_SHA

parent = json.loads(
    parent_path.read_text(encoding="utf-8")
)

assert parent["decision"] == "PASS"
assert parent["content_sha256"] == PARENT_CONTENT_SHA

parent_body = dict(parent)
parent_body.pop("content_sha256")

assert canonical_sha(parent_body) == PARENT_CONTENT_SHA

assert parent["comparison_count"] == 24
assert parent["resolution"]["production_patch_implemented"] is False
assert parent["resolution"]["implementation_authorized"] is False
assert parent["resolution"]["full_B12_authorized"] is False

for rel, expected in LOCKS.items():
    path = ROOT / rel

    assert path.is_file(), rel
    assert not path.is_symlink(), rel
    assert sha_file(path) == expected, rel

print("[PASS] PARENT_AND_FOUR_SOURCE_TEST_LOCKS")

# 2. Define only the proposed missing-column additions.

CONFIGS = {
    PROVIDER_REL: {
        "function": "suppressed_peak_provider",
        "frame": "result",
        "exception": "RuntimeError",
        "message": (
            "Source-derived suppressed-peak provider output "
            "is missing required dataset column."
        ),
    },
    B12_REL: {
        "function": "materialize_b12_dataset",
        "frame": "b12",
        "exception": "AssertionError",
        "message": (
            "Portable B12 output is missing required dataset column."
        ),
    },
}


def get_function(tree, name):
    matches = [
        node
        for node in tree.body
        if isinstance(node, ast.FunctionDef)
        and node.name == name
    ]

    assert len(matches) == 1, name
    return matches[0]


def proposed_source(rel, config):
    raw = (ROOT / rel).read_bytes()

    assert b"\r\n" not in raw
    assert raw.endswith(b"\n")

    original = raw.decode("utf-8")
    original_lines = original.splitlines(keepends=True)

    original_tree = ast.parse(original, filename=rel)
    original_fn = get_function(
        original_tree,
        config["function"],
    )

    old_condition = (
        f"'dataset' in {config['frame']}.columns"
    )

    guards = [
        node
        for node in original_fn.body
        if isinstance(node, ast.If)
        and ast.unparse(node.test) == old_condition
    ]

    assert len(guards) == 1

    guard = guards[0]
    guard_index = original_fn.body.index(guard)

    assert original_lines[guard.lineno - 1].startswith("    if (")

    # In B12, retain the existing DataFrame type guard before this insertion.
    if config["frame"] == "b12":
        assert guard_index > 0

        prior_statements = original_fn.body[:guard_index]

        assert any(
            isinstance(node, ast.If)
            and "isinstance(b12, pd.DataFrame)"
            in ast.unparse(node.test)
            for node in prior_statements
        )

    frame = config["frame"]
    exception = config["exception"]
    message = config["message"]

    added = (
        f'    if "dataset" not in {frame}.columns:\n'
        f"        raise {exception}(\n"
        f'            "{message}"\n'
        "        )\n"
        "\n"
    )

    added_lines = added.splitlines(keepends=True)

    insert_at = guard.lineno - 1

    revised_lines = (
        original_lines[:insert_at]
        + added_lines
        + original_lines[insert_at:]
    )

    # The revision must consist of precisely the proposed insertion.
    assert (
        revised_lines[:insert_at]
        + revised_lines[insert_at + len(added_lines):]
    ) == original_lines

    revised = "".join(revised_lines)
    revised_tree = ast.parse(revised, filename=rel)

    revised_fn = get_function(
        revised_tree,
        config["function"],
    )

    # No existing function statements may change.
    original_body = [
        ast.dump(node, include_attributes=False)
        for node in original_fn.body
    ]

    revised_body = [
        ast.dump(node, include_attributes=False)
        for node in revised_fn.body
    ]

    assert len(revised_body) == len(original_body) + 1

    assert (
        revised_body[:guard_index]
        + revised_body[guard_index + 1:]
    ) == original_body

    new_guard = revised_fn.body[guard_index]

    assert isinstance(new_guard, ast.If)
    assert ast.unparse(new_guard.test) == (
        f"'dataset' not in {frame}.columns"
    )

    assert len(new_guard.body) == 1
    assert isinstance(new_guard.body[0], ast.Raise)

    raised = new_guard.body[0].exc

    assert isinstance(raised, ast.Call)
    assert isinstance(raised.func, ast.Name)
    assert raised.func.id == exception

    assert len(raised.args) == 1
    assert ast.literal_eval(raised.args[0]) == message

    return original, revised, guard.lineno


# 3. Build and inspect both source diffs entirely in memory.

diff_parts = []
preview_hashes = {}

for rel, config in CONFIGS.items():
    original, revised, guard_line = proposed_source(rel, config)

    preview_hashes[rel] = sha_bytes(revised.encode("utf-8"))

    diff_text = "".join(
        difflib.unified_diff(
            original.splitlines(keepends=True),
            revised.splitlines(keepends=True),
            fromfile=f"a/{rel}",
            tofile=f"b/{rel}",
            n=3,
        )
    )

    assert diff_text
    assert diff_text.startswith(f"--- a/{rel}\n")

    diff_parts.append(diff_text)

    print()
    print("[PASS] ISOLATED_PATCH_AST_VERIFIED:", rel)
    print("original guard line:", guard_line)
    print("proposed source SHA:", preview_hashes[rel])

patch_text = "\n".join(diff_parts)

print()
print("=" * 88)
print("PROPOSED UNIFIED DIFF — REVIEW ONLY")
print("=" * 88)
print(patch_text)

# 4. Recheck frozen inputs before the only permitted write.

assert sha_file(parent_path) == PARENT_FILE_SHA

for rel, expected in LOCKS.items():
    assert sha_file(ROOT / rel) == expected, rel

out = ROOT / OUT_REL
assert out.parent.is_dir()
assert not out.is_symlink()

patch_bytes = patch_text.encode("utf-8")

if out.exists():
    assert out.read_bytes() == patch_bytes
    mode = "EXISTING_EXACT_PREVIEW_REUSED"
else:
    out.write_bytes(patch_bytes)
    mode = "NEW_UNAPPLIED_PATCH_PREVIEW_WRITTEN"

assert sha_file(out) == sha_bytes(patch_bytes)

# 5. Source/test immutability check.

assert sha_file(parent_path) == PARENT_FILE_SHA

for rel, expected in LOCKS.items():
    assert sha_file(ROOT / rel) == expected, rel

print()
print("=" * 88)
print("NOTEBOOK 14 / 14.9B PASS")
print("=" * 88)

print("patch preview:", OUT_REL)
print("patch SHA:", sha_file(out))
print("preview mode:", mode)

print("production patch applied: False")
print("source/test files modified: False")
print("pytest executed: False")
print("real runtime executed: False")

print("implementation authorized: False")
print("full B12: BLOCKED")
print("historical oracle: CLOSED")
print("CURRENT_BEST: UNCHANGED")
```

```python

# Notebook 14 / 14.14B-2
# Warm2 inference: det_threshold = 0.9975
# Requires the successful 14.14A and 14.14B-1 cells.

from pathlib import Path
import hashlib
import json
import os
import subprocess
from collections import deque

def b14b2_require(condition, message):
    if not condition:
        raise RuntimeError(message)

def b14b2_sha256(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

print("========== 14.14B-2 / SAFETY PRECHECK ==========", flush=True)

required_globals = [
    "ROOT", "TEST_DIR", "TEST_IDS", "SOURCE_LOCKS",
    "WEIGHTS", "PREDICT_SCRIPT", "EXTRA_PYTHONPATH",
    "WORK_DIR", "SPLITS_PATH", "PRED_995_DIR",
    "PRED_9975_DIR", "TRACK_DIR", "FINAL_CSV",
    "METHOD_995", "METHOD_9975", "USERNAME",
]
missing = [name for name in required_globals if name not in globals()]
b14b2_require(not missing, f"缺少前置 Cell 变量：{missing}")

b14b2_require(
    ROOT == Path("/Users/heqiuyan/Desktop/biohub-cell-tracking"),
    "ROOT 与冻结项目路径不一致",
)
b14b2_require(
    os.environ.get("USER") == USERNAME,
    "USER 环境变量发生变化",
)
b14b2_require(
    METHOD_995 == "submission_v1_warm2_det0995"
    and METHOD_9975 == "submission_v1_warm2_det09975",
    "method 名称发生变化",
)
b14b2_require(
    PRED_995_DIR == (
        ROOT / "predictions" / USERNAME / METHOD_995 / "split_0"
    ),
    "第一组预测目录不一致",
)
b14b2_require(
    PRED_9975_DIR == (
        ROOT / "predictions" / USERNAME / METHOD_9975 / "split_0"
    ),
    "第二组预测目录不一致",
)

# Verify the existing splits file without rewriting it.
b14b2_require(
    SPLITS_PATH.is_file() and not SPLITS_PATH.is_symlink(),
    "14.14B-1 splits 文件缺失或不安全",
)
b14b2_require(
    b14b2_sha256(SPLITS_PATH)
    == "9d04a255751e3d21894047f689d770d684b27d258b5e70e55010ff7c06ebc02a",
    "splits SHA 与 14.14B-1 不一致",
)

with SPLITS_PATH.open("r", encoding="utf-8") as f:
    actual_splits = json.load(f)

expected_splits = [{
    "split": 0,
    "train": [],
    "test": list(TEST_IDS),
}]

b14b2_require(
    actual_splits == expected_splits,
    "splits 内容与冻结 test 清单不一致",
)

# Recheck source identities.
for label, (path, expected_sha) in SOURCE_LOCKS.items():
    b14b2_require(Path(path).is_file(), f"输入文件缺失：{label}")
    b14b2_require(
        b14b2_sha256(path) == expected_sha,
        f"输入 SHA 漂移：{label}",
    )

# Verify that the first inference finished and its outputs remain present.
first_log = WORK_DIR / "14_14b1_warm2_det0995.log"
b14b2_require(first_log.is_file(), "第一组推理日志缺失")

first_log_text = first_log.read_text(
    encoding="utf-8", errors="replace"
)
b14b2_require(
    "PROCESS_EXIT_CODE=0" in first_log_text,
    "第一组推理日志未记录成功退出",
)

b14b2_require(PRED_995_DIR.is_dir(), "第一组预测目录缺失")
first_outputs = sorted(PRED_995_DIR.glob("*.geff"))

b14b2_require(
    len(first_outputs) == len(TEST_IDS)
    and {p.stem for p in first_outputs} == set(TEST_IDS),
    "第一组预测 GEFF 集合不完整",
)

for path in first_outputs:
    b14b2_require(
        not path.is_symlink()
        and (path.is_dir() or (path.is_file() and path.stat().st_size > 0)),
        f"第一组预测无效：{path}",
    )

# Critical: predictor deletes existing GEFF files in its output directory.
# Require the ENTIRE second method directory to be new.
second_method_dir = PRED_9975_DIR.parent
b14b2_require(
    not second_method_dir.exists()
    and not second_method_dir.is_symlink(),
    f"第二组 method 目录已存在，禁止覆盖：{second_method_dir}",
)

b14b2_require(
    not TRACK_DIR.exists() and not TRACK_DIR.is_symlink(),
    "跟踪输出目录存在冲突",
)
b14b2_require(
    not FINAL_CSV.exists() and not FINAL_CSV.is_symlink(),
    "最终提交 CSV 已存在",
)
b14b2_require(
    EXTRA_PYTHONPATH.is_dir(),
    "官方 scripts 导入目录不存在",
)

python_exe = ROOT / ".venv/bin/python"
b14b2_require(python_exe.is_file(), "项目 Python 不存在")

log_path = WORK_DIR / "14_14b2_warm2_det09975.log"
b14b2_require(
    not log_path.exists() and not log_path.is_symlink(),
    f"第二组日志已存在：{log_path}",
)

print("SOURCE_LOCKS=PASS", flush=True)
print("FIRST_INFERENCE_OUTPUTS=4", flush=True)
print("SPLITS_LOCK=PASS", flush=True)
print("SECOND_OUTPUT_COLLISION=PASS", flush=True)

command = [
    str(python_exe),
    str(PREDICT_SCRIPT),
    "--method", METHOD_9975,
    "--data-dir", str(TEST_DIR),
    "--splits", str(SPLITS_PATH),
    "--split", "0",
    "--weights", str(WEIGHTS),
    "--det-threshold", "0.9975",
]

env = os.environ.copy()
prior_pythonpath = env.get("PYTHONPATH", "")
env["PYTHONPATH"] = (
    str(EXTRA_PYTHONPATH)
    + (os.pathsep + prior_pythonpath if prior_pythonpath else "")
)

print("\n========== 14.14B-2 / INFERENCE ==========", flush=True)
print("COMMAND=", command, flush=True)
print("LOG_PATH=", log_path, flush=True)
print("OUTPUT_DIR=", PRED_9975_DIR, flush=True)
print("POOL_KERNEL_UM=3.0 (frozen script default)", flush=True)
print("EVALUATE_FLAG=ABSENT", flush=True)

tail = deque(maxlen=25)
line_count = 0
exit_code = None

# Exclusive creation ensures no historical log is overwritten.
with log_path.open("x", encoding="utf-8") as log:
    log.write("14.14B-2 FROZEN WARM2 DETECTION\n")
    log.write("COMMAND_JSON=" + json.dumps(command) + "\n")
    log.write("PYTHONPATH=" + env["PYTHONPATH"] + "\n")
    log.write("SPLITS_SHA256=" + b14b2_sha256(SPLITS_PATH) + "\n")
    log.write("GT_READ=NO\n")
    log.flush()

    process = subprocess.Popen(
        command,
        cwd=str(ROOT),
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",
        errors="replace",
        bufsize=1,
    )

    try:
        for line in process.stdout:
            log.write(line)
            log.flush()

            line_count += 1
            tail.append(line.rstrip())

            if (
                line_count <= 15
                or line_count % 100 == 0
                or "Traceback" in line
                or "ERROR" in line
                or "Error:" in line
            ):
                print(line.rstrip(), flush=True)

        exit_code = process.wait()

    except BaseException:
        if process.poll() is None:
            process.terminate()
            try:
                process.wait(timeout=10)
            except subprocess.TimeoutExpired:
                process.kill()
                process.wait()
        raise

    finally:
        log.write(f"\nPROCESS_EXIT_CODE={process.returncode}\n")
        log.flush()

print("\n========== 14.14B-2 / PROCESS RESULT ==========")
print(f"EXIT_CODE={exit_code}")
print(f"LOG_LINES={line_count}")
print(f"LOG_PATH={log_path}")

if exit_code != 0:
    print("\nLAST_LOG_LINES:")
    for line in tail:
        print(line)
    raise RuntimeError(
        "14.14B2_INFERENCE_FAILED；保留现场，不得自动重试"
    )

# Verify the four expected GEFF outputs.
b14b2_require(
    PRED_9975_DIR.is_dir(),
    "推理成功退出，但第二组输出目录不存在",
)

actual_geffs = sorted(PRED_9975_DIR.glob("*.geff"))
actual_ids = [p.stem for p in actual_geffs]

b14b2_require(
    len(actual_geffs) == len(TEST_IDS)
    and set(actual_ids) == set(TEST_IDS),
    f"第二组预测 GEFF 集合不符：{actual_ids}",
)

for path in actual_geffs:
    b14b2_require(
        not path.is_symlink(),
        f"第二组 GEFF 是符号链接：{path}",
    )

    if path.is_dir():
        b14b2_require(
            any(child.is_file() for child in path.rglob("*")),
            f"第二组 GEFF 目录为空：{path}",
        )
    else:
        b14b2_require(
            path.is_file() and path.stat().st_size > 0,
            f"第二组 GEFF 文件为空或无效：{path}",
        )

    print(f"OUTPUT_PRESENT={path}")

for label, (path, expected_sha) in SOURCE_LOCKS.items():
    b14b2_require(
        b14b2_sha256(path) == expected_sha,
        f"推理后输入 SHA 漂移：{label}",
    )

# Ensure the first inference's outputs are still present.
b14b2_require(
    {p.stem for p in PRED_995_DIR.glob("*.geff")} == set(TEST_IDS),
    "第一组预测在第二次推理后发生文件集合变化",
)

print("\n14.14B2_WARM2_DET09975_PASS")
print("DATASETS=4")
print("DETECTION_THRESHOLD=0.9975")
print("SOURCE_LOCKS_POST_RUN=PASS")
print("FIRST_INFERENCE_OUTPUTS_PRESERVED=YES")
print("GT_READ=NO")
print("EVALUATOR_RUN=NO")
print("FROZEN_V9_RUN=NO")
print("KAGGLE_UPLOAD=NO")
```

```python

# Notebook 14 / 14.14C
# Frozen_v9 tracking on the four test datasets.
# Requires successful 14.14A, 14.14B-1 and 14.14B-2.

from pathlib import Path
import hashlib
import importlib
import json
import os
import sys
import traceback

def c14_require(condition, message):
    if not condition:
        raise RuntimeError(message)

def c14_sha256(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

def c14_artifact_hash(path):
    """Hash a GEFF file or directory, including relative file names."""
    path = Path(path)
    c14_require(path.exists() and not path.is_symlink(),
                f"GEFF 缺失或为符号链接：{path}")

    if path.is_file():
        return c14_sha256(path)

    c14_require(path.is_dir(), f"未知 GEFF 类型：{path}")

    files = sorted(p for p in path.rglob("*") if p.is_file())
    c14_require(files, f"GEFF 目录中没有文件：{path}")

    h = hashlib.sha256()
    for item in path.rglob("*"):
        c14_require(not item.is_symlink(),
                    f"GEFF 内含符号链接：{item}")

    for item in files:
        relative = item.relative_to(path).as_posix()
        h.update(relative.encode("utf-8"))
        h.update(b"\x00")
        h.update(bytes.fromhex(c14_sha256(item)))
        h.update(b"\x00")

    return h.hexdigest()

print("========== 14.14C / PRECHECK ==========", flush=True)

required = [
    "ROOT", "TEST_DIR", "TEST_IDS", "SOURCE_LOCKS",
    "WORK_DIR", "SPLITS_PATH", "PRED_995_DIR",
    "PRED_9975_DIR", "TRACK_DIR", "FINAL_CSV",
]
missing = [name for name in required if name not in globals()]
c14_require(not missing, f"缺少前置变量：{missing}")

c14_require(
    ROOT == Path("/Users/heqiuyan/Desktop/biohub-cell-tracking"),
    "ROOT 不一致",
)

c14_require(
    len(TEST_IDS) == 4 and len(set(TEST_IDS)) == 4,
    "test dataset 清单不正确",
)

c14_require(
    c14_sha256(SPLITS_PATH)
    == "9d04a255751e3d21894047f689d770d684b27d258b5e70e55010ff7c06ebc02a",
    "splits 文件 SHA 不一致",
)

for label, (path, expected_sha) in SOURCE_LOCKS.items():
    c14_require(
        Path(path).is_file() and c14_sha256(path) == expected_sha,
        f"冻结输入 SHA 不一致：{label}",
    )

for threshold, log_name in [
    ("0.995", "14_14b1_warm2_det0995.log"),
    ("0.9975", "14_14b2_warm2_det09975.log"),
]:
    log = WORK_DIR / log_name
    c14_require(log.is_file(), f"缺少检测日志：{log}")
    content = log.read_text(encoding="utf-8", errors="replace")
    c14_require(
        "PROCESS_EXIT_CODE=0" in content,
        f"{threshold} 检测日志未确认成功退出",
    )

c14_require(
    not TRACK_DIR.exists() and not TRACK_DIR.is_symlink(),
    f"跟踪输出目录已经存在，禁止覆盖：{TRACK_DIR}",
)
c14_require(
    not FINAL_CSV.exists() and not FINAL_CSV.is_symlink(),
    f"提交 CSV 已存在，禁止覆盖：{FINAL_CSV}",
)

log_path = WORK_DIR / "14_14c_frozen_v9_tracking.log"
manifest_path = WORK_DIR / "14_14c_tracking_manifest.json"

for path in (log_path, manifest_path):
    c14_require(
        not path.exists() and not path.is_symlink(),
        f"本轮输出文件已存在，禁止覆盖：{path}",
    )

# Lock the two prediction sets before tracking.
prediction_locks = {}

for label, pred_dir in [
    ("det0995", PRED_995_DIR),
    ("det09975", PRED_9975_DIR),
]:
    c14_require(pred_dir.is_dir() and not pred_dir.is_symlink(),
                f"检测目录不存在或不安全：{pred_dir}")

    actual = sorted(pred_dir.glob("*.geff"))
    c14_require(
        len(actual) == len(TEST_IDS)
        and {p.stem for p in actual} == set(TEST_IDS),
        f"{label} 的 GEFF 集合不正确",
    )

    prediction_locks[label] = {
        dataset_id: c14_artifact_hash(
            pred_dir / f"{dataset_id}.geff"
        )
        for dataset_id in TEST_IDS
    }

print("SOURCE_LOCKS=PASS")
print("DETECTION_INPUTS=8 GEFF")
print("OUTPUT_COLLISION=PASS")
print("PREDICTION_INPUT_HASHES:")
print(json.dumps(prediction_locks, indent=2))

# Use the source-backed modules already identified in 14.13.
# Add only the required module search paths to this notebook process.
for path in (
    ROOT / "src",
    ROOT / "external/official/src",
    ROOT / "external/official/scripts",
):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

fv9 = importlib.import_module("biohub_cell_tracking.frozen_v9")
official_io = importlib.import_module("tracking_cellmot.io")

c14_require(
    Path(fv9.__file__).resolve()
    == SOURCE_LOCKS["frozen_v9"][0].resolve(),
    "frozen_v9 模块不是已锁定的源码文件",
)
c14_require(
    Path(official_io.__file__).resolve()
    == SOURCE_LOCKS["official_io"][0].resolve(),
    "tracking_cellmot.io 不是已锁定的源码文件",
)

build_frozen_v9_policies = fv9.build_frozen_v9_policies
apply_frozen_v9 = fv9.apply_frozen_v9
save_graph = official_io.save_graph

print("FROZEN_V9_IMPORT=PASS")
print("OFFICIAL_IO_IMPORT=PASS")

# No output directory is created until all prechecks and imports pass.
TRACK_DIR.mkdir(parents=True, exist_ok=False)

output_records = []

with log_path.open("x", encoding="utf-8") as log:

    def c14_log(message):
        message = str(message)
        print(message, flush=True)
        log.write(message + "\n")
        log.flush()

    try:
        c14_log("========== 14.14C / BUILD POLICIES ==========")
        c14_log(f"TEST_IDS={json.dumps(TEST_IDS)}")
        c14_log(f"PRED_995_DIR={PRED_995_DIR}")
        c14_log(f"PRED_9975_DIR={PRED_9975_DIR}")
        c14_log(f"VOXEL_SOURCE_DIR={TEST_DIR}")
        c14_log("GT_READ=NO")
        c14_log("EVALUATOR_RUN=NO")

        # train_dir is deliberately TEST_DIR:
        # frozen_v9 uses it to read zarr scale metadata.
        policies = build_frozen_v9_policies(
            dataset_names=list(TEST_IDS),
            pred_995_dir=PRED_995_DIR,
            pred_9975_dir=PRED_9975_DIR,
            train_dir=TEST_DIR,
            prefix_map=None,
            progress=False,
        )

        c14_log("POLICY_BUILD=PASS")

        for field in (
            "gap_candidates",
            "single_policy",
            "isolated_candidates",
            "isolated_scope",
            "strict_two_policy",
        ):
            table = getattr(policies, field)
            c14_log(f"POLICY_TABLE {field}: rows={len(table)}")

        c14_log("\n========== 14.14C / APPLY AND SAVE ==========")

        for dataset_id in TEST_IDS:
            c14_log(f"START_DATASET={dataset_id}")

            graph, scale, application = apply_frozen_v9(
                dataset_name=dataset_id,
                pred_995_dir=PRED_995_DIR,
                pred_9975_dir=PRED_9975_DIR,
                policies=policies,
                train_dir=TEST_DIR,
                strict_single_policy=True,
            )

            scale_values = [float(v) for v in scale]
            c14_require(
                scale_values == [1.625, 0.40625, 0.40625],
                f"{dataset_id}: voxel scale 不一致：{scale_values}",
            )

            out_path = TRACK_DIR / f"{dataset_id}.geff"
            c14_require(
                not out_path.exists() and not out_path.is_symlink(),
                f"禁止覆盖跟踪结果：{out_path}",
            )

            # overwrite=False is required even though the directory is new.
            save_graph(graph, out_path, overwrite=False)

            c14_require(
                out_path.exists() and not out_path.is_symlink(),
                f"未生成跟踪 GEFF：{out_path}",
            )

            digest = c14_artifact_hash(out_path)

            record = {
                "dataset": dataset_id,
                "scale_zyx": scale_values,
                "output_path": str(out_path),
                "artifact_sha256": digest,
                "application_summary": str(application.to_row()),
            }
            output_records.append(record)

            c14_log(f"SAVED={out_path}")
            c14_log(f"OUTPUT_SHA256={digest}")
            c14_log(f"APPLICATION={record['application_summary']}")
            c14_log(f"DATASET_COMPLETE={dataset_id}")

        c14_log("\n========== 14.14C / POST-RUN CHECKS ==========")

        actual_outputs = sorted(TRACK_DIR.glob("*.geff"))
        c14_require(
            len(actual_outputs) == len(TEST_IDS)
            and {p.stem for p in actual_outputs} == set(TEST_IDS),
            "最终跟踪 GEFF 集合不完整",
        )

        for label, pred_dir in [
            ("det0995", PRED_995_DIR),
            ("det09975", PRED_9975_DIR),
        ]:
            for dataset_id in TEST_IDS:
                current = c14_artifact_hash(
                    pred_dir / f"{dataset_id}.geff"
                )
                c14_require(
                    current == prediction_locks[label][dataset_id],
                    f"检测输入在跟踪期间发生变化：{label}/{dataset_id}",
                )

        for label, (path, expected_sha) in SOURCE_LOCKS.items():
            c14_require(
                c14_sha256(path) == expected_sha,
                f"运行后源码 SHA 漂移：{label}",
            )

        # Persist a scratchpad manifest only after all four outputs pass.
        manifest = {
            "stage": "14.14C",
            "status": "FROZEN_V9_TRACKING_PASS",
            "test_ids": list(TEST_IDS),
            "predictions_sha256": prediction_locks,
            "tracking_outputs": output_records,
            "source_sha256": {
                label: sha for label, (_, sha) in SOURCE_LOCKS.items()
            },
            "gt_read": False,
            "evaluator_run": False,
            "kaggle_upload": False,
        }

        with manifest_path.open("x", encoding="utf-8") as f:
            json.dump(manifest, f, ensure_ascii=False, indent=2)
            f.write("\n")

        c14_log(f"MANIFEST={manifest_path}")
        c14_log(f"MANIFEST_SHA256={c14_sha256(manifest_path)}")
        c14_log("PREDICTION_INPUTS_UNCHANGED=YES")
        c14_log("SOURCE_LOCKS_POST_RUN=PASS")
        c14_log("14.14C_FROZEN_V9_TRACKING_PASS")

    except BaseException:
        c14_log("14.14C_FROZEN_V9_TRACKING_FAILED")
        c14_log(traceback.format_exc())
        raise
```

```python

# Notebook 14 / 14.14D-1
# Official GEFF -> staging CSV; GT-free structural validation.
# Requires successful 14.14A through 14.14C.

from pathlib import Path
from collections import Counter, defaultdict
import csv
import hashlib
import importlib.util
import json
import sys

def d14_require(condition, message):
    if not condition:
        raise RuntimeError(message)

def d14_sha256(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

print("========== 14.14D-1 / PRECHECK ==========", flush=True)

required = [
    "ROOT", "TEST_IDS", "SOURCE_LOCKS", "WORK_DIR",
    "TRACK_DIR", "FINAL_CSV", "c14_artifact_hash",
]
missing = [name for name in required if name not in globals()]
d14_require(not missing, f"缺少前置变量：{missing}")

STAGING_CSV = WORK_DIR / "14_14d_submission_unverified.csv"
D1_REPORT = WORK_DIR / "14_14d1_csv_validation.json"
ROUNDTRIP_DIR = WORK_DIR / "14_14d_roundtrip_geffs"
ROUNDTRIP_CSV = WORK_DIR / "14_14d_roundtrip.csv"

expected_manifest_sha = (
    "7687cd599166fa5854875681091833a7ed42dbd72ec713001f3d86514e3f28f8"
)
tracking_manifest = WORK_DIR / "14_14c_tracking_manifest.json"

d14_require(
    tracking_manifest.is_file()
    and d14_sha256(tracking_manifest) == expected_manifest_sha,
    "14.14C 跟踪 manifest 缺失或 SHA 漂移",
)

with tracking_manifest.open(encoding="utf-8") as f:
    manifest = json.load(f)

d14_require(
    manifest.get("status") == "FROZEN_V9_TRACKING_PASS",
    "14.14C 未记录成功状态",
)
d14_require(
    manifest.get("test_ids") == TEST_IDS,
    "跟踪 manifest 的 dataset 列表不一致",
)

for label, (path, expected_sha) in SOURCE_LOCKS.items():
    d14_require(
        Path(path).is_file() and d14_sha256(path) == expected_sha,
        f"冻结文件 SHA 不一致：{label}",
    )

records = manifest["tracking_outputs"]
d14_require(len(records) == 4, "manifest 中跟踪结果数量不为 4")
d14_require(
    {r["dataset"] for r in records} == set(TEST_IDS),
    "manifest dataset 集合不一致",
)

d14_require(TRACK_DIR.is_dir(), "跟踪 GEFF 目录不存在")
actual_geffs = sorted(TRACK_DIR.glob("*.geff"))

d14_require(
    len(actual_geffs) == 4
    and {p.stem for p in actual_geffs} == set(TEST_IDS),
    "跟踪 GEFF 文件集合不正确",
)

for record in records:
    dataset_id = record["dataset"]
    path = TRACK_DIR / f"{dataset_id}.geff"

    d14_require(
        Path(record["output_path"]) == path,
        f"{dataset_id}: manifest 输出路径不一致",
    )
    d14_require(
        c14_artifact_hash(path) == record["artifact_sha256"],
        f"{dataset_id}: 跟踪 GEFF SHA 漂移",
    )

    print(f"TRACKING_INPUT_LOCK=PASS {dataset_id}", flush=True)

for path in (
    STAGING_CSV, D1_REPORT, ROUNDTRIP_DIR,
    ROUNDTRIP_CSV, FINAL_CSV,
):
    d14_require(
        not path.exists() and not path.is_symlink(),
        f"目标路径已有内容，禁止覆盖：{path}",
    )

d14_require(not WORK_DIR.is_symlink(), "工作目录不能是符号链接")

print("TRACKING_MANIFEST=PASS")
print("SOURCE_LOCKS=PASS")
print("OUTPUT_COLLISION=PASS")
print("GT_READ=NO")
print("EVALUATOR_RUN=NO")

# Load only the frozen official conversion module.
conversion_path = SOURCE_LOCKS["geffs_to_csv"][0]

spec = importlib.util.spec_from_file_location(
    "notebook14_official_geffs_to_csv",
    str(conversion_path),
)
d14_require(spec is not None and spec.loader is not None,
            "无法加载官方 GEFF -> CSV 模块")

module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

d14_require(
    Path(module.__file__).resolve() == conversion_path.resolve(),
    "加载的官方转换模块路径不一致",
)

print("\n========== 14.14D-1 / OFFICIAL CONVERSION ==========")
print(f"IN_DIR={TRACK_DIR}")
print(f"STAGING_CSV={STAGING_CSV}")

# The staging path was checked as absent.
module.geffs_to_csv(TRACK_DIR, STAGING_CSV)

d14_require(
    STAGING_CSV.is_file()
    and not STAGING_CSV.is_symlink()
    and STAGING_CSV.stat().st_size > 0,
    "官方转换未生成有效 CSV 文件",
)

print(f"STAGING_SHA256={d14_sha256(STAGING_CSV)}")
print(f"STAGING_BYTES={STAGING_CSV.stat().st_size}")

EXPECTED_COLUMNS = [
    "id", "dataset", "row_type", "node_id",
    "t", "z", "y", "x", "source_id", "target_id",
]

def d14_validate_csv(path):
    """
    GT-free validation.

    Returns canonical node/edge coordinate multisets for the later
    round-trip check. Does not assume that graph node IDs remain
    unchanged after CSV -> GEFF conversion.
    """
    nodes_by_dataset = defaultdict(dict)
    edges_by_dataset = defaultdict(list)
    node_counts = Counter()
    edge_counts = Counter()
    node_signatures = Counter()
    edge_signatures = Counter()

    stage_by_dataset = {}
    dataset_order = []
    seen_datasets = set()
    last_dataset = None
    total_rows = 0

    with Path(path).open(newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)

        d14_require(
            reader.fieldnames == EXPECTED_COLUMNS,
            f"CSV 列或顺序不正确：{reader.fieldnames}",
        )

        for row in reader:
            dataset = row["dataset"]

            d14_require(
                dataset in TEST_IDS,
                f"CSV 含额外 dataset：{dataset}",
            )

            if dataset != last_dataset:
                d14_require(
                    dataset not in seen_datasets,
                    f"dataset 不连续：{dataset}",
                )
                seen_datasets.add(dataset)
                dataset_order.append(dataset)
                last_dataset = dataset

            try:
                numeric = {
                    key: int(row[key])
                    for key in (
                        "id", "node_id", "t", "z", "y", "x",
                        "source_id", "target_id",
                    )
                }
            except (TypeError, ValueError) as exc:
                raise RuntimeError(
                    f"第 {total_rows} 行存在非整数或空数值"
                ) from exc

            d14_require(
                numeric["id"] == total_rows,
                f"CSV id 不连续：row={total_rows}, id={numeric['id']}",
            )

            kind = row["row_type"]

            if kind == "node":
                d14_require(
                    stage_by_dataset.get(dataset) != "edge",
                    f"{dataset}: edge 行之后出现 node 行",
                )

                stage_by_dataset[dataset] = "node"

                node_id = numeric["node_id"]
                d14_require(node_id >= 0, "node_id 为负")
                d14_require(numeric["t"] >= 0, "节点时间为负")
                d14_require(
                    numeric["source_id"] == -1
                    and numeric["target_id"] == -1,
                    "node 行的 source_id/target_id 不为 -1",
                )

                node_map = nodes_by_dataset[dataset]
                d14_require(
                    node_id not in node_map,
                    f"{dataset}: 重复 node_id={node_id}",
                )

                coord = tuple(
                    numeric[key] for key in ("t", "z", "y", "x")
                )
                node_map[node_id] = coord
                node_counts[dataset] += 1
                node_signatures[(dataset, coord)] += 1

            elif kind == "edge":
                stage_by_dataset[dataset] = "edge"

                d14_require(
                    all(
                        numeric[key] == -1
                        for key in ("node_id", "t", "z", "y", "x")
                    ),
                    "edge 行中节点专用字段不为 -1",
                )

                source = numeric["source_id"]
                target = numeric["target_id"]

                d14_require(
                    source in nodes_by_dataset[dataset]
                    and target in nodes_by_dataset[dataset],
                    f"{dataset}: edge 引用不存在的节点 "
                    f"{source}->{target}",
                )

                edges_by_dataset[dataset].append((source, target))
                edge_counts[dataset] += 1

            else:
                raise RuntimeError(
                    f"非法 row_type：{kind!r}"
                )

            total_rows += 1

    d14_require(total_rows > 0, "提交 CSV 没有数据行")
    d14_require(
        set(seen_datasets) == set(TEST_IDS)
        and len(dataset_order) == 4,
        f"CSV dataset 集合不完整：{dataset_order}",
    )
    d14_require(
        dataset_order == sorted(TEST_IDS),
        f"CSV dataset 排序不符合官方转换预期：{dataset_order}",
    )

    coordinate_duplicates = {}

    for dataset in TEST_IDS:
        node_map = nodes_by_dataset[dataset]
        d14_require(node_map, f"{dataset}: 没有 node 行")

        seen_edges = set()
        for source, target in edges_by_dataset[dataset]:
            d14_require(
                (source, target) not in seen_edges,
                f"{dataset}: 重复 edge {source}->{target}",
            )
            seen_edges.add((source, target))

            src_coord = node_map[source]
            dst_coord = node_map[target]
            edge_signatures[(dataset, src_coord, dst_coord)] += 1

        coords = Counter(node_map.values())
        coordinate_duplicates[dataset] = sum(
            count - 1 for count in coords.values() if count > 1
        )

    return {
        "rows": total_rows,
        "dataset_order": dataset_order,
        "node_counts": dict(node_counts),
        "edge_counts": dict(edge_counts),
        "coordinate_duplicates": coordinate_duplicates,
        "node_signatures": node_signatures,
        "edge_signatures": edge_signatures,
    }

print("\n========== 14.14D-1 / STRUCTURE VALIDATION ==========")

result = d14_validate_csv(STAGING_CSV)

for dataset_id in TEST_IDS:
    print(
        f"{dataset_id}: "
        f"nodes={result['node_counts'][dataset_id]}, "
        f"edges={result['edge_counts'][dataset_id]}, "
        f"duplicate_coordinates="
        f"{result['coordinate_duplicates'][dataset_id]}"
    )

print(f"CSV_ROWS={result['rows']}")
print(f"DATASETS={result['dataset_order']}")

# The report records structural success, not full submission readiness.
report = {
    "stage": "14.14D-1",
    "status": "STAGING_CSV_STRUCTURE_PASS",
    "tracking_manifest_sha256": expected_manifest_sha,
    "staging_csv": str(STAGING_CSV),
    "staging_csv_sha256": d14_sha256(STAGING_CSV),
    "staging_csv_bytes": STAGING_CSV.stat().st_size,
    "rows": result["rows"],
    "dataset_order": result["dataset_order"],
    "node_counts": result["node_counts"],
    "edge_counts": result["edge_counts"],
    "coordinate_duplicates": result["coordinate_duplicates"],
    "gt_read": False,
    "evaluator_run": False,
    "roundtrip_complete": False,
    "kaggle_upload": False,
}

with D1_REPORT.open("x", encoding="utf-8") as f:
    json.dump(report, f, ensure_ascii=False, indent=2)
    f.write("\n")

print(f"D1_REPORT={D1_REPORT}")
print(f"D1_REPORT_SHA256={d14_sha256(D1_REPORT)}")
print("\n14.14D1_STAGING_CSV_STRUCTURE_PASS")
print("ROUNDTRIP_COMPLETE=NO")
print("FINAL_SUBMISSION_CREATED=NO")
print("GT_READ=NO")
print("KAGGLE_UPLOAD=NO")
```

```python

# Notebook 14 / 14.14D-2
# Official CSV -> GEFF -> CSV round-trip validation.
# No GT, evaluator, model inference, or Kaggle upload.

from pathlib import Path
import hashlib
import importlib.util
import json
import traceback

def d2_require(condition, message):
    if not condition:
        raise RuntimeError(message)

def d2_sha256(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()

def d2_counter_difference(original, reconstructed, limit=5):
    missing = original - reconstructed
    extra = reconstructed - original
    return {
        "missing_total": sum(missing.values()),
        "extra_total": sum(extra.values()),
        "missing_examples": [
            (repr(key), count)
            for key, count in missing.most_common(limit)
        ],
        "extra_examples": [
            (repr(key), count)
            for key, count in extra.most_common(limit)
        ],
    }

print("========== 14.14D-2 / PRECHECK ==========", flush=True)

required = [
    "ROOT", "TEST_IDS", "SOURCE_LOCKS", "WORK_DIR",
    "TRACK_DIR", "FINAL_CSV", "STAGING_CSV",
    "D1_REPORT", "ROUNDTRIP_DIR", "ROUNDTRIP_CSV",
    "d14_validate_csv", "c14_artifact_hash",
]
missing_globals = [name for name in required if name not in globals()]

d2_require(
    not missing_globals,
    f"缺少前置 Cell 变量：{missing_globals}",
)

d2_require(
    ROOT == Path("/Users/heqiuyan/Desktop/biohub-cell-tracking"),
    "项目 ROOT 不一致",
)

EXPECTED_STAGING_SHA = (
    "eb1d384ab97e77bbd215ab902c7635cc69c8241dee10dae4a55595add785d807"
)
EXPECTED_D1_REPORT_SHA = (
    "901a545bb39e791384488408ee190cc7882733a96f7c096ccc548bce259479ad"
)
EXPECTED_TRACK_MANIFEST_SHA = (
    "7687cd599166fa5854875681091833a7ed42dbd72ec713001f3d86514e3f28f8"
)

tracking_manifest_path = (
    WORK_DIR / "14_14c_tracking_manifest.json"
)
D2_REPORT = WORK_DIR / "14_14d2_roundtrip_validation.json"

d2_require(
    STAGING_CSV.is_file()
    and not STAGING_CSV.is_symlink()
    and d2_sha256(STAGING_CSV) == EXPECTED_STAGING_SHA,
    "暂存 CSV 缺失、非普通文件或 SHA 漂移",
)

d2_require(
    D1_REPORT.is_file()
    and not D1_REPORT.is_symlink()
    and d2_sha256(D1_REPORT) == EXPECTED_D1_REPORT_SHA,
    "14.14D-1 报告缺失或 SHA 漂移",
)

d2_require(
    tracking_manifest_path.is_file()
    and not tracking_manifest_path.is_symlink()
    and d2_sha256(tracking_manifest_path) == EXPECTED_TRACK_MANIFEST_SHA,
    "14.14C manifest 缺失或 SHA 漂移",
)

with D1_REPORT.open(encoding="utf-8") as f:
    d1_report = json.load(f)

with tracking_manifest_path.open(encoding="utf-8") as f:
    tracking_manifest = json.load(f)

d2_require(
    d1_report.get("status") == "STAGING_CSV_STRUCTURE_PASS",
    "14.14D-1 未通过",
)
d2_require(
    d1_report.get("staging_csv_sha256") == EXPECTED_STAGING_SHA,
    "D-1 报告与暂存 CSV 不一致",
)
d2_require(
    tracking_manifest.get("status") == "FROZEN_V9_TRACKING_PASS",
    "14.14C 未通过",
)
d2_require(
    tracking_manifest.get("test_ids") == list(TEST_IDS),
    "14.14C dataset 列表不一致",
)

for label, (path, expected_sha) in SOURCE_LOCKS.items():
    d2_require(
        Path(path).is_file()
        and d2_sha256(path) == expected_sha,
        f"冻结输入 SHA 漂移：{label}",
    )

d2_require(
    TRACK_DIR.is_dir() and not TRACK_DIR.is_symlink(),
    "原始跟踪目录不存在或为符号链接",
)

tracking_locks = {}
for record in tracking_manifest["tracking_outputs"]:
    dataset_id = record["dataset"]
    d2_require(dataset_id in TEST_IDS, f"非法 dataset：{dataset_id}")

    path = TRACK_DIR / f"{dataset_id}.geff"
    current_sha = c14_artifact_hash(path)

    d2_require(
        current_sha == record["artifact_sha256"],
        f"跟踪 GEFF 漂移：{dataset_id}",
    )
    tracking_locks[dataset_id] = current_sha

d2_require(
    set(tracking_locks) == set(TEST_IDS),
    "四份跟踪 GEFF 不完整",
)

for path in (
    ROUNDTRIP_DIR,
    ROUNDTRIP_CSV,
    D2_REPORT,
    FINAL_CSV,
):
    d2_require(
        not path.exists() and not path.is_symlink(),
        f"目标路径已存在，禁止覆盖：{path}",
    )

d2_require(
    WORK_DIR.is_dir() and not WORK_DIR.is_symlink(),
    "工作目录不安全",
)

print("SOURCE_LOCKS=PASS")
print("TRACKING_INPUT_LOCKS=PASS")
print("STAGING_CSV_LOCK=PASS")
print("D1_REPORT_LOCK=PASS")
print("OUTPUT_COLLISION=PASS")
print("GT_READ=NO")
print("EVALUATOR_RUN=NO")

print("\n========== 14.14D-2 / REVALIDATE ORIGINAL ==========")

original = d14_validate_csv(STAGING_CSV)

d2_require(
    original["rows"] == d1_report["rows"],
    "暂存 CSV 行数与 D-1 报告不符",
)
d2_require(
    original["node_counts"] == d1_report["node_counts"]
    and original["edge_counts"] == d1_report["edge_counts"],
    "暂存 CSV 的节点或边计数与 D-1 报告不符",
)

d2_require(
    all(
        original["coordinate_duplicates"][dataset_id] == 0
        for dataset_id in TEST_IDS
    ),
    "原始 CSV 存在重复坐标，不能仅靠坐标标识节点",
)

print(f"ORIGINAL_ROWS={original['rows']}")
print(f"ORIGINAL_NODES={sum(original['node_counts'].values())}")
print(f"ORIGINAL_EDGES={sum(original['edge_counts'].values())}")

print("\n========== 14.14D-2 / OFFICIAL ROUNDTRIP ==========")

csv_to_geffs_path = SOURCE_LOCKS["csv_to_geffs"][0]
geffs_to_csv_path = SOURCE_LOCKS["geffs_to_csv"][0]

def d2_load_official_module(name, path):
    spec = importlib.util.spec_from_file_location(name, str(path))
    d2_require(
        spec is not None and spec.loader is not None,
        f"无法加载模块：{path}",
    )
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    d2_require(
        Path(module.__file__).resolve() == path.resolve(),
        f"官方模块路径不符：{path}",
    )
    return module

reverse_module = d2_load_official_module(
    "notebook14_d2_csv_to_geffs",
    csv_to_geffs_path,
)
forward_module = d2_load_official_module(
    "notebook14_d2_geffs_to_csv",
    geffs_to_csv_path,
)

print(f"CSV_TO_GEFFS={csv_to_geffs_path}")
print(f"GEFFS_TO_CSV={geffs_to_csv_path}")
print(f"ROUNDTRIP_DIR={ROUNDTRIP_DIR}")

# overwrite=False prevents replacement of any existing GEFF.
# ROUNDTRIP_DIR was confirmed absent before calling.
converted_paths = reverse_module.csv_to_geffs(
    STAGING_CSV,
    ROUNDTRIP_DIR,
    overwrite=False,
)

d2_require(
    ROUNDTRIP_DIR.is_dir()
    and not ROUNDTRIP_DIR.is_symlink(),
    "官方 CSV -> GEFF 未生成输出目录",
)

actual_paths = sorted(ROUNDTRIP_DIR.glob("*.geff"))

d2_require(
    len(actual_paths) == len(TEST_IDS)
    and {p.stem for p in actual_paths} == set(TEST_IDS),
    f"往返 GEFF 集合不正确：{[p.stem for p in actual_paths]}",
)

d2_require(
    len(converted_paths) == len(TEST_IDS)
    and {Path(p).stem for p in converted_paths} == set(TEST_IDS),
    "官方转换器返回的路径集合不正确",
)

roundtrip_geff_hashes = {}

for dataset_id in TEST_IDS:
    path = ROUNDTRIP_DIR / f"{dataset_id}.geff"
    digest = c14_artifact_hash(path)
    roundtrip_geff_hashes[dataset_id] = digest
    print(f"ROUNDTRIP_GEFF={dataset_id} SHA256={digest}")

# The second conversion writes a distinct, previously nonexistent CSV.
forward_module.geffs_to_csv(
    ROUNDTRIP_DIR,
    ROUNDTRIP_CSV,
)

d2_require(
    ROUNDTRIP_CSV.is_file()
    and not ROUNDTRIP_CSV.is_symlink()
    and ROUNDTRIP_CSV.stat().st_size > 0,
    "往返 CSV 未生成或为空",
)

print(f"ROUNDTRIP_CSV_SHA256={d2_sha256(ROUNDTRIP_CSV)}")
print(f"ROUNDTRIP_CSV_BYTES={ROUNDTRIP_CSV.stat().st_size}")

print("\n========== 14.14D-2 / SEMANTIC COMPARISON ==========")

reconstructed = d14_validate_csv(ROUNDTRIP_CSV)

for dataset_id in TEST_IDS:
    before_nodes = original["node_counts"][dataset_id]
    after_nodes = reconstructed["node_counts"][dataset_id]
    before_edges = original["edge_counts"][dataset_id]
    after_edges = reconstructed["edge_counts"][dataset_id]

    print(
        f"{dataset_id}: "
        f"nodes={before_nodes}->{after_nodes}, "
        f"edges={before_edges}->{after_edges}, "
        f"duplicate_coordinates="
        f"{reconstructed['coordinate_duplicates'][dataset_id]}"
    )

d2_require(
    reconstructed["rows"] == original["rows"],
    "往返前后 CSV 行数不一致",
)

d2_require(
    reconstructed["dataset_order"] == original["dataset_order"],
    "往返前后 dataset 顺序不一致",
)

d2_require(
    reconstructed["node_counts"] == original["node_counts"],
    "往返前后每个 dataset 的节点数不一致",
)

d2_require(
    reconstructed["edge_counts"] == original["edge_counts"],
    "往返前后每个 dataset 的边数不一致",
)

# Compare (dataset, t, z, y, x) node identities, independent of node IDs.
node_difference = d2_counter_difference(
    original["node_signatures"],
    reconstructed["node_signatures"],
)

d2_require(
    original["node_signatures"] == reconstructed["node_signatures"],
    f"节点坐标语义不一致：{node_difference}",
)

# Compare directed coordinate-to-coordinate edges.
edge_difference = d2_counter_difference(
    original["edge_signatures"],
    reconstructed["edge_signatures"],
)

d2_require(
    original["edge_signatures"] == reconstructed["edge_signatures"],
    f"边的方向或端点语义不一致：{edge_difference}",
)

d2_require(
    all(
        reconstructed["coordinate_duplicates"][dataset_id] == 0
        for dataset_id in TEST_IDS
    ),
    "往返结果出现重复节点坐标",
)

# Confirm that neither the source data nor the staging artifact changed.
d2_require(
    d2_sha256(STAGING_CSV) == EXPECTED_STAGING_SHA,
    "往返转换后暂存 CSV SHA 漂移",
)

for dataset_id, previous_sha in tracking_locks.items():
    d2_require(
        c14_artifact_hash(TRACK_DIR / f"{dataset_id}.geff")
        == previous_sha,
        f"往返过程中原始 GEFF 发生变化：{dataset_id}",
    )

for label, (path, expected_sha) in SOURCE_LOCKS.items():
    d2_require(
        d2_sha256(path) == expected_sha,
        f"往返后冻结文件发生变化：{label}",
    )

print("NODE_SEMANTICS=PASS")
print("EDGE_SEMANTICS=PASS")
print("SOURCE_LOCKS_POST_RUN=PASS")
print("STAGING_CSV_UNCHANGED=YES")
print("ORIGINAL_TRACKING_GEFFS_UNCHANGED=YES")

report = {
    "stage": "14.14D-2",
    "status": "ROUNDTRIP_SEMANTICS_PASS",
    "tracking_manifest_sha256": EXPECTED_TRACK_MANIFEST_SHA,
    "d1_report_sha256": EXPECTED_D1_REPORT_SHA,
    "staging_csv": str(STAGING_CSV),
    "staging_csv_sha256": EXPECTED_STAGING_SHA,
    "roundtrip_dir": str(ROUNDTRIP_DIR),
    "roundtrip_geff_sha256": roundtrip_geff_hashes,
    "roundtrip_csv": str(ROUNDTRIP_CSV),
    "roundtrip_csv_sha256": d2_sha256(ROUNDTRIP_CSV),
    "rows": original["rows"],
    "node_counts": original["node_counts"],
    "edge_counts": original["edge_counts"],
    "node_semantics_equal": True,
    "directed_edge_semantics_equal": True,
    "source_locks_post_run": True,
    "gt_read": False,
    "evaluator_run": False,
    "final_submission_created": False,
    "kaggle_upload": False,
}

# Create evidence only after all semantic and SHA checks pass.
with D2_REPORT.open("x", encoding="utf-8") as f:
    json.dump(report, f, ensure_ascii=False, indent=2)
    f.write("\n")

print(f"D2_REPORT={D2_REPORT}")
print(f"D2_REPORT_SHA256={d2_sha256(D2_REPORT)}")

print("\n14.14D2_ROUNDTRIP_SEMANTICS_PASS")
print("FINAL_SUBMISSION_CREATED=NO")
print("GT_READ=NO")
print("EVALUATOR_RUN=NO")
print("KAGGLE_UPLOAD=NO")
```

```python

# Notebook 14 / 14.15A-1
# Build a source-and-weights bundle for Kaggle.
# Does NOT install packages, run inference, read GT, or submit.

from pathlib import Path
import hashlib
import json
import shutil
import zipfile

ROOT = Path("/Users/heqiuyan/Desktop/biohub-cell-tracking")
RUN_DIR = ROOT / "scratchpad/notebook14_14_15a_kaggle_bundle_v1"
STAGE = RUN_DIR / "bundle"
ZIP_PATH = RUN_DIR / "kaggle_source_warm2_v1.zip"
REPORT_PATH = RUN_DIR / "14_15a1_source_bundle_report.json"

def require(condition, message):
    if not condition:
        raise RuntimeError(message)

def sha256(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

# Frozen identities from the successful local 14.14 experiment.
LOCKS = {
    "models/official_baseline/official_baseline_fold0_warm2_lr1e5/split_0/edge_predictor_best.pth":
        "31d209bfde1fd5bbf3b2a661cdb4a25bb72922b832d315ee5a75cc32f44656c1",
    "models/official_baseline/official_baseline_fold0_warm2_lr1e5/split_0/config.json":
        "e9b4e396c58081bca08adf8275bd0bd1c2d3fd6eb091a1912a5116cb6de7b50a",
    "scripts/predict_unet_transformer_mps.py":
        "23b76f7c90ffe698e955efee2b4333277a07f5b58b7eff6e580f2bb2ff49bc55",
    "scripts/train_unet_transformer_mps.py":
        "3e9d0fb8f7f3e95e21d0af0c788a9e45d2bf67b771895d28efc4aa853ab506ac",
    "src/biohub_cell_tracking/frozen_v9.py":
        "486ae177e8709ddd17bf20d61dc628477a78d7434cd54a7ecf3729915b36ab17",
    "external/official/src/tracking_cellmot/io.py":
        "efae135b088cecaab463d889f16c885ef6da3ad27b0747327d8ddc28d866b7bd",
    "external/official/scripts/geffs_to_csv.py":
        "ecb8b121effb99f7889e6c9936f495fba6d98ea1bb541a2d1b1fb9f633bc9796",
    "external/official/scripts/csv_to_geffs.py":
        "914d151495c2c0f6bcab1feeebc6d0452b0ec63175cbce3b8774d20f79fa6f9a",
    "external/official/scripts/evaluate.py":
        "03ad4049530d3682c77435194e5d921981f331df3462abcde4a7156d1a57b7d3",
    "external/official/scripts/dataspec.py":
        "a1454dc5feb9a794dff4e6b0a9f98a75ad69e0bb1d459911de6faa491d18c2f6",
}

SAMPLE_REL = (
    "data/raw/competition/"
    "biohub-cell-tracking-during-development/sample_submission.csv"
)
SAMPLE_SHA = (
    "263dec32a126192f0ce4d5443b7432940fbeedf0d7ae4130156bf83657defc40"
)

EXTRA_SCRIPTS = [
    "external/official/scripts/augmentations.py",
]

PACKAGE_DIRS = [
    "src/biohub_cell_tracking",
    "external/official/src/tracking_cellmot",
]

ALLOWED_SUFFIXES = {".py", ".pyi", ".json", ".yaml", ".yml", ".toml"}

print("========== 14.15A-1 / PRECHECK ==========")

require(ROOT.is_dir(), "ROOT 不存在")
require(
    (ROOT / "scratchpad").is_dir()
    and not (ROOT / "scratchpad").is_symlink(),
    "scratchpad 目录不存在或不安全",
)
require(
    not RUN_DIR.exists() and not RUN_DIR.is_symlink(),
    f"输出目录已存在，禁止覆盖：{RUN_DIR}",
)

files = {}

def register(relative, expected_sha=None):
    relative = Path(relative)
    source = ROOT / relative

    require(source.is_file() and not source.is_symlink(),
            f"源文件缺失或为符号链接：{source}")

    digest = sha256(source)

    if expected_sha is not None:
        require(digest == expected_sha,
                f"冻结 SHA 不匹配：{relative} / {digest}")

    key = relative.as_posix()
    require(key not in files, f"重复文件：{key}")

    files[key] = {
        "source": source,
        "sha256": digest,
        "bytes": source.stat().st_size,
    }

for relative, expected_sha in LOCKS.items():
    register(relative, expected_sha)

register(SAMPLE_REL, SAMPLE_SHA)

for relative in EXTRA_SCRIPTS:
    register(relative)

omitted_resources = []

for relative_dir in PACKAGE_DIRS:
    package_dir = ROOT / relative_dir
    require(package_dir.is_dir() and not package_dir.is_symlink(),
            f"源码包缺失：{package_dir}")

    for path in sorted(package_dir.rglob("*")):
        require(not path.is_symlink(),
                f"源码包中存在符号链接：{path}")

        if not path.is_file():
            continue

        relative = path.relative_to(ROOT).as_posix()

        if path.suffix in ALLOWED_SUFFIXES:
            if relative not in files:
                register(relative)
        elif "__pycache__" not in path.parts:
            omitted_resources.append(relative)

require(
    sum(item["bytes"] for item in files.values()) < 100_000_000,
    "源码与权重体积异常，停止打包",
)

print(f"FROZEN_SHA_LOCKS=PASS ({len(LOCKS)})")
print(f"FILE_COUNT={len(files)}")
print(f"TOTAL_INPUT_BYTES={sum(v['bytes'] for v in files.values())}")
print(f"OMITTED_NON_CODE_RESOURCES={len(omitted_resources)}")

for relative in omitted_resources[:20]:
    print(f"OMITTED={relative}")

# All prechecks have passed. Begin writing only to this new run directory.
RUN_DIR.mkdir(parents=False, exist_ok=False)
STAGE.mkdir(exist_ok=False)

for relative, item in sorted(files.items()):
    destination = STAGE / relative
    destination.parent.mkdir(parents=True, exist_ok=True)
    shutil.copyfile(item["source"], destination)

    require(
        sha256(destination) == item["sha256"],
        f"复制后 SHA 不一致：{relative}",
    )

print("\n========== 14.15A-1 / WRITE MANIFEST ==========")

manifest = {
    "stage": "14.15A-1",
    "status": "SOURCE_AND_WEIGHTS_BUNDLE_ONLY",
    "pipeline": "FROZEN_V9_WARM2_PORTABLE_SUBMISSION_V1",
    "files": {
        relative: {
            "sha256": item["sha256"],
            "bytes": item["bytes"],
        }
        for relative, item in sorted(files.items())
    },
    "omitted_non_code_resources": omitted_resources,
    "runtime_dependencies_included": False,
    "tracksdata_linux_wheel_included": False,
    "kaggle_inference_run": False,
    "gt_read": False,
    "evaluator_run": False,
    "kaggle_submission": False,
}

manifest_path = STAGE / "bundle_manifest.json"

with manifest_path.open("x", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2, ensure_ascii=False)
    f.write("\n")

manifest_sha = sha256(manifest_path)

print(f"MANIFEST_SHA256={manifest_sha}")

print("\n========== 14.15A-1 / BUILD ZIP ==========")

with zipfile.ZipFile(
    ZIP_PATH,
    mode="x",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=6,
    allowZip64=True,
) as archive:

    for path in sorted(STAGE.rglob("*")):
        if path.is_file():
            archive.write(
                path,
                arcname=Path("bundle") / path.relative_to(STAGE),
            )

expected_members = {
    "bundle/" + relative for relative in files
} | {"bundle/bundle_manifest.json"}

with zipfile.ZipFile(ZIP_PATH, "r") as archive:
    actual_members = set(archive.namelist())

    require(
        actual_members == expected_members,
        "ZIP 文件集合与 manifest 不一致",
    )
    require(
        archive.testzip() is None,
        "ZIP 完整性检查失败",
    )

    for relative, item in files.items():
        archived_bytes = archive.read("bundle/" + relative)
        archived_sha = hashlib.sha256(archived_bytes).hexdigest()

        require(
            archived_sha == item["sha256"],
            f"ZIP 内文件 SHA 不一致：{relative}",
        )

    require(
        hashlib.sha256(
            archive.read("bundle/bundle_manifest.json")
        ).hexdigest() == manifest_sha,
        "ZIP 内 manifest SHA 不一致",
    )

zip_sha = sha256(ZIP_PATH)

report = {
    "stage": "14.15A-1",
    "status": "SOURCE_BUNDLE_INTEGRITY_PASS",
    "zip_path": str(ZIP_PATH),
    "zip_sha256": zip_sha,
    "zip_bytes": ZIP_PATH.stat().st_size,
    "manifest_sha256": manifest_sha,
    "file_count": len(files),
    "runtime_dependencies_included": False,
    "tracksdata_linux_wheel_included": False,
    "kaggle_runtime_verified": False,
    "gt_read": False,
    "evaluator_run": False,
    "kaggle_upload": False,
}

with REPORT_PATH.open("x", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)
    f.write("\n")

print("\n========== 14.15A-1 / RESULT ==========")
print(f"ZIP_PATH={ZIP_PATH}")
print(f"ZIP_SHA256={zip_sha}")
print(f"ZIP_BYTES={ZIP_PATH.stat().st_size}")
print(f"FILE_COUNT={len(files)}")
print(f"REPORT_SHA256={sha256(REPORT_PATH)}")
print("14.15A1_SOURCE_BUNDLE_INTEGRITY_PASS")
print("KAGGLE_RUNTIME_VERIFIED=NO")
print("KAGGLE_UPLOAD=NO")
print("GT_READ=NO")
```
